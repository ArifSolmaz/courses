## Part 1: What Is a Tree?

A **tree** is a hierarchical data structure made of **nodes** connected by **edges**. Unlike a list (which is linear), a tree **branches out** — each node can have multiple children.

### Tree Terminology

```
            A          ← Root (top node, no parent)
          /   \
        B       C      ← B and C are children of A
       / \       \
      D   E       F    ← D, E are children of B; F is child of C
         / \
        G   H          ← G, H are children of E
```

| Term | Definition | Example |
|------|-----------|--------|
| **Root** | The topmost node (no parent) | A |
| **Node** | Any element in the tree | A, B, C, D, E, F, G, H |
| **Leaf** | A node with no children | D, F, G, H |
| **Parent** | A node that has children | A is parent of B, C |
| **Child** | A node connected below another | B, C are children of A |
| **Edge** | The connection between two nodes | A—B, A—C, B—D, etc. |
| **Depth** | Distance from root to a node | depth(A)=0, depth(B)=1, depth(G)=3 |
| **Height** | Distance from a node to its deepest leaf | height(A)=3, height(B)=2 |
| **Subtree** | A node and all its descendants | B, D, E, G, H form a subtree |

### Real-World Examples of Trees

| Example | Root | Children | Leaves |
|---------|------|----------|--------|
| **File system** | `/` (root directory) | Folders | Files |
| **Family tree** | Ancestor | Children | People with no kids |
| **HTML DOM** | `<html>` | `<head>`, `<body>` | Text nodes |
| **Company org chart** | CEO | Department heads | Individual employees |
| **Decision tree** | First question | Next questions | Final decisions |

**Figure 1.1** — A file system is a tree

````python
# A file system is a tree!
# Let's visualise a simple one using text

file_system = {
    "/": {
        "home": {
            "alice": {
                "documents": ["report.pdf", "notes.txt"],
                "pictures": ["vacation.jpg"],
            },
            "bob": {
                "code": ["main.py", "utils.py"],
            },
        },
        "etc": ["config.cfg"],
    }
}

def print_tree(tree, indent=0):
    """Recursively print a nested dict as a tree."""
    if isinstance(tree, dict):
        for key, value in tree.items():
            print(" " * indent + f"\u251c\u2500\u2500 {key}/")
            print_tree(value, indent + 4)
    elif isinstance(tree, list):
        for item in tree:
            print(" " * indent + f"\u251c\u2500\u2500 {item}")

print("File System Tree:")
print_tree(file_system)
````

## Part 2: Binary Trees

A **binary tree** is a tree where each node has **at most 2 children**: a **left** child and a **right** child.

```
         10
        /  \
       5    15
      / \     \
     3   7    20
```

Binary trees are the most commonly used tree type because:
- Simple to implement
- Foundation for BSTs, heaps, expression trees
- Efficient for searching and sorting

**Figure 2.1** — The Node class: building block of a binary tree

````python
class Node:
    """A single node in a binary tree."""

    def __init__(self, value):
        self.value = value
        self.left = None   # left child
        self.right = None  # right child

    def __repr__(self):
        return f"Node({self.value})"


# Create a simple tree:
#        10
#       /  \
#      5    15

root = Node(10)
root.left = Node(5)
root.right = Node(15)

print(f"Root:       {root}")
print(f"Left child: {root.left}")
print(f"Right child:{root.right}")
````

**Figure 2.2** — Building a larger binary tree

````python
# Build this tree:
#          1
#        /   \
#       2     3
#      / \   / \
#     4   5 6   7

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.left = Node(6)
root.right.right = Node(7)

# Let's verify the structure
print(f"Root: {root.value}")
print(f"Root -> Left: {root.left.value}")
print(f"Root -> Right: {root.right.value}")
print(f"Root -> Left -> Left: {root.left.left.value}")
print(f"Root -> Left -> Right: {root.left.right.value}")
print(f"Root -> Right -> Left: {root.right.left.value}")
print(f"Root -> Right -> Right: {root.right.right.value}")
````

**Figure 2.3** — A helper function to pretty-print binary trees

````python
def print_binary_tree(node, level=0, prefix="Root: "):
    """Print a binary tree in a readable sideways format."""
    if node is not None:
        print(" " * (level * 4) + prefix + str(node.value))
        if node.left is not None or node.right is not None:
            if node.left:
                print_binary_tree(node.left, level + 1, "L--- ")
            else:
                print(" " * ((level + 1) * 4) + "L--- (empty)")
            if node.right:
                print_binary_tree(node.right, level + 1, "R--- ")
            else:
                print(" " * ((level + 1) * 4) + "R--- (empty)")

print_binary_tree(root)
````

**Figure 2.4** — Counting nodes, leaves, and computing tree height

````python
def count_nodes(node):
    """Count total number of nodes in the tree."""
    if node is None:
        return 0
    return 1 + count_nodes(node.left) + count_nodes(node.right)

def count_leaves(node):
    """Count the number of leaf nodes."""
    if node is None:
        return 0
    if node.left is None and node.right is None:
        return 1  # This is a leaf
    return count_leaves(node.left) + count_leaves(node.right)

def tree_height(node):
    """Compute the height of the tree."""
    if node is None:
        return -1  # Convention: empty tree has height -1
    return 1 + max(tree_height(node.left), tree_height(node.right))

print(f"Total nodes:  {count_nodes(root)}")
print(f"Leaf nodes:   {count_leaves(root)}")
print(f"Tree height:  {tree_height(root)}")
````

## Part 3: Tree Traversals

**Traversal** means visiting every node in the tree exactly once. There are three main ways to traverse a binary tree:

| Traversal | Order | Mnemonic |
|-----------|-------|---------|
| **In-order** | Left → Root → Right | "Left, Me, Right" |
| **Pre-order** | Root → Left → Right | "Me first, then kids" |
| **Post-order** | Left → Right → Root | "Kids first, then me" |

All three use **recursion** — the function calls itself on subtrees.

### Step-by-Step Walkthrough

Using this tree:
```
         1
        / \
       2   3
      / \
     4   5
```

**In-order (Left, Root, Right):**
- Go left from 1 → Go left from 2 → Visit 4 → Visit 2 → Visit 5 → Visit 1 → Visit 3
- Result: **4, 2, 5, 1, 3**

**Pre-order (Root, Left, Right):**
- Visit 1 → Visit 2 → Visit 4 → Visit 5 → Visit 3
- Result: **1, 2, 4, 5, 3**

**Post-order (Left, Right, Root):**
- Visit 4 → Visit 5 → Visit 2 → Visit 3 → Visit 1
- Result: **4, 5, 2, 3, 1**

**Figure 3.1** — Implementing in-order traversal

````python
def inorder(node):
    """In-order traversal: Left -> Root -> Right"""
    if node is None:
        return []
    return inorder(node.left) + [node.value] + inorder(node.right)

# Build the tree:
#        1
#       / \
#      2   3
#     / \
#    4   5
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

result = inorder(root)
print(f"In-order: {result}")
# Expected: [4, 2, 5, 1, 3]
````

**Figure 3.2** — Implementing pre-order traversal

````python
def preorder(node):
    """Pre-order traversal: Root -> Left -> Right"""
    if node is None:
        return []
    return [node.value] + preorder(node.left) + preorder(node.right)

result = preorder(root)
print(f"Pre-order: {result}")
# Expected: [1, 2, 4, 5, 3]
````

**Figure 3.3** — Implementing post-order traversal

````python
def postorder(node):
    """Post-order traversal: Left -> Right -> Root"""
    if node is None:
        return []
    return postorder(node.left) + postorder(node.right) + [node.value]

result = postorder(root)
print(f"Post-order: {result}")
# Expected: [4, 5, 2, 3, 1]
````

**Figure 3.4** — All three traversals side by side

````python
# Build a larger tree for demonstration:
#          1
#        /   \
#       2     3
#      / \   / \
#     4   5 6   7

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.left = Node(6)
root.right.right = Node(7)

print("Tree:")
print_binary_tree(root)
print()
print(f"In-order   (L, Root, R): {inorder(root)}")
print(f"Pre-order  (Root, L, R): {preorder(root)}")
print(f"Post-order (L, R, Root): {postorder(root)}")
````

## Part 4: Step-by-Step Traversal Tracing

Let's add **print statements** inside the traversals to see exactly what happens at each step.

**Figure 4.1** — Verbose in-order traversal showing recursion steps

````python
def inorder_verbose(node, depth=0):
    """In-order traversal with step-by-step output."""
    indent = "  " * depth
    if node is None:
        print(f"{indent}(None - return)")
        return []

    print(f"{indent}Entering node {node.value}")

    # Step 1: Go left
    print(f"{indent}  Going LEFT from {node.value}")
    left_result = inorder_verbose(node.left, depth + 1)

    # Step 2: Visit this node
    print(f"{indent}  >>> VISIT {node.value} <<<")

    # Step 3: Go right
    print(f"{indent}  Going RIGHT from {node.value}")
    right_result = inorder_verbose(node.right, depth + 1)

    return left_result + [node.value] + right_result

# Use a small tree:
#      1
#     / \
#    2   3

small = Node(1)
small.left = Node(2)
small.right = Node(3)

print("=== In-Order Traversal Trace ===")
result = inorder_verbose(small)
print(f"\nResult: {result}")
````

**Figure 4.2** — Verbose pre-order traversal

````python
def preorder_verbose(node, depth=0):
    """Pre-order traversal with step-by-step output."""
    indent = "  " * depth
    if node is None:
        print(f"{indent}(None - return)")
        return []

    # Step 1: Visit this node FIRST
    print(f"{indent}>>> VISIT {node.value} <<<")

    # Step 2: Go left
    print(f"{indent}  Going LEFT from {node.value}")
    left_result = preorder_verbose(node.left, depth + 1)

    # Step 3: Go right
    print(f"{indent}  Going RIGHT from {node.value}")
    right_result = preorder_verbose(node.right, depth + 1)

    return [node.value] + left_result + right_result

print("=== Pre-Order Traversal Trace ===")
result = preorder_verbose(small)
print(f"\nResult: {result}")
````

## Part 5: Building Trees from User Input

Let's create a helper function to build binary trees more conveniently from a list.

**Figure 5.1** — Building a binary tree from a list (level-order)

````python
def build_tree_from_list(values):
    """Build a binary tree from a list (level-order).
    Use None for missing nodes.

    Example: [1, 2, 3, 4, 5, None, 7]
    Builds:
           1
          / \
         2   3
        / \   \
       4   5   7
    """
    if not values:
        return None

    root = Node(values[0])
    queue = [root]
    i = 1

    while queue and i < len(values):
        current = queue.pop(0)

        # Left child
        if i < len(values) and values[i] is not None:
            current.left = Node(values[i])
            queue.append(current.left)
        i += 1

        # Right child
        if i < len(values) and values[i] is not None:
            current.right = Node(values[i])
            queue.append(current.right)
        i += 1

    return root

# Build a tree from a list
tree = build_tree_from_list([1, 2, 3, 4, 5, None, 7])

print("Tree built from [1, 2, 3, 4, 5, None, 7]:")
print_binary_tree(tree)
print()
print(f"In-order:   {inorder(tree)}")
print(f"Pre-order:  {preorder(tree)}")
print(f"Post-order: {postorder(tree)}")
````

**Figure 5.2** — Level-order traversal (Breadth-First Search)

````python
def level_order(root):
    """Level-order traversal (BFS): visit nodes level by level."""
    if root is None:
        return []

    result = []
    queue = [root]

    while queue:
        node = queue.pop(0)
        result.append(node.value)

        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)

    return result

# Test on our tree
#          1
#        /   \
#       2     3
#      / \   / \
#     4   5 6   7

tree = build_tree_from_list([1, 2, 3, 4, 5, 6, 7])
print(f"Level-order: {level_order(tree)}")
# Expected: [1, 2, 3, 4, 5, 6, 7]
````

## Part 6: Mini Application — Verifying Traversal Output

Let's build a small tool that creates a tree and lets you check your traversal answers.

**Figure 6.1** — Traversal verifier: check your answers

````python
def verify_traversal(tree_values, traversal_type, student_answer):
    """Build a tree and check if the student's traversal answer is correct."""
    tree = build_tree_from_list(tree_values)

    if traversal_type == "inorder":
        correct = inorder(tree)
    elif traversal_type == "preorder":
        correct = preorder(tree)
    elif traversal_type == "postorder":
        correct = postorder(tree)
    else:
        print(f"Unknown traversal type: {traversal_type}")
        return

    if student_answer == correct:
        print(f"\u2705 Correct! {traversal_type}: {correct}")
    else:
        print(f"\u274c Incorrect.")
        print(f"   Your answer: {student_answer}")
        print(f"   Correct:     {correct}")

# Test tree: [10, 5, 15, 3, 7]
#       10
#      /  \
#     5    15
#    / \
#   3   7

print("Tree: [10, 5, 15, 3, 7]")
print_binary_tree(build_tree_from_list([10, 5, 15, 3, 7]))
print()

# Check answers
verify_traversal([10, 5, 15, 3, 7], "inorder",   [3, 5, 7, 10, 15])
verify_traversal([10, 5, 15, 3, 7], "preorder",  [10, 5, 3, 7, 15])
verify_traversal([10, 5, 15, 3, 7], "postorder", [3, 7, 5, 15, 10])
````

**Figure 6.2** — Expression tree: representing math with a tree

````python
# Expression trees represent mathematical expressions
# Internal nodes are operators, leaves are numbers
#
# Expression: (3 + 4) * (5 - 2)
#
#         *
#        / \
#       +   -
#      / \ / \
#     3  4 5  2

expr = Node('*')
expr.left = Node('+')
expr.right = Node('-')
expr.left.left = Node(3)
expr.left.right = Node(4)
expr.right.left = Node(5)
expr.right.right = Node(2)

print("Expression Tree for (3 + 4) * (5 - 2):")
print_binary_tree(expr)
print()
print(f"In-order (infix):    {inorder(expr)}")
print(f"Pre-order (prefix):  {preorder(expr)}")
print(f"Post-order (postfix):{postorder(expr)}")
print()
print("Notice:")
print("  In-order gives us the expression in familiar notation (infix)")
print("  Pre-order gives prefix notation (used in some calculators)")
print("  Post-order gives postfix notation (Reverse Polish Notation)")
````

**Figure 6.3** — Evaluating an expression tree

````python
def evaluate(node):
    """Evaluate an expression tree recursively."""
    # Base case: leaf node is a number
    if node.left is None and node.right is None:
        return node.value

    # Recursive case: evaluate children and apply operator
    left_val = evaluate(node.left)
    right_val = evaluate(node.right)

    if node.value == '+':
        return left_val + right_val
    elif node.value == '-':
        return left_val - right_val
    elif node.value == '*':
        return left_val * right_val
    elif node.value == '/':
        return left_val / right_val

result = evaluate(expr)
print(f"(3 + 4) * (5 - 2) = {result}")
print(f"Check: {(3 + 4) * (5 - 2)}")
````

## Part 7: Common Errors & Pitfalls

**Figure 7.1** — Forgetting the base case in recursion

````python
# ERROR 1: Missing base case leads to infinite recursion
import sys

def bad_inorder(node):
    """BROKEN: No check for None!"""
    # This will crash with AttributeError when it hits None
    return bad_inorder(node.left) + [node.value] + bad_inorder(node.right)

small_tree = Node(1)
small_tree.left = Node(2)

try:
    bad_inorder(small_tree)
except (RecursionError, AttributeError) as e:
    print(f"Error: {type(e).__name__}: {e}")
    print("\nFix: Always check 'if node is None: return []' first!")
````

**Figure 7.2** — Confusing traversal orders

````python
# ERROR 2: Mixing up the order of operations
# A common mistake: doing root THEN left THEN right for "in-order"

def wrong_inorder(node):
    """WRONG: This is actually pre-order, not in-order!"""
    if node is None:
        return []
    # Root first, then left, then right => This is PRE-order!
    return [node.value] + wrong_inorder(node.left) + wrong_inorder(node.right)

tree = build_tree_from_list([1, 2, 3, 4, 5])

print("Correct in-order:    ", inorder(tree))
print("Wrong 'in-order':    ", wrong_inorder(tree), " <- This is actually pre-order!")
print()
print("Remember:")
print("  In-order:   LEFT -> ROOT -> RIGHT")
print("  Pre-order:  ROOT -> LEFT -> RIGHT")
print("  Post-order: LEFT -> RIGHT -> ROOT")
````

**Figure 7.3** — Mutating vs returning in recursive tree functions

````python
# ERROR 3: Forgetting to return results from recursive calls

def broken_count(node):
    """BROKEN: Doesn't accumulate the count properly."""
    if node is None:
        return 0
    count = 1
    broken_count(node.left)   # Result is thrown away!
    broken_count(node.right)  # Result is thrown away!
    return count  # Always returns 1!

tree = build_tree_from_list([1, 2, 3, 4, 5])
print(f"Broken count: {broken_count(tree)}")
print(f"Correct count: {count_nodes(tree)}")
print()
print("Fix: return 1 + count(node.left) + count(node.right)")
````

## Part 8: Performance Benchmark — Traversal Methods

> **🎯 Predict first, then measure. Does reality match your prediction?**
>
> All three traversals visit every node exactly once. Before running, predict: will they all take the same time? Will any be noticeably faster or slower? Why?

How do different traversal methods compare in speed as the tree grows larger? Let's build complete binary trees of increasing depth and measure the time for in-order, pre-order, and level-order traversal.

````python
import time
import matplotlib.pyplot as plt

def build_complete_tree(depth):
    """Build a complete binary tree of given depth using level-order insertion."""
    if depth < 0:
        return None
    total_nodes = 2 ** (depth + 1) - 1
    return build_tree_from_list(list(range(1, total_nodes + 1)))

depths = [5, 8, 10, 12, 14]
num_runs = 5

inorder_times = []
preorder_times = []
levelorder_times = []

for d in depths:
    tree = build_complete_tree(d)

    # Benchmark in-order traversal
    times = []
    for _ in range(num_runs):
        start = time.time()
        inorder(tree)
        times.append(time.time() - start)
    inorder_times.append(sum(times) / num_runs)

    # Benchmark pre-order traversal
    times = []
    for _ in range(num_runs):
        start = time.time()
        preorder(tree)
        times.append(time.time() - start)
    preorder_times.append(sum(times) / num_runs)

    # Benchmark level-order traversal
    times = []
    for _ in range(num_runs):
        start = time.time()
        level_order(tree)
        times.append(time.time() - start)
    levelorder_times.append(sum(times) / num_runs)

    node_count = 2 ** (d + 1) - 1
    print(f"Depth {d:2d} ({node_count:6d} nodes): "
          f"in-order={inorder_times[-1]:.4f}s  "
          f"pre-order={preorder_times[-1]:.4f}s  "
          f"level-order={levelorder_times[-1]:.4f}s")

plt.figure(figsize=(10, 6))
plt.plot(depths, inorder_times, 'o-', label='In-order', linewidth=2)
plt.plot(depths, preorder_times, 's-', label='Pre-order', linewidth=2)
plt.plot(depths, levelorder_times, '^-', label='Level-order', linewidth=2)
plt.xlabel('Tree Depth')
plt.ylabel('Average Time (seconds)')
plt.title('Binary Tree Traversal Performance vs Tree Depth')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````

**Observations:**
- All three traversal methods visit every node exactly once, so they all have **O(n)** time complexity. The times grow roughly in proportion to the number of nodes (which doubles with each added depth level).
- **In-order** and **pre-order** use recursion and behave very similarly since they both rely on the call stack. At very large depths, you may see slight differences due to Python's recursion overhead.
- **Level-order** uses a queue (iterative approach) and may perform slightly differently due to list operations (`pop(0)`) rather than recursive function calls. Using `collections.deque` would make it faster.

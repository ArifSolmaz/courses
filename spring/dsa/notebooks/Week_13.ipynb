## Part 1: What is Dynamic Programming?

**Dynamic Programming (DP)** is a technique for solving problems by breaking them into **smaller overlapping subproblems**, solving each subproblem **only once**, and **storing** the result for future use.

### Real-World Analogy: Writing an Exam 📝

Imagine you are studying for an exam with 50 questions. Many questions share the same formulas.

| Approach | Strategy | Effort |
|----------|----------|--------|
| **Naive (no DP)** | Re-derive every formula from scratch each time | Very High |
| **Memoization (top-down)** | Write formulas on a cheat sheet as you need them | Medium |
| **Tabulation (bottom-up)** | Pre-compute a formula table before the exam | Low |

### Two Key Properties

A problem is a good fit for DP if it has:

1. **Overlapping Subproblems** — The same smaller problems appear again and again
2. **Optimal Substructure** — The best solution to the big problem can be built from the best solutions to the smaller problems

**Figure 1.1** — Fibonacci recursion tree showing overlapping subproblems

````python
# Fibonacci: the classic example of overlapping subproblems
# Let's count how many times each fib(n) is called

call_count = {}

def fib_naive(n):
    """Naive recursive Fibonacci — very slow!"""
    call_count[n] = call_count.get(n, 0) + 1
    if n <= 1:
        return n
    return fib_naive(n - 1) + fib_naive(n - 2)

call_count.clear()
result = fib_naive(10)
print(f"fib(10) = {result}")
print(f"\nTotal function calls: {sum(call_count.values())}")
print("\nCall counts per value:")
for k in sorted(call_count.keys()):
    bar = "█" * call_count[k]
    print(f"  fib({k:2d}): {call_count[k]:3d} times  {bar}")
````

Notice how `fib(2)` is called **34 times** and `fib(3)` is called **21 times**! This is a huge waste. DP eliminates this waste.

> 📌 **Recall Week 4:** You first encountered memoization in **Week 4** when we cached Fibonacci results using `@lru_cache`. Dynamic Programming formalizes this idea — identifying overlapping subproblems and storing their solutions to avoid redundant computation.

## Part 2: Top-Down (Memoization) vs Bottom-Up (Tabulation)

There are two main approaches to implementing DP:

| Feature | Top-Down (Memoization) | Bottom-Up (Tabulation) |
|---------|----------------------|------------------------|
| Direction | Start from the big problem, break down | Start from smallest, build up |
| Implementation | Recursion + cache (dict) | Loop + table (list) |
| Subproblems solved | Only the ones needed | All of them |
| Stack overflow risk | Yes (deep recursion) | No |
| Easier to write? | Often yes (natural recursion) | Sometimes more efficient |

**Figure 2.1** — Fibonacci with memoization (top-down)

````python
def fib_memo(n, memo=None):
    """Top-down: recursion + memoization."""
    if memo is None:
        memo = {}
    
    # Base cases
    if n <= 1:
        return n
    
    # Check if we already computed this
    if n in memo:
        return memo[n]
    
    # Compute and STORE the result
    memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)
    return memo[n]

print(f"fib(10) = {fib_memo(10)}")
print(f"fib(30) = {fib_memo(30)}")
print(f"fib(50) = {fib_memo(50)}")
print("\n✅ All computed instantly thanks to memoization!")
````

**Figure 2.2** — Fibonacci with tabulation (bottom-up)

````python
def fib_tab(n):
    """Bottom-up: build a table from smallest to largest."""
    if n <= 1:
        return n
    
    # Create a table
    table = [0] * (n + 1)
    table[0] = 0
    table[1] = 1
    
    # Fill the table from bottom up
    for i in range(2, n + 1):
        table[i] = table[i - 1] + table[i - 2]
    
    return table[n]

# Show the table being built
n = 10
table = [0] * (n + 1)
table[1] = 1
print("Building the DP table step by step:")
print(f"  table[0] = 0  (base case)")
print(f"  table[1] = 1  (base case)")
for i in range(2, n + 1):
    table[i] = table[i-1] + table[i-2]
    print(f"  table[{i}] = table[{i-1}] + table[{i-2}] = {table[i-1]} + {table[i-2]} = {table[i]}")
print(f"\nAnswer: fib({n}) = {table[n]}")
````

## Part 3: Classic Problem 1 — Climbing Stairs 🪨

### Problem Statement

You are climbing a staircase with **n** steps. Each time you can climb **1 step** or **2 steps**. How many **distinct ways** can you reach the top?

### Real-World Analogy

Think of walking up stairs. At each step, you decide: take one step or skip one (take two). The question is: how many different paths lead to the top?

### Working Through Small Examples

| n (steps) | Ways | Explanation |
|-----------|------|-------------|
| 1 | 1 | [1] |
| 2 | 2 | [1,1] or [2] |
| 3 | 3 | [1,1,1], [1,2], [2,1] |
| 4 | 5 | [1,1,1,1], [1,1,2], [1,2,1], [2,1,1], [2,2] |

### The Key Insight

To reach step `n`, you either:
- Came from step `n-1` (took 1 step), **or**
- Came from step `n-2` (took 2 steps)

So: `ways(n) = ways(n-1) + ways(n-2)`

This is exactly the Fibonacci sequence!

**Figure 3.1** — Climbing Stairs: Naive recursive solution

````python
def climb_naive(n):
    """Naive recursion: O(2^n) time."""
    if n <= 2:
        return n
    return climb_naive(n - 1) + climb_naive(n - 2)

# Test with small values
for i in range(1, 11):
    print(f"  stairs({i:2d}) = {climb_naive(i)} ways")
````

**Figure 3.2** — Climbing Stairs: Memoization (top-down)

````python
def climb_memo(n, memo=None):
    """Top-down with memoization: O(n) time, O(n) space."""
    if memo is None:
        memo = {}
    
    if n <= 2:
        return n
    
    if n in memo:
        return memo[n]
    
    memo[n] = climb_memo(n - 1, memo) + climb_memo(n - 2, memo)
    return memo[n]

# Now we can handle large values!
print(f"climb_memo(50)  = {climb_memo(50)}")
print(f"climb_memo(100) = {climb_memo(100)}")
````

**Figure 3.3** — Climbing Stairs: Tabulation (bottom-up)

````python
def climb_tab(n):
    """Bottom-up tabulation: O(n) time, O(n) space."""
    if n <= 2:
        return n
    
    # Build the table
    dp = [0] * (n + 1)
    dp[1] = 1
    dp[2] = 2
    
    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]
    
    return dp[n]

# Step-by-step table for n = 7
n = 7
dp = [0] * (n + 1)
dp[1] = 1
dp[2] = 2

print(f"Climbing Stairs DP Table (n = {n}):")
print(f"{'Step':<6} {'dp[i-1]':<10} {'dp[i-2]':<10} {'dp[i]':<10}")
print("-" * 36)
print(f"{'1':<6} {'-':<10} {'-':<10} {1:<10}  (base)")
print(f"{'2':<6} {'-':<10} {'-':<10} {2:<10}  (base)")

for i in range(3, n + 1):
    dp[i] = dp[i-1] + dp[i-2]
    print(f"{i:<6} {dp[i-1]:<10} {dp[i-2]:<10} {dp[i]:<10}")

print(f"\n✅ Answer: {dp[n]} distinct ways to climb {n} stairs")
````

**Figure 3.4** — Climbing Stairs: Space-optimized solution

````python
def climb_optimized(n):
    """Space-optimized: O(n) time, O(1) space.
    We only need the last two values!"""
    if n <= 2:
        return n
    
    prev2 = 1  # dp[i-2]
    prev1 = 2  # dp[i-1]
    
    for i in range(3, n + 1):
        current = prev1 + prev2
        prev2 = prev1
        prev1 = current
    
    return prev1

# Verify all methods give same results
print("Verifying all three methods agree:")
print(f"{'n':<5} {'Naive':<10} {'Memo':<10} {'Tabulation':<12} {'Optimized':<10}")
print("-" * 47)
for n in range(1, 16):
    naive = climb_naive(n) if n <= 25 else "-"
    memo = climb_memo(n)
    tab = climb_tab(n)
    opt = climb_optimized(n)
    print(f"{n:<5} {str(naive):<10} {memo:<10} {tab:<12} {opt:<10}")
````

## Part 4: Classic Problem 2 — Coin Change 🪙

### Problem Statement

Given a set of coin denominations and a target amount, find the **minimum number of coins** needed to make that amount. If it's impossible, return -1.

### Real-World Analogy

You're at a vending machine. You have unlimited coins of certain values (e.g., 1¢, 5¢, 10¢, 25¢). You want to pay exactly the right amount using as few coins as possible.

### Example

- Coins: [1, 5, 10, 25]
- Amount: 36
- Answer: 3 coins (25 + 10 + 1)

### The Key Insight

For each amount `a`, the minimum coins needed is:

```
dp[a] = min(dp[a - coin] + 1)  for each coin in coins, if a - coin >= 0
```

Translation: try removing each coin, see which remaining amount needs the fewest coins, then add 1.

**Figure 4.1** — Coin Change: Naive recursive solution

````python
def coin_change_naive(coins, amount):
    """Naive recursion: exponential time!"""
    if amount == 0:
        return 0
    if amount < 0:
        return float('inf')
    
    min_coins = float('inf')
    for coin in coins:
        result = coin_change_naive(coins, amount - coin)
        if result != float('inf'):
            min_coins = min(min_coins, result + 1)
    
    return min_coins

# Test with small amounts (large amounts would be too slow!)
coins = [1, 5, 10, 25]
for amount in [1, 5, 11, 15, 20, 30]:
    result = coin_change_naive(coins, amount)
    print(f"  Amount {amount:3d}: {result} coin(s)")
````

**Figure 4.2** — Coin Change: Memoization (top-down)

````python
def coin_change_memo(coins, amount, memo=None):
    """Top-down with memoization: O(amount * len(coins))."""
    if memo is None:
        memo = {}
    
    if amount == 0:
        return 0
    if amount < 0:
        return float('inf')
    if amount in memo:
        return memo[amount]
    
    min_coins = float('inf')
    for coin in coins:
        result = coin_change_memo(coins, amount - coin, memo)
        if result != float('inf'):
            min_coins = min(min_coins, result + 1)
    
    memo[amount] = min_coins
    return min_coins

# Now we can handle bigger amounts
coins = [1, 5, 10, 25]
for amount in [36, 49, 63, 99, 100, 250]:
    result = coin_change_memo(coins, amount)
    print(f"  Amount {amount:4d}: {result} coin(s)")
````

**Figure 4.3** — Coin Change: Tabulation (bottom-up) with step-by-step trace

````python
def coin_change_tab(coins, amount):
    """Bottom-up tabulation: O(amount * len(coins))."""
    # Initialize table: dp[i] = min coins needed for amount i
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0  # base case: 0 coins needed for amount 0
    
    for a in range(1, amount + 1):
        for coin in coins:
            if a - coin >= 0 and dp[a - coin] != float('inf'):
                dp[a] = min(dp[a], dp[a - coin] + 1)
    
    return dp[amount] if dp[amount] != float('inf') else -1

# Step-by-step trace for small example
coins = [1, 3, 4]
amount = 6
dp = [float('inf')] * (amount + 1)
dp[0] = 0

print(f"Coins: {coins}, Target Amount: {amount}")
print(f"\n{'Amount':<8} {'Try coin 1':<14} {'Try coin 3':<14} {'Try coin 4':<14} {'Result':<8}")
print("-" * 58)

for a in range(1, amount + 1):
    options = []
    for coin in coins:
        if a - coin >= 0 and dp[a - coin] != float('inf'):
            val = dp[a - coin] + 1
            options.append(f"dp[{a-coin}]+1={val}")
            dp[a] = min(dp[a], val)
        else:
            options.append("-")
    print(f"{a:<8} {options[0]:<14} {options[1]:<14} {options[2]:<14} {dp[a]:<8}")

print(f"\n✅ Minimum coins for amount {amount}: {dp[amount]}")
````

**Figure 4.4** — Coin Change: Tracking which coins were used

````python
def coin_change_with_solution(coins, amount):
    """Bottom-up tabulation that also tracks which coins to use."""
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    # Track which coin was used at each step
    coin_used = [-1] * (amount + 1)
    
    for a in range(1, amount + 1):
        for coin in coins:
            if a - coin >= 0 and dp[a - coin] + 1 < dp[a]:
                dp[a] = dp[a - coin] + 1
                coin_used[a] = coin
    
    if dp[amount] == float('inf'):
        return -1, []
    
    # Reconstruct the solution
    solution = []
    remaining = amount
    while remaining > 0:
        solution.append(coin_used[remaining])
        remaining -= coin_used[remaining]
    
    return dp[amount], solution

# Test with different coin sets
test_cases = [
    ([1, 5, 10, 25], 63),
    ([1, 3, 4], 6),
    ([2, 5], 3),      # impossible!
    ([1, 5, 10], 28),
]

for coins, amount in test_cases:
    count, solution = coin_change_with_solution(coins, amount)
    if count == -1:
        print(f"  Coins {str(coins):<16} Amount {amount:3d}: IMPOSSIBLE")
    else:
        print(f"  Coins {str(coins):<16} Amount {amount:3d}: {count} coins → {solution}")
````

## Part 5: Common Errors & Pitfalls

### Mistakes beginners make with DP

**Figure 5.1** — Common Error: Forgetting base cases

````python
# ERROR 1: Forgetting the base case
print("=" * 50)
print("ERROR 1: Missing base case")
print("=" * 50)

def climb_broken(n):
    """This will crash! No base case."""
    return climb_broken(n - 1) + climb_broken(n - 2)

try:
    climb_broken(5)
except RecursionError as e:
    print(f"\n❌ RecursionError: {str(e)[:60]}...")
    print("\n💡 Fix: Always define base cases first!")
    print("   if n <= 2: return n")
````

**Figure 5.2** — Common Error: Mutable default argument in memoization

````python
# ERROR 2: Using mutable default argument (memo={}) 
print("=" * 50)
print("ERROR 2: Mutable default argument")
print("=" * 50)

# BAD: memo dictionary is shared across all calls!
def fib_bad(n, memo={}):
    if n <= 1: return n
    if n in memo: return memo[n]
    memo[n] = fib_bad(n-1, memo) + fib_bad(n-2, memo)
    return memo[n]

print(f"First call:  fib_bad(5) = {fib_bad(5)}")
# The memo dict persists between calls - this can cause subtle bugs
print(f"memo is retained: {fib_bad.__defaults__}")

print("\n💡 Fix: Use None as default, create new dict inside:")
print("   def fib(n, memo=None):")
print("       if memo is None: memo = {}")
````

**Figure 5.3** — Common Error: Off-by-one in table size

````python
# ERROR 3: Off-by-one error in table size
print("=" * 50)
print("ERROR 3: Off-by-one in table size")
print("=" * 50)

def climb_offbyone(n):
    dp = [0] * n  # BUG: should be n+1
    dp[1] = 1
    dp[2] = 2
    for i in range(3, n + 1):  # This will crash at dp[n]
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

try:
    climb_offbyone(5)
except IndexError as e:
    print(f"\n❌ IndexError: {e}")
    print("\n💡 Fix: Table size should be (n + 1) to include index n")
    print("   dp = [0] * (n + 1)  # indices 0 through n")
````

## Part 6: Benchmark — Naive vs Memoization vs Tabulation

> **🎯 Predict first, then measure. Does reality match your prediction?**
>
> Naive recursion is O(2^n), memoization and tabulation are O(n). Before running, predict: at what value of n does naive recursion become noticeably slow (>1 second)? How fast will memoization be at n=1000?

Let's measure the actual performance differences between our three approaches.

**Figure 6.1** — Benchmark: Climbing Stairs performance comparison

````python
import time

def benchmark_climb():
    """Benchmark the three climbing stairs approaches."""
    # Naive: only small values (exponential!)
    naive_sizes = list(range(5, 36, 5))
    naive_times = []
    for n in naive_sizes:
        start = time.perf_counter()
        climb_naive(n)
        elapsed = time.perf_counter() - start
        naive_times.append(elapsed)
    
    # Memo and Tab: can handle bigger values
    dp_sizes = list(range(100, 1100, 100))
    memo_times = []
    tab_times = []
    for n in dp_sizes:
        # Memoization
        import sys
        old_limit = sys.getrecursionlimit()
        sys.setrecursionlimit(max(n + 100, old_limit))
        start = time.perf_counter()
        climb_memo(n)
        memo_times.append(time.perf_counter() - start)
        sys.setrecursionlimit(old_limit)
        
        # Tabulation
        start = time.perf_counter()
        climb_tab(n)
        tab_times.append(time.perf_counter() - start)
    
    return naive_sizes, naive_times, dp_sizes, memo_times, tab_times

naive_sizes, naive_times, dp_sizes, memo_times, tab_times = benchmark_climb()

print("Naive Recursion (small n only):")
print(f"{'n':<8} {'Time (s)':<12}")
for s, t in zip(naive_sizes, naive_times):
    print(f"{s:<8} {t:<12.6f}")

print(f"\nMemoization vs Tabulation (larger n):")
print(f"{'n':<8} {'Memo (s)':<14} {'Tab (s)':<14}")
for s, m, t in zip(dp_sizes, memo_times, tab_times):
    print(f"{s:<8} {m:<14.6f} {t:<14.6f}")
````

**Figure 6.2** — Visualization: Naive recursion time growth

````python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Naive recursion (exponential growth)
axes[0].plot(naive_sizes, naive_times, 'ro-', linewidth=2, markersize=8, label='Naive Recursion')
axes[0].set_xlabel('n (number of stairs)', fontsize=12)
axes[0].set_ylabel('Time (seconds)', fontsize=12)
axes[0].set_title('Naive Recursion: Exponential Growth', fontsize=13)
axes[0].legend(fontsize=11)
axes[0].grid(True, alpha=0.3)

# Plot 2: Memo vs Tab
axes[1].plot(dp_sizes, memo_times, 'bs-', linewidth=2, markersize=6, label='Memoization')
axes[1].plot(dp_sizes, tab_times, 'g^-', linewidth=2, markersize=6, label='Tabulation')
axes[1].set_xlabel('n (number of stairs)', fontsize=12)
axes[1].set_ylabel('Time (seconds)', fontsize=12)
axes[1].set_title('Memoization vs Tabulation: Both Linear', fontsize=13)
axes[1].legend(fontsize=11)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n📊 Observations:")
print("  1. Naive recursion time grows EXPONENTIALLY — unusable for n > 35")
print("  2. Memoization and tabulation are both LINEAR — handle n = 1000+ easily")
print("  3. Tabulation is often slightly faster (no recursion overhead)")
````

**Figure 6.3** — Benchmark: Coin Change performance comparison

````python
def benchmark_coins():
    """Benchmark coin change approaches."""
    coins = [1, 5, 10, 25]
    
    # Naive: only small amounts
    naive_amounts = [5, 10, 15, 20, 25]
    naive_times = []
    for amount in naive_amounts:
        start = time.perf_counter()
        coin_change_naive(coins, amount)
        naive_times.append(time.perf_counter() - start)
    
    # DP: larger amounts
    dp_amounts = list(range(500, 5500, 500))
    memo_times = []
    tab_times = []
    for amount in dp_amounts:
        import sys
        old_limit = sys.getrecursionlimit()
        sys.setrecursionlimit(max(amount + 200, old_limit))
        
        start = time.perf_counter()
        coin_change_memo(coins, amount)
        memo_times.append(time.perf_counter() - start)
        sys.setrecursionlimit(old_limit)
        
        start = time.perf_counter()
        coin_change_tab(coins, amount)
        tab_times.append(time.perf_counter() - start)
    
    return naive_amounts, naive_times, dp_amounts, memo_times, tab_times

c_naive_a, c_naive_t, c_dp_a, c_memo_t, c_tab_t = benchmark_coins()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(c_naive_a, c_naive_t, 'ro-', linewidth=2, markersize=8)
axes[0].set_xlabel('Amount', fontsize=12)
axes[0].set_ylabel('Time (seconds)', fontsize=12)
axes[0].set_title('Coin Change: Naive Recursion', fontsize=13)
axes[0].grid(True, alpha=0.3)

axes[1].plot(c_dp_a, c_memo_t, 'bs-', linewidth=2, markersize=6, label='Memoization')
axes[1].plot(c_dp_a, c_tab_t, 'g^-', linewidth=2, markersize=6, label='Tabulation')
axes[1].set_xlabel('Amount', fontsize=12)
axes[1].set_ylabel('Time (seconds)', fontsize=12)
axes[1].set_title('Coin Change: Memo vs Tabulation', fontsize=13)
axes[1].legend(fontsize=11)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## Part 7: Summary

### DP Approach Comparison

| Approach | Time | Space | Pros | Cons |
|----------|------|-------|------|------|
| Naive Recursion | O(2^n) | O(n) stack | Simple to write | Extremely slow |
| Memoization | O(n) | O(n) memo + stack | Natural recursive thinking | Stack overflow risk |
| Tabulation | O(n) | O(n) table | No recursion overhead | Must think bottom-up |
| Space-Optimized | O(n) | O(1) | Most efficient | Harder to understand |

### Steps to Solve Any DP Problem

1. **Identify** if the problem has overlapping subproblems + optimal substructure
2. **Define** the state: what does dp[i] represent?
3. **Write** the recurrence relation (transition formula)
4. **Set** the base cases
5. **Choose** top-down or bottom-up
6. **Optimize** space if needed

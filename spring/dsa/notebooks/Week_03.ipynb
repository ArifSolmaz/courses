## Part 1: What Is an Abstract Data Type (ADT)?

An **Abstract Data Type** defines:
- What **data** it stores
- What **operations** you can do with it
- But NOT **how** it is implemented internally

Think of it like a vending machine:
- You know you can insert coins and press a button to get a drink
- You do NOT need to know how the machine works internally

| ADT | Rule | Real-World Analogy |
|---|---|---|
| Stack | Last In, First Out (LIFO) | A stack of plates |
| Queue | First In, First Out (FIFO) | A line at a cafeteria |
| Deque | Insert/remove from both ends | A deck of cards (top and bottom) |

The ADT defines the **interface** (what you can do). The **implementation** (how it works) can vary.

## Part 2: The Stack ADT (LIFO)

A **Stack** follows the **Last In, First Out** principle:
- The last item you put in is the first item you take out
- Like a stack of plates: you always add and remove from the **top**

### Stack Operations

| Operation | Description | Big-O |
|---|---|---|
| `push(item)` | Add item to the top | O(1) |
| `pop()` | Remove and return the top item | O(1) |
| `peek()` / `top()` | Look at the top item without removing | O(1) |
| `is_empty()` | Check if the stack is empty | O(1) |
| `size()` | Return the number of items | O(1) |

### Real-World Examples of Stacks

- **Browser back button**: Each page you visit is pushed onto a stack. Clicking "Back" pops the most recent page.
- **Undo in a text editor**: Each action is pushed. Ctrl+Z pops the last action.
- **Call stack in programming**: Each function call is pushed. When a function returns, it is popped.
- **Stacking plates in a cafeteria**: You always take from the top.

**Figure 2.1** -- Implementing a Stack using a Python list

````python
class Stack:
    """Stack implementation using a Python list.
    
    The end of the list is the 'top' of the stack.
    This is efficient because list.append() and list.pop() are both O(1).
    """
    
    def __init__(self):
        self._items = []      # Internal storage
    
    def push(self, item):
        """Add item to the top of the stack. O(1)"""
        self._items.append(item)
    
    def pop(self):
        """Remove and return the top item. O(1)"""
        if self.is_empty():
            raise IndexError("Pop from empty stack")
        return self._items.pop()
    
    def peek(self):
        """Return the top item without removing it. O(1)"""
        if self.is_empty():
            raise IndexError("Peek at empty stack")
        return self._items[-1]
    
    def is_empty(self):
        """Check if the stack is empty. O(1)"""
        return len(self._items) == 0
    
    def size(self):
        """Return the number of items. O(1)"""
        return len(self._items)
    
    def __repr__(self):
        return f"Stack({self._items})  <-- top is right"


# Demo
s = Stack()
print("Empty stack:", s)
print("Is empty?", s.is_empty())

s.push("A")
s.push("B")
s.push("C")
print("\nAfter pushing A, B, C:", s)
print("Top item (peek):", s.peek())
print("Size:", s.size())

print("\nPopping:", s.pop())  # C (last in, first out)
print("Popping:", s.pop())  # B
print("Stack now:", s)
print("Popping:", s.pop())  # A
print("Is empty?", s.is_empty())
````

**Figure 2.2** -- Visualizing stack operations step by step

````python
# Let's trace through stack operations visually
s = Stack()

operations = [
    ("push", 10),
    ("push", 20),
    ("push", 30),
    ("pop", None),
    ("push", 40),
    ("pop", None),
    ("pop", None),
]

print(f"{'Operation':<15} {'Result':<10} {'Stack State'}")
print("-" * 50)

for op, val in operations:
    if op == "push":
        s.push(val)
        print(f"push({val}){'':<9} {'--':<10} {s._items}")
    elif op == "pop":
        result = s.pop()
        print(f"pop(){'':<10} {result!s:<10} {s._items}")
````

**Figure 2.3** -- Practical example: checking balanced parentheses

````python
def is_balanced(expression):
    """Check if parentheses/brackets are balanced using a stack.
    
    This is a classic stack problem!
    - When we see an opening bracket, push it
    - When we see a closing bracket, pop and check if it matches
    """
    stack = Stack()
    matching = {')': '(', ']': '[', '}': '{'}
    
    for char in expression:
        if char in '([{':
            stack.push(char)
        elif char in ')]}':
            if stack.is_empty():
                return False
            if stack.pop() != matching[char]:
                return False
    
    return stack.is_empty()

# Test cases
test_cases = [
    "(a + b) * c",
    "((a + b) * (c - d))",
    "[1, [2, 3], [4]]",
    "{a: (b + c)}",
    "(a + b))",           # Extra closing
    "((a + b)",           # Missing closing
    "([a + b)]",          # Mismatched
]

for expr in test_cases:
    result = is_balanced(expr)
    status = "Balanced" if result else "NOT balanced"
    print(f"  {expr:<25} -> {status}")
````

**Figure 2.4** -- Practical example: reversing a string with a stack

````python
def reverse_string(text):
    """Reverse a string using a stack.
    
    Push all characters, then pop them all.
    Since a stack is LIFO, characters come out in reverse order.
    """
    stack = Stack()
    
    # Push all characters
    for char in text:
        stack.push(char)
    
    # Pop all characters
    reversed_text = ""
    while not stack.is_empty():
        reversed_text += stack.pop()
    
    return reversed_text

print(reverse_string("hello"))     # olleh
print(reverse_string("Python"))    # nohtyP
print(reverse_string("12345"))     # 54321
````

## Part 3: The Queue ADT (FIFO)

A **Queue** follows the **First In, First Out** principle:
- The first item you put in is the first item you take out
- Like a line at a cafeteria: the person who arrives first gets served first

### Queue Operations

| Operation | Description | Target Big-O |
|---|---|---|
| `enqueue(item)` | Add item to the back | O(1) |
| `dequeue()` | Remove and return the front item | O(1) |
| `peek()` / `front()` | Look at the front item | O(1) |
| `is_empty()` | Check if the queue is empty | O(1) |
| `size()` | Return the number of items | O(1) |

### Real-World Examples of Queues

- **Cafeteria line**: First person in line gets served first
- **Print queue**: Documents print in the order they were sent
- **Customer service**: Calls are answered in order received
- **Task scheduling**: Operating system runs tasks in order

**Figure 3.1** -- First attempt: Queue using a Python list (naive)

````python
class QueueNaive:
    """Queue using a Python list.
    
    Problem: dequeue uses list.pop(0) which is O(n)!
    All remaining elements must shift left.
    """
    
    def __init__(self):
        self._items = []
    
    def enqueue(self, item):
        """Add to the back. O(1)"""
        self._items.append(item)
    
    def dequeue(self):
        """Remove from the front. O(n) -- THIS IS THE PROBLEM!"""
        if self.is_empty():
            raise IndexError("Dequeue from empty queue")
        return self._items.pop(0)  # <-- O(n) because all elements shift
    
    def peek(self):
        """Look at the front item. O(1)"""
        if self.is_empty():
            raise IndexError("Peek at empty queue")
        return self._items[0]
    
    def is_empty(self):
        return len(self._items) == 0
    
    def size(self):
        return len(self._items)
    
    def __repr__(self):
        return f"Queue({self._items})  <-- front is left"


# Demo
q = QueueNaive()
q.enqueue("Alice")
q.enqueue("Bob")
q.enqueue("Charlie")
print("Queue:", q)
print("Front (peek):", q.peek())

print("\nServing:", q.dequeue())  # Alice (first in, first out)
print("Serving:", q.dequeue())  # Bob
print("Queue now:", q)
````

### The Problem with `list.pop(0)`

When you remove the first element of a list, every other element must shift one position to the left:

```
Before pop(0):
  Index:   0       1       2       3
  Value: ["A"]   ["B"]   ["C"]   ["D"]

After pop(0):  (returns "A")
  Index:   0       1       2
  Value: ["B"]   ["C"]   ["D"]
           ^       ^       ^   
         shifted shifted shifted
```

With `n` elements, this is **O(n)** per dequeue. If you dequeue all `n` elements, the total is **O(n^2)**.

We need a better solution!

---
## Part 4: The Deque ADT and `collections.deque`

A **Deque** (pronounced "deck") is a **double-ended queue**: you can add and remove from **both** ends efficiently.

Python provides `collections.deque`, which is implemented as a **doubly-linked list**. This means:

| Operation | list | collections.deque |
|---|---|---|
| Append right | O(1) | O(1) |
| Pop right | O(1) | O(1) |
| Append left | O(n) | **O(1)** |
| Pop left | O(n) | **O(1)** |

The key advantage: operations at **both ends** are O(1)!

**Figure 4.1** -- Using `collections.deque`

````python
from collections import deque

# Create a deque
d = deque()

# Add to right (like append)
d.append("B")
d.append("C")
print("After append B, C:", d)

# Add to left (O(1) -- NOT possible efficiently with list!)
d.appendleft("A")
print("After appendleft A:", d)

# Remove from right
print("Pop right:", d.pop())
print("After pop:", d)

# Remove from left (O(1) -- this is the key advantage!)
print("Pop left:", d.popleft())
print("After popleft:", d)
````

**Figure 4.2** -- Efficient Queue using deque

````python
from collections import deque

class Queue:
    """Efficient Queue using collections.deque.
    
    Both enqueue and dequeue are O(1).
    """
    
    def __init__(self):
        self._items = deque()
    
    def enqueue(self, item):
        """Add to the back. O(1)"""
        self._items.append(item)
    
    def dequeue(self):
        """Remove from the front. O(1) -- MUCH better!"""
        if self.is_empty():
            raise IndexError("Dequeue from empty queue")
        return self._items.popleft()  # <-- O(1)!
    
    def peek(self):
        """Look at the front item. O(1)"""
        if self.is_empty():
            raise IndexError("Peek at empty queue")
        return self._items[0]
    
    def is_empty(self):
        return len(self._items) == 0
    
    def size(self):
        return len(self._items)
    
    def __repr__(self):
        return f"Queue({list(self._items)})  <-- front is left"


# Demo
q = Queue()
for person in ["Alice", "Bob", "Charlie", "Diana"]:
    q.enqueue(person)
    print(f"  {person} joins -> {q}")

print()
while not q.is_empty():
    person = q.dequeue()
    print(f"  Serving {person} -> {q}")
````

**Figure 4.3** -- Deque as a double-ended structure

````python
from collections import deque

# A deque lets you work from both ends
d = deque([1, 2, 3, 4, 5])
print("Initial:", d)

# All O(1) operations:
d.append(6)          # Add to right
print("append(6):", d)

d.appendleft(0)      # Add to left
print("appendleft(0):", d)

d.pop()              # Remove from right
print("pop():", d)

d.popleft()          # Remove from left
print("popleft():", d)

# Deque also supports rotation
d.rotate(2)          # Rotate right by 2
print("rotate(2):", d)

d.rotate(-2)         # Rotate left by 2
print("rotate(-2):", d)
````

## Part 5: Summary Comparison

| ADT | Principle | Python Implementation | Key Operations |
|---|---|---|---|
| Stack | LIFO | `list` (use `append`/`pop`) | push, pop, peek |
| Queue | FIFO | `collections.deque` (use `append`/`popleft`) | enqueue, dequeue, peek |
| Deque | Both ends | `collections.deque` | append, appendleft, pop, popleft |

### When to Use Each

| Situation | Use |
|---|---|
| Undo/Redo functionality | Stack |
| Parsing brackets/expressions | Stack |
| Task scheduling, print queues | Queue |
| BFS (Breadth-First Search) | Queue |
| Sliding window problems | Deque |
| Need both ends to be fast | Deque |

## Part 6: Benchmark -- list `pop(0)` vs deque `popleft()`

Now let's **prove** that `deque.popleft()` is faster than `list.pop(0)` by benchmarking at different sizes.

> **🔮 Predict first, then measure. Does reality match your prediction?** Before running: what Big-O do you expect for each? How should the ratio change as n grows?

**Figure 6.1** -- Quick comparison at one size

````python
import time
from collections import deque

n = 100_000

# Benchmark list.pop(0)
data_list = list(range(n))
start = time.time()
while data_list:
    data_list.pop(0)
time_list = time.time() - start

# Benchmark deque.popleft()
data_deque = deque(range(n))
start = time.time()
while data_deque:
    data_deque.popleft()
time_deque = time.time() - start

print(f"Removing {n:,} items from the front:")
print(f"  list.pop(0):     {time_list:.4f} sec")
print(f"  deque.popleft(): {time_deque:.4f} sec")
print(f"\n  Deque is {time_list/max(time_deque, 1e-10):.0f}x faster!")
````

**Figure 6.2** -- Full benchmark at multiple sizes

````python
import time
from collections import deque

sizes = [1_000, 5_000, 10_000, 50_000, 100_000, 200_000]
list_times = []
deque_times = []

print(f"{'Size':>10} | {'list.pop(0) (ms)':>18} | {'deque.popleft (ms)':>20} | {'Speedup':>8}")
print("-" * 65)

for n in sizes:
    # list.pop(0)
    data = list(range(n))
    start = time.time()
    while data:
        data.pop(0)
    t_list = time.time() - start
    list_times.append(t_list)
    
    # deque.popleft()
    data = deque(range(n))
    start = time.time()
    while data:
        data.popleft()
    t_deque = time.time() - start
    deque_times.append(t_deque)
    
    speedup = t_list / max(t_deque, 1e-10)
    print(f"{n:>10,} | {t_list*1000:>17.2f} | {t_deque*1000:>19.2f} | {speedup:>7.0f}x")
````

**Figure 6.3** -- Plotting the benchmark results

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(sizes, [t*1000 for t in list_times], 'rs-', 
         label='list.pop(0) - O(n) per op', linewidth=2, markersize=8)
plt.plot(sizes, [t*1000 for t in deque_times], 'go-', 
         label='deque.popleft() - O(1) per op', linewidth=2, markersize=8)
plt.xlabel('Number of Items (n)', fontsize=12)
plt.ylabel('Total Time to Pop All (ms)', fontsize=12)
plt.title('list.pop(0) vs deque.popleft(): Removing All Items', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Observation: list.pop(0) shows quadratic growth (O(n^2) total).")
print("deque.popleft() shows linear growth (O(n) total).")
print("\nConclusion: Always use deque for queue operations!")
````

**Figure 6.4** -- Benchmark: enqueue + dequeue cycle comparison

````python
import time
from collections import deque

# Simulate a real queue scenario:
# Items arrive and are processed continuously

n = 50_000

# Using list (naive queue)
start = time.time()
q_list = []
for i in range(n):
    q_list.append(i)       # enqueue
    if len(q_list) > 10:   # keep queue small
        q_list.pop(0)      # dequeue -- O(n)!
time_list = time.time() - start

# Using deque (proper queue)
start = time.time()
q_deque = deque()
for i in range(n):
    q_deque.append(i)      # enqueue
    if len(q_deque) > 10:  # keep queue small
        q_deque.popleft()  # dequeue -- O(1)!
time_deque = time.time() - start

print(f"Queue simulation ({n:,} operations):")
print(f"  List queue:  {time_list:.4f} sec")
print(f"  Deque queue: {time_deque:.4f} sec")
print(f"  Deque is {time_list/max(time_deque, 1e-10):.0f}x faster!")
````

## Part 7: Common Errors & Pitfalls

**Figure 7.1** -- Mistake: Popping from an empty stack/queue

````python
# Always check if empty before popping!
s = Stack()

# BAD: Pop without checking
try:
    s.pop()
except IndexError as e:
    print(f"Error: {e}")

# GOOD: Check first
if not s.is_empty():
    s.pop()
else:
    print("Stack is empty -- nothing to pop")

# Also works with deque
from collections import deque
d = deque()

try:
    d.popleft()
except IndexError as e:
    print(f"Deque error: {e}")
````

**Figure 7.2** -- Mistake: Using list.pop(0) for queues

````python
from collections import deque

# BAD: Using a list as a queue
queue_bad = []
queue_bad.append("task1")
queue_bad.append("task2")
queue_bad.pop(0)  # O(n) -- slow!
print("Bad approach works but is slow for large queues")

# GOOD: Using deque as a queue
queue_good = deque()
queue_good.append("task1")
queue_good.append("task2")
queue_good.popleft()  # O(1) -- fast!
print("Good approach is efficient at any size")

print("\nRule: If you need pop(0), use collections.deque instead!")
````

**Figure 7.3** -- Mistake: Confusing Stack and Queue behavior

````python
# Stack (LIFO) vs Queue (FIFO)
# Adding items 1, 2, 3 and removing them:

# Stack behavior
stack = []
for i in [1, 2, 3]:
    stack.append(i)
print("Stack removal order:", end=" ")
while stack:
    print(stack.pop(), end=" ")    # 3, 2, 1
print("(Last in, first out)")

# Queue behavior
queue = deque()
for i in [1, 2, 3]:
    queue.append(i)
print("Queue removal order:", end=" ")
while queue:
    print(queue.popleft(), end=" ")  # 1, 2, 3
print("(First in, first out)")

print("\nRemember: Stack = LIFO, Queue = FIFO")
````

**Figure 7.4** -- Mistake: Index access on deque is O(n)

````python
from collections import deque
import time

# Deque is great for ends, but random access is O(n)!
n = 1_000_000
d = deque(range(n))
lst = list(range(n))

# Random access by index
start = time.time()
for _ in range(1000):
    _ = lst[n // 2]     # O(1) for list
time_list = time.time() - start

start = time.time()
for _ in range(1000):
    _ = d[n // 2]       # O(n) for deque!
time_deque = time.time() - start

print(f"Index access (middle element, 1000 times):")
print(f"  List:  {time_list*1000:.2f} ms")
print(f"  Deque: {time_deque*1000:.2f} ms")
print(f"\nList is {time_deque/max(time_list, 1e-10):.0f}x faster for random index access!")
print("\nLesson: Use deque for front/back operations.")
print("Use list if you need frequent index-based access.")
````

## Part 8: More Practical Examples

**Figure 8.1** -- Simulating a printer queue

````python
from collections import deque

class PrinterQueue:
    """Simulates a printer queue. Jobs are printed in order received."""
    
    def __init__(self):
        self._queue = deque()
    
    def add_job(self, document_name):
        self._queue.append(document_name)
        print(f"  Added '{document_name}' to queue (position {len(self._queue)})")
    
    def print_next(self):
        if self._queue:
            doc = self._queue.popleft()
            print(f"  Printing: '{doc}' ({len(self._queue)} remaining)")
        else:
            print("  No documents in queue")
    
    def status(self):
        print(f"  Queue: {list(self._queue)}")


# Simulate
printer = PrinterQueue()
print("Adding print jobs:")
printer.add_job("Essay.pdf")
printer.add_job("Report.docx")
printer.add_job("Photo.png")

print("\nProcessing:")
printer.print_next()
printer.add_job("Homework.pdf")  # New job arrives while printing
printer.print_next()
printer.print_next()
printer.print_next()
````

**Figure 8.2** -- Using a stack to evaluate postfix expressions

````python
def evaluate_postfix(expression):
    """Evaluate a postfix expression using a stack.
    
    Postfix: operators come AFTER operands.
    Example: '3 4 + 2 *' means (3 + 4) * 2 = 14
    
    Algorithm:
    - Number? Push it
    - Operator? Pop two numbers, compute, push result
    """
    stack = []
    tokens = expression.split()
    
    print(f"Evaluating: {expression}")
    print(f"{'Token':<8} {'Action':<25} {'Stack'}")
    print("-" * 50)
    
    for token in tokens:
        if token in '+-*/':
            b = stack.pop()  # Second operand
            a = stack.pop()  # First operand
            if token == '+': result = a + b
            elif token == '-': result = a - b
            elif token == '*': result = a * b
            elif token == '/': result = a / b
            stack.append(result)
            print(f"{token:<8} {a} {token} {b} = {result:<14} {stack}")
        else:
            stack.append(int(token))
            print(f"{token:<8} {'Push':<25} {stack}")
    
    return stack[0]

# Examples
print(f"Result: {evaluate_postfix('3 4 + 2 *')}")
print()
print(f"Result: {evaluate_postfix('5 1 2 + 4 * + 3 -')}")
````

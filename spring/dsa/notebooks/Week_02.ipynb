## Part 1: Quick Review of Python Lists

A **list** is an ordered, mutable collection. Think of it as a row of numbered boxes.

| Feature | Description |
|---|---|
| Ordered | Items have a position (index) |
| Mutable | You can add, remove, and change items |
| Allows duplicates | `[1, 2, 2, 3]` is valid |
| Mixed types | `[1, "hello", 3.14]` is valid |

**Figure 1.1** -- Common list operations

````python
# Creating a list
fruits = ["apple", "banana", "cherry"]
print("Original:", fruits)

# Access by index - O(1)
print("First item:", fruits[0])
print("Last item:", fruits[-1])

# Append to end - O(1)
fruits.append("date")
print("After append:", fruits)

# Insert at beginning - O(n) !
fruits.insert(0, "avocado")
print("After insert(0):", fruits)

# Remove by value - O(n)
fruits.remove("banana")
print("After remove:", fruits)

# Check membership - O(n)
print("'cherry' in fruits:", "cherry" in fruits)
````

### Why is `insert(0, x)` slow?

When you insert at the beginning, every existing element must shift one position to the right:

```
Before insert(0, "X"):
  Index:   0       1       2
  Value: ["A"]   ["B"]   ["C"]

After insert(0, "X"):
  Index:   0       1       2       3
  Value: ["X"]   ["A"]   ["B"]   ["C"]
           ^       ^       ^       ^
          new    shifted shifted shifted
```

All `n` elements must move. That is why `insert(0, x)` is **O(n)**.

## Part 2: Quick Review of Python Dictionaries

A **dict** is an unordered collection of key-value pairs. Think of it as a real dictionary: you look up a **word** (key) and get its **definition** (value).

| Feature | Description |
|---|---|
| Key-value pairs | Each item has a unique key |
| Fast lookup | Finding a value by key is O(1) |
| Mutable | You can add, remove, and change items |
| Keys must be immutable | Strings, numbers, tuples are OK; lists are not |

**Figure 2.1** -- Common dictionary operations

````python
# Creating a dictionary
student = {
    "name": "Ayse",
    "age": 20,
    "major": "Computer Science"
}
print("Student:", student)

# Access by key - O(1)
print("Name:", student["name"])

# Add/update - O(1)
student["gpa"] = 3.8
print("After adding gpa:", student)

# Check if key exists - O(1)
print("'age' in student:", "age" in student)
print("'email' in student:", "email" in student)

# Delete - O(1)
del student["major"]
print("After delete:", student)
````

### Why are dict operations O(1)?

Dictionaries use a technique called **hashing**:

1. Take the key (e.g., `"name"`)
2. Run it through a **hash function** to get a number
3. Use that number as an index into an internal array
4. Go directly to that position

It is like knowing the exact page number in a book instead of flipping through every page. No matter how big the dictionary is, lookup takes the same time.

## Part 3: Quick Review of Python Sets

A **set** is an unordered collection of **unique** items. Think of it as a bag where each item can only appear once.

| Feature | Description |
|---|---|
| Unordered | No index, no position |
| Unique items only | Adding a duplicate has no effect |
| Fast membership test | Checking `x in set` is O(1) |
| Mutable | You can add and remove items |

**Figure 3.1** -- Common set operations

````python
# Creating a set
colors = {"red", "green", "blue"}
print("Colors:", colors)

# Add - O(1)
colors.add("yellow")
print("After add:", colors)

# Adding a duplicate - no effect
colors.add("red")
print("After adding 'red' again:", colors)

# Membership test - O(1)
print("'blue' in colors:", "blue" in colors)
print("'purple' in colors:", "purple" in colors)

# Remove - O(1)
colors.discard("green")
print("After discard:", colors)

# Remove duplicates from a list
numbers = [1, 2, 2, 3, 3, 3, 4]
unique = list(set(numbers))
print(f"\nOriginal: {numbers}")
print(f"Unique:   {unique}")
````

Sets use the same hashing technique as dictionaries, which is why membership testing is O(1).

## Part 4: Big-O Costs of Common Operations

Here is the complete comparison table. **This is one of the most important tables in this course.**

### List Operations

| Operation | Example | Big-O | Why? |
|---|---|---|---|
| Index access | `lst[i]` | O(1) | Direct memory access |
| Append | `lst.append(x)` | O(1)* | Adds to end, no shifting |
| Insert at beginning | `lst.insert(0, x)` | O(n) | Must shift all elements |
| Insert at position i | `lst.insert(i, x)` | O(n) | Must shift n-i elements |
| Pop from end | `lst.pop()` | O(1) | Removes last, no shifting |
| Pop from beginning | `lst.pop(0)` | O(n) | Must shift all elements |
| Search (membership) | `x in lst` | O(n) | Must check each element |
| Delete by value | `lst.remove(x)` | O(n) | Search + shift |
| Length | `len(lst)` | O(1) | Stored as a counter |
| Sort | `lst.sort()` | O(n log n) | Timsort algorithm |

*O(1) amortized -- occasionally needs to resize the internal array.

### Dict Operations

| Operation | Example | Big-O |
|---|---|---|
| Get by key | `d[key]` | O(1) |
| Set key | `d[key] = val` | O(1) |
| Delete key | `del d[key]` | O(1) |
| Key membership | `key in d` | O(1) |
| Iterate all | `for k in d` | O(n) |

### Set Operations

| Operation | Example | Big-O |
|---|---|---|
| Add | `s.add(x)` | O(1) |
| Remove | `s.discard(x)` | O(1) |
| Membership | `x in s` | O(1) |
| Union | `s1 | s2` | O(n+m) |
| Intersection | `s1 & s2` | O(min(n,m)) |

**Figure 4.1** -- Demonstrating the cost difference: append vs insert(0)

````python
import time

n = 100_000

# Append to end: O(1) per operation, O(n) total
start = time.time()
lst_append = []
for i in range(n):
    lst_append.append(i)
time_append = time.time() - start

# Insert at beginning: O(n) per operation, O(n^2) total!
start = time.time()
lst_insert = []
for i in range(n):
    lst_insert.insert(0, i)
time_insert = time.time() - start

print(f"append() x {n:,}: {time_append:.3f} sec")
print(f"insert(0) x {n:,}: {time_insert:.3f} sec")
print(f"\ninsert(0) is {time_insert/time_append:.0f}x slower!")
print(f"\nWhy? append is O(1) each = O(n) total")
print(f"      insert(0) is O(n) each = O(n^2) total")
````

**Figure 4.2** -- Demonstrating the cost difference: pop() vs pop(0)

````python
import time

n = 100_000

# Pop from end: O(1) per operation
lst1 = list(range(n))
start = time.time()
while lst1:
    lst1.pop()
time_pop_end = time.time() - start

# Pop from beginning: O(n) per operation
lst2 = list(range(n))
start = time.time()
while lst2:
    lst2.pop(0)
time_pop_start = time.time() - start

print(f"pop() x {n:,}:    {time_pop_end:.3f} sec")
print(f"pop(0) x {n:,}:  {time_pop_start:.3f} sec")
print(f"\npop(0) is {time_pop_start/time_pop_end:.0f}x slower!")
````

## Part 5: The Big Difference -- Membership Testing

One of the most important performance differences:

| Operation | List | Set | Dict |
|---|---|---|---|
| `x in collection` | **O(n)** | **O(1)** | **O(1)** |

### Real-World Analogy

**List membership** is like looking for a specific book on an unsorted bookshelf: you check every book one by one until you find it or reach the end.

**Set/Dict membership** is like looking up a word in a hash table (imagine a magic index that tells you the exact location instantly).

**Figure 5.1** -- Membership test: list vs set (small scale demo)

````python
import time

n = 1_000_000
data_list = list(range(n))
data_set = set(range(n))

# Search for an item near the end (worst case for list)
target = n - 1

# List membership
start = time.time()
result = target in data_list
time_list = time.time() - start

# Set membership
start = time.time()
result = target in data_set
time_set = time.time() - start

print(f"Searching for {target:,} in {n:,} items:")
print(f"  List: {time_list*1000:.3f} ms")
print(f"  Set:  {time_set*1000:.6f} ms")
print(f"\nSet is {time_list/max(time_set, 1e-10):.0f}x faster!")
````

## Part 6: Benchmark -- List vs Set Membership at Different Sizes

Let's do a proper benchmark: test membership at multiple sizes and plot the results.

> **🔮 Predict first, then measure. Does reality match your prediction?** Before running the benchmark, predict: how will list lookup time change as n grows? What about set lookup time?

**Figure 6.1** -- Benchmark: `in` operator for list vs set

````python
import time

sizes = [1_000, 5_000, 10_000, 50_000, 100_000, 500_000, 1_000_000]
list_times = []
set_times = []
num_tests = 100  # Average over 100 lookups

print(f"{'Size':>12} | {'List (ms)':>10} | {'Set (ms)':>10} | {'Speedup':>8}")
print("-" * 50)

for n in sizes:
    data_list = list(range(n))
    data_set = set(range(n))
    target = n - 1  # Worst case
    
    # Benchmark list
    start = time.time()
    for _ in range(num_tests):
        _ = target in data_list
    t_list = (time.time() - start) / num_tests
    list_times.append(t_list)
    
    # Benchmark set
    start = time.time()
    for _ in range(num_tests):
        _ = target in data_set
    t_set = (time.time() - start) / num_tests
    set_times.append(t_set)
    
    speedup = t_list / max(t_set, 1e-10)
    print(f"{n:>12,} | {t_list*1000:>9.3f} | {t_set*1000:>9.6f} | {speedup:>7.0f}x")
````

**Figure 6.2** -- Plotting list vs set membership times

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.plot(sizes, [t * 1000 for t in list_times], 'ro-', label='List - O(n)', linewidth=2, markersize=8)
plt.plot(sizes, [t * 1000 for t in set_times], 'gs-', label='Set - O(1)', linewidth=2, markersize=8)
plt.xlabel('Collection Size (n)', fontsize=12)
plt.ylabel('Lookup Time (ms)', fontsize=12)
plt.title('Membership Test: list vs set', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Observation: List lookup time grows linearly with size.")
print("Set lookup time stays flat (constant) regardless of size.")
````

**Figure 6.3** -- Zooming in: set times only (to confirm they are truly constant)

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))
plt.plot(sizes, [t * 1_000_000 for t in set_times], 'gs-', linewidth=2, markersize=8)
plt.xlabel('Collection Size (n)', fontsize=12)
plt.ylabel('Lookup Time (microseconds)', fontsize=12)
plt.title('Set Membership: Time Stays Constant (Zoomed In)', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Even zoomed in, set lookup time barely changes.")
print("This is O(1) -- constant time.")
````

## Part 7: Practical Implications -- When to Use What

| Use Case | Best Structure | Why? |
|---|---|---|
| Ordered collection, access by index | `list` | O(1) index access |
| Frequent membership checks | `set` | O(1) vs O(n) for list |
| Key-value mapping | `dict` | O(1) lookup by key |
| Removing duplicates | `set` | Automatic uniqueness |
| Maintaining insertion order | `list` or `dict` | Both preserve order |
| Frequent insertions at beginning | Neither list nor set alone | Use `collections.deque` (next week!) |

**Figure 7.1** -- Practical example: checking student attendance

````python
import time

# Scenario: Check which students from a class attended a lecture
# Class has 10,000 students, 8,000 attended

all_students = [f"student_{i}" for i in range(10_000)]
attended_list = [f"student_{i}" for i in range(8_000)]  # as a list
attended_set = set(attended_list)                        # as a set

# Check each student's attendance using LIST
start = time.time()
absent_list = []
for student in all_students:
    if student not in attended_list:
        absent_list.append(student)
time_list = time.time() - start

# Check each student's attendance using SET
start = time.time()
absent_set = []
for student in all_students:
    if student not in attended_set:
        absent_set.append(student)
time_set = time.time() - start

print(f"Absent students found: {len(absent_list)}")
print(f"Using list: {time_list:.3f} sec")
print(f"Using set:  {time_set:.4f} sec")
print(f"Set is {time_list/time_set:.0f}x faster!")
print(f"\nLesson: If you need to check membership many times,")
print(f"convert your list to a set first!")
````

**Figure 7.2** -- Practical example: counting word frequencies

````python
# Dictionaries are perfect for counting
text = "the cat sat on the mat the cat ate the rat on the mat"
words = text.split()

# Count frequencies using a dictionary
freq = {}
for word in words:
    if word in freq:        # O(1) dictionary lookup
        freq[word] += 1
    else:
        freq[word] = 1

# Display results
print("Word frequencies:")
for word, count in sorted(freq.items(), key=lambda x: -x[1]):
    print(f"  {word:>6}: {'*' * count} ({count})")

print(f"\nTotal words: {len(words)}")
print(f"Unique words: {len(freq)}")
print(f"\nThis entire process is O(n) where n = number of words.")
print(f"Each dict lookup and update is O(1).")
````

## Part 8: Common Errors & Pitfalls

**Figure 8.1** -- Mistake: Using a list when you should use a set

````python
import time

# BAD: Using a list to check if items are "allowed"
allowed_users_list = list(range(100_000))

start = time.time()
for user_id in range(1000):
    if user_id in allowed_users_list:  # O(n) each time!
        pass
time_bad = time.time() - start

# GOOD: Using a set
allowed_users_set = set(range(100_000))

start = time.time()
for user_id in range(1000):
    if user_id in allowed_users_set:  # O(1) each time!
        pass
time_good = time.time() - start

print(f"Bad (list):  {time_bad:.4f} sec")
print(f"Good (set):  {time_good:.6f} sec")
print(f"\nTip: Always use a set or dict for membership checks!")
````

**Figure 8.2** -- Mistake: Modifying a list while iterating

````python
# BAD: Removing items while iterating
numbers = [1, 2, 3, 4, 5, 6]
print("Original:", numbers)

# This gives unexpected results!
numbers_copy = numbers.copy()
for n in numbers_copy:
    if n % 2 == 0:
        numbers_copy.remove(n)  # Careful: modifying while iterating
print("After removing evens (buggy way):", numbers_copy)

# CORRECT: Use list comprehension
numbers_clean = [n for n in numbers if n % 2 != 0]
print("After removing evens (correct):  ", numbers_clean)
````

**Figure 8.3** -- Mistake: Using unhashable types as dict keys or set items

````python
# Lists are mutable and cannot be used as dict keys or in sets

try:
    bad_dict = {[1, 2, 3]: "hello"}  # List as key
except TypeError as e:
    print(f"TypeError with list as dict key: {e}")

try:
    bad_set = {[1, 2, 3]}  # List in a set
except TypeError as e:
    print(f"TypeError with list in set: {e}")

# SOLUTION: Use tuples instead (they are immutable)
good_dict = {(1, 2, 3): "hello"}
good_set = {(1, 2, 3)}
print(f"\nTuple as dict key works: {good_dict}")
print(f"Tuple in set works: {good_set}")
````

**Figure 8.4** -- Mistake: KeyError when accessing missing dict key

````python
student = {"name": "Ali", "age": 21}

# BAD: Direct access may crash
try:
    email = student["email"]  # Key doesn't exist!
except KeyError as e:
    print(f"KeyError: {e}")

# GOOD: Use .get() with a default value
email = student.get("email", "not provided")
print(f"Email: {email}")

# GOOD: Check first with 'in'
if "email" in student:
    print(student["email"])
else:
    print("No email on file")
````

## Part 9: Benchmark Lab -- Append vs Insert at Scale

Let's create a full benchmark comparing `append()` and `insert(0, x)` across multiple sizes and plot the results.

> **🔮 Predict first, then measure. Does reality match your prediction?** Before running: predict the Big-O of each operation and what the ratio should look like as n grows.

**Figure 9.1** -- Full benchmark: append vs insert(0) with plotting

````python
import time
import matplotlib.pyplot as plt

sizes = [1_000, 5_000, 10_000, 20_000, 50_000, 100_000]
append_times = []
insert_times = []

for n in sizes:
    # Benchmark append
    start = time.time()
    lst = []
    for i in range(n):
        lst.append(i)
    append_times.append(time.time() - start)
    
    # Benchmark insert(0)
    start = time.time()
    lst = []
    for i in range(n):
        lst.insert(0, i)
    insert_times.append(time.time() - start)

# Print table
print(f"{'Size':>10} | {'append (ms)':>12} | {'insert(0) (ms)':>15} | {'Ratio':>8}")
print("-" * 55)
for i, n in enumerate(sizes):
    ratio = insert_times[i] / max(append_times[i], 1e-10)
    print(f"{n:>10,} | {append_times[i]*1000:>11.2f} | {insert_times[i]*1000:>14.2f} | {ratio:>7.0f}x")

# Plot
plt.figure(figsize=(10, 5))
plt.plot(sizes, [t*1000 for t in append_times], 'go-', label='append() - O(1) each', linewidth=2, markersize=8)
plt.plot(sizes, [t*1000 for t in insert_times], 'rs-', label='insert(0) - O(n) each', linewidth=2, markersize=8)
plt.xlabel('Number of Operations', fontsize=12)
plt.ylabel('Total Time (ms)', fontsize=12)
plt.title('List: append() vs insert(0, x)', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

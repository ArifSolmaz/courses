## Part 1: Review — What Is a Binary Tree?

Before we dive into Binary Search Trees, let’s quickly recall what a **binary tree** is.

A binary tree is a tree where each node has **at most 2 children** (left and right).

| Term | Meaning |
|------|--------|
| **Root** | The top node (no parent) |
| **Leaf** | A node with no children |
| **Parent** | A node that has children |
| **Height** | Longest path from root to a leaf |
| **Subtree** | A node and all its descendants |

**Real-world analogy:** Think of a family tree where each person has at most two children.

**Figure 1.1** — A simple binary tree node class (review from Week 09)

````python
class TreeNode:
    """A node in a binary tree."""
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

# Build a small binary tree manually
#       10
#      /  \
#     5    15
#    / \
#   3   7

root = TreeNode(10)
root.left = TreeNode(5)
root.right = TreeNode(15)
root.left.left = TreeNode(3)
root.left.right = TreeNode(7)

print(f"Root: {root.value}")
print(f"Left child: {root.left.value}")
print(f"Right child: {root.right.value}")
````

## Part 2: The BST Property

A **Binary Search Tree (BST)** is a binary tree with one special rule:

> For every node:
> - All values in the **left subtree** are **less than** the node’s value
> - All values in the **right subtree** are **greater than** the node’s value

```
        8
       / \
      3   10
     / \    \
    1   6    14
       / \   /
      4   7 13
```

**Real-world analogy:** Imagine a library where books are organized:
- Books with titles A–M go to the **left** shelf
- Books with titles N–Z go to the **right** shelf
- Each shelf is further divided the same way

This makes finding a book very fast — you eliminate half the options at each step!

**Figure 2.1** — Checking if a binary tree satisfies the BST property

````python
# Let's verify the BST property on our tree from Part 1
#       10
#      /  \
#     5    15
#    / \
#   3   7

# Check: is 5 < 10? (left child < parent)
print(f"Left child {root.left.value} < Root {root.value}? {root.left.value < root.value}")

# Check: is 15 > 10? (right child > parent)
print(f"Right child {root.right.value} > Root {root.value}? {root.right.value > root.value}")

# Check: is 3 < 5? (left child of 5 < 5)
print(f"Left child {root.left.left.value} < Parent {root.left.value}? {root.left.left.value < root.left.value}")

# Check: is 7 > 5? (right child of 5 > 5)
print(f"Right child {root.left.right.value} > Parent {root.left.value}? {root.left.right.value > root.left.value}")

print("\n✅ This tree IS a valid BST!")
````

## Part 3: Why BSTs Are Useful

Why not just use a sorted list? Let’s compare:

| Operation | Sorted List | BST (balanced) |
|-----------|------------|----------------|
| Search | O(log n) with binary search | O(log n) |
| Insert | O(n) — must shift elements | O(log n) |
| Delete | O(n) — must shift elements | O(log n) |

BSTs combine the **fast search** of a sorted array with **fast insert/delete** of a linked list.

**Real-world analogy:** A phone book (sorted list) is great for looking up names, but adding a new entry means rewriting pages. A BST is like an index card system where you can slot in a new card quickly.

**Figure 3.1** — Comparing search in a list vs BST

````python
# Searching in a list: we might check every element
numbers = [1, 3, 5, 7, 10, 15]

target = 7
steps = 0
for num in numbers:
    steps += 1
    if num == target:
        break
print(f"Linear search for {target}: found in {steps} steps")

# In a BST with the same values:
#       10
#      /  \
#     5    15
#    / \
#   3   7
#  /
# 1
# To find 7: 10 -> 5 -> 7 = only 3 steps!
print(f"BST search for {target}: found in 3 steps (10 → 5 → 7)")
````

````python
class Node:
    """A node in a Binary Search Tree."""
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
    
    def __repr__(self):
        return f"Node({self.value})"


class BST:
    """A Binary Search Tree."""
    def __init__(self):
        self.root = None
    
    def is_empty(self):
        return self.root is None


# Create an empty BST
tree = BST()
print(f"Tree is empty: {tree.is_empty()}")
print(f"Root: {tree.root}")
````

## Part 5: Inserting into a BST

To insert a value into a BST:

1. Start at the **root**
2. If the new value is **less than** the current node, go **left**
3. If the new value is **greater than** the current node, go **right**
4. When you reach an empty spot (`None`), place the new node there

**Real-world analogy:** Imagine you’re filing a document in a cabinet:
- Open the top drawer (root)
- Is your document before or after the divider? Go left or right
- Keep narrowing down until you find the right slot

**Figure 5.1** — Step-by-step insertion example

````python
# Let's trace inserting values: 10, 5, 15, 3, 7

print("Step 1: Insert 10")
print("  Tree is empty → 10 becomes the root")
print("  Tree:  10")
print()

print("Step 2: Insert 5")
print("  5 < 10 → go left → empty spot → place 5")
print("  Tree:   10")
print("         /")
print("        5")
print()

print("Step 3: Insert 15")
print("  15 > 10 → go right → empty spot → place 15")
print("  Tree:   10")
print("         /  \\")
print("        5   15")
print()

print("Step 4: Insert 3")
print("  3 < 10 → go left → 3 < 5 → go left → empty → place 3")
print("  Tree:   10")
print("         /  \\")
print("        5   15")
print("       /")
print("      3")
print()

print("Step 5: Insert 7")
print("  7 < 10 → go left → 7 > 5 → go right → empty → place 7")
print("  Tree:   10")
print("         /  \\")
print("        5   15")
print("       / \\")
print("      3   7")
````

**Figure 5.2** — The `insert()` method implementation

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None
    
    def insert(self, value):
        """Insert a value into the BST."""
        new_node = Node(value)
        
        # If tree is empty, new node becomes root
        if self.root is None:
            self.root = new_node
            print(f"  Inserted {value} as root")
            return
        
        # Walk down the tree to find the right spot
        current = self.root
        while True:
            if value < current.value:
                # Go left
                if current.left is None:
                    current.left = new_node
                    print(f"  Inserted {value} to the left of {current.value}")
                    return
                current = current.left
            elif value > current.value:
                # Go right
                if current.right is None:
                    current.right = new_node
                    print(f"  Inserted {value} to the right of {current.value}")
                    return
                current = current.right
            else:
                # Duplicate value — we skip it
                print(f"  {value} already exists, skipping")
                return


# Build a BST by inserting values one by one
tree = BST()
for val in [10, 5, 15, 3, 7, 12, 20]:
    tree.insert(val)
````

## Part 6: Searching in a BST

Searching follows the same logic as inserting:

1. Start at the **root**
2. If the target equals the current node’s value → **found it!**
3. If the target is **less**, go **left**
4. If the target is **greater**, go **right**
5. If you reach `None` → **not in the tree**

**Real-world analogy:** Like looking up a word in a dictionary — you open to the middle, then go left or right based on alphabetical order.

**Figure 6.1** — The `search()` method implementation

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None
    
    def insert(self, value):
        new_node = Node(value)
        if self.root is None:
            self.root = new_node
            return
        current = self.root
        while True:
            if value < current.value:
                if current.left is None:
                    current.left = new_node
                    return
                current = current.left
            elif value > current.value:
                if current.right is None:
                    current.right = new_node
                    return
                current = current.right
            else:
                return  # duplicate
    
    def search(self, target):
        """Search for a value in the BST. Returns True/False."""
        current = self.root
        steps = 0
        
        while current is not None:
            steps += 1
            print(f"  Step {steps}: comparing {target} with {current.value}")
            
            if target == current.value:
                print(f"  ✅ Found {target} in {steps} steps!")
                return True
            elif target < current.value:
                print(f"  {target} < {current.value} → go left")
                current = current.left
            else:
                print(f"  {target} > {current.value} → go right")
                current = current.right
        
        print(f"  ❌ {target} not found (reached None after {steps} steps)")
        return False


# Build tree: 10, 5, 15, 3, 7, 12, 20
tree = BST()
for val in [10, 5, 15, 3, 7, 12, 20]:
    tree.insert(val)

print("Searching for 7:")
tree.search(7)

print("\nSearching for 13:")
tree.search(13)
````

## Part 7: In-Order Traversal — Sorted Output

One of the most useful properties of a BST: **in-order traversal** visits nodes in **sorted order**.

In-order traversal follows this pattern:
1. Visit the **left** subtree
2. Visit the **current** node
3. Visit the **right** subtree

```
        10
       /  \
      5    15
     / \   / \
    3   7 12  20
```

In-order: 3, 5, 7, 10, 12, 15, 20 — **sorted!**

**Figure 7.1** — In-order traversal implementation

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None
    
    def insert(self, value):
        new_node = Node(value)
        if self.root is None:
            self.root = new_node
            return
        current = self.root
        while True:
            if value < current.value:
                if current.left is None:
                    current.left = new_node
                    return
                current = current.left
            elif value > current.value:
                if current.right is None:
                    current.right = new_node
                    return
                current = current.right
            else:
                return
    
    def search(self, target):
        current = self.root
        while current is not None:
            if target == current.value:
                return True
            elif target < current.value:
                current = current.left
            else:
                current = current.right
        return False
    
    def inorder(self, node=None, result=None):
        """In-order traversal: Left → Node → Right (returns sorted list)."""
        if result is None:
            result = []
            node = self.root
        
        if node is not None:
            self.inorder(node.left, result)   # Visit left subtree
            result.append(node.value)          # Visit current node
            self.inorder(node.right, result)   # Visit right subtree
        
        return result


# Build BST with unsorted values
tree = BST()
values = [10, 5, 15, 3, 7, 12, 20, 1, 9]
print(f"Insertion order: {values}")

for val in values:
    tree.insert(val)

sorted_values = tree.inorder()
print(f"In-order traversal: {sorted_values}")
print(f"\n✅ The values come out sorted!")
````

**Figure 7.2** — Pre-order and post-order traversals (bonus)

````python
def preorder(node):
    """Pre-order: Node → Left → Right"""
    if node is None:
        return []
    return [node.value] + preorder(node.left) + preorder(node.right)

def postorder(node):
    """Post-order: Left → Right → Node"""
    if node is None:
        return []
    return postorder(node.left) + postorder(node.right) + [node.value]

print(f"In-order   (L-N-R): {tree.inorder()}")
print(f"Pre-order  (N-L-R): {preorder(tree.root)}")
print(f"Post-order (L-R-N): {postorder(tree.root)}")
````

## Part 8: BST Performance — Best vs Worst Case

The performance of a BST depends on its **shape**:

| Case | Tree Shape | Search Time |
|------|-----------|-------------|
| **Best case** | Balanced tree | O(log n) |
| **Worst case** | Completely unbalanced (like a linked list) | O(n) |

```
Balanced BST:           Unbalanced BST:
      10                  1
     /  \                  \
    5    15                 3
   / \   / \                 \
  3   7 12  20                5
                               \
Height: 3 (log₂7 ≈ 3)          7
                                 \
                                  10
                               Height: 5 (= n-2)
```

**When does worst case happen?** When you insert values in sorted order!

**Figure 8.1** — Demonstrating balanced vs unbalanced BST

````python
def count_steps(tree, target):
    """Count how many comparisons needed to find target."""
    current = tree.root
    steps = 0
    while current is not None:
        steps += 1
        if target == current.value:
            return steps
        elif target < current.value:
            current = current.left
        else:
            current = current.right
    return steps


# Balanced tree — insert in a good order
balanced = BST()
for val in [10, 5, 15, 3, 7, 12, 20]:
    balanced.insert(val)

# Unbalanced tree — insert in sorted order (worst case!)
unbalanced = BST()
for val in [1, 3, 5, 7, 10, 12, 15, 20]:
    unbalanced.insert(val)

# Search for the value 20 in both trees
print("Searching for 20:")
print(f"  Balanced BST:   {count_steps(balanced, 20)} steps")
print(f"  Unbalanced BST: {count_steps(unbalanced, 20)} steps")

print(f"\nWith 7 nodes:")
print(f"  Best case (balanced):   log₂(7) ≈ 3 steps")
print(f"  Worst case (unbalanced): up to 8 steps")
````

## Part 9: Complete BST with Helper Methods

Let’s put everything together in one complete, clean BST class with some useful extras.

**Figure 9.1** — Complete BST implementation

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None
    
    def insert(self, value):
        """Insert a value into the BST."""
        new_node = Node(value)
        if self.root is None:
            self.root = new_node
            return
        current = self.root
        while True:
            if value < current.value:
                if current.left is None:
                    current.left = new_node
                    return
                current = current.left
            elif value > current.value:
                if current.right is None:
                    current.right = new_node
                    return
                current = current.right
            else:
                return  # No duplicates
    
    def search(self, target):
        """Return True if target is in the BST."""
        current = self.root
        while current is not None:
            if target == current.value:
                return True
            elif target < current.value:
                current = current.left
            else:
                current = current.right
        return False
    
    def inorder(self, node=None, result=None):
        """Return values in sorted order."""
        if result is None:
            result = []
            node = self.root
        if node is not None:
            self.inorder(node.left, result)
            result.append(node.value)
            self.inorder(node.right, result)
        return result
    
    def find_min(self):
        """Find the minimum value (leftmost node)."""
        if self.root is None:
            return None
        current = self.root
        while current.left is not None:
            current = current.left
        return current.value
    
    def find_max(self):
        """Find the maximum value (rightmost node)."""
        if self.root is None:
            return None
        current = self.root
        while current.right is not None:
            current = current.right
        return current.value
    
    def height(self, node="DEFAULT"):
        """Calculate the height of the tree."""
        if node == "DEFAULT":
            node = self.root
        if node is None:
            return -1
        left_h = self.height(node.left)
        right_h = self.height(node.right)
        return 1 + max(left_h, right_h)


# Demo
tree = BST()
for val in [10, 5, 15, 3, 7, 12, 20, 1]:
    tree.insert(val)

print(f"Sorted values: {tree.inorder()}")
print(f"Min: {tree.find_min()}")
print(f"Max: {tree.find_max()}")
print(f"Height: {tree.height()}")
print(f"Search 7: {tree.search(7)}")
print(f"Search 99: {tree.search(99)}")
````

## Part 10: Common Errors

Let’s look at mistakes beginners often make with BSTs.

**Figure 10.1** — Common error: forgetting to handle None

````python
# ❌ Error: Trying to access .value on None
tree = BST()

try:
    # Accessing root.value on an empty tree
    print(tree.root.value)
except AttributeError as e:
    print(f"❌ AttributeError: {e}")
    print("   Always check if a node is None before accessing its value!")

print()

# ✅ Correct: check first
if tree.root is not None:
    print(tree.root.value)
else:
    print("✅ Tree is empty — no root to access")
````

**Figure 10.2** — Common error: confusing assignment with comparison

````python
# ❌ Common mistake: using current = current.left without checking
# This doesn't crash, but you lose the reference to the parent node

tree = BST()
tree.insert(10)
tree.insert(5)

# Mistake: overwriting current without a while loop
current = tree.root       # current = Node(10)
current = current.left    # current = Node(5) — lost reference to root!
# Now we can't go back to the root
print(f"current is now: {current.value}")
print("We've lost access to the root!")
print()

# ✅ In a search/insert, this is fine because we only go forward
# But if you need to go back up, you need a parent pointer or recursion
print("✅ Tip: BST insert/search only goes downward, so this pattern is OK there.")
````

**Figure 10.3** — Common error: inserting sorted data creates a bad tree

````python
# ⚠️ Inserting sorted data creates a "degenerate" tree (like a linked list)
bad_tree = BST()
for val in [1, 2, 3, 4, 5, 6, 7]:
    bad_tree.insert(val)

print(f"Height of tree with 7 nodes (sorted insert): {bad_tree.height()}")
print(f"Expected height if balanced: ~{2}")
print()

# ✅ Same values, better insertion order
good_tree = BST()
for val in [4, 2, 6, 1, 3, 5, 7]:  # Insert middle value first!
    good_tree.insert(val)

print(f"Height of tree with 7 nodes (balanced insert): {good_tree.height()}")
print("\n✅ Tip: To build a balanced BST, insert the median value first!")
````

## Part 12: Deleting from a BST

Deletion is the most complex BST operation because removing a node must preserve the BST property. There are **three cases** to handle:

**Case 1 — Leaf node (no children):**
Simply remove the node. Nothing else needs to change.

```
Delete 7:
      10              10
     /  \     →      /  \
    5    15         5    15
   / \             /
  3   7           3
```

**Case 2 — Node with one child:**
Replace the node with its only child. The child "moves up" to take the deleted node's place.

```
Delete 5 (has only left child):
      10              10
     /  \     →      /  \
    5    15         3    15
   /
  3
```

**Case 3 — Node with two children:**
This is the tricky case. We cannot simply remove the node because it has two subtrees. Instead:

1. Find the **in-order successor** — the smallest value in the right subtree
2. Copy that value into the node we want to delete
3. Delete the in-order successor (which will be Case 1 or Case 2)

```
Delete 10 (has two children):
  In-order successor of 10 = 12 (smallest in right subtree)
  Copy 12 into node, then delete the original 12

      10              12
     /  \     →      /  \
    5    15         5    15
   / \   /         / \
  3   7 12        3   7
```

**Why the in-order successor?** It is the smallest value that is still greater than everything in the left subtree, so replacing with it keeps the BST property intact.

**Figure 12.1** — The `delete()` method implementation

````python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


class BST:
    def __init__(self):
        self.root = None

    def insert(self, value):
        new_node = Node(value)
        if self.root is None:
            self.root = new_node
            return
        current = self.root
        while True:
            if value < current.value:
                if current.left is None:
                    current.left = new_node
                    return
                current = current.left
            elif value > current.value:
                if current.right is None:
                    current.right = new_node
                    return
                current = current.right
            else:
                return  # No duplicates

    def search(self, target):
        current = self.root
        while current is not None:
            if target == current.value:
                return True
            elif target < current.value:
                current = current.left
            else:
                current = current.right
        return False

    def inorder(self, node=None, result=None):
        if result is None:
            result = []
            node = self.root
        if node is not None:
            self.inorder(node.left, result)
            result.append(node.value)
            self.inorder(node.right, result)
        return result

    def find_min(self):
        if self.root is None:
            return None
        current = self.root
        while current.left is not None:
            current = current.left
        return current.value

    def find_max(self):
        if self.root is None:
            return None
        current = self.root
        while current.right is not None:
            current = current.right
        return current.value

    def delete(self, value):
        """Delete a value from the BST."""
        self.root = self._delete_recursive(self.root, value)

    def _delete_recursive(self, node, value):
        """Recursively delete a node and return the updated subtree root."""
        # Base case: value not found
        if node is None:
            print(f"  {value} not found in the tree")
            return None

        # Navigate to the node to delete
        if value < node.value:
            node.left = self._delete_recursive(node.left, value)
        elif value > node.value:
            node.right = self._delete_recursive(node.right, value)
        else:
            # Found the node to delete!

            # Case 1: Leaf node (no children)
            if node.left is None and node.right is None:
                print(f"  Deleted {value} (leaf node)")
                return None

            # Case 2a: Only right child
            elif node.left is None:
                print(f"  Deleted {value} (had right child {node.right.value})")
                return node.right

            # Case 2b: Only left child
            elif node.right is None:
                print(f"  Deleted {value} (had left child {node.left.value})")
                return node.left

            # Case 3: Two children
            else:
                # Find the in-order successor (smallest in right subtree)
                successor = node.right
                while successor.left is not None:
                    successor = successor.left
                print(f"  Deleted {value} (two children, replaced with successor {successor.value})")
                # Copy successor's value into this node
                node.value = successor.value
                # Delete the successor from the right subtree
                node.right = self._delete_recursive(node.right, successor.value)

        return node


# --- Demo: test all three deletion cases ---

tree = BST()
for val in [10, 5, 15, 3, 7, 12, 20]:
    tree.insert(val)

print(f"Initial tree (in-order): {tree.inorder()}")
print()

# Case 1: Delete a leaf
print("Case 1 — Delete leaf node 3:")
tree.delete(3)
print(f"  In-order: {tree.inorder()}")
print()

# Case 2: Delete node with one child
print("Case 2 — Delete node 5 (one child: 7):")
tree.delete(5)
print(f"  In-order: {tree.inorder()}")
print()

# Case 3: Delete node with two children
print("Case 3 — Delete node 15 (two children: 12 and 20):")
tree.delete(15)
print(f"  In-order: {tree.inorder()}")
````

**Figure 12.2** — Visual deletion example: removing a node with two children

````python
# Build a larger BST and delete the root (a two-children case)
tree = BST()
for val in [50, 30, 70, 20, 40, 60, 80, 35, 45, 55, 65]:
    tree.insert(val)

print("Tree structure:")
print("              50")
print("           /      \\")
print("         30        70")
print("        /  \\      /  \\")
print("      20   40    60   80")
print("          / \\   / \\")
print("        35  45 55  65")
print()

print(f"In-order before: {tree.inorder()}")
print()

# Delete root node 50 (two children)
# In-order successor = 55 (smallest in right subtree)
print("Deleting root node 50:")
tree.delete(50)
print(f"In-order after:  {tree.inorder()}")
print()

print("New tree structure:")
print("              55")
print("           /      \\")
print("         30        70")
print("        /  \\      /  \\")
print("      20   40    60   80")
print("          / \\     \\")
print("        35  45    65")
print()

# Verify BST property still holds
vals = tree.inorder()
is_sorted = all(vals[i] < vals[i+1] for i in range(len(vals)-1))
print(f"Values still sorted? {is_sorted}")
print(f"Search for 55 (new root): {tree.search(55)}")
print(f"Search for 50 (deleted):  {tree.search(50)}")
````

**Figure 12.3** — Summary of BST deletion cases

| Case | Condition | Action | Time Complexity |
|------|-----------|--------|-----------------|
| **Case 1** | Node has **no children** (leaf) | Remove the node | O(h) to find the node |
| **Case 2** | Node has **one child** | Replace node with its child | O(h) to find the node |
| **Case 3** | Node has **two children** | Replace value with in-order successor, then delete the successor | O(h) to find + O(h) to find successor |

Where **h** is the height of the tree: **O(log n)** for a balanced BST, **O(n)** in the worst case (skewed tree).

**Key insight:** Case 3 always reduces to Case 1 or Case 2, because the in-order successor (the smallest node in the right subtree) can have at most one child (a right child). It cannot have a left child, otherwise that left child would be smaller and would be the successor instead.

````python
import time
import random
import matplotlib.pyplot as plt

sizes = [100, 500, 1000, 2000, 5000]
num_runs = 5

balanced_times = []
skewed_times = []

for n in sizes:
    values = list(range(n))

    # Balanced BST: insert in random order
    times = []
    for _ in range(num_runs):
        shuffled = values[:]
        random.shuffle(shuffled)
        tree = BST()
        start = time.time()
        for v in shuffled:
            tree.insert(v)
        # Search for all values
        for v in shuffled:
            tree.search(v)
        times.append(time.time() - start)
    balanced_times.append(sum(times) / num_runs)

    # Skewed BST: insert in sorted order
    times = []
    for _ in range(num_runs):
        import sys
        old_limit = sys.getrecursionlimit()
        sys.setrecursionlimit(max(old_limit, n + 100))
        tree = BST()
        start = time.time()
        for v in values:  # sorted order → skewed tree
            tree.insert(v)
        # Search for all values
        for v in values:
            tree.search(v)
        times.append(time.time() - start)
        sys.setrecursionlimit(old_limit)
    skewed_times.append(sum(times) / num_runs)

    print(f"n={n:5d}: balanced={balanced_times[-1]:.4f}s  skewed={skewed_times[-1]:.4f}s")

plt.figure(figsize=(10, 6))
plt.plot(sizes, balanced_times, 'o-', label='Balanced (random insert)', linewidth=2)
plt.plot(sizes, skewed_times, 's-', label='Skewed (sorted insert)', linewidth=2)
plt.xlabel('Number of Elements (n)')
plt.ylabel('Average Time (seconds)')
plt.title('BST Insert + Search: Balanced vs Skewed')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
````

**Observations:**
- The **balanced BST** (random insertion order) shows much better performance, growing slowly as n increases. Each insert and search takes O(log n) on average, so the total time grows as O(n log n).
- The **skewed BST** (sorted insertion order) degrades dramatically. Each insert and search walks the entire chain, making each operation O(n) and the total O(n^2). You can see the curve bending upward steeply.
- This benchmark clearly demonstrates why balanced trees (AVL, Red-Black) are important in practice — they guarantee O(log n) operations regardless of insertion order.

## Part 1: What Is an Algorithm?

An **algorithm** is a step-by-step set of instructions to solve a problem.

You already use algorithms every day:

| Everyday Task | Algorithm |
|---|---|
| Making tea | Boil water, add tea bag, wait 3 min, remove bag |
| Finding a word in the dictionary | Open to the middle, go left or right depending on the letter |
| Sorting playing cards in your hand | Pick each card and insert it in the right position |

In programming, we write algorithms to process data. But here is the key question:

> **If two algorithms solve the same problem, how do we decide which one is "better"?**

### What Makes an Algorithm "Better"?

We care about two things:

1. **Time** - How fast does it run?
2. **Space** - How much memory does it use?

In this course, we will mostly focus on **time complexity** - how the running time grows as the input gets larger.

### Real-World Analogy: Finding a Name in a Phone Book

Imagine you need to find "Mehmet" in a phone book with 1,000 names.

**Approach 1: Linear Search**
- Start from page 1, check every name one by one
- Worst case: check all 1,000 names

**Approach 2: Binary Search**
- Open to the middle (page 500)
- "Mehmet" comes after the middle? Go to the right half
- Open to the middle of the right half (page 750)
- Keep halving until you find it
- Worst case: about 10 checks!

| Phone Book Size | Linear Search (worst) | Binary Search (worst) |
|---|---|---|
| 1,000 | 1,000 checks | 10 checks |
| 1,000,000 | 1,000,000 checks | 20 checks |
| 1,000,000,000 | 1,000,000,000 checks | 30 checks |

The difference is **dramatic** as the input grows. This is why we need a way to describe how algorithms scale.

## Part 2: Counting Operations

Before we learn Big-O, let's practice counting how many operations a piece of code performs.

An **operation** is any basic step: an assignment, a comparison, an arithmetic operation, etc.

**Figure 2.1** -- Counting operations in a simple loop

````python
def sum_list(numbers):
    total = 0                  # 1 operation (assignment)
    for num in numbers:        # n iterations (n = length of list)
        total = total + num    # 2 operations per iteration (add + assign)
    return total               # 1 operation

# Total: 1 + 2n + 1 = 2n + 2 operations
# For a list of 100 items: 2(100) + 2 = 202 operations
# For a list of 1000 items: 2(1000) + 2 = 2002 operations

result = sum_list([10, 20, 30, 40, 50])
print(f"Sum = {result}")
````

**Figure 2.2** -- Counting operations in nested loops

````python
def print_pairs(n):
    count = 0
    for i in range(n):          # n iterations
        for j in range(n):      # n iterations for EACH i
            count += 1          # 1 operation
    return count

# Total inner operations: n * n = n^2

print(f"n=5:   {print_pairs(5)} operations")    # 25
print(f"n=10:  {print_pairs(10)} operations")   # 100
print(f"n=20:  {print_pairs(20)} operations")   # 400
print(f"n=100: {print_pairs(100)} operations")  # 10,000
````

Notice how doubling `n` from 10 to 20 quadrupled the operations (100 to 400). That is the signature of **n-squared** growth.

## Part 3: Big-O Notation

**Big-O notation** describes the **growth rate** of an algorithm as the input size `n` increases.

We focus on the **dominant term** and **ignore constants**:

| Actual Count | Big-O | Why? |
|---|---|---|
| 2n + 2 | O(n) | Drop the constant 2 and the +2 |
| 5n^2 + 3n + 10 | O(n^2) | n^2 dominates as n grows |
| 100 | O(1) | Constant, doesn't depend on n |
| 3 * log(n) + 5 | O(log n) | Drop the constant 3 and +5 |

### Why Drop Constants?

When `n` is very large, constants become irrelevant:
- `2n` vs `5n`: both grow linearly, just different slopes
- `n^2` vs `1000n`: at some point `n^2` will always overtake `1000n`

### Common Big-O Classes (from fastest to slowest)

| Big-O | Name | Example | Growth |
|---|---|---|---|
| O(1) | Constant | Accessing a list by index | Does not grow |
| O(log n) | Logarithmic | Binary search | Grows very slowly |
| O(n) | Linear | Loop through a list | Grows proportionally |
| O(n log n) | Log-linear | Efficient sorting (merge sort) | Slightly faster than quadratic |
| O(n^2) | Quadratic | Nested loops | Grows fast |
| O(2^n) | Exponential | Brute-force subsets | Grows extremely fast |

**Rule of thumb:** If your algorithm is O(n^2) or worse, look for a better approach when `n` is large.

**Figure 3.1** -- Visualizing growth rates

````python
import matplotlib.pyplot as plt
import math

n_values = list(range(1, 51))

o_1      = [1 for _ in n_values]
o_log_n  = [math.log2(n) for n in n_values]
o_n      = n_values[:]
o_n_logn = [n * math.log2(n) for n in n_values]
o_n2     = [n ** 2 for n in n_values]

plt.figure(figsize=(10, 6))
plt.plot(n_values, o_1,      label="O(1) - Constant",    linewidth=2)
plt.plot(n_values, o_log_n,  label="O(log n) - Logarithmic", linewidth=2)
plt.plot(n_values, o_n,      label="O(n) - Linear",      linewidth=2)
plt.plot(n_values, o_n_logn, label="O(n log n) - Log-linear", linewidth=2)
plt.plot(n_values, o_n2,     label="O(n^2) - Quadratic",  linewidth=2)

plt.xlabel("Input Size (n)", fontsize=12)
plt.ylabel("Number of Operations", fontsize=12)
plt.title("Growth Rates of Common Big-O Classes", fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.ylim(0, 500)
plt.tight_layout()
plt.show()
````

**Figure 3.2** -- Examples of each Big-O class in Python code

````python
# O(1) - Constant: accessing by index
def get_first(data):
    return data[0]      # Always 1 step, no matter how big the list

# O(log n) - Logarithmic: binary search
def binary_search(sorted_list, target):
    low, high = 0, len(sorted_list) - 1
    while low <= high:
        mid = (low + high) // 2
        if sorted_list[mid] == target:
            return mid
        elif sorted_list[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# O(n) - Linear: simple loop
def find_max(data):
    maximum = data[0]
    for item in data:
        if item > maximum:
            maximum = item
    return maximum

# O(n^2) - Quadratic: nested loops
def has_duplicates(data):
    for i in range(len(data)):
        for j in range(i + 1, len(data)):
            if data[i] == data[j]:
                return True
    return False

# Demo
sample = [3, 7, 1, 9, 4, 6, 2, 8, 5]
print(f"First element: {get_first(sample)}")
print(f"Max element:   {find_max(sample)}")
print(f"Has duplicates: {has_duplicates(sample)}")

sorted_sample = sorted(sample)
print(f"Binary search for 6 in {sorted_sample}: index {binary_search(sorted_sample, 6)}")
````

### How to Determine Big-O: Simple Rules

1. **No loops** -> O(1)
2. **One loop** over n items -> O(n)
3. **Nested loops** (loop inside a loop) -> O(n^2)
4. **Halving the input** each step -> O(log n)
5. **Multiple sequential loops** -> take the largest: O(n) + O(n) = O(n), O(n) + O(n^2) = O(n^2)

**Figure 3.3** -- Identifying Big-O in practice

````python
# What is the Big-O of each function?

def func_a(n):
    """Just math operations, no loops"""
    return n * 2 + 5     # O(1)

def func_b(data):
    """One loop"""
    total = 0
    for x in data:       # O(n)
        total += x
    return total

def func_c(data):
    """Two sequential loops (not nested)"""
    for x in data:       # O(n)
        print(x)
    for x in data:       # O(n)
        print(x * 2)
    # Total: O(n) + O(n) = O(n)  <-- we take the dominant term

def func_d(data):
    """Nested loops"""
    for x in data:           # O(n)
        for y in data:       # O(n) for each x
            print(x, y)
    # Total: O(n * n) = O(n^2)

print("func_a: O(1)")
print("func_b: O(n)")
print("func_c: O(n) -- two sequential loops, not nested")
print("func_d: O(n^2) -- nested loops")
````

## Part 4: Introduction to Benchmarking

Theory is great, but let's **measure** actual running times!

Python's `time` module lets us measure how long code takes to run.

### The Basic Pattern

```python
import time

start = time.time()        # Record start time
# ... code to measure ...
end = time.time()          # Record end time

elapsed = end - start      # Difference = time taken
print(f"Took {elapsed:.4f} seconds")
```

**Figure 4.1** -- Basic benchmarking with `time.time()`

````python
import time

# Benchmark: summing numbers from 0 to n-1
n = 1_000_000

start = time.time()
total = 0
for i in range(n):
    total += i
end = time.time()

print(f"Sum of 0 to {n-1:,} = {total:,}")
print(f"Time taken: {end - start:.4f} seconds")
````

**Figure 4.2** -- Comparing two approaches to the same problem

````python
import time

n = 10_000_000

# Approach 1: Loop (O(n))
start = time.time()
total_loop = 0
for i in range(n):
    total_loop += i
time_loop = time.time() - start

# Approach 2: Math formula (O(1))
# Sum from 0 to n-1 = n * (n-1) / 2
start = time.time()
total_formula = n * (n - 1) // 2
time_formula = time.time() - start

print(f"Loop result:    {total_loop:,}    Time: {time_loop:.4f} sec")
print(f"Formula result: {total_formula:,}    Time: {time_formula:.6f} sec")
print(f"\nThe formula is {time_loop / max(time_formula, 0.0000001):.0f}x faster!")
print(f"Both give the same answer: {total_loop == total_formula}")
````

This demonstrates why Big-O matters: the O(1) formula gives the same answer instantly, while the O(n) loop takes noticeable time.

## Part 5: Benchmarking at Different Scales

To truly see how algorithms scale, we need to test with **different input sizes** and compare the results.

This is where benchmarking becomes really powerful.

**Figure 5.1** -- Benchmarking a linear operation at different scales

````python
import time

def benchmark_sum(n):
    """Benchmark summing numbers from 0 to n-1 using a loop."""
    start = time.time()
    total = 0
    for i in range(n):
        total += i
    elapsed = time.time() - start
    return elapsed

sizes = [1_000, 10_000, 100_000, 1_000_000]
times = []

print(f"{'n':>12} | {'Time (sec)':>12} | {'Ratio':>8}")
print("-" * 40)

prev_time = None
for n in sizes:
    t = benchmark_sum(n)
    times.append(t)
    ratio = t / prev_time if prev_time else 1.0
    print(f"{n:>12,} | {t:>12.4f} | {ratio:>7.1f}x")
    prev_time = t

print("\nNotice: when n increases 10x, time also increases ~10x.")
print("This confirms O(n) - linear growth!")
````

**Figure 5.2** -- Benchmarking a quadratic operation at different scales

````python
import time

def benchmark_nested(n):
    """Benchmark a nested loop (quadratic)."""
    start = time.time()
    count = 0
    for i in range(n):
        for j in range(n):
            count += 1
    elapsed = time.time() - start
    return elapsed

sizes = [100, 500, 1_000, 2_000, 5_000]
times = []

print(f"{'n':>8} | {'Time (sec)':>12} | {'Ratio':>8}")
print("-" * 35)

prev_time = None
for n in sizes:
    t = benchmark_nested(n)
    times.append(t)
    ratio = t / prev_time if prev_time else 1.0
    print(f"{n:>8,} | {t:>12.4f} | {ratio:>7.1f}x")
    prev_time = t

print("\nNotice: when n doubles, time increases ~4x.")
print("When n increases 5x, time increases ~25x.")
print("This confirms O(n^2) - quadratic growth!")
````

## Part 6: Plotting Benchmark Results

Numbers are good, but **graphs** make patterns much easier to see.

Let's use `matplotlib` to visualize our benchmarks.

**Figure 6.1** -- Plotting linear vs quadratic benchmarks

````python
import time
import matplotlib.pyplot as plt

def time_linear(n):
    start = time.time()
    total = 0
    for i in range(n):
        total += i
    return time.time() - start

def time_quadratic(n):
    start = time.time()
    count = 0
    for i in range(n):
        for j in range(n):
            count += 1
    return time.time() - start

# Test sizes
sizes = [100, 200, 500, 1000, 2000, 3000]
linear_times = [time_linear(n) for n in sizes]
quadratic_times = [time_quadratic(n) for n in sizes]

# Plot
plt.figure(figsize=(10, 5))
plt.plot(sizes, linear_times, 'bo-', label='O(n) - Linear', linewidth=2, markersize=8)
plt.plot(sizes, quadratic_times, 'rs-', label='O(n^2) - Quadratic', linewidth=2, markersize=8)
plt.xlabel('Input Size (n)', fontsize=12)
plt.ylabel('Time (seconds)', fontsize=12)
plt.title('Linear vs Quadratic: Measured Running Times', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

**Figure 6.2** -- Benchmark: Linear search vs Binary search

````python
import time
import matplotlib.pyplot as plt

def linear_search(data, target):
    for i, item in enumerate(data):
        if item == target:
            return i
    return -1

def binary_search(data, target):
    low, high = 0, len(data) - 1
    while low <= high:
        mid = (low + high) // 2
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

sizes = [1_000, 5_000, 10_000, 50_000, 100_000, 500_000, 1_000_000]
linear_times = []
binary_times = []

for n in sizes:
    data = list(range(n))
    target = n - 1  # Worst case: last element
    
    # Time linear search (average of 3 runs)
    t = 0
    for _ in range(3):
        start = time.time()
        linear_search(data, target)
        t += time.time() - start
    linear_times.append(t / 3)
    
    # Time binary search (average of 3 runs)
    t = 0
    for _ in range(3):
        start = time.time()
        binary_search(data, target)
        t += time.time() - start
    binary_times.append(t / 3)

plt.figure(figsize=(10, 5))
plt.plot(sizes, linear_times, 'ro-', label='Linear Search O(n)', linewidth=2)
plt.plot(sizes, binary_times, 'gs-', label='Binary Search O(log n)', linewidth=2)
plt.xlabel('Input Size (n)', fontsize=12)
plt.ylabel('Time (seconds)', fontsize=12)
plt.title('Linear Search vs Binary Search', fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f"At n=1,000,000: Linear is {linear_times[-1]/max(binary_times[-1], 1e-10):.0f}x slower than Binary")
````

## Part 7: Common Errors & Pitfalls

Let's look at common mistakes beginners make when working with algorithms and benchmarking.

**Figure 7.1** -- Mistake: Forgetting that binary search needs a sorted list

````python
def binary_search(data, target):
    low, high = 0, len(data) - 1
    while low <= high:
        mid = (low + high) // 2
        if data[mid] == target:
            return mid
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

# This will give WRONG results!
unsorted = [5, 2, 8, 1, 9, 3]
result = binary_search(unsorted, 8)
print(f"Searching unsorted list for 8: index = {result}")
print(f"But unsorted[{result}] = {unsorted[result] if result >= 0 else 'not found'}")

print()

# Correct: sort first!
sorted_data = sorted(unsorted)
result = binary_search(sorted_data, 8)
print(f"Searching sorted list {sorted_data} for 8: index = {result}")
print(f"sorted_data[{result}] = {sorted_data[result]}")
````

**Figure 7.2** -- Mistake: Off-by-one errors in range

````python
# Common mistake: range(n) goes from 0 to n-1, NOT 0 to n

n = 5
print("range(5) gives:", list(range(5)))       # [0, 1, 2, 3, 4]
print("range(1, 5) gives:", list(range(1, 5)))  # [1, 2, 3, 4]  <-- 5 is excluded!

# If you want 1 to 5 inclusive:
print("range(1, 6) gives:", list(range(1, 6)))  # [1, 2, 3, 4, 5]
````

**Figure 7.3** -- Mistake: Benchmarking too-small inputs

````python
import time

# With tiny input, times are unreliable
start = time.time()
total = sum(range(10))
elapsed = time.time() - start
print(f"n=10: {elapsed:.10f} seconds -- too fast to measure reliably!")

# Better: use larger inputs
start = time.time()
total = sum(range(1_000_000))
elapsed = time.time() - start
print(f"n=1,000,000: {elapsed:.4f} seconds -- much more reliable")

print("\nTip: Use n >= 1000 for meaningful benchmarks.")
print("Tip: Run multiple times and take the average for more accuracy.")
````

**Figure 7.4** -- Handling errors gracefully with try/except

````python
# When working with algorithms, you might encounter these errors:

# 1. IndexError - accessing beyond list bounds
try:
    data = [1, 2, 3]
    print(data[10])  # Index 10 doesn't exist!
except IndexError as e:
    print(f"IndexError: {e}")

# 2. RecursionError - too many recursive calls
try:
    def bad_recursion(n):
        return bad_recursion(n)  # Never stops!
    bad_recursion(5)
except RecursionError:
    print("RecursionError: maximum recursion depth exceeded")

# 3. TypeError - wrong type for an operation
try:
    result = "hello" + 5  # Can't add string and int
except TypeError as e:
    print(f"TypeError: {e}")

print("\nAlways handle potential errors in your algorithms!")
````

## Part 8: Benchmark Lab -- Measure Loop Performance

Now it's your turn to run a full benchmark experiment!

We'll measure how long it takes to **find an element** in a list using linear search at different scales, then plot the results.

> **🔮 Predict first, then measure. Does reality match your prediction?** Before running each benchmark, write down what you expect to see. Then run it and compare.

**Figure 8.1** -- Complete benchmark: linear search at increasing scales

````python
import time
import matplotlib.pyplot as plt

def linear_search(data, target):
    """Search for target in data, return index or -1."""
    for i in range(len(data)):
        if data[i] == target:
            return i
    return -1

# Benchmark at different sizes
sizes = [1_000, 5_000, 10_000, 50_000, 100_000, 500_000, 1_000_000]
avg_times = []
num_runs = 5  # Average over 5 runs for reliability

print(f"{'Size':>12} | {'Avg Time (ms)':>14} | {'Operations/sec':>15}")
print("-" * 50)

for n in sizes:
    data = list(range(n))
    target = n - 1  # Worst case: search for last element
    
    total_time = 0
    for _ in range(num_runs):
        start = time.time()
        linear_search(data, target)
        total_time += time.time() - start
    
    avg = total_time / num_runs
    avg_times.append(avg)
    ops_per_sec = n / avg if avg > 0 else 0
    print(f"{n:>12,} | {avg*1000:>13.2f}ms | {ops_per_sec:>14,.0f}")

# Plot the results
plt.figure(figsize=(10, 5))
plt.plot(sizes, [t * 1000 for t in avg_times], 'bo-', linewidth=2, markersize=8)
plt.xlabel('List Size (n)', fontsize=12)
plt.ylabel('Average Time (ms)', fontsize=12)
plt.title('Linear Search: Time vs Input Size', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nObservation: The graph should be roughly a straight line,")
print("confirming that linear search is O(n).")
````

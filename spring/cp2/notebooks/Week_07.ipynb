## Part 1: Why NumPy? Speed and Simplicity

NumPy (Numerical Python) is the foundation of scientific computing in Python. Two key advantages:

| Feature | Python List | NumPy Array |
|---|---|---|
| Storage | Objects scattered in memory | Contiguous typed block |
| Speed | Interpreted loop per element | Compiled C loop |
| Syntax | Manual loops | Vectorized one-liners |

````python
import numpy as np
import time

# Python list approach
size = 1_000_000
py_list = list(range(size))

start = time.time()
result_list = [x * 2 for x in py_list]
py_time = time.time() - start

# NumPy approach
arr = np.arange(size)

start = time.time()
result_arr = arr * 2
np_time = time.time() - start

print(f"Python list : {py_time:.4f} s")
print(f"NumPy array : {np_time:.4f} s")
print(f"Speedup     : {py_time / np_time:.1f}x")
````

**Figure 1.1** — Timing comparison: Python list comprehension vs. NumPy vectorized multiplication on 1 million elements.

````python
# Quick sanity check
import numpy as np
print(f"NumPy version: {np.__version__}")
print(f"Type of np.array([1,2,3]): {type(np.array([1,2,3]))}")
print(f"1 + np.array([10, 20, 30]) = {1 + np.array([10, 20, 30])}")
````

**Figure 1.2** — NumPy version check and a first taste of vectorized addition.

## Part 4: Vectorized Arithmetic

NumPy performs operations **element-wise** without explicit loops:

| Operation | Syntax | Equivalent |
|---|---|---|
| Add | `a + b` | `np.add(a, b)` |
| Subtract | `a - b` | `np.subtract(a, b)` |
| Multiply | `a * b` | `np.multiply(a, b)` |
| Divide | `a / b` | `np.divide(a, b)` |
| Power | `a ** n` | `np.power(a, n)` |
| Floor div | `a // b` | `np.floor_divide(a, b)` |

**Broadcasting**: when shapes differ, NumPy stretches the smaller array automatically (e.g., array + scalar).

````python
import numpy as np

voltaj = np.array([220, 110, 230, 120, 240])
akim   = np.array([5, 10, 4.5, 8, 6])

guc = voltaj * akim          # element-wise
print("Power (W):", guc)

# Scalar broadcasting
voltaj_kv = voltaj ** 2
print("V squared:", voltaj_kv)
````

**Figure 4.1** — Element-wise multiplication and scalar broadcasting.

````python
import numpy as np

# Celsius to Fahrenheit — vectorized
celsius = np.array([0, 20, 37, 100])
fahrenheit = celsius * 9 / 5 + 32
print("Celsius   :", celsius)
print("Fahrenheit:", fahrenheit)
````

**Figure 4.2** — Temperature conversion using vectorized arithmetic.

````python
import numpy as np

# Broadcasting with 2-D
matris = np.array([[1, 2, 3],
                   [4, 5, 6]])
satir  = np.array([10, 20, 30])   # shape (3,)

print("Matrix + row vector:\n", matris + satir)
````

**Figure 4.3** — Broadcasting a 1-D array across rows of a 2-D array.

## Part 5: Aggregation Functions

| Function | Returns |
|---|---|
| `np.sum(a)` / `a.sum()` | Total |
| `np.mean(a)` | Arithmetic mean |
| `np.std(a)` | Standard deviation |
| `np.min(a)`, `np.max(a)` | Extremes |
| `np.argmin(a)`, `np.argmax(a)` | Index of extreme |
| `np.cumsum(a)` | Cumulative sum |

````python
import numpy as np
np.random.seed(7)

notlar = np.random.randint(30, 101, size=15)
print("Grades     :", notlar)
print(f"Sum        : {notlar.sum()}")
print(f"Mean       : {notlar.mean():.1f}")
print(f"Std Dev    : {notlar.std():.1f}")
print(f"Min        : {notlar.min()} at index {notlar.argmin()}")
print(f"Max        : {notlar.max()} at index {notlar.argmax()}")
````

**Figure 5.1** — Common aggregation functions on a grades array.

````python
import numpy as np

satislar = np.array([120, 340, 200, 450, 310, 280])
kumulatif = np.cumsum(satislar)
print("Monthly sales :", satislar)
print("Cumulative    :", kumulatif)
````

**Figure 5.2** — Cumulative sum tracking running totals.

````python
import numpy as np

# axis parameter with 2-D
veri = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("Sum all     :", veri.sum())
print("Sum axis=0  :", veri.sum(axis=0))   # column sums
print("Sum axis=1  :", veri.sum(axis=1))   # row sums
print("Mean axis=1 :", veri.mean(axis=1))
````

**Figure 5.3** — Using the `axis` parameter for row-wise and column-wise aggregation.

## Part 6: Boolean Indexing

Boolean indexing lets you filter arrays with conditions:

```python
mask = arr > 50       # boolean array
filtered = arr[mask]  # only True elements
```

`np.where(condition, x, y)` returns `x` where True, `y` where False.

````python
import numpy as np
np.random.seed(42)

notlar = np.random.randint(30, 101, size=12)
print("All grades:", notlar)

# Create a boolean mask
gecen = notlar >= 60
print("Mask      :", gecen)
print("Passed    :", notlar[gecen])
print(f"Pass rate : {gecen.sum()}/{len(notlar)} = {gecen.mean()*100:.0f}%")
````

**Figure 6.1** — Boolean mask to filter passing grades.

````python
import numpy as np

sicaklik = np.array([38.1, 36.5, 39.2, 37.0, 40.1, 36.8])
hastalar = ["Elif", "Mehmet", "Zeynep", "Ahmet", "Burak", "Selin"]

# Multiple conditions
ates = sicaklik > 38.0
print("Fever patients:")
for i in np.where(ates)[0]:
    print(f"  {hastalar[i]}: {sicaklik[i]}C")
````

**Figure 6.2** — Using `np.where()` to find indices matching a condition.

````python
import numpy as np

notlar = np.array([45, 72, 88, 55, 93, 38, 67])

# np.where with replacement values
harf = np.where(notlar >= 60, "Pass", "Fail")
print("Grades:", notlar)
print("Result:", harf)
````

**Figure 6.3** — `np.where()` with two replacement values.

## Part 7: Reshaping and Slicing

| Operation | Example | Result shape |
|---|---|---|
| `reshape(r, c)` | `np.arange(6).reshape(2,3)` | (2, 3) |
| 1-D slice | `a[2:5]` | (3,) |

## Part 8: Loop vs. NumPy — Performance Comparison

The final piece of NumPy mastery is **knowing when it matters**. Let us benchmark three common operations.

````python
import numpy as np
import time

def benchmark(label, func, *args, repeats=5):
    times = []
    for _ in range(repeats):
        start = time.time()
        func(*args)
        times.append(time.time() - start)
    avg = sum(times) / len(times)
    print(f"{label:30s} : {avg*1000:8.2f} ms")
    return avg

N = 1_000_000
py_list = list(range(N))
np_arr  = np.arange(N)

# Sum
t1 = benchmark("Python sum(list)", sum, py_list)
t2 = benchmark("NumPy  np.sum(arr)", np.sum, np_arr)
print(f"{'Speedup':30s} : {t1/t2:.1f}x\n")

# Squaring
t3 = benchmark("Python [x**2 for x]", lambda l: [x**2 for x in l], py_list)
t4 = benchmark("NumPy  arr**2", lambda a: a**2, np_arr)
print(f"{'Speedup':30s} : {t3/t4:.1f}x")
````

**Figure 8.1** — Benchmark: Python loops vs. NumPy for sum and squaring.

````python
import numpy as np
import time

N = 500_000
np.random.seed(0)

py_list = [float(x) for x in np.random.rand(N)]
np_arr  = np.array(py_list)

# Filtering — boolean indexing vs list comprehension
start = time.time()
py_filtered = [x for x in py_list if x > 0.5]
py_time = time.time() - start

start = time.time()
np_filtered = np_arr[np_arr > 0.5]
np_time = time.time() - start

print(f"Python filter : {py_time*1000:.2f} ms  ({len(py_filtered)} items)")
print(f"NumPy  filter : {np_time*1000:.2f} ms  ({len(np_filtered)} items)")
print(f"Speedup       : {py_time/np_time:.1f}x")
````

**Figure 8.2** — Benchmark: Boolean indexing vs. list comprehension for filtering.

````python
import numpy as np
import time

# Dot product
N = 1_000_000
a_list = list(range(N))
b_list = list(range(N))
a_np = np.arange(N, dtype=float)
b_np = np.arange(N, dtype=float)

start = time.time()
dot_py = sum(x*y for x, y in zip(a_list, b_list))
py_t = time.time() - start

start = time.time()
dot_np = np.dot(a_np, b_np)
np_t = time.time() - start

print(f"Python dot : {py_t*1000:.2f} ms  result={dot_py:.0f}")
print(f"NumPy  dot : {np_t*1000:.2f} ms  result={dot_np:.0f}")
print(f"Speedup    : {py_t/np_t:.1f}x")
````

**Figure 8.3** — Benchmark: Dot product with zip vs. `np.dot()`.

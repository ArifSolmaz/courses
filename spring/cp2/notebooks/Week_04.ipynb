## Part 1: Dict Recap — Creation, Access, Update, Delete

Dictionaries store **key-value pairs**. Keys must be immutable (strings, numbers, tuples). Values can be anything.

| Operation | Syntax | Description |
|-----------|--------|-------------|
| Create empty | `d = {}` or `d = dict()` | Empty dictionary |
| Create with data | `d = {"a": 1, "b": 2}` | Literal syntax |
| Access | `d["a"]` | Raises `KeyError` if missing |
| Safe access | `d.get("a", 0)` | Returns default if missing |
| Update / Add | `d["c"] = 3` | Adds or overwrites |
| Delete | `del d["a"]` | Raises `KeyError` if missing |
| Membership | `"a" in d` | `True` / `False` — O(1) |
| Length | `len(d)` | Number of key-value pairs |

**Figure 1.1** — Creating dictionaries with `{}` and `dict()`

```python
# Creating dictionaries — two equivalent ways
sensor_a = {"id": "S01", "location": "Lab-A", "temp": 23.5}
sensor_b = dict(id="S02", location="Lab-B", temp=21.0)

print("sensor_a:", sensor_a)
print("sensor_b:", sensor_b)
print("Type:", type(sensor_a))
```

**Figure 1.2** — Safe access with `.get()` vs `[]`

```python
student = {"name": "Ahmet Yilmaz", "grade": 85, "dept": "EE"}

# Direct access — works when key exists
print("Name:", student["name"])

# Safe access — returns default when key is missing
print("Phone:", student.get("phone", "N/A"))

# Attempting student["phone"] would raise KeyError
# Always prefer .get() when a key might be absent
```

**Figure 1.3** — Updating and deleting entries

```python
readings = {"temp": 22.1, "humidity": 55, "pressure": 1013}
print("Before:", readings)

# Update existing key
readings["temp"] = 23.4

# Add new key
readings["wind_speed"] = 12.5

# Delete a key
del readings["pressure"]

# Check membership
print("Has 'temp'?", "temp" in readings)
print("Has 'pressure'?", "pressure" in readings)
print("After:", readings)
```

## Part 2: Iterating Dicts — Keys, Values, Items

Python dictionaries provide three iteration views:

| Method | Returns | Example |
|--------|---------|---------|
| `d.keys()` | All keys | `dict_keys(['a', 'b'])` |
| `d.values()` | All values | `dict_values([1, 2])` |
| `d.items()` | Key-value tuples | `dict_items([('a',1), ('b',2)])` |

You can also sort keys with `sorted(d)` and iterate in order.

**Figure 2.1** — Looping with `.items()`

```python
grades = {"Mehmet": 78, "Ayse": 92, "Burak": 65, "Elif": 88}

print("Student Report Card")
print("-" * 30)
for name, grade in grades.items():
    status = "PASS" if grade >= 70 else "FAIL"
    print(f"  {name:<10} {grade:>3}  [{status}]")
```

**Figure 2.2** — Sorted iteration

```python
sensor_data = {"S03": 24.1, "S01": 22.5, "S02": 23.8, "S05": 25.0, "S04": 21.9}

print("Sensors sorted by ID:")
for sid in sorted(sensor_data):
    print(f"  {sid}: {sensor_data[sid]:.1f} °C")

print("\nSensors sorted by temperature (ascending):")
for sid in sorted(sensor_data, key=sensor_data.get):
    print(f"  {sid}: {sensor_data[sid]:.1f} °C")
```

**Figure 2.3** — Extracting keys and values as lists

```python
inventory = {"resistor": 450, "capacitor": 120, "LED": 300, "transistor": 85}

all_parts = list(inventory.keys())
all_counts = list(inventory.values())

print("Parts:", all_parts)
print("Counts:", all_counts)
print("Total items:", sum(all_counts))
print("Most stocked:", max(inventory, key=inventory.get))
```

## Part 3: Counter Pattern — Frequency Counting

The **counter pattern** tallies how many times each value appears.

| Step | Code | Purpose |
|------|------|---------|
| 1. Init | `counts = {}` | Empty accumulator |
| 2. Loop | `for item in data:` | Visit every element |
| 3. Count | `counts[item] = counts.get(item, 0) + 1` | Increment safely |

This single line — `counts.get(item, 0) + 1` — is the heart of frequency counting. The `.get(item, 0)` returns `0` when the key is new, so the first occurrence becomes `0 + 1 = 1`.

**Figure 3.1** — Counting word frequencies

```python
text = "sensor sensor motor valve sensor motor sensor valve valve motor"
words = text.split()

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

print("Word frequencies:")
for word, freq in sorted(counts.items(), key=lambda x: -x[1]):
    bar = "█" * freq
    print(f"  {word:<8} {freq}  {bar}")
```

**Figure 3.2** — Counting sensor statuses

```python
statuses = ["OK", "WARN", "OK", "OK", "ERROR", "WARN", "OK",
            "ERROR", "OK", "WARN", "WARN", "OK", "OK", "ERROR"]

status_counts = {}
for s in statuses:
    status_counts[s] = status_counts.get(s, 0) + 1

total = len(statuses)
print("Sensor Status Summary")
print("-" * 35)
for status in ["OK", "WARN", "ERROR"]:
    count = status_counts.get(status, 0)
    pct = count / total * 100
    print(f"  {status:<6} {count:>3} readings  ({pct:5.1f}%)")
```

**Figure 3.3** — Counting grade letter distribution

```python
scores = [88, 72, 95, 60, 45, 78, 82, 91, 55, 67, 73, 85, 99, 40, 76]

def score_to_letter(s):
    if s >= 90: return "A"
    if s >= 80: return "B"
    if s >= 70: return "C"
    if s >= 60: return "D"
    return "F"

letter_counts = {}
for score in scores:
    letter = score_to_letter(score)
    letter_counts[letter] = letter_counts.get(letter, 0) + 1

print("Grade Distribution:")
for letter in ["A", "B", "C", "D", "F"]:
    n = letter_counts.get(letter, 0)
    print(f"  {letter}: {'█' * n} ({n})")
```

## Part 4: Histogram Pattern — Bucketing Numeric Data

The **histogram pattern** groups continuous numbers into discrete bins.

**Algorithm:**
1. Define bin edges: `[0, 10, 20, 30, 40]`
2. For each value, find which bin it falls into
3. Increment that bin's count

This is essential for summarizing sensor data, test scores, or any continuous measurement into meaningful categories.

**Figure 4.1** — Basic histogram with fixed bins

```python
temperatures = [18.2, 22.5, 25.1, 19.8, 30.2, 27.6, 21.0, 24.4,
                28.9, 31.5, 20.3, 26.7, 23.1, 29.0, 17.5, 33.2]

bins = {"<20": 0, "20-24": 0, "25-29": 0, "30+": 0}

for t in temperatures:
    if t < 20:
        bins["<20"] += 1
    elif t < 25:
        bins["20-24"] += 1
    elif t < 30:
        bins["25-29"] += 1
    else:
        bins["30+"] += 1

print("Temperature Histogram (°C)")
print("-" * 35)
for label, count in bins.items():
    bar = "█" * count
    print(f"  {label:>6}: {bar} ({count})")
```

**Figure 4.2** — Reusable `build_histogram()` function

```python
def build_histogram(data, edges):
    """Bucket data into bins defined by edges.
    edges = [0, 20, 25, 30, 40] creates bins: <20, 20-24, 25-29, 30-39, 40+
    """
    bins = {}
    for i in range(len(edges) - 1):
        label = f"{edges[i]}-{edges[i+1]-1}"
        bins[label] = 0
    bins[f"{edges[-1]}+"] = 0

    for val in data:
        placed = False
        for i in range(len(edges) - 1):
            if edges[i] <= val < edges[i+1]:
                label = f"{edges[i]}-{edges[i+1]-1}"
                bins[label] += 1
                placed = True
                break
        if not placed:
            bins[f"{edges[-1]}+"] += 1
    return bins

scores = [45, 52, 61, 67, 73, 78, 80, 85, 88, 92, 95, 99, 55, 70, 82]
hist = build_histogram(scores, [0, 50, 60, 70, 80, 90])
print("Score Distribution:")
for label, count in hist.items():
    print(f"  {label:>6}: {'█' * count} ({count})")
```

**Figure 4.3** — Histogram of signal strengths

```python
import random
random.seed(42)
signals = [random.gauss(70, 15) for _ in range(50)]

# Quick histogram using integer division trick
bucket_size = 10
hist = {}
for s in signals:
    bucket = int(s // bucket_size) * bucket_size
    label = f"{bucket}-{bucket + bucket_size - 1}"
    hist[label] = hist.get(label, 0) + 1

print("Signal Strength Histogram (dBm)")
print("-" * 40)
for label in sorted(hist.keys(), key=lambda x: int(x.split("-")[0])):
    count = hist[label]
    print(f"  {label:>7}: {'█' * count} ({count})")
```

## Part 5: Lookup Table Pattern — O(1) Mapping

A **lookup table** replaces long `if/elif` chains with a dictionary.

| Approach | Speed | Lines of Code | Maintainability |
|----------|-------|---------------|-----------------|
| `if/elif` chain | O(n) | grows with categories | Hard to update |
| Dictionary lookup | O(1) | constant | Easy — just add a key |

This pattern is especially useful for category codes, unit conversions, and status mappings.

**Figure 5.1** — Replacing if/elif with a lookup table

```python
# ❌ The if/elif way (hard to maintain)
def get_dept_name_old(code):
    if code == "EE": return "Electrical Engineering"
    elif code == "ME": return "Mechanical Engineering"
    elif code == "CE": return "Civil Engineering"
    elif code == "CS": return "Computer Science"
    elif code == "IE": return "Industrial Engineering"
    else: return "Unknown"

# ✅ The lookup table way (clean and fast)
DEPT_NAMES = {
    "EE": "Electrical Engineering",
    "ME": "Mechanical Engineering",
    "CE": "Civil Engineering",
    "CS": "Computer Science",
    "IE": "Industrial Engineering",
}

def get_dept_name(code):
    return DEPT_NAMES.get(code, "Unknown")

# Test both
for c in ["EE", "CS", "XX"]:
    print(f"  {c} -> {get_dept_name(c)}")
```

**Figure 5.2** — Unit conversion lookup

```python
UNIT_TO_METERS = {
    "mm": 0.001,
    "cm": 0.01,
    "m":  1.0,
    "km": 1000.0,
    "in": 0.0254,
    "ft": 0.3048,
    "yd": 0.9144,
}

def convert_to_meters(value, unit):
    factor = UNIT_TO_METERS.get(unit)
    if factor is None:
        return None, f"Unknown unit: {unit}"
    return value * factor, "OK"

tests = [(150, "cm"), (2.5, "km"), (6, "ft"), (10, "xx")]
for val, unit in tests:
    result, msg = convert_to_meters(val, unit)
    if result is not None:
        print(f"  {val} {unit} = {result:.4f} m")
    else:
        print(f"  {val} {unit} -> {msg}")
```

**Figure 5.3** — Status code lookup for sensor pipeline

```python
STATUS_INFO = {
    0: {"label": "OK",       "color": "green",  "action": "none"},
    1: {"label": "WARNING",  "color": "yellow", "action": "log"},
    2: {"label": "CRITICAL", "color": "red",    "action": "alert"},
    3: {"label": "OFFLINE",  "color": "gray",   "action": "restart"},
}

sensor_readings = [
    {"id": "S01", "value": 23.5, "status": 0},
    {"id": "S02", "value": 45.2, "status": 2},
    {"id": "S03", "value":  0.0, "status": 3},
    {"id": "S04", "value": 28.1, "status": 1},
]

print("Sensor Dashboard")
print("-" * 50)
for r in sensor_readings:
    info = STATUS_INFO.get(r["status"], {"label": "UNKNOWN", "action": "check"})
    print(f"  {r['id']}  value={r['value']:>6.1f}  [{info['label']:<8}]  -> {info['action']}")
```

## Part 7: Dict Comprehensions — Concise Creation

Dict comprehensions follow the syntax: `{key_expr: value_expr for item in iterable [if condition]}`

They let you create and transform dictionaries in a single, readable line.

**Figure 7.1** — Basic dict comprehension

```python
# Square numbers lookup
squares = {n: n**2 for n in range(1, 11)}
print("Squares:", squares)

# From two lists
names = ["Ahmet", "Mehmet", "Ayse", "Burak"]
scores = [85, 72, 91, 68]
grade_book = {name: score for name, score in zip(names, scores)}
print("Grades:", grade_book)
```

**Figure 7.2** — Filtering with dict comprehension

```python
readings = {"S01": 23.5, "S02": 45.2, "S03": 19.8,
            "S04": 38.7, "S05": 22.1, "S06": 41.0}

# Keep only sensors with temp > 30
hot_sensors = {k: v for k, v in readings.items() if v > 30}
print("Hot sensors:", hot_sensors)

# Keep only sensors with temp in normal range (18-30)
normal = {k: v for k, v in readings.items() if 18 <= v <= 30}
print("Normal sensors:", normal)
```

**Figure 7.3** — Transforming values

```python
temps_celsius = {"Lab-A": 22.5, "Lab-B": 25.0, "Roof": 31.2, "Basement": 18.8}

# Convert all to Fahrenheit
temps_fahrenheit = {loc: round(c * 9/5 + 32, 1) for loc, c in temps_celsius.items()}
print("Celsius:   ", temps_celsius)
print("Fahrenheit:", temps_fahrenheit)

# Invert a dictionary (swap keys and values)
inverted = {v: k for k, v in temps_celsius.items()}
print("Inverted:  ", inverted)
```

## Part 8: Combining Patterns — Counter + Histogram + Lookup Pipeline

Real data pipelines often combine multiple dictionary patterns in sequence:

1. **Lookup** — translate raw codes to meaningful labels
2. **Counter** — count occurrences of each label
3. **Histogram** — bucket numeric values into ranges

Let's build a complete pipeline using factory sensor data.

**Figure 8.1** — Complete sensor data pipeline

```python
import random
random.seed(2024)

# Simulated raw sensor data
raw_data = []
for i in range(40):
    raw_data.append({
        "sensor_id": f"S{random.randint(1,5):02d}",
        "status_code": random.choice([0, 0, 0, 1, 1, 2]),
        "temperature": round(random.gauss(25, 5), 1)
    })

# LOOKUP: translate status codes
STATUS_LABELS = {0: "OK", 1: "WARN", 2: "CRITICAL"}

# COUNTER: count statuses
status_counts = {}
for row in raw_data:
    label = STATUS_LABELS.get(row["status_code"], "UNKNOWN")
    status_counts[label] = status_counts.get(label, 0) + 1

print("=== Status Counts ===")
for label, count in sorted(status_counts.items()):
    print(f"  {label:<10} {count:>3}  {'█' * count}")

# HISTOGRAM: bucket temperatures
temp_bins = {"<20": 0, "20-24": 0, "25-29": 0, "30+": 0}
for row in raw_data:
    t = row["temperature"]
    if t < 20:   temp_bins["<20"] += 1
    elif t < 25: temp_bins["20-24"] += 1
    elif t < 30: temp_bins["25-29"] += 1
    else:        temp_bins["30+"] += 1

print("\n=== Temperature Distribution ===")
for label, count in temp_bins.items():
    print(f"  {label:>6}: {'█' * count} ({count})")
```

**Figure 8.2** — Per-sensor summary using nested dicts

```python
# Build per-sensor summary
sensor_summary = {}
for row in raw_data:
    sid = row["sensor_id"]
    if sid not in sensor_summary:
        sensor_summary[sid] = {"temps": [], "status_counts": {}}
    sensor_summary[sid]["temps"].append(row["temperature"])
    label = STATUS_LABELS.get(row["status_code"], "UNKNOWN")
    sc = sensor_summary[sid]["status_counts"]
    sc[label] = sc.get(label, 0) + 1

print("=== Per-Sensor Summary ===")
print(f"  {'ID':<5} {'Avg°C':>6} {'Min':>5} {'Max':>5}  Status Breakdown")
print("  " + "-" * 55)
for sid in sorted(sensor_summary):
    info = sensor_summary[sid]
    temps = info["temps"]
    avg = sum(temps) / len(temps)
    parts = [f"{k}:{v}" for k, v in sorted(info["status_counts"].items())]
    print(f"  {sid:<5} {avg:>6.1f} {min(temps):>5.1f} {max(temps):>5.1f}  {', '.join(parts)}")
```

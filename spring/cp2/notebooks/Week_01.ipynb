## Part 3: CP1 Review — for / while Loops, range(), enumerate()

`for` loops iterate over a sequence; `while` loops repeat while a condition is `True`.

| Pattern | Use case |
|---------|----------|
| `for x in sequence:` | Known collection |
| `for i in range(n):` | Known count |
| `for i, x in enumerate(seq):` | Index + value |
| `while condition:` | Unknown iterations |

````python
# for loop — iterating sensor readings
readings = [22.1, 23.4, 21.8, 24.0, 22.7]

total = 0
for r in readings:
    total += r
average = total / len(readings)

print(f"Readings : {readings}")
print(f"Average  : {average:.2f}°C")
````

**Figure 3.1** — Summing sensor readings with a `for` loop.

````python
# range() patterns
print("range(5)       :", list(range(5)))
print("range(2, 8)    :", list(range(2, 8)))
print("range(0, 20, 5):", list(range(0, 20, 5)))

# Generating sample timestamps
for t in range(0, 25, 5):
    print(f"  t = {t:2d} s → reading due")
````

**Figure 3.2** — Using `range()` with start, stop, step for timed sampling.

````python
# enumerate() — index + value
cities = ["Istanbul", "Ankara", "Izmir", "Bursa"]
temps  = [15, 10, 18, 12]

for i, (city, temp) in enumerate(zip(cities, temps)):
    print(f"  [{i}] {city:10s} : {temp}°C")
````

**Figure 3.3** — `enumerate()` with `zip()` for parallel lists.

````python
# while loop — convergence check (Newton's method for sqrt)
target = 50.0
guess = target / 2
tolerance = 1e-9
iterations = 0

while abs(guess * guess - target) > tolerance:
    guess = (guess + target / guess) / 2
    iterations += 1

print(f"sqrt({target}) ≈ {guess:.12f}")
print(f"Converged in {iterations} iterations")
````

**Figure 3.4** — A `while` loop for Newton’s square-root method.

## Part 5: CP1 Review — Functions, Parameters, Return Values

Functions let you **name** a piece of logic and **reuse** it.

```python
def function_name(param1, param2):
    """Brief description."""
    # body
    return result
```

Key ideas: parameters are *inputs*, `return` sends back the *output*, and the docstring explains *what* the function does.

````python
# Simple function — Mehmet's voltage divider calculator
def voltage_divider(vin, r1, r2):
    """Calculate output voltage of a voltage divider."""
    vout = vin * r2 / (r1 + r2)
    return vout

result = voltage_divider(12.0, 10_000, 4_700)
print(f"Vout = {result:.2f} V")
````

**Figure 5.1** — A function encapsulating a voltage divider formula.

````python
# Multiple return values
def analyze_readings(readings):
    """Return min, max, and mean of a list of numbers."""
    mn = min(readings)
    mx = max(readings)
    avg = sum(readings) / len(readings)
    return mn, mx, avg

data = [22.1, 23.4, 21.8, 24.0, 22.7, 25.1]
lo, hi, mean = analyze_readings(data)
print(f"Min: {lo}, Max: {hi}, Mean: {mean:.2f}")
````

**Figure 5.2** — Returning multiple values as a tuple.

````python
# Functions calling functions — building blocks
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

def classify_temp_f(f):
    if f > 100:
        return "HOT"
    elif f > 60:
        return "WARM"
    else:
        return "COLD"

temp_c = 38.5
temp_f = celsius_to_fahrenheit(temp_c)
label = classify_temp_f(temp_f)
print(f"{temp_c}°C = {temp_f:.1f}°F → {label}")
````

**Figure 5.3** — Composing small functions for a temperature pipeline.

## Part 6: What is Refactoring? The Four Moves

**Refactoring** is changing the *structure* of code without changing its *behavior*.

Why refactor?
- Makes code easier to **read** (you read code 10x more than you write it)
- Makes code easier to **debug** (clear names reveal intent)
- Makes code easier to **extend** (small functions compose well)

### The Four Refactoring Moves

| Move | What you do | Before | After |
|------|-------------|--------|-------|
| **Rename** | Give a better name | `x = 9.81` | `gravity = 9.81` |
| **Extract** | Pull code into a function | 10 lines inline | `calc_force(m, a)` |
| **Simplify** | Remove duplication/nesting | nested ifs | flat elif chain |
| **Document** | Add docstrings & comments | bare code | explained intent |

````python
# ❌ BEFORE refactoring — Zeynep's first draft
x = [23, 45, 12, 67, 34, 89, 21]
s = 0
for i in x:
    s = s + i
a = s / len(x)
print(a)
````

**Figure 6.1** — Working but unclear code (poor names, no context).

````python
# ✅ AFTER refactoring — all four moves applied

def calculate_average(values):
    """Return the arithmetic mean of a list of numbers."""  # Document
    total = sum(values)                                      # Simplify (built-in)
    return total / len(values)

# Rename: descriptive names
exam_scores = [23, 45, 12, 67, 34, 89, 21]
average_score = calculate_average(exam_scores)               # Extract

print(f"Average exam score: {average_score:.1f}")
````

**Figure 6.2** — The same logic after Rename, Extract, Simplify, Document.

````python
# Another example — Ali's messy bridge load checker
# ❌ Before
w = 1500
if w > 2000:
    print("no")
elif w > 1000:
    print("maybe")
else:
    print("yes")
````

**Figure 6.3** — Unclear variable names and print outputs.

````python
# ✅ After — refactored
MAX_LOAD_KG = 2000
WARNING_THRESHOLD_KG = 1000

def check_bridge_load(load_kg):
    """Classify bridge load status."""
    if load_kg > MAX_LOAD_KG:
        return "OVERLOADED"
    elif load_kg > WARNING_THRESHOLD_KG:
        return "WARNING"
    else:
        return "SAFE"

vehicle_weight = 1500  # kg
status = check_bridge_load(vehicle_weight)
print(f"Bridge load: {vehicle_weight} kg → {status}")
````

**Figure 6.4** — Refactored with constants, a function, and clear names.

## Part 7: Code Smells — Magic Numbers, Poor Names, Long Functions, No Comments

A **code smell** is a surface indication that something *might* be wrong. It does not mean the code is broken — it means the code is hard to maintain.

### Common Code Smells

| Smell | Description | Fix |
|-------|-------------|-----|
| **Magic number** | Unexplained literal like `9.81` | Create a named constant |
| **Poor name** | `x`, `tmp`, `data2` | Rename to reveal intent |
| **Long function** | 30+ lines doing many things | Extract into smaller functions |
| **No comments** | Reader must guess your intent | Add docstrings and key comments |
| **Copy-paste code** | Same block repeated | Extract a function |

````python
# ❌ Smell: Magic numbers everywhere
def calc(m, h):
    return m * 9.81 * h + 0.5 * 1.225 * m * 3.5**2

result = calc(75, 10)
print(result)
````

**Figure 7.1** — Magic numbers: what are 9.81, 1.225, 3.5?

````python
# ✅ Fix: Named constants
GRAVITY = 9.81              # m/s^2
AIR_DENSITY = 1.225         # kg/m^3 at sea level
WIND_SPEED = 3.5            # m/s (average)

def total_load_on_structure(mass_kg, height_m):
    """Calculate gravitational + wind load on a structure."""
    gravity_load = mass_kg * GRAVITY * height_m
    wind_load = 0.5 * AIR_DENSITY * mass_kg * WIND_SPEED ** 2
    return gravity_load + wind_load

load = total_load_on_structure(mass_kg=75, height_m=10)
print(f"Total load: {load:.1f} N")
````

**Figure 7.2** — Named constants make the physics transparent.

````python
# ❌ Smell: Long function doing everything
def process(data):
    # validate
    clean = []
    for d in data:
        if d is not None and d > 0:
            clean.append(d)
    # statistics
    avg = sum(clean) / len(clean)
    mx = max(clean)
    mn = min(clean)
    # report
    print(f"Count: {len(clean)}")
    print(f"Avg: {avg:.2f}")
    print(f"Range: {mn} - {mx}")
    return avg

process([10, None, 25, -3, 30, 18, None, 22])
````

**Figure 7.3** — One function doing validation, stats, and reporting.

````python
# ✅ Fix: Extract into small functions
def clean_readings(raw_data):
    """Remove None and non-positive values."""
    return [d for d in raw_data if d is not None and d > 0]

def compute_stats(values):
    """Return dict with count, mean, min, max."""
    return {
        "count": len(values),
        "mean": sum(values) / len(values),
        "min": min(values),
        "max": max(values),
    }

def print_report(stats):
    """Display statistics in a readable format."""
    print(f"Count: {stats['count']}")
    print(f"Avg  : {stats['mean']:.2f}")
    print(f"Range: {stats['min']} – {stats['max']}")

# Pipeline: clean → compute → report
raw = [10, None, 25, -3, 30, 18, None, 22]
cleaned = clean_readings(raw)
stats = compute_stats(cleaned)
print_report(stats)
````

**Figure 7.4** — Each function has a single responsibility.

## Part 8: The CONFIG Cell Pattern

In notebooks, you often have constants scattered across many cells. The **CONFIG cell pattern** collects all constants at the top of the notebook in one clearly labeled cell.

**Benefits:**
- Change a parameter in *one place* and re-run
- Easy to see all assumptions at a glance
- No more hunting for magic numbers

**Convention:** Use `ALL_CAPS` for constants that do not change during a run.

````python
# ═══════════════════════════════════════════
# CONFIG — change parameters here only
# ═══════════════════════════════════════════
SENSOR_COUNT = 5
SAMPLE_RATE_HZ = 10
DURATION_SEC = 60
TEMP_THRESHOLD = 75.0     # °C — alarm above this
OUTPUT_FILE = "readings.csv"
# ═══════════════════════════════════════════
````

**Figure 8.1** — A CONFIG cell with all tunable parameters.

````python
# Using CONFIG values downstream — Fatma's data pipeline
import random
random.seed(42)

total_samples = SENSOR_COUNT * SAMPLE_RATE_HZ * DURATION_SEC
print(f"Total samples to collect: {total_samples:,}")

# Simulated readings
readings = [random.gauss(70, 10) for _ in range(SENSOR_COUNT)]
alarms = [r for r in readings if r > TEMP_THRESHOLD]

print(f"Current readings: {[f'{r:.1f}' for r in readings]}")
print(f"Alarms (>{TEMP_THRESHOLD}°C): {len(alarms)}")
````

**Figure 8.2** — Downstream cells reference CONFIG constants, never raw numbers.

````python
# Full mini-example: refactored sensor report using CONFIG
# (All constants come from the CONFIG cell above)

def generate_sensor_data(n_sensors, mean_temp=70, std_temp=10):
    """Simulate temperature readings for n sensors."""
    import random
    return [round(random.gauss(mean_temp, std_temp), 1) for _ in range(n_sensors)]

def find_alarms(readings, threshold):
    """Return readings above the threshold."""
    return [r for r in readings if r > threshold]

def print_sensor_report(readings, alarms, threshold):
    """Print a formatted sensor status report."""
    print("=" * 40)
    print("SENSOR STATUS REPORT")
    print("=" * 40)
    for i, r in enumerate(readings):
        flag = " ⚠️" if r > threshold else ""
        print(f"  Sensor {i+1}: {r:6.1f}°C{flag}")
    print("-" * 40)
    print(f"  Alarms: {len(alarms)} / {len(readings)}")
    print("=" * 40)

# Pipeline
data = generate_sensor_data(SENSOR_COUNT)
alarm_list = find_alarms(data, TEMP_THRESHOLD)
print_sensor_report(data, alarm_list, TEMP_THRESHOLD)
````

**Figure 8.3** — A complete mini-pipeline using CONFIG + extracted functions.

## 1. What is Capacitance?

A **capacitor** is a device that stores electric charge (and therefore energy) by maintaining a potential difference between two conductors.

**Capacitance** is defined as the ratio of the charge stored to the voltage across the capacitor:

$$C = \frac{Q}{V}$$

where:
- $C$ = capacitance (in Farads, F)
- $Q$ = charge on one plate (in Coulombs, C)
- $V$ = voltage (potential difference) between the plates (in Volts, V)

**Key insight:** Capacitance is a geometric property -- it depends only on the shape, size, and materials of the capacitor, not on $Q$ or $V$.

A Farad is a very large unit. Typical capacitors range from picofarads (pF = $10^{-12}$ F) to millifarads (mF = $10^{-3}$ F).

## 2. The Parallel Plate Capacitor

The simplest and most common capacitor has two parallel conducting plates of area $A$, separated by a distance $d$.

From **Gauss's law** (Week 03), the electric field between two parallel plates with surface charge density $\pm\sigma$ is:

$$E = \frac{\sigma}{\varepsilon_0} = \frac{Q}{\varepsilon_0 A}$$

The voltage across the plates is:

$$V = E \cdot d = \frac{Qd}{\varepsilon_0 A}$$

Therefore the capacitance is:

$$\boxed{C = \frac{\varepsilon_0 A}{d}}$$

**Larger plates** $\rightarrow$ more capacitance. **Closer plates** $\rightarrow$ more capacitance.

### Interactive Demo 1: Parallel Plate Capacitor Explorer

Adjust the plate area, separation distance, and dielectric constant to see how capacitance, electric field, and voltage change.

```python
def parallel_plate_capacitor(A_cm2=100.0, d_mm=1.0, kappa=1.0, Q_nC=10.0):
    """
    Interactive parallel plate capacitor.
    Left: diagram of capacitor with field lines.
    Right: computed values.
    """
    eps0 = 8.854e-12
    A = A_cm2 * 1e-4       # cm^2 to m^2
    d = d_mm * 1e-3         # mm to m
    Q = Q_nC * 1e-9         # nC to C
    
    C = kappa * eps0 * A / d
    V = Q / C
    E = V / d
    U = 0.5 * Q * V
    u_density = 0.5 * kappa * eps0 * E**2  # energy density
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # --- Left: Capacitor diagram ---
    ax1.set_xlim(-3, 3)
    ax1.set_ylim(-2.5, 2.5)
    ax1.set_aspect('equal')
    
    plate_half = min(1.8, 0.5 + np.sqrt(A_cm2) / 15)
    gap_half = min(1.0, 0.2 + d_mm / 5)
    
    # Plates
    ax1.fill_between([-plate_half, plate_half], gap_half, gap_half + 0.15,
                      color='#1565C0', alpha=0.9)
    ax1.fill_between([-plate_half, plate_half], -gap_half - 0.15, -gap_half,
                      color='#C62828', alpha=0.9)
    
    # Charge labels
    for x in np.linspace(-plate_half + 0.3, plate_half - 0.3, 5):
        ax1.text(x, gap_half + 0.22, '+', fontsize=12, ha='center', color='blue', fontweight='bold')
        ax1.text(x, -gap_half - 0.27, '-', fontsize=14, ha='center', color='red', fontweight='bold')
    
    # Dielectric fill
    if kappa > 1.0:
        ax1.fill_between([-plate_half, plate_half], -gap_half, gap_half,
                          color='#FFF9C4', alpha=0.6)
        ax1.text(0, 0, f'Dielectric\n$\\kappa$ = {kappa:.1f}', fontsize=11,
                 ha='center', va='center', style='italic',
                 bbox=dict(boxstyle='round', facecolor='#FFF176', alpha=0.8))
    
    # Electric field arrows
    n_arrows = 6
    for x in np.linspace(-plate_half + 0.3, plate_half - 0.3, n_arrows):
        arrow_alpha = min(1.0, 0.3 + 0.7 * E / 1e6) if E < 1e7 else 1.0
        ax1.annotate('', xy=(x, -gap_half + 0.05), xytext=(x, gap_half - 0.05),
                     arrowprops=dict(arrowstyle='->', color='green', lw=1.5, alpha=min(arrow_alpha, 1)))
    
    ax1.text(plate_half + 0.2, gap_half + 0.07, '+Q', fontsize=12, color='blue')
    ax1.text(plate_half + 0.2, -gap_half - 0.07, '-Q', fontsize=12, color='red')
    
    # Dimension annotations
    ax1.annotate('', xy=(plate_half + 0.8, gap_half), xytext=(plate_half + 0.8, -gap_half),
                 arrowprops=dict(arrowstyle='<->', color='black', lw=1.5))
    ax1.text(plate_half + 1.0, 0, f'd = {d_mm:.1f} mm', fontsize=10, va='center')
    
    ax1.set_title('Parallel Plate Capacitor', fontsize=14)
    ax1.axis('off')
    
    # --- Right: Computed values ---
    ax2.axis('off')
    
    info_lines = [
        ('INPUTS', '', 'black', 14, 'bold'),
        (f'  Plate area A', f'{A_cm2:.0f} cm\u00b2 = {A*1e4:.1f} cm\u00b2', '#333', 12, 'normal'),
        (f'  Separation d', f'{d_mm:.1f} mm', '#333', 12, 'normal'),
        (f'  Dielectric \u03ba', f'{kappa:.1f}', '#333', 12, 'normal'),
        (f'  Charge Q', f'{Q_nC:.1f} nC', '#333', 12, 'normal'),
        ('', '', 'black', 8, 'normal'),
        ('RESULTS', '', 'black', 14, 'bold'),
        (f'  Capacitance C', f'{C*1e12:.2f} pF', '#1565C0', 13, 'bold'),
        (f'  Voltage V', f'{V:.2f} V', '#C62828', 13, 'bold'),
        (f'  Electric field E', f'{E:.2e} V/m', '#2E7D32', 13, 'bold'),
        (f'  Energy stored U', f'{U*1e9:.3f} nJ', '#E65100', 13, 'bold'),
        (f'  Energy density u', f'{u_density:.3e} J/m\u00b3', '#6A1B9A', 12, 'normal'),
    ]
    
    y_pos = 0.95
    for label, value, color, size, weight in info_lines:
        if value:
            ax2.text(0.05, y_pos, f'{label}:', fontsize=size, color='#555',
                     transform=ax2.transAxes, fontweight=weight)
            ax2.text(0.55, y_pos, value, fontsize=size, color=color,
                     transform=ax2.transAxes, fontweight=weight)
        else:
            ax2.text(0.05, y_pos, label, fontsize=size, color=color,
                     transform=ax2.transAxes, fontweight=weight)
        y_pos -= 0.08
    
    # Formula reminder
    ax2.text(0.05, 0.02, r'$C = \kappa \varepsilon_0 A / d$    |    $U = \frac{1}{2}CV^2 = \frac{1}{2}QV$',
             fontsize=12, transform=ax2.transAxes,
             bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
    
    plt.tight_layout()
    plt.show()

interact(parallel_plate_capacitor,
         A_cm2=FloatSlider(value=100, min=10, max=500, step=10, description='Area (cm^2):', style={'description_width': 'initial'}),
         d_mm=FloatSlider(value=1.0, min=0.1, max=10.0, step=0.1, description='d (mm):', style={'description_width': 'initial'}),
         kappa=FloatSlider(value=1.0, min=1.0, max=10.0, step=0.1, description='Dielectric k:', style={'description_width': 'initial'}),
         Q_nC=FloatSlider(value=10.0, min=1.0, max=100.0, step=1.0, description='Q (nC):', style={'description_width': 'initial'}));
```

## 3. Energy Stored in a Capacitor

To charge a capacitor, we must do work moving charge from one plate to the other against the growing electric field. The total energy stored is:

$$U = \frac{1}{2}CV^2 = \frac{1}{2}QV = \frac{Q^2}{2C}$$

We can also think of this energy as being stored in the **electric field** itself, with energy density:

$$u = \frac{U}{\text{Volume}} = \frac{1}{2}\kappa \varepsilon_0 E^2 = \frac{1}{2}\varepsilon E^2$$

where $\varepsilon = \kappa \varepsilon_0$ is the permittivity of the dielectric material. This result is general -- it applies to any electric field, not just in capacitors.

### Interactive Demo 2: Energy Stored in a Capacitor

Visualize how the stored energy depends on capacitance and voltage. The bar chart shows all three equivalent energy expressions.

```python
def energy_in_capacitor(C_pF=100.0, V_volts=10.0):
    """
    Interactive visualization of energy stored in a capacitor.
    Shows the energy as a function of voltage (left) and as bar chart (right).
    """
    C = C_pF * 1e-12  # pF to F
    Q = C * V_volts
    
    U1 = 0.5 * C * V_volts**2
    U2 = 0.5 * Q * V_volts
    U3 = Q**2 / (2 * C)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))
    
    # --- Left: U vs V curve ---
    V_range = np.linspace(0, V_volts * 1.5, 200)
    U_range = 0.5 * C * V_range**2
    
    ax1.plot(V_range, U_range * 1e9, 'b-', linewidth=2, label=r'$U = \frac{1}{2}CV^2$')
    ax1.fill_between(V_range[V_range <= V_volts], 0,
                      0.5 * C * V_range[V_range <= V_volts]**2 * 1e9,
                      alpha=0.3, color='blue', label='Energy stored')
    ax1.plot(V_volts, U1 * 1e9, 'ro', markersize=12, zorder=5,
             label=f'V = {V_volts:.1f} V, U = {U1*1e9:.2f} nJ')
    ax1.axvline(x=V_volts, color='red', linestyle=':', alpha=0.5)
    
    ax1.set_xlabel('Voltage V (V)', fontsize=12)
    ax1.set_ylabel('Energy U (nJ)', fontsize=12)
    ax1.set_title(f'Energy vs Voltage (C = {C_pF:.0f} pF)', fontsize=13)
    ax1.legend(fontsize=10)
    ax1.grid(True, alpha=0.3)
    ax1.set_xlim(0, V_volts * 1.5)
    ax1.set_ylim(0, None)
    
    # --- Right: Three equivalent expressions ---
    expressions = [r'$\frac{1}{2}CV^2$', r'$\frac{1}{2}QV$', r'$\frac{Q^2}{2C}$']
    values = [U1 * 1e9, U2 * 1e9, U3 * 1e9]
    colors = ['#1E88E5', '#43A047', '#E53935']
    
    bars = ax2.bar(expressions, values, color=colors, width=0.5, edgecolor='black', alpha=0.85)
    
    for bar, val in zip(bars, values):
        ax2.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + max(values) * 0.02,
                 f'{val:.3f} nJ', ha='center', fontsize=12, fontweight='bold')
    
    ax2.set_ylabel('Energy (nJ)', fontsize=12)
    ax2.set_title('Three Equivalent Energy Formulas\n(all give the same answer!)', fontsize=13)
    ax2.set_ylim(0, max(values) * 1.2)
    ax2.grid(True, alpha=0.3, axis='y')
    
    fig.suptitle(f'C = {C_pF:.0f} pF,  V = {V_volts:.1f} V,  Q = {Q*1e9:.2f} nC',
                 fontsize=13, y=1.02)
    plt.tight_layout()
    plt.show()

interact(energy_in_capacitor,
         C_pF=FloatSlider(value=100, min=10, max=1000, step=10, description='C (pF):', style={'description_width': 'initial'}),
         V_volts=FloatSlider(value=10.0, min=1.0, max=50.0, step=1.0, description='V (volts):', style={'description_width': 'initial'}));
```

## 4. Capacitors in Series and Parallel

### Parallel Combination
Capacitors connected in parallel share the **same voltage**. The total charge is the sum:

$$C_{\text{eq}} = C_1 + C_2 + C_3 + \ldots$$

Parallel capacitors **add directly** (like resistors in series).

### Series Combination
Capacitors connected in series carry the **same charge**. The voltages add:

$$\frac{1}{C_{\text{eq}}} = \frac{1}{C_1} + \frac{1}{C_2} + \frac{1}{C_3} + \ldots$$

Series capacitors **add reciprocally** (like resistors in parallel).

**Memory aid:** Capacitors combine in the *opposite* way from resistors!

### Interactive Demo 3: Series and Parallel Capacitor Calculator

Adjust the values of up to three capacitors and see the equivalent capacitance for both series and parallel connections.

```python
def series_parallel_calculator(C1=10.0, C2=20.0, C3=30.0, n_caps=2):
    """
    Interactive series/parallel capacitor calculator.
    Shows circuit diagrams and computed equivalent capacitance.
    """
    caps = [C1, C2, C3][:n_caps]
    
    # Calculate equivalents
    C_parallel = sum(caps)
    C_series = 1.0 / sum(1.0 / c for c in caps)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    
    # --- Left: Parallel ---
    ax = axes[0]
    ax.set_xlim(-1, 5)
    ax.set_ylim(-0.5, n_caps + 0.5)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title(f'Parallel: $C_{{eq}}$ = {C_parallel:.1f} pF', fontsize=14, color='#1565C0')
    
    for i, c in enumerate(caps):
        y = n_caps - 1 - i
        # Wire to capacitor
        ax.plot([0, 1.5], [y, y], 'k-', linewidth=1.5)
        # Capacitor symbol (two parallel lines)
        ax.plot([1.5, 1.5], [y - 0.2, y + 0.2], 'b-', linewidth=3)
        ax.plot([2.0, 2.0], [y - 0.2, y + 0.2], 'r-', linewidth=3)
        # Wire from capacitor
        ax.plot([2.0, 3.5], [y, y], 'k-', linewidth=1.5)
        # Label
        ax.text(1.75, y + 0.35, f'$C_{i+1}$={c:.0f}', fontsize=10, ha='center')
    
    # Connecting wires (vertical)
    ax.plot([0, 0], [-0.0, n_caps - 1], 'k-', linewidth=2)
    ax.plot([3.5, 3.5], [0, n_caps - 1], 'k-', linewidth=2)
    # Terminals
    ax.plot([-0.5, 0], [n_caps / 2 - 0.5, n_caps / 2 - 0.5], 'k-', linewidth=2)
    ax.plot([3.5, 4.0], [n_caps / 2 - 0.5, n_caps / 2 - 0.5], 'k-', linewidth=2)
    ax.text(-0.7, n_caps / 2 - 0.5, '+', fontsize=14, color='blue', fontweight='bold')
    ax.text(4.1, n_caps / 2 - 0.5, '-', fontsize=16, color='red', fontweight='bold')
    
    # --- Right: Series ---
    ax = axes[1]
    total_width = n_caps * 1.5 + 0.5
    ax.set_xlim(-0.5, total_width + 0.5)
    ax.set_ylim(-1.5, 1.5)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title(f'Series: $C_{{eq}}$ = {C_series:.2f} pF', fontsize=14, color='#C62828')
    
    x_pos = 0
    for i, c in enumerate(caps):
        # Wire
        ax.plot([x_pos, x_pos + 0.5], [0, 0], 'k-', linewidth=1.5)
        # Capacitor symbol
        ax.plot([x_pos + 0.5, x_pos + 0.5], [-0.3, 0.3], 'b-', linewidth=3)
        ax.plot([x_pos + 0.8, x_pos + 0.8], [-0.3, 0.3], 'r-', linewidth=3)
        # Wire after
        ax.plot([x_pos + 0.8, x_pos + 1.5], [0, 0], 'k-', linewidth=1.5)
        # Label
        ax.text(x_pos + 0.65, 0.5, f'$C_{i+1}$={c:.0f}', fontsize=10, ha='center')
        x_pos += 1.5
    
    ax.text(-0.7, 0, '+', fontsize=14, color='blue', fontweight='bold')
    ax.text(x_pos + 0.1, 0, '-', fontsize=16, color='red', fontweight='bold')
    
    # Summary text
    formula_par = ' + '.join([f'{c:.0f}' for c in caps])
    formula_ser = ' + '.join([f'1/{c:.0f}' for c in caps])
    
    fig.text(0.25, -0.02, f'$C_{{eq}}$ = {formula_par} = {C_parallel:.1f} pF',
             ha='center', fontsize=12, color='#1565C0')
    fig.text(0.75, -0.02, f'$1/C_{{eq}}$ = {formula_ser} = {1/C_series:.4f}  =>  $C_{{eq}}$ = {C_series:.2f} pF',
             ha='center', fontsize=12, color='#C62828')
    
    plt.tight_layout()
    plt.show()

interact(series_parallel_calculator,
         C1=FloatSlider(value=10, min=1, max=100, step=1, description='C1 (pF):', style={'description_width': 'initial'}),
         C2=FloatSlider(value=20, min=1, max=100, step=1, description='C2 (pF):', style={'description_width': 'initial'}),
         C3=FloatSlider(value=30, min=1, max=100, step=1, description='C3 (pF):', style={'description_width': 'initial'}),
         n_caps=IntSlider(value=2, min=2, max=3, step=1, description='# of caps:', style={'description_width': 'initial'}));
```

## 5. Animation: Charging a Capacitor

When a capacitor is connected to a battery, charge builds up on the plates over time. The voltage across the capacitor increases until it equals the battery voltage. Watch the charge accumulate and the voltage rise!

```python
def create_charging_animation():
    """
    Animation: Charging a parallel plate capacitor.
    Shows charges building up on plates and voltage rising.
    """
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5.5))
    
    # --- Left: Capacitor diagram ---
    ax1.set_xlim(-3, 3)
    ax1.set_ylim(-2.5, 2.5)
    ax1.set_aspect('equal')
    ax1.set_title('Capacitor Charging', fontsize=14)
    
    # Static elements: plates
    ax1.fill_between([-1.5, 1.5], 0.5, 0.65, color='gray', alpha=0.8)
    ax1.fill_between([-1.5, 1.5], -0.65, -0.5, color='gray', alpha=0.8)
    
    # Battery symbol
    ax1.plot([-2.5, -2.5], [-0.3, 0.3], 'k-', linewidth=4)  # short line (negative)
    ax1.plot([-2.2, -2.2], [-0.5, 0.5], 'k-', linewidth=2)  # long line (positive)
    ax1.text(-2.35, 0.7, 'Battery', fontsize=10, ha='center')
    ax1.text(-2.05, 0.3, '+', fontsize=10, color='blue')
    ax1.text(-2.7, -0.1, '-', fontsize=12, color='red')
    
    # Wires
    ax1.plot([-2.2, -2.2, 0, 0], [0.5, 1.5, 1.5, 0.65], 'k-', linewidth=1.5)
    ax1.plot([-2.5, -2.5, 0, 0], [-0.3, -1.5, -1.5, -0.65], 'k-', linewidth=1.5)
    
    # Dynamic elements
    charge_texts_top = []
    charge_texts_bot = []
    n_charge_positions = 8
    for i in range(n_charge_positions):
        x = -1.2 + i * 2.4 / (n_charge_positions - 1)
        t_top = ax1.text(x, 0.75, '', fontsize=10, ha='center', va='center',
                          color='blue', fontweight='bold')
        t_bot = ax1.text(x, -0.75, '', fontsize=12, ha='center', va='center',
                          color='red', fontweight='bold')
        charge_texts_top.append(t_top)
        charge_texts_bot.append(t_bot)
    
    # Field arrows (will appear gradually)
    field_arrows = []
    for x in np.linspace(-1.0, 1.0, 5):
        arrow = ax1.annotate('', xy=(x, -0.45), xytext=(x, 0.45),
                              arrowprops=dict(arrowstyle='->', color='green', lw=0, alpha=0))
        field_arrows.append(arrow)
    
    time_text = ax1.text(-2.8, -2.2, '', fontsize=11)
    
    # --- Right: Voltage and charge vs time ---
    ax2.set_xlim(0, 5)
    ax2.set_ylim(0, 1.1)
    ax2.set_xlabel('Time (RC units)', fontsize=12)
    ax2.set_ylabel('Fraction of maximum', fontsize=12)
    ax2.set_title('Charge and Voltage vs Time', fontsize=14)
    ax2.grid(True, alpha=0.3)
    
    t_full = np.linspace(0, 5, 200)
    ax2.plot(t_full, 1 - np.exp(-t_full), 'b--', alpha=0.3, linewidth=1)
    
    line_q, = ax2.plot([], [], 'b-', linewidth=2.5, label='Q(t) / Q_max')
    line_v, = ax2.plot([], [], 'r-', linewidth=2.5, label='V(t) / V_max')
    marker, = ax2.plot([], [], 'ko', markersize=8, zorder=5)
    ax2.axhline(y=1.0, color='gray', linestyle=':', alpha=0.5)
    ax2.text(4.2, 1.02, 'Maximum', fontsize=10, color='gray')
    ax2.legend(fontsize=11)
    
    def init():
        line_q.set_data([], [])
        line_v.set_data([], [])
        marker.set_data([], [])
        time_text.set_text('')
        for t in charge_texts_top + charge_texts_bot:
            t.set_text('')
        return [line_q, line_v, marker, time_text] + charge_texts_top + charge_texts_bot
    
    def animate(frame):
        t = frame / 12.0  # maps frames to RC time constants
        frac = 1 - np.exp(-t)  # charging curve
        
        # Update plot
        t_arr = np.linspace(0, t, max(2, frame))
        q_arr = 1 - np.exp(-t_arr)
        line_q.set_data(t_arr, q_arr)
        line_v.set_data(t_arr, q_arr)  # V proportional to Q
        marker.set_data([t], [frac])
        
        # Update charges on plates
        n_visible = int(frac * n_charge_positions)
        for i in range(n_charge_positions):
            if i < n_visible:
                charge_texts_top[i].set_text('+')
                charge_texts_bot[i].set_text('-')
            else:
                charge_texts_top[i].set_text('')
                charge_texts_bot[i].set_text('')
        
        # Update field arrows
        for arrow in field_arrows:
            arrow.arrow_patch.set_mutation_scale(10)
            arrow.arrow_patch.set_linewidth(frac * 2)
            arrow.arrow_patch.set_alpha(frac * 0.8)
        
        time_text.set_text(f't = {t:.1f} RC,  Q/Q_max = {frac:.1%}')
        
        return [line_q, line_v, marker, time_text] + charge_texts_top + charge_texts_bot
    
    anim = FuncAnimation(fig, animate, init_func=init, frames=60, interval=100, blit=False)
    plt.close(fig)
    return anim

anim = create_charging_animation()
HTML(anim.to_jshtml())
```

## 6. Dielectrics

A **dielectric** is an insulating material placed between the plates of a capacitor. It increases the capacitance by a factor $\kappa$ (the **dielectric constant**):

$$C = \kappa \, C_0 = \kappa \frac{\varepsilon_0 A}{d}$$

**How it works:** The electric field partially polarizes the dielectric molecules, creating an opposing internal field that reduces the total field. This allows more charge to be stored for the same voltage.

**Effects of inserting a dielectric (with battery connected):**
- Voltage $V$: stays the same (battery maintains it)
- Capacitance $C$: increases by factor $\kappa$
- Charge $Q = CV$: increases by factor $\kappa$
- Electric field $E = V/d$: stays the same
- Energy $U = \frac{1}{2}CV^2$: increases by factor $\kappa$

**Effects of inserting a dielectric (battery disconnected):**
- Charge $Q$: stays the same (nowhere to go)
- Capacitance $C$: increases by factor $\kappa$
- Voltage $V = Q/C$: decreases by factor $\kappa$
- Electric field $E = V/d$: decreases by factor $\kappa$
- Energy $U = Q^2/(2C)$: decreases by factor $\kappa$

| Material | Dielectric constant $\kappa$ |
|----------|---------------------------|
| Vacuum | 1.000 |
| Air | 1.001 |
| Paper | 3.7 |
| Glass | 5 - 10 |
| Water | 80 |

### Interactive Demo 4: Effect of Inserting a Dielectric

Compare capacitor properties before and after inserting a dielectric, in both scenarios: battery connected and battery disconnected.

```python
def dielectric_effect(kappa=3.0, C0_pF=50.0, V0=12.0, scenario='Battery connected'):
    """
    Compare before/after inserting a dielectric.
    Two scenarios: battery connected vs disconnected.
    """
    C0 = C0_pF * 1e-12
    Q0 = C0 * V0
    U0 = 0.5 * C0 * V0**2
    
    C_new = kappa * C0
    
    if scenario == 'Battery connected':
        V_new = V0  # battery holds voltage constant
        Q_new = C_new * V_new
        U_new = 0.5 * C_new * V_new**2
        subtitle = 'Battery keeps V constant'
    else:  # Battery disconnected
        Q_new = Q0  # charge is trapped
        V_new = Q_new / C_new
        U_new = Q_new**2 / (2 * C_new)
        subtitle = 'Charge Q stays constant'
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    properties = ['Capacitance', 'Voltage', 'Charge', 'Energy']
    before = [C0 * 1e12, V0, Q0 * 1e9, U0 * 1e9]
    after = [C_new * 1e12, V_new, Q_new * 1e9, U_new * 1e9]
    units = ['pF', 'V', 'nC', 'nJ']
    ratios = [a / b if b != 0 else 0 for a, b in zip(after, before)]
    
    # Bar comparison
    x = np.arange(len(properties))
    width = 0.35
    
    axes[0].bar(x - width / 2, before, width, color='#90CAF9', edgecolor='black', label='Before (vacuum)')
    axes[0].bar(x + width / 2, after, width, color='#FFF176', edgecolor='black', label=f'After (\u03ba={kappa:.1f})')
    axes[0].set_xticks(x)
    axes[0].set_xticklabels([f'{p}\n({u})' for p, u in zip(properties, units)], fontsize=10)
    axes[0].set_ylabel('Value', fontsize=12)
    axes[0].set_title(f'Before vs After\n({subtitle})', fontsize=13)
    axes[0].legend(fontsize=10)
    axes[0].grid(True, alpha=0.3, axis='y')
    
    # Ratio chart
    colors_ratio = ['#4CAF50' if r >= 1 else '#F44336' for r in ratios]
    axes[1].barh(properties, ratios, color=colors_ratio, edgecolor='black', height=0.5)
    axes[1].axvline(x=1, color='black', linewidth=1.5, linestyle='--')
    for i, (r, p) in enumerate(zip(ratios, properties)):
        axes[1].text(r + 0.05, i, f'{r:.2f}x', va='center', fontsize=12, fontweight='bold')
    axes[1].set_xlabel('Ratio (After / Before)', fontsize=12)
    axes[1].set_title('Change Factors', fontsize=13)
    axes[1].grid(True, alpha=0.3, axis='x')
    
    # Diagram
    ax = axes[2]
    ax.set_xlim(-2, 2)
    ax.set_ylim(-2, 2)
    ax.axis('off')
    ax.set_title(f'Dielectric Inserted (\u03ba = {kappa:.1f})', fontsize=13)
    
    # Plates
    ax.fill_between([-1.2, 1.2], 0.6, 0.72, color='#1565C0', alpha=0.9)
    ax.fill_between([-1.2, 1.2], -0.72, -0.6, color='#C62828', alpha=0.9)
    # Dielectric
    ax.fill_between([-1.2, 1.2], -0.6, 0.6, color='#FFF9C4', alpha=0.7)
    ax.text(0, 0, f'\u03ba = {kappa:.1f}', fontsize=14, ha='center', va='center', fontweight='bold')
    
    # Field arrows (thinner if field is reduced)
    if scenario == 'Battery connected':
        arrow_lw = 2
    else:
        arrow_lw = max(0.5, 2 / kappa)
    for x_pos in np.linspace(-0.8, 0.8, 4):
        ax.annotate('', xy=(x_pos, -0.55), xytext=(x_pos, 0.55),
                     arrowprops=dict(arrowstyle='->', color='green', lw=arrow_lw))
    
    ax.text(0, -1.5, f'C = {C_new*1e12:.1f} pF\nV = {V_new:.2f} V\nQ = {Q_new*1e9:.2f} nC\nU = {U_new*1e9:.3f} nJ',
            fontsize=11, ha='center', bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))
    
    plt.tight_layout()
    plt.show()

interact(dielectric_effect,
         kappa=FloatSlider(value=3.0, min=1.0, max=10.0, step=0.5, description='Dielectric k:', style={'description_width': 'initial'}),
         C0_pF=FloatSlider(value=50.0, min=10, max=200, step=10, description='C0 (pF):', style={'description_width': 'initial'}),
         V0=FloatSlider(value=12.0, min=1.0, max=50.0, step=1.0, description='V0 (volts):', style={'description_width': 'initial'}),
         scenario=widgets.Dropdown(options=['Battery connected', 'Battery disconnected'],
                                    value='Battery connected', description='Scenario:', style={'description_width': 'initial'}));
```

## 7. Worked Example 1: Capacitance of a Parallel Plate Capacitor

**Problem:** A parallel plate capacitor has plates of area $A = 200$ cm$^2$ separated by $d = 2.0$ mm. The space between the plates is filled with paper ($\kappa = 3.7$).

(a) What is the capacitance?  
(b) If a voltage of 100 V is applied, how much charge is stored?  
(c) What is the electric field between the plates?

```python
# Worked Example 1: Parallel plate capacitor with dielectric
print("=" * 60)
print("WORKED EXAMPLE 1: Parallel Plate Capacitor with Paper")
print("=" * 60)

eps0 = 8.854e-12
A = 200e-4        # 200 cm^2 in m^2
d = 2.0e-3         # 2.0 mm in m
kappa = 3.7        # paper
V = 100.0          # volts

print(f"\nGiven:")
print(f"  A = 200 cm^2 = {A:.4f} m^2")
print(f"  d = 2.0 mm = {d:.4f} m")
print(f"  kappa = {kappa} (paper)")
print(f"  V = {V} V")

# Part (a)
C = kappa * eps0 * A / d
print(f"\n--- Part (a): Capacitance ---")
print(f"  C = kappa * eps0 * A / d")
print(f"  C = {kappa} * {eps0:.3e} * {A:.4f} / {d:.4f}")
print(f"  C = {C:.4e} F = {C*1e12:.1f} pF = {C*1e9:.3f} nF")

# Without dielectric for comparison
C0 = eps0 * A / d
print(f"  (Without paper: C0 = {C0*1e12:.1f} pF -- paper multiplied it by {kappa}x)")

# Part (b)
Q = C * V
print(f"\n--- Part (b): Charge stored ---")
print(f"  Q = C * V = {C:.4e} * {V}")
print(f"  Q = {Q:.4e} C = {Q*1e9:.1f} nC")

# Part (c)
E = V / d
print(f"\n--- Part (c): Electric field ---")
print(f"  E = V / d = {V} / {d:.4f}")
print(f"  E = {E:.1f} V/m = {E/1e3:.1f} kV/m")
```

## 8. Worked Example 2: Energy Stored and Capacitor Combinations

**Problem:** Two capacitors, $C_1 = 4.0 \; \mu$F and $C_2 = 6.0 \; \mu$F, are connected to a 12 V battery.

(a) Find the equivalent capacitance and total energy stored when connected in **parallel**.  
(b) Find the equivalent capacitance and total energy stored when connected in **series**.  
(c) Compare the results.

```python
# Worked Example 2: Series and parallel capacitor combinations
print("=" * 60)
print("WORKED EXAMPLE 2: Capacitor Combinations")
print("=" * 60)

C1 = 4.0e-6  # 4.0 uF
C2 = 6.0e-6  # 6.0 uF
V = 12.0

print(f"\nGiven: C1 = {C1*1e6:.1f} uF, C2 = {C2*1e6:.1f} uF, V = {V} V")

# Part (a): Parallel
C_par = C1 + C2
U_par = 0.5 * C_par * V**2
Q_par = C_par * V

print(f"\n--- Part (a): PARALLEL connection ---")
print(f"  C_eq = C1 + C2 = {C1*1e6:.1f} + {C2*1e6:.1f} = {C_par*1e6:.1f} uF")
print(f"  Each capacitor has V = {V} V across it")
print(f"  Total charge: Q = C_eq * V = {Q_par*1e6:.1f} uC")
print(f"  Total energy: U = (1/2)*C_eq*V^2 = 0.5 * {C_par*1e6:.1f}e-6 * {V}^2")
print(f"  U = {U_par*1e6:.1f} uJ = {U_par*1e3:.3f} mJ")

# Part (b): Series
C_ser = 1 / (1/C1 + 1/C2)
U_ser = 0.5 * C_ser * V**2
Q_ser = C_ser * V
V1_ser = Q_ser / C1
V2_ser = Q_ser / C2

print(f"\n--- Part (b): SERIES connection ---")
print(f"  1/C_eq = 1/C1 + 1/C2 = 1/{C1*1e6:.1f} + 1/{C2*1e6:.1f} = {1/C_ser/1e-6:.4f} (1/uF)")
print(f"  C_eq = {C_ser*1e6:.2f} uF")
print(f"  Each capacitor has Q = {Q_ser*1e6:.2f} uC")
print(f"  V1 = Q/C1 = {V1_ser:.2f} V,  V2 = Q/C2 = {V2_ser:.2f} V")
print(f"  Check: V1 + V2 = {V1_ser + V2_ser:.2f} V (should be {V} V)")
print(f"  Total energy: U = (1/2)*C_eq*V^2 = {U_ser*1e6:.1f} uJ = {U_ser*1e3:.3f} mJ")

# Part (c): Comparison
print(f"\n--- Part (c): Comparison ---")
print(f"  Parallel: C_eq = {C_par*1e6:.1f} uF,  U = {U_par*1e3:.3f} mJ")
print(f"  Series:   C_eq = {C_ser*1e6:.2f} uF,  U = {U_ser*1e3:.3f} mJ")
print(f"  Ratio: U_parallel / U_series = {U_par/U_ser:.2f}")
print(f"  Parallel stores {U_par/U_ser:.1f}x more energy!")
```

## 9. Interactive: Energy Density Visualization

Explore how the energy density in the electric field between capacitor plates depends on the field strength. Remember: $u = \frac{1}{2}\kappa \varepsilon_0 E^2$.

```python
def energy_density_viz(V_volts=100.0, d_mm=2.0, kappa=1.0):
    """
    Visualize energy density between capacitor plates.
    Shows color-coded field strength and energy density.
    """
    eps0 = 8.854e-12
    d = d_mm * 1e-3
    E = V_volts / d
    u = 0.5 * kappa * eps0 * E**2
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5.5))
    
    # --- Left: Energy density as color map ---
    # Create a 2D grid representing the space between plates
    x = np.linspace(-1, 1, 100)
    y = np.linspace(-1, 1, 100)
    X, Y = np.meshgrid(x, y)
    
    # Energy density is uniform between the plates
    U_grid = np.where((np.abs(Y) < 0.7) & (np.abs(X) < 0.8), u, 0)
    
    im = ax1.imshow(U_grid, extent=[-1, 1, -1, 1], cmap='hot', origin='lower',
                     vmin=0, vmax=max(u, 1e-6))
    plt.colorbar(im, ax=ax1, label='Energy density (J/m^3)')
    
    # Draw plates
    ax1.fill_between([-0.8, 0.8], 0.7, 0.8, color='blue', alpha=0.8)
    ax1.fill_between([-0.8, 0.8], -0.8, -0.7, color='red', alpha=0.8)
    ax1.text(0, 0.85, '+', fontsize=14, ha='center', color='blue', fontweight='bold')
    ax1.text(0, -0.88, '-', fontsize=16, ha='center', color='red', fontweight='bold')
    
    ax1.set_title(f'Energy Density Between Plates\nu = {u:.2e} J/m\u00b3', fontsize=13)
    ax1.set_xlabel('x (arbitrary)')
    ax1.set_ylabel('y (arbitrary)')
    
    # --- Right: Energy density vs E-field ---
    E_range = np.linspace(0, E * 1.5, 200)
    u_range = 0.5 * kappa * eps0 * E_range**2
    
    ax2.plot(E_range / 1e3, u_range, 'b-', linewidth=2, label=r'$u = \frac{1}{2}\kappa\varepsilon_0 E^2$')
    ax2.plot(E / 1e3, u, 'ro', markersize=12, zorder=5,
             label=f'E = {E/1e3:.1f} kV/m\nu = {u:.2e} J/m\u00b3')
    ax2.fill_between(E_range[E_range <= E] / 1e3, 0,
                      0.5 * kappa * eps0 * E_range[E_range <= E]**2, alpha=0.2, color='blue')
    
    ax2.set_xlabel('Electric field E (kV/m)', fontsize=12)
    ax2.set_ylabel('Energy density u (J/m\u00b3)', fontsize=12)
    ax2.set_title('Energy Density vs Electric Field', fontsize=13)
    ax2.legend(fontsize=10)
    ax2.grid(True, alpha=0.3)
    ax2.set_xlim(0, E * 1.5 / 1e3)
    ax2.set_ylim(0, None)
    
    plt.tight_layout()
    plt.show()

interact(energy_density_viz,
         V_volts=FloatSlider(value=100, min=10, max=500, step=10, description='V (volts):', style={'description_width': 'initial'}),
         d_mm=FloatSlider(value=2.0, min=0.5, max=10.0, step=0.5, description='d (mm):', style={'description_width': 'initial'}),
         kappa=FloatSlider(value=1.0, min=1.0, max=10.0, step=0.5, description='Dielectric k:', style={'description_width': 'initial'}));
```

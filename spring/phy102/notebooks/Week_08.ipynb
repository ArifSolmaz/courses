## 1. The Biot-Savart Law

The **Biot-Savart law** tells us how a small segment of current-carrying wire creates a magnetic field at a point in space:

$$d\vec{B} = \frac{\mu_0}{4\pi} \frac{I \, d\vec{l} \times \hat{r}}{r^2}$$

where:
- $\mu_0 = 4\pi \times 10^{-7}$ T$\cdot$m/A is the **permeability of free space**
- $I$ is the current
- $d\vec{l}$ is a small length element along the wire in the direction of current
- $\hat{r}$ is the unit vector from the wire element to the field point
- $r$ is the distance from the wire element to the field point

This is the magnetic analog of Coulomb's law. To find the total B field, we integrate over the entire wire.

---
## 2. Magnetic Field of a Long Straight Wire

Using the Biot-Savart law for an infinitely long straight wire carrying current $I$, we get:

$$B = \frac{\mu_0 I}{2\pi r}$$

where $r$ is the perpendicular distance from the wire.

**Key features:**
- The field lines form **concentric circles** around the wire
- The field decreases as $1/r$
- Direction: use the **right-hand rule** -- wrap fingers of right hand around wire with thumb in current direction; fingers curl in the direction of B

### Interactive Demo 1: B Field Around a Long Straight Wire

Adjust the current to see how the magnetic field lines and magnitude change.

```python
def wire_field_demo(I=5.0):
    """
    Visualize B field around a long straight wire.
    Wire is at origin, current out of the page (for I>0).
    """
    if abs(I) < 0.01:
        print("No current, no magnetic field.")
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
        ax1.set_title('B Field Lines (I = 0 A)', fontsize=13)
        ax1.set_xlim(-5, 5)
        ax1.set_ylim(-5, 5)
        ax1.set_aspect('equal')
        ax1.set_xlabel('x (m)')
        ax1.set_ylabel('y (m)')
        ax1.grid(True, alpha=0.2)
        ax1.text(0, 0, 'No current', ha='center', va='center', fontsize=14, color='gray')
        ax2.set_xlabel('Distance r (m)', fontsize=12)
        ax2.set_ylabel('|B| (uT)', fontsize=12)
        ax2.set_title('Magnetic Field Magnitude vs Distance', fontsize=13)
        ax2.grid(True, alpha=0.3)
        ax2.set_xlim(0, 5)
        ax2.axhline(0, color='blue', lw=2.5)
        plt.tight_layout()
        plt.show()
        return

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Left panel: field lines (circles) and field direction
    ax1.set_title(f'B Field Lines (I = {I:.1f} A, {"out of" if I > 0 else "into"} page)', fontsize=13)
    ax1.set_xlim(-5, 5)
    ax1.set_ylim(-5, 5)
    ax1.set_aspect('equal')
    ax1.set_xlabel('x (m)')
    ax1.set_ylabel('y (m)')
    ax1.grid(True, alpha=0.2)
    
    # Wire cross-section
    if I > 0:
        ax1.plot(0, 0, 'o', color='black', markersize=15, zorder=5)
        ax1.plot(0, 0, '.', color='white', markersize=4, zorder=6)
        ax1.text(0.4, 0.4, 'I (out)', fontsize=10)
    else:
        ax1.plot(0, 0, 'o', color='black', markersize=15, zorder=5)
        ax1.plot(0, 0, 'x', color='white', markersize=8, mew=2, zorder=6)
        ax1.text(0.4, 0.4, 'I (in)', fontsize=10)
    
    # Concentric field lines
    theta = np.linspace(0, 2 * np.pi, 100)
    radii = [0.5, 1.0, 1.5, 2.0, 3.0, 4.0]
    for r in radii:
        B_val = mu_0 * abs(I) / (2 * np.pi * r)
        alpha = min(1.0, B_val / (mu_0 * abs(I) / (2 * np.pi * 0.5)) + 0.2)
        ax1.plot(r * np.cos(theta), r * np.sin(theta), 'b-', alpha=alpha, lw=1.5)
        
        # Arrow showing direction
        direction = 1 if I > 0 else -1
        arrow_angle = np.pi / 4
        ax1.annotate('',
                     xy=(r * np.cos(arrow_angle + direction * 0.1),
                         r * np.sin(arrow_angle + direction * 0.1)),
                     xytext=(r * np.cos(arrow_angle - direction * 0.1),
                             r * np.sin(arrow_angle - direction * 0.1)),
                     arrowprops=dict(arrowstyle='->', color='blue', lw=1.5, alpha=alpha))
    
    # Right panel: B vs r
    r_range = np.linspace(0.1, 5, 200)
    B_range = mu_0 * abs(I) / (2 * np.pi * r_range)
    
    ax2.plot(r_range, B_range * 1e6, 'b-', lw=2.5)
    ax2.set_xlabel('Distance r (m)', fontsize=12)
    ax2.set_ylabel('|B| (uT)', fontsize=12)
    ax2.set_title(f'Magnetic Field Magnitude vs Distance', fontsize=13)
    ax2.grid(True, alpha=0.3)
    ax2.set_xlim(0, 5)
    
    # Mark a few values
    for r_mark in [1.0, 2.0, 3.0]:
        B_mark = mu_0 * abs(I) / (2 * np.pi * r_mark)
        ax2.plot(r_mark, B_mark * 1e6, 'ro', markersize=8)
        ax2.annotate(f'{B_mark*1e6:.2f} uT',
                     xy=(r_mark, B_mark * 1e6),
                     xytext=(r_mark + 0.3, B_mark * 1e6 + 0.2),
                     fontsize=9)
    
    plt.tight_layout()
    
    print(f"B = mu_0 * I / (2*pi*r)")
    print(f"At r = 1 m: B = {mu_0*abs(I)/(2*np.pi*1):.2e} T = {mu_0*abs(I)/(2*np.pi*1)*1e6:.2f} uT")
    print(f"At r = 2 m: B = {mu_0*abs(I)/(2*np.pi*2):.2e} T = {mu_0*abs(I)/(2*np.pi*2)*1e6:.2f} uT")
    plt.show()

interact(wire_field_demo,
         I=FloatSlider(min=-20, max=20, step=1, value=5.0, description='I (A)'));
```

## 3. Ampere's Law

**Ampere's law** relates the line integral of the magnetic field around a closed loop (Amperian loop) to the total current passing through the loop:

$$\oint \vec{B} \cdot d\vec{l} = \mu_0 I_{\text{enc}}$$

This is the magnetic analog of Gauss's law. It is most useful when there is **symmetry** that makes B constant along the Amperian loop.

**Steps to apply Ampere's law:**
1. Identify the symmetry of the current distribution
2. Choose an Amperian loop where B is constant and parallel (or perpendicular) to $d\vec{l}$
3. Evaluate the line integral: $\oint B \, dl = B \oint dl = B(2\pi r)$ for circular loops
4. Find $I_{\text{enc}}$ and solve for B

## 4. The Solenoid

A **solenoid** is a long coil of wire with many closely spaced turns. For an ideal (infinitely long) solenoid:

**Inside the solenoid:**
$$B = \mu_0 n I$$

where $n = N/L$ is the number of turns per unit length.

**Outside the solenoid:**
$$B \approx 0$$

Key features:
- The field inside is **uniform** (does not depend on position)
- The field outside is approximately zero
- This is derived using Ampere's law with a rectangular Amperian loop

### Interactive Demo 2: Solenoid Magnetic Field

Adjust the number of turns per length and the current to see how the solenoid field behaves.

```python
def solenoid_demo(n=100, I=2.0, R=0.5):
    """
    Visualize the B field of a solenoid.
    n: turns per meter, I: current, R: solenoid radius.
    """
    B_inside = mu_0 * n * I
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Left panel: cross-section view of solenoid
    L = 3.0  # solenoid length for display
    
    # Draw solenoid outline
    rect = patches.FancyBboxPatch((-L/2, -R), L, 2*R,
                                   boxstyle="round,pad=0.05",
                                   facecolor='lightyellow', edgecolor='black', lw=2)
    ax1.add_patch(rect)
    
    # Draw wire turns (cross-sections)
    n_shown = min(int(n * L / 30), 40)  # limit displayed turns
    x_turns = np.linspace(-L/2 + 0.05, L/2 - 0.05, n_shown)
    for xt in x_turns:
        # Top: current out of page (dot)
        ax1.plot(xt, R, 'o', color='red', markersize=5)
        ax1.plot(xt, R, '.', color='white', markersize=1.5)
        # Bottom: current into page (cross)
        ax1.plot(xt, -R, 'o', color='red', markersize=5)
        ax1.plot(xt, -R, 'x', color='white', markersize=3, mew=1)
    
    # Field lines inside (horizontal, uniform)
    n_lines = 5
    for yi in np.linspace(-R * 0.7, R * 0.7, n_lines):
        ax1.annotate('', xy=(L/2 - 0.1, yi), xytext=(-L/2 + 0.1, yi),
                     arrowprops=dict(arrowstyle='->', color='blue', lw=1.5))
    
    # Label
    ax1.text(0, 0, f'B = {B_inside*1e3:.3f} mT', ha='center', va='center',
             fontsize=12, fontweight='bold', color='blue',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))
    ax1.text(0, R + 0.3, 'B $\\approx$ 0 (outside)', ha='center', fontsize=10, color='gray')
    ax1.text(0, -R - 0.3, 'B $\\approx$ 0 (outside)', ha='center', fontsize=10, color='gray')
    
    ax1.set_xlim(-2, 2)
    ax1.set_ylim(-1.5, 1.5)
    ax1.set_aspect('equal')
    ax1.set_xlabel('x (m)', fontsize=12)
    ax1.set_ylabel('y (m)', fontsize=12)
    ax1.set_title('Solenoid Cross-Section', fontsize=13)
    ax1.grid(True, alpha=0.2)
    
    # Right panel: B field profile
    r_range = np.linspace(-2, 2, 400)
    B_profile = np.where(np.abs(r_range) <= R, B_inside, B_inside * 0.02)  # small outside for display
    
    # Smooth transition at edges (more realistic for finite solenoid)
    transition_width = 0.05
    for i in range(len(r_range)):
        if abs(abs(r_range[i]) - R) < transition_width:
            B_profile[i] = B_inside * 0.5
    
    ax2.fill_between(r_range, 0, B_profile * 1e3, alpha=0.3, color='blue')
    ax2.plot(r_range, B_profile * 1e3, 'b-', lw=2.5)
    ax2.axvline(-R, color='red', linestyle='--', alpha=0.5, label=f'r = R = {R} m')
    ax2.axvline(R, color='red', linestyle='--', alpha=0.5)
    
    ax2.set_xlabel('Radial distance from axis (m)', fontsize=12)
    ax2.set_ylabel('|B| (mT)', fontsize=12)
    ax2.set_title('B Field Profile: Inside vs Outside', fontsize=13)
    ax2.legend(fontsize=10)
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    
    print(f"Solenoid parameters: n = {n} turns/m, I = {I:.1f} A")
    print(f"B_inside = mu_0 * n * I = {B_inside:.2e} T = {B_inside*1e3:.3f} mT")
    print(f"B_outside ~ 0")
    print(f"Note: B inside does NOT depend on position or radius!")
    plt.show()

interact(solenoid_demo,
         n=IntSlider(min=10, max=500, step=10, value=100, description='n (turns/m)'),
         I=FloatSlider(min=0.5, max=10, step=0.5, value=2.0, description='I (A)'),
         R=FloatSlider(min=0.2, max=1.0, step=0.1, value=0.5, description='R (m)'));
```

## 5. The Right-Hand Rule

The **right-hand rule** is essential for finding the direction of the magnetic field:

**For a straight wire:**
- Point your right thumb in the direction of the current
- Your fingers curl in the direction of the magnetic field

**For a solenoid:**
- Curl the fingers of your right hand in the direction of the current
- Your thumb points in the direction of B inside the solenoid

### Interactive Demo 3: Right-Hand Rule Trainer

Given a current direction, determine the magnetic field direction. The answer is shown below the diagram.

```python
def right_hand_rule_trainer(scenario=1):
    """
    Practice the right-hand rule with various current configurations.
    """
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.2)
    
    theta = np.linspace(0, 2 * np.pi, 100)
    
    scenarios = {
        1: {
            'title': 'Wire with current in +y direction',
            'description': 'Current flows upward (+y). What direction does B point at (x=1, y=0)?',
            'answer': 'B points OUT of the page (+z) at (1,0)\nB points INTO the page (-z) at (-1,0)',
        },
        2: {
            'title': 'Wire with current out of page (+z)',
            'description': 'Current flows out of the page. What direction does B point at (x=1, y=0)?',
            'answer': 'B points in the -y direction at (1,0)\nField lines circle counterclockwise (seen from above)',
        },
        3: {
            'title': 'Wire with current in -x direction',
            'description': 'Current flows to the left (-x). What direction does B point at (x=0, y=1)?',
            'answer': 'B points OUT of the page (+z) at (0,1)\nB points INTO the page (-z) at (0,-1)',
        },
        4: {
            'title': 'Wire with current into page (-z)',
            'description': 'Current flows into the page. What direction does B point at (x=1, y=0)?',
            'answer': 'B points in the +y direction at (1,0)\nField lines circle clockwise (seen from above)',
        },
    }
    
    s = scenarios[scenario]
    ax.set_title(s['title'], fontsize=14, fontweight='bold')
    
    if scenario == 1:
        # Current in +y
        ax.annotate('', xy=(0, 2), xytext=(0, -2),
                     arrowprops=dict(arrowstyle='->', color='red', lw=3))
        ax.text(0.2, 0, 'I', color='red', fontsize=16, fontweight='bold')
        ax.plot(1, 0, 'gs', markersize=15, label='Point of interest')
        ax.plot(-1, 0, 'bs', markersize=15)
        # Answer: B field circles
        for r in [1.0, 1.5, 2.0]:
            # In x-z plane (we show the projection)
            ax.plot(r * np.cos(theta), r * np.sin(theta), 'b--', alpha=0.3)
    
    elif scenario == 2:
        # Current out of page
        ax.plot(0, 0, 'o', color='red', markersize=25, zorder=5)
        ax.plot(0, 0, '.', color='white', markersize=8, zorder=6)
        ax.text(0.3, -0.4, 'I (out)', color='red', fontsize=12)
        ax.plot(1, 0, 'gs', markersize=15, label='Point of interest')
        # Show CCW circles
        for r in [1.0, 1.5, 2.0]:
            ax.plot(r * np.cos(theta), r * np.sin(theta), 'b--', alpha=0.3)
            angle = np.pi/4
            ax.annotate('', xy=(r*np.cos(angle+0.15), r*np.sin(angle+0.15)),
                        xytext=(r*np.cos(angle-0.15), r*np.sin(angle-0.15)),
                        arrowprops=dict(arrowstyle='->', color='blue', lw=1.5, alpha=0.5))
    
    elif scenario == 3:
        # Current in -x
        ax.annotate('', xy=(-2, 0), xytext=(2, 0),
                     arrowprops=dict(arrowstyle='->', color='red', lw=3))
        ax.text(0, 0.2, 'I', color='red', fontsize=16, fontweight='bold')
        ax.plot(0, 1, 'gs', markersize=15, label='Point of interest')
        ax.plot(0, -1, 'bs', markersize=15)
    
    elif scenario == 4:
        # Current into page
        ax.plot(0, 0, 'o', color='red', markersize=25, zorder=5)
        ax.plot(0, 0, 'x', color='white', markersize=12, mew=3, zorder=6)
        ax.text(0.3, -0.4, 'I (in)', color='red', fontsize=12)
        ax.plot(1, 0, 'gs', markersize=15, label='Point of interest')
        # Show CW circles
        for r in [1.0, 1.5, 2.0]:
            ax.plot(r * np.cos(theta), r * np.sin(theta), 'b--', alpha=0.3)
            angle = np.pi/4
            ax.annotate('', xy=(r*np.cos(angle-0.15), r*np.sin(angle-0.15)),
                        xytext=(r*np.cos(angle+0.15), r*np.sin(angle+0.15)),
                        arrowprops=dict(arrowstyle='->', color='blue', lw=1.5, alpha=0.5))
    
    ax.legend(fontsize=10)
    plt.tight_layout()
    
    print(f"QUESTION: {s['description']}")
    print(f"")
    print(f"Use the right-hand rule to figure it out, then check below:")
    print(f"")
    print(f"ANSWER: {s['answer']}")
    plt.show()

interact(right_hand_rule_trainer,
         scenario=IntSlider(min=1, max=4, step=1, value=1, description='Scenario'));
```

## 6. Animation: Magnetic Field Around a Current-Carrying Wire

Watch animated compass needles align with the magnetic field around a current-carrying wire.

```python
def create_compass_animation():
    """
    Animate compass needles aligning with B field around a wire.
    Wire carries current out of the page.
    """
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.set_xlim(-4, 4)
    ax.set_ylim(-4, 4)
    ax.set_aspect('equal')
    ax.set_title('Compass Needles Around a Current-Carrying Wire', fontsize=13)
    ax.set_xlabel('x (m)', fontsize=12)
    ax.set_ylabel('y (m)', fontsize=12)
    ax.grid(True, alpha=0.2)
    
    # Wire at origin (current out of page)
    wire = ax.plot(0, 0, 'o', color='red', markersize=20, zorder=5)[0]
    ax.plot(0, 0, '.', color='white', markersize=6, zorder=6)
    ax.text(0.4, 0.4, 'I (out)', fontsize=10, color='red')
    
    # Compass positions (grid, excluding center)
    compass_pos = []
    for xi in np.arange(-3, 4, 1.5):
        for yi in np.arange(-3, 4, 1.5):
            r = np.sqrt(xi**2 + yi**2)
            if r > 0.8:  # not too close to wire
                compass_pos.append((xi, yi))
    
    n_compasses = len(compass_pos)
    
    # Target angles (B field direction at each compass)
    target_angles = []
    for (cx, cy) in compass_pos:
        # B field direction: tangent to circle, counterclockwise
        # At point (cx, cy), B direction is (-cy, cx) / r
        angle = np.arctan2(cx, -cy)  # angle of B direction
        target_angles.append(angle)
    
    # Initial random angles
    np.random.seed(42)
    initial_angles = np.random.uniform(0, 2 * np.pi, n_compasses)
    
    # Draw compass needles
    needle_length = 0.4
    needles_red = []
    needles_white = []
    for i, (cx, cy) in enumerate(compass_pos):
        angle = initial_angles[i]
        dx = needle_length * np.cos(angle)
        dy = needle_length * np.sin(angle)
        line_r, = ax.plot([cx, cx + dx], [cy, cy + dy], 'r-', lw=2.5)
        line_w, = ax.plot([cx, cx - dx], [cy, cy - dy], 'gray', lw=2, alpha=0.6)
        needles_red.append(line_r)
        needles_white.append(line_w)
    
    n_frames = 60
    
    def animate(frame):
        t = frame / n_frames  # 0 to 1
        # Smooth interpolation (ease in-out)
        t_smooth = 0.5 * (1 - np.cos(np.pi * t))
        
        for i, (cx, cy) in enumerate(compass_pos):
            # Interpolate between initial and target angle
            # Handle angle wrapping
            diff = target_angles[i] - initial_angles[i]
            diff = (diff + np.pi) % (2 * np.pi) - np.pi  # shortest path
            current_angle = initial_angles[i] + t_smooth * diff
            
            dx = needle_length * np.cos(current_angle)
            dy = needle_length * np.sin(current_angle)
            
            needles_red[i].set_data([cx, cx + dx], [cy, cy + dy])
            needles_white[i].set_data([cx, cx - dx], [cy, cy - dy])
        
        return needles_red + needles_white
    
    anim = FuncAnimation(fig, animate, frames=n_frames, interval=80, blit=True)
    plt.close(fig)
    return anim

anim = create_compass_animation()
HTML(anim.to_jshtml())
```

## 7. Force Between Two Parallel Wires

Two parallel wires carrying currents $I_1$ and $I_2$, separated by distance $d$, exert forces on each other.

The force per unit length between the wires is:

$$\frac{F}{L} = \frac{\mu_0 I_1 I_2}{2\pi d}$$

**Direction:**
- **Parallel currents** (same direction): wires **attract** each other
- **Antiparallel currents** (opposite directions): wires **repel** each other

This is how the **ampere** was originally defined: 1 A is the current that, flowing in each of two parallel wires 1 m apart, produces a force of $2 \times 10^{-7}$ N per meter of wire.

### Interactive Demo 4: Two Parallel Current-Carrying Wires

Adjust the currents and distance to see the magnetic force between two parallel wires.

```python
def parallel_wires_demo(I1=5.0, I2=5.0, d=1.0):
    """
    Visualize the force between two parallel current-carrying wires.
    Both wires carry current out of the page (positive = out, negative = in).
    """
    F_per_L = mu_0 * abs(I1 * I2) / (2 * np.pi * d)  # force per unit length
    
    # Determine attract or repel, handle zero current
    if abs(I1) < 0.01 or abs(I2) < 0.01:
        force_type = "No force (zero current)"
        same_direction = None
    else:
        same_direction = (I1 > 0 and I2 > 0) or (I1 < 0 and I2 < 0)
        force_type = "ATTRACT" if same_direction else "REPEL"
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Left panel: wire cross-sections with force arrows
    ax1.set_title(f'Two Parallel Wires (d = {d:.1f} m)', fontsize=13)
    ax1.set_xlim(-3, 3)
    ax1.set_ylim(-2, 2)
    ax1.set_aspect('equal')
    ax1.grid(True, alpha=0.2)
    
    # Wire 1 at (-d/2, 0)
    x1 = -d / 2
    ax1.plot(x1, 0, 'o', color='red', markersize=25, zorder=5)
    if I1 > 0:
        ax1.plot(x1, 0, '.', color='white', markersize=6, zorder=6)
        ax1.text(x1, -0.5, f'I1={I1:.0f}A (out)', ha='center', fontsize=9, color='red')
    else:
        ax1.plot(x1, 0, 'x', color='white', markersize=10, mew=2, zorder=6)
        ax1.text(x1, -0.5, f'I1={I1:.0f}A (in)', ha='center', fontsize=9, color='red')
    
    # Wire 2 at (+d/2, 0)
    x2 = d / 2
    ax1.plot(x2, 0, 'o', color='blue', markersize=25, zorder=5)
    if I2 > 0:
        ax1.plot(x2, 0, '.', color='white', markersize=6, zorder=6)
        ax1.text(x2, -0.5, f'I2={I2:.0f}A (out)', ha='center', fontsize=9, color='blue')
    else:
        ax1.plot(x2, 0, 'x', color='white', markersize=10, mew=2, zorder=6)
        ax1.text(x2, -0.5, f'I2={I2:.0f}A (in)', ha='center', fontsize=9, color='blue')
    
    # Force arrows
    if abs(I1) < 0.01 or abs(I2) < 0.01:
        ax1.text(0, 0.7, 'No force (zero current)', ha='center', fontsize=16, color='gray', fontweight='bold')
    elif same_direction:  # attract
        arrow_scale = min(0.8, F_per_L * 1e6 * 0.5)  # scale for visibility
        ax1.annotate('', xy=(x1 + arrow_scale, 0), xytext=(x1 + 0.01, 0),
                     arrowprops=dict(arrowstyle='->', color='green', lw=3))
        ax1.annotate('', xy=(x2 - arrow_scale, 0), xytext=(x2 - 0.01, 0),
                     arrowprops=dict(arrowstyle='->', color='green', lw=3))
        ax1.text(0, 0.7, 'ATTRACT', ha='center', fontsize=16, color='green', fontweight='bold')
    else:  # repel
        arrow_scale = min(0.8, F_per_L * 1e6 * 0.5)  # scale for visibility
        ax1.annotate('', xy=(x1 - arrow_scale, 0), xytext=(x1 - 0.01, 0),
                     arrowprops=dict(arrowstyle='->', color='orange', lw=3))
        ax1.annotate('', xy=(x2 + arrow_scale, 0), xytext=(x2 + 0.01, 0),
                     arrowprops=dict(arrowstyle='->', color='orange', lw=3))
        ax1.text(0, 0.7, 'REPEL', ha='center', fontsize=16, color='orange', fontweight='bold')
    
    # Distance annotation
    ax1.annotate('', xy=(x2, -1.2), xytext=(x1, -1.2),
                 arrowprops=dict(arrowstyle='<->', color='black', lw=1.5))
    ax1.text(0, -1.4, f'd = {d:.1f} m', ha='center', fontsize=10)
    
    ax1.set_xlabel('x (m)', fontsize=12)
    ax1.set_ylabel('y (m)', fontsize=12)
    
    # Right panel: F/L vs d
    d_range = np.linspace(0.1, 3, 200)
    F_range = mu_0 * abs(I1 * I2) / (2 * np.pi * d_range)
    
    ax2.plot(d_range, F_range * 1e6, 'b-', lw=2.5)
    ax2.plot(d, F_per_L * 1e6, 'ro', markersize=12, zorder=5,
             label=f'd={d:.1f} m: F/L={F_per_L*1e6:.2f} uN/m')
    ax2.set_xlabel('Distance d (m)', fontsize=12)
    ax2.set_ylabel('F/L (uN/m)', fontsize=12)
    ax2.set_title('Force per Unit Length vs Distance', fontsize=13)
    ax2.legend(fontsize=10)
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    
    print(f"I1 = {I1:.1f} A, I2 = {I2:.1f} A, d = {d:.2f} m")
    print(f"F/L = mu_0*I1*I2 / (2*pi*d) = {F_per_L:.2e} N/m = {F_per_L*1e6:.2f} uN/m")
    if abs(I1) < 0.01 or abs(I2) < 0.01:
        print(f"One or both currents are zero -> {force_type}")
    else:
        print(f"Currents are {'parallel' if same_direction else 'antiparallel'} -> wires {force_type}")
    plt.show()

interact(parallel_wires_demo,
         I1=FloatSlider(min=-20, max=20, step=1, value=5.0, description='I1 (A)'),
         I2=FloatSlider(min=-20, max=20, step=1, value=5.0, description='I2 (A)'),
         d=FloatSlider(min=0.1, max=3, step=0.1, value=1.0, description='d (m)'));
```

## 8. The Toroid

A **toroid** is a solenoid bent into a donut shape. Using Ampere's law with a circular Amperian loop of radius $r$ inside the toroid:

$$B(2\pi r) = \mu_0 N I$$

$$B = \frac{\mu_0 N I}{2\pi r}$$

where $N$ is the total number of turns.

- Inside the toroid: $B = \mu_0 N I / (2\pi r)$
- Outside the toroid: $B = 0$ (no enclosed current)

## 9. Worked Examples

### Example 1: B Field of a Solenoid

**Problem:** A solenoid has 500 turns, is 25 cm long, and carries a current of 3 A. Find the magnetic field inside the solenoid.

```python
# Worked Example 1: Solenoid B field

N = 500          # total turns
L_sol = 0.25     # length in meters
I_sol = 3.0      # current in amps

n_sol = N / L_sol  # turns per meter
B_sol = mu_0 * n_sol * I_sol

print("=" * 50)
print("Worked Example 1: Solenoid Magnetic Field")
print("=" * 50)
print(f"Given:")
print(f"  N = {N} turns")
print(f"  L = {L_sol} m = {L_sol*100:.0f} cm")
print(f"  I = {I_sol} A")
print()
print(f"Solution:")
print(f"  Step 1: Find turns per unit length")
print(f"    n = N/L = {N}/{L_sol} = {n_sol:.0f} turns/m")
print()
print(f"  Step 2: Apply B = mu_0 * n * I")
print(f"    B = (4*pi*1e-7)({n_sol:.0f})({I_sol})")
print(f"    B = {B_sol:.4e} T")
print(f"    B = {B_sol*1e3:.3f} mT")
print()
print(f"  This is about {B_sol/5e-5:.0f} times Earth's magnetic field.")
```

### Example 2: Force Between Two Parallel Wires

**Problem:** Two parallel wires are 10 cm apart. Wire 1 carries 8 A and wire 2 carries 5 A in the same direction. Find the force per unit length and state whether the wires attract or repel.

```python
# Worked Example 2: Force between parallel wires

I_1 = 8.0       # A
I_2 = 5.0       # A
d_wires = 0.10   # m

F_per_L = mu_0 * I_1 * I_2 / (2 * np.pi * d_wires)

print("=" * 50)
print("Worked Example 2: Force Between Parallel Wires")
print("=" * 50)
print(f"Given:")
print(f"  I1 = {I_1} A")
print(f"  I2 = {I_2} A")
print(f"  d = {d_wires} m = {d_wires*100:.0f} cm")
print(f"  Currents in the SAME direction")
print()
print(f"Solution:")
print(f"  F/L = mu_0 * I1 * I2 / (2*pi*d)")
print(f"      = (4*pi*1e-7)({I_1})({I_2}) / (2*pi*{d_wires})")
print(f"      = {mu_0*I_1*I_2:.2e} / {2*np.pi*d_wires:.4f}")
print(f"      = {F_per_L:.2e} N/m")
print(f"      = {F_per_L*1e5:.2f} x 10^-5 N/m")
print()
print(f"  Since the currents are in the same direction,")
print(f"  the wires ATTRACT each other.")
print()
print(f"  For a 1 m length of wire: F = {F_per_L:.2e} N = {F_per_L*1e6:.1f} uN")
```

## 10. Summary

| Concept | Key Equation |
|---------|-------------|
| Biot-Savart law | $d\vec{B} = \frac{\mu_0}{4\pi} \frac{I \, d\vec{l} \times \hat{r}}{r^2}$ |
| B field of long wire | $B = \frac{\mu_0 I}{2\pi r}$ |
| Ampere's law | $\oint \vec{B} \cdot d\vec{l} = \mu_0 I_{\text{enc}}$ |
| Solenoid (inside) | $B = \mu_0 n I$ |
| Toroid (inside) | $B = \frac{\mu_0 N I}{2\pi r}$ |
| Force between parallel wires | $\frac{F}{L} = \frac{\mu_0 I_1 I_2}{2\pi d}$ |
| Parallel currents attract | Antiparallel currents repel |

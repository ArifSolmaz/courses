## 1. Introduction to Wave Optics

In previous weeks we treated light as rays (geometric optics). However, light is an **electromagnetic wave**, and when the objects it interacts with are comparable in size to its wavelength, wave effects become important.

The two key wave phenomena we study this week are:

1. **Interference** -- the superposition of two or more coherent waves
2. **Diffraction** -- the bending and spreading of waves around obstacles or through apertures

### Principle of Superposition

When two waves overlap, the resulting displacement is the **sum** of the individual displacements:

$$E_{\text{total}} = E_1 + E_2$$

- **Constructive interference**: waves in phase, amplitudes add ($\Delta \phi = 0, 2\pi, 4\pi, \ldots$)
- **Destructive interference**: waves out of phase, amplitudes cancel ($\Delta \phi = \pi, 3\pi, 5\pi, \ldots$)

## 2. Young's Double-Slit Experiment

Thomas Young's experiment (1801) provided the first clear evidence that light is a wave.

**Setup**: Monochromatic light passes through two narrow slits separated by distance $d$. On a screen at distance $L$, an interference pattern of bright and dark fringes appears.

### Key Equations

**Path difference**:
$$\Delta r = d \sin\theta$$

**Bright fringes** (constructive interference):
$$d \sin\theta = m\lambda, \quad m = 0, \pm 1, \pm 2, \ldots$$

**Dark fringes** (destructive interference):
$$d \sin\theta = \left(m + \tfrac{1}{2}\right)\lambda$$

**Position on screen** (small angle approximation $\sin\theta \approx \tan\theta = y/L$):
$$y_m = \frac{m\lambda L}{d}$$

**Fringe spacing**:
$$\Delta y = \frac{\lambda L}{d}$$

```python
def double_slit_pattern(d_um=100.0, wavelength_nm=550.0, L_m=1.5):
    """
    Interactive Young's Double-Slit Experiment.
    Shows the intensity pattern on the screen.
    """
    d = d_um * 1e-6        # slit separation in meters
    lam = wavelength_nm * 1e-9  # wavelength in meters
    L = L_m                # screen distance in meters
    
    # Screen coordinates
    y = np.linspace(-0.02, 0.02, 2000)  # +/- 2 cm on screen
    theta = np.arctan(y / L)
    
    # Double-slit intensity: I = I0 * cos^2(pi*d*sin(theta)/lambda)
    phase = np.pi * d * np.sin(theta) / lam
    intensity = np.cos(phase)**2
    
    # Fringe spacing
    fringe_spacing = lam * L / d * 1000  # in mm
    
    # Color based on wavelength
    color = wavelength_to_rgb(wavelength_nm)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # 1D intensity plot
    ax1.plot(y * 1000, intensity, color=color, linewidth=2)
    ax1.fill_between(y * 1000, intensity, alpha=0.3, color=color)
    ax1.set_xlabel('Position on Screen (mm)')
    ax1.set_ylabel('Relative Intensity')
    ax1.set_title('Double-Slit Intensity Pattern')
    ax1.set_ylim(-0.05, 1.15)
    ax1.grid(True, alpha=0.3)
    
    # 2D intensity visualization
    y2d = np.linspace(-0.02, 0.02, 500)
    theta2d = np.arctan(y2d / L)
    phase2d = np.pi * d * np.sin(theta2d) / lam
    intensity2d = np.cos(phase2d)**2
    img = np.outer(np.ones(50), intensity2d)
    
    cmap = plt.cm.gray if wavelength_nm < 380 or wavelength_nm > 750 else None
    if cmap is None:
        # Create a colormap from black to the wavelength color
        from matplotlib.colors import LinearSegmentedColormap
        cmap = LinearSegmentedColormap.from_list('wave', ['black', color])
    
    ax2.imshow(img, extent=[y2d[0]*1000, y2d[-1]*1000, -1, 1],
               aspect='auto', cmap=cmap)
    ax2.set_xlabel('Position on Screen (mm)')
    ax2.set_yticks([])
    ax2.set_title('Screen View')
    
    fig.suptitle(f'Fringe spacing: {fringe_spacing:.2f} mm | '
                 f'd = {d_um:.0f} um, lambda = {wavelength_nm:.0f} nm, L = {L_m:.1f} m',
                 fontsize=11, y=1.02)
    plt.tight_layout()
    plt.show()

def wavelength_to_rgb(wavelength_nm):
    """Approximate visible wavelength to RGB color."""
    w = wavelength_nm
    if w < 380: return (0.5, 0.0, 1.0)
    elif w < 440: return (0.5 - 0.5*(w-380)/(440-380), 0, 1.0)
    elif w < 490: return (0, (w-440)/(490-440), 1.0)
    elif w < 510: return (0, 1.0, 1.0 - (w-490)/(510-490))
    elif w < 580: return ((w-510)/(580-510), 1.0, 0)
    elif w < 645: return (1.0, 1.0 - (w-580)/(645-580), 0)
    elif w <= 750: return (1.0, 0, 0)
    else: return (0.55, 0.0, 0.0)

interact(double_slit_pattern,
         d_um=FloatSlider(min=20, max=500, step=10, value=100,
                          description='d (um):', style={'description_width': 'initial'}),
         wavelength_nm=FloatSlider(min=380, max=750, step=10, value=550,
                                   description='lambda (nm):', style={'description_width': 'initial'}),
         L_m=FloatSlider(min=0.5, max=3.0, step=0.1, value=1.5,
                         description='L (m):', style={'description_width': 'initial'}));
```

### Worked Example 1: Double-Slit Fringe Spacing

**Problem**: In a Young's double-slit experiment, the slit separation is $d = 0.12$ mm, the wavelength is $\lambda = 600$ nm, and the screen is $L = 2.0$ m from the slits. Find:

(a) The fringe spacing  
(b) The position of the 3rd bright fringe  
(c) The position of the 2nd dark fringe

```python
# Worked Example 1: Double-Slit Calculations

d = 0.12e-3    # slit separation in meters
lam = 600e-9   # wavelength in meters
L = 2.0        # screen distance in meters

# (a) Fringe spacing
delta_y = lam * L / d
print(f"(a) Fringe spacing: delta_y = lambda * L / d")
print(f"    = ({lam*1e9:.0f} nm)({L:.1f} m) / ({d*1e3:.2f} mm)")
print(f"    = {delta_y*1000:.2f} mm = {delta_y*100:.3f} cm")

# (b) Position of 3rd bright fringe (m=3)
m = 3
y_3 = m * lam * L / d
print(f"\n(b) 3rd bright fringe (m = {m}):")
print(f"    y_3 = m * lambda * L / d = {m} * {delta_y*1000:.2f} mm")
print(f"    y_3 = {y_3*1000:.2f} mm = {y_3*100:.3f} cm")

# (c) Position of 2nd dark fringe (m=1, since first dark is m=0)
m_dark = 1  # 2nd dark fringe
y_dark = (m_dark + 0.5) * lam * L / d
print(f"\n(c) 2nd dark fringe (m = {m_dark}):")
print(f"    y = (m + 1/2) * lambda * L / d = ({m_dark} + 0.5) * {delta_y*1000:.2f} mm")
print(f"    y = {y_dark*1000:.2f} mm = {y_dark*100:.3f} cm")
```

## 3. Single-Slit Diffraction

When light passes through a **single slit** of width $a$, it spreads out and produces a **diffraction pattern** with a bright central maximum and dimmer side maxima.

### Key Equations

**Minima** (dark fringes) occur at:
$$a \sin\theta = m\lambda, \quad m = \pm 1, \pm 2, \pm 3, \ldots$$

Note: $m = 0$ is **not** a minimum -- it is the central maximum!

**Intensity distribution**:
$$I(\theta) = I_0 \left[\frac{\sin(\beta/2)}{\beta/2}\right]^2$$

where $\beta = \frac{2\pi a \sin\theta}{\lambda}$

The central maximum has **twice** the width of the side maxima.

```python
def single_slit_diffraction(a_um=50.0, wavelength_nm=550.0):
    """
    Interactive Single-Slit Diffraction Pattern.
    """
    a = a_um * 1e-6         # slit width in meters
    lam = wavelength_nm * 1e-9  # wavelength in meters
    
    # Angular range
    theta = np.linspace(-0.05, 0.05, 3000)
    theta[len(theta)//2] = 1e-15  # avoid division by zero
    
    # Intensity: sinc^2 function
    beta_half = np.pi * a * np.sin(theta) / lam
    intensity = (np.sin(beta_half) / beta_half)**2
    
    # Angular position of first minimum
    theta_1 = np.arcsin(lam / a) if lam / a <= 1 else np.pi/2
    
    color = wavelength_to_rgb(wavelength_nm)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Intensity vs angle
    ax1.plot(np.degrees(theta), intensity, color=color, linewidth=2)
    ax1.fill_between(np.degrees(theta), intensity, alpha=0.2, color=color)
    ax1.axvline(np.degrees(theta_1), color='gray', linestyle='--', alpha=0.7, label=f'1st min: {np.degrees(theta_1):.2f} deg')
    ax1.axvline(-np.degrees(theta_1), color='gray', linestyle='--', alpha=0.7)
    ax1.set_xlabel('Angle (degrees)')
    ax1.set_ylabel('Relative Intensity')
    ax1.set_title('Single-Slit Diffraction Pattern')
    ax1.legend(fontsize=10)
    ax1.grid(True, alpha=0.3)
    
    # 2D screen view
    from matplotlib.colors import LinearSegmentedColormap
    cmap = LinearSegmentedColormap.from_list('wave', ['black', color])
    img = np.outer(np.ones(50), intensity)
    ax2.imshow(img, extent=[np.degrees(theta[0]), np.degrees(theta[-1]), -1, 1],
               aspect='auto', cmap=cmap)
    ax2.set_xlabel('Angle (degrees)')
    ax2.set_yticks([])
    ax2.set_title('Screen View')
    
    fig.suptitle(f'Slit width a = {a_um:.0f} um, lambda = {wavelength_nm:.0f} nm | '
                 f'Central max width: {2*np.degrees(theta_1):.2f} deg',
                 fontsize=11, y=1.02)
    plt.tight_layout()
    plt.show()

interact(single_slit_diffraction,
         a_um=FloatSlider(min=10, max=200, step=5, value=50,
                          description='a (um):', style={'description_width': 'initial'}),
         wavelength_nm=FloatSlider(min=380, max=750, step=10, value=550,
                                   description='lambda (nm):', style={'description_width': 'initial'}));
```

## 4. Wave Interference Animation

Let us visualize how two circular waves from two point sources interfere with each other. Bright regions correspond to constructive interference and dark regions to destructive interference.

```python
# Animation: Two-source wave interference

fig, ax = plt.subplots(figsize=(8, 8))

# Grid
x = np.linspace(-5, 5, 300)
y = np.linspace(-5, 5, 300)
X, Y = np.meshgrid(x, y)

# Two source positions
s1 = (-1.0, 0)
s2 = (1.0, 0)

r1 = np.sqrt((X - s1[0])**2 + (Y - s1[1])**2)
r2 = np.sqrt((X - s2[0])**2 + (Y - s2[1])**2)

k = 4.0   # wave number
omega = 2.0  # angular frequency

def animate_interference(frame):
    ax.clear()
    t = frame * 0.1
    
    # Two circular waves
    wave1 = np.sin(k * r1 - omega * t) / np.sqrt(r1 + 0.1)
    wave2 = np.sin(k * r2 - omega * t) / np.sqrt(r2 + 0.1)
    total = wave1 + wave2
    
    ax.imshow(total, extent=[-5, 5, -5, 5], cmap='RdBu', vmin=-2, vmax=2)
    ax.plot(*s1, 'ko', markersize=8)
    ax.plot(*s2, 'ko', markersize=8)
    ax.set_title(f'Two-Source Wave Interference (t = {t:.1f})', fontsize=13)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    return []

anim = FuncAnimation(fig, animate_interference, frames=60, interval=80, blit=False)
plt.close(fig)
HTML(anim.to_jshtml())
```

In the animation above:
- **Red/blue** regions show positive/negative wave displacement
- **Lines of maximum intensity** (antinodal lines) radiate outward where waves constructively interfere
- **Lines of zero intensity** (nodal lines) appear where waves destructively interfere
- The two black dots mark the source positions

## 5. Diffraction Grating

A **diffraction grating** consists of many equally spaced slits (typically thousands per mm). It produces very **sharp, bright maxima** because the contributions from $N$ slits add coherently.

### Key Equations

**Principal maxima** occur at the same angles as the double slit:
$$d \sin\theta = m\lambda, \quad m = 0, \pm 1, \pm 2, \ldots$$

where $d$ is the spacing between adjacent slits.

**Intensity pattern** for $N$ slits:
$$I(\theta) = I_0 \left[\frac{\sin(N\phi/2)}{N\sin(\phi/2)}\right]^2$$

where $\phi = \frac{2\pi d \sin\theta}{\lambda}$

As $N$ increases, the principal maxima become **narrower and brighter**.

```python
def diffraction_grating(N=5, d_um=10.0, wavelength_nm=550.0):
    """
    Interactive Diffraction Grating Pattern.
    Shows how N slits produce sharp peaks.
    """
    d = d_um * 1e-6
    lam = wavelength_nm * 1e-9
    
    theta = np.linspace(-0.15, 0.15, 5000)
    
    # N-slit intensity
    phi = np.pi * d * np.sin(theta) / lam
    
    # Handle division by zero
    with np.errstate(divide='ignore', invalid='ignore'):
        numerator = np.sin(N * phi)
        denominator = N * np.sin(phi)
        intensity = np.where(np.abs(denominator) < 1e-12, 1.0,
                            (numerator / denominator)**2)
    
    color = wavelength_to_rgb(wavelength_nm)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
    
    # Intensity pattern
    ax1.plot(np.degrees(theta), intensity, color=color, linewidth=1.5)
    ax1.fill_between(np.degrees(theta), intensity, alpha=0.2, color=color)
    ax1.set_xlabel('Angle (degrees)')
    ax1.set_ylabel('Relative Intensity')
    ax1.set_title(f'Diffraction Grating: N = {N} slits, d = {d_um:.1f} um, lambda = {wavelength_nm:.0f} nm')
    ax1.set_ylim(-0.05, 1.15)
    ax1.grid(True, alpha=0.3)
    
    # Compare N=2 vs current N
    phi2 = np.pi * d * np.sin(theta) / lam
    intensity_2 = np.cos(phi2)**2  # double slit
    ax2.plot(np.degrees(theta), intensity_2, 'b--', linewidth=1, alpha=0.5, label='N = 2 (double slit)')
    ax2.plot(np.degrees(theta), intensity, color=color, linewidth=1.5, label=f'N = {N}')
    ax2.set_xlabel('Angle (degrees)')
    ax2.set_ylabel('Relative Intensity')
    ax2.set_title('Comparison with Double Slit')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

interact(diffraction_grating,
         N=IntSlider(min=2, max=30, step=1, value=5,
                     description='N slits:', style={'description_width': 'initial'}),
         d_um=FloatSlider(min=2, max=50, step=1, value=10,
                          description='d (um):', style={'description_width': 'initial'}),
         wavelength_nm=FloatSlider(min=380, max=750, step=10, value=550,
                                   description='lambda (nm):', style={'description_width': 'initial'}));
```

### Worked Example 2: Diffraction Grating Wavelength Determination

**Problem**: A diffraction grating has 5000 lines/cm. When white light shines on it, the second-order ($m = 2$) maximum for a certain spectral line appears at $\theta = 32.0°$. 

(a) What is the wavelength of this spectral line?  
(b) At what angle does the first-order maximum appear?  
(c) Is a third-order maximum possible for this wavelength?

```python
# Worked Example 2: Diffraction Grating

lines_per_cm = 5000
d = 1.0 / (lines_per_cm * 100)  # slit spacing in meters
theta_2 = np.radians(32.0)       # angle for m=2
m = 2

print(f"Grating: {lines_per_cm} lines/cm")
print(f"Slit spacing: d = 1/{lines_per_cm*100} m = {d*1e6:.2f} um")

# (a) Wavelength
lam = d * np.sin(theta_2) / m
print(f"\n(a) From d sin(theta) = m * lambda:")
print(f"    lambda = d sin(theta) / m")
print(f"    = ({d*1e6:.2f} um) * sin({np.degrees(theta_2):.1f} deg) / {m}")
print(f"    = {lam*1e9:.1f} nm")

# (b) First-order angle
sin_theta_1 = lam / d
theta_1 = np.degrees(np.arcsin(sin_theta_1))
print(f"\n(b) For m = 1: sin(theta) = lambda / d = {sin_theta_1:.4f}")
print(f"    theta_1 = {theta_1:.2f} degrees")

# (c) Third-order check
sin_theta_3 = 3 * lam / d
print(f"\n(c) For m = 3: sin(theta) = 3*lambda / d = {sin_theta_3:.4f}")
if sin_theta_3 <= 1:
    theta_3 = np.degrees(np.arcsin(sin_theta_3))
    print(f"    Yes! Third order appears at theta = {theta_3:.2f} degrees")
else:
    print(f"    Since sin(theta) > 1, the third-order maximum does NOT exist.")
```

## 6. Thin-Film Interference

When light encounters a thin transparent film (such as a soap bubble or oil film), reflections from the **top** and **bottom** surfaces interfere with each other.

### Key Concepts

1. **Phase change on reflection**: When light reflects from a medium with a **higher** refractive index, it undergoes a $\pi$ phase shift (equivalent to half a wavelength).

2. **Path difference**: Light reflecting from the bottom surface travels an extra distance $2t$ within the film (at normal incidence), where $t$ is the film thickness.

3. **Wavelength in the film**: $\lambda_n = \lambda / n$, where $n$ is the refractive index of the film.

### Conditions (for a film in air, $n_{\text{air}} < n_{\text{film}}$)

**Constructive** (bright reflection):
$$2 n t = \left(m + \tfrac{1}{2}\right)\lambda, \quad m = 0, 1, 2, \ldots$$

**Destructive** (dark reflection):
$$2 n t = m\lambda, \quad m = 0, 1, 2, \ldots$$

The factor of $1/2$ comes from the net phase shift of $\pi$ (one surface has a phase change, the other does not).

```python
def thin_film_interference(thickness_nm=300.0, n_film=1.5):
    """
    Interactive Thin-Film Interference.
    Shows which visible wavelengths are constructively reflected.
    """
    t = thickness_nm * 1e-9  # film thickness in meters
    n = n_film
    
    # Visible wavelengths
    wavelengths = np.linspace(380, 750, 1000)  # nm
    
    # For a film in air (n_air < n_film > n_air):
    # Constructive: 2*n*t = (m + 1/2)*lambda
    # Reflectance depends on the path difference
    reflectance = []
    for lam_nm in wavelengths:
        lam = lam_nm * 1e-9
        # Phase difference (including the pi shift)
        delta = 2 * np.pi * (2 * n * t) / lam + np.pi  # pi from reflection
        # Reflected intensity proportional to cos^2(delta/2)
        R = np.sin(delta / 2)**2
        reflectance.append(R)
    
    reflectance = np.array(reflectance)
    
    # Calculate the perceived reflected color
    r_total, g_total, b_total = 0, 0, 0
    for i, lam_nm in enumerate(wavelengths):
        rgb = wavelength_to_rgb(lam_nm)
        if isinstance(rgb, tuple):
            r_total += rgb[0] * reflectance[i]
            g_total += rgb[1] * reflectance[i]
            b_total += rgb[2] * reflectance[i]
    
    # Normalize
    max_val = max(r_total, g_total, b_total, 1e-10)
    perceived_color = (r_total/max_val, g_total/max_val, b_total/max_val)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5),
                                    gridspec_kw={'width_ratios': [3, 1]})
    
    # Spectrum plot with colored fill
    for i in range(len(wavelengths) - 1):
        c = wavelength_to_rgb(wavelengths[i])
        ax1.fill_between(wavelengths[i:i+2], reflectance[i:i+2], alpha=0.6, color=c)
    
    ax1.plot(wavelengths, reflectance, 'k-', linewidth=1)
    ax1.set_xlabel('Wavelength (nm)')
    ax1.set_ylabel('Reflectance (relative)')
    ax1.set_title(f'Thin Film Reflectance Spectrum\nt = {thickness_nm:.0f} nm, n = {n_film:.2f}')
    ax1.set_xlim(380, 750)
    ax1.set_ylim(0, 1.1)
    ax1.grid(True, alpha=0.3)
    
    # Find constructive wavelengths
    for m in range(10):
        lam_c = 2 * n * t / (m + 0.5)
        if 380e-9 <= lam_c <= 750e-9:
            ax1.axvline(lam_c*1e9, color='gray', linestyle=':', alpha=0.5)
            ax1.text(lam_c*1e9, 1.05, f'm={m}', ha='center', fontsize=8)
    
    # Color swatch
    ax2.add_patch(plt.Rectangle((0, 0), 1, 1, facecolor=perceived_color))
    ax2.set_xlim(0, 1)
    ax2.set_ylim(0, 1)
    ax2.set_xticks([])
    ax2.set_yticks([])
    ax2.set_title('Reflected Color')
    ax2.set_aspect('equal')
    
    plt.tight_layout()
    plt.show()

interact(thin_film_interference,
         thickness_nm=FloatSlider(min=50, max=1000, step=10, value=300,
                                  description='Thickness (nm):', style={'description_width': 'initial'}),
         n_film=FloatSlider(min=1.1, max=2.5, step=0.05, value=1.5,
                            description='n (film):', style={'description_width': 'initial'}));
```

## 7. Combined Double-Slit and Single-Slit Pattern

In practice, each slit in a double-slit experiment has a finite width $a$. The observed pattern is the **product** of the double-slit interference pattern and the single-slit diffraction envelope:

$$I(\theta) = I_0 \cos^2\!\left(\frac{\pi d \sin\theta}{\lambda}\right) \left[\frac{\sin\left(\frac{\pi a \sin\theta}{\lambda}\right)}{\frac{\pi a \sin\theta}{\lambda}}\right]^2$$

```python
def combined_pattern(d_um=100.0, a_um=20.0, wavelength_nm=550.0):
    """
    Combined double-slit interference with single-slit diffraction envelope.
    """
    d = d_um * 1e-6
    a = a_um * 1e-6
    lam = wavelength_nm * 1e-9
    
    theta = np.linspace(-0.06, 0.06, 4000)
    theta_safe = np.where(np.abs(theta) < 1e-15, 1e-15, theta)
    
    # Double-slit interference
    interference = np.cos(np.pi * d * np.sin(theta) / lam)**2
    
    # Single-slit diffraction envelope
    beta_half = np.pi * a * np.sin(theta_safe) / lam
    diffraction = (np.sin(beta_half) / beta_half)**2
    
    # Combined
    combined = interference * diffraction
    
    color = wavelength_to_rgb(wavelength_nm)
    
    fig, ax = plt.subplots(figsize=(12, 5))
    ax.plot(np.degrees(theta), diffraction, 'r--', linewidth=1.5, alpha=0.6, label='Diffraction envelope')
    ax.plot(np.degrees(theta), combined, color=color, linewidth=1.5, label='Combined pattern')
    ax.fill_between(np.degrees(theta), combined, alpha=0.2, color=color)
    ax.set_xlabel('Angle (degrees)')
    ax.set_ylabel('Relative Intensity')
    ax.set_title(f'Combined Pattern: d = {d_um:.0f} um, a = {a_um:.0f} um, lambda = {wavelength_nm:.0f} nm')
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

interact(combined_pattern,
         d_um=FloatSlider(min=50, max=300, step=10, value=100,
                          description='d (um):', style={'description_width': 'initial'}),
         a_um=FloatSlider(min=5, max=80, step=5, value=20,
                          description='a (um):', style={'description_width': 'initial'}),
         wavelength_nm=FloatSlider(min=380, max=750, step=10, value=550,
                                   description='lambda (nm):', style={'description_width': 'initial'}));
```

## 8. Summary

| Phenomenon | Key Equation | What to Look For |
|:-----------|:-------------|:------------------|
| Double slit | $d\sin\theta = m\lambda$ | Equally spaced bright fringes |
| Single slit | $a\sin\theta = m\lambda$ (minima) | Broad central max, weaker side maxima |
| Diffraction grating | $d\sin\theta = m\lambda$ | Very sharp, bright principal maxima |
| Thin film | $2nt = (m+\frac{1}{2})\lambda$ | Colored reflections depend on thickness |

### Key Takeaways

1. Interference arises from the superposition of coherent waves
2. The double-slit fringe spacing $\Delta y = \lambda L / d$ increases with wavelength and screen distance
3. Single-slit diffraction produces a pattern dominated by a wide central maximum
4. Diffraction gratings produce much sharper peaks than double slits due to the contribution of many slits
5. Thin-film colors arise from wavelength-selective constructive interference

## 1. Electric Flux

**Electric flux** measures how much electric field "passes through" a surface. Think of it like counting how many field lines pierce through a window.

For a uniform electric field $\vec{E}$ passing through a flat surface of area $A$:

$$\Phi_E = \vec{E} \cdot \vec{A} = E \, A \, \cos\theta$$

where $\theta$ is the angle between the electric field direction and the **surface normal** (the direction perpendicular to the surface).

Key observations:
- When $\theta = 0^\circ$ (field perpendicular to surface): $\Phi_E = EA$ (maximum flux)
- When $\theta = 90^\circ$ (field parallel to surface): $\Phi_E = 0$ (no flux)
- Units: N$\cdot$m$^2$/C (or equivalently V$\cdot$m)

### Interactive Demo 1: Electric Flux Calculator

Adjust the electric field strength, area, and angle to see how the flux changes. Watch how the projection of the surface onto the field direction determines the flux.

```python
def flux_calculator(E=100.0, A=1.0, theta_deg=0.0):
    """
    Interactive electric flux calculator.
    Shows the surface, field lines, surface normal, and computed flux.
    """
    theta = np.radians(theta_deg)
    flux = E * A * np.cos(theta)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
    
    # --- Left panel: Side view of surface and field ---
    ax1.set_xlim(-2, 3)
    ax1.set_ylim(-1.5, 1.5)
    ax1.set_aspect('equal')
    ax1.set_title('Side View: Surface and Electric Field', fontsize=13)
    
    # Draw the surface as a tilted line
    half_len = 0.6 * np.sqrt(A)
    # Surface is perpendicular to the normal
    surface_angle = theta + np.pi / 2
    x_surf = [0 - half_len * np.cos(surface_angle), 0 + half_len * np.cos(surface_angle)]
    y_surf = [0 - half_len * np.sin(surface_angle), 0 + half_len * np.sin(surface_angle)]
    ax1.plot(x_surf, y_surf, 'b-', linewidth=4, label='Surface')
    
    # Draw the surface normal
    normal_len = 0.8
    ax1.annotate('', xy=(normal_len * np.cos(theta), normal_len * np.sin(theta)),
                 xytext=(0, 0),
                 arrowprops=dict(arrowstyle='->', color='green', lw=2))
    ax1.text(normal_len * np.cos(theta) + 0.1, normal_len * np.sin(theta) + 0.1,
             r'$\hat{n}$', fontsize=15, color='green', fontweight='bold')
    
    # Draw electric field arrows (horizontal, pointing right)
    for y_pos in np.linspace(-1.0, 1.0, 7):
        ax1.annotate('', xy=(2.5, y_pos), xytext=(-1.5, y_pos),
                     arrowprops=dict(arrowstyle='->', color='red', lw=1.2, alpha=0.5))
    ax1.text(2.6, 0.0, r'$\vec{E}$', fontsize=15, color='red', fontweight='bold')
    
    # Draw the angle arc
    if abs(theta_deg) > 2:
        angle_arc = np.linspace(0, theta, 30)
        arc_r = 0.4
        ax1.plot(arc_r * np.cos(angle_arc), arc_r * np.sin(angle_arc), 'k-', lw=1.5)
        mid_angle = theta / 2
        ax1.text(0.5 * np.cos(mid_angle), 0.5 * np.sin(mid_angle),
                 f'{theta_deg:.0f}'+r'$^\circ$', fontsize=12, ha='center')
    
    ax1.set_xlabel('x')
    ax1.set_ylabel('y')
    ax1.grid(True, alpha=0.3)
    
    # --- Right panel: Flux bar chart ---
    colors = ['#2196F3' if flux >= 0 else '#f44336']
    ax2.bar(['Electric Flux'], [flux], color=colors, width=0.4, edgecolor='black')
    max_flux = E * A * 1.1 if E * A > 0 else 100
    ax2.set_ylim(-max_flux, max_flux)
    ax2.set_ylabel(r'$\Phi_E$ (N$\cdot$m$^2$/C)', fontsize=13)
    ax2.set_title('Computed Flux', fontsize=13)
    ax2.axhline(y=0, color='black', linewidth=0.8)
    ax2.text(0, flux + (max_flux * 0.05 * np.sign(flux) if flux != 0 else max_flux * 0.05),
             f'{flux:.1f}', ha='center', fontsize=14, fontweight='bold')
    
    fig.suptitle(f'$\\Phi_E = E \\cdot A \\cdot \\cos\\theta = {E:.0f} \\times {A:.2f} \\times \\cos({theta_deg:.0f}^\\circ) = {flux:.1f}$ N$\\cdot$m$^2$/C',
                 fontsize=13, y=1.02)
    plt.tight_layout()
    plt.show()

interact(flux_calculator,
         E=FloatSlider(value=100, min=0, max=500, step=10, description='E (N/C):', style={'description_width': 'initial'}),
         A=FloatSlider(value=1.0, min=0.1, max=3.0, step=0.1, description='A (m^2):', style={'description_width': 'initial'}),
         theta_deg=FloatSlider(value=0, min=0, max=180, step=5, description='Angle (deg):', style={'description_width': 'initial'}));
```

## 2. Gauss's Law

**Gauss's Law** relates the total electric flux through a **closed surface** to the charge enclosed within it:

$$\oint \vec{E} \cdot d\vec{A} = \frac{Q_{\text{enc}}}{\varepsilon_0}$$

where:
- The integral is over a **closed** surface (called a **Gaussian surface**)
- $Q_{\text{enc}}$ is the total charge enclosed by the surface
- $\varepsilon_0 = 8.854 \times 10^{-12}$ C$^2$/(N$\cdot$m$^2$) is the permittivity of free space

**Important points:**
- The Gaussian surface is an **imaginary** mathematical surface -- you choose it!
- Charges outside the surface contribute zero net flux
- Gauss's law is always true, but it is most **useful** when there is high symmetry

### The Three Key Symmetries

| Symmetry | Charge Distribution | Gaussian Surface | E depends on |
|----------|-------------------|-----------------|-------------|
| **Spherical** | Point charge, charged sphere | Sphere | $r$ (distance from center) |
| **Cylindrical** | Infinite line charge, infinite cylinder | Cylinder | $r$ (distance from axis) |
| **Planar** | Infinite charged plane | Pill-box (short cylinder) | Nothing (uniform!) |

## 3. Spherical Symmetry: Charged Sphere

Consider a solid sphere of radius $R$ with total charge $Q$ uniformly distributed throughout its volume.

**Outside the sphere** ($r > R$): The enclosed charge is the total charge $Q$.

$$E(4\pi r^2) = \frac{Q}{\varepsilon_0} \quad \Rightarrow \quad E = \frac{Q}{4\pi \varepsilon_0 r^2} = \frac{kQ}{r^2}$$

This is identical to a **point charge** at the center!

**Inside the sphere** ($r < R$): Only the charge within radius $r$ is enclosed.

$$Q_{\text{enc}} = Q \left(\frac{r}{R}\right)^3$$

$$E = \frac{Q}{4\pi \varepsilon_0} \frac{r}{R^3} = \frac{kQ}{R^3} r$$

The field **increases linearly** from zero at the center.

### Interactive Demo 2: Gaussian Surface Around a Charged Sphere

Move the Gaussian surface radius to see how the enclosed charge and electric field change inside and outside the charged sphere.

```python
def gauss_sphere(r_gauss=1.5, Q=1.0, R=1.0):
    """
    Interactive Gaussian surface around a uniformly charged sphere.
    Left: visual of sphere and Gaussian surface.
    Right: E-field vs r plot with marker at current position.
    """
    eps0 = 8.854e-12
    k = 1 / (4 * np.pi * eps0)
    Q_SI = Q * 1e-6  # microcoulombs
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5.5))
    
    # --- Left: Visual diagram ---
    ax1.set_xlim(-3.5, 3.5)
    ax1.set_ylim(-3.5, 3.5)
    ax1.set_aspect('equal')
    
    # Charged sphere
    sphere = plt.Circle((0, 0), R, color='#FF7043', alpha=0.6, label=f'Charged sphere (R={R:.1f} m)')
    ax1.add_patch(sphere)
    
    # Gaussian surface
    gauss = plt.Circle((0, 0), r_gauss, fill=False, color='blue', linewidth=2.5,
                        linestyle='--', label=f'Gaussian surface (r={r_gauss:.2f} m)')
    ax1.add_patch(gauss)
    
    # Draw some + signs inside the charged sphere
    for angle in np.linspace(0, 2 * np.pi, 8, endpoint=False):
        for rad in [0, 0.4 * R, 0.7 * R]:
            if rad == 0 and angle > 0:
                continue
            ax1.text(rad * np.cos(angle), rad * np.sin(angle), '+',
                     fontsize=10, ha='center', va='center', color='red', fontweight='bold')
    
    # E-field arrows pointing outward from Gaussian surface
    if r_gauss > 0.15:
        if r_gauss < R:
            Q_enc = Q_SI * (r_gauss / R) ** 3
        else:
            Q_enc = Q_SI
        E_val = k * abs(Q_enc) / r_gauss**2 if r_gauss > 0 else 0
        arrow_len = min(0.8, 0.3 + 0.5 * E_val / (k * abs(Q_SI) / R**2)) if E_val > 0 else 0
        for angle in np.linspace(0, 2 * np.pi, 12, endpoint=False):
            x0 = r_gauss * np.cos(angle)
            y0 = r_gauss * np.sin(angle)
            dx = arrow_len * np.cos(angle)
            dy = arrow_len * np.sin(angle)
            ax1.annotate('', xy=(x0 + dx, y0 + dy), xytext=(x0, y0),
                         arrowprops=dict(arrowstyle='->', color='green', lw=1.5))
    
    ax1.legend(loc='upper left', fontsize=10)
    ax1.set_title('Gaussian Surface Diagram', fontsize=13)
    ax1.grid(True, alpha=0.2)
    ax1.set_xlabel('x (m)')
    ax1.set_ylabel('y (m)')
    
    # --- Right: E vs r plot ---
    r = np.linspace(0.01, 3.5, 500)
    E_field = np.where(
        r < R,
        k * abs(Q_SI) * r / R**3,
        k * abs(Q_SI) / r**2
    )
    
    ax2.plot(r, E_field / 1e3, 'b-', linewidth=2, label='E(r)')
    ax2.axvline(x=R, color='red', linestyle=':', alpha=0.7, label=f'Sphere surface R={R:.1f} m')
    
    # Current position marker
    if r_gauss < R:
        E_current = k * abs(Q_SI) * r_gauss / R**3
        Q_enc_val = Q * (r_gauss / R) ** 3
        region = 'INSIDE'
    else:
        E_current = k * abs(Q_SI) / r_gauss**2
        Q_enc_val = Q
        region = 'OUTSIDE'
    
    ax2.plot(r_gauss, E_current / 1e3, 'ro', markersize=12, zorder=5, label=f'Current: r={r_gauss:.2f} m')
    
    ax2.set_xlabel('Distance r (m)', fontsize=12)
    ax2.set_ylabel('E (kN/C)', fontsize=12)
    ax2.set_title('Electric Field vs Distance', fontsize=13)
    ax2.legend(fontsize=10)
    ax2.grid(True, alpha=0.3)
    ax2.set_xlim(0, 3.5)
    
    fig.suptitle(f'Region: {region}  |  $Q_{{enc}}$ = {Q_enc_val:.3f} $\\mu$C  |  E = {E_current/1e3:.2f} kN/C',
                 fontsize=13, y=1.02)
    plt.tight_layout()
    plt.show()

interact(gauss_sphere,
         r_gauss=FloatSlider(value=1.5, min=0.1, max=3.5, step=0.05, description='r (Gauss):', style={'description_width': 'initial'}),
         Q=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='Q (uC):', style={'description_width': 'initial'}),
         R=FloatSlider(value=1.0, min=0.3, max=2.0, step=0.1, description='R (sphere):', style={'description_width': 'initial'}));
```

## 4. Cylindrical Symmetry: Infinite Line Charge

Consider an infinitely long straight wire with uniform linear charge density $\lambda$ (charge per unit length, in C/m).

We choose a **cylindrical Gaussian surface** of radius $r$ and length $L$ centered on the wire.

- The flux through the two flat end caps is **zero** (field is perpendicular to their normals).
- The flux through the curved surface: $E(2\pi r L) = \frac{\lambda L}{\varepsilon_0}$

$$\boxed{E = \frac{\lambda}{2\pi \varepsilon_0 r}}$$

The field falls off as $1/r$ -- slower than a point charge ($1/r^2$) but faster than a plane (constant).

### Interactive Demo 3: Gaussian Surface for Infinite Line Charge

Adjust the linear charge density and the Gaussian cylinder radius to see how the electric field varies.

```python
def gauss_line_charge(r_gauss=0.5, lam=1.0):
    """
    Interactive: E-field from an infinite line charge using Gauss's law.
    Left: cross-section showing wire and cylindrical Gaussian surface.
    Right: E vs r plot.
    """
    eps0 = 8.854e-12
    lam_SI = lam * 1e-6  # uC/m to C/m
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5.5))
    
    # --- Left: Cross-section ---
    ax1.set_xlim(-3, 3)
    ax1.set_ylim(-3, 3)
    ax1.set_aspect('equal')
    
    # Wire (center dot)
    ax1.plot(0, 0, 'ro', markersize=10, label='Line charge (cross section)', zorder=5)
    ax1.text(0.15, 0.15, '+', fontsize=14, color='red', fontweight='bold')
    
    # Gaussian cylinder cross-section
    gauss_circle = plt.Circle((0, 0), r_gauss, fill=False, color='blue',
                               linewidth=2.5, linestyle='--', label=f'Gaussian cyl. (r={r_gauss:.2f} m)')
    ax1.add_patch(gauss_circle)
    
    # E-field arrows pointing radially outward
    E_at_r = lam_SI / (2 * np.pi * eps0 * r_gauss)
    arrow_scale = min(1.0, 0.3 + 0.7 * r_gauss / 2)
    for angle in np.linspace(0, 2 * np.pi, 12, endpoint=False):
        x0 = r_gauss * np.cos(angle)
        y0 = r_gauss * np.sin(angle)
        dx = arrow_scale * np.cos(angle)
        dy = arrow_scale * np.sin(angle)
        ax1.annotate('', xy=(x0 + dx, y0 + dy), xytext=(x0, y0),
                     arrowprops=dict(arrowstyle='->', color='green', lw=1.8))
    
    ax1.set_title('Cross-section View', fontsize=13)
    ax1.set_xlabel('x (m)')
    ax1.set_ylabel('y (m)')
    ax1.legend(loc='upper left', fontsize=10)
    ax1.grid(True, alpha=0.2)
    
    # --- Right: E vs r ---
    r = np.linspace(0.05, 3, 500)
    E_field = lam_SI / (2 * np.pi * eps0 * r)
    
    ax2.plot(r, E_field / 1e3, 'b-', linewidth=2, label=r'$E = \lambda / (2\pi\varepsilon_0 r)$')
    ax2.plot(r_gauss, E_at_r / 1e3, 'ro', markersize=12, zorder=5,
             label=f'r = {r_gauss:.2f} m, E = {E_at_r/1e3:.1f} kN/C')
    
    ax2.set_xlabel('Distance r (m)', fontsize=12)
    ax2.set_ylabel('E (kN/C)', fontsize=12)
    ax2.set_title('Electric Field vs Distance (line charge)', fontsize=13)
    ax2.legend(fontsize=10)
    ax2.grid(True, alpha=0.3)
    ax2.set_xlim(0, 3)
    ax2.set_ylim(0, None)
    
    plt.tight_layout()
    plt.show()

interact(gauss_line_charge,
         r_gauss=FloatSlider(value=0.5, min=0.05, max=3.0, step=0.05, description='r (m):', style={'description_width': 'initial'}),
         lam=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='lambda (uC/m):', style={'description_width': 'initial'}));
```

## 5. Planar Symmetry: Infinite Charged Plane

For an infinite plane with uniform surface charge density $\sigma$ (C/m$^2$), we use a **pill-box** (short cylinder) as the Gaussian surface, with its flat faces parallel to the plane.

The flux goes through both flat faces (each of area $A$):

$$2EA = \frac{\sigma A}{\varepsilon_0}$$

$$\boxed{E = \frac{\sigma}{2\varepsilon_0}}$$

Remarkable result: The electric field is **constant** -- it does not depend on distance from the plane! This is because for every field line that spreads out, new area of the infinite plane compensates.

## 6. Animation: Field Lines Through a Gaussian Surface

Watch how electric field lines from a point charge pass through an expanding and contracting spherical Gaussian surface. Notice that the **total number of lines** (proportional to flux) remains constant regardless of the surface size -- this is Gauss's law in action!

```python
def create_gauss_animation():
    """
    Animation: Electric field lines from a point charge passing through
    an expanding/contracting Gaussian surface.
    """
    fig, ax = plt.subplots(figsize=(7, 7))
    ax.set_xlim(-4, 4)
    ax.set_ylim(-4, 4)
    ax.set_aspect('equal')
    ax.set_title('Field Lines Through a Gaussian Surface', fontsize=14)
    ax.grid(True, alpha=0.2)
    
    # Point charge at origin
    charge_dot, = ax.plot([], [], 'ro', markersize=14, zorder=10)
    charge_label = ax.text(0.2, 0.2, '+Q', fontsize=12, color='red', fontweight='bold', zorder=10)
    
    # Field lines (radial)
    n_lines = 16
    angles = np.linspace(0, 2 * np.pi, n_lines, endpoint=False)
    field_lines = []
    for angle in angles:
        r_vals = np.linspace(0.3, 3.8, 50)
        x_line = r_vals * np.cos(angle)
        y_line = r_vals * np.sin(angle)
        line, = ax.plot(x_line, y_line, 'r-', alpha=0.4, linewidth=1)
        field_lines.append(line)
        # Arrowhead at midpoint
        mid = 25
        ax.annotate('', xy=(x_line[mid + 1], y_line[mid + 1]), xytext=(x_line[mid], y_line[mid]),
                     arrowprops=dict(arrowstyle='->', color='red', lw=1.5))
    
    # Gaussian surface (circle that will animate)
    gauss_circle = plt.Circle((0, 0), 1.0, fill=False, color='blue',
                               linewidth=2.5, linestyle='--')
    ax.add_patch(gauss_circle)
    
    # Flux text
    flux_text = ax.text(-3.8, 3.5, '', fontsize=12, color='blue',
                        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))
    radius_text = ax.text(-3.8, 2.8, '', fontsize=11)
    
    def init():
        charge_dot.set_data([0], [0])
        gauss_circle.set_radius(1.0)
        flux_text.set_text('')
        radius_text.set_text('')
        return [gauss_circle, flux_text, radius_text, charge_dot]
    
    def animate(frame):
        # Radius oscillates between 0.5 and 3.0
        t = frame / 60  # normalize
        r = 1.75 + 1.25 * np.sin(2 * np.pi * t)
        gauss_circle.set_radius(r)
        flux_text.set_text(r'$\Phi_E = Q/\varepsilon_0$ = const!')
        radius_text.set_text(f'Gaussian surface radius: {r:.2f} m')
        return [gauss_circle, flux_text, radius_text, charge_dot]
    
    anim = FuncAnimation(fig, animate, init_func=init, frames=60, interval=80, blit=True)
    plt.close(fig)
    return anim

anim = create_gauss_animation()
HTML(anim.to_jshtml())
```

## 7. Comparison: E-field for Different Symmetries

Let us compare how the electric field behaves with distance for the three geometries:

| Geometry | E-field dependence | Formula |
|----------|-------------------|--------|
| Sphere (point charge) | $E \propto 1/r^2$ | $E = kQ/r^2$ |
| Cylinder (line charge) | $E \propto 1/r$ | $E = \lambda / (2\pi\varepsilon_0 r)$ |
| Plane (surface charge) | $E = $ const | $E = \sigma / (2\varepsilon_0)$ |

### Interactive Demo 4: Compare E-field from Sphere, Cylinder, and Plane

```python
def compare_symmetries(Q=1.0, lam=1.0, sigma=1.0):
    """
    Compare E-field vs distance for spherical, cylindrical, and planar symmetry.
    All charge values in micro-units.
    """
    eps0 = 8.854e-12
    k = 1 / (4 * np.pi * eps0)
    
    Q_SI = Q * 1e-6
    lam_SI = lam * 1e-6
    sigma_SI = sigma * 1e-6
    
    r = np.linspace(0.1, 5, 500)
    
    E_sphere = k * Q_SI / r**2
    E_cylinder = lam_SI / (2 * np.pi * eps0 * r)
    E_plane = np.full_like(r, sigma_SI / (2 * eps0))
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))
    
    # Linear scale
    ax1.plot(r, E_sphere / 1e3, 'r-', linewidth=2, label=r'Sphere: $E \propto 1/r^2$')
    ax1.plot(r, E_cylinder / 1e3, 'b-', linewidth=2, label=r'Cylinder: $E \propto 1/r$')
    ax1.plot(r, E_plane / 1e3, 'g-', linewidth=2, label=r'Plane: $E = $ const')
    ax1.set_xlabel('Distance r (m)', fontsize=12)
    ax1.set_ylabel('E (kN/C)', fontsize=12)
    ax1.set_title('E-field Comparison (Linear Scale)', fontsize=13)
    ax1.legend(fontsize=11)
    ax1.grid(True, alpha=0.3)
    ax1.set_ylim(0, max(E_sphere[10] / 1e3, E_cylinder[10] / 1e3, E_plane[0] / 1e3) * 1.3)
    
    # Log-log scale
    ax2.loglog(r, E_sphere, 'r-', linewidth=2, label=r'Sphere ($\propto r^{-2}$)')
    ax2.loglog(r, E_cylinder, 'b-', linewidth=2, label=r'Cylinder ($\propto r^{-1}$)')
    ax2.loglog(r, E_plane, 'g-', linewidth=2, label=r'Plane ($\propto r^{0}$)')
    ax2.set_xlabel('Distance r (m)', fontsize=12)
    ax2.set_ylabel('E (N/C)', fontsize=12)
    ax2.set_title('E-field Comparison (Log-Log Scale)', fontsize=13)
    ax2.legend(fontsize=11)
    ax2.grid(True, alpha=0.3, which='both')
    
    plt.tight_layout()
    plt.show()

interact(compare_symmetries,
         Q=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='Q (uC):', style={'description_width': 'initial'}),
         lam=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='lambda (uC/m):', style={'description_width': 'initial'}),
         sigma=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='sigma (uC/m^2):', style={'description_width': 'initial'}));
```

## 8. Worked Example 1: E-field of a Uniformly Charged Sphere

**Problem:** A solid insulating sphere of radius $R = 0.10$ m carries a total charge $Q = 2.0 \; \mu$C distributed uniformly. Find the electric field at:

(a) $r = 0.05$ m (inside the sphere)  
(b) $r = 0.20$ m (outside the sphere)

```python
# Worked Example 1: E-field of a uniformly charged sphere
print("=" * 60)
print("WORKED EXAMPLE 1: Uniformly Charged Sphere")
print("=" * 60)

eps0 = 8.854e-12
k = 1 / (4 * np.pi * eps0)
Q = 2.0e-6    # 2.0 uC
R = 0.10       # 0.10 m

print(f"\nGiven: Q = {Q*1e6:.1f} uC, R = {R:.2f} m")
print(f"k = {k:.3e} N m^2/C^2")

# Part (a): Inside (r = 0.05 m)
r_a = 0.05
Q_enc_a = Q * (r_a / R)**3
E_a = k * Q_enc_a / r_a**2

print(f"\n--- Part (a): r = {r_a} m (inside) ---")
print(f"Step 1: Find enclosed charge")
print(f"  Q_enc = Q * (r/R)^3 = {Q*1e6:.1f} * ({r_a}/{R})^3 = {Q_enc_a*1e6:.4f} uC")
print(f"Step 2: Apply Gauss's law: E * 4*pi*r^2 = Q_enc / eps0")
print(f"  E = k * Q_enc / r^2 = {k:.3e} * {Q_enc_a:.3e} / {r_a}^2")
print(f"  E = {E_a:.3e} N/C = {E_a/1e3:.1f} kN/C")

# Alternatively using the linear formula:
E_a_alt = k * Q * r_a / R**3
print(f"  Check (linear formula): E = kQ*r/R^3 = {E_a_alt:.3e} N/C  [matches!]")

# Part (b): Outside (r = 0.20 m)
r_b = 0.20
E_b = k * Q / r_b**2

print(f"\n--- Part (b): r = {r_b} m (outside) ---")
print(f"Step 1: Enclosed charge = total charge Q = {Q*1e6:.1f} uC")
print(f"Step 2: Apply Gauss's law (same as point charge):")
print(f"  E = k * Q / r^2 = {k:.3e} * {Q:.3e} / {r_b}^2")
print(f"  E = {E_b:.3e} N/C = {E_b/1e3:.1f} kN/C")

print(f"\n--- Summary ---")
print(f"  E at r = {r_a} m (inside):  {E_a/1e3:.1f} kN/C")
print(f"  E at r = {R} m (surface): {k*Q/R**2/1e3:.1f} kN/C  [maximum]")
print(f"  E at r = {r_b} m (outside): {E_b/1e3:.1f} kN/C")
```

## 9. Worked Example 2: E-field of an Infinite Charged Plane

**Problem:** A large flat sheet carries a uniform surface charge density $\sigma = 5.0 \; \mu$C/m$^2$.

(a) Find the electric field near the sheet.  
(b) If a second sheet with $\sigma = -5.0 \; \mu$C/m$^2$ is placed parallel to the first, what is the field between and outside the sheets?

```python
# Worked Example 2: E-field of an infinite charged plane
print("=" * 60)
print("WORKED EXAMPLE 2: Infinite Charged Plane")
print("=" * 60)

eps0 = 8.854e-12
sigma = 5.0e-6  # 5.0 uC/m^2

print(f"\nGiven: sigma = {sigma*1e6:.1f} uC/m^2")

# Part (a)
E_plane = sigma / (2 * eps0)
print(f"\n--- Part (a): Single infinite plane ---")
print(f"Using Gauss's law with a pill-box surface:")
print(f"  2*E*A = sigma*A / eps0")
print(f"  E = sigma / (2*eps0) = {sigma:.3e} / (2 * {eps0:.3e})")
print(f"  E = {E_plane:.3e} N/C = {E_plane/1e3:.1f} kN/C")
print(f"  Direction: perpendicular to the plane, pointing away from it")

# Part (b)
E_between = sigma / eps0
E_outside = 0
print(f"\n--- Part (b): Two parallel sheets (+ and -) ---")
print(f"  Between the sheets: fields ADD")
print(f"    E = sigma / eps0 = {E_between:.3e} N/C = {E_between/1e3:.1f} kN/C")
print(f"  Outside the sheets: fields CANCEL")
print(f"    E = 0 N/C")
print(f"\n  This is the basis for a PARALLEL PLATE CAPACITOR!")
print(f"  (We will study this in detail next week)")
```

## 11. Interactive Summary: Choosing the Right Gaussian Surface

```python
def gauss_summary(geometry='Sphere'):
    """
    Summary visualization showing the Gaussian surface, field pattern,
    and formula for each geometry.
    """
    fig, ax = plt.subplots(figsize=(8, 7))
    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.2)
    
    if geometry == 'Sphere':
        # Point charge + spherical Gaussian surface
        ax.plot(0, 0, 'ro', markersize=15, zorder=10)
        ax.text(0, 0, '+', fontsize=14, ha='center', va='center', color='white', fontweight='bold', zorder=11)
        gauss = plt.Circle((0, 0), 2.0, fill=False, color='blue', linewidth=2.5, linestyle='--')
        ax.add_patch(gauss)
        for angle in np.linspace(0, 2 * np.pi, 16, endpoint=False):
            x0 = 2.0 * np.cos(angle)
            y0 = 2.0 * np.sin(angle)
            dx = 0.6 * np.cos(angle)
            dy = 0.6 * np.sin(angle)
            ax.annotate('', xy=(x0 + dx, y0 + dy), xytext=(x0, y0),
                         arrowprops=dict(arrowstyle='->', color='green', lw=1.5))
        ax.set_title('Spherical Symmetry: Point Charge / Charged Sphere\n'
                      r'$E = kQ/r^2$ (outside),  $E = kQr/R^3$ (inside)', fontsize=13)
    
    elif geometry == 'Cylinder':
        # Line charge + cylindrical Gaussian surface
        ax.plot([0, 0], [-2.5, 2.5], 'r-', linewidth=4, label='Line charge')
        for y in np.linspace(-2, 2, 5):
            ax.text(0.15, y, '+', fontsize=10, color='red', fontweight='bold')
        rect = patches.FancyBboxPatch((-1.5, -1.8), 3.0, 3.6, boxstyle='round,pad=0.1',
                                       fill=False, color='blue', linewidth=2.5, linestyle='--')
        ax.add_patch(rect)
        for y_pos in np.linspace(-1.5, 1.5, 5):
            ax.annotate('', xy=(2.2, y_pos), xytext=(1.5, y_pos),
                         arrowprops=dict(arrowstyle='->', color='green', lw=1.5))
            ax.annotate('', xy=(-2.2, y_pos), xytext=(-1.5, y_pos),
                         arrowprops=dict(arrowstyle='->', color='green', lw=1.5))
        ax.set_title('Cylindrical Symmetry: Line Charge\n'
                      r'$E = \lambda / (2\pi\varepsilon_0 r)$', fontsize=13)
    
    elif geometry == 'Plane':
        # Charged plane + pill-box
        ax.fill_between([-0.05, 0.05], -2.8, 2.8, color='red', alpha=0.5)
        for y in np.linspace(-2.5, 2.5, 8):
            ax.text(0, y, '+', fontsize=10, ha='center', va='center', color='darkred', fontweight='bold')
        rect = patches.FancyBboxPatch((-1.5, -0.8), 3.0, 1.6, boxstyle='round,pad=0.1',
                                       fill=False, color='blue', linewidth=2.5, linestyle='--')
        ax.add_patch(rect)
        for y_pos in np.linspace(-0.5, 0.5, 3):
            ax.annotate('', xy=(2.2, y_pos), xytext=(1.5, y_pos),
                         arrowprops=dict(arrowstyle='->', color='green', lw=1.5))
            ax.annotate('', xy=(-2.2, y_pos), xytext=(-1.5, y_pos),
                         arrowprops=dict(arrowstyle='->', color='green', lw=1.5))
        ax.set_title('Planar Symmetry: Infinite Charged Plane\n'
                      r'$E = \sigma / (2\varepsilon_0)$', fontsize=13)
    
    ax.text(-2.8, -2.8, 'Blue dashed: Gaussian surface\nGreen arrows: E-field direction',
            fontsize=10, bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
    plt.tight_layout()
    plt.show()

interact(gauss_summary,
         geometry=widgets.Dropdown(options=['Sphere', 'Cylinder', 'Plane'],
                                   value='Sphere', description='Geometry:'));
```

## 12. Key Formulas Summary

| Concept | Formula |
|---------|--------|
| Electric flux (uniform field) | $\Phi_E = E A \cos\theta$ |
| Gauss's law | $\oint \vec{E} \cdot d\vec{A} = Q_{\text{enc}} / \varepsilon_0$ |
| Spherical (outside) | $E = kQ / r^2$ |
| Spherical (inside uniform) | $E = kQr / R^3$ |
| Cylindrical (line charge) | $E = \lambda / (2\pi\varepsilon_0 r)$ |
| Planar (infinite plane) | $E = \sigma / (2\varepsilon_0)$ |
| Two parallel planes ($\pm\sigma$) | $E_{\text{between}} = \sigma/\varepsilon_0$, $E_{\text{outside}} = 0$ |

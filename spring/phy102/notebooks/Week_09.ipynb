## 1. Magnetic Flux

**Magnetic flux** measures how much magnetic field passes through a surface.

$$\Phi_B = \vec{B} \cdot \vec{A} = B\,A\,\cos\theta$$

where:
- $B$ = magnitude of the magnetic field (T)
- $A$ = area of the loop (m$^2$)
- $\theta$ = angle between $\vec{B}$ and the area normal $\hat{n}$
- Units: **Weber (Wb)** = T $\cdot$ m$^2$

When $\theta = 0$, the field is perpendicular to the loop surface and flux is maximum.  
When $\theta = 90°$, the field is parallel to the surface and flux is zero.

### Interactive: Magnetic Flux Calculator

Use the sliders below to adjust the magnetic field strength $B$, loop area $A$, and the angle $\theta$. The left panel shows the geometric arrangement and the right panel displays the calculated flux.

```python
def plot_flux(B=0.5, A=0.04, theta_deg=0):
    theta = np.radians(theta_deg)
    flux = B * A * np.cos(theta)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

    # Left panel: geometric picture
    ax1.set_xlim(-1.5, 1.5)
    ax1.set_ylim(-1.5, 1.5)
    ax1.set_aspect('equal')
    ax1.set_title('Loop and Field Orientation')

    # Draw the loop as an ellipse (tilted view)
    width = 1.2
    height = 1.2 * abs(np.cos(theta))
    ellipse = patches.Ellipse((0, 0), width, max(height, 0.05),
                               fill=False, edgecolor='steelblue', linewidth=2.5)
    ax1.add_patch(ellipse)

    # Area normal vector
    n_x = np.sin(theta)
    n_y = np.cos(theta)
    ax1.annotate('', xy=(0.7*n_x, 0.7*n_y), xytext=(0, 0),
                 arrowprops=dict(arrowstyle='->', color='green', lw=2))
    ax1.text(0.7*n_x + 0.1, 0.7*n_y + 0.1, r'$\hat{n}$', fontsize=14, color='green')

    # B field arrows (pointing up)
    for x in [-0.8, -0.4, 0.0, 0.4, 0.8]:
        ax1.annotate('', xy=(x, 1.2), xytext=(x, -1.2),
                     arrowprops=dict(arrowstyle='->', color='red', lw=1, alpha=0.5))
    ax1.text(1.0, 0.0, r'$\vec{B}$', fontsize=14, color='red')

    # Angle arc
    arc_angles = np.linspace(np.pi/2, np.pi/2 - theta, 30)
    arc_r = 0.4
    ax1.plot(arc_r*np.cos(arc_angles), arc_r*np.sin(arc_angles), 'k--', lw=1)
    ax1.text(0.15, 0.45, f'{theta_deg:.0f}' + r'$^\circ$', fontsize=11)

    ax1.set_xlabel('x')
    ax1.set_ylabel('y')
    ax1.grid(True, alpha=0.3)

    # Right panel: flux bar
    max_flux = 2.0 * 0.10  # max B * max A
    colors = 'dodgerblue' if flux >= 0 else 'orangered'
    ax2.barh(['$\\Phi_B$'], [flux], color=colors, height=0.4, edgecolor='black')
    ax2.set_xlim(-max_flux * 1.1, max_flux * 1.1)
    ax2.axvline(0, color='black', lw=0.8)
    ax2.set_xlabel('Magnetic Flux (Wb)')
    ax2.set_title(f'$\\Phi_B = B A \\cos\\theta = {flux:.4f}$ Wb')
    ax2.grid(True, alpha=0.3, axis='x')

    plt.tight_layout()
    plt.show()

interact(plot_flux,
         B=FloatSlider(min=0.0, max=2.0, step=0.05, value=0.5,
                       description='B (T)', style={'description_width': '60px'}),
         A=FloatSlider(min=0.001, max=0.10, step=0.001, value=0.04,
                       description='A (m²)', readout_format='.3f',
                       style={'description_width': '60px'}),
         theta_deg=FloatSlider(min=0, max=180, step=1, value=0,
                               description='θ (deg)', style={'description_width': '60px'}));
```

## 2. Faraday's Law of Induction

Michael Faraday discovered (1831) that a **changing** magnetic flux through a loop induces an electromotive force (EMF):

$$\mathcal{E} = -\frac{d\Phi_B}{dt}$$

For a coil with $N$ turns:

$$\mathcal{E} = -N\,\frac{d\Phi_B}{dt}$$

Key points:
- The **minus sign** is Lenz's law (we will discuss this soon).
- The EMF is induced whether or not there is a closed circuit.
- If the circuit is closed, the EMF drives an induced **current**.
- The flux can change because $B$, $A$, or $\theta$ changes with time.

### Interactive: Faraday's Law -- Changing B Field Through a Loop

Suppose the magnetic field through a loop of area $A$ changes linearly from $B_i$ to $B_f$ over a time $\Delta t$. The induced EMF is:

$$\mathcal{E} = -N\,A\,\frac{\Delta B}{\Delta t}$$

Adjust the parameters below and observe the induced EMF.

```python
def faraday_demo(N=1, A=0.05, Bi=0.0, Bf=1.0, dt=0.5):
    dB = Bf - Bi
    emf = -N * A * dB / dt

    t = np.linspace(0, 1.0, 300)
    # B rises linearly from Bi to Bf between t=0.2 and t=0.2+dt, constant outside
    t_start = 0.1
    t_end = t_start + dt
    B_t = np.piecewise(t,
                       [t < t_start, (t >= t_start) & (t <= t_end), t > t_end],
                       [Bi, lambda t: Bi + dB * (t - t_start) / dt, Bf])
    flux_t = N * A * B_t
    emf_t = -np.gradient(flux_t, t)

    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 8), sharex=True)

    ax1.plot(t, B_t, 'b-', lw=2)
    ax1.set_ylabel('B (T)')
    ax1.set_title('Magnetic Field vs Time')
    ax1.grid(True, alpha=0.3)

    ax2.plot(t, flux_t * 1e3, 'g-', lw=2)
    ax2.set_ylabel('$\\Phi_B$ (mWb)')
    ax2.set_title('Magnetic Flux vs Time')
    ax2.grid(True, alpha=0.3)

    ax3.plot(t, emf_t * 1e3, 'r-', lw=2)
    ax3.set_ylabel('$\\mathcal{E}$ (mV)')
    ax3.set_xlabel('Time (s)')
    ax3.set_title(f'Induced EMF vs Time   |   Peak EMF = {emf*1e3:.2f} mV')
    ax3.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

interact(faraday_demo,
         N=IntSlider(min=1, max=100, step=1, value=10, description='N turns'),
         A=FloatSlider(min=0.001, max=0.1, step=0.001, value=0.05,
                       description='A (m²)', readout_format='.3f'),
         Bi=FloatSlider(min=0.0, max=2.0, step=0.05, value=0.0, description='B_i (T)'),
         Bf=FloatSlider(min=0.0, max=2.0, step=0.05, value=1.0, description='B_f (T)'),
         dt=FloatSlider(min=0.05, max=0.8, step=0.05, value=0.5, description='Δt (s)'));
```

## 3. Lenz's Law

Lenz's law gives the **direction** of the induced current:

> The induced current flows in a direction such that the magnetic field it creates **opposes the change** in flux that produced it.

This is the physical meaning of the minus sign in Faraday's law.

**How to apply Lenz's law:**
1. Determine whether the flux through the loop is increasing or decreasing.
2. The induced current creates a field that opposes that change:
   - If flux is **increasing**, induced B opposes the external B (points opposite).
   - If flux is **decreasing**, induced B supports the external B (points same way).
3. Use the right-hand rule to find the current direction from the induced B direction.

### Interactive: Lenz's Law Direction Finder

A bar magnet approaches or recedes from a conducting loop. Observe how the induced current direction changes based on whether the magnet approaches or moves away, and which pole faces the loop.

```python
def lenz_direction(pole='North', motion='Approaching'):
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.set_xlim(-3, 3)
    ax.set_ylim(-2, 2)
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title("Lenz's Law: Magnet and Loop", fontsize=15)

    # Draw the loop (circle)
    circle = plt.Circle((1.5, 0), 0.8, fill=False, edgecolor='steelblue', lw=3)
    ax.add_patch(circle)
    ax.text(1.5, -1.2, 'Conducting\nLoop', ha='center', fontsize=10, color='steelblue')

    # Draw the magnet
    magnet_x = -1.5 if motion == 'Approaching' else -0.5
    if pole == 'North':
        colors_lr = ['red', 'blue']
        labels_lr = ['N', 'S']
    else:
        colors_lr = ['blue', 'red']
        labels_lr = ['S', 'N']

    rect1 = patches.FancyBboxPatch((magnet_x - 0.8, -0.3), 0.8, 0.6,
                                    boxstyle='round,pad=0.05', facecolor=colors_lr[0], alpha=0.7)
    rect2 = patches.FancyBboxPatch((magnet_x, -0.3), 0.8, 0.6,
                                    boxstyle='round,pad=0.05', facecolor=colors_lr[1], alpha=0.7)
    ax.add_patch(rect1)
    ax.add_patch(rect2)
    ax.text(magnet_x - 0.4, 0, labels_lr[0], ha='center', va='center', fontsize=14,
            fontweight='bold', color='white')
    ax.text(magnet_x + 0.4, 0, labels_lr[1], ha='center', va='center', fontsize=14,
            fontweight='bold', color='white')

    # Motion arrow
    if motion == 'Approaching':
        ax.annotate('', xy=(magnet_x + 1.2, 0), xytext=(magnet_x + 0.85, 0),
                     arrowprops=dict(arrowstyle='->', color='black', lw=2.5))
        ax.text(magnet_x + 1.0, 0.5, 'Moving\ntoward loop', ha='center', fontsize=9)
    else:
        ax.annotate('', xy=(magnet_x - 0.5, 0), xytext=(magnet_x - 0.1, 0),
                     arrowprops=dict(arrowstyle='->', color='black', lw=2.5))
        ax.text(magnet_x - 0.3, 0.5, 'Moving\naway', ha='center', fontsize=9)

    # Determine physics
    # B field from N pole points away from N (to the right if N faces loop)
    if pole == 'North':
        b_dir_through_loop = 'right (into loop)'
    else:
        b_dir_through_loop = 'left (out of loop)'

    if (pole == 'North' and motion == 'Approaching') or \
       (pole == 'South' and motion == 'Receding'):
        flux_change = 'INCREASING'
        current_dir = 'Counterclockwise (as seen from magnet side)'
        induced_opposes = 'Induced B opposes external B (repels magnet)'
        # Draw CCW arrow on loop
        arc_t = np.linspace(0.3, 2*np.pi - 0.3, 50)
        ax.plot(1.5 + 0.55*np.cos(arc_t), 0.55*np.sin(arc_t), 'orange', lw=2.5)
        ax.annotate('', xy=(1.5 + 0.55*np.cos(2*np.pi-0.3), 0.55*np.sin(2*np.pi-0.3)),
                     xytext=(1.5 + 0.55*np.cos(2*np.pi-0.5), 0.55*np.sin(2*np.pi-0.5)),
                     arrowprops=dict(arrowstyle='->', color='orange', lw=2))
    else:
        flux_change = 'DECREASING'
        current_dir = 'Clockwise (as seen from magnet side)'
        induced_opposes = 'Induced B supports external B (attracts magnet)'
        # Draw CW arrow on loop
        arc_t = np.linspace(0.3, 2*np.pi - 0.3, 50)
        ax.plot(1.5 + 0.55*np.cos(-arc_t), 0.55*np.sin(-arc_t), 'orange', lw=2.5)
        ax.annotate('', xy=(1.5 + 0.55*np.cos(-(2*np.pi-0.3)), 0.55*np.sin(-(2*np.pi-0.3))),
                     xytext=(1.5 + 0.55*np.cos(-(2*np.pi-0.5)), 0.55*np.sin(-(2*np.pi-0.5))),
                     arrowprops=dict(arrowstyle='->', color='orange', lw=2))

    info = (f"Flux through loop is {flux_change}\n"
            f"Induced current: {current_dir}\n"
            f"{induced_opposes}")
    ax.text(0, -1.7, info, ha='center', fontsize=11,
            bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))

    plt.tight_layout()
    plt.show()

interact(lenz_direction,
         pole=widgets.Dropdown(options=['North', 'South'], value='North',
                               description='Pole facing loop:'),
         motion=widgets.Dropdown(options=['Approaching', 'Receding'], value='Approaching',
                                 description='Magnet is:'));
```

## 4. Motional EMF

When a conducting bar of length $L$ moves with velocity $v$ through a uniform magnetic field $B$, a **motional EMF** is induced:

$$\mathcal{E} = BLv$$

**Physical picture:** Free charges inside the moving conductor experience a magnetic force $\vec{F} = q\vec{v} \times \vec{B}$, which pushes positive charges to one end and negative charges to the other, creating a potential difference.

### Bar on Rails

A classic setup: a conducting bar slides along two parallel rails separated by distance $L$, connected by a resistor $R$, in a uniform field $B$ (perpendicular to the rail plane).

- Induced EMF: $\mathcal{E} = BLv$
- Induced current: $I = \mathcal{E}/R = BLv/R$
- Force on bar (opposing motion): $F = BIL = B^2L^2v/R$

### Animation: Bar Sliding on Rails

Watch the conducting bar slide to the right on the rails. The enclosed area increases, so the flux increases, and an EMF is induced. The bottom plot shows the EMF and current as functions of time.

```python
# Animation: sliding bar on rails
B_val = 0.5   # T
L_val = 0.3   # m (rail separation)
v_val = 2.0   # m/s
R_val = 5.0   # Ohm

fig_bar, (ax_bar, ax_emf) = plt.subplots(2, 1, figsize=(10, 7),
                                          gridspec_kw={'height_ratios': [2, 1]})

# Rail setup
rail_length = 6.0
ax_bar.set_xlim(-0.5, rail_length)
ax_bar.set_ylim(-0.5, 2.0)
ax_bar.set_aspect('equal')
ax_bar.set_title('Bar Sliding on Rails in Uniform B Field (into page)')

# Draw rails
ax_bar.plot([0, rail_length], [0, 0], 'k-', lw=3)
ax_bar.plot([0, rail_length], [L_val * 4, L_val * 4], 'k-', lw=3)  # scaled for visibility
# Resistor symbol
ax_bar.plot([0, 0], [0, L_val*4], 'k-', lw=2)
ax_bar.text(-0.3, L_val*2, f'R={R_val}Ω', fontsize=10, ha='center', rotation=90)

# B field markers (x's for into page)
for xi in np.arange(0.5, rail_length, 0.8):
    for yi in np.arange(0.2, L_val*4, 0.4):
        ax_bar.plot(xi, yi, 'x', color='gray', markersize=5, alpha=0.4)
ax_bar.text(rail_length - 0.3, 1.5, r'$\vec{B}$ (into page)', fontsize=10, color='gray')

bar_line, = ax_bar.plot([], [], 'r-', lw=5)
bar_label = ax_bar.text(0, 1.4, '', fontsize=10, color='red', ha='center')
area_patch = ax_bar.fill_betweenx([0, L_val*4], 0, 0, alpha=0.15, color='cyan')

n_frames = 80
t_max = (rail_length - 1.0) / v_val
times = np.linspace(0, t_max, n_frames)

ax_emf.set_xlim(0, t_max)
emf_max = B_val * L_val * v_val
ax_emf.set_ylim(0, emf_max * 1.3 * 1e3)
ax_emf.set_xlabel('Time (s)')
ax_emf.set_ylabel('EMF (mV)')
ax_emf.set_title(f'Induced EMF = BLv = {emf_max*1e3:.1f} mV (constant speed)')
ax_emf.axhline(emf_max * 1e3, color='red', ls='--', alpha=0.5)
ax_emf.grid(True, alpha=0.3)
emf_line, = ax_emf.plot([], [], 'r-', lw=2)

plt.tight_layout()

def init_bar():
    bar_line.set_data([], [])
    emf_line.set_data([], [])
    return bar_line, emf_line

def animate_bar(i):
    t = times[i]
    x_pos = 0.5 + v_val * t

    bar_line.set_data([x_pos, x_pos], [0, L_val * 4])
    bar_label.set_position((x_pos, L_val * 4 + 0.15))
    bar_label.set_text(f'v = {v_val} m/s →')

    # Update shaded area
    for coll in ax_bar.collections:
        coll.remove()
    ax_bar.fill_betweenx([0, L_val*4], 0, x_pos, alpha=0.15, color='cyan')
    # Redraw B field markers
    # (already static, so no need)

    emf_line.set_data(times[:i+1], [emf_max * 1e3] * (i + 1))

    return bar_line, emf_line

anim_bar = FuncAnimation(fig_bar, animate_bar, init_func=init_bar,
                          frames=n_frames, interval=60, blit=False)
plt.close(fig_bar)
HTML(anim_bar.to_jshtml())
```

## 5. AC Generators

An **AC generator** converts mechanical rotation into electrical energy using electromagnetic induction.

A rectangular coil of $N$ turns and area $A$ rotates with angular velocity $\omega$ in a uniform magnetic field $B$.

The flux at time $t$:
$$\Phi_B(t) = NBA\cos(\omega t)$$

The induced EMF (by Faraday's law):
$$\mathcal{E}(t) = NBA\omega\,\sin(\omega t) = \mathcal{E}_0\,\sin(\omega t)$$

where $\mathcal{E}_0 = NBA\omega$ is the **peak EMF**.

Key observations:
- The output is **sinusoidal** -- this is why generators produce AC.
- Peak EMF increases with $N$, $B$, $A$, and $\omega$ (rotation speed).
- The frequency of the AC output equals the rotation frequency: $f = \omega/(2\pi)$.

### Animation: Rotating Coil Generator

Watch the coil rotate inside the magnetic field. The right panel shows the sinusoidal EMF output in real time.

```python
# Animation: rotating coil in B field (generator)
N_gen = 50
B_gen = 0.4    # T
A_gen = 0.02   # m^2
omega_gen = 2 * np.pi * 2  # 2 Hz rotation
E0_gen = N_gen * B_gen * A_gen * omega_gen

fig_gen, (ax_coil, ax_signal) = plt.subplots(1, 2, figsize=(13, 5))

# Left: coil view
ax_coil.set_xlim(-2, 2)
ax_coil.set_ylim(-2, 2)
ax_coil.set_aspect('equal')
ax_coil.set_title('Rotating Coil (side view)')
ax_coil.set_xlabel('x')
ax_coil.set_ylabel('z')

# Draw B field arrows
for y_pos in [-1.2, -0.4, 0.4, 1.2]:
    ax_coil.annotate('', xy=(1.8, y_pos), xytext=(-1.8, y_pos),
                      arrowprops=dict(arrowstyle='->', color='red', lw=1, alpha=0.4))
ax_coil.text(1.5, 1.6, r'$\vec{B}$', fontsize=14, color='red')

coil_line, = ax_coil.plot([], [], 'b-', lw=4)
normal_arrow = ax_coil.annotate('', xy=(0, 0), xytext=(0, 0),
                                 arrowprops=dict(arrowstyle='->', color='green', lw=2))
angle_text = ax_coil.text(-1.8, -1.8, '', fontsize=11)

# Right: EMF signal
t_full = np.linspace(0, 2.0, 500)
emf_full = E0_gen * np.sin(omega_gen * t_full)
ax_signal.set_xlim(0, 2.0)
ax_signal.set_ylim(-E0_gen * 1.2, E0_gen * 1.2)
ax_signal.set_xlabel('Time (s)')
ax_signal.set_ylabel('EMF (V)')
ax_signal.set_title(f'Generator Output  |  $\\mathcal{{E}}_0$ = {E0_gen:.2f} V')
ax_signal.grid(True, alpha=0.3)
ax_signal.axhline(0, color='black', lw=0.5)
ax_signal.plot(t_full, emf_full, 'r--', alpha=0.3, lw=1)
signal_line, = ax_signal.plot([], [], 'r-', lw=2)
dot_marker, = ax_signal.plot([], [], 'ro', markersize=8)

plt.tight_layout()

n_gen_frames = 120
t_gen = np.linspace(0, 2.0, n_gen_frames)

def init_gen():
    coil_line.set_data([], [])
    signal_line.set_data([], [])
    dot_marker.set_data([], [])
    return coil_line, signal_line, dot_marker

def animate_gen(i):
    t = t_gen[i]
    theta = omega_gen * t

    # Coil appears as a line in side view, width depends on cos(theta)
    half_w = 0.8 * np.cos(theta)
    coil_line.set_data([half_w, -half_w], [-1.0, 1.0])

    # Normal vector
    nx = 1.0 * np.sin(theta)
    ny = 0  # normal is in the x-direction when coil face is in y-z plane
    # For side view: normal rotates in x direction
    if hasattr(animate_gen, 'arrow'):
        animate_gen.arrow.remove()
    new_arrow = ax_coil.annotate('', xy=(nx, 0), xytext=(0, 0),
                                  arrowprops=dict(arrowstyle='->', color='green', lw=2.5))
    animate_gen.arrow = new_arrow

    angle_text.set_text(f'$\\omega t$ = {np.degrees(theta) % 360:.0f}$^\\circ$')

    # EMF trace
    idx = int(i * len(t_full) / n_gen_frames)
    signal_line.set_data(t_full[:idx], emf_full[:idx])
    emf_now = E0_gen * np.sin(theta)
    dot_marker.set_data([t], [emf_now])

    return coil_line, signal_line, dot_marker

# Initialize arrow reference
animate_gen.arrow = normal_arrow

anim_gen = FuncAnimation(fig_gen, animate_gen, init_func=init_gen,
                          frames=n_gen_frames, interval=50, blit=False)
plt.close(fig_gen)
HTML(anim_gen.to_jshtml())
```

## 6. Eddy Currents

When a bulk conductor (not just a wire loop) experiences a changing magnetic flux, **eddy currents** are induced throughout the volume of the conductor.

**Effects of eddy currents:**
- **Heating** (energy dissipation): Used in induction cooktops and induction furnaces.
- **Braking** (opposing motion): Used in magnetic braking systems for trains and roller coasters.
- **Energy loss in transformers**: Undesirable; reduced by using laminated cores.

**Reducing eddy currents:**
- Use **laminated** (layered) cores with insulating layers between thin metal sheets.
- This forces eddy currents into smaller loops, reducing their magnitude.

## 7. Worked Examples

### Example 1: EMF from Changing Flux

A circular coil of 200 turns and radius 5 cm is placed in a uniform magnetic field. The field increases uniformly from 0 to 0.8 T in 0.1 s. What is the magnitude of the induced EMF?

```python
# Worked Example 1: EMF from changing flux
N_ex1 = 200
r_ex1 = 0.05        # m
A_ex1 = np.pi * r_ex1**2
dB_ex1 = 0.8 - 0.0  # T
dt_ex1 = 0.1         # s

emf_ex1 = N_ex1 * A_ex1 * dB_ex1 / dt_ex1

print("Worked Example 1: EMF from Changing Flux")
print("=" * 45)
print(f"Given:  N = {N_ex1} turns")
print(f"        r = {r_ex1*100:.0f} cm = {r_ex1} m")
print(f"        ΔB = {dB_ex1} T in Δt = {dt_ex1} s")
print(f"\nStep 1: Area = πr² = π × ({r_ex1})² = {A_ex1:.6f} m²")
print(f"Step 2: |EMF| = N × A × ΔB/Δt")
print(f"             = {N_ex1} × {A_ex1:.6f} × {dB_ex1}/{dt_ex1}")
print(f"             = {emf_ex1:.4f} V")
print(f"\n→ The induced EMF is {emf_ex1:.2f} V ≈ {emf_ex1:.1f} V")
```

### Example 2: Generator Output

A generator has a coil of 100 turns, each of area 0.01 m². The coil rotates at 60 Hz in a magnetic field of 0.5 T. Find:
(a) The peak EMF  
(b) The EMF at $t = 5$ ms  
(c) The RMS EMF

```python
# Worked Example 2: Generator output
N_ex2 = 100
A_ex2 = 0.01     # m^2
f_ex2 = 60       # Hz
B_ex2 = 0.5      # T
omega_ex2 = 2 * np.pi * f_ex2

E0_ex2 = N_ex2 * B_ex2 * A_ex2 * omega_ex2
t_point = 5e-3   # 5 ms
emf_at_t = E0_ex2 * np.sin(omega_ex2 * t_point)
E_rms = E0_ex2 / np.sqrt(2)

print("Worked Example 2: Generator Output")
print("=" * 45)
print(f"Given:  N = {N_ex2}, A = {A_ex2} m², f = {f_ex2} Hz, B = {B_ex2} T")
print(f"        ω = 2πf = {omega_ex2:.2f} rad/s")
print(f"\n(a) Peak EMF:")
print(f"    E₀ = NBAω = {N_ex2} × {B_ex2} × {A_ex2} × {omega_ex2:.2f}")
print(f"    E₀ = {E0_ex2:.2f} V")
print(f"\n(b) EMF at t = {t_point*1e3:.0f} ms:")
print(f"    E = E₀ sin(ωt) = {E0_ex2:.2f} × sin({omega_ex2:.2f} × {t_point})")
print(f"    E = {E0_ex2:.2f} × sin({omega_ex2 * t_point:.4f})")
print(f"    E = {E0_ex2:.2f} × {np.sin(omega_ex2 * t_point):.4f}")
print(f"    E = {emf_at_t:.2f} V")
print(f"\n(c) RMS EMF:")
print(f"    E_rms = E₀/√2 = {E0_ex2:.2f}/√2 = {E_rms:.2f} V")
```

### Example 3: Motional EMF -- Sliding Bar

A metal bar of length 0.50 m slides at 3.0 m/s along conducting rails in a 0.80 T field. The rails are connected by a 6.0 $\Omega$ resistor. Find:
(a) The induced EMF  
(b) The induced current  
(c) The force needed to maintain constant velocity

```python
# Worked Example 3: Motional EMF
L_ex3 = 0.50    # m
v_ex3 = 3.0     # m/s
B_ex3 = 0.80    # T
R_ex3 = 6.0     # Ohm

emf_ex3 = B_ex3 * L_ex3 * v_ex3
I_ex3 = emf_ex3 / R_ex3
F_ex3 = B_ex3 * I_ex3 * L_ex3
P_ex3 = emf_ex3 * I_ex3

print("Worked Example 3: Motional EMF (Sliding Bar)")
print("=" * 48)
print(f"Given:  L = {L_ex3} m, v = {v_ex3} m/s, B = {B_ex3} T, R = {R_ex3} Ω")
print(f"\n(a) Induced EMF:")
print(f"    E = BLv = {B_ex3} × {L_ex3} × {v_ex3} = {emf_ex3:.2f} V")
print(f"\n(b) Induced current:")
print(f"    I = E/R = {emf_ex3:.2f}/{R_ex3} = {I_ex3:.3f} A = {I_ex3*1e3:.1f} mA")
print(f"\n(c) Force to maintain constant velocity:")
print(f"    F = BIL = {B_ex3} × {I_ex3:.3f} × {L_ex3}")
print(f"    F = {F_ex3:.4f} N = {F_ex3*1e3:.1f} mN")
print(f"\n    Power dissipated in R: P = EI = {P_ex3:.4f} W = {P_ex3*1e3:.1f} mW")
print(f"    Power input by force:  P = Fv = {F_ex3:.4f} × {v_ex3} = {F_ex3*v_ex3:.4f} W  ✓")
```

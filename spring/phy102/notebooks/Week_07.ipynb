## 1. The Magnetic Field

The **magnetic field** $\vec{B}$ is a vector field that exerts forces on moving charges and current-carrying wires.

Key properties:
- The SI unit of magnetic field is the **Tesla** (T): $1 \; \text{T} = 1 \; \frac{\text{kg}}{\text{A} \cdot \text{s}^2}$
- Earth's magnetic field is approximately $5 \times 10^{-5}$ T
- Unlike electric field lines, magnetic field lines always form **closed loops** (no magnetic monopoles)
- Magnetic fields are produced by moving charges (currents) and by changing electric fields

---
## 2. The Lorentz Force

When a charged particle moves through a magnetic field, it experiences a force:

$$\vec{F} = q\vec{v} \times \vec{B}$$

The magnitude of this force is:

$$F = |q| v B \sin\theta$$

where $\theta$ is the angle between $\vec{v}$ and $\vec{B}$.

**Important properties:**
- The force is **perpendicular** to both $\vec{v}$ and $\vec{B}$
- The magnetic force does **no work** on the particle (it changes direction, not speed)
- If $\vec{v} \parallel \vec{B}$, then $F = 0$

The complete Lorentz force (including electric field) is:

$$\vec{F} = q(\vec{E} + \vec{v} \times \vec{B})$$

### Interactive Demo 1: Lorentz Force on a Moving Charge

Adjust the charge, velocity components, and magnetic field to see the resulting force vector. The magnetic field points in the $z$-direction. We display a pseudo-3D view.

```python
def lorentz_force_demo(q=1.0, vx=1.0, vy=0.0, vz=0.0, Bz=1.0):
    """
    Visualize the Lorentz force F = q(v x B).
    B is along the z-axis. We show a 3D-like projection.
    """
    v = np.array([vx, vy, vz])
    B = np.array([0, 0, Bz])
    F = q * np.cross(v, B)
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Left panel: top-down view (x-y plane)
    ax1.set_title('Top-Down View (x-y plane)', fontsize=13)
    ax1.set_xlim(-5, 5)
    ax1.set_ylim(-5, 5)
    ax1.set_xlabel('x')
    ax1.set_ylabel('y')
    ax1.set_aspect('equal')
    ax1.grid(True, alpha=0.3)
    ax1.axhline(0, color='gray', lw=0.5)
    ax1.axvline(0, color='gray', lw=0.5)
    
    # Draw B field indicator
    if Bz > 0:
        ax1.plot(4, 4, 'o', color='blue', markersize=10)
        ax1.plot(4, 4, '.', color='blue', markersize=3)
        ax1.text(4, 3.3, 'B (out)', ha='center', color='blue', fontsize=10)
    elif Bz < 0:
        ax1.plot(4, 4, 'o', color='blue', markersize=10)
        ax1.plot(4, 4, 'x', color='blue', markersize=8, mew=2)
        ax1.text(4, 3.3, 'B (in)', ha='center', color='blue', fontsize=10)
    
    # Velocity vector (x-y components)
    if vx != 0 or vy != 0:
        ax1.annotate('', xy=(vx, vy), xytext=(0, 0),
                     arrowprops=dict(arrowstyle='->', color='green', lw=2.5))
        ax1.text(vx * 0.5 + 0.3, vy * 0.5 + 0.3, 'v', color='green', fontsize=14, fontweight='bold')
    
    # Force vector (x-y components)
    if F[0] != 0 or F[1] != 0:
        ax1.annotate('', xy=(F[0], F[1]), xytext=(0, 0),
                     arrowprops=dict(arrowstyle='->', color='red', lw=2.5))
        ax1.text(F[0] * 0.5 + 0.3, F[1] * 0.5 - 0.4, 'F', color='red', fontsize=14, fontweight='bold')
    
    # Charge at origin
    charge_color = 'red' if q > 0 else 'blue' if q < 0 else 'gray'
    charge_label = '+' if q > 0 else '-' if q < 0 else '0'
    ax1.plot(0, 0, 'o', color=charge_color, markersize=20, zorder=5)
    ax1.text(0, 0, charge_label, ha='center', va='center', color='white', fontsize=14, fontweight='bold', zorder=6)
    
    # Right panel: side view (x-z plane)
    ax2.set_title('Side View (x-z plane)', fontsize=13)
    ax2.set_xlim(-5, 5)
    ax2.set_ylim(-5, 5)
    ax2.set_xlabel('x')
    ax2.set_ylabel('z')
    ax2.set_aspect('equal')
    ax2.grid(True, alpha=0.3)
    ax2.axhline(0, color='gray', lw=0.5)
    ax2.axvline(0, color='gray', lw=0.5)
    
    # Velocity vector (x-z components)
    if vx != 0 or vz != 0:
        ax2.annotate('', xy=(vx, vz), xytext=(0, 0),
                     arrowprops=dict(arrowstyle='->', color='green', lw=2.5))
        ax2.text(vx * 0.5 + 0.3, vz * 0.5 + 0.3, 'v', color='green', fontsize=14, fontweight='bold')
    
    # B field vector (z-component)
    if Bz != 0:
        ax2.annotate('', xy=(0, Bz), xytext=(0, 0),
                     arrowprops=dict(arrowstyle='->', color='blue', lw=2.5))
        ax2.text(0.3, Bz * 0.5, 'B', color='blue', fontsize=14, fontweight='bold')
    
    # Force vector (x-z components)
    if F[0] != 0 or F[2] != 0:
        ax2.annotate('', xy=(F[0], F[2]), xytext=(0, 0),
                     arrowprops=dict(arrowstyle='->', color='red', lw=2.5))
        ax2.text(F[0] * 0.5 + 0.3, F[2] * 0.5 - 0.4, 'F', color='red', fontsize=14, fontweight='bold')
    
    ax2.plot(0, 0, 'o', color=charge_color, markersize=20, zorder=5)
    ax2.text(0, 0, charge_label, ha='center', va='center', color='white', fontsize=14, fontweight='bold', zorder=6)
    
    plt.tight_layout()
    
    # Print numerical values
    print(f"Velocity:  v = ({vx:.1f}, {vy:.1f}, {vz:.1f}) m/s")
    print(f"B field:   B = (0, 0, {Bz:.1f}) T")
    print(f"Charge:    q = {q:.1f} C")
    print(f"Force:     F = q(v x B) = ({F[0]:.2f}, {F[1]:.2f}, {F[2]:.2f}) N")
    print(f"|F| = {np.linalg.norm(F):.2f} N")
    plt.show()

interact(lorentz_force_demo,
         q=FloatSlider(min=-3, max=3, step=0.5, value=1.0, description='q (C)'),
         vx=FloatSlider(min=-3, max=3, step=0.5, value=2.0, description='vx (m/s)'),
         vy=FloatSlider(min=-3, max=3, step=0.5, value=0.0, description='vy (m/s)'),
         vz=FloatSlider(min=-3, max=3, step=0.5, value=0.0, description='vz (m/s)'),
         Bz=FloatSlider(min=-3, max=3, step=0.5, value=1.0, description='Bz (T)'));
```

## 3. Circular Motion in a Magnetic Field

When a charged particle moves **perpendicular** to a uniform magnetic field, the magnetic force acts as a centripetal force, causing the particle to move in a **circle**.

Setting the magnetic force equal to the centripetal force:

$$qvB = \frac{mv^2}{r}$$

Solving for the **radius of the circular orbit** (cyclotron radius):

$$r = \frac{mv}{qB}$$

The **period** of revolution (cyclotron period):

$$T = \frac{2\pi r}{v} = \frac{2\pi m}{qB}$$

Notice: The period is **independent of the speed**! This is the principle behind the cyclotron.

### Interactive Demo 2: Circular Motion of a Charged Particle in B Field

Adjust the speed, charge, mass, and magnetic field to see how the circular orbit changes.

```python
def circular_motion_demo(v=2.0, q=1.0, m=1.0, B=1.0):
    """
    Show the circular orbit of a charged particle in a uniform B field.
    """
    if q == 0 or B == 0:
        print("q and B must be nonzero for circular motion.")
        return
    
    r = m * v / (abs(q) * B)
    T = 2 * np.pi * m / (abs(q) * B)
    omega = abs(q) * B / m
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Left panel: circular orbit
    theta = np.linspace(0, 2 * np.pi, 200)
    # Positive charge in B out of page: clockwise
    sign = 1 if q > 0 else -1
    x_circle = r * np.cos(theta)
    y_circle = -sign * r + sign * r * np.sin(theta)  # shifted so particle starts at origin moving in +x
    
    # Actually let's place center at (0, -sign*r) so particle starts at (0,0) moving in +x
    cx, cy = 0, -sign * r
    x_circle = cx + r * np.sin(theta)
    y_circle = cy + r * np.cos(theta)
    
    ax1.plot(x_circle, y_circle, 'b-', lw=2, label=f'r = {r:.2f} m')
    ax1.plot(x_circle[0], y_circle[0], 'ro', markersize=12, zorder=5, label='Start')
    
    # Show velocity direction at start
    ax1.annotate('', xy=(0.3 * r, y_circle[0]), xytext=(x_circle[0], y_circle[0]),
                 arrowprops=dict(arrowstyle='->', color='green', lw=2))
    ax1.text(0.15 * r, y_circle[0] + 0.15 * r, 'v', color='green', fontsize=13, fontweight='bold')
    
    # Show radius
    ax1.plot([cx, x_circle[0]], [cy, y_circle[0]], 'k--', lw=1.5)
    ax1.plot(cx, cy, 'k+', markersize=10, mew=2)
    ax1.text(cx + 0.1 * r, cy + 0.5 * r, f'r = {r:.2f} m', fontsize=11)
    
    pad = max(r * 1.5, 1)
    ax1.set_xlim(-pad, pad)
    ax1.set_ylim(cy - r - 0.5, cy + r + 0.5)
    ax1.set_aspect('equal')
    ax1.set_xlabel('x (m)')
    ax1.set_ylabel('y (m)')
    ax1.set_title('Circular Orbit (B out of page)', fontsize=13)
    ax1.legend(fontsize=10)
    ax1.grid(True, alpha=0.3)
    
    # Right panel: r vs v and r vs B
    v_range = np.linspace(0.1, 5, 100)
    r_vs_v = m * v_range / (abs(q) * B)
    
    B_range = np.linspace(0.1, 5, 100)
    r_vs_B = m * v / (abs(q) * B_range)
    
    ax2.plot(v_range, r_vs_v, 'b-', lw=2, label=f'r(v), B={B:.1f} T')
    ax2.plot(v, r, 'bo', markersize=10, zorder=5)
    
    ax2_twin = ax2.twinx()
    ax2_twin.plot(B_range, r_vs_B, 'r--', lw=2, label=f'r(B), v={v:.1f} m/s')
    ax2_twin.plot(B, r, 'rs', markersize=10, zorder=5)
    ax2_twin.set_ylabel('r (m) [red, vs B]', color='red')
    
    ax2.set_xlabel('v (m/s) [blue] / B (T) [red]')
    ax2.set_ylabel('r (m) [blue, vs v]', color='blue')
    ax2.set_title('Cyclotron Radius Dependencies', fontsize=13)
    ax2.legend(loc='upper left', fontsize=10)
    ax2_twin.legend(loc='upper right', fontsize=10)
    ax2.grid(True, alpha=0.3)
    
    plt.tight_layout()
    
    print(f"Cyclotron radius:  r = mv/(qB) = {r:.3f} m")
    print(f"Period:            T = 2*pi*m/(qB) = {T:.3f} s")
    print(f"Cyclotron freq:    omega = qB/m = {omega:.3f} rad/s")
    print(f"Note: T does NOT depend on v!")
    plt.show()

interact(circular_motion_demo,
         v=FloatSlider(min=0.5, max=5, step=0.25, value=2.0, description='v (m/s)'),
         q=FloatSlider(min=-3.0, max=3, step=0.5, value=1.0, description='q (C)'),
         m=FloatSlider(min=0.5, max=5, step=0.25, value=1.0, description='m (kg)'),
         B=FloatSlider(min=0.5, max=3, step=0.25, value=1.0, description='B (T)'));
```

### Animation: Charged Particle in a Magnetic Field

Watch a charged particle orbit in a uniform magnetic field (top-down 2D view with B pointing out of the screen).

```python
def create_circular_orbit_animation():
    """
    Animate a charged particle undergoing circular motion in a B field.
    Top-down view with B out of the page.
    """
    # Parameters
    q = 1.0
    m = 1.0
    B = 1.0
    v = 2.0
    r = m * v / (q * B)
    omega = q * B / m
    
    # Time array for 3 full orbits
    T_period = 2 * np.pi * m / (q * B)
    t_max = 3 * T_period
    n_frames = 120
    t_arr = np.linspace(0, t_max, n_frames)
    
    # Position: center at (0, -r), starts at (0, 0) moving in +x
    cx, cy = 0, -r
    x_pos = cx + r * np.sin(omega * t_arr)
    y_pos = cy + r * np.cos(omega * t_arr)
    
    fig, ax = plt.subplots(figsize=(7, 7))
    ax.set_xlim(-r - 1, r + 1)
    ax.set_ylim(-2 * r - 1, 1)
    ax.set_aspect('equal')
    ax.set_xlabel('x (m)', fontsize=12)
    ax.set_ylabel('y (m)', fontsize=12)
    ax.set_title('Charged Particle in Uniform B Field (out of page)', fontsize=13)
    ax.grid(True, alpha=0.3)
    
    # Draw full orbit path
    theta_full = np.linspace(0, 2 * np.pi, 200)
    ax.plot(cx + r * np.sin(theta_full), cy + r * np.cos(theta_full), 'b--', alpha=0.3, lw=1)
    
    # B field indicators
    for xi in np.linspace(-r, r, 5):
        for yi in np.linspace(-2 * r, 0, 5):
            ax.plot(xi, yi, '.', color='blue', markersize=4, alpha=0.3)
            ax.plot(xi, yi, 'o', color='blue', markersize=8, alpha=0.15, mew=0.5)
    
    trail, = ax.plot([], [], 'b-', lw=2, alpha=0.6)
    particle, = ax.plot([], [], 'ro', markersize=14, zorder=5)
    vel_arrow = ax.annotate('', xy=(0, 0), xytext=(0, 0),
                            arrowprops=dict(arrowstyle='->', color='green', lw=2))
    time_text = ax.text(0.02, 0.95, '', transform=ax.transAxes, fontsize=11)
    
    def init():
        trail.set_data([], [])
        particle.set_data([], [])
        time_text.set_text('')
        return trail, particle, time_text
    
    def animate(i):
        # Trail
        trail.set_data(x_pos[:i+1], y_pos[:i+1])
        # Particle
        particle.set_data([x_pos[i]], [y_pos[i]])
        
        # Velocity arrow
        vx = v * np.cos(omega * t_arr[i])
        vy = -v * np.sin(omega * t_arr[i])
        scale = 0.5
        vel_arrow.xy = (x_pos[i] + scale * vx, y_pos[i] + scale * vy)
        vel_arrow.set_position((x_pos[i], y_pos[i]))
        
        time_text.set_text(f't = {t_arr[i]:.2f} s, orbit {t_arr[i]/T_period:.1f}')
        return trail, particle, time_text
    
    anim = FuncAnimation(fig, animate, init_func=init, frames=n_frames,
                         interval=50, blit=False)
    plt.close(fig)
    return anim

anim = create_circular_orbit_animation()
HTML(anim.to_jshtml())
```

## 4. The Velocity Selector

A **velocity selector** uses crossed electric and magnetic fields to select particles with a specific speed.

Setup:
- Electric field $\vec{E}$ points upward
- Magnetic field $\vec{B}$ points out of the page
- Particle moves to the right with speed $v$

For the particle to pass straight through (no deflection):

$$qE = qvB \quad \Rightarrow \quad v = \frac{E}{B}$$

- Particles with $v > E/B$: magnetic force dominates, deflected downward
- Particles with $v < E/B$: electric force dominates, deflected upward
- Particles with $v = E/B$: pass straight through

### Interactive Demo 3: Velocity Selector

Adjust the particle speed and see whether it passes through the selector or gets deflected.

```python
def velocity_selector_demo(v=3.0, E=2.0, B=1.0, q=1.0):
    """
    Simulate a velocity selector with crossed E and B fields.
    E points up (y), B points out of page (z). Particle enters moving in +x.
    """
    v_selected = E / B
    
    fig, ax = plt.subplots(figsize=(12, 6))
    
    # Draw selector plates
    plate_x = [1, 5]
    ax.fill_between(plate_x, 2.2, 2.5, color='red', alpha=0.5, label='+ plate')
    ax.fill_between(plate_x, -2.5, -2.2, color='blue', alpha=0.5, label='- plate')
    
    # E field arrows (pointing up)
    for xi in np.linspace(1.5, 4.5, 4):
        ax.annotate('', xy=(xi, 1.8), xytext=(xi, -1.8),
                     arrowprops=dict(arrowstyle='->', color='orange', lw=1.5, alpha=0.5))
    ax.text(4.7, 0, 'E', color='orange', fontsize=14, fontweight='bold')
    
    # B field indicators (out of page)
    for xi in np.linspace(1.5, 4.5, 4):
        for yi in np.linspace(-1.5, 1.5, 3):
            ax.plot(xi, yi, '.', color='blue', markersize=3, alpha=0.4)
            ax.plot(xi, yi, 'o', color='blue', markersize=7, alpha=0.2)
    ax.text(1.0, -1.8, 'B (out)', color='blue', fontsize=10)
    
    # Entrance and exit slits
    ax.plot([1, 1], [-2.2, -0.15], 'k-', lw=3)
    ax.plot([1, 1], [0.15, 2.2], 'k-', lw=3)
    ax.plot([5, 5], [-2.2, -0.15], 'k-', lw=3)
    ax.plot([5, 5], [0.15, 2.2], 'k-', lw=3)
    
    # Particle trajectory
    # F_E = qE (up), F_B = qvB (down for positive charge moving right in B out)
    # Net force in y: F_net = q(E - vB)
    a_y = q * (E - v * B)  # assuming m=1 for visualization
    
    dt_vals = np.linspace(0, 4 / max(v, 0.5), 300)
    x_traj = v * dt_vals
    y_traj = 0.5 * a_y * dt_vals**2
    
    # Clip to selector region
    mask = (x_traj <= 7) & (np.abs(y_traj) <= 2.2)
    x_traj = x_traj[mask]
    y_traj = y_traj[mask]
    
    if abs(v - v_selected) < 0.05:
        color = 'green'
        status = 'PASSES THROUGH'
    elif v > v_selected:
        color = 'red'
        status = 'DEFLECTED DOWN (B force wins)'
    else:
        color = 'purple'
        status = 'DEFLECTED UP (E force wins)'
    
    ax.plot(x_traj, y_traj, '-', color=color, lw=2.5)
    if len(x_traj) > 0:
        ax.plot(x_traj[-1], y_traj[-1], 'o', color=color, markersize=10, zorder=5)
    ax.plot(0, 0, 'o', color='red', markersize=10, zorder=5)
    
    ax.set_xlim(-0.5, 7)
    ax.set_ylim(-3, 3)
    ax.set_xlabel('x (m)', fontsize=12)
    ax.set_ylabel('y (m)', fontsize=12)
    ax.set_title(f'Velocity Selector: v_selected = E/B = {v_selected:.2f} m/s', fontsize=13)
    ax.grid(True, alpha=0.3)
    ax.set_aspect('equal')
    
    print(f"Selected speed: v = E/B = {E:.1f}/{B:.1f} = {v_selected:.2f} m/s")
    print(f"Particle speed: v = {v:.2f} m/s")
    print(f"Status: {status}")
    print(f"Electric force (up): F_E = qE = {q*E:.2f} N")
    print(f"Magnetic force (down): F_B = qvB = {q*v*B:.2f} N")
    plt.show()

interact(velocity_selector_demo,
         v=FloatSlider(min=0.5, max=6, step=0.1, value=2.0, description='v (m/s)'),
         E=FloatSlider(min=0.5, max=5, step=0.25, value=2.0, description='E (V/m)'),
         B=FloatSlider(min=0.5, max=3, step=0.25, value=1.0, description='B (T)'),
         q=FloatSlider(min=0.5, max=3, step=0.25, value=1.0, description='q (C)'));
```

## 5. The Mass Spectrometer

A **mass spectrometer** separates ions by mass. After passing through a velocity selector (so all ions have the same speed $v$), ions enter a region with only a magnetic field.

In this region, each ion follows a semicircular path with radius:

$$r = \frac{mv}{qB}$$

Since $v$, $q$, and $B$ are the same for all ions, heavier ions have a **larger radius**. By measuring the radius (or the diameter, the distance from the entrance slit to where the ion hits the detector), we can determine the mass:

$$m = \frac{qBr}{v} = \frac{qB(d/2)}{v} = \frac{qBd}{2v}$$

where $d$ is the diameter of the semicircle.

### Interactive Demo 4: Mass Spectrometer

See how ions of different masses trace different semicircular paths.

```python
def mass_spectrometer_demo(v=2.0, B=1.0, q=1.0, m1=1.0, m2=2.0, m3=4.0):
    """
    Simulate a mass spectrometer showing semicircular paths for different masses.
    Ions enter from the left moving in +x, B is out of the page.
    """
    masses = [m1, m2, m3]
    colors = ['red', 'blue', 'green']
    labels = ['m1', 'm2', 'm3']
    
    fig, ax = plt.subplots(figsize=(12, 7))
    
    # Draw entrance slit
    ax.plot([0, 0], [0.3, 3], 'k-', lw=3)
    ax.plot([0, 0], [-3, -0.3], 'k-', lw=3)
    ax.annotate('Entrance\nSlit', xy=(-0.8, 1.5), fontsize=10, ha='center')
    
    # Draw detector plate
    ax.fill_between([-8, 0], -0.15, 0.15, color='gray', alpha=0.3)
    ax.text(-4, 0.35, 'Detector Plate', ha='center', fontsize=10, color='gray')
    
    max_r = 0
    for i, m in enumerate(masses):
        r = m * v / (q * B)
        max_r = max(max_r, r)
        d = 2 * r  # diameter (distance on detector)
        
        # Semicircular path: center at (0, -r), from (0,0) to (0, -2r)
        theta = np.linspace(0, np.pi, 200)
        x_path = -r * np.sin(theta)  # goes to negative x
        y_path = -r + r * np.cos(theta)
        
        ax.plot(x_path, y_path, '-', color=colors[i], lw=2.5,
                label=f'{labels[i]} = {m:.1f} kg: r = {r:.2f} m, d = {d:.2f} m')
        
        # Mark landing point on detector
        ax.plot(0, -2 * r, 's', color=colors[i], markersize=12, zorder=5)
        ax.text(0.3, -2 * r, f'd = {d:.2f} m', color=colors[i], fontsize=10)
    
    # Entry point
    ax.plot(0, 0, 'ko', markersize=8, zorder=5)
    ax.annotate('', xy=(0.8, 0), xytext=(0, 0),
                 arrowprops=dict(arrowstyle='->', color='black', lw=2))
    ax.text(0.9, 0.15, f'v = {v:.1f} m/s', fontsize=10)
    
    pad = max_r * 2.5
    ax.set_xlim(-pad, 2)
    ax.set_ylim(-pad, 1.5)
    ax.set_aspect('equal')
    ax.set_xlabel('x (m)', fontsize=12)
    ax.set_ylabel('y (m)', fontsize=12)
    ax.set_title('Mass Spectrometer: Semicircular Paths', fontsize=13)
    ax.legend(loc='lower left', fontsize=10)
    ax.grid(True, alpha=0.3)
    
    print(f"v = {v:.2f} m/s, B = {B:.2f} T, q = {q:.2f} C")
    print(f"r = mv/(qB), so heavier ions have larger radii.")
    for i, m in enumerate(masses):
        r = m * v / (q * B)
        print(f"  {labels[i]} = {m:.2f} kg -> r = {r:.2f} m, d = {2*r:.2f} m")
    plt.show()

interact(mass_spectrometer_demo,
         v=FloatSlider(min=0.5, max=4, step=0.25, value=2.0, description='v (m/s)'),
         B=FloatSlider(min=0.5, max=3, step=0.25, value=1.0, description='B (T)'),
         q=FloatSlider(min=0.5, max=2, step=0.25, value=1.0, description='q (C)'),
         m1=FloatSlider(min=0.5, max=4, step=0.25, value=1.0, description='m1 (kg)'),
         m2=FloatSlider(min=0.5, max=6, step=0.25, value=2.0, description='m2 (kg)'),
         m3=FloatSlider(min=0.5, max=8, step=0.25, value=4.0, description='m3 (kg)'));
```

## 6. Worked Examples

### Example 1: Force on a Moving Charge

**Problem:** A proton ($q = 1.6 \times 10^{-19}$ C) moves with velocity $v = 3 \times 10^6$ m/s in the $+x$ direction through a magnetic field $B = 0.5$ T in the $+z$ direction. Find the magnitude and direction of the magnetic force.

```python
# Worked Example 1: Force on a moving proton

q = 1.6e-19       # C (proton charge)
v = 3e6            # m/s
B = 0.5            # T

# v = (v, 0, 0), B = (0, 0, B)
v_vec = np.array([v, 0, 0])
B_vec = np.array([0, 0, B])

# F = q(v x B)
F_vec = q * np.cross(v_vec, B_vec)
F_mag = np.linalg.norm(F_vec)

print("=" * 50)
print("Worked Example 1: Force on a Moving Proton")
print("=" * 50)
print(f"Given:")
print(f"  q = {q:.2e} C")
print(f"  v = ({v:.2e}, 0, 0) m/s")
print(f"  B = (0, 0, {B}) T")
print()
print(f"Solution:")
print(f"  F = q(v x B)")
print(f"  v x B = ({v:.2e})(i x k_hat)... ")
print(f"  Using cross product: i x k = -j")
print(f"  v x B = ({v*B:.2e}) * (-j) = (0, {-v*B:.2e}, 0)")
print(f"  F = q(v x B) = ({F_vec[0]:.2e}, {F_vec[1]:.2e}, {F_vec[2]:.2e}) N")
print(f"  |F| = {F_mag:.2e} N")
print(f"  Direction: -y direction (the proton is deflected downward)")
print()
print(f"  Alternatively: F = qvB sin(90) = {q*v*B:.2e} N  (check!)")
```

### Example 2: Radius of Circular Orbit

**Problem:** An electron ($m = 9.11 \times 10^{-31}$ kg, $q = 1.6 \times 10^{-19}$ C) moves at $v = 2 \times 10^7$ m/s perpendicular to a magnetic field of $B = 0.01$ T. Find the radius and period of its circular orbit.

```python
# Worked Example 2: Circular orbit of an electron

m_e = 9.11e-31     # kg (electron mass)
q_e = 1.6e-19      # C (magnitude of electron charge)
v_e = 2e7           # m/s
B_e = 0.01          # T

# Cyclotron radius
r_e = m_e * v_e / (q_e * B_e)

# Cyclotron period
T_e = 2 * np.pi * m_e / (q_e * B_e)

# Cyclotron frequency
f_e = 1 / T_e

print("=" * 50)
print("Worked Example 2: Electron Circular Orbit")
print("=" * 50)
print(f"Given:")
print(f"  m = {m_e:.2e} kg")
print(f"  q = {q_e:.2e} C")
print(f"  v = {v_e:.2e} m/s")
print(f"  B = {B_e} T")
print()
print(f"Solution:")
print(f"  r = mv/(qB)")
print(f"    = ({m_e:.2e})({v_e:.2e}) / ({q_e:.2e})({B_e})")
print(f"    = {m_e*v_e:.2e} / {q_e*B_e:.2e}")
print(f"    = {r_e:.4f} m = {r_e*100:.2f} cm")
print()
print(f"  T = 2*pi*m/(qB)")
print(f"    = 2*pi*({m_e:.2e}) / ({q_e:.2e})({B_e})")
print(f"    = {T_e:.2e} s = {T_e*1e9:.2f} ns")
print()
print(f"  f = 1/T = {f_e:.2e} Hz = {f_e/1e6:.1f} MHz")
print()
print(f"  Note: The period T does not depend on v!")
```

| Cyclotron period | $T = 2\pi m/(qB)$ |

## 1. The Missing Piece: Displacement Current

Before Maxwell, Ampere's law stated:

$$\oint \vec{B} \cdot d\vec{l} = \mu_0 I_{\text{enc}}$$

But Maxwell noticed a problem: when a capacitor is charging, current flows in the wires but **no real current** crosses the gap between the plates. Yet the magnetic field must be continuous!

Maxwell's fix was to add a **displacement current**:

$$I_d = \varepsilon_0 \frac{d\Phi_E}{dt}$$

This is not a real current of moving charges -- it is the effect of a **changing electric field** producing a magnetic field, just as a changing magnetic field produces an electric field (Faraday's law).

The corrected Ampere-Maxwell law becomes:

$$\oint \vec{B} \cdot d\vec{l} = \mu_0 \left( I_{\text{enc}} + \varepsilon_0 \frac{d\Phi_E}{dt} \right)$$

## 3. Electromagnetic Wave Propagation

Maxwell's equations predict that **changing electric fields create magnetic fields**, and **changing magnetic fields create electric fields**. This mutual generation allows disturbances to propagate through space as **electromagnetic waves**.

Key properties of EM waves:
- $\vec{E}$ and $\vec{B}$ are **perpendicular** to each other
- Both are **perpendicular** to the direction of propagation (transverse wave)
- They oscillate **in phase** (reach peaks and zeros at the same time)
- The speed in vacuum is:

$$c = \frac{1}{\sqrt{\mu_0 \varepsilon_0}} = 3.00 \times 10^8 \text{ m/s}$$

- The amplitudes are related by: $E_0 = c B_0$

```python
# ============================================================
# Animation 1: EM Wave Propagation (Classic 3D-like view)
# E field in y-direction, B field in x-direction, propagation in z
# ============================================================

fig = plt.figure(figsize=(12, 7))
ax = fig.add_subplot(111, projection='3d')

# Wave parameters
wavelength = 2 * np.pi
z = np.linspace(0, 3 * wavelength, 400)
n_frames = 60

def update_em_wave(frame):
    ax.cla()
    
    phase = 2 * np.pi * frame / n_frames
    
    # E field (oscillates in y-direction)
    E = np.sin(z - phase)
    # B field (oscillates in x-direction, in phase with E)
    B = np.sin(z - phase)
    
    # Plot E field (red) in the y-z plane
    ax.plot(np.zeros_like(z), z, E, color='#e74c3c', linewidth=2.5, label='E field')
    # Plot B field (blue) in the x-z plane
    ax.plot(B, z, np.zeros_like(z), color='#3498db', linewidth=2.5, label='B field')
    
    # Draw arrows at selected points to show field directions
    arrow_indices = np.arange(0, len(z), len(z) // 12)
    for i in arrow_indices:
        if abs(E[i]) > 0.05:
            ax.quiver(0, z[i], 0, 0, 0, E[i], color='#e74c3c', alpha=0.5,
                     arrow_length_ratio=0.15, linewidth=1.2)
        if abs(B[i]) > 0.05:
            ax.quiver(0, z[i], 0, B[i], 0, 0, color='#3498db', alpha=0.5,
                     arrow_length_ratio=0.15, linewidth=1.2)
    
    # Propagation direction arrow
    ax.quiver(0, 0, 0, 0, 3, 0, color='green', arrow_length_ratio=0.05,
             linewidth=2.5, alpha=0.7)
    
    ax.set_xlabel('B (x-axis)', fontsize=11, color='#3498db')
    ax.set_ylabel('Propagation z', fontsize=11, color='green')
    ax.set_zlabel('E (y-axis)', fontsize=11, color='#e74c3c')
    ax.set_title('Electromagnetic Wave Propagation', fontsize=14, fontweight='bold')
    
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(0, 3 * wavelength)
    ax.set_zlim(-1.5, 1.5)
    ax.view_init(elev=20, azim=-60)
    ax.legend(loc='upper left', fontsize=10)

anim = FuncAnimation(fig, update_em_wave, frames=n_frames, interval=80)
plt.close(fig)
HTML(anim.to_jshtml())
```

### Understanding the Animation

In the animation above:
- The **red wave** represents the electric field $\vec{E}$, oscillating in the y-direction
- The **blue wave** represents the magnetic field $\vec{B}$, oscillating in the x-direction
- The wave **propagates** along the z-direction (green arrow)
- $\vec{E}$ and $\vec{B}$ are always perpendicular to each other and to the direction of travel
- They oscillate **in phase** -- both reach their peaks and zeros together

## 4. The Speed of Light

One of the most remarkable results in physics: the speed of electromagnetic waves comes directly from two constants measured in the lab:

$$c = \frac{1}{\sqrt{\mu_0 \varepsilon_0}}$$

where:
- $\varepsilon_0 = 8.854 \times 10^{-12}$ F/m (permittivity of free space)
- $\mu_0 = 4\pi \times 10^{-7}$ T$\cdot$m/A (permeability of free space)

When Maxwell computed this, it matched the known speed of light -- leading him to declare that **light is an electromagnetic wave**!

```python
# ============================================================
# Worked Example 1: Speed of Light from epsilon_0 and mu_0
# ============================================================

print("="*60)
print("WORKED EXAMPLE: Calculating the Speed of Light")
print("="*60)

epsilon_0 = 8.854e-12   # F/m
mu_0 = 4 * np.pi * 1e-7  # T*m/A

print(f"\nGiven:")
print(f"  epsilon_0 = {epsilon_0:.3e} F/m")
print(f"  mu_0      = {mu_0:.3e} T*m/A")

print(f"\nFormula: c = 1 / sqrt(mu_0 * epsilon_0)")

product = mu_0 * epsilon_0
print(f"\nStep 1: mu_0 * epsilon_0 = {product:.4e} s^2/m^2")

c_calculated = 1 / np.sqrt(product)
print(f"Step 2: c = 1 / sqrt({product:.4e})")
print(f"         c = {c_calculated:.4e} m/s")

c_known = 2.998e8
print(f"\nKnown speed of light: c = {c_known:.3e} m/s")
print(f"Our calculation:      c = {c_calculated:.3e} m/s")
print(f"\nThey match! Light IS an electromagnetic wave.")
print("="*60)
```

## 5. EM Wave Properties Explorer

For any electromagnetic wave in vacuum:

$$c = \lambda f$$

where $\lambda$ is the wavelength and $f$ is the frequency. The energy of a photon is:

$$E = hf$$

where $h = 6.626 \times 10^{-34}$ J$\cdot$s is Planck's constant.

Use the slider below to explore how frequency determines wavelength, photon energy, and the type of electromagnetic radiation.

```python
# ============================================================
# Interactive 2: EM Wave Properties - Frequency Slider
# ============================================================

def get_em_region(freq):
    """Return EM spectrum region name and color for a given frequency."""
    if freq < 3e9:
        return "Radio Waves", "#a0522d"
    elif freq < 3e11:
        return "Microwaves", "#d2691e"
    elif freq < 4.0e14:
        return "Infrared", "#cc0000"
    elif freq < 7.9e14:
        return "Visible Light", "#228b22"
    elif freq < 3e16:
        return "Ultraviolet", "#6a0dad"
    elif freq < 3e19:
        return "X-rays", "#1e90ff"
    else:
        return "Gamma Rays", "#000080"

def get_visible_color(wavelength_nm):
    """Return approximate color for visible wavelengths."""
    if 380 <= wavelength_nm < 450:
        return '#8B00FF'  # Violet
    elif 450 <= wavelength_nm < 495:
        return '#0000FF'  # Blue
    elif 495 <= wavelength_nm < 570:
        return '#00FF00'  # Green
    elif 570 <= wavelength_nm < 590:
        return '#FFFF00'  # Yellow
    elif 590 <= wavelength_nm < 620:
        return '#FF7F00'  # Orange
    elif 620 <= wavelength_nm <= 750:
        return '#FF0000'  # Red
    return '#888888'

def em_wave_properties(log_freq=9.0):
    freq = 10**log_freq
    c = 3e8
    h = 6.626e-34
    eV = 1.602e-19
    
    wavelength = c / freq
    energy_J = h * freq
    energy_eV = energy_J / eV
    region, color = get_em_region(freq)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Left panel: wave visualization
    ax1 = axes[0]
    x = np.linspace(0, 4, 500)
    y = np.sin(2 * np.pi * x)
    ax1.plot(x, y, color=color, linewidth=2.5)
    ax1.fill_between(x, y, alpha=0.1, color=color)
    ax1.set_ylim(-1.5, 1.5)
    ax1.set_xlabel('Distance (wavelengths)', fontsize=12)
    ax1.set_ylabel('Field Amplitude', fontsize=12)
    ax1.set_title(f'{region}', fontsize=16, fontweight='bold', color=color)
    
    # Wavelength annotation
    ax1.annotate('', xy=(1, -1.3), xytext=(0, -1.3),
                arrowprops=dict(arrowstyle='<->', color='black', lw=2))
    ax1.text(0.5, -1.45, r'$\lambda$', fontsize=14, ha='center')
    
    # Right panel: properties display
    ax2 = axes[1]
    ax2.axis('off')
    
    # Format wavelength nicely
    if wavelength >= 1:
        wl_str = f"{wavelength:.2f} m"
    elif wavelength >= 1e-3:
        wl_str = f"{wavelength*1e3:.2f} mm"
    elif wavelength >= 1e-6:
        wl_str = f"{wavelength*1e6:.2f} \u03bcm"
    elif wavelength >= 1e-9:
        wl_str = f"{wavelength*1e9:.2f} nm"
    elif wavelength >= 1e-12:
        wl_str = f"{wavelength*1e12:.3f} pm"
    else:
        wl_str = f"{wavelength:.2e} m"
    
    props = [
        (f"Frequency:  {freq:.2e} Hz", 0.85),
        (f"Wavelength: {wl_str}", 0.70),
        (f"Photon Energy: {energy_eV:.4e} eV", 0.55),
        (f"Region: {region}", 0.40),
    ]
    
    for text, ypos in props:
        ax2.text(0.1, ypos, text, fontsize=14, transform=ax2.transAxes,
                family='monospace', verticalalignment='center')
    
    # Color swatch for visible light
    wl_nm = wavelength * 1e9
    if 380 <= wl_nm <= 750:
        vis_color = get_visible_color(wl_nm)
        rect = mpatches.FancyBboxPatch((0.1, 0.08), 0.8, 0.15,
                                       transform=ax2.transAxes,
                                       facecolor=vis_color, alpha=0.7,
                                       edgecolor='black', linewidth=2,
                                       boxstyle='round,pad=0.02')
        ax2.add_patch(rect)
        ax2.text(0.5, 0.155, f"Visible color at {wl_nm:.0f} nm",
                fontsize=12, ha='center', transform=ax2.transAxes,
                fontweight='bold')
    
    plt.tight_layout()
    plt.show()

freq_slider = FloatSlider(
    value=14.7, min=6, max=22, step=0.1,
    description='log\u2081\u2080(f):',
    layout=Layout(width='600px'),
    style={'description_width': '80px'},
    readout_format='.1f'
)

print("Drag the slider to change frequency (logarithmic scale):")
print("  log10(f) = 6 -> Radio  |  14.7 -> Visible  |  22 -> Gamma Rays")
interact(em_wave_properties, log_freq=freq_slider);
```

## 7. Energy in Electromagnetic Waves

EM waves carry energy. The energy per unit volume (energy density) has two equal contributions:

$$u = u_E + u_B = \frac{1}{2}\varepsilon_0 E^2 + \frac{1}{2\mu_0}B^2$$

Since $E = cB$ and $c = 1/\sqrt{\mu_0 \varepsilon_0}$, both contributions are always equal:

$$u_E = u_B$$

The **Poynting vector** $\vec{S}$ describes the rate of energy flow per unit area:

$$\vec{S} = \frac{1}{\mu_0} \vec{E} \times \vec{B}$$

Its magnitude gives the **intensity** (power per area) and its direction gives the direction of energy flow. The time-averaged intensity is:

$$I = \langle S \rangle = \frac{E_0 B_0}{2\mu_0} = \frac{E_0^2}{2\mu_0 c} = \frac{c B_0^2}{2\mu_0}$$

```python
# ============================================================
# Worked Example 2: Energy Density of an EM Wave
# ============================================================

print("="*60)
print("WORKED EXAMPLE: Energy in an EM Wave")
print("="*60)
print("\nProblem: A radio station broadcasts at a point where the")
print("electric field amplitude is E_0 = 0.02 V/m.")
print("Find: (a) B_0, (b) average energy density, (c) intensity.")

E_0 = 0.02  # V/m
c = 3e8
mu_0 = 4 * np.pi * 1e-7
epsilon_0 = 8.854e-12

# Part (a)
B_0 = E_0 / c
print(f"\n(a) B_0 = E_0 / c = {E_0} / {c:.2e}")
print(f"    B_0 = {B_0:.4e} T")

# Part (b)
u_E = 0.5 * epsilon_0 * E_0**2
u_B = B_0**2 / (2 * mu_0)
u_avg = 0.5 * (u_E + u_B)  # time-averaged total
print(f"\n(b) Average energy density:")
print(f"    <u_E> = (1/2) * epsilon_0 * E_0^2 / 2 = {u_E/2:.4e} J/m^3")
print(f"    <u_B> = B_0^2 / (4*mu_0) = {u_B/2:.4e} J/m^3")
print(f"    Note: <u_E> = <u_B>  (always equal!)")
print(f"    <u_total> = <u_E> + <u_B> = {u_avg:.4e} J/m^3")

# Part (c)
I = E_0**2 / (2 * mu_0 * c)
I_alt = c * u_avg
print(f"\n(c) Intensity:")
print(f"    I = E_0^2 / (2*mu_0*c) = {I:.4e} W/m^2")
print(f"    Check: I = c * <u> = {I_alt:.4e} W/m^2")
print(f"\n    This is about {I*1e6:.2f} microwatts per square meter.")
print("="*60)
```

## 8. The Poynting Vector and Energy Flow

The Poynting vector tells us the direction and rate of energy flow. For a wave traveling in the $+z$ direction:

$$\vec{S} = \frac{1}{\mu_0}\vec{E} \times \vec{B}$$

Since $\vec{E} \perp \vec{B}$, the cross product always points in the propagation direction.

```python
# ============================================================
# Animation 2: Synchronized E and B fields in a plane wave
# showing Poynting vector
# ============================================================

fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)

z_line = np.linspace(0, 4 * np.pi, 500)
n_frames_2 = 60

def update_eb_sync(frame):
    phase = 2 * np.pi * frame / n_frames_2
    
    for ax in axes:
        ax.cla()
        ax.set_xlim(0, 4 * np.pi)
        ax.grid(True, alpha=0.3)
    
    E = np.sin(z_line - phase)
    B = np.sin(z_line - phase)  # In phase with E
    S = E * B  # Proportional to Poynting vector magnitude
    
    # E field
    axes[0].plot(z_line, E, color='#e74c3c', linewidth=2.5)
    axes[0].fill_between(z_line, E, alpha=0.15, color='#e74c3c')
    axes[0].set_ylabel('E(z,t)', fontsize=13, color='#e74c3c')
    axes[0].set_ylim(-1.5, 1.5)
    axes[0].set_title('Electric and Magnetic Fields in a Plane Wave', 
                      fontsize=14, fontweight='bold')
    
    # B field
    axes[1].plot(z_line, B, color='#3498db', linewidth=2.5)
    axes[1].fill_between(z_line, B, alpha=0.15, color='#3498db')
    axes[1].set_ylabel('B(z,t)', fontsize=13, color='#3498db')
    axes[1].set_ylim(-1.5, 1.5)
    
    # Poynting vector (always non-negative for a traveling wave)
    axes[2].plot(z_line, S, color='#2ecc71', linewidth=2.5)
    axes[2].fill_between(z_line, S, alpha=0.2, color='#2ecc71')
    axes[2].set_ylabel('S(z,t)', fontsize=13, color='#2ecc71')
    axes[2].set_ylim(-0.5, 1.5)
    axes[2].set_xlabel('z (propagation direction)', fontsize=13)
    axes[2].axhline(y=0.5, color='green', linestyle='--', alpha=0.5,
                   label=r'$\langle S \rangle$ (time average)')
    axes[2].legend(loc='upper right', fontsize=10)

anim2 = FuncAnimation(fig, update_eb_sync, frames=n_frames_2, interval=80)
plt.tight_layout()
plt.close(fig)
HTML(anim2.to_jshtml())
```

### Key observations from the animation:

- $E(z,t)$ and $B(z,t)$ oscillate **in phase** -- they reach their peaks and zeros at the same positions and times
- The Poynting vector $S \propto E \times B$ oscillates at **twice** the frequency
- $S$ is always **non-negative** (energy always flows in the propagation direction)
- The time average $\langle S \rangle$ (dashed line) gives the **intensity**

## 9. Radiation Pressure

Since EM waves carry energy, they also carry **momentum** and can exert **radiation pressure**:

$$P_{\text{absorbed}} = \frac{I}{c} \qquad \text{(complete absorption)}$$

$$P_{\text{reflected}} = \frac{2I}{c} \qquad \text{(perfect reflection)}$$

This is a tiny effect in everyday life, but it is important for:
- Comet tails (pushed by solar radiation)
- Solar sails for spacecraft propulsion
- Laser cooling of atoms
- Stellar structure (radiation pressure holds up stars)

```python
# ============================================================
# Interactive 4: Radiation Pressure Calculator
# ============================================================

def radiation_pressure_demo(log_intensity=3.0, surface_type='Perfect Absorber'):
    I = 10**log_intensity  # W/m^2
    c = 3e8
    
    if surface_type == 'Perfect Absorber':
        P = I / c
        factor = 1
    else:  # Perfect Reflector
        P = 2 * I / c
        factor = 2
    
    # Force on 1 m^2 area
    F = P * 1.0  # 1 m^2
    
    fig, ax = plt.subplots(1, 1, figsize=(10, 5))
    ax.axis('off')
    
    ax.text(0.5, 0.95, 'Radiation Pressure Calculator', fontsize=18,
           fontweight='bold', ha='center', va='top', transform=ax.transAxes)
    
    info_text = (
        f"Intensity I = {I:.2e} W/m\u00b2\n\n"
        f"Surface: {surface_type}\n"
        f"Formula: P = {factor}I/c\n\n"
        f"Radiation Pressure: P = {P:.4e} Pa\n"
        f"Force on 1 m\u00b2: F = {F:.4e} N\n\n"
    )
    
    # Comparison
    atm = 1.013e5
    ratio = P / atm
    info_text += f"Compare: {ratio:.2e} times atmospheric pressure"
    
    if I > 1e10:
        info_text += "\n\nThis is an extremely intense beam (like a focused laser)!"
    elif I > 1000:
        info_text += f"\n\nSolar intensity at Earth is about 1400 W/m\u00b2."
    
    ax.text(0.1, 0.75, info_text, fontsize=13, va='top',
           transform=ax.transAxes, family='monospace', linespacing=1.5)
    
    plt.tight_layout()
    plt.show()

intensity_slider = FloatSlider(
    value=3.0, min=0, max=15, step=0.5,
    description='log\u2081\u2080(I):',
    layout=Layout(width='500px'),
    style={'description_width': '80px'}
)

surface_dropdown = Dropdown(
    options=['Perfect Absorber', 'Perfect Reflector'],
    value='Perfect Absorber',
    description='Surface:',
    layout=Layout(width='300px'),
    style={'description_width': '70px'}
)

interact(radiation_pressure_demo, 
         log_intensity=intensity_slider,
         surface_type=surface_dropdown);
```

## 10. Polarization of EM Waves

The electric field of an EM wave oscillates in a specific direction perpendicular to propagation. The direction of $\vec{E}$ defines the **polarization** of the wave.

- **Linear polarization**: $\vec{E}$ oscillates along one fixed direction
- **Circular polarization**: $\vec{E}$ rotates in a circle as the wave propagates
- **Unpolarized light**: $\vec{E}$ randomly changes direction (like sunlight)

Malus's Law governs the intensity transmitted through a polarizer:

$$I = I_0 \cos^2\theta$$

where $\theta$ is the angle between the polarization direction and the polarizer axis.

```python
# ============================================================
# Interactive 5: Malus's Law Visualization
# ============================================================

def malus_law(theta_deg=0):
    theta_rad = np.radians(theta_deg)
    I_ratio = np.cos(theta_rad)**2
    
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    
    # Left: polar plot showing transmission
    ax1 = fig.add_subplot(121, projection='polar')
    angles = np.linspace(0, 2*np.pi, 360)
    intensities = np.cos(angles)**2
    ax1.plot(angles, intensities, 'b-', linewidth=2, alpha=0.5)
    ax1.fill(angles, intensities, alpha=0.1, color='blue')
    ax1.plot(theta_rad, I_ratio, 'ro', markersize=12, zorder=5)
    ax1.set_title(f"Malus's Law\nI/I\u2080 = cos\u00b2({theta_deg}\u00b0) = {I_ratio:.3f}",
                 fontsize=13, fontweight='bold', pad=20)
    
    axes[0].remove()
    
    # Right: bar showing transmitted intensity
    ax2 = axes[1]
    ax2.barh(0, 1.0, height=0.4, color='gold', alpha=0.3, label='Incident I\u2080')
    ax2.barh(0, I_ratio, height=0.4, color='gold', alpha=0.9, label=f'Transmitted: {I_ratio:.1%}')
    ax2.set_xlim(0, 1.1)
    ax2.set_xlabel('I / I\u2080', fontsize=13)
    ax2.set_yticks([])
    ax2.legend(fontsize=12, loc='upper right')
    ax2.set_title(f'Polarizer angle: {theta_deg}\u00b0', fontsize=13)
    
    plt.tight_layout()
    plt.show()

theta_slider = FloatSlider(
    value=0, min=0, max=90, step=1,
    description='\u03b8 (deg):',
    layout=Layout(width='500px'),
    style={'description_width': '70px'}
)

interact(malus_law, theta_deg=theta_slider);
```

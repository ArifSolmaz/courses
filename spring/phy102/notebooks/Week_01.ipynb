## 1. Electric Charge: Fundamental Properties

Electric charge is one of the fundamental properties of matter. Here are the key facts:

**Two types of charge:**
- **Positive (+):** carried by protons
- **Negative (-):** carried by electrons

**Key principles:**
- Like charges **repel**, opposite charges **attract**
- Charge is **conserved** -- it cannot be created or destroyed, only transferred
- Charge is **quantized** -- it comes in integer multiples of the elementary charge

$$e = 1.602 \times 10^{-19} \text{ C}$$

Any charge $q$ can be written as:

$$q = n \cdot e \quad \text{where } n = 0, \pm 1, \pm 2, \ldots$$

```python
# ============================================================
# Interactive: Charge Quantization Explorer
# ============================================================

def charge_quantization(n_charges=5):
    """Visualize charge quantization."""
    e = 1.602e-19  # elementary charge in Coulombs
    q_total = n_charges * e
    
    fig, ax = plt.subplots(1, 1, figsize=(10, 3))
    
    # Draw individual charges
    for i in range(abs(n_charges)):
        color = 'blue' if n_charges < 0 else 'red'
        sign = '-' if n_charges < 0 else '+'
        circle = plt.Circle((i * 1.2 + 0.6, 0.5), 0.4, color=color, alpha=0.6)
        ax.add_patch(circle)
        ax.text(i * 1.2 + 0.6, 0.5, sign, ha='center', va='center',
                fontsize=20, fontweight='bold', color='white')
    
    ax.set_xlim(-0.5, max(abs(n_charges) * 1.2 + 0.5, 2))
    ax.set_ylim(-0.5, 1.5)
    ax.set_aspect('equal')
    ax.set_title(f'n = {n_charges}  |  Total charge q = n x e = {n_charges} x (1.602 x 10$^{{-19}}$) = {q_total:.3e} C',
                fontsize=13)
    ax.axis('off')
    plt.tight_layout()
    plt.show()

interact(charge_quantization,
         n_charges=IntSlider(min=-12, max=12, step=1, value=5,
                               description='n (count):',
                               style={'description_width': 'initial'}));
```

## 2. Coulomb's Law

The force between two point charges $q_1$ and $q_2$ separated by a distance $r$ is given by **Coulomb's Law**:

$$F = k_e \frac{|q_1 \, q_2|}{r^2}$$

where:
- $k_e = 8.988 \times 10^9 \; \text{N m}^2/\text{C}^2$ is Coulomb's constant
- The force is **attractive** if the charges have opposite signs
- The force is **repulsive** if the charges have the same sign
- The force acts along the line joining the two charges

Coulomb's constant is also written as:

$$k_e = \frac{1}{4\pi \varepsilon_0}$$

where $\varepsilon_0 = 8.854 \times 10^{-12} \; \text{C}^2/(\text{N m}^2)$ is the permittivity of free space.

```python
# ============================================================
# Interactive Demo 1: Coulomb's Law Explorer
# ============================================================

def coulombs_law_explorer(q1=1.0, q2=-1.0, d=1.0):
    """
    Interactive visualization of Coulomb's Law.
    q1, q2 in microcoulombs; d in meters.
    """
    q1_C = q1 * 1e-6  # convert uC to C
    q2_C = q2 * 1e-6
    
    # Calculate force magnitude
    F = k_e * abs(q1_C * q2_C) / d**2
    attractive = (q1_C * q2_C) < 0
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # --- Left panel: Charge diagram with force arrows ---
    ax1.set_xlim(-3, 3)
    ax1.set_ylim(-1.5, 1.5)
    
    x1, x2 = -d/2, d/2
    
    # Draw charges
    c1_color = 'red' if q1 > 0 else 'blue'
    c2_color = 'red' if q2 > 0 else 'blue'
    size1 = 0.15 + 0.05 * abs(q1)
    size2 = 0.15 + 0.05 * abs(q2)
    
    circle1 = plt.Circle((x1, 0), size1, color=c1_color, alpha=0.7, zorder=5)
    circle2 = plt.Circle((x2, 0), size2, color=c2_color, alpha=0.7, zorder=5)
    ax1.add_patch(circle1)
    ax1.add_patch(circle2)
    
    # Labels
    sign1 = '+' if q1 > 0 else '-' if q1 < 0 else '0'
    sign2 = '+' if q2 > 0 else '-' if q2 < 0 else '0'
    ax1.text(x1, 0, sign1, ha='center', va='center', fontsize=16,
             fontweight='bold', color='white', zorder=6)
    ax1.text(x2, 0, sign2, ha='center', va='center', fontsize=16,
             fontweight='bold', color='white', zorder=6)
    ax1.text(x1, -0.6, f'$q_1$ = {q1:.1f} $\\mu$C', ha='center', fontsize=11)
    ax1.text(x2, -0.6, f'$q_2$ = {q2:.1f} $\\mu$C', ha='center', fontsize=11)
    
    # Force arrows
    arrow_scale = min(0.8, F * 1e-2)  # scale for visibility
    arrow_scale = max(0.1, arrow_scale)
    
    if q1 != 0 and q2 != 0:
        if attractive:
            # Arrows point toward each other
            ax1.annotate('', xy=(x1 + size1 + 0.05, 0),
                        xytext=(x1 + size1 + 0.05 + arrow_scale, 0),
                        arrowprops=dict(arrowstyle='->', color='green', lw=2.5))
            ax1.annotate('', xy=(x2 - size2 - 0.05, 0),
                        xytext=(x2 - size2 - 0.05 - arrow_scale, 0),
                        arrowprops=dict(arrowstyle='->', color='green', lw=2.5))
            ax1.text(0, 0.8, 'ATTRACTIVE', ha='center', fontsize=14,
                    color='green', fontweight='bold')
        else:
            # Arrows point away from each other
            ax1.annotate('', xy=(x1 - size1 - 0.05 - arrow_scale, 0),
                        xytext=(x1 - size1 - 0.05, 0),
                        arrowprops=dict(arrowstyle='->', color='orange', lw=2.5))
            ax1.annotate('', xy=(x2 + size2 + 0.05 + arrow_scale, 0),
                        xytext=(x2 + size2 + 0.05, 0),
                        arrowprops=dict(arrowstyle='->', color='orange', lw=2.5))
            ax1.text(0, 0.8, 'REPULSIVE', ha='center', fontsize=14,
                    color='orange', fontweight='bold')
    
    # Distance annotation
    ax1.annotate('', xy=(x2, -1.0), xytext=(x1, -1.0),
                arrowprops=dict(arrowstyle='<->', color='gray', lw=1.5))
    ax1.text(0, -1.2, f'd = {d:.2f} m', ha='center', fontsize=11, color='gray')
    
    ax1.set_title('Charge Configuration & Forces', fontsize=13)
    ax1.set_aspect('equal')
    ax1.grid(True, alpha=0.2)
    ax1.set_xlabel('x (m)')
    
    # --- Right panel: Force vs distance curve ---
    distances = np.linspace(0.1, 3.0, 200)
    forces = k_e * abs(q1_C * q2_C) / distances**2
    
    ax2.plot(distances, forces, 'b-', linewidth=2, label='F(r)')
    ax2.axvline(x=d, color='red', linestyle='--', alpha=0.7, label=f'd = {d:.2f} m')
    ax2.plot(d, F, 'ro', markersize=10, zorder=5)
    ax2.annotate(f'F = {F:.4f} N', xy=(d, F), xytext=(d + 0.3, F * 1.2),
               fontsize=11, color='red',
               arrowprops=dict(arrowstyle='->', color='red'))
    
    ax2.set_xlabel('Distance r (m)', fontsize=12)
    ax2.set_ylabel('Force |F| (N)', fontsize=12)
    ax2.set_title('Force vs. Distance (Inverse Square Law)', fontsize=13)
    ax2.legend(fontsize=11)
    ax2.set_ylim(bottom=0)
    
    plt.tight_layout()
    plt.show()
    
    # Print summary
    nature = "attractive" if attractive else "repulsive"
    if q1 == 0 or q2 == 0:
        nature = "zero (one charge is zero)"
    print(f"Force magnitude: |F| = k |q1 q2| / d^2 = {F:.6f} N")
    print(f"Nature of force: {nature}")

# Create interactive widget
interact(coulombs_law_explorer,
         q1=FloatSlider(min=-5.0, max=5.0, step=0.1, value=2.0,
                        description='q1 (uC):', style={'description_width': 'initial'},
                        layout=Layout(width='500px')),
         q2=FloatSlider(min=-5.0, max=5.0, step=0.1, value=-3.0,
                        description='q2 (uC):', style={'description_width': 'initial'},
                        layout=Layout(width='500px')),
         d=FloatSlider(min=0.1, max=3.0, step=0.05, value=1.0,
                       description='Distance (m):', style={'description_width': 'initial'},
                       layout=Layout(width='500px')));
```

## 3. Superposition Principle

When more than two charges are present, the **net force** on any charge is the **vector sum** of the individual forces from all other charges:

$$\vec{F}_{\text{net}} = \vec{F}_1 + \vec{F}_2 + \vec{F}_3 + \ldots = \sum_i \vec{F}_i$$

This is the **superposition principle** -- each pair of charges interacts independently, and we simply add up all the force vectors.

**Important:** The forces must be added as **vectors** (not just magnitudes). You must account for both the magnitude and direction of each force.

```python
# ============================================================
# Worked Example 1: Force Between Two Charges
# ============================================================

print("=" * 60)
print("WORKED EXAMPLE 1: Coulomb's Law")
print("=" * 60)
print()
print("Problem: Two charges q1 = +3 uC and q2 = -5 uC are")
print("separated by 0.20 m. Find the force between them.")
print()
print("Solution:")
print("-" * 40)

q1 = 3e-6   # C
q2 = -5e-6  # C
r = 0.20     # m

F = k_e * abs(q1 * q2) / r**2

print(f"  q1 = {q1:.1e} C")
print(f"  q2 = {q2:.1e} C")
print(f"  r  = {r} m")
print()
print(f"  F = k |q1 q2| / r^2")
print(f"    = ({k_e:.3e}) x |({q1:.1e})({q2:.1e})| / ({r})^2")
print(f"    = {F:.4f} N")
print(f"    = {F:.2f} N")
print()
print(f"  Since q1 > 0 and q2 < 0, the force is ATTRACTIVE.")
print(f"  The charges pull toward each other.")
```

```python
# ============================================================
# Worked Example 2: Superposition -- Net Force on a Charge
# ============================================================

print("=" * 60)
print("WORKED EXAMPLE 2: Superposition of Forces")
print("=" * 60)
print()
print("Problem: Three charges are placed along the x-axis:")
print("  q1 = +2 uC at x = 0")
print("  q2 = -3 uC at x = 0.30 m")
print("  q3 = +4 uC at x = 0.60 m")
print("Find the net force on q2.")
print()

q1_val = 2e-6
q2_val = -3e-6
q3_val = 4e-6
x1, x2, x3 = 0.0, 0.30, 0.60

# Force on q2 from q1
r12 = abs(x2 - x1)
F12_mag = k_e * abs(q1_val * q2_val) / r12**2
# q1 is positive, q2 is negative -> attractive -> F12 points LEFT (toward q1)
F12_x = -F12_mag  # negative x-direction

print(f"Force on q2 from q1:")
print(f"  r12 = {r12} m")
print(f"  |F12| = k|q1 q2|/r12^2 = {F12_mag:.4f} N")
print(f"  Direction: toward q1 (negative x) -> F12_x = {F12_x:.4f} N")
print()

# Force on q2 from q3
r23 = abs(x3 - x2)
F23_mag = k_e * abs(q2_val * q3_val) / r23**2
# q3 is positive, q2 is negative -> attractive -> F23 points RIGHT (toward q3)
F23_x = +F23_mag  # positive x-direction

print(f"Force on q2 from q3:")
print(f"  r23 = {r23} m")
print(f"  |F23| = k|q2 q3|/r23^2 = {F23_mag:.4f} N")
print(f"  Direction: toward q3 (positive x) -> F23_x = {F23_x:.4f} N")
print()

# Net force
F_net_x = F12_x + F23_x
print(f"Net force on q2:")
print(f"  F_net = F12 + F23 = {F12_x:.4f} + {F23_x:.4f} = {F_net_x:.4f} N")
direction = "positive x" if F_net_x > 0 else "negative x"
print(f"  Direction: {direction}")
print(f"  |F_net| = {abs(F_net_x):.4f} N")

# Visualization
fig, ax = plt.subplots(figsize=(10, 3))
charges = [(x1, q1_val, 'q1'), (x2, q2_val, 'q2'), (x3, q3_val, 'q3')]
for x, q, label in charges:
    color = 'red' if q > 0 else 'blue'
    ax.plot(x, 0, 'o', color=color, markersize=20, zorder=5)
    sign = '+' if q > 0 else '-'
    ax.text(x, 0, sign, ha='center', va='center', color='white',
            fontsize=14, fontweight='bold', zorder=6)
    ax.text(x, -0.4, f'{label}={q*1e6:.0f}uC', ha='center', fontsize=10)

# Force arrows on q2
scale = 0.1 / max(F12_mag, F23_mag)
ax.annotate('', xy=(x2 + F12_x * scale, 0.15), xytext=(x2, 0.15),
           arrowprops=dict(arrowstyle='->', color='purple', lw=2))
ax.annotate('', xy=(x2 + F23_x * scale, 0.25), xytext=(x2, 0.25),
           arrowprops=dict(arrowstyle='->', color='green', lw=2))
ax.annotate('', xy=(x2 + F_net_x * scale, -0.15), xytext=(x2, -0.15),
           arrowprops=dict(arrowstyle='->', color='red', lw=3))
ax.text(x2 - 0.08, 0.35, 'F12', color='purple', fontsize=10)
ax.text(x2 + 0.15, 0.35, 'F23', color='green', fontsize=10)
ax.text(x2 + 0.05, -0.30, 'F_net', color='red', fontsize=10, fontweight='bold')

ax.set_xlim(-0.2, 0.8)
ax.set_ylim(-0.6, 0.6)
ax.set_xlabel('x (m)')
ax.set_title('Superposition: Net Force on q2')
ax.axhline(y=0, color='gray', linewidth=0.5)
plt.tight_layout()
plt.show()
```

## 4. The Electric Field

The **electric field** $\vec{E}$ at a point in space is defined as the force per unit positive test charge:

$$\vec{E} = \frac{\vec{F}}{q_0}$$

For a point charge $Q$, the electric field at a distance $r$ from the charge is:

$$E = k_e \frac{|Q|}{r^2}$$

**Direction of E:**
- Points **away** from positive charges
- Points **toward** negative charges

**Units:** N/C (Newtons per Coulomb) or equivalently V/m (Volts per meter)

The electric field is a **vector field** -- at every point in space, it has both a magnitude and a direction.

```python
# ============================================================
# Helper functions for electric field calculations
# ============================================================

def E_field_point_charge(q, xq, yq, x, y):
    """
    Calculate E-field components (Ex, Ey) at point (x, y)
    due to a charge q located at (xq, yq).
    """
    dx = x - xq
    dy = y - yq
    r = np.sqrt(dx**2 + dy**2)
    r = np.where(r < 0.01, 0.01, r)  # avoid division by zero
    
    E_mag = k_e * q / r**2
    Ex = E_mag * dx / r
    Ey = E_mag * dy / r
    return Ex, Ey

def E_field_total(charges, x, y):
    """
    Calculate total E-field from a list of charges.
    charges: list of (q, xq, yq) tuples; q in Coulombs.
    """
    Ex_total = np.zeros_like(x, dtype=float)
    Ey_total = np.zeros_like(y, dtype=float)
    for q, xq, yq in charges:
        ex, ey = E_field_point_charge(q, xq, yq, x, y)
        Ex_total += ex
        Ey_total += ey
    return Ex_total, Ey_total

print("Electric field helper functions defined.")
```

```python
# ============================================================
# Interactive Demo 2: Electric Field Visualizer
# Place 2-3 charges and see field vectors on a grid
# ============================================================

def field_visualizer(q1=2.0, x1=-1.0, y1=0.0,
                     q2=-2.0, x2=1.0, y2=0.0,
                     q3=0.0, x3=0.0, y3=1.5):
    """
    Visualize electric field from 2-3 point charges.
    Charges in microcoulombs, positions in meters.
    Set q3=0 to use only two charges.
    """
    # Build charge list (convert uC to C)
    charge_list = [(q1 * 1e-6, x1, y1), (q2 * 1e-6, x2, y2)]
    if q3 != 0:
        charge_list.append((q3 * 1e-6, x3, y3))
    
    # Create grid
    grid_x = np.linspace(-3, 3, 20)
    grid_y = np.linspace(-3, 3, 20)
    X, Y = np.meshgrid(grid_x, grid_y)
    
    # Calculate field
    Ex, Ey = E_field_total(charge_list, X, Y)
    E_mag = np.sqrt(Ex**2 + Ey**2)
    
    # Normalize arrows for display
    E_mag_safe = np.where(E_mag < 1e-10, 1e-10, E_mag)
    Ex_norm = Ex / E_mag_safe
    Ey_norm = Ey / E_mag_safe
    
    # Plot
    fig, ax = plt.subplots(figsize=(9, 8))
    
    # Color map based on field magnitude (log scale)
    E_log = np.log10(E_mag + 1)
    
    ax.quiver(X, Y, Ex_norm, Ey_norm, E_log,
              cmap='inferno', alpha=0.8, scale=25, width=0.004)
    
    # Draw charges
    for q, xq, yq in charge_list:
        color = 'red' if q > 0 else 'blue'
        size = 200 + 100 * abs(q * 1e6)
        ax.scatter(xq, yq, c=color, s=size, zorder=10, edgecolors='black', linewidth=2)
        sign = '+' if q > 0 else '-'
        ax.text(xq, yq, sign, ha='center', va='center',
                fontsize=16, fontweight='bold', color='white', zorder=11)
    
    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)
    ax.set_xlabel('x (m)', fontsize=12)
    ax.set_ylabel('y (m)', fontsize=12)
    ax.set_title('Electric Field Vectors (arrow direction = E direction, color = magnitude)',
                fontsize=12)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.2)
    plt.tight_layout()
    plt.show()

# Create interactive widget
style = {'description_width': 'initial'}
layout = Layout(width='450px')

interact(field_visualizer,
         q1=FloatSlider(min=-5, max=5, step=0.5, value=2.0,
                        description='q1 (uC):', style=style, layout=layout),
         x1=FloatSlider(min=-2.5, max=2.5, step=0.25, value=-1.0,
                        description='x1 (m):', style=style, layout=layout),
         y1=FloatSlider(min=-2.5, max=2.5, step=0.25, value=0.0,
                        description='y1 (m):', style=style, layout=layout),
         q2=FloatSlider(min=-5, max=5, step=0.5, value=-2.0,
                        description='q2 (uC):', style=style, layout=layout),
         x2=FloatSlider(min=-2.5, max=2.5, step=0.25, value=1.0,
                        description='x2 (m):', style=style, layout=layout),
         y2=FloatSlider(min=-2.5, max=2.5, step=0.25, value=0.0,
                        description='y2 (m):', style=style, layout=layout),
         q3=FloatSlider(min=-5, max=5, step=0.5, value=0.0,
                        description='q3 (uC) [0=off]:', style=style, layout=layout),
         x3=FloatSlider(min=-2.5, max=2.5, step=0.25, value=0.0,
                        description='x3 (m):', style=style, layout=layout),
         y3=FloatSlider(min=-2.5, max=2.5, step=0.25, value=1.5,
                        description='y3 (m):', style=style, layout=layout));
```

## 5. Electric Field Lines

**Electric field lines** are a powerful way to visualize the electric field. The rules are:

1. Field lines **start** on positive charges and **end** on negative charges
2. The **direction** of the field line at any point gives the direction of $\vec{E}$ at that point
3. The **density** of field lines (number per unit area) is proportional to the field strength
4. Field lines **never cross** each other
5. The number of lines leaving/entering a charge is proportional to the magnitude of the charge

## 6. Electric Field from Multiple Charges: Superposition

Just as forces obey superposition, so does the electric field. The total electric field at any point is the vector sum of the fields from all individual charges:

$$\vec{E}_{\text{total}} = \sum_i \vec{E}_i = \sum_i k_e \frac{q_i}{r_i^2} \hat{r}_i$$

This is extremely powerful: we can compute the field from any charge distribution by adding up contributions from individual charges.

```python
# ============================================================
# Interactive Demo 4: Superposition -- Move Charges Around
# Shows field magnitude as a heatmap with vectors
# ============================================================

def superposition_explorer(q1=3.0, x1=-1.5, y1=0.0,
                           q2=-3.0, x2=1.5, y2=0.0,
                           q3=2.0, x3=0.0, y3=2.0):
    """
    Explore superposition with 3 movable charges.
    Shows field magnitude heatmap + vector arrows.
    """
    charge_list = [
        (q1 * 1e-6, x1, y1),
        (q2 * 1e-6, x2, y2),
        (q3 * 1e-6, x3, y3)
    ]
    # Remove zero charges
    charge_list = [(q, x, y) for q, x, y in charge_list if q != 0]
    
    # Fine grid for heatmap
    gx = np.linspace(-4, 4, 150)
    gy = np.linspace(-4, 4, 150)
    X_fine, Y_fine = np.meshgrid(gx, gy)
    
    # Coarse grid for arrows
    gx_c = np.linspace(-3.5, 3.5, 16)
    gy_c = np.linspace(-3.5, 3.5, 16)
    X_c, Y_c = np.meshgrid(gx_c, gy_c)
    
    if len(charge_list) == 0:
        print("All charges are zero. Set at least one charge to a nonzero value.")
        return
    
    Ex_f, Ey_f = E_field_total(charge_list, X_fine, Y_fine)
    E_mag_f = np.sqrt(Ex_f**2 + Ey_f**2)
    
    Ex_c, Ey_c = E_field_total(charge_list, X_c, Y_c)
    E_mag_c = np.sqrt(Ex_c**2 + Ey_c**2)
    E_mag_c_safe = np.where(E_mag_c < 1e-10, 1e-10, E_mag_c)
    
    fig, ax = plt.subplots(figsize=(9, 8))
    
    # Heatmap of field magnitude (log scale)
    im = ax.pcolormesh(X_fine, Y_fine, np.log10(E_mag_f + 1),
                       cmap='YlOrRd', shading='auto', alpha=0.6)
    plt.colorbar(im, ax=ax, label='log10(|E| + 1)', shrink=0.8)
    
    # Quiver (normalized arrows)
    ax.quiver(X_c, Y_c, Ex_c / E_mag_c_safe, Ey_c / E_mag_c_safe,
              color='black', alpha=0.6, scale=25, width=0.004)
    
    # Draw charges
    for q, xq, yq in charge_list:
        color = 'red' if q > 0 else 'blue'
        size = 300 + 150 * abs(q * 1e6)
        ax.scatter(xq, yq, c=color, s=size, zorder=10,
                   edgecolors='black', linewidth=2)
        sign = '+' if q > 0 else '-'
        ax.text(xq, yq, sign, ha='center', va='center',
                fontsize=16, fontweight='bold', color='white', zorder=11)
    
    ax.set_xlim(-4, 4)
    ax.set_ylim(-4, 4)
    ax.set_aspect('equal')
    ax.set_xlabel('x (m)', fontsize=12)
    ax.set_ylabel('y (m)', fontsize=12)
    ax.set_title('Superposition: Field Magnitude (color) + Direction (arrows)', fontsize=12)
    ax.grid(True, alpha=0.15)
    plt.tight_layout()
    plt.show()

style = {'description_width': 'initial'}
layout = Layout(width='450px')

interact(superposition_explorer,
         q1=FloatSlider(min=-5, max=5, step=0.5, value=3.0,
                        description='q1 (uC):', style=style, layout=layout),
         x1=FloatSlider(min=-3, max=3, step=0.25, value=-1.5,
                        description='x1 (m):', style=style, layout=layout),
         y1=FloatSlider(min=-3, max=3, step=0.25, value=0.0,
                        description='y1 (m):', style=style, layout=layout),
         q2=FloatSlider(min=-5, max=5, step=0.5, value=-3.0,
                        description='q2 (uC):', style=style, layout=layout),
         x2=FloatSlider(min=-3, max=3, step=0.25, value=1.5,
                        description='x2 (m):', style=style, layout=layout),
         y2=FloatSlider(min=-3, max=3, step=0.25, value=0.0,
                        description='y2 (m):', style=style, layout=layout),
         q3=FloatSlider(min=-5, max=5, step=0.5, value=2.0,
                        description='q3 (uC):', style=style, layout=layout),
         x3=FloatSlider(min=-3, max=3, step=0.25, value=0.0,
                        description='x3 (m):', style=style, layout=layout),
         y3=FloatSlider(min=-3, max=3, step=0.25, value=2.0,
                        description='y3 (m):', style=style, layout=layout));
```

```python
# ============================================================
# Worked Example 3: Electric Field at a Point
# ============================================================

print("=" * 60)
print("WORKED EXAMPLE 3: Electric Field from a Point Charge")
print("=" * 60)
print()
print("Problem: A charge Q = +5 uC is at the origin.")
print("Find the electric field at point P = (0.40, 0) m.")
print()

Q = 5e-6  # C
r = 0.40  # m

E = k_e * abs(Q) / r**2

print(f"Solution:")
print(f"  E = k |Q| / r^2")
print(f"    = ({k_e:.3e}) x ({abs(Q):.1e}) / ({r})^2")
print(f"    = {E:.4e} N/C")
print(f"    = {E/1e3:.2f} kN/C")
print()
print(f"  Direction: Since Q > 0, E points AWAY from Q,")
print(f"  i.e., in the +x direction at point P.")
print(f"  E_vec = ({E:.4e}, 0) N/C")
```

```python
# ============================================================
# Worked Example 4: Net Field from Two Charges
# ============================================================

print("=" * 60)
print("WORKED EXAMPLE 4: Net Electric Field at a Point")
print("=" * 60)
print()
print("Problem: Two charges are placed on the x-axis:")
print("  q1 = +4 uC at x = -0.30 m")
print("  q2 = -2 uC at x = +0.30 m")
print("Find the net electric field at point P = (0, 0.40) m.")
print()

q1_val = 4e-6
q2_val = -2e-6
# Charge positions
xq1, yq1 = -0.30, 0.0
xq2, yq2 = 0.30, 0.0
# Point P
xP, yP = 0.0, 0.40

# Distance from q1 to P
r1 = np.sqrt((xP - xq1)**2 + (yP - yq1)**2)
# Distance from q2 to P
r2 = np.sqrt((xP - xq2)**2 + (yP - yq2)**2)

print(f"Step 1: Find distances")
print(f"  r1 = sqrt(({xP}-({xq1}))^2 + ({yP}-{yq1})^2) = {r1:.4f} m")
print(f"  r2 = sqrt(({xP}-{xq2})^2 + ({yP}-{yq2})^2) = {r2:.4f} m")
print()

# E-field magnitudes
E1 = k_e * abs(q1_val) / r1**2
E2 = k_e * abs(q2_val) / r2**2

print(f"Step 2: Find field magnitudes")
print(f"  |E1| = k|q1|/r1^2 = {E1:.4e} N/C")
print(f"  |E2| = k|q2|/r2^2 = {E2:.4e} N/C")
print()

# Components using our helper function
E1x, E1y = E_field_point_charge(q1_val, xq1, yq1, xP, yP)
E2x, E2y = E_field_point_charge(q2_val, xq2, yq2, xP, yP)

print(f"Step 3: Find components")
print(f"  E1: ({E1x:.4e}, {E1y:.4e}) N/C  [points away from q1, since q1 > 0]")
print(f"  E2: ({E2x:.4e}, {E2y:.4e}) N/C  [points toward q2, since q2 < 0]")
print()

# Net field
Ex_net = E1x + E2x
Ey_net = E1y + E2y
E_net_mag = np.sqrt(Ex_net**2 + Ey_net**2)
theta = np.degrees(np.arctan2(Ey_net, Ex_net))

print(f"Step 4: Superposition (vector sum)")
print(f"  E_net_x = {E1x:.4e} + {E2x:.4e} = {Ex_net:.4e} N/C")
print(f"  E_net_y = {E1y:.4e} + {E2y:.4e} = {Ey_net:.4e} N/C")
print(f"  |E_net| = {E_net_mag:.4e} N/C")
print(f"  Direction: {theta:.1f} degrees from +x axis")

# Diagram
fig, ax = plt.subplots(figsize=(8, 7))

# Charges
ax.plot(xq1, yq1, 'ro', markersize=15, zorder=10)
ax.plot(xq2, yq2, 'bo', markersize=15, zorder=10)
ax.text(xq1, yq1 - 0.08, '+', ha='center', va='center', color='white',
        fontsize=12, fontweight='bold', zorder=11)
ax.text(xq2, yq2 - 0.08, '-', ha='center', va='center', color='white',
        fontsize=12, fontweight='bold', zorder=11)
ax.text(xq1, -0.15, f'q1={q1_val*1e6:.0f}uC', ha='center', fontsize=10)
ax.text(xq2, -0.15, f'q2={q2_val*1e6:.0f}uC', ha='center', fontsize=10)

# Point P
ax.plot(xP, yP, 'k^', markersize=12, zorder=10)
ax.text(xP + 0.05, yP + 0.05, 'P', fontsize=14, fontweight='bold')

# E-field vectors at P (scaled for visibility)
scale = 3e-6
ax.quiver(xP, yP, E1x * scale, E1y * scale, angles='xy', scale_units='xy',
          scale=1, color='red', width=0.008, label='E1', zorder=8)
ax.quiver(xP, yP, E2x * scale, E2y * scale, angles='xy', scale_units='xy',
          scale=1, color='blue', width=0.008, label='E2', zorder=8)
ax.quiver(xP, yP, Ex_net * scale, Ey_net * scale, angles='xy', scale_units='xy',
          scale=1, color='green', width=0.01, label='E_net', zorder=9)

ax.set_xlim(-0.8, 0.8)
ax.set_ylim(-0.3, 1.0)
ax.set_aspect('equal')
ax.set_xlabel('x (m)')
ax.set_ylabel('y (m)')
ax.set_title('Net Electric Field at Point P', fontsize=13)
ax.legend(fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 8. Special Configurations: Dipole and Symmetric Arrangements

An **electric dipole** consists of two equal and opposite charges ($+q$ and $-q$) separated by a distance $d$. The **dipole moment** is:

$$\vec{p} = q \vec{d}$$

where $\vec{d}$ points from the negative to the positive charge.

Dipoles are important because many molecules are electric dipoles (e.g., water), and the dipole field pattern appears frequently in nature.

```python
# ============================================================
# Interactive: Common Charge Configurations
# ============================================================

def show_configuration(config='Dipole', charge_mag=2.0, separation=2.0):
    """
    Show field lines for common charge configurations.
    """
    q = charge_mag * 1e-6
    d = separation / 2
    
    if config == 'Dipole':
        charges = [(q, -d, 0), (-q, d, 0)]
        title = 'Electric Dipole (+q / -q)'
    elif config == 'Two Positive':
        charges = [(q, -d, 0), (q, d, 0)]
        title = 'Two Equal Positive Charges'
    elif config == 'Two Negative':
        charges = [(-q, -d, 0), (-q, d, 0)]
        title = 'Two Equal Negative Charges'
    elif config == 'Unequal (2:1)':
        charges = [(2*q, -d, 0), (-q, d, 0)]
        title = 'Unequal Charges (+2q / -q)'
    
    gx = np.linspace(-5, 5, 120)
    gy = np.linspace(-5, 5, 120)
    X, Y = np.meshgrid(gx, gy)
    
    Ex, Ey = E_field_total(charges, X, Y)
    E_mag = np.sqrt(Ex**2 + Ey**2)
    
    fig, ax = plt.subplots(figsize=(9, 8))
    
    lw = 2 * np.log1p(E_mag) / np.log1p(E_mag).max()
    ax.streamplot(X, Y, Ex, Ey, color=np.log1p(E_mag),
                  cmap='coolwarm', density=2.5, linewidth=lw,
                  arrowsize=1.5)
    
    for qi, xq, yq in charges:
        color = 'red' if qi > 0 else 'blue'
        ax.plot(xq, yq, 'o', color=color, markersize=18,
                markeredgecolor='black', markeredgewidth=2, zorder=10)
        sign = '+' if qi > 0 else '-'
        ax.text(xq, yq, sign, ha='center', va='center',
                fontsize=14, fontweight='bold', color='white', zorder=11)
    
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.set_aspect('equal')
    ax.set_xlabel('x (m)')
    ax.set_ylabel('y (m)')
    ax.set_title(title, fontsize=14)
    ax.grid(True, alpha=0.15)
    plt.tight_layout()
    plt.show()

interact(show_configuration,
         config=widgets.Dropdown(
             options=['Dipole', 'Two Positive', 'Two Negative', 'Unequal (2:1)'],
             value='Dipole', description='Configuration:',
             style={'description_width': 'initial'}),
         charge_mag=FloatSlider(min=0.5, max=5, step=0.5, value=2.0,
                                description='|q| (uC):', style={'description_width': 'initial'},
                                layout=Layout(width='450px')),
         separation=FloatSlider(min=0.5, max=4, step=0.25, value=2.0,
                                description='Separation (m):',
                                style={'description_width': 'initial'},
                                layout=Layout(width='450px')));
```

```python
# ============================================================
# Animation: Dipole Field Rotation
# Shows how field lines look as we rotate a dipole
# ============================================================

fig_anim, ax_anim = plt.subplots(figsize=(8, 8))

n_frames_rot = 30
angles_rot = np.linspace(0, 2 * np.pi, n_frames_rot, endpoint=False)

gx = np.linspace(-4, 4, 40)
gy = np.linspace(-4, 4, 40)
X_r, Y_r = np.meshgrid(gx, gy)

def init_dipole():
    ax_anim.clear()
    return []

def animate_dipole(frame):
    ax_anim.clear()
    angle = angles_rot[frame]
    d = 1.0
    q_val = 2e-6
    
    x_pos = d * np.cos(angle)
    y_pos = d * np.sin(angle)
    
    charges = [(q_val, x_pos, y_pos), (-q_val, -x_pos, -y_pos)]
    Ex, Ey = E_field_total(charges, X_r, Y_r)
    E_mag = np.sqrt(Ex**2 + Ey**2)
    
    lw = 1.5 * np.log1p(E_mag) / np.log1p(E_mag).max()
    ax_anim.streamplot(X_r, Y_r, Ex, Ey, color=np.log1p(E_mag),
                       cmap='coolwarm', density=1.8, linewidth=lw, arrowsize=1.2)
    
    ax_anim.plot(x_pos, y_pos, 'ro', markersize=14, markeredgecolor='black',
                markeredgewidth=2, zorder=10)
    ax_anim.plot(-x_pos, -y_pos, 'bo', markersize=14, markeredgecolor='black',
                markeredgewidth=2, zorder=10)
    ax_anim.text(x_pos, y_pos, '+', ha='center', va='center',
                color='white', fontsize=11, fontweight='bold', zorder=11)
    ax_anim.text(-x_pos, -y_pos, '-', ha='center', va='center',
                color='white', fontsize=11, fontweight='bold', zorder=11)
    
    ax_anim.set_xlim(-4, 4)
    ax_anim.set_ylim(-4, 4)
    ax_anim.set_aspect('equal')
    ax_anim.set_title(f'Rotating Electric Dipole (angle = {np.degrees(angle):.0f} deg)', fontsize=13)
    ax_anim.grid(True, alpha=0.15)

anim_dipole = FuncAnimation(fig_anim, animate_dipole, init_func=init_dipole,
                            frames=n_frames_rot, interval=150)
plt.close(fig_anim)
HTML(anim_dipole.to_jshtml())
```

- Field lines are a visual tool: they start on + charges, end on - charges

**P5.** *(Capacitors + DC Circuits)* A 24 V battery with internal resistance $r = 0.50\;\Omega$ is connected to two capacitors: $C_1 = 4.7\;\mu$F and $C_2 = 10\;\mu$F in series, through an external resistance $R = 220\;\Omega$. (a) Find the equivalent capacitance of the series combination. (b) After the capacitors are fully charged, what is the voltage across each capacitor? (c) How much total energy is stored in the capacitor combination? (d) If the battery is then disconnected and a dielectric slab ($\kappa = 3.0$) is inserted into $C_1$, what are the new voltages across each capacitor?

<details><summary>Answer</summary>

**(a)** Series capacitance:
$$\frac{1}{C_{\rm eq}} = \frac{1}{C_1} + \frac{1}{C_2} = \frac{1}{4.7\;\mu\text{F}} + \frac{1}{10\;\mu\text{F}} = \frac{10 + 4.7}{47} = \frac{14.7}{47}$$
$$C_{\rm eq} = \frac{47}{14.7} = 3.197\;\mu\text{F} \approx 3.20\;\mu\text{F}$$

**(b)** When fully charged, no current flows, so no voltage drop across $R$ or $r$. The full battery voltage appears across the series combination:
$$Q = C_{\rm eq} \cdot V = 3.197 \times 10^{-6} \times 24 = 76.7\;\mu\text{C}$$
Both capacitors carry the same charge $Q$ (series):
$$V_1 = \frac{Q}{C_1} = \frac{76.7\;\mu\text{C}}{4.7\;\mu\text{F}} = 16.32\;\text{V}$$
$$V_2 = \frac{Q}{C_2} = \frac{76.7\;\mu\text{C}}{10\;\mu\text{F}} = 7.67\;\text{V}$$
Check: $V_1 + V_2 = 16.32 + 7.67 \approx 24.0$ V. ✓

**(c)** Total stored energy:
$$U = \frac{Q^2}{2C_{\rm eq}} = \frac{(76.7 \times 10^{-6})^2}{2 \times 3.197 \times 10^{-6}} = \frac{5.883 \times 10^{-9}}{6.394 \times 10^{-6}} = 9.20 \times 10^{-4}\;\text{J} \approx 0.92\;\text{mJ}$$

**(d)** Battery disconnected → charge $Q = 76.7\;\mu$C is conserved. Inserting dielectric changes $C_1' = \kappa C_1 = 3.0 \times 4.7 = 14.1\;\mu$F. The new equivalent capacitance is $C_{\rm eq}' = \frac{C_1' C_2}{C_1' + C_2} = \frac{14.1 \times 10}{24.1} = 5.85\;\mu$F, but the charge stays at $Q = 76.7\;\mu$C. New voltages:
$$V_1' = \frac{Q}{C_1'} = \frac{76.7}{14.1} = 5.44\;\text{V}, \quad V_2' = \frac{Q}{C_2} = \frac{76.7}{10} = 7.67\;\text{V}$$
Total voltage drops from 24 V to $5.44 + 7.67 = 13.1$ V — the energy difference was converted to work pulling the dielectric into the capacitor.

</details>

**P6.** *(Magnetic Force + Faraday Induction)* A straight copper rod of length $L = 0.40$ m and mass $m = 50$ g slides without friction on two horizontal, parallel conducting rails separated by $L$, in a region of uniform vertical magnetic field $B = 0.80$ T. The rails are connected at one end by a resistor $R = 2.0\;\Omega$. The rod is given an initial push to the right with speed $v_0 = 5.0$ m/s and then released (no external force after the push). Neglect the resistance of the rod and rails.

(a) What is the initial induced EMF and current?
(b) What is the initial braking force on the rod?
(c) Show that the rod decelerates exponentially: $v(t) = v_0\, e^{-t/\tau}$, and find the time constant $\tau$.
(d) How far does the rod travel before it effectively stops (to 1% of $v_0$)?

<details><summary>Answer</summary>

**(a)** The motional EMF is $\mathcal{E} = BLv_0 = 0.80 \times 0.40 \times 5.0 = 1.60$ V.
Current: $I = \mathcal{E}/R = 1.60/2.0 = 0.80$ A. By Lenz's law the current flows so as to oppose the rod's motion.

**(b)** The magnetic braking force on the current-carrying rod:
$$F = BIL = 0.80 \times 0.80 \times 0.40 = 0.256\;\text{N}$$
This force opposes the velocity (to the left).

**(c)** At any instant: $\mathcal{E} = BLv$, $I = BLv/R$, braking force $F = BIL = B^2 L^2 v / R$. Newton's second law:
$$m\frac{dv}{dt} = -\frac{B^2 L^2}{R}\,v$$
This gives exponential decay: $v(t) = v_0\,e^{-t/\tau}$ where
$$\tau = \frac{mR}{B^2 L^2} = \frac{0.050 \times 2.0}{(0.80)^2 (0.40)^2} = \frac{0.10}{0.1024} = 0.977\;\text{s} \approx 0.98\;\text{s}$$

**(d)** Distance: $x = \int_0^\infty v_0 e^{-t/\tau}\,dt = v_0 \tau = 5.0 \times 0.977 = 4.88$ m (total distance). To reach 1% of $v_0$: $t_{1\%} = \tau \ln(100) = 0.977 \times 4.605 = 4.50$ s. Distance to that point:
$$x = v_0 \tau (1 - e^{-t_{1\%}/\tau}) = 5.0 \times 0.977 \times (1 - 0.01) = 4.83\;\text{m}$$

</details>

## 3. Mini-Project Option 1: RC/RLC Filter Design

**Goal**: Design a filter circuit for a target frequency and visualize its frequency response.

### Background

An **RLC series circuit** acts as a bandpass filter. Near the resonance frequency $f_0 = \frac{1}{2\pi\sqrt{LC}}$, the impedance is minimum and current is maximum.

- **Low-pass filter** (RC): passes low frequencies, attenuates high frequencies
- **High-pass filter** (RC): passes high frequencies, attenuates low frequencies  
- **Bandpass filter** (RLC): passes frequencies near resonance

```python
def rc_rlc_filter(R_ohm=100.0, C_uF=1.0, L_mH=10.0, filter_type='Low-Pass (RC)'):
    """
    Mini-Project 1: RC/RLC Filter Design.
    Plot the frequency response (Bode plot) for different filter types.
    """
    R = R_ohm
    C = C_uF * 1e-6
    L = L_mH * 1e-3
    
    f = np.logspace(1, 6, 1000)  # 10 Hz to 1 MHz
    omega = 2 * np.pi * f
    
    if filter_type == 'Low-Pass (RC)':
        # V_out/V_in = 1/(1 + j*omega*R*C)
        H = 1 / np.sqrt(1 + (omega * R * C)**2)
        f_c = 1 / (2 * np.pi * R * C)
        title = f'Low-Pass RC Filter (R={R:.0f} Ohm, C={C_uF:.1f} uF)'
        fc_label = f'Cutoff: {f_c:.1f} Hz'
    elif filter_type == 'High-Pass (RC)':
        # V_out/V_in = j*omega*R*C / (1 + j*omega*R*C)
        H = (omega * R * C) / np.sqrt(1 + (omega * R * C)**2)
        f_c = 1 / (2 * np.pi * R * C)
        title = f'High-Pass RC Filter (R={R:.0f} Ohm, C={C_uF:.1f} uF)'
        fc_label = f'Cutoff: {f_c:.1f} Hz'
    else:  # Bandpass RLC
        X_L = omega * L
        X_C = 1 / (omega * C)
        Z = np.sqrt(R**2 + (X_L - X_C)**2)
        H = R / Z  # voltage across R / total voltage
        f_c = 1 / (2 * np.pi * np.sqrt(L * C))
        Q_factor = (1/R) * np.sqrt(L/C)
        title = f'Bandpass RLC Filter (R={R:.0f} Ohm, C={C_uF:.1f} uF, L={L_mH:.1f} mH)'
        fc_label = f'Resonance: {f_c:.1f} Hz, Q={Q_factor:.1f}'
    
    # Convert to dB
    H_dB = 20 * np.log10(H + 1e-15)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
    
    # Magnitude plot
    ax1.semilogx(f, H_dB, 'b-', linewidth=2)
    ax1.axvline(f_c, color='r', linestyle='--', alpha=0.7, label=fc_label)
    ax1.axhline(-3, color='gray', linestyle=':', alpha=0.5, label='-3 dB')
    ax1.set_xlabel('Frequency (Hz)')
    ax1.set_ylabel('Gain (dB)')
    ax1.set_title(title)
    ax1.legend(fontsize=10)
    ax1.grid(True, alpha=0.3, which='both')
    ax1.set_ylim(-40, 5)
    
    # Linear magnitude plot
    ax2.semilogx(f, H, 'b-', linewidth=2)
    ax2.axvline(f_c, color='r', linestyle='--', alpha=0.7)
    ax2.axhline(1/np.sqrt(2), color='gray', linestyle=':', alpha=0.5, label='0.707 (-3 dB)')
    ax2.fill_between(f, H, alpha=0.1, color='blue')
    ax2.set_xlabel('Frequency (Hz)')
    ax2.set_ylabel('|H(f)| (linear)')
    ax2.set_title('Linear Scale')
    ax2.legend(fontsize=10)
    ax2.grid(True, alpha=0.3, which='both')
    
    plt.tight_layout()
    plt.show()
    
    print(f"Critical frequency: {f_c:.2f} Hz")
    if 'Bandpass' in filter_type:
        print(f"Quality factor Q: {Q_factor:.2f}")
        bw = f_c / Q_factor
        print(f"Bandwidth: {bw:.2f} Hz")

interact(rc_rlc_filter,
         R_ohm=FloatSlider(min=10, max=1000, step=10, value=100,
                           description='R (Ohm):', style={'description_width': 'initial'}),
         C_uF=FloatSlider(min=0.1, max=100, step=0.1, value=1.0,
                          description='C (uF):', style={'description_width': 'initial'}),
         L_mH=FloatSlider(min=1, max=100, step=1, value=10,
                          description='L (mH):', style={'description_width': 'initial'}),
         filter_type=Dropdown(options=['Low-Pass (RC)', 'High-Pass (RC)', 'Bandpass (RLC)'],
                              value='Low-Pass (RC)', description='Filter:',
                              style={'description_width': 'initial'}));
```

## 4. Mini-Project Option 2: Electromagnetic Generator Model

**Goal**: Model a rotating coil in a magnetic field, analyze the induced EMF, and examine energy conversion.

### Background

A rectangular coil of $N$ turns, area $A$, rotating at angular velocity $\omega$ in a uniform magnetic field $B$ produces an EMF:

$$\mathcal{E}(t) = N B A \omega \sin(\omega t) = \mathcal{E}_0 \sin(\omega t)$$

where $\mathcal{E}_0 = NBA\omega$ is the peak EMF.

```python
def em_generator(N_turns=100, B_T=0.5, A_cm2=50.0, rpm=1800, R_load=50.0):
    """
    Mini-Project 2: Electromagnetic Generator Model.
    Rotating coil EMF and energy analysis.
    """
    A = A_cm2 * 1e-4  # area in m^2
    omega = rpm * 2 * np.pi / 60  # angular velocity in rad/s
    f = rpm / 60  # frequency in Hz
    
    # Peak EMF
    E0 = N_turns * B_T * A * omega
    E_rms = E0 / np.sqrt(2)
    
    # Time array (3 full cycles)
    T = 1 / f
    t = np.linspace(0, 3 * T, 1000)
    
    # EMF, current, power
    emf = E0 * np.sin(omega * t)
    current = emf / R_load
    power = emf**2 / R_load
    P_avg = E_rms**2 / R_load
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 9))
    
    # EMF vs time
    axes[0, 0].plot(t * 1000, emf, 'b-', linewidth=2)
    axes[0, 0].axhline(E0, color='r', linestyle='--', alpha=0.5, label=f'E0 = {E0:.1f} V')
    axes[0, 0].axhline(-E0, color='r', linestyle='--', alpha=0.5)
    axes[0, 0].axhline(E_rms, color='g', linestyle=':', alpha=0.5, label=f'E_rms = {E_rms:.1f} V')
    axes[0, 0].set_xlabel('Time (ms)')
    axes[0, 0].set_ylabel('EMF (V)')
    axes[0, 0].set_title('Induced EMF')
    axes[0, 0].legend(fontsize=9)
    axes[0, 0].grid(True, alpha=0.3)
    
    # Current vs time
    axes[0, 1].plot(t * 1000, current * 1000, 'r-', linewidth=2)
    axes[0, 1].set_xlabel('Time (ms)')
    axes[0, 1].set_ylabel('Current (mA)')
    axes[0, 1].set_title(f'Current through R = {R_load:.0f} Ohm load')
    axes[0, 1].grid(True, alpha=0.3)
    
    # Power vs time
    axes[1, 0].plot(t * 1000, power, 'g-', linewidth=2)
    axes[1, 0].axhline(P_avg, color='orange', linestyle='--', linewidth=2,
                        label=f'P_avg = {P_avg:.2f} W')
    axes[1, 0].fill_between(t * 1000, power, alpha=0.15, color='green')
    axes[1, 0].set_xlabel('Time (ms)')
    axes[1, 0].set_ylabel('Power (W)')
    axes[1, 0].set_title('Instantaneous Power')
    axes[1, 0].legend(fontsize=10)
    axes[1, 0].grid(True, alpha=0.3)
    
    # Magnetic flux vs time
    flux = N_turns * B_T * A * np.cos(omega * t)
    axes[1, 1].plot(t * 1000, flux * 1000, 'm-', linewidth=2)
    axes[1, 1].set_xlabel('Time (ms)')
    axes[1, 1].set_ylabel('Total Flux (mWb)')
    axes[1, 1].set_title('Magnetic Flux through Coil')
    axes[1, 1].grid(True, alpha=0.3)
    
    fig.suptitle(f'EM Generator: N={N_turns}, B={B_T} T, A={A_cm2} cm^2, '
                 f'{rpm} RPM, f={f:.1f} Hz', fontsize=12, y=1.01)
    plt.tight_layout()
    plt.show()
    
    print(f"Generator Parameters:")
    print(f"  Peak EMF:      {E0:.2f} V")
    print(f"  RMS EMF:       {E_rms:.2f} V")
    print(f"  Frequency:     {f:.1f} Hz")
    print(f"  Peak Current:  {E0/R_load*1000:.1f} mA")
    print(f"  Average Power: {P_avg:.3f} W")

interact(em_generator,
         N_turns=IntSlider(min=10, max=500, step=10, value=100,
                           description='N turns:', style={'description_width': 'initial'}),
         B_T=FloatSlider(min=0.1, max=2.0, step=0.1, value=0.5,
                         description='B (T):', style={'description_width': 'initial'}),
         A_cm2=FloatSlider(min=10, max=200, step=10, value=50,
                           description='A (cm^2):', style={'description_width': 'initial'}),
         rpm=IntSlider(min=60, max=3600, step=60, value=1800,
                       description='RPM:', style={'description_width': 'initial'}),
         R_load=FloatSlider(min=10, max=500, step=10, value=50,
                            description='R_load (Ohm):', style={'description_width': 'initial'}));
```

$$\frac{1}{f} = \frac{1}{d_o} + \frac{1}{d_i}$$

The total magnification is the product of individual magnifications.

```python
def two_lens_system(f1_cm=10.0, f2_cm=5.0, d_obj_cm=25.0, sep_cm=20.0):
    """
    Mini-Project 3: Two-Lens Optical System with ray tracing.
    """
    f1 = f1_cm  # focal length of lens 1 (cm)
    f2 = f2_cm  # focal length of lens 2 (cm)
    do1 = d_obj_cm  # object distance from lens 1 (cm)
    separation = sep_cm  # distance between lenses (cm)
    
    # Lens 1: find image
    if abs(do1 - f1) < 0.01:
        print("Object at focal point of Lens 1 -- image at infinity.")
        return
    di1 = 1 / (1/f1 - 1/do1)
    M1 = -di1 / do1
    
    # Object for Lens 2
    do2 = separation - di1  # distance from lens 2
    
    if abs(do2 - f2) < 0.01:
        print("Intermediate image at focal point of Lens 2 -- final image at infinity.")
        return
    
    if abs(do2) < 0.01:
        print("Intermediate image on Lens 2 -- adjust separation.")
        return
    
    di2 = 1 / (1/f2 - 1/do2)
    M2 = -di2 / do2
    M_total = M1 * M2
    
    # Ray tracing visualization
    fig, ax = plt.subplots(figsize=(15, 6))
    
    # Positions along optical axis
    lens1_x = 0
    lens2_x = separation
    obj_x = -do1
    img1_x = di1
    img2_x = separation + di2
    
    # Object height
    h_obj = 2.0  # cm
    h_img1 = h_obj * M1
    h_img2 = h_obj * M_total
    
    # Draw optical axis
    x_range = [min(obj_x - 5, img2_x - 5), max(img2_x + 5, obj_x + 5, separation + 15)]
    # Clamp x_range to prevent extreme plot ranges when di2 is very large/negative
    x_min = max(x_range[0], -200)
    x_max = min(x_range[1], 200)
    ax.axhline(0, color='gray', linewidth=0.5, linestyle='-')
    
    # Draw lenses
    lens_h = 4
    ax.annotate('', xy=(lens1_x, lens_h), xytext=(lens1_x, -lens_h),
                arrowprops=dict(arrowstyle='<->', color='blue', lw=2))
    ax.text(lens1_x, lens_h + 0.5, f'Lens 1\nf={f1:.0f} cm', ha='center', fontsize=9, color='blue')
    
    ax.annotate('', xy=(lens2_x, lens_h), xytext=(lens2_x, -lens_h),
                arrowprops=dict(arrowstyle='<->', color='green', lw=2))
    ax.text(lens2_x, lens_h + 0.5, f'Lens 2\nf={f2:.0f} cm', ha='center', fontsize=9, color='green')
    
    # Draw focal points
    for fx in [lens1_x - f1, lens1_x + f1]:
        ax.plot(fx, 0, 'bx', markersize=8)
    for fx in [lens2_x - f2, lens2_x + f2]:
        ax.plot(fx, 0, 'gx', markersize=8)
    
    # Draw object
    ax.annotate('', xy=(obj_x, h_obj), xytext=(obj_x, 0),
                arrowprops=dict(arrowstyle='->', color='red', lw=2.5))
    ax.text(obj_x, h_obj + 0.5, 'Object', ha='center', fontsize=10, color='red')
    
    # Draw intermediate image
    style_img1 = '--' if di1 < 0 else '-'
    color_img1 = 'orange'
    ax.annotate('', xy=(img1_x, h_img1), xytext=(img1_x, 0),
                arrowprops=dict(arrowstyle='->', color=color_img1, lw=2, linestyle=style_img1))
    label1 = 'Image 1' + (' (virtual)' if di1 < 0 else '')
    ax.text(img1_x, h_img1 + 0.5 * np.sign(h_img1), label1, ha='center', fontsize=9, color=color_img1)
    
    # Draw final image
    color_img2 = 'purple'
    style_img2 = '--' if di2 < 0 else '-'
    ax.annotate('', xy=(img2_x, h_img2), xytext=(img2_x, 0),
                arrowprops=dict(arrowstyle='->', color=color_img2, lw=2.5, linestyle=style_img2))
    label2 = 'Final Image' + (' (virtual)' if di2 < 0 else '')
    ax.text(img2_x, h_img2 + 0.5 * np.sign(h_img2), label2, ha='center', fontsize=10,
            color=color_img2, fontweight='bold')
    
    # Draw principal rays through Lens 1
    # Ray 1: parallel to axis, through focal point after lens
    ax.plot([obj_x, lens1_x], [h_obj, h_obj], 'r-', alpha=0.4, linewidth=1)
    ax.plot([lens1_x, img1_x], [h_obj, h_img1], 'r-', alpha=0.4, linewidth=1)
    
    # Ray 2: through center of lens
    ax.plot([obj_x, img1_x], [h_obj, h_img1], 'r--', alpha=0.3, linewidth=1)
    
    # Set limits
    y_max = max(abs(h_obj), abs(h_img1), abs(h_img2), lens_h) + 2
    y_max = min(y_max, 50)  # clamp y as well for extreme magnifications
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(-y_max, y_max)
    ax.set_xlabel('Position along optical axis (cm)')
    ax.set_ylabel('Height (cm)')
    ax.set_title('Two-Lens Ray Tracing Diagram')
    ax.grid(True, alpha=0.2)
    
    plt.tight_layout()
    plt.show()
    
    # Print analysis
    print(f"=== Two-Lens System Analysis ===")
    print(f"\nLens 1: f1 = {f1:.1f} cm, object at do1 = {do1:.1f} cm")
    print(f"  Image 1 at di1 = {di1:.2f} cm {'(real)' if di1 > 0 else '(virtual)'}")
    print(f"  Magnification M1 = {M1:.3f}")
    print(f"\nLens 2: f2 = {f2:.1f} cm, object at do2 = {do2:.2f} cm")
    print(f"  Final image at di2 = {di2:.2f} cm from Lens 2 {'(real)' if di2 > 0 else '(virtual)'}")
    print(f"  Magnification M2 = {M2:.3f}")
    print(f"\nTotal magnification: M = M1 * M2 = {M_total:.3f}")
    print(f"Image is {'inverted' if M_total < 0 else 'upright'} and "
          f"{'magnified' if abs(M_total) > 1 else 'reduced'}")

interact(two_lens_system,
         f1_cm=FloatSlider(min=3, max=30, step=1, value=10,
                           description='f1 (cm):', style={'description_width': 'initial'}),
         f2_cm=FloatSlider(min=2, max=20, step=1, value=5,
                           description='f2 (cm):', style={'description_width': 'initial'}),
         d_obj_cm=FloatSlider(min=5, max=60, step=1, value=25,
                              description='Object dist (cm):', style={'description_width': 'initial'}),
         sep_cm=FloatSlider(min=5, max=50, step=1, value=20,
                            description='Lens sep (cm):', style={'description_width': 'initial'}));
```

## Exercise 1: Look Up a Calibration Factor (Easy)

The calibration table contains numeric factors for known sensors; `None` is not a stored factor in this exercise. For every requested sensor, print its factor. If a sensor has no factor, print `no calibration factor` instead of stopping with an error.

### Reasoning and plan

Loop over the requested names, not over the dictionary. `get()` returns `None` for an unknown sensor, so the program can print a message instead of raising `KeyError`. The dictionary itself is not changed.

**Türkçe:** Bilinmeyen anahtar için `[]` hata verir; `get()` ise `None` döndürür. Önce `get()` ile bakıp sonra karar veririz.


````python
# ✏️ [EX1] — worked solution
calibration = {"T1": 1.02, "T2": 0.98, "P1": 1.10}
requested = ["T1", "P1", "H1"]

for name in requested:
    factor = calibration.get(name)
    if factor is None:
        print(f"{name}: no calibration factor")
    else:
        print(f"{name}: factor {factor}")

assert calibration.get("H1") is None
assert calibration.get("H1", 1.0) == 1.0
assert calibration["T2"] == 0.98
assert len(calibration) == 3
````

### Expected result

```text
T1: factor 1.02
P1: factor 1.1
H1: no calibration factor
```

## Exercise 2: Count Status Codes (Easy)

A device log contains status codes. Count how many times each code appears, in the order the codes first appear. Then print the total number of codes.

### Reasoning and plan

The key is the code, the value is its count. `get(code, 0)` handles the first time a code appears. Because insertion order is kept, the report lists codes in order of first appearance. An empty log gives an empty tally.

**Türkçe:** Sayım sözlüğünde anahtar kod, değer sayıdır. `get(code, 0)` ilk görülen kodu 0'dan başlatır.


````python
# ✏️ [EX2] — worked solution
status_log = ["OK", "OK", "WARN", "OK", "ERR", "WARN", "OK", "ERR", "OK"]

tally = {}
for code in status_log:
    tally[code] = tally.get(code, 0) + 1

for code, count in tally.items():
    print(f"{code}: {count}")
print(f"Total codes: {sum(tally.values())}")

assert tally == {"OK": 5, "WARN": 2, "ERR": 2}
assert list(tally) == ["OK", "WARN", "ERR"]
assert sum(tally.values()) == len(status_log)
empty_tally = {}
for code in []:
    empty_tally[code] = empty_tally.get(code, 0) + 1
assert empty_tally == {}
````

### Expected result

```text
OK: 5
WARN: 2
ERR: 2
Total codes: 9
```

## Exercise 3: Inventory Update (Easy)

Apply the deliveries to the inventory. A delivered part that is not yet in the inventory must be added; print `New part: <name>` when that happens. Then subtract the used quantities. Finally print every part and its quantity. Assume quantities are nonnegative integers, and each used part exists after deliveries with enough stock for the requested usage.

### Reasoning and plan

Deliveries may add a new key, so use `get(part, 0)` and print a message when the key is new. Used parts already exist, so `-=` is safe here. The new part goes to the end because insertion order is kept.

**Türkçe:** Teslimat yeni bir parça ekleyebilir; `get(part, 0)` bunu güvenli yapar. Yeni anahtar sözlüğün sonuna eklenir.


````python
# ✏️ [EX3] — worked solution
inventory = {"M3 bolt": 120, "M3 nut": 80, "washer": 200}
deliveries = {"M3 bolt": 50, "washer": 100, "M4 bolt": 60}
used = {"M3 nut": 30}

for part, qty in deliveries.items():
    if part not in inventory:
        print(f"New part: {part}")
    inventory[part] = inventory.get(part, 0) + qty

for part, qty in used.items():
    inventory[part] -= qty

for part, qty in inventory.items():
    print(f"{part}: {qty}")

assert inventory == {"M3 bolt": 170, "M3 nut": 50, "washer": 300, "M4 bolt": 60}
assert list(inventory)[-1] == "M4 bolt"
assert inventory.get("M5 bolt", 0) == 0
````

### Expected result

washer: 300
M4 bolt: 60

## Exercise 4: Group Readings by Sensor (Medium)

The readings are `(sensor, value)` tuples in time order. Build a dictionary that maps each sensor to the **list** of its values, in the original order. Print each sensor with its list.

### Reasoning and plan

Each tuple is unpacked into a sensor name and a value. The first time a sensor appears, an empty list is created for it. Appending keeps the time order inside each group, and the number of stored values equals the number of readings.

**Türkçe:** Anahtar sensör adı, değer bir listedir. İlk görüşte boş liste açılır, sonra `append` ile eklenir.


````python
# ✏️ [EX4] — worked solution
readings = [("T1", 21.5), ("T2", 22.0), ("T1", 21.9), ("T3", 19.8), ("T2", 22.4), ("T1", 22.3)]

groups = {}
for sensor, value in readings:
    if sensor not in groups:
        groups[sensor] = []
    groups[sensor].append(value)

for sensor, values in groups.items():
    print(f"{sensor}: {values}")

assert groups == {"T1": [21.5, 21.9, 22.3], "T2": [22.0, 22.4], "T3": [19.8]}
assert sum(len(v) for v in groups.values()) == len(readings)
assert "T4" not in groups
````

### Expected result

```text
T1: [21.5, 21.9, 22.3]
T2: [22.0, 22.4]
T3: [19.8]
```

## Exercise 5: Average per Sensor (Medium)

Starting from the grouped dictionary, print the mean of each sensor with two decimals and the number of readings `n`. Then print the sensor with the highest mean, keeping the first sensor in insertion order if means tie. For an empty group print `<sensor>: no readings` and skip it. If no group has data, print `No sensor readings available.`

### Reasoning and plan

Each value in the dictionary is a list, so `sum` and `len` give the mean. The best sensor is tracked with a pair of variables initialised to `None`, exactly like the maximum search from Week 6. A sensor with one reading has a mean equal to that reading.

**Türkçe:** Her değer bir liste olduğu için `sum/len` ortalamayı verir. En yüksek ortalama, Week 6'daki maksimum arama gibi izlenir.


````python
# ✏️ [EX5] — worked solution
groups = {"T1": [21.5, 21.9, 22.3], "T2": [22.0, 22.4], "T3": [19.8]}

warmest = None
warmest_mean = None
for sensor, values in groups.items():
    if not values:
        print(f"{sensor}: no readings")
        continue
    mean = sum(values) / len(values)
    print(f"{sensor}: mean {mean:.2f} (n={len(values)})")
    if warmest is None or mean > warmest_mean:
        warmest = sensor
        warmest_mean = mean

if warmest is None:
    print("No sensor readings available.")
else:
    print(f"Warmest sensor: {warmest}")

assert warmest == "T2"
assert abs(warmest_mean - 22.2) < 0.001
assert abs(sum(groups["T1"]) / len(groups["T1"]) - 21.9) < 0.001
assert len(groups["T3"]) == 1
````

### Expected result

```text
T1: mean 21.90 (n=3)
T2: mean 22.20 (n=2)
T3: mean 19.80 (n=1)
Warmest sensor: T2
```

## Exercise 6: Invert a Pin Map (Medium)

The dictionary maps pin names to GPIO numbers. Assume each GPIO number is assigned to exactly one name; otherwise inversion would overwrite an earlier name. Build the **reverse** dictionary that maps GPIO numbers to names. Print it, then look up GPIO 9 and GPIO 5 (which is not assigned).

### Reasoning and plan

Inverting swaps the role of key and value. This only works because every GPIO number appears once; if two names shared a pin, the second would overwrite the first. Integer keys are fine in a dictionary.

**Türkçe:** Ters sözlükte anahtar ile değer yer değiştirir. Bu ancak değerler tekrarsızsa güvenlidir.


````python
# ✏️ [EX6] — worked solution
pins = {"LED": 13, "BUTTON": 2, "BUZZER": 9, "RELAY": 7}

reverse = {}
for name, gpio in pins.items():
    reverse[gpio] = name
print(reverse)

for gpio in [9, 5]:
    name = reverse.get(gpio)
    if name is None:
        print(f"GPIO {gpio} is not assigned")
    else:
        print(f"GPIO {gpio} is {name}")

assert reverse == {13: "LED", 2: "BUTTON", 9: "BUZZER", 7: "RELAY"}
assert reverse[pins["LED"]] == "LED"
assert reverse.get(5) is None
assert len(reverse) == len(pins)
````

### Expected result

```text

GPIO 9 is BUZZER
GPIO 5 is not assigned
```

## Exercise 7: Tally Words in a Log Line (Medium)

Split the log line into words with `.split()`. Count each whitespace-separated token exactly as written: case and punctuation remain significant. Print only the words that appear **more than once**, in order of first appearance, then the number of distinct words.

### Reasoning and plan

`split()` turns the line into a list of words. The tally pattern counts them. A second loop filters the repeated words; because the dictionary keeps insertion order, the output follows first appearance. The number of keys is the number of distinct words.

**Türkçe:** `split()` kelimeleri ayırır, sözlük sayar. `len(tally)` farklı kelime sayısıdır.


````python
# ✏️ [EX7] — worked solution
line = "ERROR sensor T2 timeout ERROR sensor T4 timeout WARN sensor T2 slow"

tally = {}
for word in line.split():
    tally[word] = tally.get(word, 0) + 1

for word, count in tally.items():
    if count > 1:
        print(f"{word}: {count}")
print(f"Distinct words: {len(tally)}")

assert tally["sensor"] == 3
assert tally["slow"] == 1
assert len(tally) == 7
assert sum(tally.values()) == len(line.split())
````

### Expected result

```text
ERROR: 2
sensor: 3
T2: 2
timeout: 2
Distinct words: 7
```

## Exercise 8: Merge Two Config Dictionaries (Medium)

`defaults` holds the standard settings. `user` holds settings the user changed. Build a merged dictionary that starts from the defaults and lets user values override them. Print every setting with its value and whether it came from `user` or `default`. Do **not** change `defaults`. `copy()` is shallow: this example replaces top-level values; mutating a shared nested list or dictionary would also affect the original.

### Reasoning and plan

`copy()` creates a separate dictionary, so writing into `merged` leaves `defaults` unchanged (the aliasing lesson from Week 6 applies to dictionaries too). Assigning a user value to an existing key overrides it but keeps its position. The source label is a simple membership test.

**Türkçe:** `copy()` ayrı bir sözlük oluşturur; `defaults` değişmez. Kullanıcı değeri aynı anahtarın üzerine yazar.


````python
# ✏️ [EX8] — worked solution
defaults = {"rate_hz": 10, "unit": "C", "threshold": 60.0, "log": False}
user = {"rate_hz": 50, "threshold": 55.5}

merged = defaults.copy()
for key, value in user.items():
    merged[key] = value

for key, value in merged.items():
    source = "user" if key in user else "default"
    print(f"{key}: {value} ({source})")

assert merged == {"rate_hz": 50, "unit": "C", "threshold": 55.5, "log": False}
assert defaults["rate_hz"] == 10
assert list(merged) == list(defaults)
````

### Expected result

```text

unit: C (default)
threshold: 55.5 (user)
log: False (default)
```

## Exercise 9: Nested Part Records Report (Medium)

Each part ID maps to a record with `name`, `qty` and `unit_cost`. Print one line per part with the stock value (`qty × unit_cost`), then the total stock value. Finally list the IDs with fewer than 5 pieces in stock.

### Reasoning and plan

The outer loop visits each ID and its inner record. Fields are read by name, which is clearer than a position in a list. The total is an accumulator; low-stock IDs are collected in a list because there may be zero, one or many.

**Türkçe:** Dış sözlük parça kimliklerini, iç sözlük alanları tutar. Alanlar adla okunur; toplam bir biriktiricidir.


````python
# ✏️ [EX9] — worked solution
parts = {
    "P-100": {"name": "bearing 608", "qty": 12, "unit_cost": 4.50},
    "P-205": {"name": "M3 bolt", "qty": 300, "unit_cost": 0.05},
    "P-310": {"name": "stepper motor", "qty": 2, "unit_cost": 18.90},
}

total = 0.0
low_stock = []
for pid, rec in parts.items():
    value = rec["qty"] * rec["unit_cost"]
    total += value
    if rec["qty"] < 5:
        low_stock.append(pid)
    print(f"{pid}  {rec['name']:<14} {rec['qty']:>4} x {rec['unit_cost']:>5.2f} = {value:>6.2f}")

print(f"Total stock value: {total:.2f}")
print(f"Low stock (< 5): {low_stock}")

assert abs(total - 106.80) < 0.001
assert low_stock == ["P-310"]
assert parts["P-205"]["qty"] == 300
assert "P-999" not in parts
````

### Expected result

P-205  M3 bolt         300 x  0.05 =  15.00

## Exercise 10: Peak Time from (t, v) Pairs (Medium)

`samples` is a non-empty list, ordered by increasing time, of `(time_s, temperature_C)` tuples. Find the highest temperature and the **first** time it occurred, using tuple unpacking. Then count how many samples are at or above 24.0 C.

### Reasoning and plan

Each sample is a fixed `(time, value)` record, so unpacking gives readable names. Using a strict `>` keeps the first of two equal peaks. `peak_t, peak_v = t, v` assigns both names in one step. The count uses `>=` as the task states.

**Türkçe:** Her örnek `(zaman, değer)` demetidir; `t, v` ile açılır. `>` kullanmak ilk tepe noktasını korur.


````python
# ✏️ [EX10] — worked solution
samples = [(0, 20.1), (10, 22.4), (20, 25.9), (30, 24.0), (40, 25.9), (50, 21.2)]

peak_t, peak_v = samples[0]
warm = 0
for t, v in samples:
    if v > peak_v:
        peak_t, peak_v = t, v
    if v >= 24.0:
        warm += 1

print(f"Peak {peak_v} C at t = {peak_t} s")
print(f"Samples at or above 24.0 C: {warm}")

assert (peak_t, peak_v) == (20, 25.9)
assert warm == 3
assert type(samples[0]) is tuple
single_t, single_v = [(5, 19.0)][0]
assert (single_t, single_v) == (5, 19.0)
````

### Expected result

```text
Peak 25.9 C at t = 20 s
Samples at or above 24.0 C: 3
```

## Exercise 11: Sparse Grid with (row, col) Keys (Medium)

An inspection camera reports defects only at some cells of a 4-row × 3-column grid. Print the grid, one row per line: `.` for an empty cell and the **first letter, uppercase** of the defect kind. Then count each defect kind. Assume all coordinates are inside the grid and every defect name is non-empty.

### Reasoning and plan

The dictionary stores only the cells that have a defect, so most cells are absent and the grid loops test membership with a `(r, c)` tuple. The tuple key works because its row and column are integers and therefore hashable; not every tuple is a valid key. Counting the values is the same tally pattern as before.

**Türkçe:** Demet `(satır, sütun)` sözlük anahtarı olabilir. Boş hücreler sözlükte yer almaz; `in` ile kontrol edilir.


````python
# ✏️ [EX11] — worked solution
defects = {(0, 2): "scratch", (1, 1): "dent", (3, 0): "scratch", (2, 2): "crack"}
rows, cols = 4, 3

for r in range(rows):
    row_text = ""
    for c in range(cols):
        if (r, c) in defects:
            row_text += defects[(r, c)][0].upper()
        else:
            row_text += "."
    print(row_text)

kinds = {}
for kind in defects.values():
    kinds[kind] = kinds.get(kind, 0) + 1
for kind, count in kinds.items():
    print(f"{kind}: {count}")

assert kinds == {"scratch": 2, "dent": 1, "crack": 1}
assert (0, 0) not in defects
assert defects.get((3, 0)) == "scratch"
assert len(defects) == 4
````

### Expected result

```text

.D.
..C
S..
scratch: 2
dent: 1
crack: 1
```

## Exercise 12: Lookup Table from Parallel Lists (Challenge)

`ids` and `limits` are two parallel lists: `limits[i]` is the maximum allowed value for `ids[i]`. Assume `ids` and `limits` have equal lengths and each table ID is unique. Build a lookup dictionary from them. Then check every `(id, value)` measurement: print `within limit` or `OVER limit` with the limit, or `no limit in table` when the ID is unknown. Finally print the list of distinct measured IDs missing from the table, in first-seen order, and the list of table IDs that were not measured.

### Reasoning and plan

Parallel lists are linked only by position, so a dictionary built with `range(len(ids))` turns the pair of lists into a lookup by name. The measurements are checked with `in`. Two lists collect the mismatches in each direction: IDs measured but unknown, and IDs known but not measured. A value equal to its limit counts as within limit.

**Türkçe:** Paralel listeler yalnızca konumla bağlıdır; sözlük onları ada göre birleştirir. Eksikler iki yönde ayrı listelenir.


````python
# ✏️ [EX12] — worked solution
ids = ["T1", "T2", "T3", "P1"]
limits = [60.0, 60.0, 75.0, 250.0]
measured = [("T1", 58.2), ("T3", 77.9), ("H1", 41.0), ("P1", 249.5), ("T4", 63.0)]

limit_of = {}
for i in range(len(ids)):
    limit_of[ids[i]] = limits[i]

missing = []
seen = []
for sensor, value in measured:
    seen.append(sensor)
    if sensor in limit_of:
        limit = limit_of[sensor]
        status = "within" if value <= limit else "OVER"
        print(f"{sensor}: {value} {status} limit {limit}")
    else:
        print(f"{sensor}: no limit in table")
        if sensor not in missing:
            missing.append(sensor)

not_measured = []
for sensor in limit_of:
    if sensor not in seen:
        not_measured.append(sensor)

print(f"Missing from table: {missing}")
print(f"Not measured: {not_measured}")

assert limit_of == {"T1": 60.0, "T2": 60.0, "T3": 75.0, "P1": 250.0}
assert missing == ["H1", "T4"]
assert not_measured == ["T2"]
assert len(limit_of) == len(ids)
assert 250.0 <= limit_of["P1"]
````

### Expected result

```text
T1: 58.2 within limit 60.0
T3: 77.9 OVER limit 75.0
H1: no limit in table
P1: 249.5 within limit 250.0
T4: no limit in table
Missing from table: ['H1', 'T4']
Not measured: ['T2']
```

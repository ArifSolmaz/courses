## Exercise 1: Create and Print (Easy)

Create a list called `fruits` containing 5 fruit names. Print the list, then print each fruit on a separate line with its index.

### Reasoning and plan

A list stores the five fruit names in order. `enumerate` visits each item together with its zero-based index. The last index is 4 even though the length is 5.

**Türkçe:** Liste sıralıdır. Beş elemanın indeksleri 0’dan 4’e gider; enumerate indeks ile değeri birlikte verir.


````python
# ✏️ [EX1] — worked solution
fruits = ['apple', 'banana', 'cherry', 'mango', 'kiwi']
print('Fruits:', fruits)
for index, fruit in enumerate(fruits):
    print(f'{index}: {fruit}')
assert len(fruits) == 5 and fruits[0] == 'apple' and fruits[-1] == 'kiwi'
assert list(enumerate(['apple'])) == [(0, 'apple')]
````

### Expected result

```text
Fruits: ['apple', 'banana', 'cherry', 'mango', 'kiwi']
0: apple
1: banana
2: cherry
3: mango
4: kiwi
```

## Exercise 2: Sum a List (Easy)

Given the list `numbers = [15, 23, 8, 42, 16, 4, 31]`, calculate the sum of all elements **without using the built-in `sum()` function**. Print the result.

### Reasoning and plan

Initialize the accumulator once, outside the loop. After processing 15, 23, and 8, it holds 15, 38, and 46. Adding rather than replacing is what keeps earlier values.

**Türkçe:** Toplam döngüden önce sıfırlanır. Her adımda eski toplamın üzerine yeni sayı eklenir.


````python
# ✏️ [EX2] — worked solution
numbers = [15, 23, 8, 42, 16, 4, 31]
total = 0
for number in numbers:
    total += number
print(f'Sum: {total}')
assert total == 139
for values, expected in [([], 0), ([-3, 2], -1)]:
    actual = 0
    for value in values:
        actual += value
    assert actual == expected
````

### Expected result

```text
Sum: 139
```

## Exercise 3: Find Maximum (Medium)

Given the list `values = [34, 67, 23, 89, 12, 56, 78, 45]`, find the maximum value **without using the built-in `max()` function**. Also print the index where the maximum was found.

### Reasoning and plan

Keep the candidate maximum and its index together. Start at index zero; replace both only when a strictly larger value appears. This selects the first maximum when equal values repeat.

**Türkçe:** En büyük değerle birlikte konumunu da saklarız. Eşit değerde güncelleme yapmayınca ilk konum korunur.


````python
# ✏️ [EX3] — worked solution
values = [34, 67, 23, 89, 12, 56, 78, 45]
maximum = values[0]
max_index = 0
for index in range(1, len(values)):
    if values[index] > maximum:
        maximum = values[index]
        max_index = index
print(f'Maximum value: {maximum}')
print(f'Found at index: {max_index}')
assert (maximum, max_index) == (89, 3)
test_values = [-2, -2, -5]
best = 0
for index in range(1, len(test_values)):
    if test_values[index] > test_values[best]:
        best = index
assert best == 0
````

### Expected result

```text
Maximum value: 89
Found at index: 3
```

## Exercise 4: Reverse a List (Medium)

Given `original = [10, 20, 30, 40, 50]`, create a new list that is the reverse of `original` **without using `reverse()`, `reversed()`, or slicing `[::-1]`**. Print both lists.

### Reasoning and plan

Visit indices from length−1 down to zero, appending each value into a fresh list. The stop argument −1 is excluded, so index zero is visited. The original list is never assigned to or changed.

**Türkçe:** Son indeksten sıfıra doğru gideriz. Yeni listeye eklemek orijinal listenin sırasını değiştirmez.


````python
# ✏️ [EX4] — worked solution
original = [10, 20, 30, 40, 50]
backwards = []
for index in range(len(original) - 1, -1, -1):
    backwards.append(original[index])
print('Original:', original)
print('Reversed:', backwards)
assert backwards == [50, 40, 30, 20, 10]
assert original == [10, 20, 30, 40, 50]
empty_result = []
for index in range(len([]) - 1, -1, -1):
    empty_result.append(index)
assert empty_result == []
````

### Expected result

```text
Original: [10, 20, 30, 40, 50]
Reversed: [50, 40, 30, 20, 10]
```

## Exercise 5: Count Occurrences (Medium)

Given the list `data = [5, 3, 8, 3, 2, 5, 3, 9, 1, 3, 5]`, count how many times the value `3` appears **without using the `count()` method**.

### Reasoning and plan

Compare every element with the target 3 and increment only on equality. Repeated target values must each be counted; unrelated repeated values do not matter.

**Türkçe:** Her eleman hedefle karşılaştırılır. Hedefin her tekrarı sayacı bir artırır.


````python
# ✏️ [EX5] — worked solution
data = [5, 3, 8, 3, 2, 5, 3, 9, 1, 3, 5]
target = 3
count = 0
for value in data:
    if value == target:
        count += 1
print(f'The value {target} appears {count} times.')
assert count == 4
for sample, expected in [([3, 3], 2), ([1, 2], 0), ([], 0)]:
    actual = 0
    for value in sample:
        if value == 3:
            actual += 1
    assert actual == expected
````

### Expected result

```text
The value 3 appears 4 times.
```

## Exercise 6: Remove Duplicates (Medium)

Given the list `items = [4, 2, 7, 2, 4, 9, 7, 1, 4, 2]`, create a new list containing only the unique values (in their original order of first appearance).

### Reasoning and plan

The result doubles as memory of what has already appeared. Append a value only when it is not in that result. Trace 4,2,7,2: `[4] → [4,2] → [4,2,7] → [4,2,7]`. Sorting would lose the required first-appearance order.

**Türkçe:** Sonuç listesi daha önce gördüklerimizi tutar. Sadece yeni değerleri ekleyerek ilk görülme sırasını koruruz.


````python
# ✏️ [EX6] — worked solution
items = [4, 2, 7, 2, 4, 9, 7, 1, 4, 2]
unique = []
for item in items:
    if item not in unique:
        unique.append(item)
print('Original:', items)
print('Unique:  ', unique)
assert unique == [4, 2, 7, 9, 1]
assert items == [4, 2, 7, 2, 4, 9, 7, 1, 4, 2]
test_result = []
for item in [0, 0, -1, -1]:
    if item not in test_result:
        test_result.append(item)
assert test_result == [0, -1]
````

### Expected result

```text
Original: [4, 2, 7, 2, 4, 9, 7, 1, 4, 2]
Unique:   [4, 2, 7, 9, 1]
```

## Exercise 7: Even Numbers Only (Medium)

Given `numbers = [13, 24, 7, 36, 9, 42, 15, 28, 3, 50]`, create a new list that contains only the even numbers. Print the new list and how many even numbers were found.

### Reasoning and plan

Build a fresh result list. The remainder test `number % 2 == 0` works for positive, negative and zero integers. Appending preserves the relative order of matching inputs.

**Türkçe:** Çiftlik koşulu negatif sayılar ve sıfır için de çalışır. Yeni listeye sırayla eklemek giriş sırasını korur.


````python
# ✏️ [EX7] — worked solution
numbers = [13, 24, 7, 36, 9, 42, 15, 28, 3, 50]
evens = []
for number in numbers:
    if number % 2 == 0:
        evens.append(number)
print('Even numbers:', evens)
print('Count:', len(evens))
assert evens == [24, 36, 42, 28, 50]
assert [n for n in [-2, -1, 0, 1, 2] if n % 2 == 0] == [-2, 0, 2]
````

### Expected result

```text
Even numbers: [24, 36, 42, 28, 50]
Count: 5
```

## Exercise 8: List Statistics (Medium)

Given `scores = [72, 85, 90, 68, 95, 78, 82, 88, 76, 93]`, calculate and print the minimum, maximum, sum, and average **without using** `min()`, `max()`, or `sum()` built-in functions.

### Reasoning and plan

One scan can update three independent pieces of state: minimum, maximum and total. Initialize both extremes from the first score; initialize total to zero because the loop includes every score. Divide by the number of scores after scanning.

**Türkçe:** Aynı döngü en küçük, en büyük ve toplamı güncelleyebilir. Ortalama için döngünün bitmesini bekleriz.


````python
# ✏️ [EX8] — worked solution
scores = [72, 85, 90, 68, 95, 78, 82, 88, 76, 93]
minimum = maximum = scores[0]
total = 0
for score in scores:
    total += score
    if score < minimum:
        minimum = score
    if score > maximum:
        maximum = score
average = total / len(scores)
print('Scores:', scores)
print('Minimum:', minimum)
print('Maximum:', maximum)
print('Sum:    ', total)
print(f'Average: {average:.2f}')
assert (minimum, maximum, total, average) == (68, 95, 827, 82.7)
assert minimum <= average <= maximum
````

### Expected result

```text
Scores: [72, 85, 90, 68, 95, 78, 82, 88, 76, 93]
Minimum: 68
Maximum: 95
Sum:     827
Average: 82.70
```

## Exercise 9: Merge Two Lists (Medium)

Given two lists `list_a = [5, 12, 3, 8]` and `list_b = [9, 1, 7, 15]`, merge them into a single list and sort it in ascending order. Print the result.

### Reasoning and plan

The `+` operation creates a combined list. `sorted` returns another list in ascending order while leaving both original lists unchanged. Duplicates should stay: merging is not deduplication.

**Türkçe:** Birleştirme tekrarları silmez. sorted yeni bir liste döndürür; giriş listelerini değiştirmez.


````python
# ✏️ [EX9] — worked solution
list_a = [5, 12, 3, 8]
list_b = [9, 1, 7, 15]
merged = sorted(list_a + list_b)
print('List A:', list_a)
print('List B:', list_b)
print('Merged and sorted:', merged)
assert merged == [1, 3, 5, 7, 8, 9, 12, 15]
assert list_a == [5, 12, 3, 8] and list_b == [9, 1, 7, 15]
assert sorted([2] + [2, 1]) == [1, 2, 2]
assert sorted([] + []) == []
````

### Expected result

```text
List A: [5, 12, 3, 8]
List B: [9, 1, 7, 15]
Merged and sorted: [1, 3, 5, 7, 8, 9, 12, 15]
```

## Exercise 10: Swap First and Last (Easy)

Given `colors = ["red", "green", "blue", "yellow", "purple"]`, swap the first and last elements. Print the list before and after.

### Reasoning and plan

Save the original first value before replacing it. Then put the original last value first and the saved value last. An empty list has no endpoints; a single-item list is unchanged.

**Türkçe:** İlk değeri geçici değişkende koruruz. Böylece üzerine yazınca bilgi kaybolmaz. Boş listede işlem yapmayız.


````python
# ✏️ [EX10] — worked solution
colors = ['red', 'green', 'blue', 'yellow', 'purple']
print('Before:', colors)
first = colors[0]
colors[0] = colors[-1]
colors[-1] = first
print('After: ', colors)
assert colors == ['purple', 'green', 'blue', 'yellow', 'red']
for sample, expected in [([], []), (['red'], ['red']), (['a', 'b'], ['b', 'a'])]:
    if len(sample) > 0:
        first = sample[0]
        sample[0] = sample[-1]
        sample[-1] = first
    assert sample == expected
````

### Expected result

```text
Before: ['red', 'green', 'blue', 'yellow', 'purple']
After:  ['purple', 'green', 'blue', 'yellow', 'red']
```

## Exercise 11: Rotate List (Medium)

Given `data = [1, 2, 3, 4, 5, 6, 7]` and `n = 3`, rotate the list to the left by `n` positions. This means the first `n` elements move to the end.

### Reasoning and plan

Split at position N. The suffix becomes the front and the prefix moves to the end. For a nonempty list, reduce large shifts modulo the length: rotating by length gives the original list. Handle empty data before taking a remainder.

**Türkçe:** Listeyi N konumunda ikiye ayırıp parçaların yerini değiştiririz. Uzunluk kadar kaydırmak başlangıç durumuna döner.


````python
# ✏️ [EX11] — worked solution
data = [1, 2, 3, 4, 5, 6, 7]
n = 3
shift = n % len(data)
rotated = data[shift:] + data[:shift]
print('Original:', data)
print(f'Rotated left by {n}:', rotated)
assert rotated == [4, 5, 6, 7, 1, 2, 3]
for sample, amount, expected in [([1, 2, 3], 0, [1, 2, 3]), ([1, 2, 3], 4, [2, 3, 1]), ([], 3, [])]:
    if sample:
        offset = amount % len(sample)
        result = sample[offset:] + sample[:offset]
    else:
        result = []
    assert result == expected
````

### Expected result

```text
Original: [1, 2, 3, 4, 5, 6, 7]
Rotated left by 3: [4, 5, 6, 7, 1, 2, 3]
```

The assertions also check the cases named in the code.


## Exercise 12: Grade Summary (Challenge)

Given a list of student grades, create a complete grade summary report:

```python
grades = [85, 92, 45, 78, 95, 67, 88, 72, 55, 90, 38, 83, 76, 91, 60]
```

Calculate and print:
1. Total number of students
2. Class average (formatted to 1 decimal place)
3. Highest and lowest grade
4. Number of students passing (grade >= 60)
5. Number of students failing (grade < 60)
6. Grade distribution: count of A (90-100), B (80-89), C (70-79), D (60-69), F (<60)

### Reasoning and plan

Use descending thresholds so each grade enters exactly one band. Count passing with grade ≥60, including exactly 60. The five grade-band counts must total 15, and passing+failing must also total 15. Average uses all scores, including failures.

**Türkçe:** Eşikleri büyükten küçüğe kontrol ederek her notu tek sınıfa yerleştiririz. 60 geçer; başarısız notlar da ortalamaya dahildir.


````python
# ✏️ [EX12] — worked solution
grades = [85, 92, 45, 78, 95, 67, 88, 72, 55, 90, 38, 83, 76, 91, 60]
count_a = count_b = count_c = count_d = count_f = 0
passing = 0
for grade in grades:
    if grade >= 60:
        passing += 1
    if grade >= 90:
        count_a += 1
    elif grade >= 80:
        count_b += 1
    elif grade >= 70:
        count_c += 1
    elif grade >= 60:
        count_d += 1
    else:
        count_f += 1
failing = len(grades) - passing
print('=== Grade Summary Report ===')
print('Total students:', len(grades))
print(f'Class average: {sum(grades) / len(grades):.1f}')
print('Highest grade:', max(grades))
print('Lowest grade:', min(grades))
print('Passing:', passing)
print('Failing:', failing)
for label, count in [('A (90-100)', count_a), ('B (80-89)', count_b), ('C (70-79)', count_c), ('D (60-69)', count_d), ('F (<60)', count_f)]:
    print(f'{label}: {count}')
assert (passing, failing) == (12, 3)
assert [count_a, count_b, count_c, count_d, count_f] == [4, 3, 3, 2, 3]
assert count_a + count_b + count_c + count_d + count_f == len(grades)
assert format(sum(grades) / len(grades), '.1f') == '74.3'
````

### Expected result

```text
=== Grade Summary Report ===
Total students: 15
Class average: 74.3
Highest grade: 95
Lowest grade: 38
Passing: 12
Failing: 3
A (90-100): 4
B (80-89): 3
C (70-79): 3
D (60-69): 2
F (<60): 3
```

## Bridge Exercise: Preview of 2D Lists

Next week we will learn about **2D lists** (lists of lists). As a preview, create a list where each element is itself a list containing a student name and their grade. Then print each student's info.

**Example:**
```python
students = [["Elif", 85], ["Mert", 92], ["Zeynep", 78]]
```

### Reasoning and plan

Each outer-list item is one student record. In a record, position 0 is the name and position 1 is the grade. A loop over the outer list therefore processes one complete student at a time.

**Türkçe:** Dış listedeki her eleman bir öğrencidir. İç listede 0 adı, 1 notu gösterir.


````python
# ✏️ [EXBridge] — worked solution
students = [['Elif', 85], ['Mert', 92], ['Zeynep', 78]]
for student in students:
    name = student[0]
    grade = student[1]
    print(f'{name:<6}: {grade}')
assert students[1][0] == 'Mert' and students[1][1] == 92
assert len(students) == 3
````

### Expected result

```text
Elif  : 85
Mert  : 92
Zeynep: 78
```

The assertions also check the cases named in the code.

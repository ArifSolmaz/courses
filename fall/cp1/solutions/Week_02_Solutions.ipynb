## Exercise 1: Operator Predictions (Easy)

### Task and supplied example


**Without running the code first**, predict the result of each expression. Then write `print()` statements to verify.

```
a) 15 // 4
b) 15 % 4
c) 2 ** 5
d) 10 / 3
e) -17 // 5
f) -17 % 5
g) 3 + 4 * 2 ** 2
h) (3 + 4) * 2 ** 2
```

**Expected output:**
```
a) 15 // 4 = 3
b) 15 % 4 = 3
c) 2 ** 5 = 32
d) 10 / 3 = 3.3333333333333335
e) -17 // 5 = -4
f) -17 % 5 = 3
g) 3 + 4 * 2 ** 2 = 19
h) (3 + 4) * 2 ** 2 = 28
```

<details>
<summary>💡 Hint</summary>
Remember: <code>**</code> has higher priority than <code>*</code>, which has higher priority than <code>+</code>. Floor division rounds toward negative infinity.
</details>

### Reasoning

Work from precedence first. A negative floor quotient rounds toward negative infinity: -17//5 is -4, and the remainder is 3 because -17 = (-4)*5 + 3. Exponentiation happens before multiplication and addition.

**TR:** Negatif bölüm aşağıya yuvarlanır; bölüm ve kalan eşitliğini kontrol et.

````python
print("a) 15 // 4 =", 15 // 4)
print("b) 15 % 4 =", 15 % 4)
print("c) 2 ** 5 =", 2 ** 5)
print("d) 10 / 3 =", 10 / 3)
print("e) -17 // 5 =", -17 // 5)
print("f) -17 % 5 =", -17 % 5)
print("g) 3 + 4 * 2 ** 2 =", 3 + 4 * 2 ** 2)
print("h) (3 + 4) * 2 ** 2 =", (3 + 4) * 2 ** 2)
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert -17 // 5 == -4 and -17 % 5 == 3
assert (-17 // 5) * 5 + (-17 % 5) == -17
assert (3 + 4 * 2 ** 2, (3 + 4) * 2 ** 2) == (19, 28)
````

## Exercise 3: Seconds Breakdown (Medium)

### Task and supplied example


Convert `total_seconds = 86399` into hours, minutes, and seconds using `//` and `%`.

**Expected output:**
```
Total seconds: 86399
Hours:   23
Minutes: 59
Seconds: 59
Time: 23h 59m 59s
```

<details>
<summary>💡 Hint</summary>
1. <code>hours = total_seconds // 3600</code><br>
2. <code>remaining = total_seconds % 3600</code><br>
3. <code>minutes = remaining // 60</code><br>
4. <code>seconds = remaining % 60</code>
</details>

### Reasoning

Remove complete 3600-second hours first. From the remaining seconds, count complete 60-second minutes; the final remainder is seconds. The reconstructed time must equal the input.

**TR:** Önce tam saatleri çıkar; kalan saniyeyi dakika ve saniyeye ayır.

````python
total_seconds = 86399
hours = total_seconds // 3600
remaining = total_seconds % 3600
minutes = remaining // 60
seconds = remaining % 60
print(f"Total seconds: {total_seconds}")
print(f"Hours:   {hours}")
print(f"Minutes: {minutes}")
print(f"Seconds: {seconds}")
print(f"Time: {hours}h {minutes}m {seconds}s")
````

````python
assert (hours, minutes, seconds) == (23, 59, 59)
assert hours * 3600 + minutes * 60 + seconds == total_seconds
assert 0 <= minutes < 60 and 0 <= seconds < 60
````

## Exercise 5: Compound Interest with Input (Medium)

### Task and supplied example


Ask the user for:
- Principal amount (TL)
- Annual interest rate (as percentage, e.g., 15)
- Number of years

Calculate and display the final amount using: A = P × (1 + r)^t

**Example interaction:**
```
Enter principal amount (TL): 10000
Enter annual interest rate (%): 15
Enter number of years: 5

==== Investment Summary ====
Principal:    10,000.00 TL
Rate:         15.0%
Duration:     5 years
Final Amount: 20,113.57 TL
Profit:       10,113.57 TL
============================
```

<details>
<summary>💡 Hint</summary>
Convert rate from percentage to decimal: <code>rate = rate_percent / 100</code>. Use <code>float(input(...))</code> to get numeric input.
</details>

### Reasoning

The fixed strings below represent what input() would return. Convert the principal and percent to numbers, divide the percent by 100, and convert years to int. Profit is final amount minus principal.

**TR:** Girdi yüzdesini 100 ile böl; kârı ana parayı çıkararak bul.

````python
principal_text = "10000"
rate_text = "15"
years_text = "5"
principal = float(principal_text)
rate = float(rate_text) / 100
years = int(years_text)
amount = principal * (1 + rate) ** years
profit = amount - principal
print("==== Investment Summary ====")
print(f"Principal:    {principal:,.2f} TL")
print(f"Rate:         {rate:.1%}")
print(f"Duration:     {years} years")
print(f"Final Amount: {amount:,.2f} TL")
print(f"Profit:       {profit:,.2f} TL")
print("============================")
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert round(amount, 2) == 20113.57
assert round(profit, 2) == 10113.57
assert principal * (1 + 0) ** years == principal
````

Voltage:    220.00 V

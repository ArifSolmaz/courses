### Task and supplied example

- Principal amount (TL)
- Annual interest rate (as percentage, e.g., 15)
- Number of years


### Reasoning

The first two digits form a two-digit integer, so `band1 * 10 + band2`. The third band multiplies by a power of ten, so `10 ** band3` keeps everything as integers.

## Exercise 1: Countdown (Easy)

### Task and supplied example


Write a program that asks the user for a positive integer `N` and counts down from `N` to 1 using a `while` loop, then prints "Go!".

**Expected output (for N=5):**
```
5 4 3 2 1
Go!
```

<details><summary>💡 Hints</summary>

- Start with the variable equal to N
- Use `while variable >= 1`
- Subtract 1 each iteration
- Use `end=" "` in print to stay on the same line

</details>

### Reasoning

Initialize the counter to the positive input. Print it, subtract one, and repeat only while it is positive. The update guarantees that the example reaches zero and stops.

**TR:** Her turda sayacı bir azalt; sıfırda döngüden çık.

````python
N = 5
counter = N
output = ""
while counter > 0:
    output += str(counter) + " "
    counter -= 1
print(output.strip())
print("Go!")
````

- **Line 4 —** `` while counter > 0: ``

  Repeat while the counter is positive; zero ends the loop.

- **Line 5 —** `` output += str(counter) + " " ``

  Append the current number and a space to the accumulated text.

- **Line 6 —** `` counter -= 1 ``

  Decrease the counter by one so the loop progresses toward zero.

- **Line 7 —** `` print(output.strip()) ``

  Remove the trailing space and print the completed countdown once.

- **Line 8 —** `` print("Go!") ``

  Print Go! after the countdown loop has finished.

## Exercise 2: Sum Until Zero (Easy)

### Task and supplied example


Write a program that keeps asking the user for numbers and adds them to a running total. When the user enters `0`, stop and print the total.

**Expected output:**
```
Enter a number (0 to stop): 10
Enter a number (0 to stop): 20
Enter a number (0 to stop): 5
Enter a number (0 to stop): 0
Total: 35
```

<details><summary>💡 Hints</summary>

- Initialize `total = 0`
- Read the first number before the loop
- Loop `while number != 0`
- Add to total and read next number inside the loop

</details>

### Reasoning

A sentinel is a stopping signal, not a data item. Read each scripted value, test for zero before adding, then stop. The finite list replaces input() solely to make this solution reproducible. Values after zero must not be consumed.

**TR:** Sıfır veri değildir; toplama eklemeden önce durma işareti olup olmadığını kontrol et.

````python
scripted_numbers = [10, 20, 5, 0, 999]
index = 0
total = 0
while index < len(scripted_numbers):
    number = scripted_numbers[index]
    index += 1
    print(f"Enter a number (0 to stop): {number}")
    if number == 0:
        break
    total += number
print(f"Total: {total}")
````

  Supply scripted entries, including a zero sentinel and an extra 999 that must remain unused.

- **Line 4 —** `` while index < len(scripted_numbers): ``

  Keep reading only while the index is inside the list.

- **Line 5 —** `` number = scripted_numbers[index] ``

  Read the entry at the current position.

- **Line 6 —** `` index += 1 ``

  Advance the index so the next pass reads a new entry.

- **Line 7 —** `` print(f"Enter a number (0 to stop): {number}") ``

  Show the supplied entry as a simulated prompt response; this line does not call input().

- **Line 8 —** `` if number == 0: ``

  Check whether the entry is the stopping sentinel, zero.

- **Line 9 —** `` break ``

  Leave the loop immediately, before adding the sentinel or reading 999.

- **Line 10 —** `` total += number ``

  Add a nonzero entry to the accumulated total.

- **Line 11 —** `` print(f"Total: {total}") ``

  Print 35 after the loop has ended.



**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert total == 35
assert index == 4
empty_total = 0
for number in [0, 99]:
    if number == 0:
        break
    empty_total += number
assert empty_total == 0
````

## Exercise 3: Charge a Capacitor to 95 % (Medium)

### Task and supplied example


A capacitor charges towards a supply of `V = 12.0` volts. In our simple model, the voltage `v` starts at 0 and each time step it rises by 20 % of the remaining gap: `v = v + (V - v) * 0.2`.

Use a `while` loop to repeat the step until `v` reaches at least 95 % of `V`. Print the voltage after every step with 2 decimals, then report how many steps were needed.

**Expected output:**
```
Step 1: 2.40 V
Step 2: 4.32 V
Step 3: 5.86 V
Step 4: 7.08 V
Step 5: 8.07 V
Step 6: 8.85 V
Step 7: 9.48 V
Step 8: 9.99 V
Step 9: 10.39 V
Step 10: 10.71 V
Step 11: 10.97 V
Step 12: 11.18 V
Step 13: 11.34 V
Step 14: 11.47 V
Charged to 95 % after 14 steps.
```

<details><summary>💡 Hints</summary>

- Compute the target once: `target = 0.95 * V`
- Loop `while v < target:` and update `v` and `steps` inside
- The gap `V - v` shrinks every step, so the loop makes progress and must end

</details>

### Reasoning

The number of steps is not known in advance, so a `while` loop with the condition `v < target` is the right tool. Each step closes 20 % of the remaining gap, so the gap shrinks by a factor 0.8 every turn; the loop is guaranteed to stop. Count the step in the same turn that updates the voltage. This is a teaching model, not an RC circuit equation.

**TR:** Adım sayısı önceden bilinmez; koşullu `while` kullan. Kalan fark her adımda 0.8 ile çarpılır, bu yüzden döngü kesinlikle biter.

````python
V = 12.0
v = 0.0
target = 0.95 * V
steps = 0
while v < target:
    v = v + (V - v) * 0.2
    steps += 1
    print(f"Step {steps}: {v:.2f} V")
print(f"Charged to 95 % after {steps} steps.")
````

**Line-by-line walkthrough**

- **Line 1 —** `` V = 12.0 ``

  Set the supply voltage to 12 volts.

- **Line 2 —** `` v = 0.0 ``

  Start the simulated capacitor voltage at zero.

- **Line 3 —** `` target = 0.95 * V ``

  Calculate the 95% target: 11.4 volts.

- **Line 4 —** `` steps = 0 ``

  Start the update count at zero.

- **Line 5 —** `` while v < target: ``

  Continue while the simulated voltage is below the target.

- **Line 6 —** `` v = v + (V - v) * 0.2 ``

  Add 20% of the remaining voltage gap; this is a discrete model update, not a measured circuit response.

- **Line 7 —** `` steps += 1 ``

  Count the update just completed.

- **Line 8 —** `` print(f"Step {steps}: {v:.2f} V") ``

  Print the update number and voltage rounded to two decimal places; the stored voltage retains its precision.

- **Line 9 —** `` print(f"Charged to 95 % after {steps} steps.") ``

  Report how many updates were required to reach or exceed the target.



**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert steps == 14
assert v >= target and v < V
assert 12.0 * (1 - 0.8 ** 13) < target
````

## Exercise 4: Wait for a Stable Reading (Medium)

### Task and supplied example


A temperature probe settles after it is plugged in. In our model the reading at step `n` is `reading = 20 + 5 / n` (so it starts at 25.0 and falls towards 20).

Start at step 1. Use a `while` loop to take the next reading until the change from the previous reading is smaller than 0.1. Print each new reading with 3 decimals and the change, then report the step at which the reading became stable. Add a safety limit of 100 steps and report separately if that limit is reached without meeting the change threshold. Here “stable” only means the change between two readings is below 0.1; it does not mean the reading is within 0.1 of the final value 20.

**Expected output:**
```
Step 2: 22.500 (change 2.500)
Step 3: 21.667 (change 0.833)
Step 4: 21.250 (change 0.417)
Step 5: 21.000 (change 0.250)
Step 6: 20.833 (change 0.167)
Step 7: 20.714 (change 0.119)
Step 8: 20.625 (change 0.089)
Stable after 8 steps: 20.625
```

<details><summary>💡 Hints</summary>

- Keep two variables: `previous` (last reading) and `reading` (new one)
- Start with `change = 1.0` so the loop runs at least once; loop `while change >= 0.1 and step < 100:`
- Inside: `step += 1`, compute the new reading, `change = abs(previous - reading)`, then `previous = reading`

</details>

### Reasoning

"Stable" is defined by the change between two readings, so the loop must remember the previous value. The condition `change >= 0.1` keeps the loop going; the safety limit `step < 100` guarantees an exit even if the model never settles. Update `previous` only after the change is computed. Step 7 still changes by 0.119, so the first stable step is 8.

**TR:** Kararlılık iki okuma arasındaki farkla tanımlanır; önceki değeri sakla. Güvenlik sınırı döngünün mutlaka bitmesini sağlar.

````python
step = 1
previous = 20 + 5 / step
reading = previous
change = 1.0
while change >= 0.1 and step < 100:
    step += 1
    reading = 20 + 5 / step
    change = abs(previous - reading)
    print(f"Step {step}: {reading:.3f} (change {change:.3f})")
    previous = reading
if change < 0.1:
    print(f"Stable after {step} steps: {reading:.3f}")
else:
    print(f"Stopped at the safety limit: {reading:.3f}; not yet stable.")
````

**Line-by-line walkthrough**

- **Line 1 —** `` step = 1 ``

  Start at step 1 so the model never divides by zero.

- **Line 2 —** `` previous = 20 + 5 / step ``

  Compute the first model reading, 25, to use as the comparison baseline.

- **Line 3 —** `` reading = previous ``

  Initialize the current reading from that baseline.

- **Line 4 —** `` change = 1.0 ``

  Use an initial change of 1.0 so the loop begins; this is a starting value, not a measured difference.

- **Line 5 —** `` while change >= 0.1 and step < 100: ``

  Continue only while the change is at least 0.1 and the step count remains below the safety limit.

- **Line 6 —** `` step += 1 ``

  Advance to the next step before evaluating its reading.

- **Line 7 —** `` reading = 20 + 5 / step ``

  Calculate the model reading for the new step.

- **Line 8 —** `` change = abs(previous - reading) ``

  Measure the absolute difference between successive readings.

- **Line 9 —** `` print(f"Step {step}: {reading:.3f} (change {change:.3f})") ``

  Print the step, reading and change to three decimal places; the stopping test uses unrounded values.

- **Line 10 —** `` previous = reading ``

  Remember the current reading for the next comparison.

- **Line 11 —** `` if change < 0.1: ``

  After stopping, check whether the change met the stability criterion.

- **Line 12 —** `` print(f"Stable after {step} steps: {reading:.3f}") ``

  Report successful convergence when the difference is below 0.1.

- **Line 13 —** `` else: ``

  Otherwise handle termination at the safety limit.

- **Line 14 —** `` print(f"Stopped at the safety limit: {reading:.3f}; not yet stable.") ``

  Report that the model stopped without establishing stability.



**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert step == 8
assert change < 0.1
assert (20 + 5 / 6) - (20 + 5 / 7) >= 0.1
assert step < 100
````

**Line-by-line walkthrough**

- **Line 1 —** `` assert step == 8 ``

  Check that this model meets the chosen change threshold at step 8.

  Check that the last change is strictly below 0.1.

```

```

```

```

<details><summary>💡 Hints</summary>

- Initialize `count = 0`
- Use `while number > 0`
- Divide by 10 each time: `number = number // 10`
- Increment count each iteration
- Handle the special case where number is 0

</details>

### Reasoning

Copy the original number because repeated //10 destroys digits. Count one iteration per removed digit. The statement asks for positive integers; the hinted extension treats zero as one digit. For 1000, trailing zero digits still count.

**TR:** Basamakları silerken asıl sayıyı koru; 0 sayısı özel olarak bir basamaklıdır.

````python
original = 12345
number = original
count = 0
if number == 0:
    count = 1
while number > 0:
    number //= 10
    count += 1
print(f"Enter a positive integer: {original}")
print(f"Number of digits: {count}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` original = 12345 ``

  Store the example integer and preserve it for the final report.

- **Line 5 —** `` count = 1 ``

  Assign one digit to zero.

- **Line 6 —** `` while number > 0: ``

  Continue while the remaining nonnegative integer has digits to remove.

- **Line 7 —** `` number //= 10 ``

  Use integer division by ten to discard the rightmost digit.

- **Line 8 —** `` count += 1 ``

  Count the digit just removed.

- **Line 9 —** `` print(f"Enter a positive integer: {original}") ``

  Display the original example value; this is a simulated prompt, not an input() call.

  Print the number of digits found.



**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert count == 5 and number == 0
for sample, expected in [(0, 1), (7, 1), (1000, 4)]:
    remaining = sample
    digits = 1 if remaining == 0 else 0
    while remaining > 0:
        remaining //= 10
        digits += 1
    assert digits == expected
````

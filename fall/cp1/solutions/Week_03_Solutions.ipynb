### Task and supplied example


- Check `num > 0`, `num < 0`, or else it's zero

### Reasoning

## Exercise 4: Leap Year (Medium)

### Task and supplied example


Determine if a given `year` is a **leap year**. The rules are:
1. If divisible by 400 -> leap year
2. If divisible by 100 -> NOT a leap year
3. If divisible by 4 -> leap year
4. Otherwise -> NOT a leap year

Expected output for `year = 2024`:
```
2024 is a leap year
```

Expected output for `year = 1900`:
```
1900 is NOT a leap year
```

<details><summary>Hints</summary>

- Check divisibility using `%` (modulo): `year % 4 == 0`
- The order of checks matters! Check 400 first, then 100, then 4
- You can also combine conditions: `(year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)`

</details>

### Reasoning

The century exception must be checked before the general divisibility-by-4 rule. A compact equivalent condition is divisible by 400 OR (divisible by 4 AND not divisible by 100). Check 2024, 1900 and 2000 to distinguish these cases.

**TR:** Yüzyıl istisnasını önce kontrol et; 1900 artık yıl değildir, 2000 artık yıldır.

````python
year = 2024
if year % 400 == 0:
    leap = True
elif year % 100 == 0:
    leap = False
elif year % 4 == 0:
    leap = True
else:
    leap = False

if leap:
    print(f"{year} is a leap year")
else:
    print(f"{year} is NOT a leap year")

# The second supplied example, using the equivalent Boolean rule.
year2 = 1900
leap2 = year2 % 400 == 0 or (year2 % 4 == 0 and year2 % 100 != 0)
if leap2:
    print(f"{year2} is a leap year")
else:
    print(f"{year2} is NOT a leap year")
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert leap is True and leap2 is False
assert 2000 % 400 == 0
assert 2023 % 4 != 0
````

## Exercise 6: Triangle Validator (Medium)

### Task and supplied example


Given three side lengths `a`, `b`, `c`, determine if they can form a valid triangle. The **triangle inequality** states that the sum of any two sides must be greater than the third side.

Expected output for `a=3, b=4, c=5`:
```
Sides 3, 4, 5 form a valid triangle
```

Expected output for `a=1, b=2, c=10`:
```
Sides 1, 2, 10 do NOT form a valid triangle
```

<details><summary>Hints</summary>

- You need to check ALL three conditions:
  - `a + b > c`
  - `a + c > b`
  - `b + c > a`
- Combine them with `and`

</details>

### Reasoning

All side lengths must be positive and all three strict triangle inequalities must hold. Equality makes a flat, degenerate shape, not a triangle. Evaluate both supplied cases before interpreting the result.

**TR:** Üç koşulun hepsi doğru olmalı; eşitlik üçgen oluşturmaz.

````python
a, b, c = 3, 4, 5
valid = a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a
if valid:
    print(f"Sides {a}, {b}, {c} form a valid triangle")
else:
    print(f"Sides {a}, {b}, {c} do NOT form a valid triangle")

a2, b2, c2 = 1, 2, 10
valid2 = a2 > 0 and b2 > 0 and c2 > 0 and a2 + b2 > c2 and a2 + c2 > b2 and b2 + c2 > a2
if valid2:
    print(f"Sides {a2}, {b2}, {c2} form a valid triangle")
else:
    print(f"Sides {a2}, {b2}, {c2} do NOT form a valid triangle")
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert valid is True and valid2 is False
assert not (1 + 2 > 3)  # Equality is excluded.
assert not (0 > 0)
````

## Exercise 7: Maximum of Three (Medium)

### Task and supplied example


Find the **maximum** of three numbers `a`, `b`, `c` **without** using the built-in `max()` function.

Expected output for `a=15, b=42, c=8`:
```
The maximum of 15, 42, 8 is 42
```

<details><summary>Hints</summary>

- Start by assuming `a` is the largest: `maximum = a`
- Then check if `b` is larger: `if b > maximum: maximum = b`
- Then check if `c` is larger: `if c > maximum: maximum = c`
- Alternatively, use nested `if/elif/else`

</details>

### Reasoning

Start with an actual input, not zero: all inputs could be negative. Compare b with the best value so far, then compare c with the updated best. Use separate if statements because both updates may be necessary.

**TR:** En büyüğü sıfırdan başlatma; bütün sayılar negatif olabilir. Her yeni sayıyı güncel en büyükle karşılaştır.

````python
a, b, c = 15, 42, 8
maximum = a
if b > maximum:
    maximum = b
if c > maximum:
    maximum = c
print(f"The maximum of {a}, {b}, {c} is {maximum}")

# A negative-only case catches the common maximum = 0 mistake.
na, nb, nc = -5, -2, -9
negative_maximum = na
if nb > negative_maximum:
    negative_maximum = nb
if nc > negative_maximum:
    negative_maximum = nc
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert maximum == 42
assert negative_maximum == -2
````

## Exercise 9: BMI Classifier (Medium)

### Task and supplied example


Given `weight` (in kg) and `height` (in meters), calculate the BMI and classify it:

**BMI Formula:** `BMI = weight / (height ** 2)`

| BMI Range | Category |
|---|---|
| Below 18.5 | Underweight |
| 18.5 ≤ BMI < 25 | Normal weight |
| 25 ≤ BMI < 30 | Overweight |
| 30.0 and above | Obese |

Expected output for `weight=70, height=1.75`:
```
BMI: 22.86 -> Normal weight
```

<details><summary>Hints</summary>

- First calculate: `bmi = weight / (height ** 2)`
- Then use `if/elif/else` to classify
- Use `round(bmi, 2)` or f-string formatting `f"{bmi:.2f}"` for display

</details>

### Reasoning

Calculate using the supplied positive height, then test increasing upper limits. Use <25 and <30 rather than <=24.9 and <=29.9: continuous decimal values must not fall into gaps.

**TR:** Sınırları kesintisiz yaz; 24.95 gibi bir değer normal aralığın içinde kalır.

````python
weight = 70
height = 1.75
if height <= 0 or weight <= 0:
    print("Height and weight must be positive.")
else:
    bmi = weight / height ** 2
    if bmi < 18.5:
        category = "Underweight"
    elif bmi < 25:
        category = "Normal weight"
    elif bmi < 30:
        category = "Overweight"
    else:
        category = "Obese"
    print(f"BMI: {bmi:.2f} -> {category}")
````

## Exercise 10: Quadrant Finder (Medium)

Given coordinates `x` and `y`, determine which **quadrant** the point is in:
- Quadrant I:   x > 0, y > 0
- Quadrant II:  x < 0, y > 0
- Quadrant III: x < 0, y < 0
- Quadrant IV:  x > 0, y < 0
- On an axis if x == 0 or y == 0

Expected output for `x=3, y=-2`:
```
Point (3, -2) is in Quadrant IV
```

<details><summary>Hints</summary>

- First check if the point is on an axis (x == 0 or y == 0)
- Then use `and` to check both x and y signs together
- For example: `x > 0 and y > 0` means Quadrant I

Given a single `letter`, determine if it is a **vowel** (a, e, i, o, u) or a **consonant**. Handle both uppercase and lowercase.

Expected output for `letter = "E"`:
```

```

<details><summary>Hints</summary>

- Convert to lowercase first: `letter.lower()`
- Check if it's in the vowels: `letter.lower() in "aeiou"`
- Or use multiple `or` comparisons

</details>

### Reasoning

Use the exercise's five English vowels. First require exactly one English letter; lowercase once, then check membership in aeiou. This prevents punctuation or multiple characters from being labelled consonants.

**TR:** Önce tek bir İngilizce harf olduğunu doğrula; sonra küçük harfe çevirip sesli harf kümesinde ara.

## Exercise 12: Simple Calculator (Challenge)

### Task and supplied example


Build a simple calculator! Given two numbers `num1`, `num2` and an `operation` string (`"+"`, `"-"`, `"*"`, `"/"`), perform the calculation and print the result. Handle **division by zero**!

Expected output for `num1=10, num2=3, operation="/"`:
```
10 / 3 = 3.3333
```

Expected output for `num1=10, num2=0, operation="/"`:
```
Error: Cannot divide by zero!
```

<details><summary>Hints</summary>

- Use `if/elif/else` to check the operation string
- For division, add a **nested if** to check if `num2 == 0`
- Use `f"{result:.4f}"` to format to 4 decimal places for division
- Add an else for invalid operations

</details>

### Reasoning

Choose the operation first, but check a zero denominator before dividing. Store a result only on a valid arithmetic branch. Invalid operations produce a message, not an undefined calculation.

**TR:** Bölme işleminden önce paydayı kontrol et; hata durumunda sonuç hesaplamaya çalışma.

````python
num1, num2, operation = 10, 3, "/"
result = None
if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    if num2 == 0:
        print("Error: Cannot divide by zero!")
    else:
        result = num1 / num2
else:
    print("Error: Unknown operation")

if result is not None:
    print(f"{num1} {operation} {num2} = {result:.4f}")

# The other supplied case: demonstrate the guard without dividing.
zero_denominator = 0
if zero_denominator == 0:
    zero_message = "Error: Cannot divide by zero!"
    print(zero_message)
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert f"{result:.4f}" == "3.3333"
assert zero_message == "Error: Cannot divide by zero!"
assert 10 + 3 == 13 and 10 - 3 == 7 and 10 * 3 == 30
````

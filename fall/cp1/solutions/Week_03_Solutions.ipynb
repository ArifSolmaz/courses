A motor controller reads the housing `temperature` in °C and reports a status:

| Temperature (°C) | Status |

| 90 and above | SHUTDOWN |

Expected output for `temperature = 78.5`:
```
Temperature: 78.5 C -> WARNING
```

<details><summary>Hints</summary>

- Check the sensor-fault range first, then use an `if/elif/else` chain
- Test from the highest threshold (90) down, or from the lowest up — but not mixed

## Exercise 4: Tolerance Check on a Shaft (Medium)

### Task and supplied example


A shaft must have a diameter of `nominal = 10.00` mm with a tolerance of `tolerance = 0.05` mm. A measurement passes when it is inside `nominal - tolerance` to `nominal + tolerance`, **including** both limits.

For each measured value, print PASS or FAIL and the deviation (measured minus nominal) with 2 decimals.

Expected output for `measured_1 = 10.04` and `measured_2 = 9.93`:
```
Measured 10.04 mm: PASS (deviation 0.04 mm)
Measured 9.93 mm: FAIL (deviation -0.07 mm)
```

<details><summary>Hints</summary>

- A chained comparison does the whole check: `nominal - tolerance <= measured <= nominal + tolerance`
- Use `if/else` for PASS/FAIL
- `f"{deviation:.2f}"` hides floating-point noise such as 0.03999999999999915

</details>

### Reasoning

Compute the limits once from the nominal and the tolerance. "Including both limits" means `<=` on both sides, so 10.05 passes and 10.051 fails. The deviation is a separate calculation for the report; it does not change the decision. Test the boundary values 9.95 and 10.05 as well as the two supplied readings.

**TR:** Alt ve üst sınırı bir kez hesapla. "Dahil" demek `<=` demektir; 10.05 geçer, 10.051 geçmez.

````python
nominal = 10.00
tolerance = 0.05
low = nominal - tolerance
high = nominal + tolerance

measured_1 = 10.04
deviation_1 = measured_1 - nominal
if low <= measured_1 <= high:
    result_1 = "PASS"
else:
    result_1 = "FAIL"
print(f"Measured {measured_1} mm: {result_1} (deviation {deviation_1:.2f} mm)")

measured_2 = 9.93
deviation_2 = measured_2 - nominal
if low <= measured_2 <= high:
    result_2 = "PASS"
else:
    result_2 = "FAIL"
print(f"Measured {measured_2} mm: {result_2} (deviation {deviation_2:.2f} mm)")
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert result_1 == "PASS" and result_2 == "FAIL"
assert f"{deviation_1:.2f}" == "0.04" and f"{deviation_2:.2f}" == "-0.07"
# Edge cases: both limits are inclusive; just outside fails.
assert low <= 9.95 <= high and low <= 10.05 <= high
assert not (low <= 10.051 <= high)
````

## Exercise 6: Triangle Validator (Medium)

### Task and supplied example


Given three side lengths `a`, `b`, `c`, determine if they can form a valid triangle. The **triangle inequality** states that the sum of any two sides must be greater than the third side.

Expected output for `a=3, b=4, c=5`:
```
Sides 3, 4, 5 form a valid triangle
```

Expected output for `a=1, b=2, c=10`:
```
Sides 1, 2, 10 do NOT form a valid triangle
```

<details><summary>Hints</summary>

- You need to check ALL three conditions:
  - `a + b > c`
  - `a + c > b`
  - `b + c > a`
- Combine them with `and`

</details>

### Reasoning

All side lengths must be positive and all three strict triangle inequalities must hold. Equality makes a flat, degenerate shape, not a triangle. Evaluate both supplied cases before interpreting the result.

**TR:** Üç koşulun hepsi doğru olmalı; eşitlik üçgen oluşturmaz.

````python
a, b, c = 3, 4, 5
valid = a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a
if valid:
    print(f"Sides {a}, {b}, {c} form a valid triangle")
else:
    print(f"Sides {a}, {b}, {c} do NOT form a valid triangle")

a2, b2, c2 = 1, 2, 10
valid2 = a2 > 0 and b2 > 0 and c2 > 0 and a2 + b2 > c2 and a2 + c2 > b2 and b2 + c2 > a2
if valid2:
    print(f"Sides {a2}, {b2}, {c2} form a valid triangle")
else:
    print(f"Sides {a2}, {b2}, {c2} do NOT form a valid triangle")
````

## Exercise 7: Maximum of Three (Medium)

### Task and supplied example


Find the **maximum** of three numbers `a`, `b`, `c` **without** using the built-in `max()` function.

Expected output for `a=15, b=42, c=8`:
```
The maximum of 15, 42, 8 is 42
```

<details><summary>Hints</summary>

- Start by assuming `a` is the largest: `maximum = a`
- Then check if `b` is larger: `if b > maximum: maximum = b`
- Then check if `c` is larger: `if c > maximum: maximum = c`
- Alternatively, use nested `if/elif/else`

</details>

### Reasoning

Start with an actual input, not zero: all inputs could be negative. Compare b with the best value so far, then compare c with the updated best. Use separate if statements because both updates may be necessary.

**TR:** En büyüğü sıfırdan başlatma; bütün sayılar negatif olabilir. Her yeni sayıyı güncel en büyükle karşılaştır.

````python
a, b, c = 15, 42, 8
maximum = a
if b > maximum:
    maximum = b
if c > maximum:
    maximum = c
print(f"The maximum of {a}, {b}, {c} is {maximum}")

# A negative-only case catches the common maximum = 0 mistake.
na, nb, nc = -5, -2, -9
negative_maximum = na
if nb > negative_maximum:
    negative_maximum = nb
if nc > negative_maximum:
    negative_maximum = nc
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert maximum == 42
assert negative_maximum == -2
````

## Exercise 9: Battery State of Charge Band (Medium)

### Task and supplied example


A lithium cell reads 3.0 V when empty and 4.2 V when full. Estimate the state of charge (SoC) from the cell `voltage` and classify it:

**SoC formula:** `soc = (voltage - 3.0) / (4.2 - 3.0) * 100`

| Voltage / SoC | Band |
|---|---|
| voltage below 3.0 or above 4.2 | INVALID READING |
| SoC below 20 % | LOW |
| 20 % ≤ SoC < 80 % | OK |
| 80 % and above | FULL |

Expected output for `voltage = 3.85`:
```
Voltage: 3.85 V -> SoC: 70.8 % -> OK
```

<details><summary>Hints</summary>

- Check the voltage range first; do not compute SoC for an invalid reading
- Then use `if/elif/else` on `soc` with `< 20` and `< 80`
- Show the SoC with `f"{soc:.1f}"`

</details>

### Reasoning

Validate the voltage first, then scale the 1.2 V span to 0–100 %. Classify the unrounded value with increasing upper limits `< 20` and `< 80`; a continuous value such as 79.95 must not fall into a gap. This is a linear teaching estimate, not a real battery model.

**TR:** Önce voltajı doğrula, sonra 1.2 V aralığını yüzdeye ölçekle. Sınıflandırmayı yuvarlanmamış değerle yap.

````python
voltage = 3.85
if voltage < 3.0 or voltage > 4.2:
    print(f"Voltage: {voltage} V -> INVALID READING")
else:
    soc = (voltage - 3.0) / (4.2 - 3.0) * 100
    if soc < 20:
        band = "LOW"
    elif soc < 80:
        band = "OK"
    else:
        band = "FULL"
    print(f"Voltage: {voltage} V -> SoC: {soc:.1f} % -> {band}")
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert f"{soc:.1f}" == "70.8"
assert band == "OK"
# Edge cases: 3.0 V is exactly 0 % (LOW), 4.2 V is 100 % (FULL); 79.95 stays OK.
assert (3.0 - 3.0) / 1.2 * 100 == 0
assert abs((4.2 - 3.0) / 1.2 * 100 - 100) < 1e-9
assert 20 <= 79.95 < 80
````

## Exercise 12: Simple Calculator (Challenge)

### Task and supplied example


Build a simple calculator! Given two numbers `num1`, `num2` and an `operation` string (`"+"`, `"-"`, `"*"`, `"/"`), perform the calculation and print the result. Handle **division by zero**!

Expected output for `num1=10, num2=3, operation="/"`:
```
10 / 3 = 3.3333
```

Expected output for `num1=10, num2=0, operation="/"`:
```
Error: Cannot divide by zero!
```

<details><summary>Hints</summary>

- Use `if/elif/else` to check the operation string
- For division, add a **nested if** to check if `num2 == 0`
- Use `f"{result:.4f}"` to format to 4 decimal places for division
- Add an else for invalid operations

</details>

### Reasoning

Choose the operation first, but check a zero denominator before dividing. Store a result only on a valid arithmetic branch. Invalid operations produce a message, not an undefined calculation.

**TR:** Bölme işleminden önce paydayı kontrol et; hata durumunda sonuç hesaplamaya çalışma.

````python
num1, num2, operation = 10, 3, "/"
result = None
if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    if num2 == 0:
        print("Error: Cannot divide by zero!")
    else:
        result = num1 / num2
else:
    print("Error: Unknown operation")

if result is not None:
    print(f"{num1} {operation} {num2} = {result:.4f}")

# The other supplied case: demonstrate the guard without dividing.
zero_denominator = 0
if zero_denominator == 0:
    zero_message = "Error: Cannot divide by zero!"
    print(zero_message)
````

**Verification:** Run the next cell after the answer. The expected values check this example and, where relevant, important boundaries. They are not additional exercises. Automated checks may use compact syntax covered later; the main answer above is the teaching route.

````python
assert f"{result:.4f}" == "3.3333"
assert zero_message == "Error: Cannot divide by zero!"
assert 10 + 3 == 13 and 10 - 3 == 7 and 10 * 3 == 30
````

Expected output for `measured_1 = 10.04` and `measured_2 = 9.93`:
```

Measured 9.93 mm: FAIL (deviation -0.07 mm)
```

<details><summary>Hints</summary>

- A chained comparison does the whole check: `nominal - tolerance <= measured <= nominal + tolerance`

</details>

### Reasoning

Compute the limits once from the nominal and the tolerance. "Including both limits" means `<=` on both sides, so 10.05 passes and 10.051 fails. ## Exercise 6: Triangle Validator (Medium)

Expected output for `a=3, b=4, c=5`:
```
Sides 3, 4, 5 form a valid triangle
```

Expected output for `a=1, b=2, c=10`:
```
Sides 1, 2, 10 do NOT form a valid triangle
```

<details><summary>Hints</summary>

- You need to check ALL three conditions:
  - `a + b > c`
  - `a + c > b`

### Reasoning

## Exercise 1: Catch Division Error (Easy)

Write a function `safe_divide(a, b)` that:
- Returns the result of `a / b`
- Catches `ZeroDivisionError` and returns the string `"Error: division by zero"`

**Expected output:**
```
safe_divide(10, 3) = 3.3333333333333335
safe_divide(10, 0) = Error: division by zero
safe_divide(0, 5) = 0.0
```

### Worked reasoning

Put the division inside try. Catch only ZeroDivisionError because that is the failure this function promises to translate into a message. A zero numerator is allowed; a zero denominator is not.

**Türkçe:** Sıfırı bölmek geçerlidir; sıfıra bölmek tanımsızdır.

**Expected check:** 10/3≈3.3333; division-by-zero message; 0/5=0.0.


````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: division by zero"

for a, b in [(10, 3), (10, 0), (0, 5)]:
    print(f"safe_divide({a}, {b}) = {safe_divide(a, b)}")
assert safe_divide(10, 0) == "Error: division by zero"
assert safe_divide(0, 5) == 0.0
assert safe_divide(-8, 2) == -4
````

## Exercise 2: Safe Integer Input (Easy)

Write a function `safe_int(text)` that:
- Tries to convert `text` to an integer
- Returns the integer if successful
- Returns `None` if it fails (catching `ValueError`)
- Prints a message when conversion fails

**Expected output:**
```
safe_int('42') = 42
safe_int('3.14') = None  (printed: Cannot convert '3.14' to integer)
safe_int('hello') = None  (printed: Cannot convert 'hello' to integer)
safe_int('0') = 0
```

### Worked reasoning

Attempt int(text) directly. An integer string may include a sign and surrounding whitespace, but a decimal string such as "3.14" is not an integer literal. Return None on failure so 0 is not confused with an invalid conversion.

**Türkçe:** None başarısızlığı gösterir; 0 ise başarılı bir dönüşüm sonucu olabilir.

**Expected check:** 42, None, None and 0, with messages for the two failures.


````python
def safe_int(text):
    try:
        return int(text)
    except (ValueError, TypeError):
        print(f"Cannot convert {text!r} to integer")
        return None

for text in ["42", "3.14", "hello", "0"]:
    print(f"safe_int({text!r}) = {safe_int(text)}")
assert safe_int(" -12 ") == -12
assert safe_int("0") == 0
assert safe_int("3.14") is None
````

## Exercise 3: Safe List Access (Medium)

Write a function `safe_get(lst, index, default=None)` that:
- Returns `lst[index]` if the index is valid
- Returns `default` if an `IndexError` occurs
- Returns `default` if a `TypeError` occurs (e.g., index is a string)

**Expected output:**
```
safe_get([10, 20, 30], 1) = 20
safe_get([10, 20, 30], 10) = None
safe_get([10, 20, 30], -1) = 30
safe_get([10, 20, 30], 'a') = None
safe_get([10, 20, 30], 10, default=-1) = -1
```

### Worked reasoning

Let normal list indexing decide validity. Catch IndexError for an index outside the list and TypeError for an unsuitable index type. Python already supports negative indices, so a separate index≥0 test would incorrectly reject -1.

**Türkçe:** Negatif indeks geçerlidir; sınır kontrolünü yanlış biçimde daraltmayın.

**Expected check:** 20, None, 30, None, then custom default -1.


````python
def safe_get(lst, index, default=None):
    try:
        return lst[index]
    except (IndexError, TypeError):
        return default

for index in [1, 10, -1, "a"]:
    print(f"safe_get([10, 20, 30], {index!r}) = {safe_get([10, 20, 30], index)}")
print("Custom default:", safe_get([10, 20, 30], 10, default=-1))
assert safe_get([10, 20, 30], -3) == 10
assert safe_get([], 0, "missing") == "missing"
assert safe_get([10], 2.5) is None
````

## Exercise 4: Calculator with Error Handling (Medium)

Write a function `calc(expression)` that takes a string like `"10 + 3"` and:
- Splits it into number, operator, number
- Supports `+`, `-`, `*`, `/`
- Handles `ValueError` (non-numeric values)
- Handles `ZeroDivisionError` (division by zero)
- Returns a helpful error message for each case

**Expected output:**
```
calc('10 + 3') = 13.0
calc('10 / 0') = Error: division by zero
calc('abc + 3') = Error: invalid number
calc('10 % 3') = Error: unknown operator '%'
```



**Finite-number rule:** after numeric conversion, reject `nan`, `inf`, and `-inf` with `math.isfinite()`. For an average with no valid numbers, return the documented empty result. Boolean values are flags, not measurements, in these validation exercises.

### Worked reasoning

Split the expression and check that it has three fields before unpacking. Convert the outer fields, reject nonfinite operands, then select an operation. Division-by-zero and malformed numbers receive different messages. Never use eval to interpret this small input language.

**Türkçe:** Önce biçimi, sonra sayıları, sonra işlemi kontrol edin; hata mesajı hangi aşamanın başarısız olduğunu söylemelidir.

**Expected check:** 13.0 and distinct division, number, operator errors; malformed and nonfinite cases are also rejected.


````python
import math

def calc(expression):
    if not isinstance(expression, str):
        return "Error: expression must be text"
    parts = expression.split()
    if len(parts) != 3:
        return "Error: use number operator number"
    left, operator, right = parts
    try:
        a, b = float(left), float(right)
    except ValueError:
        return "Error: invalid number"
    if not math.isfinite(a) or not math.isfinite(b):
        return "Error: numbers must be finite"
    try:
        if operator == "+":
            result = a + b
        elif operator == "-":
            result = a - b
        elif operator == "*":
            result = a * b
        elif operator == "/":
            result = a / b
        else:
            return f"Error: unknown operator '{operator}'"
    except ZeroDivisionError:
        return "Error: division by zero"
    return result if math.isfinite(result) else "Error: result is not finite"

for expression in ["10 + 3", "10 / 0", "abc + 3", "10 % 3"]:
    print(f"calc({expression!r}) = {calc(expression)}")
assert calc("10 + 3") == 13.0
assert calc("10 / 0") == "Error: division by zero"
assert calc("10 +") == "Error: use number operator number"
assert calc("nan + 3") == "Error: numbers must be finite"
assert calc("1e308 * 2") == "Error: result is not finite"
````

## Exercise 5: Safe Division Function (Medium)

Write a function `divide_many(numbers)` that takes a list of numbers and divides them sequentially (first / second / third / ...).

- Handle empty list (return `None`)
- Handle single item list (return that item)
- Handle `ZeroDivisionError` — skip zeros and print a warning
- Handle `TypeError` — skip non-numbers and print a warning

**Expected output:**
```
divide_many([100, 2, 5]) = 10.0
divide_many([100, 0, 5]) = 20.0  (printed: Skipping division by zero)
divide_many([100, 'a', 5]) = 20.0  (printed: Skipping non-number: a)
divide_many([]) = None
```



**Initial value:** for this exercise the first item must be a finite number. If it is invalid, return `None` with a message; only later invalid divisors are skipped. An initial zero is valid.

### Worked reasoning

The first number is the starting value. A failed later division must leave the accumulated result unchanged so processing can continue. For [100,0,5], skip the zero and calculate 100/5=20. The first item cannot simply be skipped because that would change which value is the numerator.

**Türkçe:** Geçersiz bölen atlanır; ancak başlangıç değeri geçersizse hesap için anlamlı bir temel yoktur.

**Expected check:** 10.0, 20.0, 20.0 and None; a skipped divisor does not alter the current result.


````python
def divide_many(numbers):
    if not numbers:
        return None
    first = numbers[0]
    if isinstance(first, bool) or not isinstance(first, (int, float)) or not math.isfinite(first):
        print("Invalid initial value")
        return None
    result = first
    for number in numbers[1:]:
        try:
            if isinstance(number, bool) or not isinstance(number, (int, float)):
                raise TypeError("not a numeric divisor")
            if not math.isfinite(number):
                print(f"Skipping nonfinite divisor: {number}")
                continue
            result = result / number
        except ZeroDivisionError:
            print("Skipping division by zero")
        except TypeError:
            print(f"Skipping non-number: {number}")
    return result

for values in [[100, 2, 5], [100, 0, 5], [100, "a", 5], []]:
    print(f"divide_many({values}) = {divide_many(values)}")
assert divide_many([100, 2, 5]) == 10
assert divide_many([0, 5]) == 0
assert divide_many([7]) == 7
assert divide_many(["bad", 2]) is None
assert divide_many([100, float("nan"), 5]) == 20
````

## Exercise 6: Validate and Convert (Medium)

Write a function `convert_all(values, target_type)` that converts a list of values to the specified type (`"int"`, `"float"`, or `"str"`).

- Return a list of successfully converted values
- Print a message for each failed conversion
- Count and report successes and failures

**Expected output:**
```
convert_all(['1', '2.5', 'abc', '4'], 'int')
  '1' → 1 ✓
  '2.5' → Failed (invalid literal for int()...)
  'abc' → Failed (invalid literal for int()...)
  '4' → 4 ✓
  Results: 2 succeeded, 2 failed
  Converted: [1, 4]
```

### Worked reasoning

Select the conversion function once from a dictionary, then call it on each value inside try. Conversion success is about Python representation, not a physical validation rule: float("nan") is a float, and a later domain-specific validator must reject it if finite measurements are required. Unknown target types fail clearly.

**Türkçe:** Dönüşüm ile alan doğrulaması farklıdır: float oluşturmak, geçerli ölçüm elde etmek demek değildir.

**Expected check:** Two integer conversions succeed: [1, 4]; two fail without ending the loop.


````python
def convert_all(values, target_type):
    converters = {"int": int, "float": float, "str": str}
    if target_type not in converters:
        raise ValueError("target_type must be int, float, or str")
    convert = converters[target_type]
    converted = []
    failed = 0
    for value in values:
        try:
            result = convert(value)
        except (ValueError, TypeError, OverflowError) as error:
            failed += 1
            print(f"{value!r} -> Failed ({type(error).__name__})")
        else:
            converted.append(result)
            print(f"{value!r} -> {result!r}")
    print(f"Results: {len(converted)} succeeded, {failed} failed")
    return converted

converted = convert_all(["1", "2.5", "abc", "4"], "int")
print("Converted:", converted)
assert converted == [1, 4]
assert convert_all(["1", "2.5"], "float") == [1.0, 2.5]
assert convert_all([None, 3], "str") == ["None", "3"]
````

## Exercise 7: Robust Average (Medium)

Write a function `robust_average(data)` that:
- Handles an empty list (return `0` with a message)
- Handles non-list input (return `0` with a message)
- Skips non-numeric values with a warning
- Returns the average of valid numbers

**Expected output:**
```
robust_average([10, 20, 'abc', 30, None, 40]) = 25.0
  (printed: Skipping non-numeric: 'abc', Skipping non-numeric: None)
  (4 valid values out of 6)
robust_average([]) = 0
  (printed: Empty list, returning 0)
robust_average('hello') = 0
  (printed: Expected a list, got str)
```



**Finite-number rule:** after numeric conversion, reject `nan`, `inf`, and `-inf` with `math.isfinite()`. For an average with no valid numbers, return the documented empty result. Boolean values are flags, not measurements, in these validation exercises.

### Worked reasoning

Validate the container first, then convert each candidate and reject flags/nonfinite values. Accumulate only valid numbers, and divide by the number retained. For the example the sum is 100 and count is 4, giving 25. The specified empty return is a programming sentinel 0, not a claim that the mathematical mean of no values is zero.

**Türkçe:** Ortalamanın paydası toplam girdi sayısı değil, geçerli değer sayısıdır.

**Expected check:** 25.0 for the example; [85,nan,92,inf] retains only 85 and 92, mean 88.5.


````python
def robust_average(data):
    if not isinstance(data, list):
        print(f"Expected a list, got {type(data).__name__}")
        return 0
    valid = []
    for item in data:
        try:
            if isinstance(item, bool):
                raise TypeError("boolean is a flag")
            value = float(item)
        except (TypeError, ValueError):
            print(f"Skipping non-numeric: {item!r}")
            continue
        if not math.isfinite(value):
            print(f"Skipping nonfinite: {item!r}")
            continue
        valid.append(value)
    if not valid:
        print("No valid numbers, returning 0")
        return 0
    print(f"{len(valid)} valid values out of {len(data)}")
    return sum(valid) / len(valid)

result = robust_average([10, 20, "abc", 30, None, 40])
print("Average:", result)
assert result == 25
assert robust_average([]) == 0
assert robust_average("hello") == 0
assert robust_average(["85", "nan", "92", "inf"]) == 88.5
assert robust_average([True, False]) == 0
````

## Exercise 8: Temperature Converter with Full Validation (Medium)

Write a function `convert_temperature(value, from_unit, to_unit)` that:
- Converts between "C" (Celsius), "F" (Fahrenheit), and "K" (Kelvin)
- Validates that `value` is a number
- Validates that units are "C", "F", or "K"
- Checks that temperature is above absolute zero (-273.15°C, -459.67°F, 0K)
- Returns the converted value or an error message

**Expected output:**
```
convert_temperature(100, 'C', 'F') = 212.0
convert_temperature(32, 'F', 'C') = 0.0
convert_temperature(-300, 'C', 'K') = Error: below absolute zero
convert_temperature(100, 'C', 'X') = Error: unknown unit 'X'
convert_temperature('hot', 'C', 'F') = Error: value must be a number
```



**Finite-number rule:** after numeric conversion, reject `nan`, `inf`, and `-inf` with `math.isfinite()`. For an average with no valid numbers, return the documented empty result. Boolean values are flags, not measurements, in these validation exercises.

**Boundary:** absolute zero itself is permitted (Kelvin ≥0). Use Celsius as the common intermediate unit and test the boundary as well as ordinary temperatures.

### Worked reasoning

Check conversion, finiteness, unit names and the lower physical boundary. Convert the source to Celsius, then Celsius to the target. The original-unit boundary check avoids classifying an exact absolute-zero input as invalid due to roundoff in an intermediate conversion. A result within 1e-12 of zero is displayed as zero.

**Türkçe:** Ortak ara birim dönüşümü sadeleştirir. Mutlak sıfır sınırının kendisi kabul edilir; altı reddedilir.

**Expected check:** 212.0°F and 0.0°C; bad value/unit/physical boundary each gets a distinct message.


````python
def convert_temperature(value, from_unit, to_unit):
    try:
        if isinstance(value, bool):
            raise TypeError("boolean is not a temperature")
        number = float(value)
    except (ValueError, TypeError):
        return "Error: value must be a number"
    if not math.isfinite(number):
        return "Error: value must be finite"
    lower_bounds = {"C": -273.15, "F": -459.67, "K": 0.0}
    for unit in [from_unit, to_unit]:
        if unit not in lower_bounds:
            return f"Error: unknown unit '{unit}'"
    if number < lower_bounds[from_unit]:
        return "Error: below absolute zero"
    if from_unit == "C":
        celsius = number
    elif from_unit == "F":
        celsius = (number - 32) * 5 / 9
    else:
        celsius = number - 273.15
    if to_unit == "C":
        result = celsius
    elif to_unit == "F":
        result = celsius * 9 / 5 + 32
    else:
        result = celsius + 273.15
    return 0.0 if math.isclose(result, 0, abs_tol=1e-12) else result

for args in [(100,"C","F"),(32,"F","C"),(-300,"C","K"),(100,"C","X"),("hot","C","F")]:
    print(args, "->", convert_temperature(*args))
assert convert_temperature(100, "C", "F") == 212
assert convert_temperature(-273.15, "C", "K") == 0
assert convert_temperature(-459.67, "F", "K") == 0
assert convert_temperature("nan", "C", "F") == "Error: value must be finite"
assert math.isclose(convert_temperature(273.15, "K", "C"), 0)
````

## Exercise 9: Grade Entry System (Medium)

Write a function `enter_grades(grade_strings)` that takes a list of grade strings and:
- Converts each to an integer
- Validates that each grade is between 0 and 100
- Returns a dictionary with:
  - `"valid"`: list of valid grades
  - `"invalid"`: list of `(original_value, reason)` tuples
  - `"average"`: average of valid grades (or 0 if none)

**Expected output:**
```
Results:
  Valid grades: [85, 92, 78, 100, 0]
  Invalid entries:
    'abc' - not a number
    '-5' - out of range (0-100)
    '150' - out of range (0-100)
  Average: 71.0
```

### Worked reasoning

Convert each grade string to an integer, then check the inclusive 0–100 interval. Store the original string with its rejection reason so the caller can correct it. Sum 85+92+78+100+0=355; 355/5=71.0. Return all three results in one dictionary.

**Türkçe:** Hatalı verinin özgün yazısını saklamak, kullanıcının hangi girdiyi düzelteceğini görmesini sağlar.

**Expected check:** Five valid grades, three invalid entries, average 71.0.


````python
def enter_grades(grade_strings):
    valid, invalid = [], []
    for text in grade_strings:
        try:
            grade = int(text)
        except (ValueError, TypeError):
            invalid.append((text, "not a number"))
            continue
        if not 0 <= grade <= 100:
            invalid.append((text, "out of range (0-100)"))
        else:
            valid.append(grade)
    return {"valid": valid, "invalid": invalid,
            "average": sum(valid) / len(valid) if valid else 0}

results = enter_grades(["85", "abc", "92", "-5", "78", "150", "100", "0"])
print("Valid grades:", results["valid"])
for text, reason in results["invalid"]:
    print(f"{text!r} - {reason}")
print("Average:", results["average"])
assert results["valid"] == [85, 92, 78, 100, 0]
assert len(results["invalid"]) == 3 and results["average"] == 71
assert enter_grades([])["average"] == 0
assert enter_grades(["nan"])["valid"] == []
````

## Exercise 10: Multiple Error Types (Medium)

Write a function `process_record(record)` that takes a dictionary and extracts student information. Handle each error type with a specific message.

The function should:
- Access `record["name"]` — handle `KeyError`
- Access `record["grades"][0]` — handle `IndexError`
- Calculate `sum(record["grades"]) / len(record["grades"])` — handle `ZeroDivisionError`
- Convert `record["id"]` to `int` — handle `ValueError`

**Test records and expected output:**
```python
records = [
    {"name": "Elif", "id": "12345", "grades": [85, 92, 78]},
    {"id": "12346", "grades": [90]},           # missing name
    {"name": "Cem", "id": "abc", "grades": [75]}, # bad id
    {"name": "Deniz", "id": "12348", "grades": []}, # empty grades
]
```
```
Elif (12345): avg = 85.0 ✓
KeyError: 'name' is missing
Cem: ValueError: invalid ID 'abc'
Deniz (12348): ZeroDivisionError: no grades to average
```



**Operation order:** check name, ID, first-grade access, then average. Catch an empty-list `IndexError` locally and continue to the average check; the final returned message for an empty list is the documented `ZeroDivisionError: no grades to average`. This lets both operations be demonstrated without one hiding the other.

### Worked reasoning

Treat each risky operation separately. A missing name prevents a useful record label. A bad ID prevents processing that record. An empty grades list triggers IndexError when reading position 0; catch that locally so the subsequent average check can demonstrate ZeroDivisionError and return the specified final message.

**Türkçe:** Tek geniş try bloğu ilk hatada durur. Ayrı bloklar, hangi işlemin hangi hatayı oluşturduğunu gösterir.

**Expected check:** Elif succeeds; missing name, bad ID and empty grades are explained independently.


````python
def process_record(record):
    try:
        name = record["name"]
    except KeyError:
        return "KeyError: 'name' is missing"
    try:
        student_id = int(record["id"])
    except KeyError:
        return f"{name}: KeyError: 'id' is missing"
    except (ValueError, TypeError):
        return f"{name}: ValueError: invalid ID {record.get('id')!r}"
    try:
        grades = record["grades"]
    except KeyError:
        return f"{name}: KeyError: 'grades' is missing"
    try:
        first_grade = grades[0]
    except IndexError:
        first_grade = None  # no first grade; still demonstrate the average check
    try:
        average = sum(grades) / len(grades)
    except ZeroDivisionError:
        return f"{name} ({student_id}): ZeroDivisionError: no grades to average"
    except TypeError:
        return f"{name} ({student_id}): TypeError: grades must be numbers"
    return f"{name} ({student_id}): avg = {average:.1f} ✓"

records = [
    {"name":"Elif", "id":"12345", "grades":[85,92,78]},
    {"id":"12346", "grades":[90]},
    {"name":"Cem", "id":"abc", "grades":[75]},
    {"name":"Deniz", "id":"12348", "grades":[]},
]
messages = [process_record(record) for record in records]
for message in messages:
    print(message)
assert messages[0] == "Elif (12345): avg = 85.0 ✓"
assert messages[1] == "KeyError: 'name' is missing"
assert "invalid ID 'abc'" in messages[2]
assert "no grades to average" in messages[3]
````

- Takes a prompt string, a validator function, and max attempts
- The validator function takes a string and returns `(True, converted_value)` or `(False, error_message)`

Create these validators:

Since we can't use real `input()`, simulate it with a list of test inputs.

Testing positive int validator:

**Deterministic input:** add an `inputs` list parameter (or a small input-provider parameter) so tests do not wait for a keyboard. Stop when the list is exhausted or `max_tries` is reached. For names, each whitespace-separated word must contain letters only.

A validator returns a decision and either the converted value or an explanation. The retry loop only orchestrates attempts. **Türkçe:** Doğrulama kuralı ile yeniden deneme döngüsü ayrıdır; farklı kurallar aynı döngüyü kullanabilir.

**Expected check:** The third positive-integer attempt returns 42; the third grade attempt returns 85; exhaustion returns None.


````python
def validate_positive_int(text):
    try:
        value = int(text)
    except (ValueError, TypeError):
        return False, "not a valid integer"
    return (True, value) if value > 0 else (False, "must be positive")

def validate_grade(text):
    try:
        value = int(text)
    except (ValueError, TypeError):
        return False, "not a valid integer"
    return (True, value) if 0 <= value <= 100 else (False, "must be between 0 and 100")

def validate_name(text):
    if not isinstance(text, str):
        return False, "name must be text"
    words = text.split()
    if len(words) < 2 or not all(word.isalpha() for word in words):
        return False, "use at least two words, letters only"
    return True, " ".join(words)

def get_valid_input(prompt, validator, inputs, max_tries=3):
    print(prompt)
    for text in inputs[:max(0, max_tries)]:
        valid, value_or_error = validator(text)
        label = "Valid" if valid else "Invalid"
        print(f"  {text!r} -> {label}: {value_or_error}")
        if valid:
            return value_or_error
    return None

positive = get_valid_input("Testing positive int validator:", validate_positive_int, ["abc", "-5", "42"])
grade = get_valid_input("Testing grade validator:", validate_grade, ["150", "abc", "85"])
assert positive == 42 and grade == 85
assert validate_name("  Elif   Demir ") == (True, "Elif Demir")
assert not validate_name("Elif 42")[0]
assert get_valid_input("Exhausted:", validate_grade, ["bad"]) is None
assert get_valid_input("One attempt:", validate_grade, ["bad", "85"], max_tries=1) is None
````

## Exercise 12: Robust Data Processor (Challenge)

Write a function `process_student_records(records)` that takes a list of dictionaries containing student data with potentially messy/missing fields.

Each record should have: `name`, `id`, `midterm`, `final`

The function should:
- Skip records with missing required fields (print which fields are missing)
- Convert scores to numbers (handle invalid values)
- Validate scores are 0-100
- Calculate weighted average (midterm 40%, final 60%)
- Return a list of processed records and a summary

**Test data:**
```python
records = [
    {"name": "Ayse Kara", "id": "2024001", "midterm": "85", "final": "90"},
    {"name": "Berk Yilmaz", "id": "2024002", "midterm": "abc", "final": "75"},
    {"id": "2024003", "midterm": "70", "final": "80"},
    {"name": "Ceren Demir", "id": "2024004", "midterm": "92", "final": "150"},
    {"name": "Doruk Ozkan", "id": "2024005", "midterm": "78", "final": "82"},
]
```

**Expected output (approximate):**
```
Processing 5 records...
✓ Ayse Kara: 88.0 (BA)
✗ Berk Yilmaz: invalid midterm score 'abc'
✗ Record 3: missing field 'name'
✗ Ceren Demir: final score 150 out of range
✓ Doruk Ozkan: 80.4 (BA)

Summary: 2 processed, 3 errors
```



**Finite-number rule:** after numeric conversion, reject `nan`, `inf`, and `-inf` with `math.isfinite()`. For an average with no valid numbers, return the documented empty result. Boolean values are flags, not measurements, in these validation exercises.

### Worked reasoning

Check required keys before reading them. Validate midterm and final independently, keeping the first clear rejection reason for each record. Compute 0.4×85+0.6×90=88.0 and 0.4×78+0.6×82=80.4 only after both scores pass. Reject NaN and infinity before range checks. Return accepted records together with counts and rejection details.

**Türkçe:** Eksik anahtar, dönüşüm hatası ve aralık hatası farklı sorunlardır; hiçbiri sonraki kaydın işlenmesini engellememeli.

**Expected check:** Two records accepted, three rejected; accepted averages 88.0 and 80.4, both BA.


> **Example data only:** These fictional calculation weights are for this programming exercise. The course grade uses only the midterm (50%) and final (50%).


````python
def exercise_letter(score):
    for boundary, letter in [(90,"AA"),(80,"BA"),(70,"BB"),(60,"CB"),(50,"CC")]:
        if score >= boundary:
            return letter
    return "FF"

def process_student_records(records):
    processed, errors = [], []
    required = ["name", "id", "midterm", "final"]
    for index, record in enumerate(records, start=1):
        if not isinstance(record, dict):
            errors.append(f"Record {index}: expected a dictionary")
            continue
        missing = [key for key in required if key not in record]
        if missing:
            errors.append(f"Record {index}: missing field(s) {', '.join(missing)}")
            continue
        scores = {}
        error = None
        for field in ["midterm", "final"]:
            try:
                if isinstance(record[field], bool):
                    raise TypeError("boolean is not a grade")
                number = float(record[field])
            except (ValueError, TypeError):
                error = f"{record['name']}: invalid {field} score {record[field]!r}"
                break
            if not math.isfinite(number):
                error = f"{record['name']}: {field} score must be finite"
                break
            if not 0 <= number <= 100:
                error = f"{record['name']}: {field} score {number:g} out of range"
                break
            scores[field] = number
        if error:
            errors.append(error)
            continue
        average = 0.4 * scores["midterm"] + 0.6 * scores["final"]
        processed.append({"name":record["name"], "id":record["id"],
                          "average":average, "letter":exercise_letter(average)})
    summary = {"processed":len(processed), "errors":len(errors), "details":errors}
    return processed, summary

records = [
    {"name":"Ayse Kara","id":"2024001","midterm":"85","final":"90"},
    {"name":"Berk Yilmaz","id":"2024002","midterm":"abc","final":"75"},
    {"id":"2024003","midterm":"70","final":"80"},
    {"name":"Ceren Demir","id":"2024004","midterm":"92","final":"150"},
    {"name":"Doruk Ozkan","id":"2024005","midterm":"78","final":"82"},
]
processed, summary = process_student_records(records)
for row in processed:
    print(f"✓ {row['name']}: {row['average']:.1f} ({row['letter']})")
for error in summary["details"]:
    print("✗", error)
print(f"Summary: {summary['processed']} processed, {summary['errors']} errors")
assert [round(row["average"],1) for row in processed] == [88.0,80.4]
assert (summary["processed"], summary["errors"]) == (2,3)
nan_record = {"name":"Test", "id":"1", "midterm":"nan", "final":"90"}
assert process_student_records([nan_record])[1]["errors"] == 1
assert process_student_records([])[1]["processed"] == 0
````

Predict and catch two file-related exceptions without stopping the notebook:

   `FileNotFoundError`. A file must exist before it can be read.
2. Demonstrate a denied write **by simulation**: `raise PermissionError("simulated denied write")`
   inside `try/except PermissionError`. Do not try to write to system files.

**Expected output:** `FileNotFoundError: example file is absent`, then
`PermissionError: simulated denied write`.

The second exception is deliberately raised for a repeatable lesson: operating-system
permissions differ between computers. Next week you will read and write your own files.

**Türkçe:** Dosya yokluğu ile erişim izni hatasını ayrı yakalayın. İzin hatasını
göstermek için sistem dosyasını değiştirmek gerekmez; burada hatayı kontrollü üretiriz.

### Worked reasoning

Use a newly created temporary directory to guarantee the missing example file really is absent. Simulate PermissionError explicitly because filesystem permissions vary across computers. Both demonstrations are deterministic and avoid protected system files.

**Türkçe:** Hata türünü öğrenmek için gerçek bir sistem dosyasını riske atmaya gerek yoktur.

**Expected check:** Both expected file-error messages print and execution continues.


````python
from pathlib import Path
from tempfile import TemporaryDirectory

caught = []
with TemporaryDirectory(prefix="cp1_file_errors_") as folder:
    try:
        with open(Path(folder) / "missing_example.txt", "r", encoding="utf-8") as file:
            file.read()
    except FileNotFoundError:
        print("FileNotFoundError: example file is absent")
        caught.append("missing")

try:
    raise PermissionError("simulated denied write")
except PermissionError as error:
    print(f"PermissionError: {error}")
    caught.append("permission")

assert caught == ["missing", "permission"]
````

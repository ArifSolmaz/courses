## Exercise 1: Word Counter (Easy)

Count the number of words in the given sentence.

### Reasoning and plan

Split on whitespace, then count the resulting words. With no separator argument, split handles repeated spaces and tabs together and discards leading/trailing whitespace. Empty text therefore has zero words.

**Türkçe:** split boşlukları ayırıcı kabul eder. Birden fazla boşluk yeni boş kelimeler üretmez; boş metnin kelime sayısı sıfırdır.


````python
# ✏️ [EX1] — worked solution
sentence = 'Python is a great language to learn'
words = sentence.split()
print(f'The sentence has {len(words)} words.')
assert len(words) == 7
assert len('  Python   is	fun  '.split()) == 3
assert ''.split() == []
````

### Expected result

```text
The sentence has 7 words.
```

## Exercise 2: Vowel Counter (Easy)

Count the number of vowels (a, e, i, o, u) in the given text. Count both uppercase and lowercase.

### Reasoning and plan

Visit each character, convert that character to lowercase, and test membership in the specified English vowel set. Do not count y or accented letters for this exercise. Every occurrence counts separately.

**Türkçe:** Büyük/küçük harf farkını kaldırıp verilen beş sesli harfi ararız. Her tekrar ayrı sayılır.


````python
# ✏️ [EX2] — worked solution
text = 'Istanbul is a beautiful city'
count = 0
for character in text:
    if character.lower() in 'aeiou':
        count += 1
print(f'The text has {count} vowels.')
assert count == 11
for sample, expected in [('AEIOU', 5), ('rhythm', 0), ('', 0)]:
    actual = 0
    for character in sample:
        if character.lower() in 'aeiou':
            actual += 1
    assert actual == expected
````

### Expected result

```text

```

## Exercise 3: Reverse a String (Medium)

Reverse the given string **without** using slicing (`[::-1]`). Use a loop instead.

### Reasoning and plan

Start with an empty result and place each new character in front of it. For P,r,o the states are P, rP, orP. This reverses the order without slicing. Strings are immutable, so each addition makes a new string.

**Türkçe:** Her yeni karakteri sonucun başına ekleriz. Metinler yerinde değişmez; her adımda yeni metin oluşur.


````python
# ✏️ [EX3] — worked solution
text = 'Programming'
reverse_text = ''
for character in text:
    reverse_text = character + reverse_text
print('Original:', text)
print('Reversed:', reverse_text)
assert reverse_text == 'gnimmargorP'
for sample, expected in [('', ''), ('A', 'A'), ('ab', 'ba')]:
    result = ''
    for character in sample:
        result = character + result
    assert result == expected
````

### Expected result

```text
Original: Programming
Reversed: gnimmargorP
```

## Exercise 4: Palindrome Checker (Medium)

Check if the given word is a palindrome (reads the same forwards and backwards). Ignore case.

### Reasoning and plan

Normalize case, then compare with the reverse. Only case is ignored in the stated problem: spaces and punctuation remain significant. The empty string and a single character equal their own reverse.

**Türkçe:** Yalnızca harf büyüklüğünü yok sayarız. Boşluk ve noktalama bu soruda silinmez.


````python
# ✏️ [EX4] — worked solution
word = 'Racecar'
normalized = word.lower()
palindrome = normalized == normalized[::-1]
print(f"'{word}' is " + ('a palindrome.' if palindrome else 'not a palindrome.'))
assert palindrome is True
for sample, expected in [('python', False), ('Madam', True), ('A', True), ('', True), ('ab a', False)]:
    cleaned = sample.lower()
    assert (cleaned == cleaned[::-1]) == expected
````

### Expected result

```text
'Racecar' is a palindrome.
```

## Exercise 5: Capitalize Words (Medium)

Capitalize the first letter of each word in the sentence **without** using `title()`. Use `split()`, `join()`, and string indexing.

### Reasoning and plan

Split into words, capitalize the first character of each, then join with one space. split guarantees there are no empty words to index. Lowercase the remaining characters to reproduce title-like capitalization without title().

**Türkçe:** Önce kelimeleri ayırırız; ilk harfi büyütüp kalanını küçültürüz. join kelimelerin arasına tek boşluk koyar.


````python
# ✏️ [EX5] — worked solution
sentence = 'hello world from python'
capitalized = []
for word in sentence.split():
    capitalized.append(word[0].upper() + word[1:].lower())
result = ' '.join(capitalized)
print(result)
assert result == 'Hello World From Python'
for sample, expected in [('  hELLO   wORLD ', 'Hello World'), ('', ''), ('a', 'A')]:
    words = []
    for word in sample.split():
        words.append(word[0].upper() + word[1:].lower())
    assert ' '.join(words) == expected
````

### Expected result

```text
Hello World From Python
```

## Exercise 6: CSV Line Parser (Medium)

Parse the given CSV line and display each field with its label.

### Reasoning and plan

Split the simple unquoted CSV into exactly five fields. Their positions are name, age, city, major, GPA. Convert age and GPA when treating them as numbers; labels and names stay text. Quoted fields containing commas need the csv module rather than this simplified split.

## Exercise 7: Character Frequency (Medium)

Count how many times each letter appears in the string. Ignore spaces and be case-insensitive. Print only letters that appear.

### Reasoning and plan

Scan normalized text in first-appearance order. Use a seen string to avoid printing a letter twice; count its occurrences in the full normalized text. Skip nonletters. In Hello Parsle, p occurs once, while e occurs twice and l three times.

**Türkçe:** Aynı harfi iki kez yazdırmamak için görülen harfleri saklarız. Harfin toplam tekrarını metnin tamamında sayarız.


````python
# ✏️ [EX7] — worked solution
text = 'Hello Parsle'
normalized = text.lower()
seen = ''
letters = []
counts = []
for character in normalized:
    if character.isalpha() and character not in seen:
        seen += character
        count = normalized.count(character)
        letters.append(character)
        counts.append(count)
        print(f'{character}: {count}')
assert letters == ['h', 'e', 'l', 'o', 'p', 'a', 'r', 's']
assert counts == [1, 2, 3, 1, 1, 1, 1, 1]
assert sum(counts) == 11
````

### Expected result

```text

e: 2
l: 3
o: 1
p: 1
a: 1
r: 1
s: 1
```

## Exercise 8: Acronym Generator (Medium)

Generate an acronym from the given phrase by taking the first letter of each word and converting to uppercase.

## Exercise 9: Replace Digits (Medium)

Replace every digit in the string with the `#` character.

### Reasoning and plan

Build a new string, replacing each digit with # and copying every other character unchanged. Hyphens and spaces retain their positions. isdigit follows Python’s Unicode digit definition; the supplied phone number uses ordinary ASCII digits.

**Türkçe:** Rakamları # ile değiştirirken diğer karakterleri aynen kopyalarız. Böylece metnin uzunluğu ve biçimi korunur.


````python
# ✏️ [EX9] — worked solution
text = 'My phone is 0532-123-4567'
masked = ''
for character in text:
    if character.isdigit():
        masked += '#'
    else:
        masked += character
print('Original:', text)
print('Masked:  ', masked)
assert masked == 'My phone is ####-###-####'
assert len(masked) == len(text)
assert masked.count('#') == 11
````

### Expected result


## Exercise 10: Email Validator (Medium)

Check if the given email address passes these basic rules:
1. Contains exactly one `@` symbol
2. Has at least one character before `@`
3. Has at least one `.` after `@`
4. Does not contain spaces

### Reasoning and plan

Implement the four specified rules in sequence: exactly one @, a nonempty part before it, a dot after it, and no spaces. This is a classroom format check, not proof that a mailbox exists or a complete email standard validator. Keep the stated rules visible rather than adding hidden ones.

**Türkçe:** Dört basit koşulu ayrı kontrol ederiz. Bu çalışma adresin gerçek veya ulaşılabilir olduğunu kanıtlamaz.


````python
# ✏️ [EX10] — worked solution
emails = ['emre.kaya@university.edu.tr', 'invalid-email', 'no spaces@mail.com']
expected = [True, False, False]
for index in range(len(emails)):
    email = emails[index]
    at_position = email.find('@')
    valid = (email.count('@') == 1
             and at_position > 0
             and '.' in email[at_position + 1:]
             and ' ' not in email)
    print(f'{email} -> {"Valid" if valid else "Invalid"}')
    assert valid == expected[index]
for email, expected_valid in [('@mail.com', False), ('a@@b.com', False), ('a@domain', False), ('a@b.c', True)]:
    at_position = email.find('@')
    valid = (email.count('@') == 1 and at_position > 0
             and '.' in email[at_position + 1:] and ' ' not in email)
    assert valid == expected_valid
````

### Expected result

```text
emre.kaya@university.edu.tr -> Valid
invalid-email -> Invalid
no spaces@mail.com -> Invalid
```

The assertions also check the cases named in the code.


## Exercise 11: Log Parser (Challenge)

Parse the log lines below. Each line has the format `timestamp,sensor_name,value`. Find:
- The number of readings
- The average value across all readings
- The sensor with the highest reading

### Reasoning and plan

Split the text into rows, then split each row into timestamp, sensor and numeric value. Convert the value to float before comparison. Keep the sensor and timestamp whenever the maximum changes. The last reading 45.8 exceeds the earlier 45.2; total 157.3 divided by 5 gives 31.46.

**Türkçe:** En büyük değer değişince sensör ve zaman etiketi de değişmelidir. Son satırdaki 45,8 önceki 45,2’den büyüktür.


````python
# ✏️ [EX11] — worked solution
log_data = """2025-01-15 08:30:15,sensor_A,23.5
2025-01-15 08:30:16,sensor_B,45.2
2025-01-15 08:30:17,sensor_A,24.1
2025-01-15 08:30:18,sensor_C,18.7
2025-01-15 08:30:19,sensor_B,45.8"""
total = 0.0
count = 0
highest = float('-inf')
highest_sensor = highest_time = ''
for line in log_data.splitlines():
    fields = line.split(',')
    assert len(fields) == 3
    value = float(fields[2])
    total += value
    count += 1
    if value > highest:
        highest = value
        highest_sensor = fields[1]
        highest_time = fields[0]
print('Total readings:', count)
print(f'Average value: {total / count:.2f}')
print(f'Highest reading: {highest_sensor} with {highest}')
print('Highest timestamp:', highest_time)
assert count == 5 and abs(total - 157.3) < 1e-9
assert (highest, highest_sensor, highest_time) == (45.8, 'sensor_B', '2025-01-15 08:30:19')
````

### Expected result

```text
Total readings: 5
Average value: 31.46
Highest reading: sensor_B with 45.8
Highest timestamp: 2025-01-15 08:30:19
```

## Exercise 12: Caesar Cipher (Challenge)

Implement a Caesar cipher that shifts each letter by a given number of positions. Only shift letters (a-z, A-Z), leave other characters unchanged. Wrap around (z shifted by 1 becomes a).

### Reasoning and plan

Handle uppercase and lowercase using separate base codes. Subtract the base to get position 0–25, add the shift, wrap using %26, then add the base back. For z+3: (25+3)%26=2, giving c. Only A–Z/a–z are shifted; spaces, punctuation and other characters pass through.

**Türkçe:** Harf konumunu 0–25 aralığına çevirip kaydırırız. %26 alfabenin sonundan başına dönmeyi sağlar.


````python
# ✏️ [EX12] — worked solution
cases = [('Hello, World!', 3, 'Khoor, Zruog!'),
         ('Zz Aa!', 1, 'Aa Bb!'),
         ('Abc', -1, 'Zab'),
         ('Çığ 123!', 26, 'Çığ 123!'),
         ('', 3, '')]
for message, shift, expected in cases:
    encrypted = ''
    for character in message:
        if 'a' <= character <= 'z':
            base = ord('a')
            position = ord(character) - base
            encrypted += chr((position + shift) % 26 + base)
        elif 'A' <= character <= 'Z':
            base = ord('A')
            position = ord(character) - base
            encrypted += chr((position + shift) % 26 + base)
        else:
            encrypted += character
    print(f'{message!r}, shift {shift}: {encrypted!r}')
    assert encrypted == expected
````

### Expected result

```text
'Hello, World!', shift 3: 'Khoor, Zruog!'
'Zz Aa!', shift 1: 'Aa Bb!'
'Abc', shift -1: 'Zab'
'Çığ 123!', shift 26: 'Çığ 123!'
'', shift 3: ''
```

The assertions also check the cases named in the code.


## Bridge Exercise: Preview of Functions

Next week, we will learn about **functions**. A function is a reusable block of code that performs a specific task.

Try converting your palindrome checker (EX4) into a function. Don't worry if the syntax is new — just give it a try!

### Reasoning and plan

Move the palindrome calculation inside a named function. The parameter receives one word; return sends a Boolean back to the caller. Printing belongs in the outer loop so the same result can also be checked or reused elsewhere.

**Türkçe:** Hesabı fonksiyon içine taşırız. return mantıksal sonucu verir; yazdırma işlemi çağıran tarafta yapılır.


````python
# ✏️ [EXBridge] — worked solution
def is_palindrome(word):
    normalized = word.lower()
    return normalized == normalized[::-1]

words = ['racecar', 'python', 'madam']
for word in words:
    label = 'Palindrome' if is_palindrome(word) else 'Not a palindrome'
    print(f'{word} -> {label}')
assert is_palindrome('Racecar') is True
assert is_palindrome('python') is False
assert is_palindrome('') is True
````

### Expected result

```text
racecar -> Palindrome
python -> Not a palindrome
madam -> Palindrome
```

## Exercise 2: Area of Circle (Easy)

Write a function `circle_area(radius)` that takes a non-negative radius and **returns** the area of a circle. Use `3.14159` for pi.

### Reasoning and plan

Use the specified approximation of pi. Square the radius, multiply by pi, and return the full numeric result. Formatting to two decimals belongs to the caller. Radius is non-negative; doubling it quadruples the area.

**Türkçe:** Yarıçapın karesi pi ile çarpılır. Fonksiyon tam sonucu verir; iki basamaklı gösterim dışarıda yapılır.

## Exercise 3: Wheel Distance Function (Easy)

Write a function `wheel_distance(rotations, diameter_mm)` that returns the distance a wheel travels in metres, rounded to 2 decimals. Use `pi = 3.14159`. Formula: `distance = rotations * pi * diameter_mm / 1000`

Assume rolling without slipping, a positive diameter, and a non-negative number of rotations. This estimates path length, not signed displacement.

### Reasoning and plan

One rotation moves the wheel by one circumference, pi times the diameter. Divide by 1000 to turn millimetres into metres, then round for a readable result. Zero rotations is a useful check: the function must return 0.0, not fail. A 1000 mm wheel turned once travels approximately pi metres; this function returns 3.14 m after rounding.

**Türkçe:** Bir tur, tekerleği bir çevre kadar ilerletir. mm'yi metreye çevirmek için 1000'e böl; sonucu iki basamağa yuvarla.


````python
def wheel_distance(rotations, diameter_mm):
    pi = 3.14159
    return round(rotations * pi * diameter_mm / 1000, 2)
for rotations, diameter_mm in [(100, 65), (0, 65), (250, 120)]:
    print(f'{rotations} rotations of a {diameter_mm} mm wheel = {wheel_distance(rotations, diameter_mm)} m')
assert wheel_distance(100, 65) == 20.42
assert wheel_distance(250, 120) == 94.25
assert wheel_distance(0, 65) == 0.0
assert wheel_distance(1, 1000) == 3.14
````

## Exercise 4: Max of Two (Medium)

Write a function `max_of_two(a, b)` that returns the larger of two numbers **without** using the built-in `max()` function.

### Reasoning and plan

Compare the two values once. If a is at least b, return a; otherwise return b. Equality belongs to the first branch, which still returns the correct value. Initializing a candidate to zero is unnecessary and would complicate negative inputs.

**Türkçe:** Tek karşılaştırma yeterlidir. Eşitlikte birinci değer döndürülür; sonuç yine aynıdır.


````python
def max_of_two(a, b):
    if a >= b:
        return a
    return b
for a, b, expected in [(10, 20, 20), (50, 30, 50), (7, 7, 7)]:
    result = max_of_two(a, b)
    print(f'max_of_two({a}, {b}) = {result}')
    assert result == expected
assert max_of_two(-8, -3) == -3
````

## Exercise 6: Factorial Function (Medium)

Write a function `factorial(n)` that returns the factorial of n using a loop. Factorial of n is: n! = 1 * 2 * 3 * ... * n. Factorial of 0 is 1.

### Reasoning and plan

Use 1 as the multiplication identity. After factors 1,2,3,4,5, the accumulator is 1,2,6,24,120. For n=0, the loop runs zero times, leaving 1. This exercise assumes a non-negative integer; invalid types and negative values require a separate validation policy.

**Türkçe:** Çarpmanın başlangıç değeri 1’dir. n sıfırken döngü çalışmaz ve 0! = 1 sonucu korunur.


````python
def factorial(n):
    result = 1
    for factor in range(1, n + 1):
        result *= factor
    return result
for n in [0, 5, 7]:
    print(f'factorial({n}) = {factorial(n)}')
assert factorial(0) == 1 and factorial(1) == 1
assert factorial(5) == 120 and factorial(7) == 5040
assert factorial(6) == 6 * factorial(5)
````

- **Line 1 —** `` def factorial(n): ``

- **Line 2 —** `` result = 1 ``

  Initialize the product to one, the multiplication identity.

- **Line 3 —** `` for factor in range(1, n + 1): ``

  Visit integers 1 through n inclusive. When n is zero, the loop has no passes.

- **Line 4 —** `` result *= factor ``

  Multiply the running result by the current factor.

- **Line 5 —** `` return result ``

  Return the completed product, leaving 0! equal to the initial one.

- **Line 6 —** `` for n in [0, 5, 7]: ``

  Test zero and two positive inputs.

- **Line 7 —** `` print(f'factorial({n}) = {factorial(n)}') ``

  Call and print the factorial for each test input.

## Exercise 7: Is Prime (Medium)

Write a function `is_prime(n)` that returns `True` if n is a prime number, `False` otherwise. Assume an integer input. A prime number is greater than 1 and has exactly two positive divisors: 1 and itself.

### Reasoning and plan

Reject integers below 2. Test possible divisors from 2 up to n−1; finding even one divisor is enough to return False. Return True only after the loop finishes without finding any. For 2 the divisor loop is empty, so it correctly returns True.

**Türkçe:** Bir bölen bulmak asal olmadığını kanıtlar. Hiç bölen bulunmadıysa döngü bittikten sonra True döndürülür.


````python
def is_prime(n):
    if n < 2:
        return False
    for divisor in range(2, n):
        if n % divisor == 0:
            return False
    return True
for n in [2, 7, 10, 1]:
    print(f'is_prime({n}) = {is_prime(n)}')
for n, expected in [(2, True), (7, True), (10, False), (1, False), (0, False), (-3, False), (9, False), (29, True)]:
    assert is_prime(n) == expected
````

  Return False immediately for those values.

- **Line 4 —** `` for divisor in range(2, n): ``

  Try every integer divisor from 2 up to but excluding n. This is a straightforward teaching implementation rather than a square-root-bounded search.

- **Line 5 —** `` if n % divisor == 0: ``

  Check whether the candidate divides n exactly.

- **Line 6 —** `` return False ``

  Return False as soon as a divisor proves n composite.

- **Line 7 —** `` return True ``

## Exercise 8: Power Function (Medium)

Write a function `power(base, exponent)` that calculates base raised to the exponent **without** using the `**` operator. Use a loop.

### Reasoning and plan

For a non-negative integer exponent, repeated multiplication is enough. Start at 1 and multiply by base exactly exponent times. For base=3 and exponent=4, the trace is 1→3→9→27→81. Zero repetitions leave 1; negative or fractional exponents are outside the stated contract.

**Türkçe:** Taban, üs kadar kez çarpılır. Üs sıfırsa hiç çarpma yapılmaz ve başlangıç değeri 1 kalır.


````python
def power(base, exponent):
    result = 1
    for repetition in range(exponent):
        result *= base
    return result
for base, exponent in [(2, 10), (3, 4), (5, 0)]:
    print(f'power({base}, {exponent}) = {power(base, exponent)}')
assert power(2, 10) == 1024 and power(3, 4) == 81
assert power(5, 0) == 1 and power(0, 0) == 1
assert power(-2, 3) == -8 and power(-2, 4) == 16
assert power(0, 3) == 0 and power(0.5, 2) == 0.25
````

- **Line 3 —** `` for repetition in range(exponent): ``

- **Line 4 —** `` result *= base ``

  Multiply the running result by base.

- **Line 5 —** `` return result ``

## Exercise 9: Count Vowels Function (Medium)

Write a function `count_vowels(text)` that takes a string and **returns** the number of vowels (the English letters a, e, i, o, u — case-insensitive; not a Turkish vowel classifier).

### Reasoning and plan

Each call starts its own counter at zero. Check characters against the English vowel set after lowercasing; return the final count after the loop. Putting return inside the loop would stop after the first character.

**Türkçe:** Sayaç her çağrıda yeniden başlar. return döngünün dışında olmalıdır; içeride olursa ilk harften sonra işlem biter.


````python
def count_vowels(text):
    count = 0
    for character in text:
        if character.lower() in 'aeiou':
            count += 1
    return count
for text in ['Hello World', 'Python Programming', 'AEIOU']:
    print(f'count_vowels({text!r}) = {count_vowels(text)}')
assert count_vowels('Hello World') == 3
assert count_vowels('Python Programming') == 4
assert count_vowels('AEIOU') == 5
assert count_vowels('') == 0 and count_vowels('rhythm') == 0
````

## Exercise 10: List Average (Medium)

Write a function `average(numbers)` that takes a list of numbers and **returns** their average. Handle the case where the list is empty (return 0). This is an exercise convention, not a measured zero: an empty average is mathematically undefined. In a sensor report, use a separate missing-data indication.

### Reasoning and plan

Check for an empty list before dividing. Return zero there because the exercise explicitly chooses that convention; mathematically an empty average is undefined. Otherwise divide the total by the length. A one-item list averages to its only value.

**Türkçe:** Boş liste için sorunun seçtiği sonuç 0’dır; bu matematiksel ortalamanın tanımlı olduğu anlamına gelmez. Bölmeden önce boşluk kontrol edilir.

## Exercise 11: Dynamic Pressure Function (Medium)

Write a function `dynamic_pressure(density, velocity)` that returns the dynamic pressure of a moving fluid in pascals as a float:

Rules:

- For a valid density, a velocity of 0 returns `0.0`; the negative-density check takes priority.
- Assume finite numeric inputs. A signed velocity is allowed because it is squared; this is dynamic pressure, not pipe pressure loss.

Call it for air (density 1.225) at 10 and 20 m/s and for water (density 1000) at 2 m/s.

### Reasoning and plan

Validate the input first so an impossible density never reaches the formula. Then return the computed float; do not print inside the function, so the caller decides the format. Because velocity is squared, doubling it multiplies the pressure by four (61.25 → 245.0); this is an independent check that does not rely on the printed numbers. Zero velocity is the natural edge case and gives 0.0.

**Türkçe:** Önce girdiyi doğrula, sonra sayıyı döndür; yazdırmayı çağıran tarafa bırak. Hız iki katına çıkınca basınç dört katına çıkar.


````python
def dynamic_pressure(density, velocity):
    if density < 0:
        return None
    return 0.5 * density * velocity ** 2
for fluid, density, velocity in [('Air', 1.225, 10.0), ('Air', 1.225, 20.0), ('Water', 1000, 2.0)]:
    pressure = dynamic_pressure(density, velocity)
    print(f'{fluid} at {velocity} m/s: {pressure:.2f} Pa')
assert abs(dynamic_pressure(1.225, 10.0) - 61.25) < 1e-9
assert dynamic_pressure(1000, 2.0) == 2000.0
assert abs(dynamic_pressure(1.225, 20.0) - 4 * dynamic_pressure(1.225, 10.0)) < 1e-9
assert dynamic_pressure(1.225, 0) == 0.0
assert dynamic_pressure(-1, 5) is None
````

## Exercise 12: Mini Calculator (Challenge)

Create a mini calculator with separate functions for addition, subtraction, multiplication, and division. Then write a `calculate(num1, op, num2)` function that takes two numbers and an operator string (`"+"`, `"-"`, `"*"`, `"/"`) and returns the result.

Return `"Error: Division by zero"` for a zero denominator and `"Error: Unknown operator"` for an unsupported operator. Otherwise return a number. The caller must distinguish error strings from numeric results before further arithmetic.

### Reasoning and plan

Each arithmetic helper performs one operation. The dispatcher selects the appropriate helper and returns its result to the caller. Division checks its denominator before dividing; an unsupported operator returns a separate message. This keeps calculation separate from displaying results.

**Türkçe:** Her yardımcı fonksiyon tek işlem yapar. Ana fonksiyon operatöre göre doğru yardımcıyı çağırır; sıfıra bölme önceden kontrol edilir.


````python
def add(a, b):
    return a + b
def sub(a, b):
    return a - b
def mul(a, b):
    return a * b
def div(a, b):
    if b == 0:
        return 'Error: Division by zero'
    return a / b
def calculate(num1, op, num2):
    if op == '+':
        return add(num1, num2)
    if op == '-':
        return sub(num1, num2)
    if op == '*':
        return mul(num1, num2)
    if op == '/':
        return div(num1, num2)
    return 'Error: Unknown operator'
for op in ['+', '-', '*', '/']:
    print(f'10 {op} 5 = {calculate(10, op, 5)}')
print(f'10 / 0 = {calculate(10, "/", 0)}')
assert calculate(10, '+', 5) == 15 and calculate(10, '-', 5) == 5
assert calculate(10, '*', 5) == 50 and calculate(10, '/', 5) == 2.0
assert calculate(10, '/', 0) == 'Error: Division by zero'
assert calculate(0, '/', 5) == 0 and calculate(2, '%', 3) == 'Error: Unknown operator'
````

## Exercise 3: Motor Temperature Rise (Medium)

### Task and supplied example


A motor is tested for one hour. Its housing temperature is 24.5 °C at the start and 61.3 °C at the end.

Store both values in variables. Calculate the temperature rise ΔT and the end temperature in kelvin.

**Formula:** K = °C + 273.15

**Expected output:**
```
Start: 24.5 C
End: 61.3 C
Rise: 36.8 C

```

<details>
<summary>💡 Hint</summary>
Subtract start from end for the rise. Use <code>round(value, 1)</code> so that 36.800000000000004 is shown as 36.8.
</details>

### Reasoning

The rise is a difference: end minus start. Kelvin is the same scale with a shifted zero, so add 273.15. Floating-point subtraction can give 36.800000000000004; `round` only changes the display, not the physics.

**TR:** Sıcaklık artışı bitiş eksi başlangıçtır. Kelvin için 273.15 ekle. `round` yalnızca gösterimi düzenler.

- **Line 4 —** `` end_k = end_c + 273.15 ``

  Add 273.15 to the final Celsius temperature to express the same temperature in kelvin.

- **Line 5 —** `` print("Start:", start_c, "C") ``

## Exercise 5: Compound Interest (Medium)

### Task and supplied example


Ali deposits 10,000 TL in a bank with 15% annual interest. Calculate the amount after 3 years.

**Formula:** A = P × (1 + r) ^ t
- P = principal (10000)
- r = annual interest rate (0.15)
- t = time in years (3)

**Expected output:**
```
Principal: 10000 TL
Rate: 15 %
Time: 3 years
Final Amount: 15208.75 TL
```

<details>
<summary>💡 Hint</summary>
Use <code>**</code> for the exponent. The rate should be 0.15, not 15.
</details>

### Reasoning

Use rate 0.15, so each year multiplies the amount by 1.15. Three years means 1.15**3, then multiply by the principal. Round only the displayed amount; tiny floating-point tails are not a different interest calculation.

## Exercise 6: Belt Speed from Pulley RPM (Medium)

### Task and supplied example


A motor turns a pulley at 1450 rpm. The pulley diameter is 120 mm. Calculate the belt speed in m/s.

**Formulas:**
- circumference (m) = pi × diameter_mm / 1000
- belt speed (m/s) = circumference × rpm / 60

Use `pi = 3.14159`.

**Expected output:**
```
RPM: 1450
Pulley diameter: 120 mm
Circumference: 0.377 m
Belt speed: 9.11 m/s
```

<details>
<summary>💡 Hint</summary>
Convert millimetres to metres first (divide by 1000). One minute has 60 seconds, so divide by 60 to get "per second". Use <code>round(x, 3)</code> and <code>round(x, 2)</code> for display.
</details>

### Reasoning

One turn of the pulley moves the belt by one circumference. Convert the diameter to metres before multiplying, and convert "per minute" to "per second" by dividing by 60. Keep the full value in the variable; round only when printing.

**TR:** Bir tur, kayışı bir çevre kadar taşır. Önce mm'yi metreye, sonra dakikayı saniyeye çevir.

````python
rpm = 1450
diameter_mm = 120
pi = 3.14159
circumference = pi * diameter_mm / 1000
belt_speed = circumference * rpm / 60
print("RPM:", rpm)
print("Pulley diameter:", diameter_mm, "mm")
print("Circumference:", round(circumference, 3), "m")
print("Belt speed:", round(belt_speed, 2), "m/s")
````

- **Line 3 —** `` pi = 3.14159 ``

  Set `pi` to the decimal number `3.14159`. This stores the result without displaying it.

- **Line 4 —** `` circumference = pi * diameter_mm / 1000 ``

  Multiply π by the pulley diameter to get circumference in millimetres, then divide by 1000 to convert to metres.

- **Line 5 —** `` belt_speed = circumference * rpm / 60 ``

  Multiply metres per revolution by revolutions per minute, then divide by 60 for metres per second; the model assumes no belt slip.

- **Line 6 —** `` print("RPM:", rpm) ``

  Print the text `'RPM:'`; then the current value of `rpm`. Separate the arguments with spaces. Finish with a newline.

- **Line 7 —** `` print("Pulley diameter:", diameter_mm, "mm") ``

  Print the text `'Pulley diameter:'`; then the current value of `diameter_mm`; then the text `'mm'`. Separate the arguments with spaces. Finish with a newline.

- **Line 8 —** `` print("Circumference:", round(circumference, 3), "m") ``

- **Line 9 —** `` print("Belt speed:", round(belt_speed, 2), "m/s") ``

  Print the text `'Belt speed:'`; then `belt_speed` rounded to `2` decimal places; then the text `'m/s'`. Separate the arguments with spaces. Finish with a newline.

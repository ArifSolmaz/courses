## Exercise 1: Create and Read (Easy)

Write a program that:
1. Creates a file called `"my_note.txt"` with 3 lines of your choice
2. Reads the file back and prints its contents

**Expected output:**
```text
I love Python programming.
Files are really useful.
This is my third line.
```

### Worked reasoning

Write three lines, including separators, then close the file before reading it. A with block closes the file even if an exception occurs. Compare the read-back text with the intended text to check persistence, not just the printed output.

**Türkçe:** Yazılan içeriği yeniden okuyup karşılaştırmak dosyanın gerçekten doğru kaydedildiğini doğrular.

**Expected check:** Three lines are printed and match the persisted text exactly.


````python
lines = ["I love Python programming.", "Files are really useful.", "This is my third line."]
text = "\n".join(lines) + "\n"
with open("my_note.txt", "w", encoding="utf-8") as file:
    file.write(text)
with open("my_note.txt", "r", encoding="utf-8") as file:
    saved = file.read()
print(saved, end="")
assert saved == text
assert len(saved.splitlines()) == 3
````

**Türkçe:** Dosyadan gelen değerler metindir; toplamadan önce sayıya dönüştürülür.

First run the cell below to create a file. Then write a program that asks for a search keyword and prints all lines containing that keyword (case-insensitive).

**Expected output:**
```text
Enter keyword: python

  Line 1: Python is a popular programming language.
  Line 4: Many beginners choose Python as their first language.
```

### Worked reasoning

Use a preset keyword instead of live input so the example is repeatable. Compare lowercase versions while retaining the original line for display. enumerate(start=1) gives human-readable line numbers. Collect matches before reporting their count.

**Türkçe:** Karşılaştırmayı küçük harflerle yapın; ekranda özgün metni ve 1’den başlayan satır numarasını koruyun.

**Expected check:** Two matching lines, numbered 1 and 4; matching is case-insensitive.


**Contract:** Use a non-empty keyword. Matching uses Python lowercasing, not Turkish locale-specific matching. The worked answer uses a supplied keyword for repeatability.


````python
with open('article.txt', "w", encoding="utf-8") as file:
    file.write("""Python is a popular programming language.
It was created by Guido van Rossum.
Java and C++ are also widely used.
Many beginners choose Python as their first language.
Programming skills are in high demand.
""")
def search_file(filename, keyword):
    if not keyword:
        raise ValueError("Use a non-empty keyword")
    matches = []
    with open(filename, "r", encoding="utf-8") as file:
        for number, line in enumerate(file, start=1):
            if keyword.lower() in line.lower():
                matches.append((number, line.rstrip("\n")))
    return matches
keyword = "python"
matches = search_file("article.txt", keyword)
print("Enter keyword:", keyword)
print(f"Found {len(matches)} matching lines:")
for number, line in matches:
    print(f"  Line {number}: {line}")
assert [number for number, _ in matches] == [1, 4]
assert search_file("article.txt", "PYTHON") == matches
assert search_file("article.txt", "unmatched-word") == []
````

## Exercise 8: CSV Writer (Medium)

Given the data below, write it to a CSV file called `"inventory.csv"` with a header row. Then read and print the file to verify.

```python
inventory = [
    ["Laptop", 5, 15000.00],
    ["Mouse", 25, 150.50],
    ["Keyboard", 15, 350.75],
    ["Monitor", 8, 4500.00]
]
```

**Expected output:**
```text
product,quantity,price
Laptop,5,15000.0
Mouse,25,150.5
Keyboard,15,350.75
Monitor,8,4500.0
```

### Worked reasoning

Write the column names once, then one row per product. The quantities and prices are formatted as text in a CSV line. Read back and parse them to verify that the values and header survived the file boundary.

**Türkçe:** CSV dosyasında her şey metindir; yeniden okurken sayısal alanların türünü geri kurarız.

**Expected check:** Header plus four product rows; rereading recovers every original field value.


**Contract:** Use the CSV module for writing and reading. Convert quantity back to int and price back to float before comparing restored records.


````python
inventory = [
    ["Laptop", 5, 15000.00], ["Mouse", 25, 150.50],
    ["Keyboard", 15, 350.75], ["Monitor", 8, 4500.00],
]
import csv
with open("inventory.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["product", "quantity", "price"])
    writer.writerows(inventory)
with open("inventory.csv", "r", encoding="utf-8") as file:
    print(file.read(), end="")
with open("inventory.csv", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)
    assert next(reader) == ["product", "quantity", "price"]
    restored = [[p, int(q), float(price)] for p,q,price in reader]
assert restored == inventory
````

**Line-by-line walkthrough**

- **Line 1 —** `` inventory = [ ``

  Start an inventory list; each row holds a product, quantity, and unit price.

- **Line 2 —** `` ["Laptop", 5, 15000.00], ["Mouse", 25, 150.50], ``

  Add laptop and mouse records with their quantities and prices.

- **Line 3 —** `` ["Keyboard", 15, 350.75], ["Monitor", 8, 4500.00], ``

  Add keyboard and monitor records.

- **Line 4 —** `` ] ``

Read the two exam columns, calculate their equally weighted score, and write the derived report to a different file. Reopen it to verify the rows.


````python
import csv
with open("grades.csv", "w", encoding="utf-8") as file:
    file.write('name,midterm,final\nElif Demir,82,90\nBurak Yilmaz,65,72\nZeynep Kaya,94,96\nMert Ozturk,45,50\nAyse Celik,78,80\nCem Arslan,68,74\nSelin Tas,88,91\n')
def grade_report(source, destination):
    results = []
    with open(source, newline="", encoding="utf-8") as file:
        reader = csv.reader(file)
        if next(reader, None) != ["name", "midterm", "final"]:
            raise ValueError("Unexpected grade header")
        for name, midterm, final in reader:
            midterm, final = float(midterm), float(final)
            weighted = 0.5 * midterm + 0.5 * final
            status = "PASS" if weighted >= 60 else "FAIL"
            results.append([name, weighted, status])
    with open(destination, "w", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        writer.writerow(["name", "weighted", "status"])
        for name, weighted, status in results:
            writer.writerow([name, f"{weighted:.1f}", status])
    return results
results = grade_report("grades.csv", "final_grades.csv")
print("Grade Report\n============")
for name, weighted, status in results:
    print(f"{name:<16}: {weighted:.1f} {status}")
print("Results written to final_grades.csv")
assert [r[1] for r in results] == [86,68.5,95,47.5,79,71,89.5]
with open("final_grades.csv", newline="", encoding="utf-8") as file:
    saved = list(csv.reader(file))
assert len(saved) == 8
assert saved[1] == ["Elif Demir", "86.0", "PASS"]
````

## Exercise 12: Data Cleaner (Challenge)

First run the cell below to create a messy CSV file. Then write a program that:
1. Reads the file
2. Removes rows with empty values
3. Removes rows where age is not a valid number

Check field count before indexing. Reject blank required fields, then convert the age and check its inclusive range. Save the original data-row number and reason for rejected rows. **Türkçe:** Alan sayısı kontrolü indeks hatasını önler. Hatalı satırın numarası ve nedeni kaybolmamalıdır.

**Contract:** This deliberately restricted parser accepts exactly three unquoted fields per physical line. Age must be integer text in 0–120. Row numbers count data rows, excluding the header; CSV physical line = row + 1. Retain raw data and write a separate rejection CSV. For general quoted CSV, use the Part 8 reader.


````python
with open('messy_data.csv', "w", encoding="utf-8") as file:
    file.write("""name,age,city
Elif Demir,21,Istanbul
Burak Yilmaz,22,Ankara
Zeynep Kaya,,Izmir
Mert Ozturk,23,Bursa
Ayse Celik,abc,Antalya
Cem Arslan,24,Trabzon
Selin Tas,-5,Eskisehir
Deniz Korkmaz,20,Konya
""")
def clean_people(lines):
    clean, errors = [], []
    for number, line in enumerate(lines, start=1):
        parts = [part.strip() for part in line.strip().split(",")]
        if len(parts) != 3:
            errors.append((number, "expected 3 fields"))
            continue
        name, age_text, city = parts
        if not name or not age_text or not city:
            reason = "empty value (age is empty)" if not age_text else "empty required value"
            errors.append((number, reason))
            continue
        try:
            age = int(age_text)
        except ValueError:
            errors.append((number, f"invalid age {age_text!r}"))
            continue
        if not 0 <= age <= 120:
            errors.append((number, f"age out of range ({age})"))
            continue
        clean.append([name, str(age), city])
    return clean, errors
with open("messy_data.csv", "r", encoding="utf-8") as file:
    header = file.readline().strip()
    lines = file.readlines()
clean, errors = clean_people(lines)
with open("clean_data.csv", "w", encoding="utf-8") as file:
    file.write(header + "\n")
    for row in clean:
        file.write(",".join(row) + "\n")
import csv
with open("clean_rejections.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["source_line", "raw_record", "reason"])
    for number, reason in errors:
        writer.writerow([number + 1, lines[number - 1].rstrip("\n"), reason])
print("Data Cleaning Report\n====================")
print(f"Total rows: {len(lines)}\nRemoved: {len(errors)} rows")
for number, reason in errors:
    print(f"  - Row {number}: {reason}")
print(f"Clean rows: {len(clean)}\nClean data written to clean_data.csv")
assert [number for number, _ in errors] == [3, 5, 7]
assert len(clean) == 5
assert clean_people(["A,120,B", "A,121,B", "too,few", "A,20,"])[0] == [["A","120","B"]]
with open("clean_data.csv", "r", encoding="utf-8") as file:
    assert len(file.read().splitlines()) == 6
````

## Bridge Exercise: Preview of Tables, NumPy and Plotting

Read `temperatures.csv`, skip the header, collect the `temp_celsius` values in a list, and compute the mean with a loop. Next week NumPy does the same arithmetic in one line and matplotlib draws it.

### Reasoning and plan

Write the same seven-row fixture first so this cell runs on its own. The loop keeps a running total and a count; the mean is total divided by count, computed once at the end. The seven values sum to 167.8, so the mean is 23.971…, printed 23.97. Keeping the values in a list also prepares them for `np.array(temps)` and `plt.plot(temps)` next week.

**Türkçe:** Döngü toplamı ve sayacı biriktirir; ortalama en sonda bir kez hesaplanır. Gelecek hafta aynı işi NumPy tek satırda yapar.


For a header-only file return an empty list and mean `None`; the caller should print a no-data message.


**Expected output:**
```text
Read 7 temperatures from temperatures.csv
Mean temperature: 23.97 °C
```


````python
with open("temperatures.csv", "w", encoding="utf-8") as file:
    file.write("""date,temp_celsius
2024-01-01,22.5
2024-01-02,24.3
2024-01-03,19.8
2024-01-04,18.5
2024-01-05,31.2
2024-01-06,27.4
2024-01-07,24.1
""")
def mean_from_file(filename):
    temps = []
    total = 0.0
    with open(filename, "r", encoding="utf-8") as file:
        header = file.readline()
        for line in file:
            parts = line.strip().split(",")
            value = float(parts[1])
            temps.append(value)
            total += value
    return temps, total / len(temps) if temps else None
temps, mean_temp = mean_from_file("temperatures.csv")
print(f"Read {len(temps)} temperatures from temperatures.csv")
print(f"Mean temperature: {mean_temp:.2f} °C")
assert len(temps) == 7 and temps[0] == 22.5
assert abs(mean_temp - 167.8 / 7) < 0.001
with open("one_temp.csv", "w", encoding="utf-8") as file:
    file.write("date,temp_celsius\n2024-02-01,-3.5\n")
assert mean_from_file("one_temp.csv") == ([-3.5], -3.5)
````

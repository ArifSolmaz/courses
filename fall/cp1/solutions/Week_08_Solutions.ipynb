## Exercise 1: Word Counter (Easy)

Count the number of words in the given sentence.

### Reasoning and plan

Split on whitespace, then count the resulting words. With no separator argument, split handles repeated spaces and tabs together and discards leading/trailing whitespace. Empty text therefore has zero words.

````python
sentence = 'Python is a great language to learn'
words = sentence.split()
print(f'The sentence has {len(words)} words.')
assert len(words) == 7
assert len('  Python   is	fun  '.split()) == 3
assert ''.split() == []
````

### Reasoning and plan

### Reasoning and plan

## Exercise 7: Character Frequency (Medium)

Count how many times each letter appears in the string. Ignore spaces and be case-insensitive. Print only letters that appear.

### Reasoning and plan

Scan normalized text in first-appearance order. Use a seen string to avoid printing a letter twice; count its occurrences in the full normalized text. Skip nonletters. In Hello Parsle, p occurs once, while e occurs twice and l three times.

**Türkçe:** Aynı harfi iki kez yazdırmamak için görülen harfleri saklarız. Harfin toplam tekrarını metnin tamamında sayarız.


````python
text = 'Hello Parsle'
normalized = text.lower()
seen = ''
letters = []
counts = []
for character in normalized:
    if character.isalpha() and character not in seen:
        seen += character
        count = normalized.count(character)
        letters.append(character)
        counts.append(count)
        print(f'{character}: {count}')
assert letters == ['h', 'e', 'l', 'o', 'p', 'a', 'r', 's']
assert counts == [1, 2, 3, 1, 1, 1, 1, 1]
assert sum(counts) == 11
````

**Line-by-line walkthrough**

- **Line 1 —** `` text = 'Hello Parsle' ``

  Store text containing repeated letters and a space.

  Normalize case once for the counting process.

- **Line 3 —** `` seen = '' ``

  Start an empty string recording letters already reported.

- **Line 4 —** `` letters = [] ``

  Start a list to record each distinct letter in first-appearance order.

- **Line 5 —** `` counts = [] ``

  Start a parallel list of their counts.

- **Line 6 —** `` for character in normalized: ``

  Visit each normalized character.

- **Line 7 —** `` if character.isalpha() and character not in seen: ``

  Process it only if it is alphabetic and has not already been reported. isalpha() uses Unicode letter classification.

- **Line 8 —** `` seen += character ``

  Mark this letter as seen before subsequent occurrences are visited.

- **Line 9 —** `` count = normalized.count(character) ``

  Count its occurrences throughout the normalized text.

- **Line 10 —** `` letters.append(character) ``

  Append the letter to the report order.

- **Line 11 —** `` counts.append(count) ``

  Append its count at the matching list position.

- **Line 12 —** `` print(f'{character}: {count}') ``

## Exercise 9: Replace Digits (Medium)

Replace every digit in the string with the `#` character.

### Reasoning and plan

Build a new string, replacing each digit with # and copying every other character unchanged. Hyphens and spaces retain their positions. isdigit follows Python’s Unicode digit definition; the supplied phone number uses ordinary ASCII digits.

**Türkçe:** Rakamları # ile değiştirirken diğer karakterleri aynen kopyalarız. Böylece metnin uzunluğu ve biçimi korunur.


````python
text = 'My phone is 0532-123-4567'
masked = ''
for character in text:
    if character.isdigit():
        masked += '#'
    else:
        masked += character
print('Original:', text)
print('Masked:  ', masked)
assert masked == 'My phone is ####-###-####'
assert len(masked) == len(text)
assert masked.count('#') == 11
````

Check if the given email address passes these basic rules:
1. Contains exactly one `@` symbol
2. Has at least one character before `@`
3. Has at least one `.` after `@`
4. Does not contain spaces

### Reasoning and plan

Implement the four specified rules in sequence: exactly one @, a nonempty part before it, a dot after it, and no spaces. This is a classroom format check, not proof that a mailbox exists or a complete email standard validator. Keep the stated rules visible rather than adding hidden ones.

**Türkçe:** Dört basit koşulu ayrı kontrol ederiz. Bu çalışma adresin gerçek veya ulaşılabilir olduğunu kanıtlamaz.

“Valid” here only means these four classroom rules pass. It does not mean a complete email syntax check or a reachable mailbox. The space rule checks the literal space character only.


````python
emails = ['emre.kaya@university.edu.tr', 'invalid-email', 'no spaces@mail.com']
expected = [True, False, False]
for index in range(len(emails)):
    email = emails[index]
    at_position = email.find('@')
    valid = (email.count('@') == 1
             and at_position > 0
             and '.' in email[at_position + 1:]
             and ' ' not in email)
    print(f'{email} -> {"Valid" if valid else "Invalid"}')
    assert valid == expected[index]
for email, expected_valid in [('@mail.com', False), ('a@@b.com', False), ('a@domain', False), ('a@b.c', True)]:
    at_position = email.find('@')
    valid = (email.count('@') == 1 and at_position > 0
             and '.' in email[at_position + 1:] and ' ' not in email)
    assert valid == expected_valid
````

Parse the log lines below. Each line has the format `timestamp,sensor_name,value`. Assume nonblank rows contain exactly three unquoted fields and a finite numeric value. All sensors here measure the same quantity in the same unit. Ignore blank lines. For an empty log, report that no readings are available. Find:
- The number of readings
- The average value across all readings
- The sensor and timestamp of the highest reading (keep the first occurrence if readings tie)

### Reasoning and plan

Split the text into rows, then split each row into timestamp, sensor and numeric value. Convert the value to float before comparison. Keep the sensor and timestamp whenever the maximum changes. The last reading 45.8 exceeds the earlier 45.2; total 157.3 divided by 5 gives 31.46.

**Türkçe:** En büyük değer değişince sensör ve zaman etiketi de değişmelidir. Son satırdaki 45,8 önceki 45,2’den büyüktür.

## Exercise 12: Caesar Cipher (Challenge)

Implement a Caesar cipher that shifts each letter by a given number of positions. Only shift letters (a-z, A-Z), leave other characters unchanged. Wrap around (z shifted by 1 becomes a).

### Reasoning and plan

Handle uppercase and lowercase using separate base codes. Subtract the base to get position 0–25, add the shift, wrap using %26, then add the base back. For z+3: (25+3)%26=2, giving c. Only A–Z/a–z are shifted; spaces, punctuation and other characters pass through.

**Türkçe:** Harf konumunu 0–25 aralığına çevirip kaydırırız. %26 alfabenin sonundan başına dönmeyi sağlar.


````python
cases = [('Hello, World!', 3, 'Khoor, Zruog!'),
         ('Zz Aa!', 1, 'Aa Bb!'),
         ('Abc', -1, 'Zab'),
         ('Çığ 123!', 26, 'Çığ 123!'),
         ('', 3, '')]
for message, shift, expected in cases:
    encrypted = ''
    for character in message:
        if 'a' <= character <= 'z':
            base = ord('a')
            position = ord(character) - base
            encrypted += chr((position + shift) % 26 + base)
        elif 'A' <= character <= 'Z':
            base = ord('A')
            position = ord(character) - base
            encrypted += chr((position + shift) % 26 + base)
        else:
            encrypted += character
    if message == 'Hello, World!' and shift == 3:
        print('Original: ', message)
        print('Encrypted:', encrypted)
    assert encrypted == expected
````

  Start an empty transformed string for this case.

## Bridge Exercise: Preview of Functions

Next week, we will learn about **functions**. A function is a reusable block of code that performs a specific task.

Try converting your palindrome checker (EX4) into a function. Don't worry if the syntax is new — just give it a try!

### Reasoning and plan

Move the palindrome calculation inside a named function. The parameter receives one word; return sends a Boolean back to the caller. Printing belongs in the outer loop so the same result can also be checked or reused elsewhere.

**Türkçe:** Hesabı fonksiyon içine taşırız. return mantıksal sonucu verir; yazdırma işlemi çağıran tarafta yapılır.


````python
def is_palindrome(word):
    normalized = word.lower()
    return normalized == normalized[::-1]
words = ['racecar', 'python', 'madam']
for word in words:
    label = 'Palindrome' if is_palindrome(word) else 'Not a palindrome'
    print(f'{word} -> {label}')
assert is_palindrome('Racecar') is True
assert is_palindrome('python') is False
assert is_palindrome('') is True
````

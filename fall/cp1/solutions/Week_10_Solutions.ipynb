```python

    greeting = "Merhaba"
```

## Exercise 3: Temperature Library (Medium)

Create a temperature conversion mini-library with these functions:

| Function | Formula |
|----------|---------|
| `c_to_f(celsius)` | F = C × 9/5 + 32 |
| `f_to_c(fahrenheit)` | C = (F − 32) × 5/9 |
| `c_to_k(celsius)` | K = C + 273.15 |
| `k_to_c(kelvin)` | C = K − 273.15 |

**Expected output:**
```
100°C = 212.0°F
32°F = 0.0°C
0°C = 273.15K
300K = 26.85°C
```

### Worked reasoning

Each formula maps one scale to another. Fahrenheit adds an offset after multiplying; reversing it requires subtracting 32 first. Kelvin and Celsius have equal step sizes and differ by 273.15. Round only when printing; test a round trip as well as reference points.

**Türkçe:** Ters dönüşümde işlemleri ters sırada uygulayın; hesap içinde erken yuvarlamayın.

**Expected check:** 212.0°F, 0.0°C, 273.15K and 26.85°C; round trips recover the starting number.


````python
import math

def c_to_f(celsius):
    return celsius * 9 / 5 + 32

def f_to_c(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

def c_to_k(celsius):
    return celsius + 273.15

def k_to_c(kelvin):
    return kelvin - 273.15

print(f"100°C = {c_to_f(100):.1f}°F")
print(f"32°F = {f_to_c(32):.1f}°C")
print(f"0°C = {c_to_k(0):.2f}K")
print(f"300K = {k_to_c(300):.2f}°C")
assert c_to_f(-40) == -40
assert math.isclose(f_to_c(c_to_f(23.4)), 23.4)
assert math.isclose(k_to_c(c_to_k(-10)), -10)
````

## Exercise 4: Math Helper Library (Medium)

Create a math helper library with these functions:

- `is_prime(n)` — returns `True` if `n` is prime, `False` otherwise
- `factorial(n)` — returns n! (e.g., factorial(5) = 120)
- `gcd(a, b)` — returns the greatest common divisor
- `lcm(a, b)` — returns the least common multiple (use `gcd` inside!)

**Expected output:**
```
is_prime(7) = True
is_prime(10) = False
factorial(5) = 120
gcd(12, 8) = 4
lcm(12, 8) = 24
```

### Worked reasoning

A composite integer has a factor at most its square root, so stop trial division when `divisor*divisor > n`. Factorial repeatedly multiplies, starting at 1 (therefore 0! = 1). Euclid replaces `(a,b)` with `(b,a%b)` until the remainder is zero. Compose LCM from GCD, treating a zero argument separately.

**Türkçe:** GCD döngüsünde kalan küçülür. LCM, GCD sonucunu yeniden kullanır; sıfır durumu ayrıca ele alınır.

**Expected check:** True, False, 120, 4, 24; boundary tests cover 0!, small primes and zero LCM.


````python
def is_prime(n):
    if n < 2:
        return False
    divisor = 2
    while divisor * divisor <= n:
        if n % divisor == 0:
            return False
        divisor += 1
    return True

def factorial(n):
    if n < 0:
        raise ValueError("factorial requires a non-negative integer")
    result = 1
    for factor in range(2, n + 1):
        result *= factor
    return result

def gcd(a, b):
    a, b = abs(a), abs(b)
    while b != 0:
        a, b = b, a % b
    return a

def lcm(a, b):
    if a == 0 or b == 0:
        return 0
    return abs(a * b) // gcd(a, b)

print("is_prime(7) =", is_prime(7))
print("is_prime(10) =", is_prime(10))
print("factorial(5) =", factorial(5))
print("gcd(12, 8) =", gcd(12, 8))
print("lcm(12, 8) =", lcm(12, 8))
assert not is_prime(1) and is_prime(2) and not is_prime(49)
assert factorial(0) == 1
assert gcd(-12, 8) == 4 and gcd(0, 8) == 8
assert lcm(0, 0) == 0 and lcm(3, 4) == 12
````

## Exercise 5: String Helper Library (Medium)

Create a string helper library with these functions:

- `is_palindrome(text)` — returns `True` if text reads the same forwards and backwards (ignore case and spaces)
- `count_vowels(text)` — returns the count of vowels (a, e, i, o, u)
- `reverse_string(text)` — returns the reversed text

**Expected output:**
```
is_palindrome('racecar') = True
is_palindrome('hello') = False
count_vowels('Hello World') = 3
reverse_string('Python') = nohtyP
```

### Worked reasoning

Normalize case and remove spaces only for palindrome comparison; preserve the original text for reversal. A vowel counter increases once for every matching character. The exercise’s vowel set is the five listed English vowels, so Turkish ı/ö/ü are not silently added.

**Türkçe:** Palindrom için normalleştirme yapıyoruz; ters çevirme fonksiyonu özgün harfleri korur.

**Expected check:** True, False, 3, nohtyP; spaces and mixed case work for palindrome comparison.


````python
def is_palindrome(text):
    cleaned = text.lower().replace(" ", "")
    return cleaned == cleaned[::-1]

def count_vowels(text):
    count = 0
    for char in text.lower():
        if char in "aeiou":
            count += 1
    return count

def reverse_string(text):
    return text[::-1]

print("is_palindrome('racecar') =", is_palindrome("racecar"))
print("is_palindrome('hello') =", is_palindrome("hello"))
print("count_vowels('Hello World') =", count_vowels("Hello World"))
print("reverse_string('Python') =", reverse_string("Python"))
assert is_palindrome("Never odd or even")
assert count_vowels("AEIOU") == 5
assert reverse_string("") == ""
````

## Exercise 6: Statistics Library (Medium)

Create a statistics library with:

- `mean(numbers)` — returns the average
- `median(numbers)` — returns the middle value (sort first!)
- `mode(numbers)` — returns the most frequent value

**Expected output:**
```
Data: [4, 7, 2, 7, 3, 7, 5]
Mean: 5.0
Median: 5
Mode: 7
```



**Input contract:** use a nonempty list of numbers. For tied modes, return the value that appears first in the original list. Do not mutate the original list while finding the median. The bridge below deliberately explores an empty-list error.

### Worked reasoning

The sum is 35, so the mean is 35/7 = 5. Sorting a copy gives `[2,3,4,5,7,7,7]`; index 3 is the median. Count each candidate for the mode; update the winner only for a strictly larger count to retain the first occurrence in a tie. Nonempty input is the contract; the bridge explores the empty mean failure.

**Türkçe:** Ortanca için sıralı kopya kullanılır. En sık değerde eşitlik varsa ilk görüleni koruyoruz.

**Expected check:** Mean 5.0, median 5, mode 7; the input order remains unchanged.


````python
def mean(numbers):
    return sum(numbers) / len(numbers)

def median(numbers):
    ordered = sorted(numbers)
    middle = len(ordered) // 2
    if len(ordered) % 2 == 1:
        return ordered[middle]
    return (ordered[middle - 1] + ordered[middle]) / 2

def mode(numbers):
    winner = numbers[0]
    best_count = 0
    for number in numbers:
        count = numbers.count(number)
        if count > best_count:
            winner = number
            best_count = count
    return winner

data = [4, 7, 2, 7, 3, 7, 5]
before = data.copy()
print("Data:", data)
print("Mean:", mean(data))
print("Median:", median(data))
print("Mode:", mode(data))
assert (mean(data), median(data), mode(data)) == (5.0, 5, 7)
assert data == before
assert median([4, 1, 2, 3]) == 2.5
assert mode([2, 1, 1, 2]) == 2
````

## Exercise 7: Geometry Library (Medium)

Create a geometry library with:

- `circle_area(radius)` — returns π × r² (use `PI = 3.14159` as a global constant)
- `rect_area(width, height)` — returns width × height
- `triangle_area(base, height)` — returns 0.5 × base × height

Also create a `shape_report(shapes)` function that takes a list of shape dictionaries and prints a report.

**Expected output:**
```
circle (r=5): area = 78.54
rectangle (4x6): area = 24.00
triangle (b=3, h=8): area = 12.00
```

### Worked reasoning

Give each formula a name, and store dimensions under descriptive dictionary keys. `shape_report` selects the relevant formula from each shape’s type. The constant PI is read but never rebound inside a function. Areas are 3.14159×25, 4×6, and 0.5×3×8.

**Türkçe:** Sözlük, şeklin türünü ve ölçülerini isimli alanlarda tutar; rapor doğru alanları doğru fonksiyona verir.

**Expected check:** Areas 78.54, 24.00 and 12.00 square units.


````python
PI = 3.14159

def circle_area(radius):
    return PI * radius ** 2

def rect_area(width, height):
    return width * height

def triangle_area(base, height):
    return 0.5 * base * height

def shape_report(shapes):
    areas = []
    for shape in shapes:
        if shape["type"] == "circle":
            area = circle_area(shape["radius"])
            label = f"circle (r={shape['radius']})"
        elif shape["type"] == "rectangle":
            area = rect_area(shape["width"], shape["height"])
            label = f"rectangle ({shape['width']}x{shape['height']})"
        elif shape["type"] == "triangle":
            area = triangle_area(shape["base"], shape["height"])
            label = f"triangle (b={shape['base']}, h={shape['height']})"
        else:
            raise ValueError("unknown shape type")
        print(f"{label}: area = {area:.2f}")
        areas.append(area)
    return areas

shapes = [
    {"type": "circle", "radius": 5},
    {"type": "rectangle", "width": 4, "height": 6},
    {"type": "triangle", "base": 3, "height": 8},
]
areas = shape_report(shapes)
assert [round(value, 2) for value in areas] == [78.54, 24.0, 12.0]
assert rect_area(0, 6) == 0
assert shape_report([]) == []
````

## Exercise 8: Refactor — Inline to Functions (Medium)

Refactor the following inline code into well-organized functions. The output should remain **exactly the same**.

```python
# Inline code to refactor:
prices = [120, 250, 80, 310, 45, 190]
total = 0
discounted_total = 0
for p in prices:
    total += p
    if p > 200:
        discounted_total += p * 0.9
    elif p > 100:
        discounted_total += p * 0.95
    else:
        discounted_total += p
print(f"Original total: {total} TL")
print(f"Discounted total: {discounted_total:.2f} TL")
print(f"You saved: {total - discounted_total:.2f} TL")
```

**Expected output:**
```
Original total: 995 TL
Discounted total: 923.50 TL
You saved: 71.50 TL
```

### Worked reasoning

Preserve the original strict inequalities: prices above 200 receive 10%, otherwise prices above 100 receive 5%. The transformed prices are 114, 225, 80, 279, 45, 180.5; their sum is 923.5. Savings are 995−923.5 = 71.5. A refactoring must preserve the algorithm’s output, even when the old printed answer key was wrong.

**Türkçe:** Refaktör işlemin davranışını korur. İndirim sınırlarında > ile ≥ aynı değildir; 100 ve 200 değerlerini test edin.

**Expected check:** Original 995 TL; discounted 923.50 TL; saved 71.50 TL.


````python
def apply_discount(price):
    if price > 200:
        return price * 0.9
    if price > 100:
        return price * 0.95
    return price

def calculate_total(prices):
    return sum(prices)

def calculate_discounted_total(prices):
    total = 0
    for price in prices:
        total += apply_discount(price)
    return total

def print_summary(prices):
    total = calculate_total(prices)
    discounted = calculate_discounted_total(prices)
    print(f"Original total: {total} TL")
    print(f"Discounted total: {discounted:.2f} TL")
    print(f"You saved: {total - discounted:.2f} TL")

prices = [120, 250, 80, 310, 45, 190]
print_summary(prices)
assert calculate_total(prices) == 995
assert calculate_discounted_total(prices) == 923.5
assert apply_discount(100) == 100 and apply_discount(200) == 190
assert calculate_discounted_total([]) == 0
````

## Exercise 9: Validator Library (Medium)

Create a validator library with:

- `validate_age(age)` — returns `True` if age is between 0 and 150
- `validate_email(email)` — returns `True` if email contains `@` and `.` after `@`
- `validate_id(student_id)` — returns `True` if student_id is exactly 10 digits

Also create `validate_student(name, age, email, student_id)` that checks all three and returns a list of error messages (empty list if all valid).

**Expected output:**
```
Valid student: []
Invalid student: ['Invalid age', 'Invalid email']
```

### Worked reasoning

Keep each validation rule separate and collect all failures instead of stopping at the first. The email rule is a deliberately simple exercise rule, not a complete email-address standard. Interpret ID digits as ten ASCII characters 0–9 so digit-like Unicode symbols do not accidentally pass. The name is a label; this exercise only requests checks on age, email and ID.

**Türkçe:** Her koşulu ayrı fonksiyonda kontrol edin; birden çok hata varsa hepsini listeleyin.

**Expected check:** The valid student has []; the invalid student has ['Invalid age', 'Invalid email'].


````python
def validate_age(age):
    return isinstance(age, (int, float)) and not isinstance(age, bool) and 0 <= age <= 150

def validate_email(email):
    if not isinstance(email, str) or "@" not in email:
        return False
    after_at = email.split("@", 1)[1]
    return "." in after_at

def validate_id(student_id):
    if not isinstance(student_id, str) or len(student_id) != 10:
        return False
    for char in student_id:
        if char not in "0123456789":
            return False
    return True

def validate_student(name, age, email, student_id):
    errors = []
    if not validate_age(age):
        errors.append("Invalid age")
    if not validate_email(email):
        errors.append("Invalid email")
    if not validate_id(student_id):
        errors.append("Invalid ID")
    return errors

print("Valid student:", validate_student("Elif", 21, "elif@example.org", "0123456789"))
bad = validate_student("Cem", -2, "cem.example.org", "0123456789")
print("Invalid student:", bad)
assert bad == ["Invalid age", "Invalid email"]
assert validate_age(0) and validate_age(150) and not validate_age(151)
assert not validate_id("²" * 10)
assert not validate_email("first.last@example")
````

## Exercise 10: Composition (Medium)

Build a `process_grade(raw_score, bonus, curve)` function that:

1. Calls `apply_bonus(score, bonus)` — adds bonus points
2. Calls `apply_curve(score, curve)` — adds curve points
3. Calls `clamp(score, 0, 100)` — ensures score is between 0 and 100
4. Calls `to_letter(score)` — converts to letter grade

**Expected output:**
```
Raw: 72, Bonus: 5, Curve: 8 -> Score: 85, Grade: BA
Raw: 95, Bonus: 10, Curve: 5 -> Score: 100, Grade: AA
Raw: 30, Bonus: 0, Curve: 5 -> Score: 35, Grade: FF
```



**Exercise grading scale:** AA ≥90, BA ≥80, BB ≥70, CB ≥60, CC ≥50, otherwise FF; only FF fails. These are rules for this programming exercise, not the course assessment weights.

### Worked reasoning

Compose the functions in the stated order. First 72+5+8=85; second 95+10+5=110 is clamped to 100; third 30+0+5=35 remains 35. Return score and letter as a pair so a caller can display both without recomputing. The exercise scale is AA≥90, BA≥80, BB≥70, CB≥60, CC≥50, otherwise FF.

**Türkçe:** İşlemlerin sırası önemlidir: ekle, sınırla, sonra harf notuna dönüştür.

**Expected check:** 85 BA; 100 AA; 35 FF; clamping also prevents negative final scores.


````python
def apply_bonus(score, bonus):
    return score + bonus

def apply_curve(score, curve):
    return score + curve

def clamp(value, lower, upper):
    return max(lower, min(value, upper))

def to_letter(score):
    for boundary, letter in [(90, "AA"), (80, "BA"), (70, "BB"), (60, "CB"), (50, "CC")]:
        if score >= boundary:
            return letter
    return "FF"

def process_grade(raw_score, bonus, curve):
    score = apply_bonus(raw_score, bonus)
    score = apply_curve(score, curve)
    score = clamp(score, 0, 100)
    return score, to_letter(score)

for raw, bonus, curve in [(72, 5, 8), (95, 10, 5), (30, 0, 5)]:
    score, letter = process_grade(raw, bonus, curve)
    print(f"Raw: {raw}, Bonus: {bonus}, Curve: {curve} -> Score: {score}, Grade: {letter}")
assert process_grade(72, 5, 8) == (85, "BA")
assert process_grade(95, 10, 5) == (100, "AA")
assert process_grade(-10, 0, 0) == (0, "FF")
assert to_letter(50) == "CC" and to_letter(49.9) == "FF"
````

## Exercise 11: Grade Report Generator (Challenge)

Build a complete grade report system that combines multiple helper functions. Given a list of students with their exam scores, generate a formatted report.

Requirements:
- Calculate weighted average (Midterm: 40%, Final: 60%)
- Determine letter grade and pass/fail status
- Calculate class statistics (highest, lowest, class average)
- Print a formatted report

**Test data:**
```python
students = [
    {"name": "Ayse Yildiz", "midterm": 78, "final": 85},
    {"name": "Kemal Ozturk", "midterm": 92, "final": 88},
    {"name": "Deniz Arslan", "midterm": 45, "final": 40},
    {"name": "Hakan Cetin", "midterm": 65, "final": 72},
    {"name": "Sibel Kara", "midterm": 88, "final": 95},
]
```

**Expected output (approximate):**
```
═══════════════════════════════════════
         GRADE REPORT
═══════════════════════════════════════
Ayse Yildiz     : 82.2 (BA) - PASS
Kemal Ozturk    : 89.6 (BA) - PASS
Deniz Arslan    : 42.0 (FF) - FAIL
Hakan Cetin     : 69.2 (CB) - PASS
Sibel Kara      : 92.2 (AA) - PASS
═══════════════════════════════════════
Class Average: 75.0
Highest: 92.2 (Sibel Kara)
Lowest: 42.0 (Deniz Arslan)
Pass Rate: 80.0%
═══════════════════════════════════════
```



**Exercise grading scale:** AA ≥90, BA ≥80, BB ≥70, CB ≥60, CC ≥50, otherwise FF; only FF fails. These are rules for this programming exercise, not the course assessment weights.

### Worked reasoning

Compute each weighted score before any rounding: Ayse 0.4×78+0.6×85=82.2; Kemal 89.6; Deniz 42.0; Hakan 69.2; Sibel 92.2. Sum 375.2 / 5 = 75.04, printed 75.0. Four scores pass, so 4/5×100=80%. Reuse the letter mapping from EX10; the illustrative exam weights are not CP1’s course assessment weights.

**Türkçe:** Önce bireysel sonuçları tam hassasiyetle hesaplayın; sınıf ortalamasını yuvarlanmış metinlerden üretmeyin.

**Expected check:** Class average 75.0; highest Sibel 92.2, lowest Deniz 42.0; pass rate 80.0%.


> **Example data only:** These fictional calculation weights are for this programming exercise. The course grade uses only the midterm (50%) and final (50%).


````python
def weighted_average(midterm, final):
    return 0.4 * midterm + 0.6 * final

def student_result(student):
    score = weighted_average(student["midterm"], student["final"])
    letter = to_letter(score)
    return {"name": student["name"], "score": score, "letter": letter,
            "status": "FAIL" if letter == "FF" else "PASS"}

def class_stats(results):
    if not results:
        return {"average": None, "highest": None, "lowest": None, "pass_rate": None}
    total = 0
    passed = 0
    highest = lowest = results[0]
    for row in results:
        total += row["score"]
        passed += row["status"] == "PASS"
        if row["score"] > highest["score"]:
            highest = row
        if row["score"] < lowest["score"]:
            lowest = row
    return {"average": total / len(results), "highest": highest, "lowest": lowest,
            "pass_rate": 100 * passed / len(results)}

students = [
    {"name": "Ayse Yildiz", "midterm": 78, "final": 85},
    {"name": "Kemal Ozturk", "midterm": 92, "final": 88},
    {"name": "Deniz Arslan", "midterm": 45, "final": 40},
    {"name": "Hakan Cetin", "midterm": 65, "final": 72},
    {"name": "Sibel Kara", "midterm": 88, "final": 95},
]
results = [student_result(student) for student in students]
stats = class_stats(results)
print("GRADE REPORT")
for row in results:
    print(f"{row['name']:<15}: {row['score']:.1f} ({row['letter']}) - {row['status']}")
print(f"Class Average: {stats['average']:.1f}")
for label in ["highest", "lowest"]:
    row = stats[label]
    print(f"{label.title()}: {row['score']:.1f} ({row['name']})")
print(f"Pass Rate: {stats['pass_rate']:.1f}%")
assert math.isclose(stats["average"], 75.04)
assert stats["pass_rate"] == 80
assert stats["highest"]["name"] == "Sibel Kara"
assert class_stats([])["average"] is None
````

|---|---|---|
| `velocity(distance, time)` | `distance / time` | **Average** speed for straight forward motion |
| `acceleration(v_final, v_initial, time)` | `(v_final - v_initial) / time` | Constant acceleration |
| `force(mass, acceleration)` | `mass * acceleration` | Net force along the motion |
| `kinetic_energy(mass, velocity)` | `0.5 * mass * velocity**2` | Energy at the stated instantaneous speed |
| `potential_energy(mass, height)` | `mass * 9.81 * height` | Gravitational energy relative to zero height |
| `momentum(mass, velocity)` | `mass * velocity` | Momentum along the motion |

**Physical model:** a 10 kg cart moves in a straight line from rest with constant
acceleration, covering 100 m in 5 s. Its height is 20 m above the chosen reference.
There is no reversal. Time is positive and mass is non-negative. The force requested
is the net force **along the motion**; height is only used to report gravitational energy.

The average speed is `100/5 = 20 m/s`. It is not the final speed.
From `distance = (v_initial + v_final)*time/2`, with `v_initial = 0`, obtain
`v_final = 2*distance/time = 40 m/s`, then `a = (40-0)/5 = 8 m/s²`.
Use the final speed for final kinetic energy and momentum.

**Expected output:**
```text
Average speed: 20.00 m/s
Final velocity: 40.00 m/s
Acceleration: 8.00 m/s²
Net force: 80.00 N
Final kinetic energy: 8000.00 J
Potential energy: 1962.00 J
Final momentum: 400.00 kg·m/s
```

**Türkçe:** `yol/zaman` ortalama sürattir. Durarak başlayıp sabit ivmeyle hızlanan
cisimde son sürat bunun iki katıdır. Kuvvet, enerji ve momentumda aynı fiziksel
modeli kullanın. Fonksiyonları kısa tutun; rapor fonksiyonu bunları sırayla çağırır.

### Worked reasoning

From rest and constant acceleration: d=½at² ⇒ a=2d/t²=200/25=8 m/s². Then vf=at=40 m/s while average speed d/t=20 m/s. Net force=10×8=80 N, final KE=½×10×40²=8000 J, gravitational PE=10×9.81×20=1962 J, final momentum=10×40=400 kg·m/s. This is forward straight-line motion; the reported height defines only the gravitational reference energy.

**Türkçe:** Ortalama sürat ile son sürati ayrı tutuyoruz. d=½at² bağıntısı, tüm fonksiyonların aynı hareket modelini kullanmasını sağlar.

**Expected check:** 20 m/s average, 40 m/s final, 8 m/s², 80 N, 8000 J, 1962 J, 400 kg·m/s.


````python
GRAVITY = 9.81

def velocity(distance, time):
    """Average speed for forward straight-line motion; time must be positive."""
    return distance / time

def acceleration(v_final, v_initial, time):
    return (v_final - v_initial) / time

def force(mass, acceleration):
    return mass * acceleration

def kinetic_energy(mass, velocity):
    return 0.5 * mass * velocity ** 2

def potential_energy(mass, height):
    return mass * GRAVITY * height

def momentum(mass, velocity):
    return mass * velocity

def physics_report(mass, distance, time, height):
    average = velocity(distance, time)
    final = 2 * average  # starts at rest; constant acceleration
    a = acceleration(final, 0, time)
    report = {"average": average, "final": final, "acceleration": a,
              "force": force(mass, a), "kinetic": kinetic_energy(mass, final),
              "potential": potential_energy(mass, height), "momentum": momentum(mass, final)}
    for label, key, unit in [
        ("Average speed", "average", "m/s"), ("Final velocity", "final", "m/s"),
        ("Acceleration", "acceleration", "m/s²"), ("Net force", "force", "N"),
        ("Final kinetic energy", "kinetic", "J"), ("Potential energy", "potential", "J"),
        ("Final momentum", "momentum", "kg·m/s")]:
        print(f"{label}: {report[key]:.2f} {unit}")
    return report

physics = physics_report(10, 100, 5, 20)
assert physics["acceleration"] == 8 and physics["final"] == 40
assert physics["force"] == 80 and physics["kinetic"] == 8000
assert physics["momentum"] == 400
assert 0.5 * physics["acceleration"] * 5 ** 2 == 100
assert physics["force"] * 100 == physics["kinetic"]  # work-energy check
````

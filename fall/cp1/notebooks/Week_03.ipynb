## Part 2: Boolean Expressions

A **Boolean expression** is any expression that evaluates to either `True` or `False`. These are the building blocks of decision-making.

### Comparison Operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 3` | `True` |
| `>` | greater than | `7 > 3` | `True` |
| `<` | less than | `2 < 1` | `False` |
| `>=` | greater than or equal | `5 >= 5` | `True` |
| `<=` | less than or equal | `3 <= 2` | `False` |

> **Note:** The result of a comparison is always a `bool` type — either `True` or `False`.

**Figure 2.1: Comparison operators in action**

````python
# Boolean expressions with comparison operators
x = 10
y = 7

print("x == y :", x == y)   # Is 10 equal to 7?
print("x != y :", x != y)   # Is 10 not equal to 7?
print("x > y  :", x > y)    # Is 10 greater than 7?
print("x < y  :", x < y)    # Is 10 less than 7?
print("x >= 10:", x >= 10)  # Is 10 greater than or equal to 10?
print("x <= 5 :", x <= 5)   # Is 10 less than or equal to 5?
````

**Figure 2.2: The bool type**

````python
# Boolean values are a data type
result = 5 > 3
print("Value:", result)       # True
print("Type:", type(result))  # <class 'bool'>

# You can also assign True/False directly
is_student = True
is_graduated = False
print("Is student?", is_student)
print("Is graduated?", is_graduated)
````

**Figure 2.3: Comparing strings**

````python
# You can compare strings too!
name1 = "Ahmet"
name2 = "ahmet"

print(name1 == name2)           # False - case matters!
print(name1.lower() == name2)   # True - now both lowercase
print("apple" < "banana")       # True - alphabetical order
````

## Part 3: The `if` Statement

The `if` statement lets your program **execute code only when a condition is True**.

### Syntax
```python
if condition:
    # This code runs ONLY if condition is True
    # Notice the 4-space indentation!
```

### Key Rules
1. The condition must be followed by a **colon** `:`
2. The code inside must be **indented by 4 spaces**
3. Everything at the same indentation level is part of the same **code block**
4. Code after the if block (not indented) **always runs**

**Figure 3.1: Basic if statement**

````python
# Basic if statement
temperature = 35

if temperature > 30:
    print("It's hot today!")
    print("Don't forget to drink water.")

print("Have a nice day!")  # This ALWAYS runs (not indented under if)
````

**Figure 3.2: When the condition is False**

````python
# When the condition is False, the indented block is SKIPPED
temperature = 20

if temperature > 30:
    print("It's hot today!")          # This does NOT run
    print("Don't forget to drink water.")  # This does NOT run

print("Have a nice day!")  # This ALWAYS runs
````

**Figure 3.3: Indentation defines the code block**

````python
# Indentation is CRUCIAL in Python
score = 85

if score >= 60:
    print("You passed!")    # Part of the if block
    print("Well done!")     # Also part of the if block
print("End of program.")    # NOT part of if - always runs

print("---")

# What if score is 40?
score = 40

if score >= 60:
    print("You passed!")    # Skipped
    print("Well done!")     # Skipped
print("End of program.")    # Still runs!
````

> **Note:** Python uses **indentation** (whitespace) to define code blocks, unlike many other languages that use `{ }` braces. Always use **4 spaces** for each level of indentation. Most code editors do this automatically when you press Tab.

## Part 4: `if/else` — Two-Way Decisions

Often you want to do **one thing** if a condition is True, and **something else** if it's False.


## Part 5: `if/elif/else` — Multi-Way Decisions

When you have **more than two** possibilities, use `elif` (short for "else if").

```python
if condition1:
    # runs if condition1 is True
elif condition2:
    # runs if condition1 is False AND condition2 is True
elif condition3:
    # runs if conditions 1 and 2 are False AND condition3 is True
else:
    # runs if ALL conditions above are False
```

> **Note:** Python checks conditions **from top to bottom** and executes **only the first** matching block. Order matters!

**Figure 5.1: Grading system**

````python
# Turkish university grading system
score = 78

if score >= 90:
    grade = "AA"
elif score >= 85:
    grade = "BA"
elif score >= 80:
    grade = "BB"
elif score >= 75:
    grade = "CB"
elif score >= 70:
    grade = "CC"
elif score >= 65:
    grade = "DC"
elif score >= 60:
    grade = "DD"
elif score >= 50:
    grade = "FD"
else:
    grade = "FF"

print(f"Score: {score} -> Grade: {grade}")
````

**Figure 5.2: Why order matters**

````python
# WRONG order - this gives incorrect results!
score = 95

# Bad example - checking >= 60 first catches everything!
if score >= 60:
    grade = "DD"   # 95 is >= 60, so this runs!
elif score >= 70:
    grade = "CC"   # Never reached for high scores
elif score >= 90:
    grade = "AA"   # Never reached!

print(f"WRONG: Score {score} -> Grade {grade}")

# Correct order - check highest first
if score >= 90:
    grade = "AA"   # 95 is >= 90, correct!
elif score >= 70:
    grade = "CC"
elif score >= 60:
    grade = "DD"

print(f"RIGHT: Score {score} -> Grade {grade}")
````

**Figure 5.3: Temperature classifier**

````python
# Temperature classifier
temp = 22

if temp < 0:
    description = "Freezing"
elif temp < 10:
    description = "Cold"
elif temp < 20:
    description = "Cool"
elif temp < 30:
    description = "Warm"
else:
    description = "Hot"

print(f"{temp} degrees Celsius is: {description}")
````

## Part 6: Logical Operators — `and`, `or`, `not`

You can **combine** multiple conditions using logical operators.

| Operator | Meaning | Example |
|---|---|---|
| `and` | Both must be True | `age >= 18 and has_id` |
| `or` | At least one must be True | `is_student or is_teacher` |
| `not` | Reverses True/False | `not is_raining` |

### Truth Tables

**`and` operator:**

| A | B | A and B |
|---|---|---|
| True | True | **True** |
| True | False | False |
| False | True | False |
| False | False | False |

**`or` operator:**

| A | B | A or B |
|---|---|---|
| True | True | **True** |
| True | False | **True** |
| False | True | **True** |
| False | False | False |

**`not` operator:**

| A | not A |
|---|---|
| True | False |
| False | True |

**Figure 6.1: Using `and`**

````python
# Using 'and' - BOTH conditions must be True
age = 20
has_license = True

if age >= 18 and has_license:
    print("You can drive!")
else:
    print("You cannot drive.")

print("---")

# Another example: range check
score = 75
if score >= 0 and score <= 100:
    print(f"Score {score} is valid")
else:
    print(f"Score {score} is invalid")
````

**Figure 6.2: Using `or`**

````python
# Using 'or' - at least ONE condition must be True
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
else:
    print("It's a weekday.")

print("---")

# Discount eligibility
age = 15
is_student = True

if age < 12 or age > 65 or is_student:
    print("You get a discount!")
else:
    print("Regular price.")
````

**Figure 6.3: Using `not`**

````python
# Using 'not' - reverses the Boolean value
is_raining = False

if not is_raining:
    print("Let's go for a walk!")
else:
    print("Better stay inside.")

print("---")

# Combining operators
username = "admin"
password = "1234"

if username == "admin" and not password == "1234":
    print("Wrong password!")
elif username == "admin" and password == "1234":
    print("Welcome, admin!")
else:
    print("Unknown user.")
````

### Short-Circuit Evaluation

Python is smart about evaluating logical operators:
- `and`: If the first condition is `False`, Python **skips** the second (result is already `False`)
- `or`: If the first condition is `True`, Python **skips** the second (result is already `True`)

**Figure 6.4: Short-circuit evaluation**

````python
# Short-circuit evaluation
x = 0

# Without short-circuit, this would cause division by zero!
# But Python stops at 'x != 0' because it's False
if x != 0 and 10 / x > 2:
    print("Condition met")
else:
    print("x is zero, division was safely skipped!")
````

## Part 7: Chained Comparisons

Python has a special, readable syntax for range checks:

```python
# Instead of this:
if x >= 0 and x < 100:

# You can write this:
if 0 <= x < 100:
```

This is unique to Python and reads like natural math notation!

**Figure 7.1: Chained comparisons**

````python
# Chained comparisons - Python's special feature
age = 25

# Traditional way
if age >= 18 and age <= 65:
    print("Working age (traditional check)")

# Pythonic way - much cleaner!
if 18 <= age <= 65:
    print("Working age (chained comparison)")

print("---")

# Score validation
score = 85
if 0 <= score <= 100:
    print(f"Score {score} is valid")
else:
    print(f"Score {score} is out of range")
````

## Part 8: Nested Conditionals

You can put an `if` statement **inside** another `if` statement. This is called **nesting**.

```python
if outer_condition:
    if inner_condition:
        # runs when BOTH conditions are True
    else:
        # runs when outer is True but inner is False
```

> **Note:** Try to avoid nesting more than 2-3 levels deep. Deep nesting makes code hard to read. You can often **flatten** nested conditions using `and`.

**Figure 8.1: Nested conditionals**

````python
# Nested conditionals - movie ticket pricing
age = 15
is_student = True

if age < 18:
    if is_student:
        price = 20
        print(f"Child student ticket: {price} TL")
    else:
        price = 25
        print(f"Child ticket: {price} TL")
else:
    if is_student:
        price = 30
        print(f"Adult student ticket: {price} TL")
    else:
        price = 50
        print(f"Adult ticket: {price} TL")
````

**Figure 8.2: Flattening nested conditionals**

````python
# The same logic, flattened with 'and' - often easier to read
age = 15
is_student = True

if age < 18 and is_student:
    price = 20
    print(f"Child student ticket: {price} TL")
elif age < 18 and not is_student:
    price = 25
    print(f"Child ticket: {price} TL")
elif age >= 18 and is_student:
    price = 30
    print(f"Adult student ticket: {price} TL")
else:
    price = 50
    print(f"Adult ticket: {price} TL")
````

## Part 10: Common Pitfalls

Let's look at mistakes that beginners commonly make with conditionals.

### Pitfall 1: `=` vs `==`

| Symbol | Meaning | Example |
|---|---|---|
| `=` | **Assignment** (store a value) | `x = 5` |
| `==` | **Comparison** (check equality) | `x == 5` |

**Figure 10.1: Assignment vs comparison**

````python
x = 10

# CORRECT - comparison
if x == 10:
    print("x is 10")

# WRONG - this would be an error in an if statement:
# if x = 10:    # SyntaxError!
#     print("x is 10")

print("Remember: = assigns, == compares!")
````

### Pitfall 2: Unreachable Branches

**Figure 10.2: Unreachable code**

````python
# BAD: Unreachable branch
score = 95

if score >= 90:
    print("Excellent!")    # This runs for 95
elif score >= 95:           # This can NEVER be reached!
    print("Outstanding!")   # Because 95 >= 90 is already True above

print("---")

# GOOD: Check the more specific condition first
if score >= 95:
    print("Outstanding!")  # Now this can be reached
elif score >= 90:
    print("Excellent!")
````

### Pitfall 3: Forgetting the Colon

**Figure 10.3: Don't forget the colon!**

````python
# Common syntax errors (commented out so they don't crash):

# Missing colon:
# if x == 10     <- SyntaxError! Need a colon
#     print("hi")

# Wrong indentation:
# if x == 10:
# print("hi")    <- IndentationError! Need 4 spaces

# Correct:
x = 10
if x == 10:
    print("This is correct!")
````

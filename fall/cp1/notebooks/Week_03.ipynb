## Part 2: Boolean Expressions

A **Boolean expression** is any expression that evaluates to either `True` or `False`. These are the building blocks of decision-making.

### Comparison Operators

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | equal to | `5 == 5` | `True` |
| `!=` | not equal to | `5 != 3` | `True` |
| `>` | greater than | `7 > 3` | `True` |

| `>=` | greater than or equal | `5 >= 5` | `True` |
| `<=` | less than or equal | `3 <= 2` | `False` |

> **Note:** The result of a comparison is always a `bool` type — either `True` or `False`.

## Part 3: The `if` Statement

The `if` statement lets your program **execute code only when a condition is True**.

### Syntax
```python
if condition:
    # This code runs ONLY if condition is True
    # Notice the 4-space indentation!
```

### Key Rules
1. The condition must be followed by a **colon** `:`
2. The code inside must be **indented by 4 spaces**
3. Everything at the same indentation level is part of the same **code block**
4. Code after the if block (not indented) **always runs**

````python
temperature = 35
if temperature > 30:
    print("It's hot today!")
    print("Don't forget to drink water.")
print("Have a nice day!")
````


For `x != 0 and 10/x > 2`, Python checks from left to right.

| x | First condition | Is division evaluated? | Whole result |

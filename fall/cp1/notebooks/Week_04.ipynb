## Part 2: The `for` Loop

The `for` loop repeats a block of code **for each item** in a sequence.

### Syntax
```python
for variable in sequence:
    # This code runs once for each item
    # 'variable' takes the value of each item, one at a time
```

**Line-by-line walkthrough of the syntax template**

- **Line 1 —** `for variable in sequence:` visits the items of an existing sequence in order. Each pass assigns one item to the chosen variable name before entering the indented body. Replace these placeholder names with your real data and loop variable.
- **Line 2 —** The indented comment says where the repeated instructions belong; it does not perform an action.
- **Line 3 —** The second comment explains the changing value of the loop variable; it also performs no action.

This is a syntax template, not a runnable cell: `sequence` needs a value, and a loop body must contain an executable statement rather than only comments. The next example supplies both.


### How It Works
1. Python takes the **first item** from the sequence
2. Assigns it to the **loop variable**
3. Runs the **indented code block**
4. Goes back and takes the **next item**
5. Repeats until there are **no more items**

**Figure 2.1: Looping through a list of names**

````python
# Looping through a list of names
students = ["Elif", "Burak", "Deniz", "Selin", "Omer"]

# student holds one name at a time, in list order.
for student in students:
    # The f-string inserts the current name into the greeting.
    print(f"Hello, {student}!")

# No indentation: this runs once after the loop.
# \n starts a new line before the final message.
print("\nAll students greeted!")
````

**Line-by-line walkthrough**

- **Line 2 —** `` students = ["Elif", "Burak", "Deniz", "Selin", "Omer"] ``

  Store five strings in a list named `students`. The brackets hold the items in the order Elif, Burak, Deniz, Selin, Omer.

- **Line 5 —** `` for student in students: ``

  Take one name from that list on each pass and assign it to `student`. Finish the indented body before taking the next name.

- **Line 7 —** `` print(f"Hello, {student}!") ``

  Build an f-string using the current name: the first pass prints `Hello, Elif!`, the second `Hello, Burak!`, and so on.

- **Line 11 —** `` print("\nAll students greeted!") ``

  At the left margin, this runs once after all five greetings. `\n` inserts a blank line before `All students greeted!`.


**Figure 2.2: The loop variable changes each iteration**

````python
# Let's trace what happens step by step
colors = ["red", "green", "blue"]

# First pass: red; second: green; third: blue.
for color in colors:
    # This line repeats because it is indented.
    print(f"Current color is: {color}")

# This nonempty loop leaves 'color' set to its last item.
# An empty loop would not assign a new value to 'color'.
print(f"\nAfter the loop, color = {color}")
````

- **Line 5 —** `` for color in colors: ``

  Assign `red` to `color`, run the body, then repeat for `green` and `blue`. The colon and indentation identify that body.

  Insert the current `color` into the f-string and print it. This line runs three times, showing red, green and blue in order.

- **Line 11 —** `` print(f"\nAfter the loop, color = {color}") ``

  After the nonempty loop, `color` still holds `blue`; Python does not delete the loop variable. The leading `\n` adds a blank line. An empty loop would not assign a new value.


**Figure 2.3: Loop with multiple statements in the body**

````python
# You can have multiple lines inside the loop body
numbers = [3, 7, 2, 9, 5]

# Take one number, then finish all three indented lines.
for num in numbers:
    # For the first item 3: doubled becomes 6.
    doubled = num * 2
    # ** means power: 3 ** 2 is 9, not 3 times 2.
    squared = num ** 2
    print(f"{num} -> doubled: {doubled}, squared: {squared}")
````

## Part 3: The `range()` Function

The `range()` function represents a sequence of integers. It is the **most common** way to control how many times a `for` loop runs.

### Three Forms of `range()`

| Form | Meaning | Example | Numbers Generated |
|---|---|---|---|
| `range(n)` | 0 to n-1 | `range(5)` | 0, 1, 2, 3, 4 |
| `range(start, stop)` | start to stop-1 | `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(start, stop, step)` | start, then add step; stop is excluded | `range(0, 10, 2)` | 0, 2, 4, 6, 8 |

> **Note:** The `stop` value is **never included**! `range(1, 5)` gives 1, 2, 3, 4 — not 5. Think of it as: "up to but not including."

A negative step counts downward (for example, `range(5, 0, -1)`). A step of zero raises `ValueError`. A step pointing away from the stop produces an empty range, such as `range(1, 5, -1)`.

**Figure 3.1: `range(n)` — one argument**

````python
# range(n) - generates 0, 1, 2, ..., n-1
print("range(5):")
# With one argument, range starts at 0 and excludes 5.
for i in range(5):
    # end=" " puts a space after each value instead of a newline.
    print(i, end=" ")
print()  # New line

print("\nrange(3):")
for i in range(3):
    # end=" " puts a space after each value instead of a newline.
    print(i, end=" ")
print()
````

- **Line 4 —** `` for i in range(5): ``

  Use one-argument `range(5)`: start at 0, advance by 1, and stop before 5. `i` therefore takes five values, 0 through 4.

**Figure 3.2: `range(start, stop)` — two arguments**

````python
# range(start, stop) - generates start, start+1, ..., stop-1
print("range(1, 6):")
# Visit 1, 2, 3, 4, 5; the stop value 6 is excluded.
for i in range(1, 6):
    print(i, end=" ")
print()

print("\nrange(5, 11):")
# Start at 5 and stop before 11: six values.
for i in range(5, 11):
    print(i, end=" ")
print()

# Common use: print numbers 1 to 10
print("\nNumbers 1 to 10:")
for i in range(1, 11):   # Note: 11, not 10!
    print(i, end=" ")
print()
````

## Part 5: Accumulator Patterns

An **accumulator** is a variable that "accumulates" (collects) results as the loop runs. This is one of the most important patterns in programming!

### The Pattern
```python
# 1. Initialize the accumulator BEFORE the loop
total = 0

# 2. Update it INSIDE the loop
for item in sequence:
    total = total + item    # or: total += item

# 3. Use the result AFTER the loop
print(total)
```

**Line-by-line walkthrough of the pattern template**

- **Line 1 —** This comment describes initialization; Python ignores the comment itself.
- **Line 2 —** `total = 0` creates the starting sum once, before the loop. A value of zero means nothing has been added yet.
- **Line 4 —** This comment identifies the update stage; it does not update the total.
- **Line 5 —** `for item in sequence:` takes one item per pass from a sequence you supply. For example, use a previously defined list `[2, 4, 6]`.
- **Line 6 —** `total = total + item` adds the current item to the previous total, then stores the new sum. With `[2, 4, 6]`, the totals become 2, 6 and 12. The comment shows the equivalent shorthand `total += item`.
- **Line 8 —** This comment identifies the reporting stage; it does not print anything.
- **Line 9 —** `print(total)` sits outside the loop and prints the final sum once. Indenting it would instead show a running total after each item.

Blank lines 3 and 7 only separate the stages visually. Define `sequence` before using this template.


**Figure 5.1: Running sum**

````python
# Accumulator pattern: Sum of numbers 1 to 10
total = 0    # Step 1: Initialize

# Include 10 by using the stop value 11.
for i in range(1, 11):
    total += i    # Step 2: Accumulate (same as total = total + i)
    # This indented print shows the running total each pass.
    print(f"After adding {i}: total = {total}")

print(f"\nFinal sum: {total}")    # Step 3: Use result
````

**Line-by-line walkthrough**

- **Line 2 —** `` total = 0    # Step 1: Initialize ``

  Start the running sum at 0, before any values have been added.

- **Line 5 —** `` for i in range(1, 11): ``

  Assign 1 through 10 to `i` in order. Using stop 11 includes the number 10 in the sum.

- **Line 6 —** `` total += i    # Step 2: Accumulate (same as total = total + i) ``

  Add the current `i` to the previous `total`. The first three totals are 1, 3 and 6; the previous total is not discarded.

- **Line 8 —** `` print(f"After adding {i}: total = {total}") ``

  Print the new total during this pass, after the addition. The first line says `After adding 1: total = 1`; there are ten such progress lines.

- **Line 10 —** `` print(f"\nFinal sum: {total}")    # Step 3: Use result ``

  After the loop, print a blank line and `Final sum: 55`. Its indentation makes it a single final report, not another running-total line.


### Trace the accumulator, not just its final value

For `total = 0` followed by `for i in range(1, 4): total += i`:

| Iteration | i | total before | Operation | total after |
|---|---:|---:|---|---:|
| 1 | 1 | 0 | 0 + 1 | 1 |
| 2 | 2 | 1 | 1 + 2 | 3 |
| 3 | 3 | 3 | 3 + 3 | 6 |

Pseudocode: start the total once; visit each number; add it to the previous total; report after the loop. Putting `total = 0` inside the loop erases earlier work. The stop value 4 is excluded.

**Figure 5.3: Running product (factorial)**

## Part 7: Nested Loops

A **nested loop** is a loop inside another loop. The inner loop runs **completely** for each iteration of the outer loop.

```python
for i in range(outer_count):    # Outer loop
    for j in range(inner_count):    # Inner loop
        # This code runs outer_count * inner_count times!
```

**Line-by-line walkthrough of the nested-loop template**

- **Line 1 —** `for i in range(outer_count):` starts an outer loop. For a nonnegative integer count of 2, `i` takes 0 and 1.
- **Line 2 —** `for j in range(inner_count):` is indented inside the outer body. For a count of 3, `j` restarts at 0 and visits 0, 1 and 2 for each outer value.
- **Line 3 —** The doubly indented comment marks the place for the innermost action. A real instruction here would run six times for counts 2 and 3; this comment itself performs no action.

With those counts the pairs are (0,0), (0,1), (0,2), (1,0), (1,1), (1,2). This is a template: define the two counts and replace the comment with an executable statement. The next example does that.


**Figure 7.1: Understanding nested loop execution**

````python
# Nested loop - see how inner loop runs fully for each outer iteration
for i in range(1, 4):       # Outer: 1, 2, 3
    # For EACH i, restart j at 1 and run all three inner passes.
    for j in range(1, 4):   # Inner: 1, 2, 3
        print(f"  i={i}, j={j}")
    # This is inside the outer loop but outside the inner loop.
    # It prints once per i, after all three j values.
    print(f"--- End of outer loop i={i} ---")
````

**Line-by-line walkthrough**

- **Line 2 —** `` for i in range(1, 4):       # Outer: 1, 2, 3 ``

  Start the outer loop with `i` taking 1, 2 and 3. Each value of `i` must finish its whole body before the next outer pass.

- **Line 4 —** `` for j in range(1, 4):   # Inner: 1, 2, 3 ``

## Part 9: Off-by-One Errors

An **off-by-one error** happens when your loop runs one time too many or one time too few. This is one of the most common bugs in programming!

### Key Things to Remember
- For integer `n >= 0`, `range(n)` gives **n numbers**: 0 through n-1
- For integer `n >= 1`, `range(1, n)` gives **n-1 numbers**: 1 through n-1
- For integer `n >= 0`, `range(1, n+1)` gives **n numbers**: 1 through n
- Counting upward by 1 (`a <= b`), include `b` with `range(a, b+1)`. Counting downward by 1 (`a >= b`), use `range(a, b-1, -1)`.

**Figure 9.1: Common off-by-one mistakes**

````python
# Common off-by-one error examples

# WRONG: Print numbers 1 to 5
# The loop runs correctly, but its values do not match the task.
print("WRONG - range(5) starts at 0:")
for i in range(5):
    print(i, end=" ")  # Prints: 0 1 2 3 4
print()

# WRONG: Print numbers 1 to 5
print("WRONG - range(1, 5) stops at 4:")
for i in range(1, 5):
    print(i, end=" ")  # Prints: 1 2 3 4
print()

# CORRECT: Print numbers 1 to 5
# To include an upper endpoint, set stop one step beyond it.
print("RIGHT - range(1, 6) includes 5:")
for i in range(1, 6):
    print(i, end=" ")  # Prints: 1 2 3 4 5
print()
````

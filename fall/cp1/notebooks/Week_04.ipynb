## Part 2: The `for` Loop

The `for` loop repeats a block of code **for each item** in a sequence.

### Syntax
```python
for variable in sequence:
    # This code runs once for each item
    # 'variable' takes the value of each item, one at a time
```

### How It Works
1. Python takes the **first item** from the sequence
2. Assigns it to the **loop variable**
3. Runs the **indented code block**
4. Goes back and takes the **next item**
5. Repeats until there are **no more items**

**Figure 2.1: Looping through a list of names**

````python
# Looping through a list of names
students = ["Elif", "Burak", "Deniz", "Selin", "Omer"]

for student in students:
    print(f"Hello, {student}!")

print("\nAll students greeted!")
````

**Figure 2.2: The loop variable changes each iteration**

````python
# Let's trace what happens step by step
colors = ["red", "green", "blue"]

for color in colors:
    print(f"Current color is: {color}")

# After the loop, 'color' still has the LAST value
print(f"\nAfter the loop, color = {color}")
````

**Figure 2.3: Loop with multiple statements in the body**

````python
# You can have multiple lines inside the loop body
numbers = [3, 7, 2, 9, 5]

for num in numbers:
    doubled = num * 2
    squared = num ** 2
    print(f"{num} -> doubled: {doubled}, squared: {squared}")
````

## Part 3: The `range()` Function

The `range()` function generates a sequence of numbers. It is the **most common** way to control how many times a `for` loop runs.

### Three Forms of `range()`

| Form | Meaning | Example | Numbers Generated |
|---|---|---|---|
| `range(n)` | 0 to n-1 | `range(5)` | 0, 1, 2, 3, 4 |
| `range(start, stop)` | start to stop-1 | `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(start, stop, step)` | start to stop-1, by step | `range(0, 10, 2)` | 0, 2, 4, 6, 8 |

> **Note:** The `stop` value is **never included**! `range(1, 5)` gives 1, 2, 3, 4 — not 5. Think of it as: "up to but not including."

**Figure 3.1: `range(n)` — one argument**

````python
# range(n) - generates 0, 1, 2, ..., n-1
print("range(5):")
for i in range(5):
    print(i, end=" ")
print()  # New line

print("\nrange(3):")
for i in range(3):
    print(i, end=" ")
print()
````

**Figure 3.2: `range(start, stop)` — two arguments**

````python
# range(start, stop) - generates start, start+1, ..., stop-1
print("range(1, 6):")
for i in range(1, 6):
    print(i, end=" ")
print()

print("\nrange(5, 11):")
for i in range(5, 11):
    print(i, end=" ")
print()

# Common use: print numbers 1 to 10
print("\nNumbers 1 to 10:")
for i in range(1, 11):   # Note: 11, not 10!
    print(i, end=" ")
print()
````

**Figure 3.3: `range(start, stop, step)` — three arguments**

````python
# range(start, stop, step) - generates start, start+step, start+2*step, ...
print("Even numbers 0-10:")
for i in range(0, 11, 2):
    print(i, end=" ")
print()

print("\nOdd numbers 1-10:")
for i in range(1, 11, 2):
    print(i, end=" ")
print()

print("\nMultiples of 5 up to 50:")
for i in range(5, 51, 5):
    print(i, end=" ")
print()

# Negative step - counting DOWN
print("\nCountdown from 10 to 1:")
for i in range(10, 0, -1):
    print(i, end=" ")
print("\nBlast off!")
````

**Figure 3.4: Visualizing range boundaries**

````python
# Understanding range boundaries
# range(start, stop) -> start is INCLUDED, stop is EXCLUDED

print("range(1, 5) gives:", list(range(1, 5)))       # [1, 2, 3, 4]
print("range(0, 3) gives:", list(range(0, 3)))       # [0, 1, 2]
print("range(5) gives:   ", list(range(5)))          # [0, 1, 2, 3, 4]
print("range(2, 8, 2):   ", list(range(2, 8, 2)))   # [2, 4, 6]
print("range(10, 0, -3): ", list(range(10, 0, -3))) # [10, 7, 4, 1]
````

## Part 4: Looping Through Strings

A string is a **sequence of characters**, so you can loop through it just like a list!

**Figure 4.1: Iterating over a string**

````python
# Looping through each character in a string
name = "Python"

for char in name:
    print(char, end=" ")
print()  # New line
````

**Figure 4.2: Counting characters in a string**

````python
# Count vowels in a name
name = "Abdulkadir"
vowel_count = 0

for char in name:
    if char.lower() in "aeiou":
        vowel_count += 1

print(f"The name '{name}' has {vowel_count} vowels.")
````

**Figure 4.3: Building a new string from a loop**

````python
# Converting a string to uppercase manually
message = "hello world"
result = ""

for char in message:
    result = result + char.upper()

print(f"Original: {message}")
print(f"Uppercase: {result}")
````

## Part 5: Accumulator Patterns

An **accumulator** is a variable that "accumulates" (collects) results as the loop runs. This is one of the most important patterns in programming!

### The Pattern
```python
# 1. Initialize the accumulator BEFORE the loop
total = 0

# 2. Update it INSIDE the loop
for item in sequence:
    total = total + item    # or: total += item

# 3. Use the result AFTER the loop
print(total)
```

**Figure 5.1: Running sum**

````python
# Accumulator pattern: Sum of numbers 1 to 10
total = 0    # Step 1: Initialize

for i in range(1, 11):
    total += i    # Step 2: Accumulate (same as total = total + i)
    print(f"After adding {i}: total = {total}")

print(f"\nFinal sum: {total}")    # Step 3: Use result
````

**Figure 5.2: Running count**

````python
# Accumulator pattern: Count
scores = [85, 92, 45, 78, 60, 33, 95, 71]
pass_count = 0

for score in scores:
    if score >= 60:
        pass_count += 1

print(f"Out of {len(scores)} students, {pass_count} passed.")
print(f"Pass rate: {pass_count / len(scores) * 100:.1f}%")
````

**Figure 5.3: Running product (factorial)**

````python
# Accumulator pattern: Product (computing factorial)
# 5! = 5 * 4 * 3 * 2 * 1 = 120
n = 5
factorial = 1    # Initialize to 1 for multiplication!

for i in range(1, n + 1):
    factorial *= i    # same as factorial = factorial * i
    print(f"Step {i}: factorial = {factorial}")

print(f"\n{n}! = {factorial}")
````

> **Note:** When accumulating a **sum** or **count**, initialize to `0`. When accumulating a **product**, initialize to `1`. Think about why: adding 0 doesn't change a sum, and multiplying by 1 doesn't change a product.

## Part 6: Loop with `if` — Filtering

Combining a `for` loop with an `if` statement lets you **process only certain items** from a sequence.

**Figure 6.1: Counting even numbers**

````python
# Count how many even numbers are between 1 and 20
even_count = 0

for num in range(1, 21):
    if num % 2 == 0:
        even_count += 1
        print(f"  Found even: {num}")

print(f"\nTotal even numbers: {even_count}")
````

**Figure 6.2: Sum of positive numbers only**

````python
# Sum only positive numbers from a list
numbers = [4, -3, 7, -1, 8, -5, 2]
positive_sum = 0

for num in numbers:
    if num > 0:
        positive_sum += num
        print(f"  Adding {num}, running sum = {positive_sum}")
    else:
        print(f"  Skipping {num} (not positive)")

print(f"\nSum of positive numbers: {positive_sum}")
````

**Figure 6.3: Finding the maximum value**

````python
# Finding the maximum value in a list (without using max())
grades = [72, 85, 91, 68, 45, 93, 78]
maximum = grades[0]    # Start with the first element

for grade in grades:
    if grade > maximum:
        maximum = grade
        print(f"  New maximum found: {maximum}")

print(f"\nHighest grade: {maximum}")
````

## Part 7: Nested Loops

A **nested loop** is a loop inside another loop. The inner loop runs **completely** for each iteration of the outer loop.

```python
for i in range(outer_count):    # Outer loop
    for j in range(inner_count):    # Inner loop
        # This code runs outer_count * inner_count times!
```

**Figure 7.1: Understanding nested loop execution**

````python
# Nested loop - see how inner loop runs fully for each outer iteration
for i in range(1, 4):       # Outer: 1, 2, 3
    for j in range(1, 4):   # Inner: 1, 2, 3
        print(f"  i={i}, j={j}")
    print(f"--- End of outer loop i={i} ---")
````

**Figure 7.2: Multiplication table**

````python
# Multiplication table for 1 to 5
print("Multiplication Table (1-5)")
print("=" * 30)

# Print header
print("   ", end="")
for j in range(1, 6):
    print(f"{j:4}", end="")
print()
print("   " + "----" * 5)

# Print table body
for i in range(1, 6):
    print(f"{i:2}|", end="")
    for j in range(1, 6):
        print(f"{i*j:4}", end="")
    print()  # New line after each row
````

**Figure 7.3: Star patterns with nested loops**

````python
# Right triangle pattern using stars
# Row 1: *
# Row 2: **
# Row 3: ***
# Row 4: ****
# Row 5: *****

rows = 5
for i in range(1, rows + 1):
    for j in range(i):
        print("*", end="")
    print()  # New line after each row
````

**Figure 7.4: More star patterns**

````python
# Inverted triangle
print("Inverted triangle:")
rows = 5
for i in range(rows, 0, -1):
    for j in range(i):
        print("*", end="")
    print()

print()

# Simple trick: you can also use string multiplication!
print("Same pattern using string multiplication:")
for i in range(rows, 0, -1):
    print("*" * i)
````

> **Note:** You can use **string multiplication** (`"*" * 5` gives `"*****"`) as a shortcut instead of an inner loop for simple patterns. Both approaches work!

## Part 8: Common Loop Patterns

Here are patterns you'll use frequently:

**Figure 8.1: Countdown**

````python
# Countdown pattern
print("Countdown:")
for i in range(10, 0, -1):
    print(i, end="... ")
print("Go!")
````

**Figure 8.2: Step by 2 (skip counting)**

````python
# Skipping numbers with step
print("Every 3rd number from 0 to 30:")
for i in range(0, 31, 3):
    print(i, end=" ")
print()

print("\nEven numbers from 2 to 20:")
for i in range(2, 21, 2):
    print(i, end=" ")
print()
````

**Figure 8.3: Running average**

````python
# Computing the average of exam scores
scores = [85, 92, 78, 95, 88, 76, 90]
total = 0
count = 0

for score in scores:
    total += score
    count += 1

average = total / count
print(f"Scores: {scores}")
print(f"Total: {total}")
print(f"Count: {count}")
print(f"Average: {average:.2f}")
````

**Figure 8.4: Using `end` parameter in print**

````python
# The 'end' parameter controls what print() adds after the text
# Default is end="\n" (new line)

# Normal print - each on new line
print("Normal:")
for i in range(5):
    print(i)

# Using end=" " to print on same line
print("\nWith end=' ':")
for i in range(5):
    print(i, end=" ")
print()  # Final newline

# Using end=", "
print("\nWith end=', ':")
for i in range(5):
    print(i, end=", ")
print()  # Final newline
````

## Part 9: Off-by-One Errors

An **off-by-one error** happens when your loop runs one time too many or one time too few. This is one of the most common bugs in programming!

### Key Things to Remember
- `range(n)` gives **n numbers**: 0 through n-1
- `range(1, n)` gives **n-1 numbers**: 1 through n-1
- `range(1, n+1)` gives **n numbers**: 1 through n
- To include the endpoint, add 1 to stop: `range(a, b+1)`

**Figure 9.1: Common off-by-one mistakes**

````python
# Common off-by-one error examples

# WRONG: Print numbers 1 to 5
print("WRONG - range(5) starts at 0:")
for i in range(5):
    print(i, end=" ")  # Prints: 0 1 2 3 4
print()

# WRONG: Print numbers 1 to 5
print("WRONG - range(1, 5) stops at 4:")
for i in range(1, 5):
    print(i, end=" ")  # Prints: 1 2 3 4
print()

# CORRECT: Print numbers 1 to 5
print("RIGHT - range(1, 6) includes 5:")
for i in range(1, 6):
    print(i, end=" ")  # Prints: 1 2 3 4 5
print()
````

**Figure 9.2: Counting iterations**

````python
# How many times does each loop run?

count = 0
for i in range(10):
    count += 1
print(f"range(10) runs {count} times")       # 10 times

count = 0
for i in range(1, 10):
    count += 1
print(f"range(1, 10) runs {count} times")    # 9 times

count = 0
for i in range(1, 11):
    count += 1
print(f"range(1, 11) runs {count} times")    # 10 times

count = 0
for i in range(0, 10, 2):
    count += 1
print(f"range(0, 10, 2) runs {count} times") # 5 times
````

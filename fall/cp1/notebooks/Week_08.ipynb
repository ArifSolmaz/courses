### Indexing

Each character in a string has a position (index). Indexing starts at **0**.

| Index | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
|-------|---|---|---|---|---|---|---|---|
| Char  | I | s | t | a | n | b | u | l |
| Neg.  | -8| -7| -6| -5| -4| -3| -2| -1|

**Figure 1.2: Accessing characters by index**

````python
city = "Istanbul"
print(city[0])
print(city[3])
print(city[-1])
print(city[-2])
````

**Line-by-line walkthrough**

- **Line 1 —** `` city = "Istanbul" ``

  Store the eight-character string Istanbul.

- **Line 2 —** `` print(city[0]) ``

  Read the first character at index 0, I.

  Read index 3, the fourth character, a.

  Read the last character with index -1, l.

- **Line 5 —** `` print(city[-2]) ``

### Slicing

Slicing extracts a portion of a string using `[start:stop:step]`.

- `start` — where the slice begins (inclusive)
- `stop` — where the slice ends (exclusive)
- `step` — the index increment: `2` selects every second character, skipping one between selections; `-1` reverses the direction. A step of zero raises `ValueError`.

**Figure 1.3: String slicing examples**

````python
text = "Hello, World!"
print(text[0:5])
print(text[7:12])
print(text[:5])
print(text[7:])
print(text[::2])
print(text[::-1])
````

**Line-by-line walkthrough**

- **Line 1 —** `` text = "Hello, World!" ``

  Store the text, including its comma, space and exclamation mark.

- **Line 2 —** `` print(text[0:5]) ``

  Slice indices 0 through 4; the stop index 5 is excluded, giving Hello.

- **Line 3 —** `` print(text[7:12]) ``

  Slice indices 7 through 11, giving World.

- **Line 4 —** `` print(text[:5]) ``

  Omit the start to begin at index zero, again giving Hello.

- **Line 5 —** `` print(text[7:]) ``

  Omit the stop to take everything from index 7 through the end, including the exclamation mark.

- **Line 6 —** `` print(text[::2]) ``

  Take every second character starting at index zero, giving Hlo ol!.

- **Line 7 —** `` print(text[::-1]) ``

  Use a step of -1 to create a reversed string. None of these slices changes text.



> 💡 **Note:** Strings are **immutable** in Python. You cannot change a character directly. For example, `text[0] = 'h'` will cause an error. You must create a new string instead.

## Part 2: String Methods

Python provides many built-in methods to work with strings. A **method** is a function that belongs to an object — you call it with a dot: `string.method()`.

### Case Methods

| Method | Description | Example | Result |
|--------|-------------|---------|--------|
| `upper()` | All uppercase | `"hello".upper()` | `"HELLO"` |
| `lower()` | All lowercase | `"HELLO".lower()` | `"hello"` |
| `title()` | Capitalize each word | `"hello world".title()` | `"Hello World"` |
| `capitalize()` | Capitalize first character and lowercase the rest | `"hello world".capitalize()` | `"Hello world"` |
| `swapcase()` | Swap upper/lower | `"Hello".swapcase()` | `"hELLO"` |
These are Unicode case operations, not Turkish locale-specific name formatting. For example, `"I".lower()` gives `"i"`, not `"ı"`.


**Figure 2.1: Case conversion methods**

````python
student = "mehmet kaya"
print(student.upper())
print(student.lower())
print(student.title())
print(student.capitalize())
print(student.swapcase())
````

- **Line 3 —** `` print(student.lower()) ``

- **Line 4 —** `` print(student.title()) ``

- **Line 5 —** `` print(student.capitalize()) ``

- **Line 6 —** `` print(student.swapcase()) ``

### Search and Replace Methods

| Method | Description |
|--------|-------------|

## Part 3: split() and join()

These two methods are among the most important string operations.

### split() — String to List

`split()` breaks a string into a list of substrings. By default, it splits on whitespace.
With no separator, repeated whitespace is combined and empty text gives `[]`. With an explicit separator, empty fields are kept: `"T1,,C".split(",")` gives `["T1", "", "C"]`.


**Figure 3.1: Splitting strings into lists**

````python
sentence = "Ayse likes to code in Python"
words = sentence.split()
print(words)
print(len(words))
date = "2025-11-15"
parts = date.split("-")
print(parts)
csv_line = "Burak,22,Istanbul,Engineering"
fields = csv_line.split(",")
print(fields)
````

**Line-by-line walkthrough**

- **Line 1 —** `` sentence = "Ayse likes to code in Python" ``

  Store a sentence whose words are separated by spaces.

- **Line 2 —** `` words = sentence.split() ``

  Split on whitespace; repeated whitespace is collapsed and no empty boundary words are produced.

- **Line 3 —** `` print(words) ``

  Print the six resulting word strings in a list.

- **Line 4 —** `` print(len(words)) ``

### join() — List to String

`join()` combines an iterable of strings into one string, with a separator between elements. It does not always undo `split()`: splitting on whitespace discards the original spacing.

The syntax is: `separator.join(list)`

**Figure 3.2: Joining lists into strings**

````python
words = ["I", "love", "Python"]
sentence = " ".join(words)
print(sentence)
print("-".join(words))
print(", ".join(words))
print("".join(words))
parts = ["2025", "11", "15"]
print("/".join(parts))
````

**Line-by-line walkthrough**

- **Line 1 —** `` words = ["I", "love", "Python"] ``

  Create a list containing three word strings.

- **Line 2 —** `` sentence = " ".join(words) ``

  Join the words with one space between adjacent items, without a leading or trailing separator.

- **Line 3 —** `` print(sentence) ``

### find() vs index()

Both find the position of a substring, but they behave differently when the substring is not found.

| Method | Not Found | Use When |
|--------|-----------|----------|
| `find()` | Returns `-1` | You want to check without errors |
| `index()` | Raises `ValueError` | You are sure it exists |

**Figure 5.3: Finding all occurrences of a substring**

Advancing by one allows overlapping matches (`"aa"` in `"aaa"` starts at 0 and 1); `count()` counts non-overlapping matches. This example requires a non-empty search term.


````python
text = "the cat sat on the mat with the hat"
search = "the"
if not search:
    print("Enter a non-empty search term.")
positions = []
start = 0
while search:
    pos = text.find(search, start)
    if pos == -1:
        break
    positions.append(pos)
    start = pos + 1
print(f"'{search}' found at positions: {positions}")
````

  Begin searching at index zero.

- **Line 8 —** `` pos = text.find(search, start) ``

### Example 3: Character Category Checker

**Figure 9.3: Checking character categories in a sample string**

This is a string-method exercise: it reports character categories and a length threshold, not password strength. Use the supplied example text, not a real password.

````python
sample_text = "MyP@ss123"
has_upper = False
has_lower = False
has_digit = False
has_special = False
for char in sample_text:
    if char.isupper():
        has_upper = True
    elif char.islower():
        has_lower = True
    elif char.isdigit():
        has_digit = True
    elif not char.isalnum() and not char.isspace():
        has_special = True
score = sum([has_upper, has_lower, has_digit, has_special])
long_enough = len(sample_text) >= 8
print(f"Sample text: {sample_text}")
print(f"Length: {len(sample_text)} {'✓' if long_enough else '✗'}")
print(f"Uppercase: {'✓' if has_upper else '✗'}")
print(f"Lowercase: {'✓' if has_lower else '✗'}")
print(f"Digit: {'✓' if has_digit else '✗'}")
print(f"Punctuation/symbol: {'✓' if has_special else '✗'}")
print(f"Character categories present: {score}/4")
````

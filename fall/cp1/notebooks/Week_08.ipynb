## Part 2: What is a 2D List?

A **2D list** (two-dimensional list) is a list where each element is itself a list. Think of it as a **table** with rows and columns.

### Real-World Examples

| Scenario | Rows | Columns |
|---|---|---|
| Student grades | Students | Exams |
| Temperature data | Days | Hours |
| Seating chart | Rows of seats | Seats per row |
| Game board (chess) | Board rows | Board columns |

## Part 3: Accessing 2D Elements

To access a specific element in a 2D list, you need **two indices**:

```python
matrix[row][col]
```

- **First index** `[row]` selects which inner list (row)
- **Second index** `[col]` selects which element within that row (column)

**Figure 3.1: Accessing elements in a 2D list**

````python
grades = [
    [85, 90, 78],   # Row 0: Elif
    [92, 88, 95],   # Row 1: Mert
    [78, 85, 82],   # Row 2: Zeynep
    [95, 70, 88],   # Row 3: Can
    [88, 92, 90]    # Row 4: Seda
]

students = ["Elif", "Mert", "Zeynep", "Can", "Seda"]

# Access a full row
print("Elif's grades (row 0):", grades[0])        # [85, 90, 78]

# Access a specific element
print("Mert's Exam 2 (row 1, col 2):", grades[1][2])  # 95
print("Can's Exam 0 (row 3, col 0):", grades[3][0])    # 95

# Modify a specific element
print("\nBefore change — Zeynep's Exam 1:", grades[2][1])  # 85
grades[2][1] = 87
print("After change  — Zeynep's Exam 1:", grades[2][1])   # 87
````

> 💡 **Note:** `grades[1]` gives you the entire row (a list). `grades[1][2]` gives you one specific number. Always think: **row first, column second**.

## Part 4: Iterating Over 2D Lists

To process every element in a 2D list, you need **nested loops** — a loop inside a loop.

### Pattern: Row by Row

**Figure 4.1: Iterating row by row**

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Print each row
print("=== Row by row ===")
for row in matrix:
    print(row)
````

**Figure 4.2: Iterating element by element with nested loops**

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Method 1: for row in matrix, for element in row
print("=== Element by element (by value) ===")
for row in matrix:
    for element in row:
        print(element, end="  ")
    print()  # newline after each row

print()

# Method 2: Using indices (when you need row/col numbers)
print("=== Element by element (by index) ===")
for i in range(len(matrix)):          # i = row index
    for j in range(len(matrix[i])):   # j = column index
        print(f"[{i}][{j}]={matrix[i][j]}", end="  ")
    print()
````

> 💡 **Note:** The outer loop goes through **rows**, the inner loop goes through **columns** within each row. The inner loop completes entirely for each iteration of the outer loop.

## Part 5: Common 2D Operations

### Sum of All Elements

**Figure 5.1: Summing all elements in a 2D list**

````python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

# Sum all elements
total = 0
for row in matrix:
    for element in row:
        total += element
print(f"Sum of all elements: {total}")  # 450
````

**Figure 5.2: Sum per row and sum per column**

````python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

# Sum per row
print("=== Row Sums ===")
for i in range(len(matrix)):
    row_sum = 0
    for j in range(len(matrix[i])):
        row_sum += matrix[i][j]
    print(f"Row {i}: {row_sum}")

print()

# Sum per column
print("=== Column Sums ===")
num_cols = len(matrix[0])
for j in range(num_cols):          # outer loop: columns
    col_sum = 0
    for i in range(len(matrix)):   # inner loop: rows
        col_sum += matrix[i][j]
    print(f"Column {j}: {col_sum}")
````

> 💡 **Note:** For **row sums**, the outer loop goes through rows. For **column sums**, the outer loop goes through columns and the inner loop goes through rows — the loops are "flipped" in purpose!

## Part 6: Row and Column Statistics

We can calculate min, max, and average for each row or each column.

**Figure 6.1: Statistics per row**

````python
# Student grades: 4 students, 3 exams
grades = [
    [85, 90, 78],   # Elif
    [92, 88, 95],   # Mert
    [78, 65, 82],   # Zeynep
    [95, 70, 88]    # Can
]
students = ["Elif", "Mert", "Zeynep", "Can"]

print(f"{'Student':<10} {'Min':>5} {'Max':>5} {'Avg':>7}")
print("-" * 30)

for i in range(len(grades)):
    row = grades[i]
    row_min = row[0]
    row_max = row[0]
    row_sum = 0
    for val in row:
        if val < row_min:
            row_min = val
        if val > row_max:
            row_max = val
        row_sum += val
    row_avg = row_sum / len(row)
    print(f"{students[i]:<10} {row_min:>5} {row_max:>5} {row_avg:>7.1f}")
````

**Figure 6.2: Statistics per column (per exam)**

````python
grades = [
    [85, 90, 78],
    [92, 88, 95],
    [78, 65, 82],
    [95, 70, 88]
]
exam_names = ["Exam 1", "Exam 2", "Exam 3"]

print(f"{'Exam':<10} {'Min':>5} {'Max':>5} {'Avg':>7}")
print("-" * 30)

num_rows = len(grades)
num_cols = len(grades[0])

for j in range(num_cols):
    col_min = grades[0][j]
    col_max = grades[0][j]
    col_sum = 0
    for i in range(num_rows):
        val = grades[i][j]
        if val < col_min:
            col_min = val
        if val > col_max:
            col_max = val
        col_sum += val
    col_avg = col_sum / num_rows
    print(f"{exam_names[j]:<10} {col_min:>5} {col_max:>5} {col_avg:>7.1f}")
````

## Part 7: Building 2D Lists

You can build 2D lists dynamically using loops.

**Figure 7.1: Building a 2D list programmatically**

````python
# Build a 4x3 matrix filled with zeros
rows = 4
cols = 3

matrix = []
for i in range(rows):
    row = []
    for j in range(cols):
        row.append(0)
    matrix.append(row)

print("4x3 zero matrix:")
for row in matrix:
    print(row)
````

**Figure 7.2: Building a matrix with computed values**

````python
# Build a 5x5 matrix where each element = row * col
size = 5
matrix = []

for i in range(size):
    row = []
    for j in range(size):
        row.append(i * j)
    matrix.append(row)

print("Multiplication grid (i * j):")
for row in matrix:
    for val in row:
        print(f"{val:4}", end="")
    print()
````

**Figure 7.3: Simulating user input for a 2D list**

````python
# Simulated student grades input (normally you would use input())
# Here we use predefined values to demonstrate the pattern

student_names = ["Ayse", "Burak", "Ceren"]
# Simulated grades (in a real program, these would come from input())
simulated_grades = [
    [75, 80, 92],
    [88, 72, 85],
    [90, 95, 78]
]

# Build the table
grade_table = []
for i in range(len(student_names)):
    grade_table.append(simulated_grades[i])
    print(f"Added grades for {student_names[i]}: {simulated_grades[i]}")

# Display the table
print(f"\n{'Student':<10} {'Exam1':>6} {'Exam2':>6} {'Exam3':>6}")
print("-" * 30)
for i in range(len(grade_table)):
    print(f"{student_names[i]:<10}", end="")
    for grade in grade_table[i]:
        print(f"{grade:>6}", end="")
    print()
````

> 💡 **Note:** When building a 2D list, always create a **new inner list** for each row. Do NOT use `[[0]*cols]*rows` — this creates rows that all reference the same list object, leading to unexpected behavior!

## Part 8: Practical Examples

**Figure 8.1: Student grades table with full analysis**

````python
# Complete student grades analysis
students = ["Elif", "Mert", "Zeynep", "Can", "Seda"]
exams = ["Midterm 1", "Midterm 2", "Final"]
grades = [
    [85, 90, 78],
    [92, 88, 95],
    [78, 65, 82],
    [95, 70, 88],
    [88, 92, 90]
]

# Print the table
print(f"{'Student':<10}", end="")
for exam in exams:
    print(f"{exam:>12}", end="")
print(f"{'Average':>10}")
print("=" * 55)

for i in range(len(students)):
    print(f"{students[i]:<10}", end="")
    total = 0
    for j in range(len(grades[i])):
        print(f"{grades[i][j]:>12}", end="")
        total += grades[i][j]
    avg = total / len(grades[i])
    print(f"{avg:>10.1f}")

# Class average per exam
print("=" * 55)
print(f"{'Class Avg':<10}", end="")
for j in range(len(exams)):
    col_sum = 0
    for i in range(len(students)):
        col_sum += grades[i][j]
    col_avg = col_sum / len(students)
    print(f"{col_avg:>12.1f}", end="")
print()
````

**Figure 8.2: Temperature grid — daily temperatures per city**

````python
# Daily temperatures for 3 cities over 5 days
cities = ["Istanbul", "Ankara", "Izmir"]
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
temps = [
    [15, 17, 14, 18, 16],   # Istanbul
    [8,  10, 7,  12, 9],    # Ankara
    [20, 22, 19, 24, 21]    # Izmir
]

# Display the grid
print(f"{'City':<12}", end="")
for day in days:
    print(f"{day:>6}", end="")
print(f"{'Avg':>8}")
print("-" * 45)

for i in range(len(cities)):
    print(f"{cities[i]:<12}", end="")
    total = 0
    for j in range(len(temps[i])):
        print(f"{temps[i][j]:>5}°", end="")
        total += temps[i][j]
    avg = total / len(temps[i])
    print(f"{avg:>7.1f}°")
````

## Part 9: Matrix Operations

### Transpose

**Transposing** a matrix means swapping its rows and columns. Row `i` becomes column `i`.

```
Original:        Transpose:
1  2  3          1  4  7
4  5  6          2  5  8
7  8  9          3  6  9
```

**Figure 9.1: Transposing a matrix**

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

rows = len(matrix)
cols = len(matrix[0])

# Build the transpose: new matrix has cols rows and rows cols
transpose = []
for j in range(cols):
    new_row = []
    for i in range(rows):
        new_row.append(matrix[i][j])
    transpose.append(new_row)

print("Original:")
for row in matrix:
    print(row)

print("\nTranspose:")
for row in transpose:
    print(row)
````

### Diagonal Sum

The **main diagonal** of a square matrix consists of elements where the row index equals the column index: `matrix[0][0]`, `matrix[1][1]`, `matrix[2][2]`, etc.

**Figure 9.2: Summing the main diagonal**

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Main diagonal: elements where row == col
diag_sum = 0
for i in range(len(matrix)):
    print(f"Diagonal element [{i}][{i}] = {matrix[i][i]}")
    diag_sum += matrix[i][i]

print(f"\nMain diagonal sum: {diag_sum}")  # 1 + 5 + 9 = 15
````

## Part 2: Indexing and Negative Indexing

Every item has an **index**. Counting from the left starts at `0`. Counting from the right starts at `-1`.

```
temps_c = [21.5, 22.0, 22.4, 23.1, 22.8]
index      0     1     2     3     4
negative  -5    -4    -3    -2    -1
```

`temps_c[0]` is the first reading and `temps_c[-1]` is the last one. The last valid nonnegative index is `len(temps_c) - 1`. An index outside the list raises an `IndexError`.


**Figure 2.1: Reading one item by index**

````python
temps_c = [21.5, 22.0, 22.4, 23.1, 22.8]
print("First reading:", temps_c[0])
print("Third reading:", temps_c[2])
print("Last reading: ", temps_c[-1])
print("One before last:", temps_c[-2])
print("Last valid index:", len(temps_c) - 1)
````

**Line-by-line walkthrough**

- **Line 1 —** `` temps_c = [21.5, 22.0, 22.4, 23.1, 22.8] ``

  Create a five-reading temperature log; its valid nonnegative indices are 0 through 4.

- **Line 2 —** `` print("First reading:", temps_c[0]) ``

  Read index 0, the first value, 21.5.

- **Line 3 —** `` print("Third reading:", temps_c[2]) ``

  Read index 2, the third value, 22.4. List positions start at zero.

- **Line 4 —** `` print("Last reading: ", temps_c[-1]) ``

  Read index -1, the final value, 22.8.

- **Line 5 —** `` print("One before last:", temps_c[-2]) ``

  Read index -2, one position before the end, giving 23.1.

- **Line 6 —** `` print("Last valid index:", len(temps_c) - 1) ``

  Subtract one from the length to find the final nonnegative index, 4. Accessing index 5 would raise IndexError; an empty list has no valid item index.



> 💡 **Note:** Index `3` selects the **fourth** reading. Off-by-one mistakes are the most common list bug. Check `len()` before you trust an index.

## Part 3: Slicing

A **slice** selects a range of items and returns a **new list**.

```python
values[start:stop:step]
```

| Part | Meaning | Default |
|---|---|---|
| `start` | first index included | first item for positive step; last item for negative step |
| `stop` | first index **excluded** | past the last item for positive step; before the first for negative step |
| `step` | distance between selected indices | `1` |

| Slice | Meaning |
|---|---|
| `v[:3]` | first three items |
| `v[3:]` | everything from index 3 |
| `v[-3:]` | last three items |
| `v[::2]` | every second item |
| `v[::-1]` | reversed copy |
| `v[:]` | full copy |

A slice is a **shallow** copy: nested objects can still be shared. Out-of-range slice bounds are clipped; a step of zero raises `ValueError`. With a negative step, an omitted stop is different from explicitly writing `-1`: `v[::-1]` reverses the list, while `v[:-1:-1]` is empty.


**Figure 3.1: Slicing a sample log**

````python
rpm = [0, 480, 1150, 1490, 1502, 1498, 1501, 1499]
print("Samples 2 to 4:", rpm[2:5])
print("Warm-up:       ", rpm[:3])
print("Steady state:  ", rpm[3:])
print("Last two:      ", rpm[-2:])
print("Every second:  ", rpm[::2])
print("Reversed:      ", rpm[::-1])
print("Original:      ", rpm)
````

**Line-by-line walkthrough**

- **Line 1 —** `` rpm = [0, 480, 1150, 1490, 1502, 1498, 1501, 1499] ``

  Store eight speed samples in rpm, preserving their acquisition order.

- **Line 2 —** `` print("Samples 2 to 4:", rpm[2:5]) ``

  Copy indices 2, 3 and 4 into a slice; the stop index 5 is excluded.

- **Line 3 —** `` print("Warm-up:       ", rpm[:3]) ``

  Take the first three samples, indices 0 through 2, as warm-up readings.

- **Line 4 —** `` print("Steady state:  ", rpm[3:]) ``

  Take all readings from index 3 to the end as the selected steady-state segment.

- **Line 5 —** `` print("Last two:      ", rpm[-2:]) ``

  Take the last two readings using a negative starting index.

- **Line 6 —** `` print("Every second:  ", rpm[::2]) ``

  Take every second sample starting at index 0: indices 0, 2, 4 and 6.

- **Line 7 —** `` print("Reversed:      ", rpm[::-1]) ``

  Create a reversed copy using a step of -1.

- **Line 8 —** `` print("Original:      ", rpm) ``

  Print the original list to show that these slices did not change it. Slicing creates a new outer list.

## Part 4: Changing, Adding and Removing

Lists are **mutable**. Assign to an index to replace one item.

| Add | Effect |
|---|---|
| `v.append(x)` | add `x` at the end |
| `v.insert(i, x)` | add `x` at index `i` |
| `v.extend([a, b])` | add each item at the end |

| Remove | Effect | Returns |
|---|---|---|
| `v.remove(x)` | delete the first `x` | `None` |
| `v.pop(i)` | delete index `i` (default: last) | the deleted item |
| `del v[i]` | delete index `i` | nothing |

`remove(x)` raises `ValueError` if `x` is absent. `pop()` on an empty list, or `pop(i)` / `del v[i]` with an invalid index, raises `IndexError`. The add methods above return `None`; `del` is a statement, not a function call.

## Part 6: Accumulator Patterns

An **accumulator** is a variable that remembers something while the loop runs. Every pattern has the same four steps:

```python
total = 0                 # 1. initialise before the loop
for value in readings:    # 2. visit each item
    total += value        # 3. update the accumulator
print(total)              # 4. use the result after the loop
```

| Question | Accumulator | Start value | Update |
|---|---|---|---|
| How much in total? | `total` | `0` | `total += value` |
| How many match? | `count` | `0` | `count += 1` if the test is true |
| What is the average? | `total`, `count` | `0`, `0` | divide **after** the loop |
| Largest / smallest? | `highest`, `lowest` | **the first reading** | replace when a better value appears |

Start `highest` from the first reading, not from `0`. With readings `[-5, -2, -7]` a tracker started at `0` reports `0`, a value that never occurred. An empty list has no first reading: report "no data" instead.


**Figure 6.1: Sum, count and average of motor current samples**

````python
current_a = [1.8, 2.1, 2.4, 3.6, 2.9, 3.8]
limit_a = 3.0
total = 0
count_high = 0
for current in current_a:
    total += current
    if current > limit_a:
        count_high += 1
if current_a:
    average = total / len(current_a)
else:
    average = None
print(f"Total:   {total:.1f} A")
if average is None:
    print("Average: no data")
else:
    print(f"Average: {average:.2f} A")
print(f"Samples above {limit_a} A: {count_high} of {len(current_a)}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` current_a = [1.8, 2.1, 2.4, 3.6, 2.9, 3.8] ``

  Store six current readings in amperes.

- **Line 2 —** `` limit_a = 3.0 ``

  Set the comparison threshold to 3.0 A.

- **Line 3 —** `` total = 0 ``

  Initialize the sum before processing readings.

- **Line 4 —** `` count_high = 0 ``

  Initialize the count of readings strictly above the threshold.

- **Line 5 —** `` for current in current_a: ``

  Visit each current value in the log.

- **Line 6 —** `` total += current ``

  Add the current reading to the running sum.

- **Line 7 —** `` if current > limit_a: ``

  Test whether this reading exceeds 3.0; exactly 3.0 would not qualify.

- **Line 8 —** `` count_high += 1 ``

  Count one qualifying reading; the supplied data has two.

- **Line 9 —** `` if current_a: ``

  Check whether the list is nonempty before calculating its mean.

- **Line 10 —** `` average = total / len(current_a) ``

  Divide the total by the number of readings to get the arithmetic mean.

- **Line 11 —** `` else: ``

- **Line 13 —** `` print(f"Total:   {total:.1f} A") ``

  Print the sum to one decimal place. It is an intermediate sum of readings, not current at a single instant or an energy measurement.

**Figure 6.2: Maximum and minimum with their index**

````python
temps_c = [23.5, 19.8, 31.2, 27.6, 15.4, 28.9]
if not temps_c:
    print("No data")
else:
    highest = temps_c[0]
    high_index = 0
    lowest = temps_c[0]
    low_index = 0
    for i, temp in enumerate(temps_c):
        if temp > highest:
            highest = temp
            high_index = i
        if temp < lowest:
            lowest = temp
            low_index = i
    print(f"Hottest: {highest} C at index {high_index}")
    print(f"Coldest: {lowest} C at index {low_index}")
    if high_index + 1 < len(temps_c):
        print(f"What followed the peak: {temps_c[high_index + 1]} C")
    else:
        print("The peak is the last reading; no following sample.")
````

## Part 7: Search, Filter and Transform

Three more questions about the same list:

- **Search:** is a value present, and where is the first match? `x in v` answers yes or no. A loop with `break` finds the position and stops early.
- **Filter:** build a new list with only the items that pass a test.
- **Transform:** build a new list with a changed version of every item.

Filter and transform both start with an empty list and `append()` inside the loop.


**Figure 7.1: Search for the first reading above a limit**

````python
current_a = [1.8, 2.1, 2.4, 3.6, 2.9, 3.8]
print("Is 2.4 in the log?", 2.4 in current_a)
first_index = -1
for i, current in enumerate(current_a):
    if current > 3.0:
        first_index = i
        break
print("First index above 3.0 A:", first_index)
````

**Line-by-line walkthrough**

- **Line 1 —** `` current_a = [1.8, 2.1, 2.4, 3.6, 2.9, 3.8] ``

  Store six current readings in time order.

- **Line 2 —** `` print("Is 2.4 in the log?", 2.4 in current_a) ``

  Test membership: in checks whether an equal value occurs anywhere in this list and returns a boolean.

- **Line 3 —** `` first_index = -1 ``

  Use -1 as a not-found marker. Do not index the list with this marker, because Python would interpret it as the final item.

- **Line 4 —** `` for i, current in enumerate(current_a): ``

  Visit indices and current readings together.

- **Line 5 —** `` if current > 3.0: ``

  Check for a reading strictly above 3.0 A.

- **Line 6 —** `` first_index = i ``

  Save the index of the matching reading.

- **Line 7 —** `` break ``

  Stop at the first match, so later high readings do not replace it.

- **Line 8 —** `` print("First index above 3.0 A:", first_index) ``

  Print the found index, 3 here, or the -1 marker if no reading qualifies.



**Figure 7.2: Filter and transform**

````python
lengths_mm = [1200, 850, 60, 2400, 15]
long_parts = []
for length in lengths_mm:
    if length >= 1000:
        long_parts.append(length)
print("Long parts (mm):", long_parts)
lengths_m = []
for length in lengths_mm:
    lengths_m.append(length / 1000)
print("All parts (m):", lengths_m)
print("Original unchanged:", lengths_mm)
````

  Visit each original length.

- **Line 4 —** `` if length >= 1000: ``

  Accept lengths at least 1000 mm, including exactly 1000.

- **Line 5 —** `` long_parts.append(length) ``

  Append each accepted length, preserving its original order.

- **Line 6 —** `` print("Long parts (mm):", long_parts) ``

  Print the filtered lengths, still in millimetres.

- **Line 7 —** `` lengths_m = [] ``

### The short form: list comprehension

A **list comprehension** is the short form of the filter loop and the transform loop:

```python
long_parts = [length for length in lengths_mm if length >= 1000]
lengths_m = [length / 1000 for length in lengths_mm]
```

Read it as: **visit each `length` → keep it if the `if` test holds → put the first expression into the new list**. The final `if` filters; it needs no `else`. Both forms make a new list and leave the original unchanged. The full loop is always acceptable.

| `length` | `>= 1000`? | `long_parts` so far |
|---:|---|---|
| 1200 | yes | `[1200]` |
| 850 | no | `[1200]` |
| 60 | no | `[1200]` |
| 2400 | yes | `[1200, 2400]` |

**Türkçe:** Liste üreteci, filtre döngüsünün kısa yazımıdır. Sondaki `if` hangi elemanların alınacağını seçer. Önce uzun döngüyü anlayın.


**Figure 7.3: The same filter and transform as comprehensions**

````python
lengths_mm = [1200, 850, 60, 2400, 15]
long_parts = [length for length in lengths_mm if length >= 1000]
lengths_m = [length / 1000 for length in lengths_mm]
over_limit_count = len([length for length in lengths_mm if length > 100])
print("Long parts:", long_parts)
print("In metres: ", lengths_m)
print("Parts over 100 mm:", over_limit_count)
````

**Line-by-line walkthrough**

- **Line 1 —** `` lengths_mm = [1200, 850, 60, 2400, 15] ``

  Store the original millimetre lengths.

- **Line 2 —** `` long_parts = [length for length in lengths_mm if length >= 1000] ``

  Build a new list by keeping only lengths at least 1000 mm; this comprehension combines looping and filtering.

- **Line 3 —** `` lengths_m = [length / 1000 for length in lengths_mm] ``

## Part 8: Aliasing versus Copy

`view = raw` does **not** make a second list. Both names now point to the same list, so sorting `view` also sorts `raw`. `view = raw.copy()` (or `raw[:]`) creates a separate list, and the original log keeps its acquisition order.


**Figure 8.1: An alias changes the original; a copy does not**

````python
raw = [3, 1, 3]
view = raw
view.sort()
print("After sorting the alias -> raw:", raw, "view:", view)
raw = [3, 1, 3]
view = raw.copy()
view.sort()
print("After sorting the copy  -> raw:", raw, "view:", view)
print("Same object?", view is raw)
````

**Line-by-line walkthrough**

- **Line 1 —** `` raw = [3, 1, 3] ``

  Create a list whose original order is 3, 1, 3.

- **Line 2 —** `` view = raw ``

  Assign the same list object to a second name; this creates an alias, not a copy.

- **Line 3 —** `` view.sort() ``

  Sort through the alias; because both names refer to one list, raw changes too.

- **Line 4 —** `` print("After sorting the alias -> raw:", raw, "view:", view) ``

  Print both names to show that each now displays 1, 3, 3.

- **Line 5 —** `` raw = [3, 1, 3] ``

  Create a fresh list in the original order for a separate comparison.

- **Line 6 —** `` view = raw.copy() ``

  Make a shallow copy: view is a different outer list. Nested mutable items would still be shared, but these items are integers.

- **Line 7 —** `` view.sort() ``

  Sort only the copied list.

- **Line 8 —** `` print("After sorting the copy  -> raw:", raw, "view:", view) ``

  Print both lists: raw keeps 3, 1, 3 while view is sorted.

- **Line 9 —** `` print("Same object?", view is raw) ``

  Use is to compare object identity, giving False even if two lists might contain equal values.



> 💡 **Note:** `copy()` makes a shallow copy. If the list contains inner lists, those inner lists are still shared. We return to this in Week 13.

## Part 2: The Sum Pattern

The **sum pattern** is the most fundamental accumulation pattern. You initialize a variable to 0, then add values to it inside a loop.

### The Template:
```python
total = 0              # Step 1: Initialize
for item in data:      # Step 2: Loop
    total += item      # Step 3: Accumulate
print(total)           # Step 4: Use the result
```

This is like adding up a grocery receipt line by line — you start at 0 and keep adding each item's price.

**Figure 2.1: Sum of a list of numbers**

````python
# Sum pattern: add up all exam scores
scores = [85, 92, 78, 95, 88, 76, 90]

total = 0
for score in scores:
    total += score

print(f"Total of all scores: {total}")
````

**Figure 2.2: Conditional sum — sum only values meeting a condition**

````python
# Sum only the passing scores (>= 60)
scores = [85, 42, 92, 55, 78, 38, 95, 88]

passing_total = 0
for score in scores:
    if score >= 60:
        passing_total += score

print(f"Sum of passing scores: {passing_total}")
````

**Figure 2.3: Sum with range**

````python
# Sum of multiples of 3 between 1 and 50
total = 0
for i in range(1, 51):
    if i % 3 == 0:
        total += i

print(f"Sum of multiples of 3 (1-50): {total}")
````

**Figure 2.4: Product pattern (variation of sum)**

````python
# Product pattern: multiply all values (like factorial)
# Note: initialize to 1, not 0!
n = 5
factorial = 1
for i in range(1, n + 1):
    factorial *= i

print(f"{n}! = {factorial}")  # 5! = 120
````

> 💡 **Note:** For the sum pattern, always initialize to `0`. For the product pattern, always initialize to `1`. Using the wrong initial value is a common bug!

## Part 3: The Count Pattern

The **count pattern** counts how many items meet a certain condition. Initialize a counter to 0, then increment it when the condition is True.

### The Template:
```python
count = 0              # Step 1: Initialize
for item in data:      # Step 2: Loop
    if condition:       # Step 3: Check
        count += 1     # Step 4: Increment
print(count)           # Step 5: Use the result
```

**Figure 3.1: Count passing students**

````python
# Count how many students passed (score >= 60)
scores = [85, 42, 92, 55, 78, 38, 95, 88, 61, 49]

pass_count = 0
for score in scores:
    if score >= 60:
        pass_count += 1

total_students = len(scores)
fail_count = total_students - pass_count

print(f"Total students: {total_students}")
print(f"Passed: {pass_count}")
print(f"Failed: {fail_count}")
````

**Figure 3.2: Count even and odd numbers**

````python
# Count even and odd numbers in a range
even_count = 0
odd_count = 0

for i in range(1, 21):
    if i % 2 == 0:
        even_count += 1
    else:
        odd_count += 1

print(f"Even numbers (1-20): {even_count}")
print(f"Odd numbers (1-20): {odd_count}")
````

**Figure 3.3: Count with multiple categories**

````python
# Categorize exam scores
scores = [95, 82, 67, 45, 91, 73, 58, 88, 76, 52, 99, 64]

excellent = 0  # 90-100
good = 0       # 75-89
average = 0    # 60-74
failing = 0    # below 60

for score in scores:
    if score >= 90:
        excellent += 1
    elif score >= 75:
        good += 1
    elif score >= 60:
        average += 1
    else:
        failing += 1

print(f"Excellent (90+): {excellent}")
print(f"Good (75-89):    {good}")
print(f"Average (60-74): {average}")
print(f"Failing (<60):   {failing}")
````

## Part 4: The Min/Max Pattern

The **min/max pattern** finds the smallest or largest value in a collection.

### Two approaches:

**Approach 1:** Initialize with the first value
```python
max_val = data[0]           # Start with first element
for item in data[1:]:       # Check the rest
    if item > max_val:
        max_val = item
```

**Approach 2:** Initialize with infinity
```python
min_val = float('inf')      # Positive infinity
for item in data:
    if item < min_val:
        min_val = item
```

| To find | Initialize with | Compare using |
|---|---|---|
| Maximum | First value or `float('-inf')` | `>` |
| Minimum | First value or `float('inf')` | `<` |

**Figure 4.1: Find the maximum value**

````python
# Find the highest temperature
temperatures = [22, 28, 19, 35, 31, 24, 27]

max_temp = temperatures[0]   # Start with the first value
for temp in temperatures[1:]:  # Check the rest
    if temp > max_temp:
        max_temp = temp

print(f"Highest temperature: {max_temp}°C")
````

**Figure 4.2: Find the minimum value using float('inf')**

````python
# Find the lowest price
prices = [29.99, 15.50, 42.00, 8.99, 22.75]

min_price = float('inf')   # Start with infinity
for price in prices:
    if price < min_price:
        min_price = price

print(f"Cheapest item: {min_price} TL")
````

**Figure 4.3: What is float('inf')?**

````python
# float('inf') is a special value: positive infinity
# Any number is less than infinity
print(999999999 < float('inf'))    # True
print(-5 < float('inf'))           # True

# float('-inf') is negative infinity
# Any number is greater than negative infinity
print(-999999999 > float('-inf'))  # True
````

**Figure 4.4: Find both min and max in one loop**

````python
# Find both the highest and lowest scores in one pass
scores = [85, 42, 92, 55, 78, 38, 95, 88]

min_score = scores[0]
max_score = scores[0]

for score in scores[1:]:
    if score < min_score:
        min_score = score
    if score > max_score:
        max_score = score

print(f"Lowest score:  {min_score}")
print(f"Highest score: {max_score}")
print(f"Range: {max_score - min_score}")
````

> 💡 **Note:** Python has built-in `min()` and `max()` functions, but it's important to understand how they work internally. In exams, you may be asked to implement these patterns without using the built-in functions.

## Part 5: The Average Pattern

The **average pattern** combines the sum pattern and the count pattern. Calculate the sum, count the items, then divide.

### The Template:
```python
total = 0
count = 0
for item in data:
    total += item
    count += 1
average = total / count    # Watch out for division by zero!
```

**Figure 5.1: Calculate average score**

````python
# Average of Elif's quiz scores
quiz_scores = [8, 9, 7, 10, 6, 8, 9]

total = 0
count = 0
for score in quiz_scores:
    total += score
    count += 1

average = total / count
print(f"Elif's quiz average: {average:.1f}")
````

**Figure 5.2: Conditional average — average of passing scores only**

````python
# Average of passing scores only
scores = [85, 42, 92, 55, 78, 38, 95, 88]

pass_total = 0
pass_count = 0

for score in scores:
    if score >= 60:
        pass_total += score
        pass_count += 1

if pass_count > 0:    # Avoid division by zero!
    pass_avg = pass_total / pass_count
    print(f"Average of passing scores: {pass_avg:.1f}")
else:
    print("No passing scores found.")

print(f"({pass_count} out of {len(scores)} students passed)")
````

**Figure 5.3: Average with user input**

````python
# Calculate average of N numbers entered by user
n = int(input("How many numbers? "))

total = 0
for i in range(n):
    num = float(input(f"Enter number {i + 1}: "))
    total += num

if n > 0:
    average = total / n
    print(f"\nAverage: {average:.2f}")
else:
    print("No numbers entered.")
````

> 💡 **Note:** Always check for division by zero before calculating an average! If count is 0, dividing by it will cause a `ZeroDivisionError`.

## Part 6: The Search/Scan Pattern

The **search pattern** looks for a specific value or condition. There are two variations:

1. **Find first match** — stop as soon as you find one (use `break`)
2. **Find all matches** — collect all items that match (scan the entire data)

**Figure 6.1: Find first match**

````python
# Find the first student who scored above 90
names = ["Mehmet", "Zeynep", "Can", "Defne", "Burak", "Selin"]
scores = [72, 85, 91, 68, 95, 88]

found = False
for i in range(len(names)):
    if scores[i] > 90:
        print(f"First student above 90: {names[i]} ({scores[i]})")
        found = True
        break

if not found:
    print("No student scored above 90.")
````

**Figure 6.2: Find all matches**

````python
# Find ALL students who scored above 80
names = ["Mehmet", "Zeynep", "Can", "Defne", "Burak", "Selin"]
scores = [72, 85, 91, 68, 95, 88]

print("Students scoring above 80:")
found_count = 0
for i in range(len(names)):
    if scores[i] > 80:
        print(f"  {names[i]}: {scores[i]}")
        found_count += 1

if found_count == 0:
    print("  None found.")
else:
    print(f"\nTotal: {found_count} student(s)")
````

**Figure 6.3: Search with a flag variable**

````python
# Check if a number exists in data
data = [14, 27, 33, 42, 56, 68, 79, 85, 91]
target = 42

found = False
position = -1

for i in range(len(data)):
    if data[i] == target:
        found = True
        position = i
        break

if found:
    print(f"{target} found at position {position}")
else:
    print(f"{target} not found in the data")
````

## Part 7: Combining Patterns

In real programs, you often need **multiple patterns at once**. The good news is you can combine them in a single loop!

### Key Insight:
Instead of writing separate loops for sum, count, min, and max, you can do everything in **one pass** through the data.

**Figure 7.1: Sum + Count + Min + Max in one loop**

````python
# Analyze Ayşe's monthly expenses
expenses = [250, 180, 320, 95, 410, 275, 150, 380, 200, 165, 290, 350]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

total = 0
count = 0
min_expense = expenses[0]
max_expense = expenses[0]
min_month = months[0]
max_month = months[0]

for i in range(len(expenses)):
    total += expenses[i]
    count += 1
    if expenses[i] < min_expense:
        min_expense = expenses[i]
        min_month = months[i]
    if expenses[i] > max_expense:
        max_expense = expenses[i]
        max_month = months[i]

average = total / count

print(f"Total expenses: {total} TL")
print(f"Average monthly: {average:.1f} TL")
print(f"Lowest:  {min_expense} TL ({min_month})")
print(f"Highest: {max_expense} TL ({max_month})")
````

**Figure 7.2: Multiple accumulators — tracking categories**

````python
# Separate positive and negative values
values = [15, -3, 28, -7, 0, 42, -12, 8, -1, 33]

pos_sum = 0
neg_sum = 0
pos_count = 0
neg_count = 0
zero_count = 0

for val in values:
    if val > 0:
        pos_sum += val
        pos_count += 1
    elif val < 0:
        neg_sum += val
        neg_count += 1
    else:
        zero_count += 1

print(f"Positive: count={pos_count}, sum={pos_sum}")
print(f"Negative: count={neg_count}, sum={neg_sum}")
print(f"Zero:     count={zero_count}")
print(f"Net total: {pos_sum + neg_sum}")
````

**Figure 8.1: Using parallel lists for related data**

````python
# Student data stored in parallel lists
students = ["Emre", "Ayşe", "Mert", "Deniz", "Ceren", "Özge", "Ali", "Buse"]
midterm =  [75,     88,     62,     95,      70,       83,      55,     91]
final   =  [80,     92,     58,     98,      75,       79,      60,     87]

print(f"{'Student':<10} {'Midterm':<10} {'Final':<10} {'Average':<10}")
print("-" * 40)

for i in range(len(students)):
    avg = (midterm[i] + final[i]) / 2
    print(f"{students[i]:<10} {midterm[i]:<10} {final[i]:<10} {avg:<10.1f}")
````

````python
# Daily sales data for a small shop
days  = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [1200, 950, 1100, 1350, 1500, 2200, 1800]

# Find the best and worst days
best_day = days[0]
worst_day = days[0]
best_sales = sales[0]
worst_sales = sales[0]
total_sales = 0

for i in range(len(days)):
    total_sales += sales[i]
    if sales[i] > best_sales:
        best_sales = sales[i]
        best_day = days[i]
    if sales[i] < worst_sales:
        worst_sales = sales[i]
        worst_day = days[i]

avg_sales = total_sales / len(days)

print(f"Best day:  {best_day} ({best_sales} TL)")
print(f"Worst day: {worst_day} ({worst_sales} TL)")
print(f"Average:   {avg_sales:.0f} TL")
print(f"Total:     {total_sales} TL")
````

## Part 9: Building a Summary Report

Let's put everything together by building a complete summary report from data. This is a very common real-world task.

**Figure 9.1: Complete student grade report**

````python
# Complete Grade Report
students = ["Emre", "Ayşe", "Mert", "Deniz", "Ceren", "Özge", "Ali", "Buse"]
scores   = [75,      88,     42,     95,      70,       83,     55,     91]

# Initialize all our pattern variables
total = 0
count = 0
pass_count = 0
fail_count = 0
highest = scores[0]
lowest = scores[0]
top_student = students[0]
bottom_student = students[0]

# One loop does everything
for i in range(len(students)):
    score = scores[i]
    total += score
    count += 1
    
    if score >= 60:
        pass_count += 1
    else:
        fail_count += 1
    
    if score > highest:
        highest = score
        top_student = students[i]
    if score < lowest:
        lowest = score
        bottom_student = students[i]

average = total / count

# Print the report
print("=" * 35)
print("     GRADE REPORT")
print("=" * 35)
print(f"Students:    {count}")
print(f"Passed:      {pass_count}")
print(f"Failed:      {fail_count}")
print(f"Pass rate:   {pass_count / count * 100:.0f}%")
print(f"-" * 35)
print(f"Average:     {average:.1f}")
print(f"Highest:     {highest} ({top_student})")
print(f"Lowest:      {lowest} ({bottom_student})")
print("=" * 35)
````

**Figure 9.2: Weather data summary**

````python
# Weekly weather summary for Istanbul
days  = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temps = [18, 22, 25, 20, 17, 23, 26]

total = 0
hot_days = 0     # above 22
cool_days = 0    # 22 or below
min_temp = temps[0]
max_temp = temps[0]
min_day = days[0]
max_day = days[0]

for i in range(len(temps)):
    total += temps[i]
    if temps[i] > 22:
        hot_days += 1
    else:
        cool_days += 1
    if temps[i] < min_temp:
        min_temp = temps[i]
        min_day = days[i]
    if temps[i] > max_temp:
        max_temp = temps[i]
        max_day = days[i]

avg_temp = total / len(temps)

print("\u2600\ufe0f WEEKLY WEATHER REPORT")
print(f"Average temp: {avg_temp:.1f}\u00b0C")
print(f"Warmest:      {max_temp}\u00b0C ({max_day})")
print(f"Coolest:      {min_temp}\u00b0C ({min_day})")
print(f"Hot days:     {hot_days}")
print(f"Cool days:    {cool_days}")
````

### Trace: keeping two distinct leaders

For values `[4, 9, 9, 6]`, keep the largest and second-largest **different** values. A new record moves the old record into second place; a repeated record changes neither variable.

| Value | Largest after update | Second after update |
|---:|---:|---:|
| 4 | 4 | missing |
| 9 | 9 | 4 |
| 9 | 9 | 4 |
| 6 | 9 | 6 |

Pseudocode: if value exceeds first, move first to second, then replace first; otherwise, if second < value < first, replace second. Order matters: save the old first **before** overwriting it.

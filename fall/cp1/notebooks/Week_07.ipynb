## Part 1: Why a Key?

Last week a list kept readings in order. The position (index) told us *when* a reading arrived.

Often the question is not "which position?" but "which name?". We want the calibration factor of sensor `T2`, the GPIO number of pin `LED`, or the message for error code `E07`.

With a list of pairs, finding a name needs a loop.


**Figure 1.1: Finding one sensor in a list of pairs needs a loop**

````python
pairs = [["T1", 1.02], ["T2", 0.98], ["P1", 1.10]]
wanted = "T2"
factor = None
for pair in pairs:
    if pair[0] == wanted:
        factor = pair[1]
        break
print(f"Factor for {wanted}: {factor}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` pairs = [["T1", 1.02], ["T2", 0.98], ["P1", 1.10]] ``

  Store three sensor-name/factor pairs as nested lists.

- **Line 2 —** `` wanted = "T2" ``

  Choose T2 as the name to find.

- **Line 3 —** `` factor = None ``

  Use None to represent no match yet; a numeric zero would still be a valid stored value.

- **Line 4 —** `` for pair in pairs: ``

  Visit the pairs in their stored order.

- **Line 5 —** `` if pair[0] == wanted: ``

  Compare the name at position 0 with the requested name.

- **Line 6 —** `` factor = pair[1] ``

  Save the matching pair's factor from position 1.

- **Line 7 —** `` break ``

  Stop at the first match rather than continuing through later pairs.

- **Line 8 —** `` print(f"Factor for {wanted}: {factor}") ``

  Print the requested name and its factor, or None if no pair matched.



A **dictionary** (`dict`) stores **key → value** pairs. You give the key; Python finds the value. No loop is needed.


**Figure 1.2: The same lookup with a dictionary**

````python
factors = {"T1": 1.02, "T2": 0.98, "P1": 1.10}
print(f"Factor for T2: {factors['T2']}")
print(f"Factor for P1: {factors['P1']}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` factors = {"T1": 1.02, "T2": 0.98, "P1": 1.10} ``

  Map each sensor name directly to its calibration factor using a dictionary.

- **Line 2 —** `` print(f"Factor for T2: {factors['T2']}") ``

  Look up T2 by key and print 0.98; an absent key used with brackets would raise KeyError.

- **Line 3 —** `` print(f"Factor for P1: {factors['P1']}") ``

  Look up P1 by key and print its factor, 1.1.



> 💡 **Note:** In a list you ask *"what is at position 2?"*. In a dictionary you ask *"what belongs to the name T2?"*. The key describes the meaning of the value.

## Part 2: Creating and Reading a Dictionary

Write a dictionary with curly braces: `{key: value, key: value}`. Keys are usually strings or numbers. Each key appears **once**. Repeating a key when creating a dictionary keeps the last supplied value for that key. An empty dictionary is `{}`.

| Operation | Meaning |
|---|---|
| `d[key]` | The value for `key`; stops with `KeyError` if the key is absent |
| `key in d` | `True` if the key exists |
| `d.get(key)` | The value, or `None` if the key is absent |
| `d.get(key, default)` | The value, or `default` if the key is absent |
| `len(d)` | Number of entries |


**Figure 2.1: A pin-name → GPIO number map**

````python
pins = {"LED": 13, "BUTTON": 2, "BUZZER": 9}
print(pins)
print(pins["LED"])
print(len(pins))
print(type(pins))
````

**Line-by-line walkthrough**

- **Line 1 —** `` pins = {"LED": 13, "BUTTON": 2, "BUZZER": 9} ``

  Associate each device name with its GPIO pin number.

- **Line 2 —** `` print(pins) ``

  Print the complete dictionary in insertion order.

- **Line 3 —** `` print(pins["LED"]) ``

  Read the value stored under LED, giving 13.

- **Line 4 —** `` print(len(pins)) ``

  Print the number of key-value pairs, three.

- **Line 5 —** `` print(type(pins)) ``

  Print the object's type, dict.



### A missing key is an error

`d[key]` with an unknown key raises `KeyError` and stops the program. Check first with `in`.


**Figure 2.2: Checking a key before reading it**

````python
error_messages = {"E01": "Sensor timeout", "E02": "Voltage low", "E07": "Motor overheated"}
code = "E05"
print("E07" in error_messages)
print(code in error_messages)
if code in error_messages:
    print(error_messages[code])
else:
    print(f"{code}: unknown code")
````

### `.get()` gives a default instead of an error

`d.get(key, default)` returns the value if the key exists, otherwise `default`. If the key is absent and no default is supplied, it returns `None`. If a key exists with the value `None`, `get()` also returns `None`; use `key in d` to distinguish these cases.


**Figure 2.3: Reading with a default value**

````python
error_messages = {"E01": "Sensor timeout", "E02": "Voltage low", "E07": "Motor overheated"}
print(error_messages.get("E02"))
print(error_messages.get("E05"))
print(error_messages.get("E05", "unknown code"))
````

**Line-by-line walkthrough**

- **Line 1 —** `` error_messages = {"E01": "Sensor timeout", "E02": "Voltage low", "E07": "Motor overheated"} ``

  Create the error-code lookup table.

- **Line 2 —** `` print(error_messages.get("E02")) ``

  Use get() to retrieve the existing E02 description.

- **Line 3 —** `` print(error_messages.get("E05")) ``

  Request an absent key without an explicit default; get() returns None without adding the key.

- **Line 4 —** `` print(error_messages.get("E05", "unknown code")) ``

  Request that absent key with a fallback message; the dictionary itself is still unchanged.



> 💡 **Note:** Use `d[key]` when the key *must* exist (a missing key is a bug you want to see). Use `.get()` when a missing key is normal and you have a sensible default.

## Part 3: Adding, Updating and Deleting Entries

Assigning to a key **adds** the entry if the key is new, or **updates** it if the key exists. `del d[key]` removes an entry. `d.pop(key)` removes it and gives you the value.

A dictionary remembers the **insertion order**. New keys go to the end. Updating a value does not move the key.

Deleting and reinserting a key places it at the end. `del d[key]` and `d.pop(key)` raise `KeyError` for an absent key; `d.pop(key, default)` returns the default instead.


**Figure 3.1: Updating a part inventory**

````python
inventory = {"M3 bolt": 120, "M3 nut": 80}
inventory["washer"] = 200
inventory["M3 bolt"] = inventory["M3 bolt"] - 20
inventory["M3 nut"] += 40
print(inventory)
del inventory["washer"]
print(inventory)
removed = inventory.pop("M3 nut")
print(removed, inventory)
````

**Line-by-line walkthrough**

- **Line 1 —** `` inventory = {"M3 bolt": 120, "M3 nut": 80} ``

  Create the initial stock quantities for two parts.

- **Line 2 —** `` inventory["washer"] = 200 ``

  Add a new washer key with quantity 200.

- **Line 3 —** `` inventory["M3 bolt"] = inventory["M3 bolt"] - 20 ``

  Read the bolt quantity, subtract 20, and store the updated quantity under the same key.

- **Line 4 —** `` inventory["M3 nut"] += 40 ``

  Increase the nut quantity by 40 using an augmented assignment.

- **Line 5 —** `` print(inventory) ``

  Print the updated inventory, including washers.

- **Line 6 —** `` del inventory["washer"] ``

  Delete the washer key and its value; deleting an absent key this way would raise KeyError.

- **Line 7 —** `` print(inventory) ``

  Print the dictionary after deletion.

- **Line 8 —** `` removed = inventory.pop("M3 nut") ``

  Remove the nut entry and keep its previous quantity, 120, in removed.

- **Line 9 —** `` print(removed, inventory) ``

  Print the removed quantity and the remaining inventory.



**Figure 3.2: Insertion order is kept**

````python
latest = {}
latest["T2"] = 22.0
latest["T1"] = 21.5
latest["T3"] = 19.8
latest["T2"] = 22.4
print(latest)
print(list(latest))
````

**Line-by-line walkthrough**

- **Line 1 —** `` latest = {} ``

  Start an empty dictionary for the latest sensor readings.

- **Line 2 —** `` latest["T2"] = 22.0 ``

  Insert T2 first with value 22.0.

- **Line 3 —** `` latest["T1"] = 21.5 ``

## Part 4: Iterating Over a Dictionary

A `for` loop over a dictionary visits the **keys**. Three methods give other views:

| Loop | Each turn gives |
|---|---|
| `for key in d:` | a key |
| `for key in d.keys():` | a key (same as above) |
| `for value in d.values():` | a value |
| `for key, value in d.items():` | a key and its value together |

Do not add or delete dictionary keys while iterating over that same dictionary; this can raise `RuntimeError`. The tally examples iterate over a separate log while updating the tally.


**Figure 4.1: Three ways to loop over a unit conversion table**

### Counting with a dictionary

A very common pattern: the key is the thing you count, the value is how many times you saw it. If the key is new, start at 0.


**Figure 4.2: Tally of status codes from a log**

````python
status_log = ["OK", "OK", "WARN", "OK", "ERR", "WARN", "OK"]
tally = {}
for code in status_log:
    if code not in tally:
        tally[code] = 0
    tally[code] += 1
print(tally)
tally2 = {}
for code in status_log:
    tally2[code] = tally2.get(code, 0) + 1
print(tally2 == tally)
````

**Line-by-line walkthrough**

- **Line 1 —** `` status_log = ["OK", "OK", "WARN", "OK", "ERR", "WARN", "OK"] ``

  Store the seven status observations in order.

- **Line 2 —** `` tally = {} ``

  Create an empty dictionary of counts.

- **Line 3 —** `` for code in status_log: ``

  Visit each observed status code.

- **Line 4 —** `` if code not in tally: ``

  Check whether this code has no count yet.

- **Line 5 —** `` tally[code] = 0 ``

  Initialize a newly seen code's count to zero.

- **Line 6 —** `` tally[code] += 1 ``

  Add one for the current observation, whether the key was new or already present.

- **Line 7 —** `` print(tally) ``

  Print the completed counts: OK 4, WARN 2 and ERR 1.

- **Line 8 —** `` tally2 = {} ``

  Start a separate dictionary to demonstrate the shorter equivalent pattern.

- **Line 9 —** `` for code in status_log: ``

  Visit the same status observations again.

- **Line 10 —** `` tally2[code] = tally2.get(code, 0) + 1 ``

  Get the previous count or zero, add one, and store it. get() alone would not insert the missing key; the assignment does.

- **Line 11 —** `` print(tally2 == tally) ``

  Compare the two dictionaries by their key-value contents, giving True.



**Figure 4.3: The most frequent code**

````python
tally = {"OK": 4, "WARN": 2, "ERR": 1}
most_common = None
for code, count in tally.items():
    if most_common is None or count > tally[most_common]:
        most_common = code
if most_common is None:
    print("No status codes to count.")
else:
    print(f"Most frequent: {most_common} ({tally[most_common]} times)")
````

## Part 5: Grouping Data

A value can be a **list**. A dictionary of lists groups records that share a key, for example all readings of one sensor.

The pattern has three steps: if the key is new, create an empty list; then append to the list for that key; finally read each group.


**Figure 5.1: Grouping readings by sensor**

````python
readings = [["T1", 21.5], ["T2", 22.0], ["T1", 21.9], ["T3", 19.8], ["T2", 22.4]]
groups = {}
for reading in readings:
    sensor = reading[0]
    value = reading[1]
    if sensor not in groups:
        groups[sensor] = []
    groups[sensor].append(value)
print(groups)
for sensor, values in groups.items():
    mean = sum(values) / len(values)
    print(f"{sensor}: {len(values)} readings, mean {mean:.2f}")
````

- **Line 4 —** `` sensor = reading[0] ``

  Read its sensor name from position zero.

- **Line 5 —** `` value = reading[1] ``

- **Line 6 —** `` if sensor not in groups: ``

## Part 6: Tuples: Fixed Records

A **tuple** is a sequence like a list, but its element references **cannot be replaced, added or removed** after creation. A mutable object inside a tuple, such as a list, can still change. Write it with parentheses: `(1200, 23.5)`. Use a tuple for a small fixed record, such as `(time, value)` or `(row, col)`.

- Index a tuple like a list: `pair[0]`.
- **Unpack** it into names: `t, v = pair`.
- `pair[1] = 24.0` raises `TypeError`, because tuples are immutable.
- `d.items()` gives `(key, value)` tuples; that is why `for key, value in d.items()` works.

The comma makes a non-empty tuple: `(5,)` is a one-item tuple, while `(5)` is the integer 5. Ordinary unpacking needs the same number of names and items; otherwise it raises `ValueError`.


**Figure 6.1: Creating, indexing and unpacking a tuple**

````python
sample = (1200, 23.5)
print(sample[0], sample[1])
print(type(sample))
print(len(sample))
t, v = sample
print(f"t = {t} ms, T = {v} C")
````

**Line-by-line walkthrough**

- **Line 1 —** `` sample = (1200, 23.5) ``

  Create a tuple containing time in milliseconds and temperature in Celsius. Its item references cannot be reassigned.

- **Line 2 —** `` print(sample[0], sample[1]) ``

  Read both tuple positions and print their values.

- **Line 3 —** `` print(type(sample)) ``

  Print the object's type, tuple.

- **Line 4 —** `` print(len(sample)) ``

  Print its number of items, two.

- **Line 5 —** `` t, v = sample ``

  Unpack the two items into t and v; the number of variables must match the number of items.

- **Line 6 —** `` print(f"t = {t} ms, T = {v} C") ``

  Print labelled time and temperature. Assigning sample[1] would raise TypeError; create a new tuple to represent a changed record.



**Figure 6.2: A list of (time, value) tuples, unpacked in a loop**

````python
samples = [(0, 20.1), (10, 22.4), (20, 25.9), (30, 24.0)]
for t, v in samples:
    print(f"t = {t:>2} s   T = {v:.1f} C")
first_t, first_v = samples[0]
last_t, last_v = samples[-1]
print(f"Change: {last_v - first_v:.1f} C in {last_t - first_t} s")
````

## Part 7: Choosing List, Dictionary or Tuple

| Structure | Use it when | Example |
|---|---|---|
| `list` | Order matters, the size changes, you find items by position | readings in time order |
| `dict` | You find values by a **name**, each name appears once | calibration factor per sensor |
| `tuple` | A small **fixed** sequence of fields; nested mutable objects may still change | `(time, value)`, `(row, col)` |

The three often work together: a list of tuples keeps every observation in order; a dictionary keeps one summary per name.


**Figure 7.1: All three structures in one small log**

````python
log = [("T1", 21.5), ("T2", 22.0), ("T1", 21.9), ("T2", 22.4), ("T1", 22.3)]
latest = {}
count = {}
for sensor, value in log:
    latest[sensor] = value
    count[sensor] = count.get(sensor, 0) + 1
print("Observations:", len(log))
print("Latest:", latest)
print("Count: ", count)
````

**Line-by-line walkthrough**

- **Line 1 —** `` log = [("T1", 21.5), ("T2", 22.0), ("T1", 21.9), ("T2", 22.4), ("T1", 22.3)] ``

  Store all observations as a list of sensor-value tuples, including repeated sensor names.

- **Line 2 —** `` latest = {} ``

  Create a dictionary for the last encountered reading per sensor.

- **Line 3 —** `` count = {} ``

  Create a separate dictionary for the observation count per sensor.

- **Line 4 —** `` for sensor, value in log: ``

  Unpack each observation into its sensor name and value in log order.

- **Line 5 —** `` latest[sensor] = value ``

  Replace that sensor's stored value. This is newest only when log order represents acquisition order; older observations still remain in log.

- **Line 6 —** `` count[sensor] = count.get(sensor, 0) + 1 ``

  Increment the sensor's count, starting from zero for its first appearance.

- **Line 7 —** `` print("Observations:", len(log)) ``

  Print the total number of observations, including repeated names.

- **Line 8 —** `` print("Latest:", latest) ``

  Print the last value encountered for each sensor.

- **Line 9 —** `` print("Count: ", count) ``

  Print the number of observations assigned to each sensor.

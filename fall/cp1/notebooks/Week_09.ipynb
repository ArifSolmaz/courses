## Part 1: Why Functions?

Imagine you need to calculate the area of a rectangle in many places in your program. Without functions, you would copy and paste the same code again and again.

**Problems with copy-pasting:**
- If you find a bug, you must fix it in every copy
- Your code becomes long and hard to read
- Changes are error-prone

**Functions solve these problems by letting you:**
- Write the code once, use it many times
- Give a name to a block of code
- Organize your program into logical pieces
- Test each piece independently

**Figure 1.1: Without functions — repeated code**

````python
width1 = 5
height1 = 3
area1 = width1 * height1
print(f"Area 1: {area1}")
width2 = 10
height2 = 7
area2 = width2 * height2
print(f"Area 2: {area2}")
width3 = 8
height3 = 4
area3 = width3 * height3
print(f"Area 3: {area3}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` width1 = 5 ``

  Set the first rectangle's width to 5.

- **Line 2 —** `` height1 = 3 ``

  Set its height to 3, using the same length unit as the width.

- **Line 3 —** `` area1 = width1 * height1 ``

  Multiply the dimensions to calculate area in square units.

- **Line 4 —** `` print(f"Area 1: {area1}") ``

  Print the first area, 15.

- **Line 5 —** `` width2 = 10 ``

  Set the second width to 10.

- **Line 6 —** `` height2 = 7 ``

  Set its height to 7.

- **Line 7 —** `` area2 = width2 * height2 ``

  Repeat the multiplication for this rectangle.

- **Line 8 —** `` print(f"Area 2: {area2}") ``

  Print the second area, 70.

- **Line 9 —** `` width3 = 8 ``

  Set the third width to 8.

- **Line 10 —** `` height3 = 4 ``

  Set its height to 4.

- **Line 11 —** `` area3 = width3 * height3 ``

  Repeat the same calculation again; this duplication motivates a reusable function.

- **Line 12 —** `` print(f"Area 3: {area3}") ``

  Print the third area, 32.

  Call the same function with 10 and 7 to print 70.

## Part 2: Defining Functions

A function definition has these parts:

```
def function_name(parameters):
    """Docstring (optional but recommended)"""
    # body — indented code
    return value  # optional
```

| Part | Description |
|------|-------------|
| `def` | Keyword that starts a function definition |
| `function_name` | Name you choose (follow variable naming rules) |
| `parameters` | Inputs the function receives (in parentheses) |
| `:` | Colon after the parentheses |
| body | Indented code that runs when the function is called |
| `return` | Sends a value back to the caller (optional) |

**Figure 2.1: A simple function — definition and call**

````python
def say_hello():
    print("Hello! Welcome to the course.")
say_hello()
say_hello()
````

### Returning Multiple Values

`return smallest, largest` returns one tuple containing two values. The caller can unpack that tuple into two names.


**Figure 4.2: Returning a pair**

For an empty list, this function returns `(None, None)` because no minimum or maximum exists. Check for `None` before using those values in calculations.


````python
def min_max(numbers):
    """Return (minimum, maximum), or (None, None) for an empty list."""
    if not numbers:
        return None, None
    smallest = numbers[0]
    largest = numbers[0]
    for n in numbers:
        if n < smallest:
            smallest = n
        if n > largest:
            largest = n
    return smallest, largest
grades = [85, 92, 78, 95, 60, 88]
low, high = min_max(grades)
print(f"Lowest: {low}, Highest: {high}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` def min_max(numbers): ``

  Define a function that receives a list of numbers.

- **Line 2 —** `` """Return (minimum, maximum), or (None, None) for an empty list.""" ``

  Document the result contract: a pair of extrema, or two None values when the list is empty.

- **Line 3 —** `` if not numbers: ``

  Check for an empty list before accessing its first item.

- **Line 4 —** `` return None, None ``

  Return a two-item tuple containing None values and exit immediately for empty input.

- **Line 5 —** `` smallest = numbers[0] ``

  Initialize the minimum from actual data rather than zero.

- **Line 6 —** `` largest = numbers[0] ``

  Initialize the maximum from the same first item.

- **Line 7 —** `` for n in numbers: ``

  Visit every number.

- **Line 8 —** `` if n < smallest: ``

  Check whether it is below the current minimum.

- **Line 9 —** `` smallest = n ``

  Replace the minimum when that test succeeds.

- **Line 10 —** `` if n > largest: ``

  Independently check whether it is above the current maximum.

- **Line 11 —** `` largest = n ``

### What Happens Without return?

````python
def say_hi(name):
    print(f"Hi, {name}!")
result = say_hi("Deniz")
print(f"Return value: {result}")
print(f"Type: {type(result)}")
````

### Follow a value through a function call

For `def double(number): return number * 2`, calling `double(7)` binds `number` to 7, computes 14, and gives 14 back to the caller. `answer = double(7)` stores that returned value; `print(answer)` displays it.

| Step | Inside the function | Caller |
|---|---|---|
| call | `number = 7` | waits for result |
| calculate | `number * 2 = 14` | still waiting |
| return | sends 14 back | `answer = 14` |

Pseudocode: receive input → calculate → return result. Keep printing outside when callers may need to reuse the result. A function that only prints and reaches its end returns `None`.

## Part 6: Functions That Print vs Return

Printing displays a value; returning gives a value to the caller. A function can do both, but keeping calculation and display separate makes reuse easier.

| | Print-only function | Function returning a number |
|---|---|---|
| Display | Shows text when called | Caller decides whether to print |
| Assignment | Stores `None` if the function reaches its end | Stores the returned number |
| Next calculation | `None + 1` raises `TypeError` | The number can be used in arithmetic |

`print()` itself returns `None`. `return` may return any object, including `None`; it does not guarantee a numeric result.


**Figure 6.1: Print vs Return — the critical difference**

````python
def add_print(a, b):
    print(a + b)
def add_return(a, b):
    return a + b
result1 = add_print(3, 4)
print(f"result1 = {result1}")
print()
result2 = add_return(3, 4)
print(f"result2 = {result2}")
total = add_return(3, 4) + add_return(5, 6)
print(f"total = {total}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` def add_print(a, b): ``

  Define a function intended only to display a sum.

- **Line 2 —** `` print(a + b) ``

  Print the sum; the function implicitly returns None afterward.

- **Line 3 —** `` def add_return(a, b): ``

  Define a separate calculation function with the same two parameters.

- **Line 4 —** `` return a + b ``

  Return the sum without printing it.

- **Line 5 —** `` result1 = add_print(3, 4) ``

  Call the display function: it prints 7, while result1 receives None.

- **Line 6 —** `` print(f"result1 = {result1}") ``

  Print result1 to reveal that no numerical result was returned. Adding two add_print() results would attempt None + None and raise TypeError.

- **Line 7 —** `` print() ``

  Print a blank line separating the demonstrations.

- **Line 8 —** `` result2 = add_return(3, 4) ``

  Call the returning version and store 7; the call itself prints nothing.

- **Line 9 —** `` print(f"result2 = {result2}") ``

  Print its reusable numerical return value.

- **Line 10 —** `` total = add_return(3, 4) + add_return(5, 6) ``

  Add two returned sums, 7 and 11, giving 18.

- **Line 11 —** `` print(f"total = {total}") ``

  Print the combined result.



> 💡 **Note:** As a rule of thumb: most functions should **return** values. Only use `print()` inside a function when the purpose of the function is specifically to display something to the user.

## Part 7: Documenting Functions

A **docstring** is a string literal that is the first statement in a function body. Describe the parameters, units, returned value and invalid-input policy.


**Figure 7.1: Documenting the force-sensor conversion**

The opening calibration gives `(100 - 0) / (4.5 - 0.5) = 25 N/V`. Only 0.5–4.5 V is supported by this teaching model. An out-of-range reading returns `None`, not a force of zero. Inputs are finite numeric voltages.


````python
def voltage_to_force(voltage):
    """Return force in N for a finite voltage in the calibrated range.

    Calibration: 0.5 V = 0 N; 4.5 V = 100 N, linear in between.
    Return None outside the inclusive 0.5–4.5 V range.
    """
    if not 0.5 <= voltage <= 4.5:
        return None
    return (voltage - 0.5) * 25
for voltage in [0.5, 2.5, 4.5, 5.0]:
    force = voltage_to_force(voltage)
    if force is None:
        print(f"{voltage:.1f} V: outside calibrated range")
    else:
        print(f"{voltage:.1f} V: {force:.1f} N")
````

### Bridge Exercise: Preview of Scope

Next week, we will learn about **scope** — where variables are visible in your program.

Run the code below and try to predict what will happen. Can the function see the variable `x` defined outside? Can the main code see the variable `y` defined inside the function?

**Expected Output:**
```
Inside function: x = 10
Inside function: y = 20
Outside function: x = 10
```

<details><summary>💡 Hint</summary>
The assignment to <code>y</code> creates a local name. With no global <code>y</code>, uncommenting the last line raises <code>NameError</code>. If a global <code>y</code> already exists, that separate value is printed instead. Functions can read globals; rebinding a global name requires <code>global</code>, while mutating an existing list is a different operation. We will explore this in depth next week.
</details>

````python
x = 10
def my_function():
    y = 20
    print(f"Inside function: x = {x}")
    print(f"Inside function: y = {y}")
my_function()
print(f"Outside function: x = {x}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` x = 10 ``

  Bind x to 10 in the surrounding notebook namespace.

- **Line 2 —** `` def my_function(): ``

  Define a function with no parameters.

- **Line 3 —** `` y = 20 ``

  Create y as a local variable belonging to this call.

- **Line 4 —** `` print(f"Inside function: x = {x}") ``

  Read the surrounding x, since this function does not assign its own local x.

- **Line 5 —** `` print(f"Inside function: y = {y}") ``

  Print the local y, 20.

- **Line 6 —** `` my_function() ``

  Call the function, executing both internal print statements.

- **Line 7 —** `` print(f"Outside function: x = {x}") ``

  Print the surrounding x, which remains 10. Accessing y outside would raise NameError in a fresh namespace; a pre-existing global y would be a separate variable, not this local one.

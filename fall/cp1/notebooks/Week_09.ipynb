## Part 1: Review — Strings Basics

Strings are sequences of characters. We have used them since Week 1, but now we will explore them in depth.

### Creating Strings

You can create strings with single quotes, double quotes, or triple quotes for multi-line text.

**Figure 1.1: Creating strings in different ways**

````python
# Single quotes
name = 'Elif'

# Double quotes
city = "Istanbul"

# Triple quotes for multi-line
poem = """Roses are red,
Violets are blue,
Python is fun,
And so are you!"""

print(name)
print(city)
print(poem)
````

### Indexing

Each character in a string has a position (index). Indexing starts at **0**.

| Index | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
|-------|---|---|---|---|---|---|---|---|
| Char  | I | s | t | a | n | b | u | l |
| Neg.  | -8| -7| -6| -5| -4| -3| -2| -1|

**Figure 1.2: Accessing characters by index**

````python
city = "Istanbul"

print(city[0])     # First character: I
print(city[3])     # Fourth character: a
print(city[-1])    # Last character: l
print(city[-2])    # Second to last: u
````

### Slicing

Slicing extracts a portion of a string using `[start:stop:step]`.

- `start` — where the slice begins (inclusive)
- `stop` — where the slice ends (exclusive)
- `step` — how many characters to skip

**Figure 1.3: String slicing examples**

````python
text = "Hello, World!"

print(text[0:5])     # Hello
print(text[7:12])    # World
print(text[:5])      # Hello (start defaults to 0)
print(text[7:])      # World! (stop defaults to end)
print(text[::2])     # Hlo ol! (every 2nd character)
print(text[::-1])    # !dlroW ,olleH (reversed)
````

> 💡 **Note:** Strings are **immutable** in Python. You cannot change a character directly. For example, `text[0] = 'h'` will cause an error. You must create a new string instead.

## Part 2: String Methods

Python provides many built-in methods to work with strings. A **method** is a function that belongs to an object — you call it with a dot: `string.method()`.

### Case Methods

| Method | Description | Example | Result |
|--------|-------------|---------|--------|
| `upper()` | All uppercase | `"hello".upper()` | `"HELLO"` |
| `lower()` | All lowercase | `"HELLO".lower()` | `"hello"` |
| `title()` | Capitalize each word | `"hello world".title()` | `"Hello World"` |
| `capitalize()` | Capitalize first letter | `"hello world".capitalize()` | `"Hello world"` |
| `swapcase()` | Swap upper/lower | `"Hello".swapcase()` | `"hELLO"` |

**Figure 2.1: Case conversion methods**

````python
student = "mehmet kaya"

print(student.upper())       # MEHMET KAYA
print(student.lower())       # mehmet kaya
print(student.title())       # Mehmet Kaya
print(student.capitalize())  # Mehmet kaya
print(student.swapcase())    # MEHMET KAYA
````

### Whitespace Methods

| Method | Description |
|--------|-------------|
| `strip()` | Remove whitespace from both ends |
| `lstrip()` | Remove whitespace from the left |
| `rstrip()` | Remove whitespace from the right |

**Figure 2.2: Removing whitespace with strip()**

````python
messy = "   Hello, Zeynep!   "

print(f"'{messy}'")
print(f"'{messy.strip()}'")
print(f"'{messy.lstrip()}'")
print(f"'{messy.rstrip()}'")
````

### Search and Replace Methods

| Method | Description |
|--------|-------------|
| `find(sub)` | Returns index of first occurrence, or -1 if not found |
| `index(sub)` | Like find(), but raises error if not found |
| `count(sub)` | Counts non-overlapping occurrences |
| `replace(old, new)` | Replaces all occurrences |
| `startswith(prefix)` | Returns True if string starts with prefix |
| `endswith(suffix)` | Returns True if string ends with suffix |

**Figure 2.3: Searching and replacing in strings**

````python
sentence = "Python is fun and Python is powerful"

print(sentence.find("Python"))        # 0 (first occurrence)
print(sentence.find("Java"))          # -1 (not found)
print(sentence.count("Python"))       # 2
print(sentence.replace("Python", "Programming"))
print(sentence.startswith("Python"))  # True
print(sentence.endswith("powerful"))  # True
````

> 💡 **Note:** `replace()` does not change the original string. It returns a **new** string. Remember, strings are immutable!

## Part 3: split() and join()

These two methods are among the most important string operations.

### split() — String to List

`split()` breaks a string into a list of substrings. By default, it splits on whitespace.

**Figure 3.1: Splitting strings into lists**

````python
sentence = "Ayse likes to code in Python"
words = sentence.split()
print(words)        # ['Ayse', 'likes', 'to', 'code', 'in', 'Python']
print(len(words))   # 6

# Split on a specific character
date = "2025-11-15"
parts = date.split("-")
print(parts)        # ['2025', '11', '15']

# Split CSV-like data
csv_line = "Burak,22,Istanbul,Engineering"
fields = csv_line.split(",")
print(fields)       # ['Burak', '22', 'Istanbul', 'Engineering']
````

### join() — List to String

`join()` is the opposite of `split()`. It combines a list of strings into one string, with a separator between each element.

The syntax is: `separator.join(list)`

**Figure 3.2: Joining lists into strings**

````python
words = ["I", "love", "Python"]

sentence = " ".join(words)
print(sentence)     # I love Python

# Join with different separators
print("-".join(words))    # I-love-Python
print(", ".join(words))   # I, love, Python
print("".join(words))     # IlovePython

# Rebuild a date
parts = ["2025", "11", "15"]
print("/".join(parts))    # 2025/11/15
````

> 💡 **Note:** `join()` only works with lists of **strings**. If your list contains numbers, convert them first: `", ".join(str(x) for x in numbers)`

## Part 4: String Iteration

You can iterate over a string character by character using a `for` loop. This is one of the most powerful techniques for string processing.

**Figure 4.1: Iterating over characters in a string**

````python
name = "Deniz"

# Method 1: Direct iteration
for char in name:
    print(char, end=" ")
print()  # D e n i z

# Method 2: Using index
for i in range(len(name)):
    print(f"Index {i}: {name[i]}")
````

**Figure 4.2: Counting specific characters**

````python
text = "Hello, how are you today?"
vowels = "aeiouAEIOU"
vowel_count = 0

for char in text:
    if char in vowels:
        vowel_count += 1

print(f"The text has {vowel_count} vowels.")  # 9
````

**Figure 4.3: Building a new string character by character**

````python
# Remove all spaces from a string
original = "H e l l o   W o r l d"
no_spaces = ""

for char in original:
    if char != " ":
        no_spaces += char

print(no_spaces)  # HelloWorld
````

## Part 5: String Searching

Python provides several ways to search for text within strings.

### The `in` Operator

The simplest way to check if a substring exists.

**Figure 5.1: Using the `in` operator for searching**

````python
message = "Welcome to Computer Programming I"

print("Computer" in message)    # True
print("computer" in message)    # False (case-sensitive!)
print("Java" in message)        # False
print("Java" not in message)    # True

# Common pattern: case-insensitive search
search = "computer"
if search.lower() in message.lower():
    print(f"Found '{search}' in the message!")
````

### find() vs index()

Both find the position of a substring, but they behave differently when the substring is not found.

| Method | Not Found | Use When |
|--------|-----------|----------|
| `find()` | Returns `-1` | You want to check without errors |
| `index()` | Raises `ValueError` | You are sure it exists |

**Figure 5.2: Finding positions of substrings**

````python
text = "Learning Python is fun. Python is everywhere."

# find() — safe, returns -1 if not found
pos = text.find("Python")
print(f"First 'Python' at index: {pos}")   # 9

# Find second occurrence
pos2 = text.find("Python", pos + 1)
print(f"Second 'Python' at index: {pos2}")  # 23

# Not found
pos3 = text.find("Java")
print(f"'Java' at index: {pos3}")            # -1
````

**Figure 5.3: Finding all occurrences of a substring**

````python
text = "the cat sat on the mat with the hat"
search = "the"

positions = []
start = 0

while True:
    pos = text.find(search, start)
    if pos == -1:
        break
    positions.append(pos)
    start = pos + 1

print(f"'{search}' found at positions: {positions}")  # [0, 15, 28]
````

## Part 6: String Formatting Review

f-strings (formatted string literals) are the modern way to format strings in Python. We have used them before, but let's review some advanced features.

### Basic f-string Syntax

Put `f` before the opening quote and use `{}` to embed expressions.

**Figure 6.1: f-string basics and formatting**

````python
name = "Selin"
age = 20
gpa = 3.7542

# Basic embedding
print(f"Name: {name}, Age: {age}")

# Expressions inside {}
print(f"Next year: {age + 1}")
print(f"Name in upper: {name.upper()}")

# Number formatting
print(f"GPA: {gpa:.2f}")            # 3.75 (2 decimal places)
print(f"GPA: {gpa:.1f}")            # 3.8 (1 decimal place)

# Alignment and padding
print(f"{'Left':<15}|")
print(f"{'Center':^15}|")
print(f"{'Right':>15}|")

# Number formatting with separators
population = 85000000
print(f"Population: {population:,}")  # 85,000,000
````

**Figure 6.2: Formatted table with f-strings**

````python
students = [
    ("Emre Yilmaz", 85, 92, 78),
    ("Sude Demir", 90, 88, 95),
    ("Cem Ozturk", 72, 65, 80),
]

print(f"{'Name':<15} {'Midterm':>8} {'Final':>8} {'HW':>8} {'Avg':>8}")
print("-" * 52)

for name, midterm, final, hw in students:
    avg = (midterm + final + hw) / 3
    print(f"{name:<15} {midterm:>8} {final:>8} {hw:>8} {avg:>8.1f}")
````

## Part 7: Parsing Structured Text

Many real-world tasks involve reading and processing structured text data. CSV (Comma-Separated Values) is one of the most common formats.

**Figure 7.1: Parsing a CSV-like line**

````python
# A CSV line with student data
csv_line = "2024001234,Fatma Sahin,Computer Engineering,3.45"

# Split into fields
fields = csv_line.split(",")

student_id = fields[0]
name = fields[1]
department = fields[2]
gpa = float(fields[3])

print(f"Student: {name}")
print(f"ID: {student_id}")
print(f"Department: {department}")
print(f"GPA: {gpa:.2f}")
````

**Figure 7.2: Processing multiple CSV lines**

````python
data = """Ali Koc,85,92,78
Zeynep Arslan,90,88,95
Murat Celik,72,65,80
Beren Yildiz,88,91,85"""

lines = data.strip().split("\n")

total_avg = 0
count = 0

for line in lines:
    parts = line.split(",")
    name = parts[0]
    scores = [int(parts[1]), int(parts[2]), int(parts[3])]
    avg = sum(scores) / len(scores)
    total_avg += avg
    count += 1
    print(f"{name:<18} Average: {avg:.1f}")

print(f"\nClass average: {total_avg / count:.1f}")
````

## Part 8: Building Strings

When you need to construct a string piece by piece, you have two main approaches.

### Approach 1: Concatenation in a Loop

**Figure 8.1: Building strings with concatenation**

````python
# Build a string of even numbers
result = ""
for i in range(2, 11, 2):
    result += str(i) + " "

print(result.strip())  # 2 4 6 8 10
````

### Approach 2: The join() Pattern

This is the **preferred** method. Build a list, then join it.

**Figure 8.2: Building strings with join() — the better way**

````python
# Build a string of even numbers — join pattern
parts = []
for i in range(2, 11, 2):
    parts.append(str(i))

result = " ".join(parts)
print(result)  # 2 4 6 8 10

# Another example: build a comma-separated list of names
names = ["Arda", "Buse", "Can", "Defne"]
name_list = ", ".join(names)
print(f"Students: {name_list}")  # Students: Arda, Buse, Can, Defne
````

> 💡 **Note:** For large strings, `join()` is much faster than repeated `+=` concatenation. Make it a habit to use the join pattern.

## Part 9: Practical Examples

Let's apply what we have learned to real-world scenarios.

### Example 1: Log Line Parser

**Figure 9.1: Parsing server log lines**

````python
log_lines = [
    "2025-01-15 08:30:15,sensor_A,23.5",
    "2025-01-15 08:30:16,sensor_B,45.2",
    "2025-01-15 08:30:17,sensor_A,24.1",
    "2025-01-15 08:30:18,sensor_C,18.7",
    "2025-01-15 08:30:19,sensor_B,44.8",
]

print(f"{'Timestamp':<22} {'Sensor':<12} {'Value':>8}")
print("-" * 44)

for line in log_lines:
    parts = line.split(",")
    timestamp = parts[0]
    sensor = parts[1]
    value = float(parts[2])
    print(f"{timestamp:<22} {sensor:<12} {value:>8.1f}")
````

### Example 2: Simple Data Extraction

**Figure 9.2: Extracting email addresses from text**

````python
text = """Contact us at support@example.com for help.
Send feedback to feedback@example.com.
For sales, reach out to sales@example.com."""

words = text.split()
emails = []

for word in words:
    # Clean punctuation from the end
    clean_word = word.strip(".,!?;:")
    if "@" in clean_word and "." in clean_word:
        emails.append(clean_word)

print("Found emails:")
for email in emails:
    print(f"  - {email}")
````

### Example 3: Password Strength Checker

**Figure 9.3: Checking password strength**

````python
password = "MyP@ss123"

has_upper = False
has_lower = False
has_digit = False
has_special = False

for char in password:
    if char.isupper():
        has_upper = True
    elif char.islower():
        has_lower = True
    elif char.isdigit():
        has_digit = True
    else:
        has_special = True

score = sum([has_upper, has_lower, has_digit, has_special])
long_enough = len(password) >= 8

print(f"Password: {password}")
print(f"Length: {len(password)} {'✓' if long_enough else '✗'}")
print(f"Uppercase: {'✓' if has_upper else '✗'}")
print(f"Lowercase: {'✓' if has_lower else '✗'}")
print(f"Digit: {'✓' if has_digit else '✗'}")
print(f"Special: {'✓' if has_special else '✗'}")
print(f"Strength: {score}/4 — {'Strong' if score == 4 and long_enough else 'Weak'}")
````

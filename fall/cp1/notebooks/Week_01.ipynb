````python
# Run once at the start of class
_checkpoint_results = {}

def record_checkpoint(number, checks):
    """Show immediate feedback and keep results only in this Colab runtime."""
    results = [(str(label), bool(result)) for label, result in checks]
    passed = sum(result for _, result in results)
    total = len(results)
    for label, result in results:
        print(("✅" if result else "❌"), label)
    _checkpoint_results[int(number)] = (passed, total)
    print(f"Checkpoint {number}: {passed}/{total} passed")
    if passed == total:
        print("🟢 Ready to continue.")
    elif passed == 0:
        print("🔴 Review the preceding examples, correct your work, and run this cell again.")
    else:
        print("🟡 Almost there—correct the failed item and try again.")
    return passed, total

def exercise_checkpoint(number, expected):
    """Count exercise cells that contain work and were executed in this runtime."""
    import re
    completed = set()
    for source in globals().get("In", []):
        match = re.search(r"#\s*✏️\s*\[EX(\d+)\]", str(source))
        if not match:
            continue
        answer = re.sub(
            r"^.*?#\s*✏️\s*\[EX\d+\]", "", str(source),
            count=1, flags=re.S
        ).strip()
        if answer and answer != "pass":
            completed.add(int(match.group(1)))
    checks = [
        (f"Exercise {index} edited and run", index in completed)
        for index in range(1, expected + 1)
    ]
    return record_checkpoint(number, checks)

def show_progress_summary():
    print("\n=== My checkpoint summary ===")
    total_passed = 0
    total_checks = 0
    for number in range(1, 6):
        if number not in _checkpoint_results:
            print(f"Checkpoint {number}: not run yet")
            continue
        passed, total = _checkpoint_results[number]
        total_passed += passed
        total_checks += total
        print(f"Checkpoint {number}: {passed}/{total}")
    if total_checks:
        print(f"Total so far: {total_passed}/{total_checks}")
    print("These results exist only in your current runtime.")

print("✅ Self-check tools ready")
````

### The `sep` Parameter

The `sep` parameter controls what goes **between** multiple values (default is a space).

- Temporarily disable code during testing

## Part 5: Variables

A **variable** is a named container that stores a value. Think of it as a labeled box where you can put data.

### Creating Variables

In Python, you create a variable simply by assigning a value to a name with `=`.

**Figure 1.9** — Creating and using variables

````python
# Creating variables
student_name = "Ayse"
age = 19
gpa = 3.45
is_enrolled = True

# Using variables
print("Name:", student_name)
print("Age:", age)
print("GPA:", gpa)
print("Enrolled:", is_enrolled)
````

### Variable Naming Rules

| Rule | Valid | Invalid |
|---|---|---|
| Must start with letter or `_` | `name`, `_count` | `2name` |
| Can contain letters, digits, `_` | `student_1` | `student-1` |
| Case-sensitive | `Age` ≠ `age` | — |
| Cannot be a Python keyword | `my_class` | `class` |

### Python Keywords (Reserved Words)

These words have special meaning and **cannot** be used as variable names:

```
False  True   None   and    or     not    if     else   elif
for    while  break  continue  return  def   class  import  from
```

### Naming Conventions

| Convention | Example | When to Use |
|---|---|---|
| `snake_case` | `student_name` | Variables, functions |
| `UPPER_CASE` | `MAX_SPEED` | Constants |
| `PascalCase` | `StudentRecord` | Classes (later in course) |

> 💡 **Tip:** Always use descriptive names! `speed` is better than `s`, and `student_count` is better than `sc`.

### Multiple Assignment & Swapping

**Figure 1.10** — Multiple assignment and variable swapping

````python
# Multiple assignment in one line
x, y, z = 10, 20, 30
print("x =", x, "| y =", y, "| z =", z)

# Same value to multiple variables
a = b = c = 0
print("a =", a, "| b =", b, "| c =", c)

# Swapping variables (Python makes this easy!)
first = "Mehmet"
second = "Ali"
print("Before swap:", first, second)

first, second = second, first
print("After swap:", first, second)
````

### Changing Variable Values

**Figure 1.11** — Variables can change

````python
# Variables can be reassigned
score = 85
print("Original score:", score)

score = 92
print("Updated score:", score)

# Variables can even change type (but this is not recommended)
data = 42
print("data is", type(data))

data = "Hello"
print("data is now", type(data))
````

## Part 6: Data Types

Every value in Python has a **type**. The four basic types are:

| Type | Name | Example | Description |
|---|---|---|---|
| `int` | Integer | `42`, `-7`, `0` | Whole numbers |
| `float` | Float | `3.14`, `-0.5`, `2.0` | Decimal numbers |
| `str` | String | `"Hello"`, `'Python'` | Text |
| `bool` | Boolean | `True`, `False` | Yes/No values |

> 💡 **Note:** `True` and `False` must be capitalized. `true` or `TRUE` will cause an error.

### The `type()` Function

**Figure 1.12** — Checking data types with `type()`

````python
# Check the type of different values
print(type(42))        # <class 'int'>
print(type(3.14))      # <class 'float'>
print(type("Hello"))   # <class 'str'>
print(type(True))      # <class 'bool'>

# Check the type of variables
student_name = "Zeynep"
student_age = 21
student_gpa = 3.67
is_active = True

print(type(student_name))  # str
print(type(student_age))   # int
print(type(student_gpa))   # float
print(type(is_active))     # bool
````

### Important Type Details

**Figure 1.13** — Subtle type differences

````python
# 5 is an int, 5.0 is a float
print(type(5))    # int
print(type(5.0))  # float

# Strings with numbers are still strings!
print(type("42"))   # str, NOT int
print(type("3.14")) # str, NOT float

# Boolean values are special
print(type(True))   # bool
print(type("True")) # str (it's text, not a boolean!)
````

## Part 7: Basic Arithmetic

Python can work as a powerful calculator! Here are the arithmetic operators:

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `10 - 4` | `6` |
| `*` | Multiplication | `6 * 7` | `42` |
| `/` | Division | `15 / 4` | `3.75` |
| `//` | Floor Division | `15 // 4` | `3` |
| `%` | Modulo (remainder) | `15 % 4` | `3` |
| `**` | Exponentiation | `2 ** 3` | `8` |

**Figure 1.14** — Basic arithmetic operations

````python
# Addition and Subtraction
print("5 + 3 =", 5 + 3)
print("10 - 4 =", 10 - 4)

# Multiplication
print("6 * 7 =", 6 * 7)

# Division (always returns a float)
print("15 / 4 =", 15 / 4)
print("10 / 2 =", 10 / 2)  # Even this is a float: 5.0

# Floor Division (rounds down to integer)
print("15 // 4 =", 15 // 4)
print("17 // 5 =", 17 // 5)

# Modulo (remainder after division)
print("15 % 4 =", 15 % 4)
print("10 % 3 =", 10 % 3)

# Exponentiation (power)
print("2 ** 3 =", 2 ** 3)   # 2 cubed
print("5 ** 2 =", 5 ** 2)   # 5 squared
````

### Operator Precedence (PEMDAS)

Python follows standard math rules for the order of operations:

| Priority | Operator | Name |
|---|---|---|
| 1 (highest) | `()` | Parentheses |
| 2 | `**` | Exponentiation |
| 3 | `*`, `/`, `//`, `%` | Multiplication, Division |
| 4 (lowest) | `+`, `-` | Addition, Subtraction |

> 💡 **Tip:** When in doubt, use parentheses `()` to make your intent clear!

**Figure 1.15** — Operator precedence examples

````python
# Without parentheses: multiplication first, then addition
result1 = 2 + 3 * 4
print("2 + 3 * 4 =", result1)  # 14, not 20

# With parentheses: addition first
result2 = (2 + 3) * 4
print("(2 + 3) * 4 =", result2)  # 20

# More examples
print("10 - 2 * 3 =", 10 - 2 * 3)        # 4
print("(10 - 2) * 3 =", (10 - 2) * 3)    # 24
print("2 ** 3 + 1 =", 2 ** 3 + 1)        # 9
print("2 ** (3 + 1) =", 2 ** (3 + 1))    # 16
````

### Engineering Examples

**Figure 1.16** — Arithmetic in engineering contexts

````python
# Distance = Speed x Time
speed = 60       # km/h
time = 2.5       # hours
distance = speed * time
print("Distance:", distance, "km")

# Area of a circle: A = pi * r^2
pi = 3.14159
radius = 5       # meters
area = pi * radius ** 2
print("Circle area:", area, "m^2")

# Ohm's Law: V = I * R
current = 2.5    # amperes
resistance = 100 # ohms
voltage = current * resistance
print("Voltage:", voltage, "volts")

# Celsius to Fahrenheit: F = C * 9/5 + 32
celsius = 37
fahrenheit = celsius * 9/5 + 32
print(celsius, "C =", fahrenheit, "F")
````

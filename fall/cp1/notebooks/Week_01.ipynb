## Part 4: Comments

Comments are notes in your code that Python **ignores**. They are for humans, not the computer.

### Why Write Comments?
- Explain **what** your code does and **why**
- Help others (and your future self!) understand the code
- Temporarily disable code during testing

### Single-Line Comments

Use the `#` symbol. Everything after `#` on that line is ignored.

### Variable Naming Rules

| Rule | Valid | Invalid |
|---|---|---|
| Must start with letter or `_` | `name`, `_count` | `2name` |
| Can contain letters, digits, `_` | `student_1` | `student-1` |
| Case-sensitive | `Age` ≠ `age` | — |
| Cannot be a Python keyword | `my_class` | `class` |

### Python Keywords (Reserved Words)

These words have special meaning and **cannot** be used as variable names:

```
False  True   None   and    or     not    if     else   elif
for    while  break  continue  return  def   class  import  from
```

### Naming Conventions

| Convention | Example | When to Use |
|---|---|---|
| `snake_case` | `student_name` | Variables, functions |
| `UPPER_CASE` | `MAX_SPEED` | Constants |
| `PascalCase` | `StudentRecord` | Classes (later in course) |

> 💡 **Tip:** Always use descriptive names! `speed` is better than `s`, and `student_count` is better than `sc`.

### Trace assignment before you run

Goal: increase a stored count by one. Python evaluates the right side using the old value, then binds the name to the result.

| Statement | Read on the right | Value stored afterward |
|---|---|---|
| `count = 4` | `4` | `4` |

| `double = count * 2` | `5 * 2` | `double = 10`; `count = 5` |

### Operator Precedence (PEMDAS)

Python follows standard math rules for the order of operations:

| Priority | Operator | Name |
|---|---|---|
| 1 (highest) | `()` | Parentheses |
| 2 | `**` | Exponentiation |
| 3 | `*`, `/`, `//`, `%` | Multiplication, Division |
| 4 (lowest) | `+`, `-` | Addition, Subtraction |

> 💡 **Tip:** When in doubt, use parentheses `()` to make your intent clear!

**Figure 1.15** — Operator precedence examples

````python
result1 = 2 + 3 * 4
print("2 + 3 * 4 =", result1)
result2 = (2 + 3) * 4
print("(2 + 3) * 4 =", result2)
print("10 - 2 * 3 =", 10 - 2 * 3)
print("(10 - 2) * 3 =", (10 - 2) * 3)
print("2 ** 3 + 1 =", 2 ** 3 + 1)
print("2 ** (3 + 1) =", 2 ** (3 + 1))
````

### Calculation and display are different steps

For compound interest, translate the formula in order: convert 15% to `0.15`; add one; raise to the number of years; multiply by the principal. In Python, `**` means a power; `^` does not.

Binary floating-point numbers can leave a tiny display tail: `10000 * 1.15**3` may print `15208.749999999998`. The arithmetic is correct to ordinary display precision. `round(amount, 2)` gives a number rounded to two decimal places; next week f-strings will control the displayed number of digits.

## Part 3: Comparison Operators

Comparison operators compare two values and return a **boolean** (`True` or `False`).

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `<` | Less than | `3 < 5` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<=` | Less than or equal | `5 <= 5` | `True` |
| `>=` | Greater than or equal | `3 >= 5` | `False` |

> 💡 **Warning:** Don't confuse `=` (assignment) with `==` (comparison)!
> - `x = 5` means "set x to 5"
> - `x == 5` means "is x equal to 5?"

**Figure 2.4** — Comparison operators in action

````python
print("5 == 5:", 5 == 5)
print("5 == 3:", 5 == 3)
print("5 != 3:", 5 != 3)
print("5 > 3:",  5 > 3)
print("5 < 3:",  5 < 3)
print("5 >= 5:", 5 >= 5)
print("5 <= 4:", 5 <= 4)
print()
ali_score = 85
ayse_score = 92
print(f"Ali: {ali_score}, Ayse: {ayse_score}")
print(f"Ali > Ayse? {ali_score > ayse_score}")
print(f"Ali == Ayse? {ali_score == ayse_score}")
````

### Format Specifications

F-strings support powerful formatting with the `:` syntax inside `{}`.

| Format | Meaning | Example | Output |
|---|---|---|---|

### Common String Methods

## Part 2: Arithmetic Operators in Depth

We introduced the 7 arithmetic operators last week. Now let's explore them in more detail.

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `7 + 3` | `10` |
| `-` | Subtraction | `7 - 3` | `4` |
| `*` | Multiplication | `7 * 3` | `21` |
| `/` | Division | `7 / 3` | `2.3333...` |
| `//` | Floor Division | `7 // 3` | `2` |
| `%` | Modulo | `7 % 3` | `1` |
| `**` | Exponentiation | `7 ** 3` | `343` |

### Integer vs Float Division

This is one of the most important distinctions for beginners.

**Figure 2.2** — Division: `/` vs `//`

````python
# Regular division (/) ALWAYS returns a float
print("10 / 3 =", 10 / 3)    # 3.3333...
print("10 / 2 =", 10 / 2)    # 5.0 (still a float!)
print("9 / 3 =", 9 / 3)      # 3.0

print()

# Floor division (//) rounds DOWN to nearest integer
print("10 // 3 =", 10 // 3)   # 3
print("10 // 2 =", 10 // 2)   # 5
print("7 // 2 =", 7 // 2)     # 3 (rounds down)

print()

# Negative floor division rounds toward negative infinity
print("-7 // 2 =", -7 // 2)   # -4 (not -3!)
````

### Modulo Use Cases

The modulo operator `%` gives the **remainder** after division. It is extremely useful!

| Use Case | How | Example |
|---|---|---|
| Check even/odd | `n % 2` | `0` = even, `1` = odd |
| Extract digits | `n % 10` | Last digit of n |
| Time breakdown | `sec % 60` | Remaining seconds |
| Wrap around | `i % length` | Cyclic indexing |

**Figure 2.3** — Practical uses of the modulo operator

````python
# Check even or odd
number = 17
remainder = number % 2
print(f"{number} % 2 = {remainder}")  # 1, so it's odd

number = 24
remainder = number % 2
print(f"{number} % 2 = {remainder}")  # 0, so it's even

print()

# Time breakdown: 4567 seconds to hours, minutes, seconds
total_seconds = 4567
hours = total_seconds // 3600
remaining = total_seconds % 3600
minutes = remaining // 60
seconds = remaining % 60

print(f"{total_seconds} seconds = {hours}h {minutes}m {seconds}s")

print()

# Extract last digit
number = 12345
last_digit = number % 10
print(f"Last digit of {number} is {last_digit}")
````

## Part 3: Comparison Operators

Comparison operators compare two values and return a **boolean** (`True` or `False`).

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `<` | Less than | `3 < 5` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<=` | Less than or equal | `5 <= 5` | `True` |
| `>=` | Greater than or equal | `3 >= 5` | `False` |

> 💡 **Warning:** Don't confuse `=` (assignment) with `==` (comparison)!
> - `x = 5` means "set x to 5"
> - `x == 5` means "is x equal to 5?"

**Figure 2.4** — Comparison operators in action

````python
# Basic comparisons
print("5 == 5:", 5 == 5)    # True
print("5 == 3:", 5 == 3)    # False
print("5 != 3:", 5 != 3)    # True
print("5 > 3:",  5 > 3)     # True
print("5 < 3:",  5 < 3)     # False
print("5 >= 5:", 5 >= 5)    # True
print("5 <= 4:", 5 <= 4)    # False

print()

# Comparing variables
ali_score = 85
ayse_score = 92
print(f"Ali: {ali_score}, Ayse: {ayse_score}")
print(f"Ali > Ayse? {ali_score > ayse_score}")
print(f"Ali == Ayse? {ali_score == ayse_score}")
````

### Chained Comparisons

Python allows you to chain comparisons, which is very convenient for range checks.

**Figure 2.5** — Chained comparisons

````python
# Chained comparisons (unique to Python!)
age = 20

# Instead of: age >= 18 and age <= 65
print(f"Is {age} between 18 and 65? {18 <= age <= 65}")

temperature = 22
print(f"Is {temperature} between 20 and 25? {20 <= temperature <= 25}")

# You can chain more than two
x = 5
print(f"1 < {x} < 10 < 20: {1 < x < 10 < 20}")
````

## Part 4: Assignment Operators

Assignment operators provide a shorthand for updating a variable's value.

| Operator | Equivalent | Example |
|---|---|---|
| `x += 5` | `x = x + 5` | Add and assign |
| `x -= 3` | `x = x - 3` | Subtract and assign |
| `x *= 2` | `x = x * 2` | Multiply and assign |
| `x /= 4` | `x = x / 4` | Divide and assign |
| `x //= 3` | `x = x // 3` | Floor divide and assign |
| `x %= 7` | `x = x % 7` | Modulo and assign |
| `x **= 2` | `x = x ** 2` | Exponent and assign |

**Figure 2.6** — Assignment operators

````python
# Starting value
score = 100
print("Start:", score)

score += 10    # score = score + 10
print("After += 10:", score)  # 110

score -= 25    # score = score - 25
print("After -= 25:", score)  # 85

score *= 2     # score = score * 2
print("After *= 2:", score)   # 170

score //= 3    # score = score // 3
print("After //= 3:", score)  # 56

score %= 10    # score = score % 10
print("After %= 10:", score)  # 6

score **= 3    # score = score ** 3
print("After **= 3:", score)  # 216
````

**Figure 2.7** — Practical example: counting and accumulating

````python
# Practical example: tracking a bank balance
balance = 5000  # Starting balance in TL
print(f"Initial balance: {balance} TL")

balance += 1500  # Salary deposit
print(f"After salary: {balance} TL")

balance -= 800   # Rent payment
print(f"After rent: {balance} TL")

balance -= 350   # Grocery shopping
print(f"After groceries: {balance} TL")

balance += 200   # Freelance work
print(f"After freelance: {balance} TL")
````

## Part 5: Operator Precedence

When an expression has multiple operators, Python follows these priority rules:

| Priority | Operators | Description |
|---|---|---|
| 1 (highest) | `()` | Parentheses |
| 2 | `**` | Exponentiation |
| 3 | `+x`, `-x` | Unary plus/minus |
| 4 | `*`, `/`, `//`, `%` | Multiplication, division |
| 5 | `+`, `-` | Addition, subtraction |
| 6 | `==`, `!=`, `<`, `>`, `<=`, `>=` | Comparisons |
| 7 (lowest) | `=`, `+=`, `-=`, etc. | Assignment |

> 💡 **Best Practice:** When in doubt, use parentheses! They make your code clearer and prevent mistakes.

**Figure 2.8** — Operator precedence with complex expressions

````python
# Complex expressions - predict the result before running!

# Example 1: * before +
result = 2 + 3 * 4 - 1
print(f"2 + 3 * 4 - 1 = {result}")  # 2 + 12 - 1 = 13

# Example 2: ** before *
result = 2 * 3 ** 2
print(f"2 * 3 ** 2 = {result}")  # 2 * 9 = 18

# Example 3: Parentheses override everything
result = (2 + 3) * (4 - 1)
print(f"(2 + 3) * (4 - 1) = {result}")  # 5 * 3 = 15

# Example 4: Mixed
result = 10 + 20 * 30 / 10 - 5
print(f"10 + 20 * 30 / 10 - 5 = {result}")  # 10 + 60.0 - 5 = 65.0

# Example 5: Use parentheses for clarity
# Physics: v = v0 + a*t
v0 = 10  # initial velocity (m/s)
a = 9.81 # acceleration (m/s^2)
t = 3    # time (seconds)
v = v0 + (a * t)  # Parentheses make intent clear
print(f"Final velocity: {v} m/s")
````

## Part 6: Type Conversion

Sometimes you need to convert a value from one type to another. Python provides built-in functions for this.

| Function | Converts to | Example | Result |
|---|---|---|---|
| `int()` | Integer | `int(3.7)` | `3` |
| `float()` | Float | `float(5)` | `5.0` |
| `str()` | String | `str(42)` | `"42"` |
| `bool()` | Boolean | `bool(1)` | `True` |

**Figure 2.9** — Type conversion examples

````python
# int() - convert to integer
print(int(3.7))     # 3 (truncates, does NOT round!)
print(int(3.2))     # 3
print(int(-3.7))    # -3 (truncates toward zero)
print(int("42"))    # 42 (string to int)
print(int(True))    # 1
print(int(False))   # 0

print()

# float() - convert to float
print(float(5))       # 5.0
print(float("3.14"))  # 3.14
print(float("42"))    # 42.0

print()

# str() - convert to string
print(str(42))     # "42"
print(str(3.14))   # "3.14"
print(str(True))   # "True"
````

### Common Traps!

> 💡 **Warning:** Not all conversions work. Some will cause errors!

**Figure 2.10** — Type conversion traps

````python
# These WORK
print(int("42"))       # 42
print(float("3.14"))   # 3.14

# These FAIL (uncomment to see errors)
# print(int("3.5"))    # ValueError! Can't convert "3.5" directly to int
# print(int("hello"))  # ValueError! Not a number at all
# print(int(""))       # ValueError! Empty string

# Fix for "3.5" to int: convert to float first, then to int
print(int(float("3.5")))  # 3 (two-step conversion)

print()

# Truncation vs Rounding
print("int(3.9) =", int(3.9))       # 3 (truncates!)
print("round(3.9) =", round(3.9))   # 4 (rounds)
print("round(3.5) =", round(3.5))   # 4 (rounds to nearest even)
print("round(2.5) =", round(2.5))   # 2 (rounds to nearest even!)
````

### Boolean Conversion

The `bool()` function follows simple rules:

| Value | `bool()` result | Rule |
|---|---|---|
| `0`, `0.0` | `False` | Zero is False |
| `""` (empty string) | `False` | Empty is False |
| `None` | `False` | None is False |
| Any non-zero number | `True` | Non-zero is True |
| Any non-empty string | `True` | Non-empty is True |

**Figure 2.11** — Boolean conversion

````python
# Falsy values (become False)
print(bool(0))       # False
print(bool(0.0))     # False
print(bool(""))      # False

print()

# Truthy values (become True)
print(bool(1))       # True
print(bool(-5))      # True
print(bool(3.14))    # True
print(bool("hello")) # True
print(bool(" "))     # True (space is not empty!)
````

## Part 8: F-Strings (Formatted String Literals)

F-strings are the modern, recommended way to format strings in Python. They start with `f` before the opening quote.

### Basic Syntax

```python
f"text {expression} more text"
```

Anything inside `{}` is evaluated as a Python expression.

**Figure 2.15** — Basic f-string usage

````python
name = "Mehmet"
age = 21
gpa = 3.456

# Basic f-string
print(f"My name is {name}")
print(f"I am {age} years old")

# Expressions inside {}
print(f"Next year I will be {age + 1}")
print(f"5 x 7 = {5 * 7}")

# Variables and math combined
width = 12
height = 8
print(f"Rectangle {width}x{height} has area {width * height}")
````

### Format Specifications

F-strings support powerful formatting with the `:` syntax inside `{}`.

| Format | Meaning | Example | Output |
|---|---|---|---|
| `:.2f` | 2 decimal places | `f"{3.14159:.2f}"` | `3.14` |
| `:.4f` | 4 decimal places | `f"{3.14159:.4f}"` | `3.1416` |
| `:>10` | Right-align, width 10 | `f"{"Hi":>10}"` | `"        Hi"` |
| `:<10` | Left-align, width 10 | `f"{"Hi":<10}"` | `"Hi        "` |
| `:^10` | Center, width 10 | `f"{"Hi":^10}"` | `"    Hi    "` |
| `:,` | Thousands separator | `f"{1000000:,}"` | `1,000,000` |
| `:.1%` | Percentage | `f"{0.856:.1%}"` | `85.6%` |
| `:.2e` | Scientific notation | `f"{12345:.2e}"` | `1.23e+04` |

**Figure 2.16** — Decimal place formatting

````python
pi = 3.14159265358979

print(f"No format:    {pi}")
print(f"1 decimal:    {pi:.1f}")
print(f"2 decimals:   {pi:.2f}")
print(f"4 decimals:   {pi:.4f}")
print(f"0 decimals:   {pi:.0f}")

print()

# Engineering example
voltage = 220.4567
current = 3.1289
power = voltage * current

print(f"Voltage: {voltage:.1f} V")
print(f"Current: {current:.2f} A")
print(f"Power:   {power:.2f} W")
````

**Figure 2.17** — Alignment and width formatting

````python
# Right-align numbers for a clean table
print(f"{'Item':<15} {'Price':>10}")
print(f"{'-'*15} {'-'*10}")
print(f"{'Laptop':<15} {'12,500':>10}")
print(f"{'Mouse':<15} {'250':>10}")
print(f"{'Keyboard':<15} {'450':>10}")
print(f"{'Monitor':<15} {'3,200':>10}")
````

**Figure 2.18** — Thousands separator, percentage, and scientific notation

````python
# Thousands separator
population = 85000000
print(f"Turkey population: {population:,}")
print(f"Turkey population: {population:_}")  # Underscore variant

print()

# Percentage
success_rate = 0.856
print(f"Success rate: {success_rate:.1%}")   # 85.6%
print(f"Success rate: {success_rate:.0%}")   # 86%

print()

# Scientific notation (useful in engineering!)
speed_of_light = 299792458
electron_mass = 0.000000000000000000000000000000911
print(f"Speed of light: {speed_of_light:.2e} m/s")
print(f"Electron mass: {electron_mass:.3e} kg")
````

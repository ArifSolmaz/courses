## Part 1: Review — for Loops

In the previous weeks, we learned about `for` loops. Let's do a quick recap.

A `for` loop repeats a block of code **a known number of times**. It iterates over a sequence (like `range()`).

| Feature | Description |
|---|---|
| **Syntax** | `for variable in sequence:` |
| **Use when** | You know how many times to repeat |
| **Common with** | `range()`, strings, lists |
| **Loop variable** | Automatically updated each iteration |

**Figure 1.1: Quick for loop recap**

````python
# Print numbers 1 to 5 using a for loop
for i in range(1, 6):
    print(i)
````

**Figure 1.2: Summing numbers with a for loop**

````python
# Sum of 1 to 10
total = 0
for i in range(1, 11):
    total += i
print("Sum:", total)  # 55
````

> 💡 **Note:** `for` loops are great when you know the number of iterations in advance. But what if you **don't know** how many times to repeat? That's where `while` loops come in!

## Part 2: The while Loop

A `while` loop repeats a block of code **as long as a condition is True**. Unlike `for` loops, you don't need to know how many iterations ahead of time.

### Syntax

```python
while condition:
    # code block
    # (must eventually make condition False!)
```

### How it works:
1. Check the condition
2. If `True`, execute the body
3. Go back to step 1
4. If `False`, skip the body and continue after the loop

### When to use `while` vs `for`?

| Use `for` when... | Use `while` when... |
|---|---|
| You know the number of iterations | You don't know when to stop |
| Iterating over a sequence | Waiting for a condition to change |
| Counting from A to B | User input validation |
| Processing each item in a collection | Running until a sentinel value |

**Figure 2.1: Basic while loop — counting from 1 to 5**

````python
# Count from 1 to 5 using while
count = 1
while count <= 5:
    print(count)
    count += 1    # Don't forget this!

print("Done! count is now:", count)
````

**Figure 2.2: Comparing for and while for the same task**

````python
# for loop version
print("=== for loop ===")
for i in range(1, 6):
    print(i)

# while loop version (same result)
print("\n=== while loop ===")
i = 1
while i <= 5:
    print(i)
    i += 1
````

> 💡 **Note:** For counting tasks, `for` loops are usually cleaner. Use `while` when the number of iterations depends on a condition that changes during execution.

## Part 3: while Loop Examples

Let's look at several practical examples where `while` loops shine.

**Figure 3.1: Countdown timer**

````python
# Countdown from 10 to 1
countdown = 10
while countdown >= 1:
    print(countdown, end=" ")
    countdown -= 1
print("\nBlast off! 🚀")
````

**Figure 3.2: Simple guessing game**

````python
# Guessing game - try to guess the secret number
secret = 7
guess = int(input("Guess a number between 1 and 10: "))

while guess != secret:
    if guess < secret:
        print("Too low!")
    else:
        print("Too high!")
    guess = int(input("Try again: "))

print("Correct! The number was", secret)
````

**Figure 3.3: Input validation**

````python
# Keep asking until user enters a valid age (1-120)
age = int(input("Enter your age (1-120): "))

while age < 1 or age > 120:
    print("Invalid age! Must be between 1 and 120.")
    age = int(input("Enter your age (1-120): "))

print(f"Your age is {age}. Thank you!")
````

**Figure 3.4: Powers of 2**

````python
# Print powers of 2 that are less than 1000
power = 1
while power < 1000:
    print(power, end=" ")
    power *= 2

print("\nNext power would be:", power)
````

### A while loop needs a route toward stopping

Pseudocode: choose a starting value; test the condition; do one step; update the value; return to the condition. For a countdown from 3:

| Value before test | `counter > 0` | Output | Value after update |
|---:|---|---:|---:|
| 3 | True | 3 | 2 |
| 2 | True | 2 | 1 |
| 1 | True | 1 | 0 |
| 0 | False | nothing | loop ends |

If the update is absent, the condition never changes. If `continue` skips the update, that iteration can repeat forever. **TR:** Döngüyü yazarken hangi satırın durma koşuluna yaklaştırdığını göster; yalnızca koşulu yazmak yetmez.

## Part 4: Infinite Loops

An **infinite loop** is a loop that never stops because its condition **never becomes False**.

### Common causes:
- Forgetting to update the loop variable
- Updating the variable in the wrong direction
- A condition that can never be False

- In Jupyter/Colab: Click the **stop button** (square icon) or go to **Runtime > Interrupt execution**
- In terminal: Press **Ctrl+C**

## Part 5: The break Statement

The `break` statement **immediately exits** the loop, no matter what the loop condition says.

### When to use `break`:
- You found what you were looking for (search pattern)
- The user wants to quit
- An error condition occurred
- You want to exit a `while True` loop

```python
while condition:
    # some code
    if something:
        break       # exit the loop immediately
    # more code
```

**Figure 5.1: Using break to exit when found**

````python
# Search for the first number divisible by 7 between 50 and 100
number = 50
while number <= 100:
    if number % 7 == 0:
        print(f"Found it! {number} is divisible by 7")
        break
    number += 1

print("Search complete.")
````

**Figure 5.2: while True with break — the classic pattern**

````python
# Keep asking for input until user types 'quit'
while True:
    text = input("Enter something (or 'quit' to stop): ")
    if text.lower() == "quit":
        break
    print(f"You entered: {text}")

print("Goodbye!")
````

**Figure 5.3: break in a for loop**

````python
# break also works in for loops
# Find the first even number in a sequence
numbers = [3, 7, 11, 4, 9, 2, 8]

for num in numbers:
    if num % 2 == 0:
        print(f"First even number found: {num}")
        break
else:
    print("No even number found.")
````

> 💡 **Note:** The `else` clause of a `for` or `while` loop runs only if the loop completed without hitting `break`. This is a unique Python feature!

## Part 6: The continue Statement

The `continue` statement **skips the rest of the current iteration** and jumps to the next one.

### When to use `continue`:
- Skip unwanted values (filtering)
- Avoid deeply nested if statements
- Skip invalid data

```python
while condition:
    # some code
    if skip_this:
        continue    # skip to next iteration
    # this code only runs when NOT skipped
```

**Figure 6.1: Skip odd numbers using continue**

````python
# Print only even numbers from 1 to 10
for i in range(1, 11):
    if i % 2 != 0:   # if odd
        continue       # skip this iteration
    print(i, end=" ")

print()  # new line
````

**Figure 6.2: continue in a while loop — be careful!**

````python
# Print numbers 1-10, but skip multiples of 3
i = 0
while i < 10:
    i += 1          # Update BEFORE continue!
    if i % 3 == 0:
        continue     # Skip multiples of 3
    print(i, end=" ")

print()
````

> 💡 **Note:** When using `continue` in a `while` loop, make sure you update the loop variable **before** the `continue` statement. Otherwise, you'll create an infinite loop!

**Figure 6.3: Comparing break vs continue**

````python
# break: stops the entire loop
print("=== break ===")
for i in range(1, 11):
    if i == 5:
        break
    print(i, end=" ")
print()

# continue: skips one iteration, loop goes on
print("=== continue ===")
for i in range(1, 11):
    if i == 5:
        continue
    print(i, end=" ")
print()
````

| Statement | Effect | Loop continues? |
|---|---|---|
| `break` | Exit the loop immediately | No |
| `continue` | Skip rest of current iteration | Yes, next iteration |

## Part 7: Loop Patterns

There are several common patterns for `while` loops. Learning these patterns will help you write better programs.

### Pattern 1: Sentinel Value Loop

A **sentinel value** is a special value that signals the end of input. The loop continues until the sentinel is entered.

**Figure 7.1: Sentinel value loop — sum until zero**

````python
# Add numbers until user enters 0 (the sentinel)
total = 0
number = int(input("Enter a number (0 to stop): "))

while number != 0:
    total += number
    number = int(input("Enter a number (0 to stop): "))

print(f"The sum is: {total}")
````

### Pattern 2: Counter-Controlled Loop

Uses a counter variable to control how many times the loop runs.

**Figure 7.2: Counter-controlled loop**

````python
# Get exactly 5 test scores
count = 0
total = 0

while count < 5:
    score = int(input(f"Enter score {count + 1} of 5: "))
    total += score
    count += 1

average = total / count
print(f"\nAverage score: {average:.1f}")
````

### Pattern 3: Flag-Controlled Loop

Uses a boolean variable (a "flag") to control when the loop stops.

**Figure 7.3: Flag-controlled loop**

````python
# Keep running until user confirms they want to quit
running = True

while running:
    answer = input("Do you want to continue? (yes/no): ")
    if answer.lower() == "no":
        running = False    # Set the flag to stop
        print("Stopping...")
    else:
        print("Let's keep going!")

print("Program ended.")
````

| Pattern | How it works | Example use |
|---|---|---|
| Sentinel value | Stop when special value entered | Sum until 0 |
| Counter-controlled | Count iterations | Get N inputs |
| Flag-controlled | Boolean flag controls loop | Menu system |

## Part 9: Common Mistakes

Here are the most common mistakes students make with `while` loops. Learn to recognize and avoid them!

**Figure 9.1: Mistake 1 — Forgetting to update the loop variable**

````python
# ❌ WRONG - infinite loop (commented out for safety)
# x = 1
# while x <= 5:
#     print(x)
#     # Forgot: x += 1

# ✅ CORRECT
x = 1
while x <= 5:
    print(x)
    x += 1  # Always update the loop variable!
````

**Figure 9.2: Mistake 2 — Off-by-one error**

````python
# Goal: print 1 to 5

# ❌ WRONG - prints 1 to 4 (off by one)
print("Wrong (< instead of <=):")
x = 1
while x < 5:   # Should be <=
    print(x, end=" ")
    x += 1
print()

# ✅ CORRECT
print("Correct:")
x = 1
while x <= 5:
    print(x, end=" ")
    x += 1
print()
````

**Figure 9.3: Mistake 3 — continue before update in while loop**

````python
# ❌ WRONG - infinite loop when i == 3 (commented out)
# i = 0
# while i < 10:
#     if i == 3:
#         continue    # Skips i += 1, so i stays 3 forever!
#     print(i)
#     i += 1

# ✅ CORRECT - update before continue
i = 0
while i < 10:
    i += 1          # Update first!
    if i == 3:
        continue     # Now safe to skip
    print(i, end=" ")
print()
````

**Figure 9.4: Mistake 4 — Wrong comparison in condition**

````python
# Goal: repeat while the answer is neither "yes" nor "no".
# Correct: both comparisons must be true for the answer to be invalid.
# Wrong: answer != "yes" or answer != "no" is always True.
answer = input("Enter yes or no: ").strip().lower()
while answer != "yes" and answer != "no":
    print("Invalid input!")
    answer = input("Enter yes or no: ").strip().lower()

print(f"You entered: {answer}")
````

### Truth table: when should validation repeat?

| answer | answer != "yes" | answer != "no" | `and`: repeat? | `or` |
|---|---|---|---|---|
| yes | False | True | False | True |
| no | True | False | False | True |
| maybe | True | True | True | True |

We repeat only for the third row. `or` cannot stop: every string differs from at least one of the two accepted words. `.strip().lower()` also allows surrounding spaces and uppercase input.

**TR:** Geçersiz olmak için iki doğru cevaptan da farklı olması gerekir; bu yüzden `and` kullanılır. `or` her satırda doğru olur.

`n % 10` takes the last digit; `n // 10` removes it.

| n before | Last digit | Sum after adding it | n after removing it |
|---:|---:|---:|---:|
| 123 | 3 | 3 | 12 |
| 12 | 2 | 5 | 1 |
| 1 | 1 | 6 | 0 |

For reversal, append that digit arithmetically: `reversed_num = reversed_num * 10 + digit`. The reversed values become 3, 32, 321. Reversing 120 gives the integer 21, because integers do not preserve a leading zero.

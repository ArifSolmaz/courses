## Explain the program: A while loop needs progress, not only a condition

A while loop checks its condition before every pass. The condition says when to keep going. The loop needs a route toward stopping: for example, update a counter or exit with `break`. A correct condition alone does not guarantee that the loop ends.

**Draw or trace.** Draw a cycle: test attempts < 3 → do one attempt → increase attempts → test again. Mark the exit where the test is false.

## Part 2: The while Loop

A `while` loop repeats a block of code **as long as a condition is True**. Use it when a condition determines whether another iteration is needed. A `for` loop processes an iterable; its length need not be known in advance.

### Syntax

```python
while condition:
    # code block
    # update state toward stopping, or use a deliberate break
```

### How it works:
1. Check the condition
2. If `True`, execute the body
3. Go back to step 1
4. If `False`, skip the body and continue after the loop

### When to use `while` vs `for`?

| Use `for` when... | Use `while` when... |
|---|---|
| You know the number of iterations | A condition decides whether to continue |
| Iterating over a sequence | Waiting for a condition to change |
| Counting from A to B | User input validation |
| Processing each item in a collection | Running until a sentinel value |

**Figure 2.1: Basic while loop — counting from 1 to 5**

````python
count = 1
while count <= 5:
    print(count)
    count += 1
print("Done! count is now:", count)
````

**Line-by-line walkthrough**

- **Line 1 —** `` count = 1 ``

  Start the counter at 1, the first number to print.

- **Line 2 —** `` while count <= 5: ``

  Before every pass, test whether the counter is at most 5. Exactly 5 passes the test; 6 does not.

- **Line 3 —** `` print(count) ``

  Print the current counter on its own line while the condition is true.

- **Line 4 —** `` count += 1 ``

  Add one after printing. After printing 5, the counter becomes 6; without this update, it would remain 1 forever.

- **Line 5 —** `` print("Done! count is now:", count) ``

  Once the condition fails, print the final counter, 6. The stopping value is different from the last printed value, 5.



**Figure 2.2: Comparing for and while for the same task**

````python
print("=== for loop ===")
for i in range(1, 6):
    print(i)
print("\n=== while loop ===")
i = 1
while i <= 5:
    print(i)
    i += 1
````

**Line-by-line walkthrough**

- **Line 1 —** `` print("=== for loop ===") ``

  Print a heading for the for-loop demonstration.

- **Line 2 —** `` for i in range(1, 6): ``

  Visit the five integers 1 through 5, stopping before 6.

- **Line 3 —** `` print(i) ``

  Print each current value on its own line.

- **Line 4 —** `` print("\n=== while loop ===") ``

  Print a blank line and the heading for the while-loop demonstration.

- **Line 5 —** `` i = 1 ``

  Reset `i` to 1 rather than reusing the previous loop's final value 5.

- **Line 6 —** `` while i <= 5: ``

  Test whether `i` is at most 5 before each pass.

- **Line 7 —** `` print(i) ``

  Print the current `i`, producing the same 1-through-5 output as above.

- **Line 8 —** `` i += 1 ``

  Increase `i` explicitly; a while loop does not automatically advance it. The final update makes it 6, stopping the loop.



> 💡 **Note:** For counting tasks, `for` loops are usually cleaner. Use `while` when the number of iterations depends on a condition that changes during execution.

## Part 3: while Loop Examples

Let's look at several practical examples where `while` loops shine.

**Figure 3.1: Countdown timer**

````python
countdown = 10
while countdown >= 1:
    print(countdown, end=" ")
    countdown -= 1
print("\nBlast off! 🚀")
````

- **Line 1 —** `` countdown = 10 ``

  Initialize the countdown to 10.

- **Line 2 —** `` while countdown >= 1: ``

  Check whether at least 1 remains before each pass. This includes 1 but excludes 0.

- **Line 3 —** `` print(countdown, end=" ") ``

  Print the current number followed by a space instead of a newline, keeping the countdown on one line.

- **Line 4 —** `` countdown -= 1 ``

  Subtract one after printing; the values move 10, 9, …, 1, then 0 for the final failed test.

- **Line 5 —** `` print("\nBlast off! 🚀") ``

  After the loop, start a new line and print the launch message once.



**Figure 3.2: Simple guessing game**

````python
secret = 7
guess = int(input("Guess a number between 1 and 10: "))
while guess != secret:
    if guess < secret:
        print("Too low!")
    else:
        print("Too high!")
    guess = int(input("Try again: "))
print("Correct! The number was", secret)
````

**Line-by-line walkthrough**

- **Line 2 —** `` guess = int(input("Guess a number between 1 and 10: ")) ``

  Ask for a guess, receive text, and convert it to an integer. This example assumes integer text: `abc` would raise ValueError, not count as a wrong guess.

- **Line 3 —** `` while guess != secret: ``

  Repeat while the numeric guess differs from 7. A correct first guess skips the loop entirely.

- **Line 4 —** `` if guess < secret: ``

  Test whether the incorrect guess is below the secret.

- **Line 5 —** `` print("Too low!") ``

  For a low guess, print `Too low!`.

- **Line 6 —** `` else: ``

  Choose the alternative branch when the incorrect guess was not below 7; it must therefore be above 7.

- **Line 7 —** `` print("Too high!") ``

  Print `Too high!` for that branch.

  Request and convert a new guess before rechecking the while condition. This update allows the loop eventually to stop; the prompt's 1–10 range is not enforced by a separate check.

- **Line 9 —** `` print("Correct! The number was", secret) ``

**Figure 3.3: Validating the range of an integer input**

### A while loop needs a route toward stopping

Pseudocode: choose a starting value; test the condition; do one step; update the value; return to the condition. For a countdown from 3:

| Value before test | `counter > 0` | Output | Value after update |
|---:|---|---:|---:|
| 3 | True | 3 | 2 |
| 2 | True | 2 | 1 |
| 1 | True | 1 | 0 |
| 0 | False | nothing | loop ends |

If the update is absent, the condition never changes. If `continue` skips the update, that iteration can repeat forever. **TR:** Döngüyü yazarken hangi satırın durma koşuluna yaklaştırdığını göster; yalnızca koşulu yazmak yetmez.

## Part 4: Infinite Loops

An **infinite loop** keeps repeating without reaching an exit. A condition that stays true is not enough to make a loop infinite: `while True` can still terminate with `break`.

### Common causes:
- Forgetting to update the loop variable
- Updating the variable in the wrong direction
- A condition that can never be False

### How to stop an infinite loop:
- In Jupyter/Colab: Click the **stop button** (square icon) or go to **Runtime > Interrupt execution**
- In terminal: Press **Ctrl+C**

> 💡 **Note:** If your notebook seems frozen and nothing is printing, you probably have an infinite loop. Interrupt the kernel!

## Part 5: The break Statement

The `break` statement **immediately exits the nearest enclosing loop**, no matter what its condition says. In nested loops, it does not exit the outer loop.

### When to use `break`:
- You found what you were looking for (search pattern)
- The user wants to quit
- An error condition occurred
- You want to exit a `while True` loop

```python
while condition:
    # some code
    if something:
        break       # exit the loop immediately
    # more code
```

**Figure 5.1: Using break to exit when found**

````python
number = 50
while number <= 100:
    if number % 7 == 0:
        print(f"Found it! {number} is divisible by 7")
        break
    number += 1
print("Search complete.")
````

**Line-by-line walkthrough**

- **Line 1 —** `` number = 50 ``

  Begin the candidate search at 50.

- **Line 2 —** `` while number <= 100: ``

  Keep searching while the candidate is at most 100, so 100 would also be tested if reached.

- **Line 3 —** `` if number % 7 == 0: ``

  Compute the candidate's remainder on division by 7. Remainder zero means divisibility without a fractional part.

- **Line 4 —** `` print(f"Found it! {number} is divisible by 7") ``

  When the test succeeds, print the candidate; the first successful candidate is 56.

- **Line 5 —** `` break ``

  Exit the while loop immediately, skipping the increment below. Consequently `number` remains 56.

- **Line 6 —** `` number += 1 ``

  If no divisor match occurred on this pass, add 1 to try the next candidate. This line is inside the while but outside the if.

- **Line 7 —** `` print("Search complete.") ``

  Print the completion message after either a break or exhaustion of the search range. The message alone does not prove a match was found.



**Figure 5.2: while True with break — the classic pattern**

````python
while True:
    text = input("Enter something (or 'quit' to stop): ")
    if text.lower() == "quit":
        break
    print(f"You entered: {text}")
print("Goodbye!")
````

**Line-by-line walkthrough**

- **Line 1 —** `` while True: ``

  Begin a loop whose condition is always true. It needs the explicit break below to finish normally.

- **Line 2 —** `` text = input("Enter something (or 'quit' to stop): ") ``

  Read one text response per pass. In the HTML version, each call consumes the next supplied input line.

- **Line 3 —** `` if text.lower() == "quit": ``

  Lowercase the response and compare it to `quit`. Thus `QUIT` works, but surrounding spaces are not removed by this code.

- **Line 4 —** `` break ``

  If the response matched, exit immediately and skip the echo statement.

- **Line 5 —** `` print(f"You entered: {text}") ``

  Otherwise print the original text as entered; lowercasing for comparison did not overwrite `text`.

- **Line 6 —** `` print("Goodbye!") ``

  After the break, print `Goodbye!` once. The sentinel word is not echoed as a normal message.



**Figure 5.3: break in a for loop**

````python
numbers = [3, 7, 11, 4, 9, 2, 8]
for num in numbers:
    if num % 2 == 0:
        print(f"First even number found: {num}")
        break
else:
    print("No even number found.")
````

**Line-by-line walkthrough**

- **Line 1 —** `` numbers = [3, 7, 11, 4, 9, 2, 8] ``

  Store the candidate integers in their search order.

- **Line 2 —** `` for num in numbers: ``

  Visit each candidate, beginning with 3, and assign it to `num`.

- **Line 3 —** `` if num % 2 == 0: ``

  Check for remainder zero on division by 2, identifying even integers.

- **Line 4 —** `` print(f"First even number found: {num}") ``

  Print the first matching value, 4. Earlier candidates 3, 7 and 11 did not satisfy the test.

- **Line 5 —** `` break ``

  Exit the for loop immediately. Values 9, 2 and 8 are not visited after the first match.

- **Line 6 —** `` else: ``

  This else belongs to the for loop, not to the if. Its body runs only if the loop finishes without a break, including an empty list.

- **Line 7 —** `` print("No even number found.") ``

  Print the no-match message only in that no-break case. For this supplied list the break at 4 prevents it.



> 💡 **Note:** The `else` clause of a `for` or `while` loop runs only if the loop completed without hitting `break`. This is a unique Python feature!

## Part 6: The continue Statement

The `continue` statement **skips the rest of the current iteration** and jumps to the next one.

### When to use `continue`:
- Skip unwanted values (filtering)
- Avoid deeply nested if statements
- Skip invalid data

```python
while condition:
    # some code
    if skip_this:
        continue    # skip to next iteration
    # this code only runs when NOT skipped
```

**Figure 6.1: Skip odd numbers using continue**

````python
for i in range(1, 11):
    if i % 2 != 0:
        continue
    print(i, end=" ")
print()
````

- **Line 1 —** `` for i in range(1, 11): ``

  Visit every integer 1 through 10, including both endpoints.

- **Line 2 —** `` if i % 2 != 0: ``

  Test for a nonzero remainder on division by 2, which identifies an odd integer.

- **Line 3 —** `` continue ``

  For odd values, skip the rest of this pass and let the for loop supply the next value automatically.

- **Line 4 —** `` print(i, end=" ") ``

  For values not skipped, print the current even integer with a space. The visible values are 2, 4, 6, 8 and 10.

- **Line 5 —** `` print() ``

  After all ten candidates, print the newline omitted by the loop's space-ending print calls.



**Figure 6.2: continue in a while loop — be careful!**

````python
i = 0
while i < 10:
    i += 1
    if i % 3 == 0:
        continue
    print(i, end=" ")
print()
````

**Line-by-line walkthrough**

- **Line 1 —** `` i = 0 ``

  Start `i` at 0 because it will be incremented before the first test inside the body.

- **Line 2 —** `` while i < 10: ``

  Enter while the current `i` is below 10. The last pass enters with 9 and then increments to 10.

- **Line 3 —** `` i += 1 ``

  Increase `i` before any possible continue. This guarantees progress even on passes whose remaining body is skipped.

- **Line 4 —** `` if i % 3 == 0: ``

  Check whether the updated value is divisible by 3.

- **Line 5 —** `` continue ``

  For 3, 6 and 9, skip the print and return to the while condition. The increment has already happened, so these values cannot trap the loop.

- **Line 6 —** `` print(i, end=" ") ``

- **Line 7 —** `` print() ``

  Finish the output line after `i` reaches 10 and the next while test fails.



> 💡 **Note:** When using `continue` in a `while` loop, make sure every path that continues can make progress toward stopping. In this counter example, update the counter **before** `continue`; skipping that update can cause an infinite loop.

## Part 7: Loop Patterns

There are several common patterns for `while` loops. Learning these patterns will help you write better programs.

### Pattern 1: Sentinel Value Loop

A **sentinel value** is a special value that signals the end of input. The loop continues until the sentinel is entered.

**Figure 7.1: Sentinel value loop — sum until zero**

````python
total = 0
number = int(input("Enter a number (0 to stop): "))
while number != 0:
    total += number
    number = int(input("Enter a number (0 to stop): "))
print(f"The sum is: {total}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` total = 0 ``

  Initialize the sum to zero before reading any values.

- **Line 2 —** `` number = int(input("Enter a number (0 to stop): ")) ``

  Read the first response and convert it to an integer. This is a priming read: the condition needs a value to test before the loop starts.

- **Line 3 —** `` while number != 0: ``

  Continue while the current number is not the sentinel 0. A first response of 0 gives no passes and leaves total 0.

- **Line 4 —** `` total += number ``

  Add the current nonzero number to the sum; negative integers are also added by this code.

- **Line 5 —** `` number = int(input("Enter a number (0 to stop): ")) ``

  Read and convert the next response. A sentinel entered here is tested before any addition on the next pass.

- **Line 6 —** `` print(f"The sum is: {total}") ``

  Print the completed sum once the sentinel ends the loop. This example assumes valid integer text rather than handling conversion errors.



### Pattern 2: Counter-Controlled Loop

Uses a counter variable to control how many times the loop runs.

**Figure 7.2: Counter-controlled loop**

````python
count = 0
total = 0
while count < 5:
    score = int(input(f"Enter score {count + 1} of 5: "))
    total += score
    count += 1
average = total / count
print(f"\nAverage score: {average:.1f}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` count = 0 ``

  Start the number of scores received at zero.

- **Line 2 —** `` total = 0 ``

  Start the score total at zero, separately from the count.

- **Line 3 —** `` while count < 5: ``

  Repeat until five scores have been processed; counts 0 through 4 enter the body.

- **Line 4 —** `` score = int(input(f"Enter score {count + 1} of 5: ")) ``

  Prompt for score number `count + 1`, so the first prompt says 1 of 5. Convert the response to an integer; this code does not enforce a 0–100 score range.

- **Line 5 —** `` total += score ``

  Add the received score to the running sum.

- **Line 6 —** `` count += 1 ``

  Increase the received-score count by one. After the fifth entry, count is 5 and the next condition fails.

- **Line 7 —** `` average = total / count ``

  Divide the finished total by the count. Normal completion guarantees a denominator of 5; input conversion or exhaustion would have stopped execution earlier.

- **Line 8 —** `` print(f"\nAverage score: {average:.1f}") ``

  Print a blank line and the mean to one decimal place. This formats the display without rounding the stored average itself.



### Pattern 3: Flag-Controlled Loop

Uses a boolean variable (a "flag") to control when the loop stops.

**Figure 7.3: Flag-controlled loop**

````python
running = True
while running:
    answer = input("Do you want to continue? (yes/no): ")
    if answer.lower() == "no":
        running = False
        print("Stopping...")
    else:
        print("Let's keep going!")
print("Program ended.")
````

**Line-by-line walkthrough**

- **Line 1 —** `` running = True ``

  Initialize the Boolean control flag to True so the loop can begin.

- **Line 2 —** `` while running: ``

  Check the flag before each pass; the body repeats while it remains True.

- **Line 3 —** `` answer = input("Do you want to continue? (yes/no): ") ``

  Read a text answer to the continue question.

- **Line 4 —** `` if answer.lower() == "no": ``

  Lowercase the answer and compare it with `no`; uppercase NO is accepted, but spaces around it are not removed.

- **Line 5 —** `` running = False ``

  Set the flag to False after a no response. This does not immediately jump out of the current body; it affects the next while test.

- **Line 6 —** `` print("Stopping...") ``

  Print `Stopping...` during that same final pass.

- **Line 7 —** `` else: ``

  Choose the alternative branch for every other response, not just the word yes.

- **Line 8 —** `` print("Let's keep going!") ``

  Print the continue message. Because the flag stays True here, the loop will ask again.

- **Line 9 —** `` print("Program ended.") ``

  Once the false flag prevents another pass, print the final message once.



| Pattern | How it works | Example use |
|---|---|---|
| Sentinel value | Stop when special value entered | Sum until 0 |
| Counter-controlled | Count iterations | Get N inputs |
| Flag-controlled | Boolean flag controls loop | Menu system |

## Part 9: Common Mistakes

Here are the most common mistakes students make with `while` loops. Learn to recognize and avoid them!

**Figure 9.1: Mistake 1 — Forgetting to update the loop variable**

````python
# ❌ WRONG - infinite loop (commented out for safety)
# x = 1
# while x <= 5:
#     print(x)
#     # Forgot: x += 1
# ✅ CORRECT
x = 1
while x <= 5:
    print(x)
    x += 1
````

**Line-by-line walkthrough**

- **Line 7 —** `` x = 1 ``

  Initialize the active, corrected counter to 1. The earlier lines beginning with # are an inactive example, not a second running loop.

- **Line 8 —** `` while x <= 5: ``

  Repeat while the counter is at most 5.

- **Line 9 —** `` print(x) ``

  Print the current counter, producing 1 through 5 on separate lines.

- **Line 10 —** `` x += 1 ``

  Increase the counter after printing. This is the missing step in the commented broken example; reaching 6 lets the loop stop.



**Figure 9.2: Mistake 2 — Off-by-one error**

````python
print("Wrong (< instead of <=):")
x = 1
while x < 5:
    print(x, end=" ")
    x += 1
print()
print("Correct:")
x = 1
while x <= 5:
    print(x, end=" ")
    x += 1
print()
````

**Line-by-line walkthrough**

- **Line 1 —** `` print("Wrong (< instead of <=):") ``

  Print a heading identifying an output-boundary mistake; the following loop is valid Python and does terminate.

- **Line 2 —** `` x = 1 ``

  Start at 1.

- **Line 3 —** `` while x < 5: ``

  Repeat only while the value is strictly below 5, excluding the desired endpoint.

- **Line 4 —** `` print(x, end=" ") ``

  Print the current value with a space, so this example displays 1 through 4.

- **Line 5 —** `` x += 1 ``

  Increase the value, eventually reaching 5 and failing the condition.

- **Line 6 —** `` print() ``

  End that demonstration's output line.

- **Line 7 —** `` print("Correct:") ``

  Print the heading for the corrected boundary.

- **Line 8 —** `` x = 1 ``

  Reset `x` to 1 so the second experiment starts independently of the first.

- **Line 9 —** `` while x <= 5: ``

  Use less-than-or-equal so 5 is also allowed into the body.

- **Line 10 —** `` print(x, end=" ") ``

  Print the current value, this time including 5.

- **Line 11 —** `` x += 1 ``

  Increase the value after printing; after 5 it becomes 6 and stops the loop.

- **Line 12 —** `` print() ``

  End the corrected output line.



**Figure 9.3: Mistake 3 — continue before update in while loop**

````python
# ❌ WRONG - infinite loop when i == 3 (commented out)
# i = 0
# while i < 10:
#     if i == 3:
#         continue    # Skips i += 1, so i stays 3 forever!
#     print(i)
#     i += 1
# ✅ CORRECT - update before continue
i = 0
while i < 10:
    i += 1
    if i == 3:
        continue
    print(i, end=" ")
print()
````

**Line-by-line walkthrough**

- **Line 9 —** `` i = 0 ``

  Initialize the corrected counter to 0; the commented loop above remains inactive.

- **Line 10 —** `` while i < 10: ``

  Check whether the counter is below 10 before each pass.

- **Line 11 —** `` i += 1 ``

  Increment first, so even a pass that executes continue has made progress.

- **Line 12 —** `` if i == 3: ``

  Check whether the updated value equals the one value to skip, 3.

- **Line 13 —** `` continue ``

  At 3, return to the while condition without printing. Because the counter was already advanced, the next pass can advance to 4.

- **Line 14 —** `` print(i, end=" ") ``

  Print all the other values, giving 1, 2, 4, 5, 6, 7, 8, 9 and 10 on one line.

- **Line 15 —** `` print() ``

  End the line after the loop finishes at counter 10.



**Figure 9.4: Mistake 4 — Wrong comparison in condition**

````python
answer = input("Enter yes or no: ").strip().lower()
while answer != "yes" and answer != "no":
    print("Invalid input!")
    answer = input("Enter yes or no: ").strip().lower()
print(f"You entered: {answer}")
````

**Line-by-line walkthrough**

- **Line 1 —** `` answer = input("Enter yes or no: ").strip().lower() ``

  Read a response, remove whitespace from its ends, then lowercase it. For example, ` YES ` becomes `yes` before being stored.

- **Line 2 —** `` while answer != "yes" and answer != "no": ``

  Repeat only when the answer differs from both accepted words. `and` is essential: yes fails the first comparison and no fails the second, so either accepted answer stops the loop.

- **Line 3 —** `` print("Invalid input!") ``

  Print an invalid-input message for a response such as maybe.

- **Line 4 —** `` answer = input("Enter yes or no: ").strip().lower() ``

  Request another response and apply the same cleanup. Replacing `answer` allows the next condition test to see the new text.

- **Line 5 —** `` print(f"You entered: {answer}") ``

  Print the accepted normalized answer after the loop. Using `or` in the condition would never accept either word because one of its two comparisons is always true.



### Truth table: when should validation repeat?

| answer | answer != "yes" | answer != "no" | `and`: repeat? | `or` |
|---|---|---|---|---|
| yes | False | True | False | True |
| no | True | False | False | True |
| maybe | True | True | True | True |

We repeat only for the third row. `or` cannot stop: every string differs from at least one of the two accepted words. `.strip().lower()` also allows surrounding spaces and uppercase input.

### Trace digit arithmetic with 123

`n % 10` takes the last digit; `n // 10` removes it.

| n before | Last digit | Sum after adding it | n after removing it |
|---:|---:|---:|---:|
| 123 | 3 | 3 | 12 |
| 12 | 2 | 5 | 1 |
| 1 | 1 | 6 | 0 |

For reversal, append that digit arithmetically: `reversed_num = reversed_num * 10 + digit`. The reversed values become 3, 32, 321. Reversing 120 gives the integer 21, because integers do not preserve a leading zero.

**TR:** `% 10` son basamağı bulur; `// 10` onu siler. Sonucu 10 ile çarpmak yeni basamak için yer açar.

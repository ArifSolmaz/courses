## 1. One law for apples and moons / Elmalar ve uydular için tek yasa

Every pair of masses attracts along the line joining them:

$$\boxed{\,F = G\frac{m_1 m_2}{r^2}\,}, \qquad G = 6.674\times10^{-11}\ \mathrm{N\,m^2/kg^2}$$

$r$ is the distance between **centres**, not between surfaces, and the force on each mass is equal and
opposite — a Newton's-third-law pair, exactly as in Week 03.

### Where $g$ came from

You have written $g = 9.81\ \mathrm{m/s^2}$ in every week of this course without asking where it comes
from. Set the weight of a mass $m$ at the Earth's surface equal to the gravitational attraction:

$$mg = G\frac{M_E m}{R_E^2} \quad\Longrightarrow\quad \boxed{\,g = \frac{GM_E}{R_E^2}\,}$$

**The mass $m$ cancels.** That is why everything falls at the same rate, and why your free-fall experiment
in the Week 04 laboratory never needed to know the mass of the ball. It is also the deepest thing on this
page: the object's own mass sets both how hard gravity pulls it (in $F = GMm/r^2$) and how hard it is to
accelerate (in $F = ma$), and the two cancel exactly.

**Numbers.** With $M_E = 5.97\times10^{24}\ \mathrm{kg}$ and $R_E = 6.37\times10^6\ \mathrm m$:

$$g = \frac{(6.674\times10^{-11})(5.97\times10^{24})}{(6.37\times10^6)^2}
= \frac{3.985\times10^{14}}{4.058\times10^{13}} = 9.82\ \mathrm{m/s^2}$$

**Limiting check and a real prediction.** $g \propto 1/r^2$, so at the altitude of the International Space
Station ($h = 400\ \mathrm{km}$, $r = 6.77\times10^6\ \mathrm m$):

$$g = 9.82\left(\frac{6.37}{6.77}\right)^2 = 8.69\ \mathrm{m/s^2}$$

That is 89% of surface gravity. **So astronauts on the ISS are not weightless because gravity is weak
there** — gravity is nearly as strong as in this room. They float because the station and everything in it
are in free fall together, continuously missing the Earth. "Zero gravity" is the most widespread physics
misconception there is, and you now have the number to refute it.

**TR:** Bütün kütle çiftleri birbirini $F = Gm_1m_2/r^2$ ile çeker; $r$ **merkezler** arası uzaklıktır.
Yüzeydeki ağırlığı bu çekime eşitleyince $g = GM_E/R_E^2$ çıkar ve **cismin kütlesi sadeleşir** — her şeyin
aynı ivmeyle düşmesinin ve 4. hafta deneyinin bilyenin kütlesini hiç sormamasının sebebi budur. Uzay
istasyonunun yüksekliğinde $g$ hâlâ yüzey değerinin %89'udur: astronotlar kütle çekimi zayıf olduğu için
değil, istasyonla birlikte **sürekli serbest düştükleri** için süzülürler. "Sıfır yerçekimi" en yaygın
fizik yanılgısıdır ve artık onu çürütecek sayıya sahipsin.

## 2. Why orbits are just circular motion / Yörünge, dairesel hareketten başka bir şey değil

A satellite of mass $m$ in a circular orbit of radius $r$ around a body of mass $M$. There is exactly one
force on it, and it points at the centre — so it **is** the centripetal force. No new physics is needed;
just set the two expressions equal:

$$G\frac{Mm}{r^2} = \frac{mv^2}{r}$$

Cancel one factor of $r$ and the satellite's mass $m$:

$$\boxed{\,v = \sqrt{\frac{GM}{r}}\,}$$

**Three things fall straight out of this, and each is worth a moment:**

1. **The orbital speed does not depend on the satellite's mass.** A bolt and a space station in the same
   orbit travel at the same speed — which is why a lost bolt stays alongside the station instead of
   drifting away.
2. **Lower orbits are faster**, as $1/\sqrt{r}$. This is genuinely counter-intuitive: to catch up with
   something ahead of you in orbit, you must *drop* to a lower, faster orbit and then climb back.
3. **You cannot choose speed and radius independently.** Pick the altitude and the speed is decided for
   you. There is exactly one circular-orbit speed at each radius.

**Limiting check.** $r \to \infty$ gives $v \to 0$ ✅ — very distant orbits are very slow.

### The period

The satellite covers one circumference in one period, so $v = 2\pi r/T$. Substituting:

$$\frac{2\pi r}{T} = \sqrt{\frac{GM}{r}} \quad\Longrightarrow\quad
\frac{4\pi^2 r^2}{T^2} = \frac{GM}{r} \quad\Longrightarrow\quad
\boxed{\,T^2 = \frac{4\pi^2}{GM}\,r^3\,}$$

Stop and look at what just happened. $T^2 \propto r^3$ is **Kepler's third law**, and it fell out of
Newton's second law plus an inverse-square force in three lines of algebra. Kepler extracted it from
twenty years of Tycho Brahe's naked-eye observations and could not explain it; Newton *derived* it. This is
the single most famous "and therefore" in the history of physics, and you are now in a position to do it
yourself.

**TR:** Yörüngedeki uyduya etki eden tek kuvvet merkeze doğrudur, yani merkezcil kuvvetin kendisidir.
İkisini eşitleyip sadeleştirince $v = \sqrt{GM/r}$ çıkar: hız **uydunun kütlesinden bağımsızdır** (kaybolan
bir cıvatanın istasyonun yanında kalmasının sebebi), alçak yörüngeler **daha hızlıdır**, ve her yarıçap için
tek bir dairesel yörünge hızı vardır. $v = 2\pi r/T$ yerine koyunca $T^2 \propto r^3$ — **Kepler'in üçüncü
yasası** — üç satır cebirle çıkar. Kepler bunu yirmi yıllık gözlemden çıkardı ama açıklayamadı; Newton
türetti.

## 3. Kepler's three laws / Kepler'in üç yasası

| | Law | Where it comes from |
| --- | --- | --- |
| **1** | Planets move on **ellipses** with the Sun at one focus | Requires the general (non-circular) solution; a circle is the special case |
| **2** | The line to the Sun sweeps **equal areas in equal times** | Conservation of **angular momentum** — Week 11 |
| **3** | $T^2 \propto a^3$, with $a$ the semi-major axis | Derived above for a circle |

### The second law is angular momentum, and nothing else

This is worth seeing, because it connects two weeks of this course that look unrelated. Gravity always
points along the line from the planet to the Sun, so it has **no lever arm about the Sun** — the
perpendicular distance from the line of action to the pivot is zero. Therefore the torque about the Sun is
zero, and by Week 11's $\sum\tau = dL/dt$, angular momentum is conserved.

The area swept per unit time turns out to be exactly $L/2m$. So *equal areas in equal times* is a
statement that $L$ is constant, written in the language of geometry because Kepler had no other language
available.

**And it tells you something physical immediately:** since $L = mvr$ for the perpendicular component, a
planet close to the Sun (small $r$) must move **fast**, and far away it must move **slowly**. Earth is
about 3% closer to the Sun in early January than in early July, and travels measurably faster then.

### Using the third law without ever looking up $G$

Write the third law for two bodies orbiting the same central mass and divide:

$$\frac{T_1^2}{T_2^2} = \frac{r_1^3}{r_2^3}$$

$G$ and $M$ have cancelled. **Worked example.** The Moon orbits at $r = 3.84\times10^8\ \mathrm m$ with
$T = 27.3\ \mathrm{days}$. What is the period of a satellite at $r = 4.22\times10^7\ \mathrm m$?

$$T_2 = T_1\left(\frac{r_2}{r_1}\right)^{3/2} = 27.3\left(\frac{4.22\times10^7}{3.84\times10^8}\right)^{3/2}
= 27.3\,(0.1099)^{3/2} = 27.3(0.03643) = 0.995\ \mathrm{days}$$

One day. That radius is geostationary orbit, and this is how you find it: a satellite whose period matches
the Earth's rotation stays above the same spot, which is why television dishes do not move. **You just
derived the altitude of the entire satellite television industry from the Moon.**

**TR:** Kepler'in ikinci yasası, açısal momentumun korunumundan başka bir şey değildir: kütle çekimi
Güneş'e doğru olduğu için Güneş'e göre moment kolu sıfırdır, tork sıfırdır, $L$ korunur (11. hafta). Bu
yüzden Güneş'e yakın gezegen **hızlı**, uzak olan **yavaş** hareket eder. Üçüncü yasayı iki cisim için
oranlayarak yazarsan $G$ ve $M$ sadeleşir: Ay'ın yörüngesinden yola çıkarak yer-durağan yörüngenin bir
günlük periyodunu — yani bütün uydu televizyonu endüstrisinin yüksekliğini — bulabilirsin.

## 4. Orbital energy and escape speed / Yörünge enerjisi ve kaçış hızı

Week 08 used $U_g = mgh$, which assumes $g$ is constant — fine near the ground, useless for orbits. The
general gravitational potential energy, taking zero at infinite separation, is

$$\boxed{\,U = -\frac{GMm}{r}\,}$$

**Why negative?** Because we chose $U = 0$ infinitely far away, and gravity does positive work as things
fall inward — so anything at finite $r$ has less than zero. The sign is a bookkeeping choice, but a useful
one: **bound systems have negative total energy.** That single sentence is how astronomers decide whether
two objects are gravitationally tied together at all.

### Escape speed

To escape, you need enough kinetic energy to reach $r \to \infty$ with nothing left over. Set the total
energy to zero — symbolic first:

$$\tfrac12 mv_{\rm esc}^2 - \frac{GMm}{R} = 0 \quad\Longrightarrow\quad
\boxed{\,v_{\rm esc} = \sqrt{\frac{2GM}{R}}\,}$$

**The mass $m$ cancels again.** Escape speed is a property of the *planet*, not of what you are launching:
a pebble and a rocket need the same speed. (Rockets need more fuel, not more speed — they carry their fuel
and fight air drag.)

**Numbers for Earth:**
$$v_{\rm esc} = \sqrt{\frac{2(6.674\times10^{-11})(5.97\times10^{24})}{6.37\times10^6}}
= \sqrt{1.251\times10^8} = 1.12\times10^4\ \mathrm{m/s} = 11.2\ \mathrm{km/s}$$

**Limiting checks.** Compare with the circular-orbit speed $v = \sqrt{GM/r}$ from §2: escape speed is
exactly $\sqrt{2}$ times the orbital speed at the same radius. So a satellite in low orbit at
$7.9\ \mathrm{km/s}$ needs only a 41% speed increase to leave the Earth entirely — a pleasingly small
number for such a large-sounding achievement. And $R \to \infty$ gives $v_{\rm esc} \to 0$ ✅: far from
everything, nothing holds you.

**Total energy of a circular orbit.** Substituting $v^2 = GM/r$:

$$E = K + U = \frac{GMm}{2r} - \frac{GMm}{r} = -\frac{GMm}{2r}$$

Negative, so bound ✅. And note $K = -E$ and $U = 2E$: **to raise a satellite to a higher orbit you must
add energy, yet it ends up moving slower.** Students find that contradictory until they notice that the
potential energy rises by twice as much as the kinetic energy falls.

**TR:** Yörüngeler için $U = -GMm/r$ kullanılır; işaret negatiftir çünkü sıfırı sonsuzda seçtik — ve bu
faydalıdır: **bağlı sistemlerin toplam enerjisi negatiftir.** Kaçış hızı $\sqrt{2GM/R}$'dir ve cismin
kütlesinden bağımsızdır: çakıl taşı ile roket aynı hıza ihtiyaç duyar. Kaçış hızı, aynı yarıçaptaki
yörünge hızının tam $\sqrt2$ katıdır. Dairesel yörüngenin toplam enerjisi $-GMm/2r$'dir: daha yüksek
yörüngeye çıkmak enerji **eklemeyi** gerektirir, ama uydu orada **daha yavaş** hareket eder — potansiyel
enerji, kinetik enerjinin düştüğünün iki katı kadar yükseldiği için.

## 5. Capstone: finding a planet you cannot see / Göremediğin bir gezegeni bulmak

Thousands of planets around other stars have been found, and the commonest method uses nothing beyond
this notebook and Week 10's circular motion. A planet crossing in front of its star blocks a little of
its light. We measure the star's brightness against time — a **transit light curve** — and read the
planet off it.

### What the depth tells you: the planet's size

The star is a disc of radius $R_\star$; the planet blocks a disc of radius $R_p$. The fractional drop in
brightness is the ratio of the areas:

$$\boxed{\,\delta = \frac{\Delta F}{F} = \frac{\pi R_p^2}{\pi R_\star^2} = \left(\frac{R_p}{R_\star}\right)^2\,}$$

So $R_p = R_\star\sqrt{\delta}$. For Jupiter across the Sun, $R_p/R_\star = 0.10$, giving
$\delta = 0.01$ — a **1%** dip. For Earth across the Sun, $R_p/R_\star = 0.0092$ and
$\delta = 8.4\times10^{-5}$, i.e. **0.0084%**. That is the measurement problem in one number: finding
another Earth means measuring a star's brightness to better than one part in ten thousand.

### What the period tells you: the orbit

Transits repeat once per orbit, so you measure $T$ directly by waiting. Then Kepler's third law from §2,
rearranged for the orbital radius:

$$a = \left(\frac{GM_\star T^2}{4\pi^2}\right)^{1/3}$$

**That is the whole method.** Brightness against time gives you the depth and the period; the depth gives
the planet's radius and the period gives its distance from the star. From a dip in a light curve you get a
planet's size and orbit — without ever seeing the planet.

### Worked example — symbolic first, then numbers

A star of radius $R_\star = 1.10\,R_\odot$ and mass $M_\star = 1.05\,M_\odot$ dims by
$\delta = 1.44\times10^{-2}$ every $3.52\ \mathrm{days}$. Find the planet's radius (in Jupiter radii) and
its orbital distance (in astronomical units).

Use $R_\odot = 6.96\times10^8\ \mathrm m$, $M_\odot = 1.99\times10^{30}\ \mathrm{kg}$,
$R_J = 7.15\times10^7\ \mathrm m$, $1\ \mathrm{AU} = 1.50\times10^{11}\ \mathrm m$.

**Radius.**
$$R_p = R_\star\sqrt{\delta} = (1.10)(6.96\times10^8)\sqrt{1.44\times10^{-2}}
= (7.656\times10^8)(0.120) = 9.19\times10^7\ \mathrm m = 1.28\,R_J$$

**Orbit.** $T = 3.52\ \mathrm{days} = 3.041\times10^5\ \mathrm s$ and
$M_\star = 2.090\times10^{30}\ \mathrm{kg}$:

$$a = \left(\frac{(6.674\times10^{-11})(2.090\times10^{30})(3.041\times10^5)^2}{4\pi^2}\right)^{1/3}
= \left(\frac{1.290\times10^{31}}{39.48}\right)^{1/3} = (3.267\times10^{29})^{1/3}$$
$$a = 6.89\times10^9\ \mathrm m = 0.0459\ \mathrm{AU}$$

**Interpret it, which is the part that matters.** A planet slightly larger than Jupiter, orbiting about eight
times closer to its star than Mercury is to the Sun, with a year of three and a half days. This is a "hot
Jupiter", and the first exoplanet ever found around a normal star was one of them. They were a genuine
shock in 1995: nothing in the theory of the time allowed a giant planet to exist there, and the discovery
forced planet-formation theory to be rebuilt around planets that migrate inward after forming far out.

**A limiting check worth doing.** $a \propto T^{2/3}$, so a planet with a 100-day year sits about
$100^{2/3}/3.52^{2/3} \approx 9$ times further out. Short-period planets are close in — which is also why
they are the easiest to find: you only have to watch for a few days to see several transits.

**TR:** Başka yıldızların binlerce gezegeni, bu nottaki fizikten fazlasını gerektirmeyen bir yöntemle
bulundu. Gezegen yıldızın önünden geçerken ışığın bir kısmını keser; düşüşün derinliği **alanların
oranıdır**, yani $\delta = (R_p/R_\star)^2$. Dünya'nın Güneş önünden geçişi yalnızca %0.0084'lük bir
düşüştür — "ikinci bir Dünya" bulmak, bir yıldızın parlaklığını on binde birden iyi ölçmek demektir.
Periyot ise beklemekle doğrudan ölçülür ve Kepler'in üçüncü yasası yörünge yarıçapını verir. Yukarıdaki
örnek, Jüpiter'den biraz büyük, yıldızına Merkür'ün Güneş'e olan uzaklığının dokuzda biri kadar yakın ve
yılı üç buçuk gün süren bir gezegen veriyor: bir "sıcak Jüpiter". 1995'te bunlar gerçek bir şoktu; dönemin
kuramında bir dev gezegenin orada bulunmasına izin veren hiçbir şey yoktu.


```python
#@title Optional: read a planet off a transit light curve {display-mode: "form"}
import numpy as np
import matplotlib.pyplot as plt

# ---------------- change these and re-run ----------------
R_star_sun  = 1.10     # stellar radius  [solar radii]
M_star_sun  = 1.05     # stellar mass    [solar masses]
depth       = 0.0144   # fractional dip in brightness
period_days = 3.52     # orbital period  [days]
# ---------------------------------------------------------

G, R_sun, M_sun, R_jup, AU = 6.674e-11, 6.96e8, 1.99e30, 7.15e7, 1.50e11
R_star, M_star, T = R_star_sun * R_sun, M_star_sun * M_sun, period_days * 86400.0

R_p = R_star * np.sqrt(depth)
a   = (G * M_star * T**2 / (4 * np.pi**2))**(1/3)
# central transit: the planet crosses one stellar diameter plus its own
t_dur = T / np.pi * np.arcsin((R_star + R_p) / a)

print(f"planet radius   R_p = {R_p:.3e} m = {R_p/R_jup:.2f} Jupiter radii")
print(f"orbital radius  a   = {a:.3e} m = {a/AU:.4f} AU")
print(f"transit depth   d   = {depth*100:.3f} %")
print(f"transit duration    = {t_dur/3600:.2f} hours  (central crossing)")
print(f"orbital speed   v   = {np.sqrt(G*M_star/a)/1000:.1f} km/s")

# a simple box-shaped light curve, which is what the depth and duration mean
half = t_dur / 2 / 3600
t = np.linspace(-3 * half, 3 * half, 2000)
flux = np.where(np.abs(t) <= half, 1 - depth, 1.0)

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(t, flux, lw=2)
ax.set_xlabel("time from mid-transit [hours]")
ax.set_ylabel("relative brightness")
ax.set_title(f"Transit light curve: depth {depth*100:.2f}%, duration {t_dur/3600:.2f} h")
ax.set_ylim(1 - 1.8 * depth, 1 + 0.6 * depth)
ax.axhline(1.0, ls=":", lw=1)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print("\nReal light curves are rounded at the edges, not square: the planet takes")
print("time to cross the limb, and the star is brighter at its centre than its rim.")
```

#### G1 — Gravity on another world  ·  L1

Mars has mass $6.42\times10^{23}\ \mathrm{kg}$ and radius $3.39\times10^6\ \mathrm m$. Find $g$ on Mars,
as a fraction of Earth's. Then find how far a ball dropped from $1.50\ \mathrm m$ would take to land.

<details><summary>Answer</summary>

$$g_M = \frac{GM}{R^2} = \frac{(6.674\times10^{-11})(6.42\times10^{23})}{(3.39\times10^6)^2}
= \frac{4.285\times10^{13}}{1.149\times10^{13}} = 3.73\ \mathrm{m/s^2}$$

That is $3.73/9.81 = 0.380$, about **38%** of Earth's.

From $h = \tfrac12 g t^2$, symbolically $t = \sqrt{2h/g}$, so
$t = \sqrt{2(1.50)/3.73} = 0.897\ \mathrm s$ — against $0.553\ \mathrm s$ on Earth. **Limiting check:**
$t \propto 1/\sqrt g$, and $\sqrt{9.81/3.73} = 1.62$, matching $0.897/0.553$ ✅.

**TR:** $g_M = 3.73\ \mathrm{m/s^2}$, Dünya'nın %38'i; düşme süresi $\sqrt{g_E/g_M} = 1.62$ kat uzar.

</details>

#### G2 — The orbit you use every day  ·  L2

Find the altitude above the Earth's surface of a geostationary satellite, whose period must equal one
sidereal day, $86\,164\ \mathrm s$.

<details><summary>Answer</summary>

**Symbolic first**, from $T^2 = 4\pi^2 r^3/(GM_E)$:

$$r = \left(\frac{GM_E T^2}{4\pi^2}\right)^{1/3}$$

$$r = \left(\frac{(6.674\times10^{-11})(5.97\times10^{24})(86\,164)^2}{39.48}\right)^{1/3}
= \left(\frac{2.958\times10^{24}}{39.48}\right)^{1/3} = (7.492\times10^{22})^{1/3} = 4.22\times10^7\ \mathrm m$$

Altitude $= r - R_E = 4.22\times10^7 - 6.37\times10^6 = 3.58\times10^7\ \mathrm m \approx 35\,800\ \mathrm{km}$.

**Check against §3:** we got this same radius from the Moon's orbit without using $G$ at all ✅. Two
independent routes to the same number is the best confirmation available.

Note it must be a **sidereal** day (one rotation relative to the stars), not 86 400 s — the 236-second
difference is the Earth's motion around the Sun, and ignoring it misplaces the satellite.

**TR:** $r = 4.22\times10^7\ \mathrm m$, yükseklik yaklaşık $35\,800\ \mathrm{km}$. **Yıldız günü**
kullanılmalı; 236 saniyelik fark uyduyu yanlış yere koyar.

</details>

#### G3 — Weighing a star with a planet  ·  L3

A transiting planet orbits its star every $8.70\ \mathrm{days}$ at $a = 0.0800\ \mathrm{AU}$. (a) Find the
star's mass, symbolically then numerically, in solar masses. (b) The transit is $2.50\times10^{-3}$ deep
and the star's radius is $0.900\,R_\odot$; find the planet's radius in Earth radii
($R_E = 6.37\times10^6\ \mathrm m$). (c) Comment on what kind of planet it is.

<details><summary>Answer</summary>

**(a) Symbolic.** Rearranging Kepler's third law for the central mass:

$$\boxed{\,M_\star = \frac{4\pi^2 a^3}{G T^2}\,}$$

With $a = 0.0800(1.50\times10^{11}) = 1.20\times10^{10}\ \mathrm m$ and
$T = 8.70(86\,400) = 7.517\times10^5\ \mathrm s$:

$$M_\star = \frac{39.48(1.20\times10^{10})^3}{(6.674\times10^{-11})(7.517\times10^5)^2}
= \frac{6.822\times10^{31}}{3.771\times10^{1}} = 1.81\times10^{30}\ \mathrm{kg} = 0.909\,M_\odot$$

**Limiting check:** $M_\star \propto a^3/T^2$, so a planet the same distance out but with a *longer* year
implies a *lighter* star ✅ — weaker gravity, slower orbit.

**(b)** $R_p = R_\star\sqrt\delta = (0.900)(6.96\times10^8)\sqrt{2.50\times10^{-3}}
= (6.264\times10^8)(0.0500) = 3.13\times10^7\ \mathrm m = 4.92\,R_E$.

**(c)** About five Earth radii and very close in — far too large to be rocky like Earth, far too small to
be a gas giant like Jupiter (11 $R_E$). This is a "sub-Neptune", the commonest type of planet known and
one with no representative in our own Solar System. Note also what the method *cannot* tell you: the
transit gives the radius but not the mass, so its density — and therefore whether it is mostly rock, water
or hydrogen — needs a second, independent measurement.

**TR:** (a) $M_\star = 4\pi^2a^3/(GT^2) = 0.909\,M_\odot$. (b) $R_p = 4.92\,R_E$. (c) Bir "alt-Neptün":
bilinen en yaygın gezegen türü ve Güneş Sistemi'nde örneği olmayan bir sınıf. Yöntemin **söyleyemediğine**
de dikkat et: geçiş yarıçapı verir, kütleyi vermez; yoğunluk için ikinci bir bağımsız ölçüm gerekir.

</details>

**Where this goes next.** The transit method's limits are what current research is about: smaller planets
need better photometry, longer periods need longer observations, and the departure of a transit from a
perfectly repeating clock — **transit timing variations** — reveals additional planets you cannot see at
all. If any of that interests you, ask.

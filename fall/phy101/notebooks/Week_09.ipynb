## 1. From Translation to Rotation — The Big Analogy

Everything you learned about linear motion has a rotational twin:

| Linear Quantity | Symbol | Rotational Quantity | Symbol |
|:---|:---:|:---|:---:|
| Position | $x$ | Angular position | $\theta$ |
| Velocity | $v$ | Angular velocity | $\omega$ |
| Acceleration | $a$ | Angular acceleration | $\alpha$ |
| Mass | $m$ | Moment of inertia | $I$ |
| Force | $F$ | Torque | $\tau$ |
| Momentum | $p = mv$ | Angular momentum | $L = I\omega$ |
| Kinetic energy | $\frac{1}{2}mv^2$ | Rotational KE | $\frac{1}{2}I\omega^2$ |
| Newton's 2nd law | $F = ma$ | Rotational 2nd law | $\tau = I\alpha$ |

> **Think of it this way:** Mass tells you how hard it is to change *linear* motion. Moment of inertia tells you how hard it is to change *rotational* motion. But unlike mass, $I$ depends on **where** the mass is relative to the axis.

## 2. Angular Kinematics

### 2.1 Definitions

For a rigid body rotating about a fixed axis:

$$\omega = \frac{d\theta}{dt}, \quad \alpha = \frac{d\omega}{dt} = \frac{d^2\theta}{dt^2}$$

### 2.2 Constant Angular Acceleration Equations

| # | Equation | Missing variable |
|:-:|:---|:---:|
| 1 | $\omega = \omega_0 + \alpha t$ | $\theta$ |
| 2 | $\theta = \omega_0 t + \frac{1}{2}\alpha t^2$ | $\omega$ (final) |
| 3 | $\omega^2 = \omega_0^2 + 2\alpha\theta$ | $t$ |
| 4 | $\theta = \frac{1}{2}(\omega_0 + \omega)t$ | $\alpha$ |

These are **identical in structure** to the linear equations — just replace $x \to \theta$, $v \to \omega$, $a \to \alpha$.

### 2.3 Connection Between Linear and Angular

For a point at distance $r$ from the rotation axis:

$$s = r\theta, \quad v_{\text{tan}} = r\omega, \quad a_{\text{tan}} = r\alpha, \quad a_{\text{cent}} = \omega^2 r$$

## 🎮 Interactive Demo 1: Animated Rotating Disk with Angular Velocity Display

Watch how a disk rotates under constant angular acceleration. The speedometer shows the instantaneous angular velocity, and the trail shows angular position history.

```python
# ============================================================
# DEMO 1: Animated Rotating Disk / Wheel
# ============================================================

def rotating_disk_animation(omega0=0.5, alpha=0.8, duration=8.0):
    """
    Animate a rotating disk with constant angular acceleration.
    Shows: the disk, a radius marker, angular velocity gauge, and theta vs t plot.
    """
    fps = 30
    n_frames = int(duration * fps)
    dt = duration / n_frames
    times = np.linspace(0, duration, n_frames)
    thetas = omega0 * times + 0.5 * alpha * times**2
    omegas = omega0 + alpha * times

    fig, axes = plt.subplots(1, 3, figsize=(14, 5),
                             gridspec_kw={'width_ratios': [1, 1, 1.2]})
    fig.suptitle('Rotating Disk — Constant Angular Acceleration', fontsize=14, fontweight='bold')

    # --- Left panel: the disk ---
    ax_disk = axes[0]
    ax_disk.set_xlim(-1.5, 1.5)
    ax_disk.set_ylim(-1.5, 1.5)
    ax_disk.set_aspect('equal')
    ax_disk.set_title('Disk (top view)')
    ax_disk.set_xticks([])
    ax_disk.set_yticks([])

    disk_circle = plt.Circle((0, 0), 1.0, fill=False, linewidth=2, color='steelblue')
    ax_disk.add_patch(disk_circle)
    inner = plt.Circle((0, 0), 0.08, fill=True, color='steelblue')
    ax_disk.add_patch(inner)
    radius_line, = ax_disk.plot([], [], 'o-', color='crimson', lw=3, markersize=8)
    # Spoke lines for visual reference
    n_spokes = 4
    spoke_lines = []
    for i in range(n_spokes):
        ln, = ax_disk.plot([], [], '-', color='steelblue', lw=1, alpha=0.5)
        spoke_lines.append(ln)
    info_text = ax_disk.text(0, -1.35, '', ha='center', fontsize=10)

    # --- Middle panel: omega gauge ---
    ax_gauge = axes[1]
    ax_gauge.set_xlim(-1.5, 1.5)
    ax_gauge.set_ylim(-0.3, 1.6)
    ax_gauge.set_aspect('equal')
    ax_gauge.set_title('Angular Velocity Gauge')
    ax_gauge.set_xticks([])
    ax_gauge.set_yticks([])

    # Draw gauge arc
    gauge_angles = np.linspace(np.pi, 0, 100)
    ax_gauge.plot(1.2 * np.cos(gauge_angles), 1.2 * np.sin(gauge_angles), 'k-', lw=2)
    omega_max = omega0 + alpha * duration
    for frac in np.linspace(0, 1, 6):
        a = np.pi * (1 - frac)
        ax_gauge.plot([1.05*np.cos(a), 1.2*np.cos(a)],
                      [1.05*np.sin(a), 1.2*np.sin(a)], 'k-', lw=1.5)
        val = frac * omega_max
        ax_gauge.text(1.35*np.cos(a), 1.35*np.sin(a), f'{val:.1f}',
                      ha='center', va='center', fontsize=8)
    needle, = ax_gauge.plot([], [], '-', color='red', lw=2.5)
    omega_text = ax_gauge.text(0, -0.15, '', ha='center', fontsize=12, fontweight='bold')

    # --- Right panel: theta(t) and omega(t) ---
    ax_plot = axes[2]
    ax_plot.set_xlim(0, duration)
    ax_plot.set_xlabel('Time (s)')
    ax_plot.set_title('Kinematics')
    ax_plot2 = ax_plot.twinx()
    theta_line, = ax_plot.plot([], [], 'b-', lw=2, label=r'$\theta$ (rad)')
    omega_line, = ax_plot2.plot([], [], 'r--', lw=2, label=r'$\omega$ (rad/s)')
    ax_plot.set_ylabel(r'$\theta$ (rad)', color='blue')
    ax_plot2.set_ylabel(r'$\omega$ (rad/s)', color='red')
    ax_plot.set_ylim(0, max(thetas)*1.1)
    ax_plot2.set_ylim(0, max(omegas)*1.1)
    dot_theta, = ax_plot.plot([], [], 'bo', markersize=6)
    dot_omega, = ax_plot2.plot([], [], 'rs', markersize=6)
    lines = [theta_line, omega_line]
    labels = [l.get_label() for l in lines]
    ax_plot.legend(lines, labels, loc='upper left', fontsize=9)

    plt.tight_layout()

    def init():
        return []

    def update(frame):
        t = times[frame]
        theta = thetas[frame]
        omega = omegas[frame]

        # Disk radius marker
        x_end = np.cos(theta)
        y_end = np.sin(theta)
        radius_line.set_data([0, x_end], [0, y_end])

        # Spokes
        for i, ln in enumerate(spoke_lines):
            a = theta + i * 2 * np.pi / n_spokes
            ln.set_data([0, 0.95*np.cos(a)], [0, 0.95*np.sin(a)])

        info_text.set_text(f't = {t:.1f} s | $\\theta$ = {theta:.1f} rad ({np.degrees(theta):.0f}$^\\circ$)')

        # Gauge needle
        frac = omega / omega_max if omega_max > 0 else 0
        frac = min(frac, 1.0)
        needle_angle = np.pi * (1 - frac)
        needle.set_data([0, 0.95*np.cos(needle_angle)], [0, 0.95*np.sin(needle_angle)])
        omega_text.set_text(f'$\\omega$ = {omega:.2f} rad/s')

        # Plots
        theta_line.set_data(times[:frame+1], thetas[:frame+1])
        omega_line.set_data(times[:frame+1], omegas[:frame+1])
        dot_theta.set_data([t], [theta])
        dot_omega.set_data([t], [omega])

        return []

    anim = FuncAnimation(fig, update, frames=n_frames, init_func=init,
                         interval=1000/fps, blit=True)
    plt.close(fig)
    return HTML(anim.to_jshtml())

rotating_disk_animation(omega0=0.5, alpha=0.8, duration=8.0)
```

## 3. Moment of Inertia

### 3.1 Definition

For a system of particles:
$$I = \sum_i m_i r_i^2$$

For a continuous body:
$$I = \int r^2 \, dm$$

where $r$ is the perpendicular distance from the rotation axis.

### 3.2 Standard Moments of Inertia

| Shape | Axis | $I$ |
|:---|:---|:---:|
| Thin ring (mass $M$, radius $R$) | Through center, perpendicular | $MR^2$ |
| Solid disk / cylinder | Through center, perpendicular | $\frac{1}{2}MR^2$ |
| Solid sphere | Through center | $\frac{2}{5}MR^2$ |
| Hollow sphere | Through center | $\frac{2}{3}MR^2$ |
| Thin rod (length $L$) | Through center, perpendicular | $\frac{1}{12}ML^2$ |
| Thin rod (length $L$) | Through end, perpendicular | $\frac{1}{3}ML^2$ |

### 3.3 Parallel-Axis Theorem

$$I = I_{\text{cm}} + Md^2$$

where $d$ is the distance between the center-of-mass axis and the new parallel axis.

> **Physical insight:** The further the mass is from the axis, the larger $I$ is. A ring has ALL its mass at distance $R$, so $I_{\text{ring}} = MR^2$. A disk spreads mass from 0 to $R$, so $I_{\text{disk}} = \frac{1}{2}MR^2$ — only half as much!

## 4. Rotational Kinetic Energy & Rolling Motion

### 4.1 Rotational Kinetic Energy

$$K_{\text{rot}} = \frac{1}{2} I \omega^2$$

### 4.2 Rolling Without Slipping

When an object rolls without slipping:
$$v_{\text{cm}} = R\omega, \quad a_{\text{cm}} = R\alpha$$

Total kinetic energy of a rolling object:
$$K_{\text{total}} = \frac{1}{2}mv_{\text{cm}}^2 + \frac{1}{2}I\omega^2 = \frac{1}{2}mv_{\text{cm}}^2\left(1 + \frac{I}{mR^2}\right)$$

### 4.3 Rolling Down an Incline

Using energy conservation for an object starting from rest at height $h$:

$$mgh = \frac{1}{2}mv^2\left(1 + \frac{I}{mR^2}\right)$$

$$v = \sqrt{\frac{2gh}{1 + I/(mR^2)}}$$

> **Key result:** The object with the smallest $I/(mR^2)$ ratio reaches the bottom first. A solid sphere ($I/(mR^2) = 2/5$) beats a solid cylinder ($1/2$), which beats a ring ($1$). **The shape matters, not the mass or radius!**

## 🎮 Interactive Demo 3: Rolling vs Sliding Down an Incline

Watch three different shapes race down an incline. Who wins? Adjust the incline angle and see the energy bar charts evolve.

```python
# ============================================================
# DEMO 3: Rolling Race Down an Incline
# ============================================================

def rolling_race_animation(angle_deg=30, h=2.0):
    """
    Animate solid sphere, solid cylinder, and ring rolling down an incline.
    Shows positions, velocities, and energy distribution.
    """
    g = 9.81
    angle = np.radians(angle_deg)
    L_incline = h / np.sin(angle)  # length along the incline

    # I/(mR^2) ratios
    objects = {
        'Solid Sphere': {'c': 2/5, 'color': '#2ecc71', 'marker': 'o'},
        'Solid Cylinder': {'c': 1/2, 'color': '#3498db', 'marker': 's'},
        'Ring': {'c': 1.0, 'color': '#e74c3c', 'marker': '^'},
    }

    # Calculate arrival times and motions
    # a_cm = g sin(theta) / (1 + c)
    results = {}
    t_max = 0
    for name, props in objects.items():
        c = props['c']
        a_cm = g * np.sin(angle) / (1 + c)
        t_arrive = np.sqrt(2 * L_incline / a_cm)
        t_max = max(t_max, t_arrive)
        results[name] = {'a': a_cm, 't_arrive': t_arrive}

    duration = t_max * 1.15
    fps = 30
    n_frames = int(duration * fps)
    times = np.linspace(0, duration, n_frames)

    fig, (ax_incline, ax_energy) = plt.subplots(1, 2, figsize=(14, 6),
                                                 gridspec_kw={'width_ratios': [1.5, 1]})
    fig.suptitle(f'Rolling Race — Incline {angle_deg}$^\\circ$, Height {h:.1f} m', fontsize=14, fontweight='bold')

    # --- Incline panel ---
    x_start = 0
    y_start = h
    x_end = h / np.tan(angle)
    y_end = 0
    ax_incline.plot([x_start, x_end], [y_start, y_end], 'k-', lw=3)
    ax_incline.plot([x_end, x_end], [y_end, y_start], 'k--', lw=1, alpha=0.3)
    ax_incline.fill_between([x_start, x_end], [y_start, y_end], alpha=0.05, color='gray')
    margin = 0.5
    ax_incline.set_xlim(-margin, x_end + margin)
    ax_incline.set_ylim(-margin, h + margin)
    ax_incline.set_aspect('equal')
    ax_incline.set_xlabel('x (m)')
    ax_incline.set_ylabel('y (m)')
    ax_incline.set_title('Incline View')
    ax_incline.axhline(y=0, color='brown', lw=2)

    # Object markers
    obj_plots = {}
    obj_labels = {}
    offsets = [-0.12, 0, 0.12]  # vertical offsets so they don't overlap
    for i, (name, props) in enumerate(objects.items()):
        p, = ax_incline.plot([], [], props['marker'], color=props['color'],
                             markersize=14, markeredgecolor='black', label=name)
        obj_plots[name] = p
        lbl = ax_incline.text(0, 0, '', fontsize=8, color=props['color'], fontweight='bold')
        obj_labels[name] = lbl
    ax_incline.legend(loc='upper right', fontsize=9)
    time_text = ax_incline.text(0.02, 0.02, '', transform=ax_incline.transAxes, fontsize=11)

    # --- Energy panel ---
    ax_energy.set_title('Energy Distribution at Bottom')
    ax_energy.set_ylabel('Energy (J)')

    # Use m=1kg for energy display
    m = 1.0
    E_total = m * g * h
    bar_names = list(objects.keys())
    x_pos = np.arange(len(bar_names))
    bar_width = 0.35
    bars_trans = ax_energy.bar(x_pos - bar_width/2, [0]*3, bar_width,
                               label='Translational KE', color='steelblue')
    bars_rot = ax_energy.bar(x_pos + bar_width/2, [0]*3, bar_width,
                              label='Rotational KE', color='coral')
    ax_energy.set_xticks(x_pos)
    ax_energy.set_xticklabels(['Sphere', 'Cylinder', 'Ring'], fontsize=9)
    ax_energy.set_ylim(0, E_total * 1.1)
    ax_energy.axhline(y=E_total, color='green', linestyle='--', label=f'Total = mgh = {E_total:.1f} J')
    ax_energy.legend(fontsize=8)

    plt.tight_layout()

    def update(frame):
        t = times[frame]
        time_text.set_text(f't = {t:.2f} s')

        for i, (name, props) in enumerate(objects.items()):
            a = results[name]['a']
            c = props['c']
            s = min(0.5 * a * t**2, L_incline)  # distance along incline
            frac = s / L_incline

            # Position on incline
            x_pos_obj = x_start + frac * (x_end - x_start) + offsets[i] * np.cos(angle + np.pi/2)
            y_pos_obj = y_start + frac * (y_end - y_start) + offsets[i] * np.sin(angle + np.pi/2) + 0.08
            obj_plots[name].set_data([x_pos_obj], [y_pos_obj])

            v = min(a * t, np.sqrt(2 * a * L_incline))
            obj_labels[name].set_position((x_pos_obj + 0.15, y_pos_obj + 0.1))
            obj_labels[name].set_text(f'v={v:.1f}')

            # Energy bars
            KE_trans = 0.5 * m * v**2
            KE_rot = 0.5 * m * v**2 * c
            bars_trans[i].set_height(KE_trans)
            bars_rot[i].set_height(KE_rot)

        return []

    anim = FuncAnimation(fig, update, frames=n_frames, interval=1000/fps, blit=True)
    plt.close(fig)
    return HTML(anim.to_jshtml())

rolling_race_animation(angle_deg=30, h=2.0)
```

**Observations:**
- The solid sphere always wins — it has the smallest $I/(mR^2) = 2/5$
- The ring always comes last — all its mass is at the maximum distance from the axis
- Notice how the ring puts more energy into rotation and less into translation
- **The result is independent of mass and radius!** (Try changing them in the code)

## 5. Pulley-Rope Systems

When a pulley has mass (and therefore moment of inertia), we must include its rotational dynamics.

### Atwood Machine with a Massive Pulley

Two masses $m_1$ and $m_2$ connected by a rope over a pulley of mass $M$ and radius $R$:

The tensions on each side are **different** (because the pulley has inertia):

$$m_1 g - T_1 = m_1 a$$
$$T_2 - m_2 g = m_2 a$$
$$T_1 R - T_2 R = I\alpha = I \frac{a}{R}$$

Solving for the acceleration:

$$a = \frac{(m_1 - m_2)g}{m_1 + m_2 + I/R^2}$$

For a solid disk pulley ($I = \frac{1}{2}MR^2$):

$$a = \frac{(m_1 - m_2)g}{m_1 + m_2 + M/2}$$

> **Key insight:** The pulley's inertia acts like extra mass in the denominator, slowing the system down compared to a massless-pulley Atwood machine.

## 🎮 Interactive Demo 4: Pulley System Simulator

Adjust the masses and pulley parameters to see how the system accelerates.

```python
# ============================================================
# DEMO 4: Atwood Machine with Massive Pulley
# ============================================================

def pulley_simulator(m1=5.0, m2=3.0, M_pulley=2.0, R_pulley=0.15):
    """
    Simulate and animate an Atwood machine with a massive pulley.
    m1 > m2 assumed. Pulley is a solid disk.
    """
    g = 9.81
    I_pulley = 0.5 * M_pulley * R_pulley**2

    # Acceleration
    a = (m1 - m2) * g / (m1 + m2 + I_pulley / R_pulley**2)
    a_massless = (m1 - m2) * g / (m1 + m2)  # comparison

    # Tensions
    T1 = m1 * (g - a)
    T2 = m2 * (g + a)

    # Animation
    duration = 3.0
    fps = 30
    n_frames = int(duration * fps)
    times = np.linspace(0, duration, n_frames)

    fig, (ax_sys, ax_info) = plt.subplots(1, 2, figsize=(13, 7),
                                           gridspec_kw={'width_ratios': [1, 1.2]})
    fig.suptitle('Atwood Machine with Massive Pulley', fontsize=14, fontweight='bold')

    # System view
    ax_sys.set_xlim(-2, 2)
    ax_sys.set_ylim(-4, 2)
    ax_sys.set_aspect('equal')
    ax_sys.set_title('System')

    # Pulley
    pulley_center = (0, 1)
    pulley_r = 0.4
    pulley_circle = plt.Circle(pulley_center, pulley_r, fill=False, lw=3, color='gray')
    ax_sys.add_patch(pulley_circle)
    ax_sys.plot([0], [1], '+', color='black', markersize=10, markeredgewidth=2)
    # Support
    ax_sys.plot([-0.5, 0.5], [1.5, 1.5], 'k-', lw=4)
    ax_sys.plot([0, 0], [1.4, 1.5], 'k-', lw=2)
    # Spoke for rotation visual
    spoke, = ax_sys.plot([], [], '-', color='gray', lw=2)

    # Ropes and masses
    rope_left, = ax_sys.plot([], [], 'k-', lw=1.5)
    rope_right, = ax_sys.plot([], [], 'k-', lw=1.5)

    mass1_box = mpatches.FancyBboxPatch((-1.3, 0), 0.6, 0.6, boxstyle="round,pad=0.05",
                                         facecolor='#e74c3c', edgecolor='black', lw=2)
    mass2_box = mpatches.FancyBboxPatch((0.7, 0), 0.6, 0.6, boxstyle="round,pad=0.05",
                                         facecolor='#3498db', edgecolor='black', lw=2)
    ax_sys.add_patch(mass1_box)
    ax_sys.add_patch(mass2_box)
    label1 = ax_sys.text(-1, 0, '', ha='center', va='center', fontsize=10,
                         color='white', fontweight='bold')
    label2 = ax_sys.text(1, 0, '', ha='center', va='center', fontsize=10,
                         color='white', fontweight='bold')

    time_text = ax_sys.text(0.02, 0.02, '', transform=ax_sys.transAxes, fontsize=10)

    # Info panel
    ax_info.axis('off')
    info_lines = [
        f'Parameters:',
        f'  m$_1$ = {m1:.1f} kg (heavier, left)',
        f'  m$_2$ = {m2:.1f} kg (lighter, right)',
        f'  M$_{{pulley}}$ = {M_pulley:.1f} kg (solid disk)',
        f'  R$_{{pulley}}$ = {R_pulley:.2f} m',
        f'  I$_{{pulley}}$ = {I_pulley:.4f} kg m$^2$',
        f'',
        f'Results:',
        f'  a = {a:.3f} m/s$^2$',
        f'  a (massless pulley) = {a_massless:.3f} m/s$^2$',
        f'  Reduction = {(1 - a/a_massless)*100:.1f}%',
        f'',
        f'  T$_1$ = {T1:.2f} N',
        f'  T$_2$ = {T2:.2f} N',
        f'  T$_1$ - T$_2$ = {T1-T2:.2f} N (net torque / R)',
        f'',
        f'Equations:',
        f'  a = (m$_1$ - m$_2$)g / (m$_1$ + m$_2$ + I/R$^2$)',
        f'  With massless pulley: a = (m$_1$ - m$_2$)g / (m$_1$ + m$_2$)',
    ]
    ax_info.text(0.05, 0.95, '\n'.join(info_lines), transform=ax_info.transAxes,
                 fontsize=11, verticalalignment='top', fontfamily='monospace',
                 bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

    plt.tight_layout()

    def update(frame):
        t = times[frame]
        s = 0.5 * a * t**2  # displacement
        theta = s / R_pulley  # pulley angle

        # m1 goes down, m2 goes up
        y1_center = 0 - s
        y2_center = -0.5 + s

        mass1_box.set_y(y1_center - 0.3)
        mass2_box.set_y(y2_center - 0.3)
        label1.set_position((-1, y1_center))
        label1.set_text(f'm$_1$\n{m1:.0f}kg')
        label2.set_position((1, y2_center))
        label2.set_text(f'm$_2$\n{m2:.0f}kg')

        # Ropes
        rope_left.set_data([-pulley_r + pulley_center[0], -pulley_r + pulley_center[0]],
                           [pulley_center[1], y1_center + 0.3])
        rope_right.set_data([pulley_r + pulley_center[0], pulley_r + pulley_center[0]],
                            [pulley_center[1], y2_center + 0.3])

        # Pulley spoke
        sx = pulley_center[0] + pulley_r * 0.9 * np.cos(theta)
        sy = pulley_center[1] + pulley_r * 0.9 * np.sin(theta)
        spoke.set_data([pulley_center[0], sx], [pulley_center[1], sy])

        v = a * t
        time_text.set_text(f't = {t:.2f} s\nv = {v:.2f} m/s\ns = {s:.3f} m')

        return []

    anim = FuncAnimation(fig, update, frames=n_frames, interval=1000/fps, blit=True)
    plt.close(fig)
    return HTML(anim.to_jshtml())

pulley_simulator(m1=5.0, m2=3.0, M_pulley=2.0, R_pulley=0.15)
```

**Try it:** Change `M_pulley` to 0 and compare the acceleration with the massless-pulley formula. Then increase `M_pulley` and watch the system slow down.

## 🔧 Worked Example 1: Angular Kinematics

**Problem:** A grinding wheel starts from rest and accelerates uniformly at $\alpha = 3.0$ rad/s$^2$ for 8 seconds, then coasts at constant angular velocity for 10 seconds, then decelerates uniformly to rest in 20 seconds. Find:
(a) The maximum angular velocity
(b) The angular deceleration in phase 3
(c) The total angle turned through

```python
# Worked Example 1: Angular Kinematics — Three-phase motion

alpha1 = 3.0   # rad/s^2 (acceleration phase)
t1 = 8.0       # s
t2 = 10.0      # s (coasting phase)
t3 = 20.0      # s (deceleration phase)

# (a) Maximum angular velocity
omega_max = 0 + alpha1 * t1
print(f"(a) Maximum angular velocity: omega_max = alpha * t1 = {alpha1} * {t1} = {omega_max:.1f} rad/s")
print(f"    That's {omega_max/(2*np.pi):.1f} rev/s = {omega_max/(2*np.pi)*60:.0f} RPM")

# (b) Angular deceleration
alpha3 = -omega_max / t3
print(f"\n(b) Deceleration: alpha3 = -omega_max / t3 = -{omega_max}/{t3} = {alpha3:.2f} rad/s^2")

# (c) Total angle
theta1 = 0.5 * alpha1 * t1**2
theta2 = omega_max * t2
theta3 = omega_max * t3 + 0.5 * alpha3 * t3**2
theta_total = theta1 + theta2 + theta3

print(f"\n(c) Phase 1: theta1 = 0.5 * {alpha1} * {t1}^2 = {theta1:.1f} rad")
print(f"    Phase 2: theta2 = {omega_max} * {t2} = {theta2:.1f} rad")
print(f"    Phase 3: theta3 = {omega_max}*{t3} + 0.5*({alpha3})*{t3}^2 = {theta3:.1f} rad")
print(f"    Total: theta = {theta1:.1f} + {theta2:.1f} + {theta3:.1f} = {theta_total:.1f} rad")
print(f"    That's {theta_total/(2*np.pi):.1f} complete revolutions")

# Visualization
t_all = np.concatenate([
    np.linspace(0, t1, 100),
    np.linspace(t1, t1+t2, 100),
    np.linspace(t1+t2, t1+t2+t3, 100)
])
omega_all = np.concatenate([
    alpha1 * np.linspace(0, t1, 100),
    omega_max * np.ones(100),
    omega_max + alpha3 * np.linspace(0, t3, 100)
])

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
ax1.plot(t_all, omega_all, 'b-', lw=2)
ax1.fill_between(t_all, omega_all, alpha=0.2)
ax1.set_ylabel(r'$\omega$ (rad/s)')
ax1.set_title('Angular Velocity vs Time')
ax1.axvline(t1, color='gray', ls='--', alpha=0.5)
ax1.axvline(t1+t2, color='gray', ls='--', alpha=0.5)
ax1.text(t1/2, omega_max*0.5, 'Phase 1\nAccel', ha='center', fontsize=10)
ax1.text(t1+t2/2, omega_max*0.5, 'Phase 2\nCoast', ha='center', fontsize=10)
ax1.text(t1+t2+t3/2, omega_max*0.5, 'Phase 3\nDecel', ha='center', fontsize=10)
ax1.grid(True, alpha=0.3)

# Compute theta for each time
theta_all_1 = 0.5 * alpha1 * np.linspace(0, t1, 100)**2
theta_all_2 = theta1 + omega_max * np.linspace(0, t2, 100)
t3_arr = np.linspace(0, t3, 100)
theta_all_3 = theta1 + theta2 + omega_max * t3_arr + 0.5 * alpha3 * t3_arr**2
theta_all = np.concatenate([theta_all_1, theta_all_2, theta_all_3])

ax2.plot(t_all, theta_all, 'r-', lw=2)
ax2.set_xlabel('Time (s)')
ax2.set_ylabel(r'$\theta$ (rad)')
ax2.set_title('Angular Position vs Time')
ax2.axvline(t1, color='gray', ls='--', alpha=0.5)
ax2.axvline(t1+t2, color='gray', ls='--', alpha=0.5)
ax2.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Joining this lesson / Derse buradan başlayanlar

**Start from familiar motion and energy equations; no prior Python is required.** For a rigid body rotating about a fixed axis:

- $\tau_{\rm net}=I\alpha$: net torque changes angular speed. $I$ measures how mass is distributed about the chosen axis; a solid disk has $I=\tfrac12MR^2$.
- With **constant** $\alpha$, $\omega=\omega_0+\alpha t$ and $\theta=\omega_0t+\tfrac12\alpha t^2$. Convert RPM using $\omega=\mathrm{RPM}\times2\pi/60$.
- Rotational energy is $K_{\rm rot}=\tfrac12I\omega^2$. Rolling **without slipping** also requires $v=R\omega$ and includes translational energy $\tfrac12Mv^2$.

**Quick entry check:** For $I=0.20\,\mathrm{kg\,m^2}$ and constant net torque $0.60\,\mathrm{N\,m}$, $\alpha=0.60/0.20=3.0\,\mathrm{rad/s^2}$. Starting from rest, two seconds gives $\omega=6.0\,\mathrm{rad/s}$. This result assumes the quoted torque is the net torque after any resistance.

**Unit bridge:** $120\,\mathrm{RPM}$ means 120 revolutions per minute, so $120\times(2\pi)/60=4\pi=12.57\,\mathrm{rad/s}$. Convert before calculating energy. Radians have no physical dimension, but writing rad helps identify an angle.

**Rolling bridge:** Without slipping, $v=R\omega$, so $\omega=v/R$. Substitute this into BOTH energy terms:
$$Mgh=\tfrac12Mv^2+\tfrac12I(v/R)^2
=\tfrac12\left(M+I/R^2\right)v^2.$$
Multiply by 2, divide by $M+I/R^2$, then take the square root. If $I=\beta MR^2$, the result is $v=\sqrt{\frac{2gh}{1+\beta}}$. Rotation uses some of the energy; the speed is smaller than the sliding value $\sqrt{2gh}$.

**Pulley bridge:** Translating masses obey $\sum F=ma$; the pulley obeys $\sum\tau=I\alpha$. The connection is $a=R\alpha$. A massive accelerating pulley generally needs unequal tensions to produce torque.

**Pause:** Doubling $R$ at fixed mass quadruples $I$ for the same shape. But in ideal rolling from the same height, $I/(MR^2)=\beta$ stays the same, so the final speed does not change.

## 2. Angular Kinematics

### 2.1 Definitions

For a rigid body rotating about a fixed axis:

$$\omega = \frac{d\theta}{dt}, \quad \alpha = \frac{d\omega}{dt} = \frac{d^2\theta}{dt^2}$$

These are **identical in structure** to the linear equations — just replace $x \to \theta$, $v \to \omega$, $a \to \alpha$.

### 2.3 Connection Between Linear and Angular

For a point at distance $r$ from the rotation axis:

$$s = r\theta, \quad v_{\text{tan}} = r\omega, \quad a_{\text{tan}} = r\alpha, \quad a_{\text{cent}} = \omega^2 r$$

**Use a unit conversion before the kinematics.** At $120\,\mathrm{rev/min}$:

$$\omega_0=120\,\frac{\mathrm{rev}}{\mathrm{min}}\frac{2\pi\,\mathrm{rad}}{1\,\mathrm{rev}}\frac{1\,\mathrm{min}}{60\,\mathrm s}=4\pi\,\mathrm{rad/s}.$$

If it then brakes uniformly to rest in $4.0\,\mathrm s$,
$$\alpha=\frac{\omega_f-\omega_0}{t}=\frac{0-4\pi}{4}=-\pi\,\mathrm{rad/s^2},\qquad\Delta\theta=\frac{4\pi+0}{2}(4)=8\pi\,\mathrm{rad}=4\,\mathrm{rev}.$$

Here $\theta$ in the table means angular displacement from the initial orientation; otherwise write $\theta-\theta_0$. **TR:** Negatif ivme burada frenlemeyi gösterir. Süre boyunca hız sıfıra yaklaşırken dönme açısı yine pozitif birikir.

```python
#@title Run once — prepare compact demonstration controls
"""Embedded in the lecture notebooks by refresh_notebook_interface.py.

Only standard ipywidgets and IPython are used; notebooks remain self-contained.
"""
import functools
import traceback
import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import display, HTML
from IPython.utils.capture import capture_output
from ipywidgets import interactive as _PhysicsInteractive
from ipywidgets.widgets.interaction import show_inline_matplotlib_plots

# Set this explicitly: otherwise a local kernel can choose a desktop backend
# and block at plt.show(), leaving the notebook apparently frozen.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

try:
    from google.colab import output as _colab_output
except ImportError:
    pass
else:
    _colab_output.enable_custom_widget_manager()

_physics_panels = []
_physics_callback_errors = []


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; }
.phy101-controls { padding: 0 0 4px; box-sizing: border-box; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-animation-panel { max-width: 100%; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1;
  background: white; color: #172433; padding: 4px; }
@media (max-width: 650px) {
  .phy101-panel { flex-direction: column !important; }
  .phy101-controls, .phy101-plot-output { width: 100% !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Fit full labels and postpone costly redraws until a slider is released."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "continuous_update"):
            control.continuous_update = False
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "310px"
        control.layout.flex = "0 1 310px"
        control.layout.max_width = "100%"
        control.layout.min_width = "0"
        control.layout.margin = "2px 6px 2px 0"
        if isinstance(control, widgets.Button):
            control.layout.width = "auto"
            control.layout.flex = "0 0 auto"
    # Include the style in each output: Colab may isolate output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.Box([style] + flat, layout=widgets.Layout(
        display="flex", flex_flow="row wrap", align_items="center",
        width="100%", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        width="100%", min_width="0", max_width="100%",
        height="auto", overflow="visible", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Keep a compact control toolbar directly above the complete result."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="column",
                              align_items="stretch", width="100%"))
    panel.add_class("phy101-panel")
    return panel


class _PhysicsStableInteractive(_PhysicsInteractive):
    """Replace one persistent output model only after a redraw has completed.

    Sending clear_output followed by an inline figure depends on callback message
    capture in the frontend. Explicit MIME records keep repeated slider changes
    visible in Jupyter/Colab, without clearing the surrounding controls.
    """
    def update(self, *args):
        self.kwargs = {}
        if self.manual:
            self.manual_button.disabled = True
        try:
            for control in self.kwargs_widgets:
                self.kwargs[control._kwarg] = control.get_interact_value()
            with capture_output(stdout=True, stderr=True, display=True) as captured:
                self.result = self.f(**self.kwargs)
                show_inline_matplotlib_plots()
                if self.auto_display and self.result is not None:
                    display(self.result)
            records = []
            if captured.stdout:
                records.append({"output_type": "stream", "name": "stdout", "text": captured.stdout})
            if captured.stderr:
                records.append({"output_type": "stream", "name": "stderr", "text": captured.stderr})
            records.extend({"output_type": "display_data", "data": output.data,
                            "metadata": output.metadata or {}} for output in captured.outputs)
            # Keep the old result visible during calculation; publish the complete
            # replacement as one widget-state update, including its PNG/HTML data.
            self.out.outputs = tuple(records)
        except Exception:
            _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), traceback.format_exc()))
            self.out.outputs = ({"output_type": "stream", "name": "stderr",
                                 "text": traceback.format_exc()},)
        finally:
            if self.manual:
                self.manual_button.disabled = False


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


def physics_interactive(function, **controls):
    """An interactive widget with the same arguments as ipywidgets.interactive."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        try:
            return function(*args, **kwargs)
        except Exception as error:
            _physics_callback_errors.append((function.__name__, repr(error)))
            raise

    panel = _PhysicsStableInteractive(checked, {"auto_display": True}, **controls)
    inputs = [child for child in panel.children if child is not panel.out]
    panel.children = (_physics_controls(inputs), _physics_output(panel.out))
    panel.layout = widgets.Layout(display="flex", flex_flow="column",
                                  align_items="stretch", width="100%")
    panel.add_class("phy101-panel")
    _physics_panels.append(panel)
    return panel


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel
    display(panel)
    return function
```

## 🎮 Interactive Demo 1: Animated Rotating Disk with Angular Velocity Display

```python
#@title Run the animation — observe the physical motion
# ============================================================
# DEMO 1: Animated Rotating Disk / Wheel
# ============================================================

def rotating_disk_animation(omega0=0.5, alpha=0.8, duration=8.0):
    """
    Animate a rotating disk with constant angular acceleration.
    Shows: the disk, a radius marker, angular velocity gauge, and theta vs t plot.
    """
    if duration <= 0:
        raise ValueError('duration must be positive')
    fps = 30
    n_frames = max(2, int(duration * fps))
    dt = duration / n_frames
    times = np.linspace(0, duration, n_frames)
    thetas = omega0 * times + 0.5 * alpha * times**2
    omegas = omega0 + alpha * times

    fig = plt.figure(figsize=(14, 6), layout='constrained')
    grid = fig.add_gridspec(2, 3, width_ratios=[1, 1, 1.3])
    axes = [fig.add_subplot(grid[:, 0]), fig.add_subplot(grid[:, 1]),
            fig.add_subplot(grid[0, 2])]
    lower_plot = fig.add_subplot(grid[1, 2], sharex=axes[2])
    fig.suptitle('Rotating Disk — Constant Angular Acceleration', fontsize=14, fontweight='bold')

    # --- Left panel: the disk ---
    ax_disk = axes[0]
    ax_disk.set_xlim(-1.5, 1.5)
    ax_disk.set_ylim(-1.5, 1.5)
    ax_disk.set_aspect('equal')
    ax_disk.set_title('Disk (top view)')
    ax_disk.set_xticks([])
    ax_disk.set_yticks([])

    disk_circle = plt.Circle((0, 0), 1.0, fill=False, linewidth=2, color='steelblue')
    ax_disk.add_patch(disk_circle)
    inner = plt.Circle((0, 0), 0.08, fill=True, color='steelblue')
    ax_disk.add_patch(inner)
    radius_line, = ax_disk.plot([], [], 'o-', color='crimson', lw=3, markersize=8)
    # Spoke lines for visual reference
    n_spokes = 4
    spoke_lines = []
    for i in range(n_spokes):
        ln, = ax_disk.plot([], [], '-', color='steelblue', lw=1, alpha=0.5)
        spoke_lines.append(ln)
    info_text = ax_disk.text(0, -1.35, '', ha='center', fontsize=10)

    # --- Middle panel: omega gauge ---
    ax_gauge = axes[1]
    ax_gauge.set_xlim(-1.5, 1.5)
    ax_gauge.set_ylim(-0.3, 1.6)
    ax_gauge.set_aspect('equal')
    ax_gauge.set_title('Angular Velocity Gauge')
    ax_gauge.set_xticks([])
    ax_gauge.set_yticks([])

    # Draw gauge arc
    gauge_angles = np.linspace(np.pi, 0, 100)
    ax_gauge.plot(1.2 * np.cos(gauge_angles), 1.2 * np.sin(gauge_angles), 'k-', lw=2)
    omega_min = min(0.0, float(np.min(omegas)))
    omega_max = max(0.0, float(np.max(omegas)))
    if omega_max == omega_min:
        omega_min, omega_max = -1.0, 1.0
    for frac in np.linspace(0, 1, 6):
        a = np.pi * (1 - frac)
        ax_gauge.plot([1.05*np.cos(a), 1.2*np.cos(a)],
                      [1.05*np.sin(a), 1.2*np.sin(a)], 'k-', lw=1.5)
        val = omega_min + frac * (omega_max - omega_min)
        ax_gauge.text(1.35*np.cos(a), 1.35*np.sin(a), f'{val:.1f}',
                      ha='center', va='center', fontsize=8)
    needle, = ax_gauge.plot([], [], '-', color='red', lw=2.5)
    omega_text = ax_gauge.text(0, -0.15, '', ha='center', fontsize=12, fontweight='bold')

    # --- Right panel: theta(t) and omega(t) ---
    ax_plot = axes[2]
    ax_plot.set_xlim(0, duration)
    ax_plot.set_xlabel('Time (s)')
    ax_plot.set_title('Angular position')
    ax_plot2 = lower_plot
    ax_plot2.set_title('Angular velocity (signed)')
    ax_plot2.set_xlabel('Time (s)')
    theta_line, = ax_plot.plot([], [], 'b-', lw=2, label=r'$\theta$ (rad)')
    omega_line, = ax_plot2.plot([], [], 'r--', lw=2, label=r'$\omega$ (rad/s)')
    ax_plot.set_ylabel(r'$\theta$ (rad)', color='blue')
    ax_plot2.set_ylabel(r'$\omega$ (rad/s)', color='red')
    for axis, values in ((ax_plot, thetas), (ax_plot2, omegas)):
        low = min(0.0, float(np.min(values)))
        high = max(0.0, float(np.max(values)))
        padding = max(0.1, 0.1 * (high - low))
        axis.set_ylim(low - padding, high + padding)
        axis.axhline(0, color='gray', linewidth=0.7)
        axis.grid(True, alpha=0.2)
    dot_theta, = ax_plot.plot([], [], 'bo', markersize=6)
    dot_omega, = ax_plot2.plot([], [], 'rs', markersize=6)
    lines = [theta_line, omega_line]
    labels = [l.get_label() for l in lines]
    ax_plot.legend(loc='upper right', fontsize=8)
    ax_plot2.legend(loc='upper right', fontsize=8)

    def init():
        return []

    def update(frame):
        t = times[frame]
        theta = thetas[frame]
        omega = omegas[frame]

        # Disk radius marker
        x_end = np.cos(theta)
        y_end = np.sin(theta)
        radius_line.set_data([0, x_end], [0, y_end])

        # Spokes
        for i, ln in enumerate(spoke_lines):
            a = theta + i * 2 * np.pi / n_spokes
            ln.set_data([0, 0.95*np.cos(a)], [0, 0.95*np.sin(a)])

        info_text.set_text(f't = {t:.1f} s\n$\\theta$ = {theta:.1f} rad ({np.degrees(theta):.0f}$^\\circ$)')

        # Gauge needle
        frac = np.clip((omega - omega_min) / (omega_max - omega_min), 0.0, 1.0)
        needle_angle = np.pi * (1 - frac)
        needle.set_data([0, 0.95*np.cos(needle_angle)], [0, 0.95*np.sin(needle_angle)])
        omega_text.set_text(f'$\\omega$ = {omega:.2f} rad/s')

        # Plots
        theta_line.set_data(times[:frame+1], thetas[:frame+1])
        omega_line.set_data(times[:frame+1], omegas[:frame+1])
        dot_theta.set_data([t], [theta])
        dot_omega.set_data([t], [omega])

        return []

    anim = FuncAnimation(fig, update, frames=physics_frames(n_frames), init_func=init,
                         interval=physics_interval(n_frames, 1000/fps), blit=True)
    plt.close(fig)
    return physics_animation_html(anim)

rotating_disk_animation(omega0=0.5, alpha=0.8, duration=8.0)
```

## 4. Rotational Kinetic Energy & Rolling Motion

### 4.1 Rotational Kinetic Energy

$$K_{\text{rot}} = \frac{1}{2} I \omega^2$$

### 4.2 Rolling Without Slipping

When an object rolls without slipping:
$$v_{\text{cm}} = R\omega, \quad a_{\text{cm}} = R\alpha$$

Total kinetic energy of a rolling object:
$$K_{\text{total}} = \frac{1}{2}mv_{\text{cm}}^2 + \frac{1}{2}I\omega^2 = \frac{1}{2}mv_{\text{cm}}^2\left(1 + \frac{I}{mR^2}\right)$$

### 4.3 Rolling Down an Incline

Using energy conservation for an object starting from rest at height $h$:

$$mgh = \frac{1}{2}mv^2\left(1 + \frac{I}{mR^2}\right)$$

$$v = \sqrt{\frac{2gh}{1 + I/(mR^2)}}$$

> **Key result:** For the same ramp, release from rest, sufficient static friction to prevent slipping, and negligible dissipative losses, the object with the smallest $I/(mR^2)$ ratio reaches the bottom first. A solid sphere ($I/(mR^2) = 2/5$) beats a solid cylinder ($1/2$), which beats a ring ($1$). **The shape matters, not the mass or radius!**

```python
#@title Run the animation — observe the physical motion
# ---- Rolling race: positions, speeds and energy together ----

def rolling_race_animation(angle_deg=30, h=2.0):
    """Show current speeds in a fixed table, beside the motion and energy."""
    g = 9.81
    angle = np.radians(angle_deg)
    L_incline = h / np.sin(angle)
    objects = {
        'Solid sphere': {'c': 2/5, 'color': '#237c49', 'marker': 'o'},
        'Solid cylinder': {'c': 1/2, 'color': '#2155a3', 'marker': 's'},
        'Ring': {'c': 1.0, 'color': '#bd3028', 'marker': '^'},
    }
    results = {}
    for name, props in objects.items():
        a_cm = g * np.sin(angle) / (1 + props['c'])
        results[name] = {'a': a_cm, 't_arrive': np.sqrt(2 * L_incline / a_cm)}
    duration = max(value['t_arrive'] for value in results.values()) * 1.15
    fps = 30
    n_frames = max(2, int(duration * fps))
    times = np.linspace(0, duration, n_frames)

    fig = plt.figure(figsize=(11.6, 4.3), layout='constrained')
    grid = fig.add_gridspec(2, 2, width_ratios=[1.35, 1], height_ratios=[2.5, 1])
    ax_incline = fig.add_subplot(grid[0, 0])
    info = fig.add_subplot(grid[1, 0])
    ax_energy = fig.add_subplot(grid[:, 1])
    fig.suptitle(rf'Rolling without slipping: incline ${angle_deg:.0f}^\circ$, height ${h:.1f}\,\mathrm{{m}}$',
                 fontsize=12)

    x_end = h / np.tan(angle)
    ax_incline.plot([0, x_end], [h, 0], color='#333333', lw=2)
    ax_incline.fill_between([0, x_end], [h, 0], alpha=0.05, color='gray')
    ax_incline.set(xlim=(-0.5, x_end+0.5), ylim=(-0.5, h+0.5), xlabel='x (m)', ylabel='y (m)')
    ax_incline.set_aspect('equal')
    ax_incline.axhline(0, color='#795548', lw=1)
    time_title = ax_incline.set_title('Positions at t = 0.00 s', fontsize=10)

    obj_plots = {}
    offsets = [-0.12, 0, 0.12]
    for name, props in objects.items():
        obj_plots[name], = ax_incline.plot([], [], props['marker'], color=props['color'],
                                          markersize=10, markeredgecolor='black')
    info.axis('off')
    table = info.table(cellText=[[name, '0.00', f'{results[name]["t_arrive"]:.2f}'] for name in objects],
                       colLabels=['Object', 'Speed (m/s)', 'Arrival time (s)'],
                       colWidths=[0.44,0.26,0.30],cellLoc='center',bbox=[0,0.06,1,0.94])
    table.auto_set_font_size(False);table.set_fontsize(9)
    for (row,col),cell in table.get_celld().items():
        cell.set_edgecolor('#d4dbe2')
        if row==0:
            cell.set_facecolor('#e7edf3');cell.get_text().set_fontweight('bold')
        else:
            cell.set_facecolor('#ffffff' if row%2 else '#f4f7fa')
            if col==0:cell.get_text().set_color(list(objects.values())[row-1]['color'])

    m = 1.0
    E_total = m * g * h
    x_pos = np.arange(len(objects));bar_width=0.35
    bars_trans = ax_energy.bar(x_pos-bar_width/2, [0]*3, bar_width,
                               label='Translational KE',color='#386f9d')
    bars_rot = ax_energy.bar(x_pos+bar_width/2, [0]*3, bar_width,
                             label='Rotational KE',color='#cf643f')
    ax_energy.set_xticks(x_pos, ['Sphere','Cylinder','Ring'])
    ax_energy.set(ylim=(0,E_total*1.45), ylabel='Energy (J)')
    ax_energy.set_title('Current kinetic energy — each mass 1 kg', fontsize=10)
    ax_energy.axhline(E_total,color='#237c49',ls='--',
                      label=rf'Initial $mgh={E_total:.1f}$ J')
    ax_energy.legend(loc='upper right',fontsize=8.5,framealpha=1)
    for ax in (ax_incline,ax_energy):
        ax.tick_params(labelsize=9)
        ax.xaxis.label.set_size(10);ax.yaxis.label.set_size(10)
    fig.supxlabel('Small marker offsets help distinguish objects. Speeds stay at their arrival values after reaching the bottom.',fontsize=8.5)

    def update(frame):
        t=times[frame]
        time_title.set_text(f'Positions at t = {t:.2f} s')
        for i,(name,props) in enumerate(objects.items()):
            a=results[name]['a'];c=props['c']
            s=min(0.5*a*t**2,L_incline)
            frac=s/L_incline
            xpos=frac*x_end+offsets[i]*np.cos(angle+np.pi/2)
            ypos=h*(1-frac)+offsets[i]*np.sin(angle+np.pi/2)+0.08
            obj_plots[name].set_data([xpos],[ypos])
            v=min(a*t,np.sqrt(2*a*L_incline))
            table[(i+1,1)].get_text().set_text(f'{v:.2f}')
            bars_trans[i].set_height(0.5*m*v**2)
            bars_rot[i].set_height(0.5*m*v**2*c)
        return []

    anim=FuncAnimation(fig,update,frames=physics_frames(n_frames),
                        interval=physics_interval(n_frames,1000/fps),blit=True)
    plt.close(fig)
    return physics_animation_html(anim)

rolling_race_animation(angle_deg=30,h=2.0)
```

## 5. Pulley-Rope Systems

When a pulley has mass (and therefore moment of inertia), we must include its rotational dynamics.

### Atwood Machine with a Massive Pulley

Two masses $m_1$ and $m_2$ connected by a rope over a pulley of mass $M$ and radius $R$:

For a nonzero pulley inertia and nonzero angular acceleration, the tensions on the two sides are **different**. Choose downward on the $m_1$ side and upward on the $m_2$ side as positive:

$$m_1 g - T_1 = m_1 a$$
$$T_2 - m_2 g = m_2 a$$
$$T_1 R - T_2 R = I\alpha = I \frac{a}{R}$$

Solving for the acceleration:

$$a = \frac{(m_1 - m_2)g}{m_1 + m_2 + I/R^2}$$

For a solid disk pulley ($I = \frac{1}{2}MR^2$):

$$a = \frac{(m_1 - m_2)g}{m_1 + m_2 + M/2}$$

> **Key insight:** The pulley's inertia acts like extra mass in the denominator, slowing the system down compared to a massless-pulley Atwood machine.

**Eliminate tensions slowly.** Add the two mass equations and use the pulley equation divided by $R$:

$$\begin{aligned}(m_1-m_2)g-(T_1-T_2)&=(m_1+m_2)a,\\T_1-T_2&=\frac{I}{R^2}a,\\(m_1-m_2)g&=\left(m_1+m_2+\frac{I}{R^2}\right)a.\end{aligned}$$

For $m_1=3.0\,\mathrm{kg}$, $m_2=1.0\,\mathrm{kg}$ and a $2.0\,\mathrm{kg}$ disk pulley,
$$a=\frac{(3-1)(9.81)}{3+1+2/2}=3.924\,\mathrm{m/s^2}.$$

Then $T_1=3(9.81-3.924)=17.658\,\mathrm N$ and $T_2=1(9.81+3.924)=13.734\,\mathrm N$. **TR:** Gerilme farkı makaraya açısal ivme kazandırır. Eşit kütle sınırında $a=0$ ve gerilmeler eşit olabilir; makaraya kütle vermek tek başına her durumda fark yaratmaz.

## 🎮 Interactive Demo 4: Pulley System Simulator

Adjust the masses and pulley parameters to see how the system accelerates.

```python
#@title Run the animation — observe the physical motion
# ============================================================
# DEMO 4: Atwood Machine with Massive Pulley
# ============================================================

def pulley_simulator(m1=5.0, m2=3.0, M_pulley=2.0, R_pulley=0.15):
    """
    Simulate and animate an Atwood machine with a massive pulley.
    m1 > m2 assumed. Pulley is a solid disk.
    """
    g = 9.81
    I_pulley = 0.5 * M_pulley * R_pulley**2

    # Acceleration
    a = (m1 - m2) * g / (m1 + m2 + I_pulley / R_pulley**2)
    a_massless = (m1 - m2) * g / (m1 + m2)  # comparison

    # Tensions
    T1 = m1 * (g - a)
    T2 = m2 * (g + a)

    # Animation
    duration = 3.0
    fps = 30
    n_frames = int(duration * fps)
    times = np.linspace(0, duration, n_frames)

    fig, (ax_sys, ax_info) = plt.subplots(1, 2, figsize=(13, 7),
                                           gridspec_kw={'width_ratios': [1, 1.2]})
    fig.suptitle('Atwood Machine with Massive Pulley', fontsize=14, fontweight='bold')

    # System view
    ax_sys.set_xlim(-2, 2)
    ax_sys.set_ylim(-4, 2)
    ax_sys.set_aspect('equal')
    ax_sys.set_title('System')

    # Pulley
    pulley_center = (0, 1)
    pulley_r = 0.4
    pulley_circle = plt.Circle(pulley_center, pulley_r, fill=False, lw=3, color='gray')
    ax_sys.add_patch(pulley_circle)
    ax_sys.plot([0], [1], '+', color='black', markersize=10, markeredgewidth=2)
    # Support
    ax_sys.plot([-0.5, 0.5], [1.5, 1.5], 'k-', lw=4)
    ax_sys.plot([0, 0], [1.4, 1.5], 'k-', lw=2)
    # Spoke for rotation visual
    spoke, = ax_sys.plot([], [], '-', color='gray', lw=2)

    # Ropes and masses
    rope_left, = ax_sys.plot([], [], 'k-', lw=1.5)
    rope_right, = ax_sys.plot([], [], 'k-', lw=1.5)

    mass1_box = mpatches.FancyBboxPatch((-1.3, 0), 0.6, 0.6, boxstyle="round,pad=0.05",
                                         facecolor='#e74c3c', edgecolor='black', lw=2)
    mass2_box = mpatches.FancyBboxPatch((0.7, 0), 0.6, 0.6, boxstyle="round,pad=0.05",
                                         facecolor='#3498db', edgecolor='black', lw=2)
    ax_sys.add_patch(mass1_box)
    ax_sys.add_patch(mass2_box)
    label1 = ax_sys.text(-1, 0, '', ha='center', va='center', fontsize=10,
                         color='white', fontweight='bold')
    label2 = ax_sys.text(1, 0, '', ha='center', va='center', fontsize=10,
                         color='white', fontweight='bold')

    time_text = ax_sys.text(0.02, 0.02, '', transform=ax_sys.transAxes, fontsize=10)

    # Info panel
    ax_info.axis('off')
    info_lines = [
        f'Parameters:',
        f'  m$_1$ = {m1:.1f} kg (heavier, left)',
        f'  m$_2$ = {m2:.1f} kg (lighter, right)',
        f'  M$_{{pulley}}$ = {M_pulley:.1f} kg (solid disk)',
        f'  R$_{{pulley}}$ = {R_pulley:.2f} m',
        f'  I$_{{pulley}}$ = {I_pulley:.4f} kg m$^2$',
        f'',
        f'Results:',
        f'  a = {a:.3f} m/s$^2$',
        f'  a (massless pulley) = {a_massless:.3f} m/s$^2$',
        f'  Reduction = {(1 - a/a_massless)*100:.1f}%',
        f'',
        f'  T$_1$ = {T1:.2f} N',
        f'  T$_2$ = {T2:.2f} N',
        f'  T$_1$ - T$_2$ = {T1-T2:.2f} N (net torque / R)',
        f'',
        f'Equations:',
        f'  a = (m$_1$ - m$_2$)g / (m$_1$ + m$_2$ + I/R$^2$)',
        f'  With massless pulley: a = (m$_1$ - m$_2$)g / (m$_1$ + m$_2$)',
    ]
    ax_info.text(0.05, 0.95, '\n'.join(info_lines), transform=ax_info.transAxes,
                 fontsize=11, verticalalignment='top', fontfamily='monospace',
                 bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

    plt.tight_layout()

    def update(frame):
        t = times[frame]
        s = 0.5 * a * t**2  # displacement
        theta = s / R_pulley  # pulley angle

        # m1 goes down, m2 goes up
        y1_center = 0 - s
        y2_center = -0.5 + s

        mass1_box.set_y(y1_center - 0.3)
        mass2_box.set_y(y2_center - 0.3)
        label1.set_position((-1, y1_center))
        label1.set_text(f'm$_1$\n{m1:.0f}kg')
        label2.set_position((1, y2_center))
        label2.set_text(f'm$_2$\n{m2:.0f}kg')

        # Ropes
        rope_left.set_data([-pulley_r + pulley_center[0], -pulley_r + pulley_center[0]],
                           [pulley_center[1], y1_center + 0.3])
        rope_right.set_data([pulley_r + pulley_center[0], pulley_r + pulley_center[0]],
                            [pulley_center[1], y2_center + 0.3])

        # Pulley spoke
        sx = pulley_center[0] + pulley_r * 0.9 * np.cos(theta)
        sy = pulley_center[1] + pulley_r * 0.9 * np.sin(theta)
        spoke.set_data([pulley_center[0], sx], [pulley_center[1], sy])

        v = a * t
        time_text.set_text(f't = {t:.2f} s\nv = {v:.2f} m/s\ns = {s:.3f} m')

        return []

    anim = FuncAnimation(fig, update, frames=physics_frames(n_frames), interval=physics_interval(n_frames, 1000/fps), blit=True)
    plt.close(fig)
    return physics_animation_html(anim)

pulley_simulator(m1=5.0, m2=3.0, M_pulley=2.0, R_pulley=0.15)
```

**Problem:** A solid cylinder (mass $4\,\mathrm{kg}$, radius $0.1\,\mathrm m$) and a hollow cylinder (same mass and radius) roll without slipping down a 2-meter high ramp ($30^\circ$). Find the speed of each at the bottom and the time each takes.

$$Mgh=\tfrac12Mv^2+\tfrac12\beta MR^2(v/R)^2
\Rightarrow v=\sqrt{\frac{2gh}{1+\beta}}.$$
For $h=2\,\mathrm m$, the solid cylinder has $v=\sqrt{\frac{39.24}{1.5}}=5.11\,\mathrm{m/s}$; the hollow cylinder has $v=\sqrt{\frac{39.24}{2}}=4.43\,\mathrm{m/s}$.

From $s=\tfrac12at^2$, multiply by 2 and divide by $a$: $t=\sqrt{2s/a}$. Thus **$t_{\rm solid}=1.564\,\mathrm s$** and **$t_{\rm hollow}=1.806\,\mathrm s$**. A thicker hollow cylinder would need its inner radius to determine $I$.

### Convert revolutions before calculating angular braking

**Problem.** A grinding wheel of mass $1.15\,\mathrm{kg}$ and diameter $22.0\,\mathrm{cm}$ spins counterclockwise at $20.0\,\mathrm{rev/s}$. When switched off, it slows uniformly to rest in $80.0\,\mathrm{s}$. Find its signed angular acceleration and the number of revolutions completed while stopping. Take counterclockwise as positive.

**Reasoning and algebra.** A full revolution is $2\pi$ radians. Convert the initial angular speed before using the radian-based acceleration equation:

$$
\omega_i=(20.0\,\mathrm{rev/s})
\frac{2\pi\,\mathrm{rad}}{1\,\mathrm{rev}}
=40\pi\,\mathrm{rad/s}.
$$

Because the acceleration is constant, the change in angular speed is divided by the elapsed time:

$$
\begin{aligned}
\alpha&=\frac{\omega_f-\omega_i}{t}
=\frac{0-40\pi\,\mathrm{rad/s}}{80.0\,\mathrm{s}}
=-\frac{\pi}{2}\,\mathrm{rad/s^2}\\
&=\boxed{-1.57\,\mathrm{rad/s^2}}.
\end{aligned}
$$

For the angle, use average angular speed. Uniform slowing makes that average the arithmetic mean of the endpoint speeds:

$$
\begin{aligned}
\Delta\theta&=\frac{\omega_i+\omega_f}{2}t
=\frac{40\pi+0}{2}(80.0)=1600\pi\,\mathrm{rad},\\
N&=\frac{\Delta\theta}{2\pi\,\mathrm{rad/rev}}
=\boxed{800\text{ revolutions}}.
\end{aligned}
$$

**Check.** The average rotation rate is $10.0\,\mathrm{rev/s}$, so $10.0\times80.0=800$. The acceleration is clockwise, opposite the still-counterclockwise rotation. The wheel's mass and diameter are unnecessary for this kinematics calculation; torque would require additional inertia information.

**Türkçe açıklama.** $20.0$ değerini doğrudan radyan/saniye sanmıyoruz; önce $2\pi$ ile çarpıyoruz. Negatif açısal ivme, bu örnekte dönmenin yavaşladığını gösterir. Tekerlek duruncaya kadar dönme yönü değişmez. Tur sayısını bulmak için toplam radyanı yeniden $2\pi$'ye böleriz.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 9, §9.2 Problems, question 4 (`09_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### Use rim speed to find angular acceleration

**Problem.** A $3.45\,\mathrm{kg}$ centrifuge starts from rest and accelerates uniformly for $100\,\mathrm{s}$. At full speed, a point $8.00\,\mathrm{cm}$ from its axis moves at $150\,\mathrm{m/s}$. Find the angular acceleration and the number of revolutions completed during the acceleration.

**Reasoning and algebra.** The given $150\,\mathrm{m/s}$ is a linear speed of one point, not the rotation rate. Convert the radius to metres and apply $v=r\omega$:

$$
\begin{aligned}
r&=8.00\,\mathrm{cm}=0.0800\,\mathrm{m},\\
\omega_f&=\frac{v_f}{r}
=\frac{150\,\mathrm{m/s}}{0.0800\,\mathrm{m}}
=1875\,\mathrm{rad/s}.
\end{aligned}
$$

Since $\omega_i=0$, constant angular acceleration gives

$$
\alpha=\frac{\omega_f-\omega_i}{t}
=\frac{1875\,\mathrm{rad/s}}{100\,\mathrm{s}}
=\boxed{18.75\,\mathrm{rad/s^2}}.
$$

The angular displacement follows from the average angular speed:

$$
\begin{aligned}
\Delta\theta&=\tfrac12(\omega_i+\omega_f)t
=\tfrac12(1875)(100)=93750\,\mathrm{rad},\\
N&=\frac{93750}{2\pi}=14920.8\,\mathrm{rev}
\approx\boxed{1.49\times10^4\text{ revolutions}}.
\end{aligned}
$$

**Interpretation and check.** The final rate is about $298.4\,\mathrm{rev/s}$, while the average rate during the uniform increase is half that. Multiplying this average by $100\,\mathrm{s}$ reproduces the turn count. The large result is consistent with the unusually high stated rim speed; no extra factor of $100$ should survive the centimetre-to-metre conversion. The mass does not enter these kinematic relations.

**Türkçe açıklama.** Önce doğrusal hızdan açısal hıza geçiyoruz: aynı $v$ için yarıçap küçüldükçe $\omega$ büyür. $8\,\mathrm{cm}$ yerine $8\,\mathrm{m}$ yazmak sonucu yüz kat değiştirir. Başlangıçta durgun olduğu için ortalama açısal hız, son açısal hızın yarısıdır; toplam turu son hızla çarparak bulamayız.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 9, §9.2 Problems, question 5 (`09_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### The second revolution takes less time

**Problem.** A $1.25\,\mathrm{kg}$ ball begins rolling from rest with constant angular acceleration. Its first complete revolution takes $3.60\,\mathrm{s}$. Find the time required for its next complete revolution. Measure all elapsed times from the instant the ball is released.

**Reasoning and algebra.** Let $t_1$ be the time at which one revolution is complete and $t_2$ the time at which two revolutions are complete. The requested interval is $t_2-t_1$, not $t_2$. Starting from rest gives

$$
\theta=\tfrac12\alpha t^2.
$$

At the first milestone, $\theta_1=2\pi\,\mathrm{rad}$. Rearrange to identify the constant acceleration:

$$
\alpha=\frac{2\theta_1}{t_1^2}
=\frac{4\pi}{(3.60)^2}\,\mathrm{rad/s^2}
=0.96963\,\mathrm{rad/s^2}.
$$

After two revolutions the total angle is $4\pi\,\mathrm{rad}$. Dividing the two displacement equations avoids unnecessary intermediate rounding:

$$
\frac{\theta_2}{\theta_1}
=\frac{t_2^2}{t_1^2}=2
\quad\Longrightarrow\quad t_2=\sqrt{2}\,t_1.
$$

Therefore

$$
\begin{aligned}
\Delta t_{\text{second}}&=t_2-t_1
=(\sqrt{2}-1)(3.60\,\mathrm{s})\\
&=\boxed{1.49\,\mathrm{s}}.
\end{aligned}
$$

**Interpretation and check.** Two revolutions have finished at $t_2=5.091\,\mathrm{s}$, so the second occupies the interval from $3.600$ to $5.091\,\mathrm{s}$. It is shorter because the ball is already rotating when that interval begins and continues to speed up. The two intervals cover equal angles, but they do not have equal average angular speeds.

**Türkçe açıklama.** Zamanın sıfır noktasını ikinci tur başında değiştirmedik. $t_2$, toplam iki turun süresidir; istenen ikinci turun süresi için $t_1$'i çıkardık. Sabit ivmede yolun zamanın karesiyle artması, eşit yolların giderek daha kısa sürede alınmasını açıklar.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 9, §9.2 Problems, question 11 (`09_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### Moving a rotation axis changes the moment of inertia

**Problem.** A slender, uniform $100.00\,\mathrm{cm}$ rod serves as a metre stick. Compare two parallel axes perpendicular to the rod: one passes through the $50\,\mathrm{cm}$ mark and the other through the $30\,\mathrm{cm}$ mark. Find $I_{30}/I_{50}$, the ratio of their moments of inertia.

**Reasoning and algebra.** Picture the rod lying horizontally, with both rotation axes pointing perpendicular to the page. The first axis passes through the centre of mass. The distance separating the axes is

$$
d=50\,\mathrm{cm}-30\,\mathrm{cm}=20\,\mathrm{cm}=0.200\,\mathrm{m}.
$$

For a uniform slender rod, its central moment of inertia is $I_{50}=ML^2/12$. The parallel-axis theorem supplies the shifted value:

$$
I_{30}=I_{50}+Md^2.
$$

Form the requested ratio before inserting a mass. The common mass cancels:

$$
\begin{aligned}
\frac{I_{30}}{I_{50}}
&=\frac{ML^2/12+Md^2}{ML^2/12}\\
&=1+\frac{Md^2}{ML^2/12}
=1+\frac{12d^2}{L^2}\\
&=1+\frac{12(0.200\,\mathrm{m})^2}{(1.0000\,\mathrm{m})^2}
=\boxed{1.48}.
\end{aligned}
$$

**Interpretation and check.** The shifted-axis inertia is $48\%$ larger, approximately $1.5$ times the central value. This must exceed one because $Md^2$ is positive. If the second axis moved back to the centre, $d$ would become zero and the ratio would become one. No numerical mass is needed for a ratio involving the same rod.

**Türkçe açıklama.** $d$, eksenin cetvel üzerindeki $30\,\mathrm{cm}$ konumu değildir; iki eksen arasındaki $20\,\mathrm{cm}$ uzaklıktır. Pay ve paydada bulunan kütleyi sadeleştiriyoruz. Kare alma işlemi hem uzaklığa hem birime uygulanır; bu nedenle son oran birimsizdir.

### Point masses connect inertia to rotational energy

**Problem.** A very light $1.00\,\mathrm{m}$ rod carries a $2.00\,\mathrm{kg}$ mass at one end and a $3.00\,\mathrm{kg}$ mass at the other. A fixed perpendicular axis lies $30.0\,\mathrm{cm}$ from the $3.00\,\mathrm{kg}$ end. The rotating system has kinetic energy $100.0\,\mathrm{J}$. Find its moment of inertia and its rotation rate in revolutions per second.

**Reasoning and algebra.** Draw the pivot between the masses. Distances in $I=\sum mr^2$ are measured from that pivot, not from an arbitrary end of the rod:

$$
r_3=0.300\,\mathrm{m},\qquad
r_2=1.00-0.300=0.700\,\mathrm{m}.
$$

Treat the attached objects as point masses and neglect the rod's own inertia:

$$
\begin{aligned}
I&=m_2r_2^2+m_3r_3^2\\
&=(2.00)(0.700)^2+(3.00)(0.300)^2\\
&=0.980+0.270=\boxed{1.25\,\mathrm{kg\,m^2}}.
\end{aligned}
$$

This provides a bridge to rotational energy. Starting with $K=\tfrac12 I\omega^2$, isolate the unknown before substituting:

$$
\begin{aligned}
\omega^2&=\frac{2K}{I},\\
\omega&=\sqrt{\frac{2(100.0\,\mathrm{J})}{1.25\,\mathrm{kg\,m^2}}}
=12.649\,\mathrm{rad/s},\\
f&=\frac{\omega}{2\pi}=\boxed{2.01\,\mathrm{rev/s}}.
\end{aligned}
$$

**Check.** Substituting back gives $\tfrac12(1.25)(12.649)^2\approx100.0\,\mathrm{J}$. The lighter mass contributes more inertia because it lies much farther from the pivot and the distance is squared. The energy determines an angular-speed magnitude; it does not tell us clockwise or counterclockwise.

**Türkçe açıklama.** Büyük kütle her zaman daha büyük atalet katkısı yapmaz; eksene uzaklık karesi önemlidir. Önce iki $mr^2$ terimini ayrı hesapladık. Enerji denkleminde $\omega^2$'yi yalnız bıraktıktan sonra karekök aldık. Radyan/saniyeyi tur/saniyeye çevirmek için sonucu $2\pi$'ye böldük.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 9, §9.2 Problems, question 16 (`09_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### Only the perpendicular force component turns the nut

**Problem.** A $1.10\,\mathrm{kg}$ wrench extends directly east from a nut. A $150.0\,\mathrm{N}$ force is applied $15.0\,\mathrm{cm}$ east of the nut, directed $30.0^\circ$ north of east. Find the torque produced by this applied force about the centre of the nut, including its direction when viewed from above.

**Reasoning and algebra.** Draw the position vector $\mathbf r$ eastward from the nut to the force application point. The angle between $\mathbf r$ and $\mathbf F$ is $30.0^\circ$. Only the northward force component is perpendicular to the wrench:

$$
\begin{aligned}
r&=15.0\,\mathrm{cm}=0.150\,\mathrm{m},\\
F_\perp&=F\sin\theta=(150.0\,\mathrm{N})\sin30.0^\circ
=75.0\,\mathrm{N},\\
\tau&=rF_\perp
=(0.150\,\mathrm{m})(75.0\,\mathrm{N})
=\boxed{11.25\,\mathrm{N\,m}}.
\end{aligned}
$$

Rounded to three significant figures, the torque is $11.3\,\mathrm{N\,m}$. The northward force acting at the eastward end turns the wrench counterclockwise in the horizontal plane; the torque vector points upward by the right-hand rule.

**Check.** The full product $rF=22.5\,\mathrm{N\,m}$ would require a force at $90^\circ$ to the wrench. Here $\sin30^\circ=1/2$, so the actual torque is half that maximum. The eastward component pushes along the wrench and has zero moment arm about the nut. The wrench's mass is unnecessary when calculating the torque of the specified applied force.

**Türkçe açıklama.** Kuvvetin tamamını doğrudan kol uzunluğuyla çarpmıyoruz. Dönmeye katkı yapan, anahtara dik olan $75.0\,\mathrm{N}$'luk bileşendir. $15\,\mathrm{cm}$'yi $0.150\,\mathrm{m}$'ye çevirerek tork birimini $\mathrm{N\,m}$ elde ederiz. Açıyı yanlışlıkla kuzey ekseninden ölçmek sinüs–kosinüs karışıklığına yol açar.

### Infer a braking torque from the stopping angle

**Problem.** A uniform solid sphere has mass $1.85\,\mathrm{kg}$ and diameter $45.0\,\mathrm{cm}$. It spins about its centre at $2.40\,\mathrm{rev/s}$, then slows uniformly to rest after $18.2$ revolutions. Find the net braking torque. Take its initial direction of rotation as positive.

**Reasoning and algebra.** First obtain the radius and the correct solid-sphere inertia. Diameter is twice the radius:

$$
\begin{aligned}
R&=0.450/2=0.225\,\mathrm{m},\\
I&=\tfrac25MR^2
=\tfrac25(1.85)(0.225)^2
=0.0374625\,\mathrm{kg\,m^2}.
\end{aligned}
$$

The stopping time is not given, so choose an angular kinematics equation that eliminates time:

$$
\omega_f^2=\omega_i^2+2\alpha\Delta\theta.
$$

Convert both rotational quantities consistently, then rearrange:

$$
\begin{aligned}
\omega_i&=2.40(2\pi)\,\mathrm{rad/s},\qquad
\Delta\theta=18.2(2\pi)\,\mathrm{rad},\\
\alpha&=\frac{0-\omega_i^2}{2\Delta\theta}
=-0.99426\,\mathrm{rad/s^2},\\
\tau_{\mathrm{net}}&=I\alpha
=(0.0374625)(-0.99426)
=\boxed{-0.03725\,\mathrm{N\,m}}.
\end{aligned}
$$

**Interpretation and check.** The magnitude is approximately $0.0372\,\mathrm{N\,m}$; the negative sign indicates a torque opposing the initial rotation. As an independent check, the constant torque does work $\tau\Delta\theta=-\tfrac12 I\omega_i^2$, removing all initial rotational kinetic energy. Keeping radians in both equations makes this energy comparison consistent.

**Türkçe açıklama.** Bu problemde zaman aramaya gerek yok; $\omega^2$ bağıntısı doğrudan verilen açıyla çalışır. Atalet momentinde çapı yarıçap yerine kullanmak sonucu dört kat büyütür. İvme ve torkun eksi çıkması, seçtiğimiz pozitif dönme yönüne karşı etki ettiklerini gösterir; torkun büyüklüğünü söylerken pozitif değer kullanırız.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 10, §10.2 Problems, question 11 (`10_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### Equal opposite forces can still produce rotation

**Problem.** A uniform thin beam of length $2.00\,\mathrm{m}$ and mass $5.00\,\mathrm{kg}$ lies horizontally east–west. A $200\,\mathrm{N}$ force pushes downward at its east end; another $200\,\mathrm{N}$ force pushes upward at its west end. Find its angular acceleration about its centre of mass.

**Reasoning and algebra.** Place the origin at the beam's centre. Each force acts at a distance $L/2=1.00\,\mathrm{m}$. Draw both arrows: they tend to lower the east end and raise the west end. Their force vectors cancel, but their torques have the same rotational sense:

$$
\begin{aligned}
\tau_{\mathrm{net}}&=F\frac{L}{2}+F\frac{L}{2}=FL,\\
&=(200\,\mathrm{N})(2.00\,\mathrm{m})=400\,\mathrm{N\,m}.
\end{aligned}
$$

This pair is called a force couple. For rotation about the perpendicular central axis of a uniform slender beam,

$$
\begin{aligned}
I_{\mathrm{cm}}&=\frac{ML^2}{12}
=\frac{(5.00)(2.00)^2}{12}
=1.6667\,\mathrm{kg\,m^2},\\
\alpha&=\frac{\tau_{\mathrm{net}}}{I_{\mathrm{cm}}}
=\frac{400}{1.6667}\,\mathrm{rad/s^2}
=\boxed{240\,\mathrm{rad/s^2}}.
\end{aligned}
$$

**Direction and check.** With east as $+x$, north as $+y$, and upward as $+z$, both $\mathbf r\times\mathbf F$ terms point north. Thus the angular-acceleration vector points north. The beam's weight acts through its centre of mass, so it contributes no torque about that point. Cancellation of the two applied forces alone does not cancel their rotational effect.

**Türkçe açıklama.** Kuvvetlerin eşit ve zıt olması, torklarının da eşit ve zıt olduğu anlamına gelmez. Her iki kuvvet aynı dönme eğilimini oluşturduğu için torkları topladık. Merkezden her uca uzaklık $1.00\,\mathrm{m}$'dir. Son adımda $\tau=I\alpha$ denkleminde iki tarafı $I$'ya bölerek açısal ivmeyi yalnız bıraktık.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 10, §10.2 Problems, question 13 (`10_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### A rolling shell accelerates more slowly than a sliding point

**Problem.** A thin cylindrical shell is released from rest on a ramp inclined $30^\circ$ above horizontal. It rolls without slipping. Find how long it takes to travel the first $3.1\,\mathrm{m}$ along the ramp. Use $g=9.81\,\mathrm{m/s^2}$.

**Reasoning and algebra.** Choose downhill as positive. Draw weight, normal force and static friction; the friction force points uphill and supplies the torque needed to spin up the shell. Let $f$ be its positive magnitude. Translation and rotation give

$$
mg\sin\theta-f=ma,\qquad fR=I\alpha.
$$

For a thin shell, $I=mR^2$. Rolling without slipping adds $a=\alpha R$. Substitute this constraint into the torque equation:

$$
fR=(mR^2)\frac{a}{R}
\quad\Longrightarrow\quad f=ma.
$$

Return to the translational equation and isolate the acceleration:

$$
mg\sin\theta-ma=ma
\quad\Longrightarrow\quad
 a=\frac{g\sin\theta}{2}
=\frac{9.81\sin30^\circ}{2}
=2.4525\,\mathrm{m/s^2}.
$$

Starting from rest with constant acceleration, $s=\tfrac12at^2$. Therefore

$$
t=\sqrt{\frac{2s}{a}}
=\sqrt{\frac{2(3.1\,\mathrm{m})}{2.4525\,\mathrm{m/s^2}}}
=\boxed{1.59\,\mathrm{s}\approx1.6\,\mathrm{s}}.
$$

**Check.** The downhill acceleration is half of $g\sin30^\circ$, because some gravitational energy becomes rotation. Mass and radius cancel for this specified shape. Static friction is compatible with mechanical-energy conservation during ideal rolling; there is no sliding at the contact point.

**Türkçe açıklama.** Önce öteleme ve dönme denklemlerini birlikte çözüyoruz. Kaymadan yuvarlanma koşulu, $a$ ile $\alpha$'yı bağlar. Sürtünmeyi yok saymak dönmeyi başlatacak torku da yok saymak olur. Zamanı bulurken karekök almayı unutmuyoruz; $2s/a$'nın birimi saniye karedir.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 10, §10.2 Problems, question 18 (`10_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### One rolling sphere links energy distance and rotation

**Problem.** A uniform solid sphere of mass $1.5\,\mathrm{kg}$ and diameter $30.0\,\mathrm{cm}$ rolls from rest without slipping down a $35^\circ$ incline of length $7.0\,\mathrm{m}$. Find its final linear speed, angular speed and total rotation angle. Explain which results depend on mass or radius. Use $g=9.81\,\mathrm{m/s^2}$.

**Reasoning and algebra.** The vertical drop is $h=s\sin35^\circ$, while the rolling distance is $s=7.0\,\mathrm{m}$. The radius is $R=0.150\,\mathrm{m}$. Gravitational energy becomes both translational and rotational kinetic energy:

$$
\begin{aligned}
mgh&=\tfrac12mv^2+\tfrac12I\omega^2,\\
I&=\tfrac25mR^2,\qquad \omega=v/R,\\
mgh&=\tfrac12mv^2+\tfrac15mv^2=\tfrac7{10}mv^2,\\
v&=\sqrt{\frac{10gh}{7}}
=\sqrt{\frac{10(9.81)(7.0\sin35^\circ)}{7}}
=\boxed{7.50\,\mathrm{m/s}}.
\end{aligned}
$$

The two no-slip relations now supply the remaining quantities:

$$
\omega=\frac{v}{R}=\frac{7.5012}{0.150}
=\boxed{50.0\,\mathrm{rad/s}},\qquad
\Delta\theta=\frac{s}{R}=\frac{7.0}{0.150}
=\boxed{46.7\,\mathrm{rad}}.
$$

**Interpretation and check.** For this fixed shape and height, $v$ depends on neither mass nor radius: both cancel in the energy equation. Angular speed depends inversely on radius, and not on mass. A larger sphere can have the same centre-of-mass speed while turning more slowly. The check $R\Delta\theta=7.0\,\mathrm{m}$ recovers the distance along the ramp, not its vertical height.

**Türkçe açıklama.** Çapı önce ikiye bölerek yarıçapı bulduk. Enerji hesabında düşey yükseklik $s\sin35^\circ$, yuvarlanma açısında ise eğim boyunca alınan $s$ kullanılır. $\omega=v/R$ yerine konduğunda dönme enerjisindeki $R^2$ sadeleşir; doğrusal hızın yarıçaptan bağımsız olması buradan gelir.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 10, §10.2 Problems, question 22 (`10_testbank.doc`). Original givens retained; wording and worked explanation expanded.

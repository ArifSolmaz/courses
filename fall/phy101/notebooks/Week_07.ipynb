## 1. Linear Momentum

### Definition

**Momentum** is the product of an object's mass and velocity:

$$\vec{p} = m\vec{v}$$

| Quantity | Symbol | SI Unit | Nature |
|----------|--------|---------|--------|
| Momentum | $\vec{p}$ | kg$\cdot$m/s | Vector |
| Mass | $m$ | kg | Scalar |
| Velocity | $\vec{v}$ | m/s | Vector |

### Analogy: The Freight Train vs. The Sports Car

Think of momentum as "how hard it is to stop something." A slow freight train (huge mass, small velocity) can have the same momentum as a fast sports car (small mass, huge velocity). Both are equally hard to stop!

### Newton's Second Law in Momentum Form

$$\vec{F}_{\text{net}} = \frac{d\vec{p}}{dt}$$

When mass is constant: $\vec{F}_{\text{net}} = m\vec{a}$ (the familiar form).

## 2. Impulse-Momentum Theorem

**Impulse** is the change in momentum produced by a force acting over a time interval:

$$\vec{J} = \int_{t_1}^{t_2} \vec{F}\, dt = \Delta\vec{p} = m\vec{v}_f - m\vec{v}_i$$

For a constant force: $\vec{J} = \vec{F}_{\text{avg}} \cdot \Delta t$

### Engineering Insight

This is why car airbags and crumple zones work! By increasing $\Delta t$, the **average force** on the occupant decreases for the same momentum change. The area under the F-t curve (impulse) stays the same, but the peak force drops.

```python
# Setup: Run this cell first to import all required libraries
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
import ipywidgets as widgets
from ipywidgets import interact, interactive, FloatSlider, IntSlider, HBox, VBox, Output
import warnings
warnings.filterwarnings('ignore')

# Enable inline plots
%matplotlib inline
print("All libraries loaded successfully!")
```

### Interactive Demo 1: Impulse-Momentum Theorem Visualizer

Explore how the same impulse (area under the F-t curve) can result from different force profiles. Adjust the collision duration to see how peak force changes.

```python
def impulse_visualizer(delta_v=10.0, mass=1500.0, collision_time=0.1):
    """
    Visualize impulse-momentum theorem with F-t graphs.
    Shows how extending collision time reduces peak force.
    """
    impulse = mass * delta_v  # Total impulse (N*s)
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # --- Panel 1: Short collision (no protection) ---
    t_short = 0.05
    t1 = np.linspace(0, t_short, 200)
    # Half-sine pulse
    F_short = (np.pi * impulse / (2 * t_short)) * np.sin(np.pi * t1 / t_short)
    peak_short = np.max(F_short)
    
    axes[0].fill_between(t1 * 1000, F_short / 1000, alpha=0.3, color='red')
    axes[0].plot(t1 * 1000, F_short / 1000, 'r-', linewidth=2)
    axes[0].set_title(f'No Protection\n$\\Delta t$ = {t_short*1000:.0f} ms', fontsize=12)
    axes[0].set_xlabel('Time (ms)')
    axes[0].set_ylabel('Force (kN)')
    axes[0].annotate(f'Peak: {peak_short/1000:.0f} kN', 
                     xy=(t_short*500, peak_short/1000), fontsize=11, color='red',
                     ha='center', va='bottom')
    
    # --- Panel 2: User-adjustable collision ---
    t2 = np.linspace(0, collision_time, 200)
    F_user = (np.pi * impulse / (2 * collision_time)) * np.sin(np.pi * t2 / collision_time)
    peak_user = np.max(F_user)
    
    axes[1].fill_between(t2 * 1000, F_user / 1000, alpha=0.3, color='blue')
    axes[1].plot(t2 * 1000, F_user / 1000, 'b-', linewidth=2)
    axes[1].set_title(f'Your Design\n$\\Delta t$ = {collision_time*1000:.0f} ms', fontsize=12)
    axes[1].set_xlabel('Time (ms)')
    axes[1].set_ylabel('Force (kN)')
    axes[1].annotate(f'Peak: {peak_user/1000:.0f} kN', 
                     xy=(collision_time*500, peak_user/1000), fontsize=11, color='blue',
                     ha='center', va='bottom')
    
    # --- Panel 3: Long collision (crumple zone + airbag) ---
    t_long = 0.3
    t3 = np.linspace(0, t_long, 200)
    F_long = (np.pi * impulse / (2 * t_long)) * np.sin(np.pi * t3 / t_long)
    peak_long = np.max(F_long)
    
    axes[2].fill_between(t3 * 1000, F_long / 1000, alpha=0.3, color='green')
    axes[2].plot(t3 * 1000, F_long / 1000, 'g-', linewidth=2)
    axes[2].set_title(f'Full Protection\n$\\Delta t$ = {t_long*1000:.0f} ms', fontsize=12)
    axes[2].set_xlabel('Time (ms)')
    axes[2].set_ylabel('Force (kN)')
    axes[2].annotate(f'Peak: {peak_long/1000:.0f} kN', 
                     xy=(t_long*500, peak_long/1000), fontsize=11, color='green',
                     ha='center', va='bottom')
    
    # Set same y-limits for comparison
    y_max = peak_short / 1000 * 1.2
    for ax in axes:
        ax.set_ylim(0, y_max)
        ax.grid(True, alpha=0.3)
    
    fig.suptitle(f'Impulse = {impulse/1000:.0f} kN$\\cdot$ms  (m={mass:.0f} kg, $\\Delta v$={delta_v:.0f} m/s)',
                 fontsize=13, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()
    
    reduction = (1 - peak_user / peak_short) * 100
    print(f"Peak force reduction vs no protection: {reduction:.1f}%")
    print(f"Same impulse (shaded area) = {impulse:.0f} N·s in all three cases!")

interact(impulse_visualizer,
         delta_v=FloatSlider(value=10, min=2, max=30, step=1, description='$\\Delta v$ (m/s)'),
         mass=FloatSlider(value=1500, min=500, max=3000, step=100, description='Mass (kg)'),
         collision_time=FloatSlider(value=0.1, min=0.02, max=0.5, step=0.01, description='$\\Delta t$ (s)'));
```

## 3. Conservation of Momentum

For an **isolated system** (no external forces), total momentum is conserved:

$$\vec{p}_{\text{total, before}} = \vec{p}_{\text{total, after}}$$

For two objects in 1D:

$$m_1 v_{1i} + m_2 v_{2i} = m_1 v_{1f} + m_2 v_{2f}$$

### Types of Collisions

| Type | Momentum Conserved? | KE Conserved? | Objects Stick? | Example |
|------|:---:|:---:|:---:|---|
| **Perfectly Elastic** | Yes | Yes | No | Billiard balls, Newton's cradle |
| **Inelastic** | Yes | No (KE lost) | No | Car crash with bounce |
| **Perfectly Inelastic** | Yes | No (max KE lost) | Yes | Bullet in block |
| **Explosive** | Yes | No (KE gained) | Separate | Explosion, spring release |

### Elastic Collision Formulas (1D)

$$v_{1f} = \frac{m_1 - m_2}{m_1 + m_2} v_{1i} + \frac{2m_2}{m_1 + m_2} v_{2i}$$

$$v_{2f} = \frac{2m_1}{m_1 + m_2} v_{1i} + \frac{m_2 - m_1}{m_1 + m_2} v_{2i}$$

### Perfectly Inelastic Collision Formula (1D)

$$v_f = \frac{m_1 v_{1i} + m_2 v_{2i}}{m_1 + m_2}$$

### Interactive Demo 2: Animated 1D Collision Simulator

Watch two objects collide! Adjust masses and initial velocities to explore different collision scenarios. The animation shows the collision in real time and displays momentum and energy data.

```python
def collision_1d_animation(m1=2.0, m2=1.0, v1i=3.0, v2i=-1.0, collision_type='Elastic'):
    """
    Animated 1D collision between two objects.
    """
    # Compute final velocities
    if collision_type == 'Elastic':
        v1f = ((m1 - m2) * v1i + 2 * m2 * v2i) / (m1 + m2)
        v2f = (2 * m1 * v1i + (m2 - m1) * v2i) / (m1 + m2)
    else:  # Perfectly Inelastic
        v1f = (m1 * v1i + m2 * v2i) / (m1 + m2)
        v2f = v1f
    
    # Energy calculations
    KE_before = 0.5 * m1 * v1i**2 + 0.5 * m2 * v2i**2
    KE_after = 0.5 * m1 * v1f**2 + 0.5 * m2 * v2f**2
    p_before = m1 * v1i + m2 * v2i
    p_after = m1 * v1f + m2 * v2f
    
    # Box sizes proportional to mass
    size1 = 0.3 + 0.15 * m1
    size2 = 0.3 + 0.15 * m2
    
    # Starting positions
    x1_start = -4.0
    x2_start = 4.0
    
    # Time to collision
    if abs(v1i - v2i) < 1e-9:
        t_collision = 100  # They never meet
    else:
        gap = (x2_start - size2/2) - (x1_start + size1/2)
        t_collision = gap / (v1i - v2i) if (v1i - v2i) > 0 else 100
    
    total_time = min(t_collision * 3, 8)
    n_frames = 150
    dt = total_time / n_frames
    
    fig, ax = plt.subplots(figsize=(12, 4))
    ax.set_xlim(-8, 8)
    ax.set_ylim(-1.5, 2.5)
    ax.set_aspect('equal')
    ax.set_xlabel('Position (m)', fontsize=12)
    ax.axhline(y=-0.05, color='gray', linewidth=2)  # ground line
    ax.set_yticks([])
    
    color1 = '#2196F3'
    color2 = '#FF5722'
    
    box1 = patches.FancyBboxPatch((x1_start - size1/2, 0), size1, size1,
                                   boxstyle="round,pad=0.05", facecolor=color1, edgecolor='black', linewidth=2)
    box2 = patches.FancyBboxPatch((x2_start - size2/2, 0), size2, size2,
                                   boxstyle="round,pad=0.05", facecolor=color2, edgecolor='black', linewidth=2)
    ax.add_patch(box1)
    ax.add_patch(box2)
    
    label1 = ax.text(x1_start, size1 + 0.2, f'm={m1:.1f} kg', ha='center', fontsize=10, fontweight='bold', color=color1)
    label2 = ax.text(x2_start, size2 + 0.2, f'm={m2:.1f} kg', ha='center', fontsize=10, fontweight='bold', color=color2)
    
    vel_text1 = ax.text(x1_start, -0.4, '', ha='center', fontsize=9, color=color1)
    vel_text2 = ax.text(x2_start, -0.4, '', ha='center', fontsize=9, color=color2)
    
    info_text = ax.text(0, 2.2, '', ha='center', fontsize=11, fontweight='bold',
                        bbox=dict(boxstyle='round,pad=0.3', facecolor='lightyellow', alpha=0.9))
    
    arrow_scale = 0.3  # scale factor for arrow length
    arrow1 = ax.annotate('', xy=(x1_start + v1i * arrow_scale, size1/2), xytext=(x1_start, size1/2),
                         arrowprops=dict(arrowstyle='->', color=color1, lw=2))
    arrow2 = ax.annotate('', xy=(x2_start + v2i * arrow_scale, size2/2), xytext=(x2_start, size2/2),
                         arrowprops=dict(arrowstyle='->', color=color2, lw=2))
    
    collided = [False]
    x1_pos = [x1_start]
    x2_pos = [x2_start]
    
    def init():
        return box1, box2, label1, label2, vel_text1, vel_text2, info_text
    
    def update(frame):
        t = frame * dt
        
        if not collided[0]:
            x1 = x1_start + v1i * t
            x2 = x2_start + v2i * t
            
            # Check collision
            if (x1 + size1/2) >= (x2 - size2/2) and (v1i - v2i) > 0:
                collided[0] = True
                x1_pos[0] = x1
                x2_pos[0] = x2
                v1_current = v1i
                v2_current = v2i
                info_text.set_text(f'COLLISION! p={p_before:.1f} kg m/s')
            else:
                v1_current = v1i
                v2_current = v2i
                info_text.set_text(f'Before: p = {p_before:.1f} kg$\\cdot$m/s, KE = {KE_before:.1f} J')
        
        if collided[0]:
            t_after = t - t_collision
            if t_after < 0:
                t_after = 0
            x1 = x1_pos[0] + v1f * t_after
            x2 = x2_pos[0] + v2f * t_after
            v1_current = v1f
            v2_current = v2f
            KE_loss = KE_before - KE_after
            info_text.set_text(f'After: p = {p_after:.1f} kg$\\cdot$m/s, KE = {KE_after:.1f} J (lost {KE_loss:.1f} J)')
        else:
            v1_current = v1i
            v2_current = v2i
        
        # Update positions
        box1.set_x(x1 - size1/2)
        box2.set_x(x2 - size2/2)
        label1.set_x(x1)
        label2.set_x(x2)
        vel_text1.set_x(x1)
        vel_text2.set_x(x2)
        vel_text1.set_text(f'v = {v1_current:.2f} m/s')
        vel_text2.set_text(f'v = {v2_current:.2f} m/s')
        
        # Update velocity arrows to follow the boxes
        arrow1.xy = (x1 + v1_current * arrow_scale, size1/2)
        arrow1.set_position((x1, size1/2))
        arrow2.xy = (x2 + v2_current * arrow_scale, size2/2)
        arrow2.set_position((x2, size2/2))
        
        return box1, box2, label1, label2, vel_text1, vel_text2, info_text
    
    anim = FuncAnimation(fig, update, init_func=init, frames=n_frames, interval=40, blit=False)
    plt.close(fig)
    return HTML(anim.to_jshtml())

display(HTML("<h4>Adjust parameters and run the cell to see the animation:</h4>"))

interact(collision_1d_animation,
         m1=FloatSlider(value=2.0, min=0.5, max=5.0, step=0.5, description='Mass 1 (kg)'),
         m2=FloatSlider(value=1.0, min=0.5, max=5.0, step=0.5, description='Mass 2 (kg)'),
         v1i=FloatSlider(value=3.0, min=-5.0, max=5.0, step=0.5, description='v1 initial'),
         v2i=FloatSlider(value=-1.0, min=-5.0, max=5.0, step=0.5, description='v2 initial'),
         collision_type=widgets.Dropdown(options=['Elastic', 'Perfectly Inelastic'], value='Elastic', description='Type'));
```

### Interactive Demo 3: Elastic vs Inelastic Collision Comparison

Compare what happens in elastic vs inelastic collisions side by side. Notice how momentum is conserved in both, but kinetic energy is only conserved in elastic collisions.

```python
def collision_comparison(m1=2.0, m2=1.0, v1i=4.0, v2i=0.0):
    """
    Side-by-side comparison of elastic vs perfectly inelastic collisions.
    Shows bar charts of momentum and energy before/after.
    """
    # Elastic
    v1f_e = ((m1 - m2) * v1i + 2 * m2 * v2i) / (m1 + m2)
    v2f_e = (2 * m1 * v1i + (m2 - m1) * v2i) / (m1 + m2)
    
    # Perfectly Inelastic
    vf_in = (m1 * v1i + m2 * v2i) / (m1 + m2)
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 8))
    
    # --- Momentum comparison ---
    categories = ['Object 1', 'Object 2', 'Total']
    
    p_before = [m1*v1i, m2*v2i, m1*v1i + m2*v2i]
    p_after_e = [m1*v1f_e, m2*v2f_e, m1*v1f_e + m2*v2f_e]
    p_after_in = [m1*vf_in, m2*vf_in, m1*vf_in + m2*vf_in]
    
    x = np.arange(len(categories))
    width = 0.35
    
    # Elastic momentum
    axes[0,0].bar(x - width/2, p_before, width, label='Before', color='#42A5F5', alpha=0.8)
    axes[0,0].bar(x + width/2, p_after_e, width, label='After', color='#1565C0', alpha=0.8)
    axes[0,0].set_title('Elastic: Momentum', fontsize=12, fontweight='bold')
    axes[0,0].set_xticks(x)
    axes[0,0].set_xticklabels(categories)
    axes[0,0].set_ylabel('Momentum (kg m/s)')
    axes[0,0].legend()
    axes[0,0].grid(axis='y', alpha=0.3)
    
    # Inelastic momentum
    axes[0,1].bar(x - width/2, p_before, width, label='Before', color='#EF5350', alpha=0.8)
    axes[0,1].bar(x + width/2, p_after_in, width, label='After', color='#B71C1C', alpha=0.8)
    axes[0,1].set_title('Inelastic: Momentum', fontsize=12, fontweight='bold')
    axes[0,1].set_xticks(x)
    axes[0,1].set_xticklabels(categories)
    axes[0,1].set_ylabel('Momentum (kg m/s)')
    axes[0,1].legend()
    axes[0,1].grid(axis='y', alpha=0.3)
    
    # --- Energy comparison ---
    KE_before = [0.5*m1*v1i**2, 0.5*m2*v2i**2, 0.5*m1*v1i**2 + 0.5*m2*v2i**2]
    KE_after_e = [0.5*m1*v1f_e**2, 0.5*m2*v2f_e**2, 0.5*m1*v1f_e**2 + 0.5*m2*v2f_e**2]
    KE_after_in = [0.5*m1*vf_in**2, 0.5*m2*vf_in**2, 0.5*m1*vf_in**2 + 0.5*m2*vf_in**2]
    
    # Elastic KE
    axes[1,0].bar(x - width/2, KE_before, width, label='Before', color='#66BB6A', alpha=0.8)
    axes[1,0].bar(x + width/2, KE_after_e, width, label='After', color='#2E7D32', alpha=0.8)
    axes[1,0].set_title('Elastic: Kinetic Energy (CONSERVED)', fontsize=12, fontweight='bold', color='green')
    axes[1,0].set_xticks(x)
    axes[1,0].set_xticklabels(categories)
    axes[1,0].set_ylabel('Kinetic Energy (J)')
    axes[1,0].legend()
    axes[1,0].grid(axis='y', alpha=0.3)
    
    # Inelastic KE
    axes[1,1].bar(x - width/2, KE_before, width, label='Before', color='#FFA726', alpha=0.8)
    axes[1,1].bar(x + width/2, KE_after_in, width, label='After', color='#E65100', alpha=0.8)
    KE_lost = KE_before[2] - KE_after_in[2]
    pct = (KE_lost / KE_before[2] * 100) if KE_before[2] > 0 else 0
    axes[1,1].set_title(f'Inelastic: Kinetic Energy (LOST {pct:.1f}%)', fontsize=12, fontweight='bold', color='red')
    axes[1,1].set_xticks(x)
    axes[1,1].set_xticklabels(categories)
    axes[1,1].set_ylabel('Kinetic Energy (J)')
    axes[1,1].legend()
    axes[1,1].grid(axis='y', alpha=0.3)
    
    plt.suptitle(f'm$_1$={m1} kg, m$_2$={m2} kg | v$_{{1i}}$={v1i} m/s, v$_{{2i}}$={v2i} m/s',
                 fontsize=13, y=1.02)
    plt.tight_layout()
    plt.show()
    
    print(f"Elastic:  v1f = {v1f_e:.2f} m/s, v2f = {v2f_e:.2f} m/s")
    print(f"Inelastic: vf = {vf_in:.2f} m/s (both objects move together)")
    print(f"\nMomentum is ALWAYS conserved: p_total = {m1*v1i + m2*v2i:.2f} kg m/s")
    print(f"Energy lost in inelastic collision: {KE_lost:.2f} J ({pct:.1f}%) -- converted to heat, sound, deformation")

interact(collision_comparison,
         m1=FloatSlider(value=2.0, min=0.5, max=5.0, step=0.5, description='Mass 1 (kg)'),
         m2=FloatSlider(value=1.0, min=0.5, max=5.0, step=0.5, description='Mass 2 (kg)'),
         v1i=FloatSlider(value=4.0, min=-5.0, max=5.0, step=0.5, description='v1 initial'),
         v2i=FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='v2 initial'));
```

### Interactive Demo 4: Newton's Cradle Simulation

The famous Newton's Cradle demonstrates conservation of momentum AND kinetic energy in elastic collisions. Watch how the number of balls released equals the number that swing out on the other side.

```python
def newtons_cradle(n_balls_released=1):
    """
    Newton's Cradle simulation with 5 balls.
    """
    n_total = 5
    L = 3.0        # string length
    r = 0.25       # ball radius
    spacing = 2 * r # ball spacing (touching)
    max_angle = 35  # degrees
    
    n_released = min(n_balls_released, n_total - 1)
    
    # Pivot points
    pivots_x = np.array([(i - (n_total-1)/2) * spacing for i in range(n_total)])
    pivot_y = L + r
    
    omega = np.sqrt(9.81 / L)  # angular frequency
    period = 2 * np.pi / omega
    total_time = 4 * period
    n_frames = 200
    
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.set_xlim(-4, 4)
    ax.set_ylim(-1, L + 2*r + 0.5)
    ax.set_aspect('equal')
    ax.set_title(f"Newton's Cradle: {n_released} ball(s) released", fontsize=14, fontweight='bold')
    ax.axis('off')
    
    # Draw frame
    frame_width = (n_total + 2) * spacing
    ax.plot([-frame_width/2, frame_width/2], [pivot_y, pivot_y], 'k-', linewidth=4)
    
    # Create ball patches and string lines
    balls = []
    strings = []
    for i in range(n_total):
        ball = plt.Circle((pivots_x[i], r), r, color='#B0BEC5', ec='#37474F', linewidth=2, zorder=5)
        ax.add_patch(ball)
        balls.append(ball)
        line, = ax.plot([pivots_x[i], pivots_x[i]], [pivot_y, r], 'k-', linewidth=1, zorder=4)
        strings.append(line)
    
    info = ax.text(0, -0.7, '', ha='center', fontsize=11,
                   bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))
    
    def get_angle(ball_idx, t):
        """Calculate angle for each ball based on phase."""
        theta_max = np.radians(max_angle)
        phase = omega * t
        cycle = phase % (2 * np.pi)
        
        # Left balls swing (released side)
        if ball_idx < n_released:
            if cycle < np.pi:  # Left balls are swinging back from left
                angle = -theta_max * np.cos(phase)  # negative = left side
                if angle > 0:
                    angle = 0  # Stop at contact
                return angle
            else:
                return 0.0  # At rest
        # Right balls swing (response side)
        elif ball_idx >= n_total - n_released:
            if cycle >= np.pi:
                angle = theta_max * np.cos(phase - np.pi)
                if angle < 0:
                    angle = 0
                return angle
            else:
                return 0.0
        else:
            return 0.0
    
    def update(frame):
        t = frame * total_time / n_frames
        
        for i in range(n_total):
            angle = get_angle(i, t)
            
            # Ball position
            bx = pivots_x[i] + L * np.sin(angle)
            by = pivot_y - L * np.cos(angle)
            
            balls[i].set_center((bx, by))
            strings[i].set_data([pivots_x[i], bx], [pivot_y, by])
            
            # Color moving balls
            if abs(angle) > 0.01:
                balls[i].set_facecolor('#FF7043')
            else:
                balls[i].set_facecolor('#B0BEC5')
        
        # Calculate total KE and PE for display
        info.set_text(f'Momentum and Kinetic Energy are both conserved!\n'
                      f'{n_released} ball(s) in = {n_released} ball(s) out')
        
        return balls + strings + [info]
    
    anim = FuncAnimation(fig, update, frames=n_frames, interval=40, blit=False)
    plt.close(fig)
    return HTML(anim.to_jshtml())

interact(newtons_cradle,
         n_balls_released=IntSlider(value=1, min=1, max=4, step=1, description='Balls released'));
```

### Example 1: Head-on Elastic Collision

A 2.0 kg ball moving at 5.0 m/s collides head-on elastically with a 3.0 kg ball initially at rest. Find the final velocities.

```python
# Worked Example 1: Elastic Collision
m1, m2 = 2.0, 3.0  # kg
v1i, v2i = 5.0, 0.0  # m/s

# Elastic collision formulas
v1f = ((m1 - m2) * v1i + 2 * m2 * v2i) / (m1 + m2)
v2f = (2 * m1 * v1i + (m2 - m1) * v2i) / (m1 + m2)

print("=" * 50)
print("ELASTIC COLLISION")
print("=" * 50)
print(f"\nBefore:")
print(f"  Ball 1: m = {m1} kg, v = {v1i} m/s, p = {m1*v1i:.1f} kg m/s")
print(f"  Ball 2: m = {m2} kg, v = {v2i} m/s, p = {m2*v2i:.1f} kg m/s")
print(f"  Total momentum: {m1*v1i + m2*v2i:.1f} kg m/s")
print(f"  Total KE: {0.5*m1*v1i**2 + 0.5*m2*v2i**2:.1f} J")

print(f"\nAfter:")
print(f"  Ball 1: v = {v1f:.2f} m/s  (bounces back!)")
print(f"  Ball 2: v = {v2f:.2f} m/s")
print(f"  Total momentum: {m1*v1f + m2*v2f:.1f} kg m/s  (conserved!)")
print(f"  Total KE: {0.5*m1*v1f**2 + 0.5*m2*v2f**2:.1f} J  (conserved!)")
```

### Example 2: Perfectly Inelastic Collision (Ballistic Pendulum)

A 10 g bullet traveling at 400 m/s embeds itself in a 2.0 kg wooden block hanging from a string. How high does the block+bullet rise?

```python
# Worked Example 2: Ballistic Pendulum
m_bullet = 0.010  # kg
m_block = 2.0     # kg
v_bullet = 400.0  # m/s
g = 9.81          # m/s^2

# Step 1: Perfectly inelastic collision (momentum conservation)
v_combined = (m_bullet * v_bullet) / (m_bullet + m_block)

# Step 2: Energy conservation (KE -> PE) for the rise
h = v_combined**2 / (2 * g)

# Energy analysis
KE_before = 0.5 * m_bullet * v_bullet**2
KE_after = 0.5 * (m_bullet + m_block) * v_combined**2
KE_lost = KE_before - KE_after

print("=" * 50)
print("BALLISTIC PENDULUM")
print("=" * 50)
print(f"\nStep 1: Collision (momentum conservation)")
print(f"  m_bullet * v_bullet = (m_bullet + m_block) * v_f")
print(f"  {m_bullet} x {v_bullet} = {m_bullet + m_block} x v_f")
print(f"  v_f = {v_combined:.3f} m/s")
print(f"\nStep 2: Rise (energy conservation)")
print(f"  (1/2)mv^2 = mgh")
print(f"  h = v^2 / (2g) = {v_combined:.3f}^2 / (2 x {g})")
print(f"  h = {h:.4f} m = {h*100:.2f} cm")
print(f"\nEnergy Analysis:")
print(f"  KE before collision: {KE_before:.1f} J")
print(f"  KE after collision:  {KE_after:.2f} J")
print(f"  Energy lost to deformation/heat: {KE_lost:.1f} J ({KE_lost/KE_before*100:.1f}%)")
```

### Mini Task: Bumper Design Challenge

You are designing a bumper for a 1200 kg vehicle that must survive a 5 m/s impact. Compare a rigid bumper (10 ms collision time) with a crumple-zone bumper (150 ms collision time). Calculate the peak force experienced by occupants in each case.

```python
# Mini Task: Bumper Design
m_car = 1200  # kg
v_impact = 5.0  # m/s

impulse = m_car * v_impact  # N*s

# Rigid bumper
dt_rigid = 0.010  # 10 ms
F_rigid = impulse / dt_rigid

# Crumple zone bumper
dt_crumple = 0.150  # 150 ms
F_crumple = impulse / dt_crumple

print("=" * 50)
print("BUMPER DESIGN COMPARISON")
print("=" * 50)
print(f"\nImpulse = m * delta_v = {m_car} x {v_impact} = {impulse:.0f} N s")
print(f"\nRigid Bumper (dt = {dt_rigid*1000:.0f} ms):")
print(f"  F_avg = {F_rigid:.0f} N = {F_rigid/1000:.0f} kN")
print(f"  Equivalent to {F_rigid/(m_car*9.81):.1f} g's on occupant!")
print(f"\nCrumple Zone (dt = {dt_crumple*1000:.0f} ms):")
print(f"  F_avg = {F_crumple:.0f} N = {F_crumple/1000:.1f} kN")
print(f"  Equivalent to {F_crumple/(m_car*9.81):.1f} g's on occupant")
print(f"\nForce reduction: {(1 - F_crumple/F_rigid)*100:.0f}%")
print(f"The crumple zone makes the collision {F_rigid/F_crumple:.0f}x safer!")
```

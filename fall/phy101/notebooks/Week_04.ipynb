## 2. Theory: Newton's Laws of Motion

### 2.1 The Three Laws

| Law | Statement | Key Equation |
|---|---|---|
| **1st** (Inertia) | An object at rest stays at rest, and an object in motion stays in uniform motion, unless acted on by a net external force. | $\vec{F}_{\text{net}} = 0 \Rightarrow \vec{a} = 0$ |
| **2nd** (F = ma) | The acceleration of an object is directly proportional to the net force and inversely proportional to its mass. | $\vec{F}_{\text{net}} = m\vec{a}$ |
| **3rd** (Action-Reaction) | For every action there is an equal and opposite reaction. | $\vec{F}_{A\to B} = -\vec{F}_{B\to A}$ |

### 2.2 Common Forces

| Force | Symbol | Direction | Magnitude |
|---|---|---|---|
| Weight | $\vec{W}$ | Straight down | $mg$ |
| Normal | $\vec{N}$ | Perpendicular to surface, away from surface | Adjusts to prevent penetration |
| Static friction | $\vec{f}_s$ | Along surface, opposing *tendency* of motion | $f_s \leq \mu_s N$ |
| Kinetic friction | $\vec{f}_k$ | Along surface, opposing motion | $f_k = \mu_k N$ |
| Tension | $\vec{T}$ | Along the string, pulling | Same throughout ideal (massless) string |
| Applied force | $\vec{F}_{\text{app}}$ | As specified | As specified |

> **Analogy:** Think of a free body diagram like a bank statement for forces — every force "deposited" on the object must be accounted for. The net balance ($\vec{F}_{\text{net}}$) determines the acceleration.

### 2.3 Friction Model

- **Static friction** prevents motion: $f_s \leq \mu_s N$. It adjusts from 0 up to the maximum.
- **Kinetic friction** acts during sliding: $f_k = \mu_k N$, typically $\mu_k < \mu_s$.
- The **normal force** $N$ is not always $mg$ — it depends on the geometry (e.g., inclined planes).

### 2.4 Strategy for Solving Force Problems

1. **Identify** all objects and the forces on each
2. **Draw** a free body diagram for each object
3. **Choose** a coordinate system (often tilted for inclines)
4. **Apply** $\sum F_x = ma_x$ and $\sum F_y = ma_y$
5. **Solve** the resulting equations

## 3. Interactive Demo 1 — Free Body Diagram Drawer

Select a scenario and see the forces drawn on the object. Toggle individual forces on/off to study the FBD.

```python
def draw_fbd(scenario='flat', show_weight=True, show_normal=True,
             show_friction=True, show_applied=True, theta=30, mu_k=0.3):
    """
    Draw a free body diagram for common scenarios.
    scenario: 'flat', 'incline', 'hanging'
    """
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.2)

    m = 5  # kg (for display)
    W = m * g
    arrow_kw = dict(head_width=0.2, head_length=0.15, linewidth=2.5)
    scale = 0.06  # scale forces to plot units

    if scenario == 'flat':
        # Block on flat surface
        block = patches.FancyBboxPatch((-1, -0.5), 2, 1, boxstyle="round,pad=0.05",
                                        facecolor='steelblue', edgecolor='black', linewidth=2)
        ax.add_patch(block)
        ax.text(0, 0, f'{m} kg', ha='center', va='center', fontsize=14, color='white', fontweight='bold')

        # Surface
        ax.plot([-4, 4], [-0.5, -0.5], 'k-', linewidth=3)
        for xh in np.arange(-4, 4, 0.5):
            ax.plot([xh, xh - 0.3], [-0.5, -0.9], 'k-', linewidth=0.8)

        F_app = 20  # N
        N_val = W
        f_val = min(mu_k * N_val, F_app)

        if show_weight:
            ax.arrow(0, 0, 0, -W * scale, color='red', **arrow_kw)
            ax.text(0.3, -W * scale / 2, f'W = {W:.1f} N', fontsize=12, color='red')
        if show_normal:
            ax.arrow(0, 0, 0, N_val * scale, color='green', **arrow_kw)
            ax.text(0.3, N_val * scale / 2, f'N = {N_val:.1f} N', fontsize=12, color='green')
        if show_applied:
            ax.arrow(1, 0, F_app * scale, 0, color='blue', **arrow_kw)
            ax.text(1 + F_app * scale / 2, 0.4, f'F = {F_app:.0f} N', fontsize=12, color='blue')
        if show_friction:
            ax.arrow(-1, 0, -f_val * scale, 0, color='orange', **arrow_kw)
            ax.text(-1 - f_val * scale - 0.3, 0.4, f'f = {f_val:.1f} N', fontsize=12, color='orange')

        net_x = F_app - f_val if show_applied and show_friction else (F_app if show_applied else (-f_val if show_friction else 0))
        a_val = net_x / m
        ax.set_title(f'FBD: Block on Flat Surface (μ_k = {mu_k})\nNet force = {net_x:.1f} N → a = {a_val:.2f} m/s²',
                     fontsize=14)

    elif scenario == 'incline':
        theta_rad = np.radians(theta)
        # Draw incline
        L = 7
        ix = [-3, -3 + L * np.cos(theta_rad), -3]
        iy = [-3, -3, -3 + L * np.sin(theta_rad)]
        ax.fill(ix, iy, color='gray', alpha=0.2)
        ax.plot(ix + [ix[0]], iy + [iy[0]], 'k-', linewidth=2)

        # Block center on slope
        cx = -3 + 0.45 * L * np.cos(theta_rad)
        cy = -3 + 0.45 * L * np.sin(theta_rad)

        # Rotated block
        bw, bh = 1.2, 0.8
        corners = np.array([[-bw/2, -bh/2], [bw/2, -bh/2],
                            [bw/2, bh/2], [-bw/2, bh/2], [-bw/2, -bh/2]])
        R_mat = np.array([[np.cos(theta_rad), -np.sin(theta_rad)],
                          [np.sin(theta_rad), np.cos(theta_rad)]])
        rotated = (R_mat @ corners.T).T + [cx, cy]
        ax.fill(rotated[:, 0], rotated[:, 1], color='steelblue', alpha=0.9)
        ax.plot(rotated[:, 0], rotated[:, 1], 'k-', linewidth=2)
        ax.text(cx, cy, f'{m} kg', ha='center', va='center', fontsize=11,
                color='white', fontweight='bold')

        # Forces
        N_val = m * g * np.cos(theta_rad)
        W_par = m * g * np.sin(theta_rad)
        W_perp = m * g * np.cos(theta_rad)
        f_val = mu_k * N_val

        if show_weight:
            ax.arrow(cx, cy, 0, -W * scale, color='red', **arrow_kw)
            ax.text(cx + 0.3, cy - W * scale / 2, f'W = {W:.1f} N', fontsize=11, color='red')

        if show_normal:
            nx = -np.sin(theta_rad) * N_val * scale
            ny = np.cos(theta_rad) * N_val * scale
            ax.arrow(cx, cy, nx, ny, color='green', **arrow_kw)
            ax.text(cx + nx + 0.2, cy + ny, f'N = {N_val:.1f} N', fontsize=11, color='green')

        if show_friction:
            fx = np.cos(theta_rad) * f_val * scale
            fy = np.sin(theta_rad) * f_val * scale
            ax.arrow(cx, cy, fx, fy, color='orange', **arrow_kw)
            ax.text(cx + fx + 0.2, cy + fy + 0.2, f'f = {f_val:.1f} N', fontsize=11, color='orange')

        # Angle arc
        arc_angles = np.linspace(0, theta_rad, 30)
        arc_r = 1.5
        ax.plot(-3 + arc_r * np.cos(arc_angles), -3 + arc_r * np.sin(arc_angles), 'k-', linewidth=1.5)
        ax.text(-3 + arc_r * 0.7, -3 + arc_r * 0.2, f'{theta}°', fontsize=12)

        a_net = g * (np.sin(theta_rad) - mu_k * np.cos(theta_rad))
        direction = "down the slope" if a_net > 0 else "stays put"
        ax.set_title(f'FBD: Block on Incline (θ={theta}°, μ_k={mu_k})\n'
                     f'a = {a_net:.2f} m/s² ({direction})', fontsize=14)

    elif scenario == 'hanging':
        # Block hanging from string
        block = patches.FancyBboxPatch((-0.8, -2.5), 1.6, 1.2, boxstyle="round,pad=0.05",
                                        facecolor='steelblue', edgecolor='black', linewidth=2)
        ax.add_patch(block)
        ax.text(0, -1.9, f'{m} kg', ha='center', va='center', fontsize=14, color='white', fontweight='bold')

        # String
        ax.plot([0, 0], [-1.3, 2.5], 'k-', linewidth=2)
        ax.plot([-1, 1], [2.5, 2.5], 'k-', linewidth=4)  # ceiling
        for xh in np.arange(-1, 1.1, 0.3):
            ax.plot([xh, xh - 0.2], [2.5, 2.8], 'k-', linewidth=0.8)

        if show_weight:
            ax.arrow(0, -1.9, 0, -W * scale, color='red', **arrow_kw)
            ax.text(0.3, -1.9 - W * scale / 2, f'W = {W:.1f} N', fontsize=12, color='red')
        if show_normal:  # here it's tension
            ax.arrow(0, -1.9, 0, W * scale, color='purple', **arrow_kw)
            ax.text(0.3, -1.9 + W * scale / 2, f'T = {W:.1f} N', fontsize=12, color='purple')

        ax.set_title(f'FBD: Hanging Block (equilibrium)\nT = W = {W:.1f} N', fontsize=14)

    ax.set_xlabel('x')
    ax.set_ylabel('y')
    plt.tight_layout()
    plt.show()


scenario_dd = widgets.Dropdown(options=['flat', 'incline', 'hanging'],
                               value='flat', description='Scenario:')
theta_sl = widgets.IntSlider(value=30, min=5, max=70, step=5, description='θ (°):')
mu_sl = widgets.FloatSlider(value=0.3, min=0.0, max=0.8, step=0.05,
                            description='μ_k:', readout_format='.2f')
w_chk = widgets.Checkbox(value=True, description='Weight')
n_chk = widgets.Checkbox(value=True, description='Normal/Tension')
f_chk = widgets.Checkbox(value=True, description='Friction')
a_chk = widgets.Checkbox(value=True, description='Applied')

@widgets.interact(scenario=scenario_dd, show_weight=w_chk, show_normal=n_chk,
                  show_friction=f_chk, show_applied=a_chk, theta=theta_sl, mu_k=mu_sl)
def _fbd(scenario, show_weight, show_normal, show_friction, show_applied, theta, mu_k):
    draw_fbd(scenario, show_weight, show_normal, show_friction, show_applied, theta, mu_k)
```

## 4. Interactive Demo 2 — Inclined Plane Simulator

A block is placed on an incline. Adjust the angle, mass, and friction coefficient to see:
- Whether the block slides or stays
- The acceleration if it slides
- An animation of the block sliding down

The critical angle at which the block just begins to slide is $\theta_c = \arctan(\mu_s)$.

```python
def incline_simulation(theta_deg=30, mass=5.0, mu_s=0.4, mu_k=0.3):
    """Simulate and animate a block on an inclined plane."""
    theta = np.radians(theta_deg)
    W = mass * g
    N = W * np.cos(theta)
    W_parallel = W * np.sin(theta)
    f_s_max = mu_s * N
    f_k = mu_k * N

    # Check if block slides
    slides = W_parallel > f_s_max
    theta_critical = np.degrees(np.arctan(mu_s))

    if slides:
        a = g * (np.sin(theta) - mu_k * np.cos(theta))
        status = f"SLIDING  (a = {a:.2f} m/s²)"
    else:
        a = 0.0
        f_s_actual = W_parallel
        status = f"STATIONARY  (f_s = {f_s_actual:.1f} N < f_s_max = {f_s_max:.1f} N)"

    # Info panel
    print(f"╔══════════════════════════════════════════╗")
    print(f"║  Inclined Plane Analysis                 ║")
    print(f"╠══════════════════════════════════════════╣")
    print(f"║  θ = {theta_deg}°    m = {mass} kg                  ")
    print(f"║  μ_s = {mu_s}   μ_k = {mu_k}                  ")
    print(f"║  W = {W:.1f} N                              ")
    print(f"║  N = {N:.1f} N                              ")
    print(f"║  W∥ = {W_parallel:.1f} N   W⊥ = {N:.1f} N          ")
    print(f"║  f_s(max) = {f_s_max:.1f} N   f_k = {f_k:.1f} N    ")
    print(f"║  Critical angle: θ_c = {theta_critical:.1f}°           ")
    print(f"║  Status: {status}")
    print(f"╚══════════════════════════════════════════╝")

    if not slides:
        # Just draw the static scene
        fig, ax = plt.subplots(figsize=(10, 7))
        L = 8
        ix = [0, L * np.cos(theta), 0]
        iy = [0, 0, L * np.sin(theta)]
        ax.fill(ix, iy, color='gray', alpha=0.2)
        ax.plot(ix + [0], iy + [0], 'k-', linewidth=2)

        cx = 0.5 * L * np.cos(theta)
        cy = 0.5 * L * np.sin(theta)
        bw, bh = 0.8, 0.6
        corners = np.array([[-bw/2, 0], [bw/2, 0], [bw/2, bh], [-bw/2, bh], [-bw/2, 0]])
        R_mat = np.array([[np.cos(theta), -np.sin(theta)],
                          [np.sin(theta), np.cos(theta)]])
        rotated = (R_mat @ corners.T).T + [cx, cy]
        ax.fill(rotated[:, 0], rotated[:, 1], color='steelblue', alpha=0.9)
        ax.plot(rotated[:, 0], rotated[:, 1], 'k-', linewidth=2)
        ax.text(cx, cy + 0.4, 'STATIC', ha='center', va='center', fontsize=11, fontweight='bold', color='white')

        ax.set_xlim(-1, L * np.cos(theta) + 1)
        ax.set_ylim(-1, L * np.sin(theta) + 1)
        ax.set_aspect('equal')
        ax.set_title(f'Block at rest on incline (θ = {theta_deg}° < θ_c = {theta_critical:.1f}°)', fontsize=14)
        plt.tight_layout()
        plt.show()
        return

    # Animate sliding
    L = 8
    t_max = np.sqrt(2 * L / a) if a > 0 else 3
    t_arr = np.linspace(0, t_max, 150)
    s_arr = 0.5 * a * t_arr**2  # distance along slope

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # Left: animation scene
    ax = axes[0]
    ix = [0, L * np.cos(theta), 0]
    iy = [0, 0, L * np.sin(theta)]
    ax.fill(ix, iy, color='gray', alpha=0.2)
    ax.plot(ix + [0], iy + [0], 'k-', linewidth=2)
    ax.set_xlim(-1, L * np.cos(theta) + 2)
    ax.set_ylim(-1, L * np.sin(theta) + 1)
    ax.set_aspect('equal')
    ax.set_title('Sliding Block')

    bw, bh = 0.6, 0.5
    corners_base = np.array([[-bw/2, 0], [bw/2, 0], [bw/2, bh], [-bw/2, bh], [-bw/2, 0]])
    R_mat = np.array([[np.cos(theta), -np.sin(theta)],
                      [np.sin(theta), np.cos(theta)]])

    block_patch, = ax.fill([], [], color='steelblue', alpha=0.9)
    block_line, = ax.plot([], [], 'k-', linewidth=2)
    time_text = ax.text(0.02, 0.95, '', transform=ax.transAxes, fontsize=11,
                        verticalalignment='top',
                        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

    # Right: velocity and position vs time
    ax2 = axes[1]
    v_arr = a * t_arr
    ax2.plot(t_arr, s_arr, 'b-', linewidth=2, label='Position s(t)')
    ax2.plot(t_arr, v_arr, 'r--', linewidth=2, label='Velocity v(t)')
    pos_dot, = ax2.plot([], [], 'bo', markersize=10)
    vel_dot, = ax2.plot([], [], 'rs', markersize=10)
    ax2.set_xlabel('Time (s)')
    ax2.set_ylabel('s (m) / v (m/s)')
    ax2.set_title('Position & Velocity vs Time')
    ax2.legend()

    start_pos = 0.85  # start near top of slope (fraction of L)

    def init():
        return block_patch, block_line, time_text, pos_dot, vel_dot

    def update(frame):
        # Block slides down from top
        frac = start_pos - s_arr[frame] / L
        frac = max(frac, 0.05)
        cx = frac * L * np.cos(theta)
        cy = frac * L * np.sin(theta)
        rotated = (R_mat @ corners_base.T).T + [cx, cy]
        block_patch.set_xy(rotated)
        block_line.set_data(rotated[:, 0], rotated[:, 1])
        time_text.set_text(f't = {t_arr[frame]:.2f} s\nv = {v_arr[frame]:.1f} m/s\ns = {s_arr[frame]:.1f} m')
        pos_dot.set_data([t_arr[frame]], [s_arr[frame]])
        vel_dot.set_data([t_arr[frame]], [v_arr[frame]])
        return block_patch, block_line, time_text, pos_dot, vel_dot

    anim = FuncAnimation(fig, update, init_func=init,
                         frames=len(t_arr), interval=30, blit=True)
    plt.close(fig)
    display(HTML(anim.to_jshtml()))


# Widget interface
theta_w = widgets.IntSlider(value=35, min=5, max=75, step=1, description='θ (°):')
mass_w = widgets.FloatSlider(value=5.0, min=1.0, max=20.0, step=0.5, description='Mass (kg):')
mus_w = widgets.FloatSlider(value=0.4, min=0.0, max=1.0, step=0.05,
                            description='μ_s:', readout_format='.2f')
muk_w = widgets.FloatSlider(value=0.3, min=0.0, max=0.8, step=0.05,
                            description='μ_k:', readout_format='.2f')
run_btn = widgets.Button(description='▶ Run', button_style='success',
                         layout=widgets.Layout(width='120px'))
incline_out = widgets.Output()

def on_run(b):
    with incline_out:
        clear_output(wait=True)
        incline_simulation(theta_w.value, mass_w.value, mus_w.value, muk_w.value)

run_btn.on_click(on_run)
display(widgets.VBox([
    widgets.HBox([theta_w, mass_w]),
    widgets.HBox([mus_w, muk_w, run_btn]),
    incline_out
]))
on_run(None)
```

## 5. Interactive Demo 3 — Atwood Machine Simulator

An **Atwood machine** consists of two masses connected by a string over a frictionless pulley.

For masses $m_1 > m_2$:
$$a = \frac{(m_1 - m_2)}{(m_1 + m_2)}\,g \qquad T = \frac{2\,m_1\,m_2}{m_1 + m_2}\,g$$

The heavier mass accelerates downward while the lighter mass accelerates upward.

```python
def atwood_machine(m1=8.0, m2=5.0):
    """Animate an Atwood machine."""
    a = (m1 - m2) / (m1 + m2) * g
    T = 2 * m1 * m2 / (m1 + m2) * g

    print(f"Atwood Machine: m₁ = {m1} kg, m₂ = {m2} kg")
    print(f"  Acceleration: a = {a:.3f} m/s²")
    print(f"  Tension:      T = {T:.2f} N")
    if abs(m1 - m2) < 0.01:
        print("  System is in equilibrium (m₁ = m₂).")

    # Animation
    t_max = 2.0  # seconds
    n_frames = 100
    t_arr = np.linspace(0, t_max, n_frames)
    s_arr = 0.5 * abs(a) * t_arr**2  # displacement
    v_arr = abs(a) * t_arr

    fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 7),
                                   gridspec_kw={'width_ratios': [1, 1.2]})

    # --- Left panel: Atwood machine diagram ---
    ax.set_xlim(-3, 3)
    ax.set_ylim(-5, 4)
    ax.set_aspect('equal')
    ax.set_title('Atwood Machine', fontsize=14)
    ax.axis('off')

    # Pulley
    pulley = plt.Circle((0, 3), 0.5, color='gray', fill=True, linewidth=2)
    ax.add_patch(pulley)
    ax.plot([0], [3], 'k+', markersize=10, markeredgewidth=2)
    # Support
    ax.plot([-1, 1], [3.8, 3.8], 'k-', linewidth=4)
    ax.plot([0, 0], [3.5, 3.8], 'k-', linewidth=3)

    # Initial positions
    y1_init = 0.0
    y2_init = 0.0

    # Strings
    string1, = ax.plot([], [], 'k-', linewidth=2)
    string2, = ax.plot([], [], 'k-', linewidth=2)

    # Mass blocks
    block1 = patches.FancyBboxPatch((-2.2, y1_init), 1.4, 1,
                                     boxstyle="round,pad=0.05",
                                     facecolor='crimson', edgecolor='black', linewidth=2)
    block2 = patches.FancyBboxPatch((0.8, y2_init), 1.4, 1,
                                     boxstyle="round,pad=0.05",
                                     facecolor='royalblue', edgecolor='black', linewidth=2)
    ax.add_patch(block1)
    ax.add_patch(block2)

    label1 = ax.text(-1.5, y1_init + 0.5, f'm₁\n{m1}kg', ha='center', va='center',
                     fontsize=11, color='white', fontweight='bold')
    label2 = ax.text(1.5, y2_init + 0.5, f'm₂\n{m2}kg', ha='center', va='center',
                     fontsize=11, color='white', fontweight='bold')

    info = ax.text(0, -4.5, '', ha='center', fontsize=12,
                   bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))

    # --- Right panel: graphs ---
    ax2.set_xlabel('Time (s)')
    ax2.set_ylabel('Displacement (m) / Velocity (m/s)')
    ax2.set_title('Displacement & Velocity vs Time')
    ax2.plot(t_arr, s_arr, 'b-', linewidth=2, label='|displacement|')
    ax2.plot(t_arr, v_arr, 'r--', linewidth=2, label='|velocity|')
    s_dot, = ax2.plot([], [], 'bo', markersize=8)
    v_dot, = ax2.plot([], [], 'rs', markersize=8)
    ax2.legend()

    def init():
        return block1, block2, string1, string2, label1, label2, info, s_dot, v_dot

    def update(frame):
        disp = s_arr[frame]
        sign = 1 if m1 > m2 else -1

        y1 = y1_init - sign * disp
        y2 = y2_init + sign * disp

        block1.set_y(y1)
        block2.set_y(y2)
        label1.set_position((-1.5, y1 + 0.5))
        label2.set_position((1.5, y2 + 0.5))

        # Strings from pulley to top of blocks
        string1.set_data([-0.5, -1.5, -1.5], [3, 3, y1 + 1])
        string2.set_data([0.5, 1.5, 1.5], [3, 3, y2 + 1])

        info.set_text(f't = {t_arr[frame]:.2f} s   |s| = {disp:.2f} m   |v| = {v_arr[frame]:.2f} m/s\n'
                      f'a = {a:.3f} m/s²   T = {T:.1f} N')

        s_dot.set_data([t_arr[frame]], [s_arr[frame]])
        v_dot.set_data([t_arr[frame]], [v_arr[frame]])
        return block1, block2, string1, string2, label1, label2, info, s_dot, v_dot

    anim = FuncAnimation(fig, update, init_func=init,
                         frames=n_frames, interval=40, blit=True)
    plt.close(fig)
    display(HTML(anim.to_jshtml()))


m1_w = widgets.FloatSlider(value=8, min=1, max=20, step=0.5, description='m₁ (kg):')
m2_w = widgets.FloatSlider(value=5, min=1, max=20, step=0.5, description='m₂ (kg):')
atwood_btn = widgets.Button(description='▶ Simulate', button_style='success',
                            layout=widgets.Layout(width='140px'))
atwood_out = widgets.Output()

def on_atwood(b):
    with atwood_out:
        clear_output(wait=True)
        atwood_machine(m1_w.value, m2_w.value)

atwood_btn.on_click(on_atwood)
display(widgets.VBox([widgets.HBox([m1_w, m2_w, atwood_btn]), atwood_out]))
on_atwood(None)
```

## 6. Interactive Demo 4 — Friction Force Visualizer

This demo shows how friction behaves as you gradually increase an applied force:
- **Static regime:** friction matches the applied force (no motion)
- **Transition:** at $F = \mu_s N$, the block just begins to slide
- **Kinetic regime:** friction drops to $\mu_k N$ and stays constant

```python
def friction_visualizer(mass=5.0, mu_s=0.5, mu_k=0.35):
    """Show friction force vs applied force."""
    N = mass * g
    f_s_max = mu_s * N
    f_k = mu_k * N

    F_applied = np.linspace(0, f_s_max * 1.8, 500)
    f_friction = np.where(F_applied <= f_s_max, F_applied, f_k)
    a_net = np.where(F_applied <= f_s_max, 0, (F_applied - f_k) / mass)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # Left: friction vs applied
    ax1.plot(F_applied, f_friction, 'b-', linewidth=2.5, label='Friction force')
    ax1.plot(F_applied, F_applied, 'k--', linewidth=1, alpha=0.4, label='f = F (reference)')
    ax1.axhline(f_s_max, color='red', linestyle=':', alpha=0.7, label=f'f_s(max) = {f_s_max:.1f} N')
    ax1.axhline(f_k, color='orange', linestyle=':', alpha=0.7, label=f'f_k = {f_k:.1f} N')
    ax1.axvline(f_s_max, color='gray', linestyle='--', alpha=0.3)

    # Shade regions
    ax1.axvspan(0, f_s_max, alpha=0.08, color='green', label='Static region')
    ax1.axvspan(f_s_max, F_applied[-1], alpha=0.08, color='red', label='Kinetic region')

    ax1.set_xlabel('Applied Force (N)', fontsize=12)
    ax1.set_ylabel('Friction Force (N)', fontsize=12)
    ax1.set_title('Friction vs Applied Force', fontsize=14)
    ax1.legend(fontsize=9, loc='upper left')
    ax1.set_xlim(0, F_applied[-1])
    ax1.set_ylim(0, f_s_max * 1.5)

    # Right: acceleration vs applied
    ax2.plot(F_applied, a_net, 'r-', linewidth=2.5)
    ax2.axvline(f_s_max, color='gray', linestyle='--', alpha=0.3)
    ax2.axvspan(0, f_s_max, alpha=0.08, color='green')
    ax2.axvspan(f_s_max, F_applied[-1], alpha=0.08, color='red')
    ax2.text(f_s_max * 0.4, max(a_net) * 0.8, 'Static\n(a = 0)', ha='center',
             fontsize=12, color='green', fontweight='bold')
    ax2.text(f_s_max * 1.4, max(a_net) * 0.5, 'Kinetic\n(sliding)', ha='center',
             fontsize=12, color='red', fontweight='bold')
    ax2.set_xlabel('Applied Force (N)', fontsize=12)
    ax2.set_ylabel('Acceleration (m/s²)', fontsize=12)
    ax2.set_title('Acceleration vs Applied Force', fontsize=14)

    plt.tight_layout()
    plt.show()


@widgets.interact(
    mass=widgets.FloatSlider(value=5, min=1, max=20, step=0.5, description='Mass (kg):'),
    mu_s=widgets.FloatSlider(value=0.5, min=0.1, max=1.0, step=0.05,
                             description='μ_s:', readout_format='.2f'),
    mu_k=widgets.FloatSlider(value=0.35, min=0.05, max=0.9, step=0.05,
                             description='μ_k:', readout_format='.2f')
)
def _friction_viz(mass, mu_s, mu_k):
    if mu_k > mu_s:
        print("⚠️  μ_k should be ≤ μ_s. Adjusting μ_k = μ_s.")
        mu_k = mu_s
    friction_visualizer(mass, mu_s, mu_k)
```

## 7. Interactive Demo 5 — Friction Parameter Sweep on Incline

How does the critical angle change with the friction coefficient? This plot sweeps $\mu_s$ from 0 to 1 and shows the corresponding critical angle $\theta_c = \arctan(\mu_s)$.

```python
def friction_sweep():
    """Plot critical angle vs friction coefficient."""
    mu_vals = np.linspace(0, 1.5, 300)
    theta_c = np.degrees(np.arctan(mu_vals))

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(mu_vals, theta_c, 'b-', linewidth=2.5)
    ax.fill_between(mu_vals, theta_c, alpha=0.1, color='green', label='Block stays (θ < θ_c)')
    ax.fill_between(mu_vals, theta_c, 90, alpha=0.1, color='red', label='Block slides (θ > θ_c)')

    # Mark common materials
    materials = {
        'Ice on ice': 0.03,
        'Wood on wood': 0.4,
        'Rubber on concrete': 0.8,
        'Steel on steel': 0.6,
    }
    for name, mu in materials.items():
        tc = np.degrees(np.arctan(mu))
        ax.plot(mu, tc, 'ko', markersize=8)
        ax.annotate(f'{name}\nμ={mu}, θ_c={tc:.0f}°',
                    (mu, tc), textcoords='offset points',
                    xytext=(10, 8), fontsize=9, arrowprops=dict(arrowstyle='->', color='gray'))

    ax.set_xlabel('Static friction coefficient μ_s', fontsize=12)
    ax.set_ylabel('Critical angle θ_c (°)', fontsize=12)
    ax.set_title('Critical Angle vs Friction Coefficient', fontsize=14)
    ax.legend(fontsize=11)
    ax.set_xlim(0, 1.5)
    ax.set_ylim(0, 70)
    plt.tight_layout()
    plt.show()

friction_sweep()
```

## 8. Worked Examples

### Example 1: Block on a Flat Surface

A 10 kg block on a horizontal surface ($\mu_k = 0.25$) is pushed with 40 N. Find the acceleration.

```python
m_ex1 = 10  # kg
F_app = 40  # N
mu_k_ex1 = 0.25

N_ex1 = m_ex1 * g
f_k_ex1 = mu_k_ex1 * N_ex1
a_ex1 = (F_app - f_k_ex1) / m_ex1

print("Block on flat surface:")
print(f"  Normal force  : N  = mg = {N_ex1:.1f} N")
print(f"  Kinetic friction: f_k = μ_k × N = {f_k_ex1:.2f} N")
print(f"  Net force     : F_net = F - f_k = {F_app - f_k_ex1:.2f} N")
print(f"  Acceleration  : a = F_net / m = {a_ex1:.2f} m/s²")
```

### Example 2: Block on an Incline

A 5 kg block is placed on a 30° incline with $\mu_s = 0.4$, $\mu_k = 0.3$. Does it slide? If yes, what is its acceleration?

```python
m_ex2 = 5  # kg
theta_ex2 = 30  # degrees
mu_s_ex2 = 0.4
mu_k_ex2 = 0.3
theta_r = np.radians(theta_ex2)

W_ex2 = m_ex2 * g
N_ex2 = W_ex2 * np.cos(theta_r)
W_par = W_ex2 * np.sin(theta_r)
f_s_max = mu_s_ex2 * N_ex2

print(f"Block on {theta_ex2}° incline:")
print(f"  Weight parallel to slope: W∥ = {W_par:.2f} N")
print(f"  Max static friction:      f_s(max) = {f_s_max:.2f} N")

if W_par > f_s_max:
    f_k_ex2 = mu_k_ex2 * N_ex2
    a_ex2 = (W_par - f_k_ex2) / m_ex2
    print(f"  W∥ > f_s(max)  →  Block SLIDES!")
    print(f"  Kinetic friction: f_k = {f_k_ex2:.2f} N")
    print(f"  Acceleration: a = {a_ex2:.2f} m/s²")
else:
    print(f"  W∥ ≤ f_s(max)  →  Block STAYS PUT")
    print(f"  Static friction adjusts to f_s = {W_par:.2f} N")
```

### Example 3: Atwood Machine

Two masses $m_1 = 7$ kg and $m_2 = 3$ kg are connected by a string over a frictionless pulley. Find the acceleration and tension.

```python
m1_ex3 = 7  # kg
m2_ex3 = 3  # kg

a_ex3 = (m1_ex3 - m2_ex3) / (m1_ex3 + m2_ex3) * g
T_ex3 = 2 * m1_ex3 * m2_ex3 / (m1_ex3 + m2_ex3) * g

print(f"Atwood machine: m₁ = {m1_ex3} kg, m₂ = {m2_ex3} kg")
print(f"  a = (m₁ - m₂)/(m₁ + m₂) × g = {a_ex3:.3f} m/s²")
print(f"  T = 2m₁m₂/(m₁ + m₂) × g = {T_ex3:.2f} N")
print(f"")
print(f"  Check: T < m₁g = {m1_ex3*g:.1f} N  ✓  (m₁ accelerates down)")
print(f"  Check: T > m₂g = {m2_ex3*g:.1f} N  ✓  (m₂ accelerates up)")
```

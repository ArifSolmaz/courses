## 1. Before you start / Başlamadan önce

**Retrieval question:** Apply $\sum\mathbf F=m\mathbf a$ to one object. Review vector components from Weeks 1–3.

### 🎯 Learning Objectives

By the end of this session you will be able to:

1. **State** Newton's three laws of motion and explain their physical meaning
2. **Draw** correct free body diagrams (FBDs) for objects on flat and inclined surfaces
3. **Distinguish** between static and kinetic friction and apply $\lvert f_s\rvert\le\mu_sN$ at rest and $\lvert f_k\rvert=\mu_kN$ during sliding
4. **Solve** for acceleration, tension, and normal forces in single-body and two-body problems
5. **Simulate** motion on an inclined plane and in Atwood machine configurations

### Before calculating: a free-body diagram becomes algebra

**Core route:** select one object → draw only forces acting on it → choose axes → write $\sum F_x=ma_x$ and $\sum F_y=ma_y$ → solve by hand. The simulations are optional checks; you are learning mechanics, not Python syntax.

**Forces are interactions.** Weight $mg$ comes from Earth, the normal force $N$ from a surface, tension $T$ from a string, and friction from contact. 'Net force' is their sum, not an extra arrow. A third-law partner acts on the other object and does not belong in the same free-body diagram.

**Algebra bridge:** for a lifted mass, take up positive. Isolate tension $T$:

$$\begin{aligned}
T-mg&=ma,\\
T&=ma+mg &&\text{add }mg\text{ to both sides},\\
T&=m(a+g) &&\text{factor the common mass }m.
\end{aligned}$$

Both terms in the parentheses have acceleration units. Thus $\mathrm{kg}\times\mathrm{m/s^2}=\mathrm N$. If $a=0$, the expression reduces to $T=mg$.

**Inclines:** choose one axis along the slope and one perpendicular. Weight components are $mg\sin\theta$ down the slope and $mg\cos\theta$ into it. If no other perpendicular force acts and there is no perpendicular acceleration, $N=mg\cos\theta$. On a horizontal surface with an angled push, $N$ need not equal $mg$.

**Friction decision:** while at rest, $0\le\lvert f_s\rvert\le\mu_sN$; static friction adjusts to what is needed. Check that limit before assuming motion. During sliding, $\lvert f_k\rvert=\mu_kN$ opposite relative sliding. Knowing only $\mu_k$ is insufficient to decide whether a resting box starts moving.

**Review stop:** a block moving right at constant speed has zero net horizontal force. It can still have two nonzero opposing horizontal forces. Draw them and show their cancellation.

**Türkçe:** Önce hangi cismi incelediğini belirle. Kuvvetler toplamını yazmadan $F=ma$ içine tek bir kuvveti koyma. Sürtünmenin yönünü ve cismin kayıp kaymadığını ayrıca kontrol et.

### First 15 minutes: repair Week 03 / İlk 15 dakika: 3. haftayı onar

Week 03 carried two topics in one session, so this week opens by repairing the part that matters most for
everything that follows: **drawing one object and every force on it.** Work through the three checks
below before the new material. If all three are comfortable, you have lost nothing; if one is not, you

**Check 1 — what belongs on the diagram?** A block is sliding to the right along a rough horizontal table,
having been pushed a moment ago and now released. List every force acting **on the block**.

<details><summary>Answer / Yanıt</summary>

Three forces: **weight** $mg$ down, **normal force** $N$ up from the table, **friction** $f$ pointing
*left*, opposing the motion.

There is no forward force. The push has finished; nothing is touching the block from behind. The most
common error in mechanics is drawing a forward arrow because the block is moving forward — motion is not
a force. Velocity and force are different arrows, and they are often in different directions.

**TR:** Üç kuvvet: ağırlık, normal kuvvet ve harekete karşı **sola** doğru sürtünme. İleri doğru kuvvet
**yoktur**: itme bitmiştir. Hareket bir kuvvet değildir; hız ve kuvvet farklı oklardır.

</details>

**Check 2 — turn the diagram into algebra.** A block on a frictionless incline of angle $\theta$. Resolve
the weight into components parallel and perpendicular to the surface, and write the two equations.

<details><summary>Answer / Yanıt</summary>

Choose axes **along** and **perpendicular to** the slope, not horizontal and vertical — that single choice
is what makes incline problems easy.

$$\boxed{\;\text{along: } a = g\sin\theta, \qquad \text{perpendicular: } N = mg\cos\theta\;}$$

**Limiting checks:** $\theta \to 0$ gives $a \to 0$ and $N \to mg$ (flat table ✅);
$\theta \to 90^\circ$ gives $a \to g$ and $N \to 0$ (free fall ✅). Note the mass cancels from $a$ — and
$\sin$ goes with the direction of motion while $\cos$ goes with the normal. Swapping those two is the
second most common error of the semester.

**TR:** Eksenleri yatay/düşey değil, **eğime paralel ve dik** seç; eğik düzlem problemlerini kolaylaştıran
tek şey bu seçimdir. Sınır kontrolleri formülü doğrular. $\sin$ hareket yönüyle, $\cos$ normalle gider;
bunları karıştırmak dönemin en sık ikinci hatasıdır.

</details>

**Check 3 — Newton's third law, precisely.** A book rests on a table. The book's weight is $mg$ down and
the table pushes up with $N = mg$. Are these two forces a third-law pair?

<details><summary>Answer / Yanıt</summary>

**No.** A third-law pair acts on **two different objects**, and these both act on the book.

- Weight: the Earth pulls the book down. Its partner is *the book pulling the Earth up.*
- Normal force: the table pushes the book up. Its partner is *the book pushing the table down.*

These two forces on the book happen to be equal here only because the book is in equilibrium. Put the
book in a lift that is accelerating upward and $N > mg$ — yet each force still has exactly the same
third-law partner. **Equal and opposite on the same object is equilibrium; equal and opposite on two
objects is the third law.** Keeping those apart is worth more marks than any formula this week.

**TR:** **Hayır.** Üçüncü yasa çifti **iki farklı cisme** etki eder; bunların ikisi de kitaba etki
ediyor. Ağırlığın eşi, kitabın Yer'i yukarı çekmesidir; normal kuvvetin eşi, kitabın masayı aşağı
itmesidir. Burada eşit olmaları yalnızca kitap dengede olduğu içindir: yukarı ivmelenen bir asansörde
$N > mg$ olur, ama eşler değişmez. **Aynı cisimde eşit ve zıt → denge; iki cisimde eşit ve zıt → üçüncü
yasa.**

</details>

**TR:** 3. hafta tek oturumda iki konu taşıdı; bu hafta en kritik parçayı onararak başlıyor: tek bir cismi
ve üzerindeki bütün kuvvetleri çizmek. Üçü de rahatsa hiçbir şey kaybetmedin; biri değilse dersin kalanını
kurtardın.


### How this week is worked / Bu hafta nasıl çalışılır

**Every problem, worked example and experiment in this course follows the same six steps.** Step 4 is the
one added in September 2026, and it is the one that separates a physicist from a calculator operator.

> **draw → choose a law → rearrange to the symbolic answer → check a limiting case → substitute with units → interpret**

1. **Symbolic answer first.** Isolate the unknown in symbols, before any number appears.
2. **One limiting case.** Push a parameter to a value whose answer you already know — a mass to zero, an
   angle to $90^\circ$, a friction coefficient to zero — and confirm the formula behaves sensibly. If it
   does not, the algebra is wrong and you have just caught it **for free**, before wasting arithmetic on it.
3. **Then numbers**, with units carried through every line.

Why bother: the symbolic route is *less* arithmetic, not more, and it is where the marks are. A correct
symbolic answer with a slip in the final multiplication is nearly full credit; a correct number with no
derivation is not.

```python
#@title Run once — prepare the demonstration controls
"""Shared demonstration interface, embedded in every PHY101 notebook.

Only standard ipywidgets, IPython and matplotlib are used, so the notebooks stay
self-contained in Colab and in a local Jupyter. The design goals are:

* A slider change must always produce a visible reaction. The status line under
  the controls says "Updating…" immediately and reports the draw time afterwards.
* The graph is replaced through the same Output-widget route that
  ``ipywidgets.interact`` uses (``clear_output(wait=True)`` followed by a fresh
  display), which is the most widely tested path in Colab and Jupyter. No
  output-capturing context is used: ipykernel 7 dispatches widget messages
  concurrently and IPython's capture object breaks that dispatch.
* Live updates while dragging are switched on when a graph draws quickly and
  switched off (update on release) when it draws slowly, so the kernel never
  falls behind a fast slider.
"""
import functools
import sys
import time
import traceback

import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import HTML, clear_output, display

# Force the inline backend. Otherwise a local kernel may choose a desktop
# backend and block at plt.show(), which looks like a frozen notebook.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

_physics_panels = []
_physics_callback_errors = []

# Draw-time thresholds (seconds) for switching live dragging on and off.
PHYSICS_LIVE_ON = 0.12
PHYSICS_LIVE_OFF = 0.25


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; background: #fff; }
.phy101-controls { padding: 0 0 2px; box-sizing: border-box; }
.phy101-status { font-size: 12px; color: #4a5a6a; padding: 0 2px 6px; min-height: 18px; }
.phy101-status.busy { color: #b45309; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-animation-panel { max-width: 100%; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1; background: white; color: #172433; padding: 4px; }
@media (max-width: 650px) {
  .phy101-controls, .phy101-plot-output { width: 100% !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Lay the controls out as a wrapping toolbar with full-length labels."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "310px"
        control.layout.flex = "0 1 310px"
        control.layout.max_width = "100%"
        control.layout.min_width = "0"
        control.layout.margin = "2px 6px 2px 0"
        if isinstance(control, widgets.Button):
            control.layout.width = "auto"
            control.layout.flex = "0 0 auto"
    # Repeat the style inside the widget tree: Colab isolates output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.Box([style] + flat, layout=widgets.Layout(
        display="flex", flex_flow="row wrap", align_items="center",
        width="100%", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        width="100%", min_width="0", max_width="100%",
        height="auto", overflow="visible", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Button-driven demos: a compact control toolbar directly above the result."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="column",
                              align_items="stretch", width="100%"))
    panel.add_class("phy101-panel")
    return panel


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


class PhysicsPanel:
    """Controls, a status line and one Output widget that shows the latest result."""

    def __init__(self, function, controls):
        self.f = function
        self.controls = controls
        self.out = _physics_output(widgets.Output())
        self.status = widgets.HTML(value="")
        self.status.add_class("phy101-status")
        self.seconds = None
        self.live = True
        self.updates = 0
        self.last_outputs = 0
        self.figures = 0
        self.error = None
        visible = []
        for control in controls.values():
            if isinstance(control, widgets.fixed):
                continue
            visible.append(control)
            if hasattr(control, "continuous_update"):
                control.continuous_update = True
            control.observe(self._changed, names="value")
        self.widget = widgets.VBox([_physics_controls(visible), self.status, self.out],
                                   layout=widgets.Layout(width="100%"))
        self.widget.add_class("phy101-panel")
        self.children = self.widget.children
        _physics_panels.append(self)
        self.render()

    # Compatibility with the earlier validation code.
    @property
    def layout(self):
        return self.widget.layout

    def _changed(self, change):
        self.render()

    def _set_live(self, live):
        if live == self.live:
            return
        self.live = live
        for control in self.controls.values():
            if hasattr(control, "continuous_update"):
                control.continuous_update = live

    def render(self):
        self.status.value = "⏳ Updating… / Güncelleniyor…"
        self.status.add_class("busy")
        started = time.perf_counter()
        kwargs = {name: control.value for name, control in self.controls.items()}
        self.error = None
        self.figures = 0
        # Count everything the demonstration shows (figures, HTML, animations, text)
        # by wrapping the display publisher and stdout for this draw only.
        shell = get_ipython()
        publisher = getattr(shell, "display_pub", None) if shell is not None else None
        if publisher is not None:
            original_publish = publisher.publish

            def counting_publish(*args, **kwargs):
                self.figures += 1
                return original_publish(*args, **kwargs)
            publisher.publish = counting_publish
        stdout = sys.stdout
        original_write = stdout.write

        def counting_write(text):
            if text.strip():
                self.figures += 1
            return original_write(text)
        stdout.write = counting_write
        try:
            # The previous result stays visible until the new one arrives.
            with self.out:
                clear_output(wait=True)
                try:
                    result = self.f(**kwargs)
                    from ipywidgets.widgets.interaction import show_inline_matplotlib_plots
                    show_inline_matplotlib_plots()
                    if result is not None:
                        display(result)
                except Exception:
                    self.error = traceback.format_exc()
                    _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), self.error))
                    print(self.error)
        finally:
            if publisher is not None and publisher.__dict__.get("publish") is counting_publish:
                del publisher.publish
            if stdout.__dict__.get("write") is counting_write:
                del stdout.write
        self.last_outputs = self.figures
        self.seconds = time.perf_counter() - started
        self.updates += 1
        if self.seconds > PHYSICS_LIVE_OFF:
            self._set_live(False)
        elif self.seconds < PHYSICS_LIVE_ON:
            self._set_live(True)
        self.status.remove_class("busy")
        mode = ("updates while you drag / sürüklerken güncellenir" if self.live
                else "updates when you release the slider / kaydırıcıyı bırakınca güncellenir")
        self.status.value = (f"✓ Drawn in {self.seconds:.2f} s · {mode}" if self.error is None
                             else "⚠ The demonstration reported an error; see the message below.")


def physics_interactive(function, **controls):
    """Build a panel like ipywidgets.interactive, returning the panel object."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        return function(*args, **kwargs)

    return PhysicsPanel(checked, controls)


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel.widget
    function.panel = panel
    display(panel.widget)
    return function
```

**Small force balance:** A $2\,\mathrm{kg}$ box is pushed downward by an additional vertical $5\,\mathrm N$ while resting on a floor. Upward positive gives

$$N-mg-5\,\mathrm N=0\quad\Rightarrow\quad N=(2)(9.81)+5=24.6\,\mathrm N.$$

Adding $mg+5\,\mathrm N$ to both sides isolates $N$. It exceeds the weight because the surface supports the extra push too.

**Türkçe:** Normal kuvvetini otomatik olarak $mg$ yazma. Düşey ivme sıfır olsa bile ek aşağı kuvvet, zeminin daha büyük tepki kuvveti vermesini gerektirir.

#### Friction Model

$$\boxed{\;0\le f_s\le\mu_s N\quad\text{(static, adjusts)},\qquad f_k=\mu_k N\quad\text{(kinetic, fixed)}\;}$$

Static friction adjusts to the needed value until its limit. At equality the contact is at impending slip; a larger demand cannot be supported by static friction. During sliding use the kinetic magnitude and the sliding direction.

**Think–Pair–Explain:** A horizontal contact has $N=20\,\mathrm N$, $\mu_s=0.40$, and $\mu_k=0.30$. A resting box is pulled with $5\,\mathrm N$. What is friction? What changes for a $10\,\mathrm N$ pull?

**Model response:** The static limit is $(0.40)(20)=8\,\mathrm N$. At a $5\,\mathrm N$ pull, static friction is $5\,\mathrm N$, not $8\,\mathrm N$. A $10\,\mathrm N$ pull exceeds the limit; once sliding begins, kinetic friction is $(0.30)(20)=6\,\mathrm N$, giving net force $10-6=4\,\mathrm N$.

**Türkçe:** $\mu_sN$ her durumda sürtünmenin kendisi değildir, sağlayabileceği en büyük değerdir. Kayma başladıktan sonra farklı bir model ve $\mu_k$ kullanılır.

#### Strategy for Solving Force Problems

1. Select an object and identify its contacts.
2. Draw only the real forces acting on it; net force is their sum, not an extra arrow.
3. Choose axes; tilt them with an incline when useful.
4. Write $\sum F_x=ma_x$ and $\sum F_y=ma_y$ with signs.
5. Rearrange symbolically, substitute units, then test direction and limiting cases.


### Interactive: Inclined Plane Simulator

A block is placed on an incline. Adjust the angle, mass, and friction coefficient to see:
- Whether the block slides or stays
- The acceleration if it slides
- An animation of the block sliding down

The critical angle at which the block just begins to slide is $\theta_c = \arctan(\mu_s)$.

**Predict before you touch the controls / Kaydırıcıya dokunmadan önce tahmin et.** Write down
what you expect to happen to the curve when you change the control, and why — one sentence and, if you
can, a sign or a limiting case. Then move it and see whether you were right. Watching a graph move
without having committed to a prediction teaches almost nothing; being wrong about a prediction you wrote
down teaches a great deal.


```python
#@title Run the demonstration — predict, adjust, observe
def incline_simulation(theta_deg=30, mass=5.0, mu_s=0.4, mu_k=0.3):
    """Simulate and animate a block on an inclined plane."""
    if mass <= 0 or not 0 <= theta_deg < 90 or not 0 <= mu_k <= mu_s:
        print("Use mass > 0, 0 <= angle < 90°, and 0 <= μ_k <= μ_s for this sliding-from-rest model.")
        return
    theta = np.radians(theta_deg)
    W = mass * g
    N = W * np.cos(theta)
    W_parallel = W * np.sin(theta)
    f_s_max = mu_s * N
    f_k = mu_k * N

    # Check if block slides
    slides = W_parallel > f_s_max
    theta_critical = np.degrees(np.arctan(mu_s))

    if slides:
        a = g * (np.sin(theta) - mu_k * np.cos(theta))
        status = f"SLIDING  (a = {a:.2f} m/s²)"
    else:
        a = 0.0
        f_s_actual = W_parallel
        status = f"STATIONARY  (f_s = {f_s_actual:.1f} N ≤ f_s_max = {f_s_max:.1f} N)"

    # Info panel
    print(f"╔══════════════════════════════════════════╗")
    print(f"║  Inclined Plane Analysis                 ║")
    print(f"╠══════════════════════════════════════════╣")
    print(f"║  θ = {theta_deg}°    m = {mass} kg                  ")
    print(f"║  μ_s = {mu_s}   μ_k = {mu_k}                  ")
    print(f"║  W = {W:.1f} N                              ")
    print(f"║  N = {N:.1f} N                              ")
    print(f"║  W∥ = {W_parallel:.1f} N   W⊥ = {N:.1f} N          ")
    print(f"║  f_s(max) = {f_s_max:.1f} N   f_k = {f_k:.1f} N    ")
    print(f"║  Critical angle: θ_c = {theta_critical:.1f}°           ")
    print(f"║  Status: {status}")
    print(f"╚══════════════════════════════════════════╝")

    if not slides:
        # Just draw the static scene
        fig, ax = plt.subplots(figsize=(10, 7))
        L = 8
        ix = [0, L * np.cos(theta), L * np.cos(theta)]
        iy = [0, 0, L * np.sin(theta)]
        ax.fill(ix, iy, color='gray', alpha=0.2)
        ax.plot(ix + [0], iy + [0], 'k-', linewidth=2)

        cx = 0.5 * L * np.cos(theta)
        cy = 0.5 * L * np.sin(theta)
        bw, bh = 0.8, 0.6
        corners = np.array([[-bw/2, 0], [bw/2, 0], [bw/2, bh], [-bw/2, bh], [-bw/2, 0]])
        R_mat = np.array([[np.cos(theta), -np.sin(theta)],
                          [np.sin(theta), np.cos(theta)]])
        rotated = (R_mat @ corners.T).T + [cx, cy]
        ax.fill(rotated[:, 0], rotated[:, 1], color='steelblue', alpha=0.9)
        ax.plot(rotated[:, 0], rotated[:, 1], 'k-', linewidth=2)
        ax.text(cx, cy + 0.4, 'STATIC', ha='center', va='center', fontsize=11, fontweight='bold', color='white')

        ax.set_xlim(-1, L * np.cos(theta) + 1)
        ax.set_ylim(-1, L * np.sin(theta) + 1)
        ax.set_aspect('equal')
        ax.set_title(f'Block at rest on incline (θ = {theta_deg}° ≤ θ_c = {theta_critical:.1f}°)', fontsize=14)
        plt.tight_layout()
        plt.show()
        return

    # Animate sliding
    L = 8
    travel_length = (0.85 - 0.05) * L
    t_max = np.sqrt(2 * travel_length / a)
    t_arr = np.linspace(0, t_max, 150)
    s_arr = 0.5 * a * t_arr**2  # distance along slope

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # Left: animation scene
    ax = axes[0]
    ix = [0, L * np.cos(theta), L * np.cos(theta)]
    iy = [0, 0, L * np.sin(theta)]
    ax.fill(ix, iy, color='gray', alpha=0.2)
    ax.plot(ix + [0], iy + [0], 'k-', linewidth=2)
    ax.set_xlim(-1, L * np.cos(theta) + 2)
    ax.set_ylim(-1, L * np.sin(theta) + 1)
    ax.set_aspect('equal')
    ax.set_title('Sliding Block')

    bw, bh = 0.6, 0.5
    corners_base = np.array([[-bw/2, 0], [bw/2, 0], [bw/2, bh], [-bw/2, bh], [-bw/2, 0]])
    R_mat = np.array([[np.cos(theta), -np.sin(theta)],
                      [np.sin(theta), np.cos(theta)]])

    block_patch, = ax.fill([], [], color='steelblue', alpha=0.9)
    block_line, = ax.plot([], [], 'k-', linewidth=2)
    time_text = ax.text(0.02, 0.95, '', transform=ax.transAxes, fontsize=11,
                        verticalalignment='top',
                        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

    # Right: velocity and position vs time
    ax2 = axes[1]
    v_arr = a * t_arr
    axv = ax2.twinx()
    ax2.plot(t_arr, s_arr, 'b-', linewidth=2, label='Position s(t)')
    axv.plot(t_arr, v_arr, 'r--', linewidth=2, label='Velocity v(t)')
    pos_dot, = ax2.plot([], [], 'bo', markersize=10)
    vel_dot, = axv.plot([], [], 'rs', markersize=10)
    ax2.set_xlabel('Time (s)')
    ax2.set_ylabel('Distance s (m)', color='blue')
    axv.set_ylabel('Speed v (m/s)', color='red')
    ax2.set_title('Position & Velocity vs Time')
    ax2.legend(loc="upper left")
    axv.legend(loc="lower right")
    fig.tight_layout()

    start_pos = 0.85  # start near top of slope (fraction of L)

    def init():
        return block_patch, block_line, time_text, pos_dot, vel_dot

    def update(frame):
        # Block slides down from top
        frac = start_pos - s_arr[frame] / L
        frac = max(frac, 0.05)
        cx = frac * L * np.cos(theta)
        cy = frac * L * np.sin(theta)
        rotated = (R_mat @ corners_base.T).T + [cx, cy]
        block_patch.set_xy(rotated)
        block_line.set_data(rotated[:, 0], rotated[:, 1])
        time_text.set_text(f't = {t_arr[frame]:.2f} s\nv = {v_arr[frame]:.1f} m/s\ns = {s_arr[frame]:.1f} m')
        pos_dot.set_data([t_arr[frame]], [s_arr[frame]])
        vel_dot.set_data([t_arr[frame]], [v_arr[frame]])
        return block_patch, block_line, time_text, pos_dot, vel_dot

    anim = FuncAnimation(fig, update, init_func=init,
                         frames=physics_frames(len(t_arr)), interval=physics_interval(len(t_arr), 30), blit=True)
    plt.close(fig)
    display(physics_animation_html(anim))


# Widget interface
theta_w = widgets.IntSlider(value=35, min=5, max=75, step=1, description='θ (°):')
mass_w = widgets.FloatSlider(value=5.0, min=1.0, max=20.0, step=0.5, description='Mass (kg):')
mus_w = widgets.FloatSlider(value=0.4, min=0.0, max=1.0, step=0.05,
                            description='μ_s:', readout_format='.2f')
muk_w = widgets.FloatSlider(value=0.3, min=0.0, max=0.8, step=0.05,
                            description='μ_k:', readout_format='.2f')
run_btn = widgets.Button(description='▶ Run', button_style='success',
                         layout=widgets.Layout(width='120px'))
incline_out = widgets.Output()

def on_run(b):
    with incline_out:
        clear_output(wait=True)
        incline_simulation(theta_w.value, mass_w.value, mus_w.value, muk_w.value)

run_btn.on_click(on_run)
display(physics_panel([theta_w, mass_w, mus_w, muk_w, run_btn], incline_out))
on_run(None)
```

### Interactive: Atwood Machine Simulator

An **Atwood machine** consists of two masses connected by a string over a frictionless pulley.

For masses $m_1 > m_2$:
$$\boxed{\;a = \frac{m_1 - m_2}{m_1 + m_2}\,g \qquad T = \frac{2\,m_1 m_2}{m_1 + m_2}\,g\;}$$

The heavier mass accelerates downward while the lighter mass accelerates upward.

**Predict before you touch the controls / Kaydırıcıya dokunmadan önce tahmin et.** Write down
what you expect to happen to the curve when you change the control, and why — one sentence and, if you
can, a sign or a limiting case. Then move it and see whether you were right. Watching a graph move
without having committed to a prediction teaches almost nothing; being wrong about a prediction you wrote
down teaches a great deal.


```python
#@title Run the demonstration — predict, adjust, observe
def atwood_machine(m1=8.0, m2=5.0):
    """Animate an Atwood machine."""
    a = (m1 - m2) / (m1 + m2) * g
    T = 2 * m1 * m2 / (m1 + m2) * g

    print(f"Atwood Machine: m₁ = {m1} kg, m₂ = {m2} kg")
    print(f"  Acceleration: a = {a:.3f} m/s²")
    print(f"  Tension:      T = {T:.2f} N")
    if abs(m1 - m2) < 0.01:
        print("  System is in equilibrium (m₁ = m₂).")

    # Animation
    if m1 <= 0 or m2 <= 0:
        raise ValueError("Both masses must be positive.")
    # Stop before the rising mass reaches the pulley; the graphs use the same time.
    t_max = min(2.0, np.sqrt(2 * 1.5 / abs(a))) if a != 0 else 2.0
    n_frames = 100
    t_arr = np.linspace(0, t_max, n_frames)
    s_arr = 0.5 * abs(a) * t_arr**2  # displacement
    v_arr = abs(a) * t_arr

    fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 7),
                                   gridspec_kw={'width_ratios': [1, 1.2]})

    # --- Left panel: Atwood machine diagram ---
    ax.set_xlim(-3, 3)
    ax.set_ylim(-5, 4)
    ax.set_aspect('equal')
    ax.set_title('Atwood Machine', fontsize=14)
    ax.axis('off')

    # Pulley
    pulley = plt.Circle((0, 3), 0.5, color='gray', fill=True, linewidth=2)
    ax.add_patch(pulley)
    ax.plot([0], [3], 'k+', markersize=10, markeredgewidth=2)
    # Support
    ax.plot([-1, 1], [3.8, 3.8], 'k-', linewidth=4)
    ax.plot([0, 0], [3.5, 3.8], 'k-', linewidth=3)

    # Initial positions
    y1_init = 0.0
    y2_init = 0.0

    # Strings
    string1, = ax.plot([], [], 'k-', linewidth=2)
    string2, = ax.plot([], [], 'k-', linewidth=2)

    # Mass blocks
    block1 = patches.FancyBboxPatch((-2.2, y1_init), 1.4, 1,
                                     boxstyle="round,pad=0.05",
                                     facecolor='crimson', edgecolor='black', linewidth=2)
    block2 = patches.FancyBboxPatch((0.8, y2_init), 1.4, 1,
                                     boxstyle="round,pad=0.05",
                                     facecolor='royalblue', edgecolor='black', linewidth=2)
    ax.add_patch(block1)
    ax.add_patch(block2)

    label1 = ax.text(-1.5, y1_init + 0.5, f'm₁\n{m1}kg', ha='center', va='center',
                     fontsize=11, color='white', fontweight='bold')
    label2 = ax.text(1.5, y2_init + 0.5, f'm₂\n{m2}kg', ha='center', va='center',
                     fontsize=11, color='white', fontweight='bold')

    info = fig.text(0.5, 0.035, '', ha='center', fontsize=10,
                   bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))

    # --- Right panel: graphs ---
    ax2.set_xlabel('Time (s)')
    ax2.set_ylabel('Displacement magnitude (m)', color='blue')
    axv = ax2.twinx()
    axv.set_ylabel('Speed (m/s)', color='red')
    ax2.set_title('Displacement & Velocity vs Time')
    ax2.plot(t_arr, s_arr, 'b-', linewidth=2, label='|displacement|')
    axv.plot(t_arr, v_arr, 'r--', linewidth=2, label='|velocity|')
    s_dot, = ax2.plot([], [], 'bo', markersize=8)
    v_dot, = axv.plot([], [], 'rs', markersize=8)
    ax2.legend(loc="upper left")
    axv.legend(loc="lower right")
    fig.tight_layout(rect=[0, 0.13, 1, 1])

    def init():
        return block1, block2, string1, string2, label1, label2, info, s_dot, v_dot

    def update(frame):
        disp = s_arr[frame]
        sign = 1 if m1 > m2 else -1

        y1 = y1_init - sign * disp
        y2 = y2_init + sign * disp

        block1.set_y(y1)
        block2.set_y(y2)
        label1.set_position((-1.5, y1 + 0.5))
        label2.set_position((1.5, y2 + 0.5))

        # Strings from pulley to top of blocks
        string1.set_data([-0.5, -1.5, -1.5], [3, 3, y1 + 1])
        string2.set_data([0.5, 1.5, 1.5], [3, 3, y2 + 1])

        info.set_text(f't = {t_arr[frame]:.2f} s   |s| = {disp:.2f} m   |v| = {v_arr[frame]:.2f} m/s\n'
                      f'a = {a:.3f} m/s²   T = {T:.1f} N')

        s_dot.set_data([t_arr[frame]], [s_arr[frame]])
        v_dot.set_data([t_arr[frame]], [v_arr[frame]])
        return block1, block2, string1, string2, label1, label2, info, s_dot, v_dot

    anim = FuncAnimation(fig, update, init_func=init,
                         frames=physics_frames(n_frames), interval=physics_interval(n_frames, 40), blit=True)
    plt.close(fig)
    display(physics_animation_html(anim))


m1_w = widgets.FloatSlider(value=8, min=1, max=20, step=0.5, description='m₁ (kg):')
m2_w = widgets.FloatSlider(value=5, min=1, max=20, step=0.5, description='m₂ (kg):')
atwood_btn = widgets.Button(description='▶ Simulate', button_style='success',
                            layout=widgets.Layout(width='140px'))
atwood_out = widgets.Output()

def on_atwood(b):
    with atwood_out:
        clear_output(wait=True)
        atwood_machine(m1_w.value, m2_w.value)

atwood_btn.on_click(on_atwood)
display(physics_panel([m1_w, m2_w, atwood_btn], atwood_out))
on_atwood(None)
```

### Interactive: Friction Force Visualizer

This demo shows how friction behaves as you gradually increase an applied force:
- **Static regime:** friction matches the applied force (no motion)
- **Transition:** at $F=\mu_sN$, the block is at impending slip; exceeding the limit starts sliding in this ideal model
- **Kinetic regime:** friction drops to $\mu_k N$ and stays constant

**Predict before you touch the controls / Kaydırıcıya dokunmadan önce tahmin et.** Write down
what you expect to happen to the curve when you change the control, and why — one sentence and, if you
can, a sign or a limiting case. Then move it and see whether you were right. Watching a graph move
without having committed to a prediction teaches almost nothing; being wrong about a prediction you wrote
down teaches a great deal.


```python
#@title Run the demonstration — predict, adjust, observe
def friction_visualizer(mass=5.0, mu_s=0.5, mu_k=0.35):
    """Show friction force vs applied force."""
    N = mass * g
    f_s_max = mu_s * N
    f_k = mu_k * N

    F_applied = np.linspace(0, max(f_s_max * 1.8, f_k * 1.8, 1.0), 500)
    f_friction = np.where(F_applied <= f_s_max, F_applied, f_k)
    a_net = np.where(F_applied <= f_s_max, 0, (F_applied - f_k) / mass)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.6, 4.2))
    fig.subplots_adjust(left=0.07, right=0.98, bottom=0.29, top=0.83, wspace=0.30)
    fig.suptitle(rf'$m={mass:.1f}\,\mathrm{{kg}},\quad N=mg={N:.2f}\,\mathrm{{N}},\quad '
                 rf'\mu_s={mu_s:.2f},\quad \mu_k={mu_k:.2f}$', fontsize=11)

    # Left: friction vs applied
    ax1.plot(F_applied, f_friction, 'b-', linewidth=2.5, label='Friction force')
    ax1.plot(F_applied, F_applied, 'k--', linewidth=1, alpha=0.4, label='f = F (reference)')
    ax1.axhline(f_s_max, color='red', linestyle=':', alpha=0.7, label=f'f_s(max) = {f_s_max:.1f} N')
    ax1.axhline(f_k, color='orange', linestyle=':', alpha=0.7, label=f'f_k = {f_k:.1f} N')
    ax1.axvline(f_s_max, color='gray', linestyle='--', alpha=0.3)

    # Shade regions
    ax1.axvspan(0, f_s_max, alpha=0.08, color='green', label='Static region')
    ax1.axvspan(f_s_max, F_applied[-1], alpha=0.08, color='red', label='Kinetic region')

    ax1.set_xlabel('Applied Force (N)', fontsize=12)
    ax1.set_ylabel('Friction Force (N)', fontsize=12)
    ax1.set_title('Friction vs Applied Force', fontsize=14)
    ax1.set_xlim(0, F_applied[-1])
    ax1.set_ylim(0, max(f_s_max * 1.5, f_k * 1.5, 1.0))

    # Right: acceleration vs applied
    ax2.plot(F_applied, a_net, 'r-', linewidth=2.5)
    ax2.axvline(f_s_max, color='gray', linestyle='--', alpha=0.3)
    ax2.axvspan(0, f_s_max, alpha=0.08, color='green')
    ax2.axvspan(f_s_max, F_applied[-1], alpha=0.08, color='red')
    ax2.set_xlabel('Applied Force (N)', fontsize=12)
    ax2.set_ylabel('Acceleration (m/s²)', fontsize=12)
    ax2.set_title('Acceleration vs Applied Force\nStatic: a = 0 | Kinetic: sliding', fontsize=11)

    # A figure-level legend has its own space, below both x-axis labels.
    handles, labels = ax1.get_legend_handles_labels()
    fig.legend(handles, labels, loc='lower center', bbox_to_anchor=(0.5, 0.02),
               ncol=3, fontsize=9, frameon=False)
    for ax in (ax1, ax2):
        ax.tick_params(labelsize=9)
        ax.xaxis.label.set_size(10)
        ax.yaxis.label.set_size(10)
        ax.title.set_fontsize(11)
    plt.show()


@physics_interact(
    mass=widgets.FloatSlider(value=5, min=1, max=20, step=0.5, description='Mass (kg):'),
    mu_s=widgets.FloatSlider(value=0.5, min=0.1, max=1.0, step=0.05,
                             description='μ_s:', readout_format='.2f'),
    mu_k=widgets.FloatSlider(value=0.35, min=0.05, max=0.9, step=0.05,
                             description='μ_k:', readout_format='.2f')
)
def _friction_viz(mass, mu_s, mu_k):
    if mu_k > mu_s:
        print("⚠️  μ_k should be ≤ μ_s. Adjusting μ_k = μ_s.")
        mu_k = mu_s
    friction_visualizer(mass, mu_s, mu_k)
```

### Friction parameter sweep on an incline (static plot)

How does the critical angle change with the friction coefficient? This plot sweeps $\mu_s$ from 0 to 1 and shows the corresponding critical angle $\theta_c = \arctan(\mu_s)$.

**Predict before you run the cell / Hücreyi çalıştırmadan önce tahmin et.** Sketch $\theta_c$ against $\mu_s$ from 0 to 1: what is $\theta_c$ at $\mu_s=0$, at $\mu_s=1$, and does the curve bend upward or downward? Then run the cell and compare.


```python
#@title Optional numerical check — the algebra is explained above
def friction_sweep():
    """Plot critical angle vs friction coefficient."""
    mu_vals = np.linspace(0, 1.5, 300)
    theta_c = np.degrees(np.arctan(mu_vals))

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(mu_vals, theta_c, 'b-', linewidth=2.5)
    ax.fill_between(mu_vals, theta_c, alpha=0.1, color='green', label='Block stays (θ < θ_c)')
    ax.fill_between(mu_vals, theta_c, 90, alpha=0.1, color='red', label='Block slides (θ > θ_c)')

    # Mark common materials
    materials = {
        'Ice on ice': 0.03,
        'Wood on wood': 0.4,
        'Rubber on concrete': 0.8,
        'Steel on steel': 0.6,
    }
    for name, mu in materials.items():
        tc = np.degrees(np.arctan(mu))
        ax.plot(mu, tc, 'ko', markersize=8)
        ax.annotate(f'{name}\nμ={mu}, θ_c={tc:.0f}°',
                    (mu, tc), textcoords='offset points',
                    xytext=(10, 8), fontsize=9, arrowprops=dict(arrowstyle='->', color='gray'))

    ax.set_xlabel('Static friction coefficient μ_s', fontsize=12)
    ax.set_ylabel('Critical angle θ_c (°)', fontsize=12)
    ax.set_title('Critical Angle vs Friction Coefficient', fontsize=14)
    ax.legend(fontsize=11)
    ax.set_xlim(0, 1.5)
    ax.set_ylim(0, 70)
    plt.tight_layout()
    plt.show()

friction_sweep()
```

#### Example 1: Block on a Flat Surface

A $10\,\mathrm{kg}$ block on a horizontal surface ($\mu_k=0.25$) is pushed with $40\,\mathrm N$. Find the acceleration.

#### Hand calculation and the missing assumption

Assume the block is already sliding in the direction of the horizontal push. Only $\mu_k$ is supplied, so we cannot determine whether a resting block starts moving.

Choose forward and upward positive. Vertical acceleration is zero. First find normal force $N$, then kinetic-friction magnitude $f_k$:

$$\begin{aligned}
N-mg&=0,\\
N&=mg=(10\,\mathrm{kg})(9.81\,\mathrm{m/s^2})=98.1\,\mathrm N,\\
f_k&=\mu_kN=(0.25)(98.1\,\mathrm N)=24.5\,\mathrm N.
\end{aligned}$$

Friction points backward. Write its minus sign in the horizontal sum, then divide both sides by mass:

$$\begin{aligned}
F-f_k&=ma,\\
40\,\mathrm N-24.5\,\mathrm N&=(10\,\mathrm{kg})a,\\
a&=\frac{15.5\,\mathrm N}{10\,\mathrm{kg}}=1.55\,\mathrm{m/s^2}.
\end{aligned}$$

**Result:** Acceleration is $1.55\,\mathrm{m/s^2}$ forward under the sliding assumption.

**Check:** It is smaller than the frictionless value $F/m=4\,\mathrm{m/s^2}$. The net force, not the applied force alone, equals $ma$.

**Türkçe:** Önce düşey dengeyle normal kuvveti bul; sürtünme bu kuvvete bağlıdır. Ardından yatay kuvvetleri işaretleriyle topla. Kinetik katsayı kayma hesabına yeter, fakat başlangıçta durgun kutunun harekete geçip geçmeyeceğini belirlemez.


```python
#@title Optional numerical check — the algebra is explained above
m_ex1 = 10  # kg
F_app = 40  # N
mu_k_ex1 = 0.25

N_ex1 = m_ex1 * g
f_k_ex1 = mu_k_ex1 * N_ex1
a_ex1 = (F_app - f_k_ex1) / m_ex1

print("Block on flat surface:")
print(f"  Normal force  : N  = mg = {N_ex1:.1f} N")
print(f"  Kinetic friction: f_k = μ_k × N = {f_k_ex1:.2f} N")
print(f"  Net force     : F_net = F - f_k = {F_app - f_k_ex1:.2f} N")
print(f"  Acceleration  : a = F_net / m = {a_ex1:.2f} m/s²")
```

#### Example 2: Block on an Incline

A $5\,\mathrm{kg}$ block is placed on a $30^\circ$ incline with $\mu_s=0.4$ and $\mu_k=0.3$. Does it slide? If yes, what is its acceleration?

#### Hand calculation: first decide whether it moves

Choose downhill positive. Let $W_{\parallel}$ be the downhill weight component. There is no acceleration perpendicular to the plane:

$$\begin{aligned}
N&=mg\cos30^\circ=(5)(9.81)\cos30^\circ=42.5\,\mathrm N,\\
W_{\parallel}&=mg\sin30^\circ=(5)(9.81)\sin30^\circ=24.5\,\mathrm N,\\
f_{s,\max}&=\mu_sN=(0.4)(42.5)=17.0\,\mathrm N.
\end{aligned}$$

The needed uphill friction exceeds the static limit: $24.5\,\mathrm N>17.0\,\mathrm N$. Therefore **the block slides downhill**.

After sliding starts, replace static friction by kinetic friction. Keep its uphill direction as a minus sign:

$$\begin{aligned}
f_k&=\mu_kN=(0.3)(42.5)=12.7\,\mathrm N,\\
W_{\parallel}-f_k&=ma,\\
a&=\frac{24.5\,\mathrm N-12.7\,\mathrm N}{5\,\mathrm{kg}}=2.36\,\mathrm{m/s^2}.
\end{aligned}$$

The same result follows symbolically by factoring and dividing by nonzero mass:

$$m\bigl(g\sin\theta-\mu_kg\cos\theta\bigr)=ma$$

$$\boxed{\;a=g(\sin\theta-\mu_k\cos\theta)\;}$$

**Check:** The acceleration lies between zero and the frictionless value $g\sin30^\circ=4.91\,\mathrm{m/s^2}$. The critical static angle satisfies $\tan\theta_c=\mu_s$, and $30^\circ>\tan^{-1}(0.4)$.

**Türkçe:** Kayma kararını kuvvetlerin büyüklüklerini karşılaştırarak ver. Kararda $\mu_s$, hareket başladıktan sonraki ivmede $\mu_k$ kullanılır. Son bölümde her terimde aynı kütle bulunduğu için denklemin iki tarafını kütleye bölüyoruz.


```python
#@title Optional numerical check — the algebra is explained above
m_ex2 = 5  # kg
theta_ex2 = 30  # degrees
mu_s_ex2 = 0.4
mu_k_ex2 = 0.3
theta_r = np.radians(theta_ex2)

W_ex2 = m_ex2 * g
N_ex2 = W_ex2 * np.cos(theta_r)
W_par = W_ex2 * np.sin(theta_r)
f_s_max = mu_s_ex2 * N_ex2

print(f"Block on {theta_ex2}° incline:")
print(f"  Weight parallel to slope: W∥ = {W_par:.2f} N")
print(f"  Max static friction:      f_s(max) = {f_s_max:.2f} N")

if W_par > f_s_max:
    f_k_ex2 = mu_k_ex2 * N_ex2
    a_ex2 = (W_par - f_k_ex2) / m_ex2
    print(f"  W∥ > f_s(max)  →  Block SLIDES!")
    print(f"  Kinetic friction: f_k = {f_k_ex2:.2f} N")
    print(f"  Acceleration: a = {a_ex2:.2f} m/s²")
else:
    print(f"  W∥ ≤ f_s(max)  →  Block STAYS PUT")
    print(f"  Static friction adjusts to f_s = {W_par:.2f} N")
```

#### Example 3: Atwood Machine

Two masses $m_1=7\,\mathrm{kg}$ and $m_2=3\,\mathrm{kg}$ are connected by a string over a frictionless pulley. Find acceleration and tension.

#### Hand calculation: add two force equations

Assume a light, inextensible string and an ideal pulley. Take downward positive for $m_1$ and upward positive for $m_2$. Their acceleration magnitudes are the same, denoted $a$; the string tension is $T$ on each side.

$$\begin{aligned}
m_1g-T&=m_1a &&\text{heavier mass},\\
T-m_2g&=m_2a &&\text{lighter mass}.
\end{aligned}$$

Add the equations. The opposite tension terms cancel:

$$\begin{aligned}
(m_1-m_2)g&=(m_1+m_2)a,\\
a&=\frac{m_1-m_2}{m_1+m_2}g,\\
 &=\frac{7-3}{7+3}(9.81\,\mathrm{m/s^2})=3.92\,\mathrm{m/s^2}.
\end{aligned}$$

To find tension, use the lighter mass and add its weight to both sides:

$$\begin{aligned}
T-m_2g&=m_2a,\\
T&=m_2(g+a),\\
 &=(3\,\mathrm{kg})(9.81+3.92)\,\mathrm{m/s^2}=41.2\,\mathrm N.
\end{aligned}$$

**Check:** The heavier mass gives the same answer, $T=m_1(g-a)=(7)(9.81-3.92)=41.2\,\mathrm N$. Acceleration is less than $g$; tension lies between the two weights.

**Türkçe:** İki kütle için pozitif yönleri farklı seçiyoruz, çünkü biri aşağı inerken diğeri yukarı çıkar. Bu seçimle aynı pozitif $a$ büyüklüğü kullanılabilir. Denklemleri toplarken gerilme yok olur; gerilmeyi daha sonra tek kütlenin denkleminden geri buluruz.


```python
#@title Optional numerical check — the algebra is explained above
m1_ex3 = 7  # kg
m2_ex3 = 3  # kg

a_ex3 = (m1_ex3 - m2_ex3) / (m1_ex3 + m2_ex3) * g
T_ex3 = 2 * m1_ex3 * m2_ex3 / (m1_ex3 + m2_ex3) * g

print(f"Atwood machine: m₁ = {m1_ex3} kg, m₂ = {m2_ex3} kg")
print(f"  a = (m₁ - m₂)/(m₁ + m₂) × g = {a_ex3:.3f} m/s²")
print(f"  T = 2m₁m₂/(m₁ + m₂) × g = {T_ex3:.2f} N")
print(f"")
print(f"  Check: T < m₁g = {m1_ex3*g:.1f} N  ✓  (m₁ accelerates down)")
print(f"  Check: T > m₂g = {m2_ex3*g:.1f} N  ✓  (m₂ accelerates up)")
```

## 5. Exit check / Çıkış kontrolü

Explain which coefficient decides whether sliding starts and which coefficient predicts acceleration after sliding.

Write one sentence naming the physical principle and one line of algebra you can now explain. **TR:** Sonuca nasıl ulaştığını bir cümleyle anlat; emin olmadığın ilk adımı işaretle.

<details>
<summary>Model responses / Örnek yanıtlar — open after trying</summary>

**Entry:** Resolve the actual forces along the chosen axes, add them with signs, and divide their sum by mass. For a ramp tilted by $\theta$, the weight components have magnitudes $mg\sin\theta$ parallel and $mg\cos\theta$ perpendicular to it.

**Exit:** Compare the needed static friction with $\mu_sN$ to decide whether rest is possible. Once the object slides, use $f_k=\mu_kN$ opposite the relative sliding direction to find its acceleration. Do not automatically set static friction equal to its maximum.

**Visual prediction:** On the horizontal surface with no other vertical forces, $N=mg$ and the threshold is $F_{\rm threshold}=\mu_smg$. Doubling mass doubles this threshold. Below it, $f_s=F_{\rm applied}$ and net horizontal force is zero.

</details>

### How this week is worked / Bu hafta nasıl çalışılır

**Every problem, worked example and experiment in this course follows the same six steps.** Explain the physical meaning of the chosen law and operation before using it.

> **draw → choose a law → rearrange to the symbolic answer → check a limiting case → substitute with units → interpret**

1. **Symbolic answer first.** Isolate the unknown in symbols, before any number appears.
2. **One limiting case.** Push a parameter to a value whose answer you already know — a mass to zero, an
   angle to $90^\circ$, a friction coefficient to zero — and confirm the formula behaves sensibly. If it
   does not, the algebra is wrong and you have just caught it **for free**, before wasting arithmetic on it.
3. **Then numbers**, with units carried through every line.

Why bother: the symbolic route is *less* arithmetic, not more, and it is where the marks are. A correct
symbolic answer with a slip in the final multiplication is nearly full credit; a correct number with no
derivation is not.

**TR:** Her problemde önce sembolik sonuç, sonra bir sınır durumu kontrolü, sonra sayılar. Sınır kontrolü
cebir hatasını aritmetiğe girmeden, bedavaya yakalar. Doğru sembolik çözüm artı son çarpımda sürçme
neredeyse tam puandır; türetmesi olmayan doğru sayı değildir.

## 1. Fully solved review routes / Çözümlü tekrar örnekleri

### Review week: reconnect the first five teaching weeks

No new topic is introduced this week. Use the 29 October holiday week for question solving and repair of the foundations already taught: units/vectors, 1D/2D motion, Newton's laws/friction, work and kinetic energy. Potential energy and its conservation account begin after the midterm.

Start by rating each skill **can explain / need a hint / need help**: unit conversion; choosing signs; velocity components; drawing one object's forces; rearranging an equation; checking work signs. Choose two weak skills, then use the worked review activities below. **TR:** Amaç yeni formül eklemek değil, bildiğimiz konuları birbirine bağlamak ve eksik adımı bulmak.

**Objectives / Hedefler.** By the end of the session you can (1) turn a short engineering situation into a labelled diagram and one governing equation from Weeks 01–05; (2) reach a symbolic answer, test a limiting case, then substitute with units; (3) mark your own exam-style solution against a scheme; (4) check your work against the eight recurring midterm errors.


### Choose an operation from the physical question

Before selecting an equation, annotate the diagram with the quantity being sought and its type. A force balance combines vectors; work projects a force along displacement; acceleration is a rate of velocity change. The same object can need different operations for different questions.

**Try without numbers.** A motor pulls a carriage on a straight horizontal rail with a constant angled cable force. Draw a free-body diagram and the displacement arrow. Choose a route for (a) acceleration, (b) speed after a stated distance, (c) average motor power over a stated time. State one assumption for each.

<details><summary>Compare three routes</summary>

(a) Add force components and use $\sum F_x=ma_x$; if friction matters, first find the normal force. (b) Add each force's work and use $W_{\rm net}=\Delta K$; energy determines speed, not its direction. (c) Divide the motor's work by elapsed time; net work and motor work need not be equal. The cable angle belongs in its projection onto the displacement. If you double the path at the same constant force, cable work doubles; speed generally does not.

</details>

Use the following worked review routes to practise the same translation. After checking a result, change one assumption—drag, contact, constant acceleration or fixed direction—and identify which equation needs reconsideration. This is a review of Weeks 01–05, not a new topic.

**TR:** Formülün görünüşünden değil, fiziksel sorudan başla: bileşke mi, doğrultudaki etki mi, değişim hızı mı?

### Review A — motion plus force, with a complete worked route

A 2.0 kg cart begins at rest on a horizontal track. A 10 N force pulls right while a 2 N friction force acts left for 3.0 s. Assume the cart is already in the sliding regime once motion begins and these forces stay constant.

1. Draw the cart: right pull, left friction, upward normal, downward weight. Choose right positive.
2. Net horizontal force is $10-2=8$ N. From $F_{\rm net}=ma$, divide by 2.0 kg: $a=4.0$ m/s².
3. $v=0+at=4.0(3.0)=12.0$ m/s and $d=0+at^2/2=4.0(9.0)/2=18.0$ m.
4. Check the same motion with work: $W_{\rm net}=F_{\rm net}d=8(18)=144$ J. The change in kinetic energy is $\Delta K=\tfrac12(2)(12^2)-0=144$ J.

**Explain:** Why is $10/2$ the wrong acceleration? Why is friction work negative even while speed increases? **Answer:** acceleration uses the net 8 N, and the negative friction work is smaller than the positive pulling work. **TR:** Aynı durumu iki yöntemle kontrol etmek, yalnızca son sayıyı kontrol etmekten daha değerlidir.

### Review B — one clock for two directions

A ball leaves a 1.25 m high table horizontally at 3.0 m/s. Ignore drag. Set upward positive and floor height to zero.
$$0=1.25-\tfrac12gt^2\Rightarrow t^2=\frac{2(1.25)}{9.81}
\Rightarrow t=0.505\,\mathrm s.$$
Horizontal range is $x=v_xt=3.0(0.505)=1.51$ m. Just before impact $v_y=-gt=-4.95$ m/s. Speed is
$$v=\sqrt{v_x^2+v_y^2}=\sqrt{3.0^2+4.95^2}=5.79\,\mathrm{m/s}.$$
The vertical force is $mg$ down throughout the flight; horizontal net force is zero. At release the vertical speed is zero but acceleration is already downward.

**Try then check:** If horizontal launch speed doubles, time stays 0.505 s and range doubles to 3.03 m. **TR:** Yatay hız iki katına çıkınca düşey denklem değişmez; ortak süre değişmeden yatay yol artar.


### Review C — repair a tempting wrong solution

A 1000 kg vehicle slows from 72 km/h to 36 km/h. A learner writes “energy halves because speed halves.” Correct the reasoning.

Convert the speeds first: $72/3.6=20$ m/s and $36/3.6=10$ m/s.
$$K_i=\tfrac12(1000)(20^2)=200000\,\mathrm J,
\quad K_f=\tfrac12(1000)(10^2)=50000\,\mathrm J.$$
$$W_{\rm net}=K_f-K_i=-150000\,\mathrm J.$$
Energy becomes one quarter, because speed is squared. If the net retarding force is constant at 3000 N, take forward positive and write $W=-Fd$:
$$-150000=-3000d\Rightarrow d=50.0\,\mathrm m.$$
**Check:** $a=-3000/1000=-3.0$ m/s², and $10^2=20^2+2(-3)d$ also gives 50.0 m. **TR:** Eksi işaretini kaybetme; fren kuvveti hareket yönüne terstir.

### Before you open any answer / Yanıtları açmadan önce

Each problem shows its **answer** so you can check yourself. The **full worked solution** is in the
separate solutions collection and opens on the date given under each problem.

**Write your own attempt first — symbolic answer, limiting check, then numbers — and only then open the
answer.** Opening first turns a problem into a worked example, and worked examples do not build the skill
the exam tests. If you have already opened it, that problem is spent; do a different one.

Levels: **L1 core** is the lesson's route and is enough for the week. **L2** combines two ideas. **L3** is
engineering-style, multi-step, and closest to the harder common-exam questions.

**TR:** Her problemin **yanıtı** görünür, tam çözümü gecikmeli açılır. Önce kendi denemeni yaz — sembolik
sonuç, sınır kontrolü, sonra sayılar — ancak ondan sonra yanıtı aç. Önce açarsan problem çözümlü örneğe
dönüşür ve sınavın ölçtüğü beceriyi kazandırmaz. Yanıtı açtıysan o problem tükenmiştir; başka birini çöz.


**Explain your operation.** Alongside the first equation, write one sentence connecting it to your diagram. After solving, change one parameter or assumption and predict what would change; use the worked lesson examples if you need a model. This is a study routine for the existing questions, not an extra examinable topic.


### Midterm format and how marks are given / Sınav biçimi

**What a common-exam question looks like.** A small situation in three or four sentences, with a sketch or the instruction to draw one, and two to four parts that build on each other: a diagram or a component, then a symbolic result, then a number, then a check. The context is new; every move was rehearsed in Weeks 01–05: convert a unit, resolve a vector, write one kinematic equation, draw one body's forces, apply $\sum \vec F = m\vec a$, add the works and compare with $\Delta K$. *"Find $a$ in terms of $m$, $\theta$, $\mu_k$ and $g$; then evaluate it"* is typical: *in terms of* means the symbolic answer is a marked step of its own.

**How partial credit works in this section.** Marks sit on steps, not on the final number: diagram, governing equation, symbolic result, limiting-case check, substitution with units, interpretation.

- A **correct symbolic result with a correct limiting case** earns most of the marks for a part even if the final multiplication slips: about one mark in ten is lost.
- A **bare number** with no equation earns little even when it is right; a derived number and a guessed one look the same.
- A **wrong formula with correct arithmetic** loses the formula and answer marks but keeps the diagram marks and any later part that uses the wrong result consistently (*error carried forward*).
- **Units, signs and significant figures** are marked once per part: an unexplained negative acceleration on a body that is speeding up loses the interpretation mark; $3.2058$ loses the significant-figure mark.

**TR:** Puan adımlara verilir: sınır kontrolü yapılmış doğru sembolik sonuç, son çarpımda sürçse bile puanın çoğunu alır; türetmesiz çıplak sayı az puan alır.

**Before handing in — six lines / Teslimden önce altı satır.** In the last five minutes, on every question:

1. **Units** — every answer carries a unit; km/h, cm and minutes were converted to SI before the first equation.
2. **Signs** — each sign matches the axis drawn; a surprising sign is explained or fixed.
3. **Limiting case** — one parameter pushed to $0$, $90^\circ$ or $\infty$; the check is written, not just thought.
4. **Significant figures** — three in the final answer; full precision in between.
5. **Diagram labelled** — forces, axes and the positive direction on the sketch, in the equations' symbols.
6. **Answer boxed** — each part's final result boxed with its unit.

**TR:** 1 birimler, 2 işaretler, 3 yazılı sınır durumu, 4 üç anlamlı basamak, 5 etiketli şekil, 6 kutulanmış sonuç.


### Two sample midterm questions with a marking scheme / Örnek sorular

Sit each question on paper in **20 minutes** before opening its scheme, then mark yourself line by line; each scheme is out of 20. **TR:** Her soruyu 20 dakikada, anahtarı açmadan çöz; sonra kendini puanla.

#### Sample question 1 — Crate pulled up a rough loading ramp / Örnek soru 1

A $30.0\ \mathrm{kg}$ crate is pulled up a ramp inclined at $\theta = 25.0^\circ$ by a rope that makes an angle $\phi = 15.0^\circ$ with the ramp surface and carries a constant tension $T = 220\ \mathrm N$. The coefficient of kinetic friction is $\mu_k = 0.250$. The crate starts from rest and moves up the ramp.

**(a)** Draw the free-body diagram of the crate with axes along and perpendicular to the ramp; mark the positive direction. *(4 marks)*

**(b)** Find the acceleration in terms of $T$, $\phi$, $\theta$, $\mu_k$, $m$ and $g$. Check one limiting case, then evaluate. *(9 marks)*

**(c)** The crate moves $d = 3.00\ \mathrm m$ up the ramp. Find the work done by each force and, from the work–kinetic energy theorem, the crate's speed there. Show that it agrees with (b). *(7 marks)*

**Predict first:** Is $N$ larger or smaller than $mg\cos\theta$? Which work is zero?

**TR:** (a) Serbest cisim diyagramı, (b) ivme: sembolik, sınır durumu, sayı, (c) $3.00\ \mathrm m$ boyunca işler ve hız.

<details><summary>Marking scheme / Puanlama</summary>

**(a) Free-body diagram — 4 marks.** $m\vec g$ down (1); $\vec N$ perpendicular to the ramp (1); $\vec T$ at $\phi$ above the surface and $\vec f_k$ down the ramp (1); axes with up-the-ramp positive (1).

**(b) Acceleration — 9 marks.** Perpendicular balance including the rope's lift, $N + T\sin\phi - mg\cos\theta = 0$ (2), so $N = mg\cos\theta - T\sin\phi$ (1) and $f_k = \mu_k\,(mg\cos\theta - T\sin\phi)$ (1). Along the ramp, $T\cos\phi - f_k - mg\sin\theta = ma$ (2):
$$a = \frac{T\cos\phi - \mu_k\,(mg\cos\theta - T\sin\phi) - mg\sin\theta}{m} \qquad (1)$$
Limiting case (1): $\phi \to 0$, $\mu_k \to 0$ gives $a = T/m - g\sin\theta$, the frictionless incline; $\theta \to 0$, $\phi \to 0$ gives $(T - \mu_k mg)/m$, a flat floor. Numbers (1): $mg\cos\theta = 267\ \mathrm N$, $T\sin\phi = 56.9\ \mathrm N$, $N = 210\ \mathrm N$, $f_k = 52.4\ \mathrm N$, $T\cos\phi = 213\ \mathrm N$, $mg\sin\theta = 124\ \mathrm N$; net $35.7\ \mathrm N$, so $a = 35.7/30.0 = 1.19\ \mathrm{m/s^2}$ up the ramp.

**(c) Works and speed — 7 marks.** $W_T = T\cos\phi\,d = 638\ \mathrm J$ (1). $W_N = 0$, since $\vec N \perp$ displacement (1). $W_f = -f_k d = -157\ \mathrm J$, negative because friction opposes the displacement (2). $W_g = -mg\sin\theta\,d = -373\ \mathrm J$ (1). $W_{\rm net} = 107\ \mathrm J = \tfrac12 m v^2$ gives $v = 2.67\ \mathrm{m/s}$, and $v^2 = 2ad$ gives the same (2).

**Where marks were lost in past sittings.** $N = mg\cos\theta$ with the rope's lift ignored (2 of 9 in (b), carried into $f_k$); $\sin\theta$ and $\cos\theta$ swapped; $W_f$ positive "because the crate speeds up"; $W_T = Td$ without $\cos\phi$; $a = 1.189352$.

</details>


#### Sample question 2 — Ball launched from a moving test cart / Örnek soru 2

A test cart moves along a horizontal rail at constant $u = 2.00\ \mathrm{m/s}$. A launcher on it fires a ball of mass $m = 0.150\ \mathrm{kg}$ at $v' = 6.00\ \mathrm{m/s}$ **relative to the cart**, at $\alpha = 60.0^\circ$ above the horizontal in the direction of travel. The ball is launched from, and caught at, rail height. Ignore air resistance.

**(a)** Find the ball's velocity relative to the ground at launch: components, magnitude, direction. *(4 marks)*

**(b)** Find, in symbols and numbers, the time of flight, the maximum height, the distance travelled along the rail, and how far ahead of the cart the ball lands. *(8 marks)*

**(c)** Using the work–kinetic energy theorem, find the work done by gravity from launch to the top, and the kinetic energy when the ball is caught. State the launch kinetic energy measured on the cart and explain in one sentence why it differs without contradiction. *(8 marks)*

**Predict first:** Does the ball land ahead of the cart, behind it, or on it? Does the time of flight depend on $u$?

**TR:** (a) Yere göre ilk hız, (b) uçuş süresi, yükseklik, menzil, arabaya göre düşüş yeri, (c) kütle çekiminin işi ve iki sistemdeki kinetik enerjiler.

<details><summary>Marking scheme / Puanlama</summary>

**(a) Ground velocity — 4 marks.** $\vec v = \vec u + \vec v\,'$ written or drawn (1). $v_x = u + v'\cos\alpha = 5.00\ \mathrm{m/s}$, $v_y = v'\sin\alpha = 5.20\ \mathrm{m/s}$ (2). $v = 7.21\ \mathrm{m/s}$ at $46.1^\circ$, flatter than $60^\circ$ because the cart adds horizontal velocity only (1).

**(b) Flight — 8 marks.** Up positive: $0 = v_y t - \tfrac12 g t^2$, so $t = 2v_y/g$ (1); no $u$ appears, and $v_y \to 0$ gives $t \to 0$ (1). $t = 2(5.196)/9.81 = 1.06\ \mathrm s$ (1). $H = v_y^2/2g = 1.38\ \mathrm m$ (1). Along the rail $R = v_x t = 5.30\ \mathrm m$ (1). Relative to the cart the horizontal velocity is $3.00\ \mathrm{m/s}$, so the ball lands $(v_x - u)\,t = 3.18\ \mathrm m$ **ahead** of the cart (2); in the cart frame this is the same problem with $u = 0$ (1).

**(c) Energy — 8 marks.** Launch to top: $W_g = -mgH = -\tfrac12 m v_y^2 = -2.03\ \mathrm J$, equivalently $\Delta K = \tfrac12 m v_x^2 - \tfrac12 m v^2$ (2); $v_y \to 0$ gives $W_g \to 0$ (1). Launch to catch: same height, $W_g = 0$, so $K_{\rm catch} = \tfrac12 m v^2 = \tfrac12(0.150)(7.211)^2 = 3.90\ \mathrm J$ (2). On the cart, $K' = \tfrac12 m v'^2 = 2.70\ \mathrm J$ (1). Explanation (2): kinetic energy depends on the frame because velocity does; the theorem holds in each frame with its own velocities, and gravity does the same $-mgH$ in both.

**Where marks were lost in past sittings.** $6.00\ \mathrm{m/s}$ at $60^\circ$ used as the ground velocity; $u$ added to the vertical component; $t$ solved with $7.21$ in place of $v_y$; $W_g = +mgH$ on the upward leg; the rail range $5.30\ \mathrm m$ reported as the distance from the cart.

</details>

#### P1 — Pump lifting water to a header tank  ·  L2

A pilot-plant pump delivers water ($\rho = 1000\ \mathrm{kg/m^3}$) at $Q = 12.0\ \mathrm{L/min}$ from a sump, where it is at rest, to a header tank whose outlet is $h = 8.50\ \mathrm m$ higher; the water leaves the outlet at $v = 2.40\ \mathrm{m/s}$. (a) Convert $Q$ to SI and find the mass flow rate $\dot m$. (b) Find the mechanical power the pump gives the water in terms of $\dot m$, $g$, $h$ and $v$; check a limiting case; evaluate. (c) The motor is $60.0\ \%$ efficient; what electrical power does it draw?

**Predict first:** Which is larger, the power to lift the water or the power to give it its outlet speed, and by roughly what factor?

**TR:** (a) Kütle debisi, (b) suya verilen mekanik güç, (c) %60 verimli motorun çektiği güç.

<details><summary>Answer / Yanıt</summary>

**(a)** $Q = \dfrac{12.0 \times 10^{-3}\ \mathrm{m^3}}{60\ \mathrm s} = 2.00 \times 10^{-4}\ \mathrm{m^3/s}$, so $\dot m = \rho Q = 0.200\ \mathrm{kg/s}$; litre and minute are exact definitions.

**(b)** In a time $\Delta t$ a mass $\dot m\,\Delta t$ passes. The pump works against gravity, $(\dot m\,\Delta t)\,gh$, and by the work–kinetic energy theorem also supplies $\tfrac12(\dot m\,\Delta t)v^2$, since the water starts at rest. Dividing by $\Delta t$,
$$P = \dot m\left(gh + \tfrac12 v^2\right).$$
Limiting cases: $h \to 0$ leaves $\tfrac12 \dot m v^2$, a horizontal jet; $v \to 0$ leaves $\dot m gh$, a slow lift. Numbers: $\dot m gh = (0.200)(9.81)(8.50) = 16.7\ \mathrm W$, $\tfrac12 \dot m v^2 = 0.576\ \mathrm W$, so $P = 17.3\ \mathrm W$.

**(c)** $P_{\rm elec} = P/\eta = 17.25/0.600 = 28.8\ \mathrm W$.

**Interpretation:** the lift term is about thirty times the kinetic term, which is why pump sizing starts from $\dot m gh$; no potential-energy bookkeeping is needed, only the work done against the weight of the water that moves.

</details>


#### P2 — Unit error in a logged motion trace  ·  L2

A $1.50\ \mathrm{kg}$ test cart is pushed from rest by a constant force along a straight track while a microcontroller logs its position. The header says `t [s], x [m]` and the rows read `t`: 0, 500, 1000, 1500, 2000 and `x`: 0.000, 0.125, 0.500, 1.125, 2.000. (a) Show from the data alone that the acceleration is constant. (b) The header taken literally gives an absurd acceleration; find the unit error, the true acceleration and the speed at the last row. (c) Find the net force and check that its work over the run equals the cart's kinetic energy.

**Predict first:** Is $x$ growing like $t$ or like $t^2$? What order of magnitude should a cart-on-a-track acceleration have?

**TR:** (a) Sabit ivmeyi veriden göster, (b) birim hatasını bul; gerçek ivme ve son hız, (c) net kuvvetin işi ile kinetik enerji.

<details><summary>Answer / Yanıt</summary>

**(a)** Successive differences of $x$: $0.125,\ 0.375,\ 0.625,\ 0.875$; their differences are all $0.250$. Equal second differences at equal time steps mean $x \propto t^2$, constant acceleration: for $x = \tfrac12 a t^2$ the second difference is $a\,(\Delta t)^2$.

**(b)** Taken literally, $\Delta t = 500\ \mathrm s$ gives $a = 0.250/500^2 = 1.00 \times 10^{-6}\ \mathrm{m/s^2}$: half an hour to cross two metres. The time column is in **milliseconds**. With $\Delta t = 0.500\ \mathrm s$,
$$a = \frac{\text{second difference}}{(\Delta t)^2} = \frac{0.250\ \mathrm m}{(0.500\ \mathrm s)^2} = 1.00\ \mathrm{m/s^2},$$
and $x = \tfrac12(1.00)t^2$ reproduces every row; zero second differences would return $a = 0$, uniform motion. At $t = 2.00\ \mathrm s$, $v = at = 2.00\ \mathrm{m/s}$.

**(c)** $F_{\rm net} = ma = 1.50\ \mathrm N$; $W = F_{\rm net}\,x = 3.00\ \mathrm J = \tfrac12(1.50)(2.00)^2$.

**Interpretation:** the physics caught the software error; a factor of $1000$ between header and data is almost always a millisecond timestamp.

</details>


#### P3 — Tote released on an AGV loading ramp  ·  L2

An automated guided vehicle places a $25.0\ \mathrm{kg}$ tote at rest on a ramp inclined at $\theta = 30.0^\circ$ and releases it; $\mu_s = 0.350$, $\mu_k = 0.200$. (a) Show whether the tote stays or slides. (b) If it slides, find its acceleration in terms of $g$, $\theta$ and $\mu_k$; check a limiting case; evaluate. (c) Find its speed and the elapsed time after $d = 4.00\ \mathrm m$ down the ramp.

**Predict first:** Does (a) depend on the mass? Is the acceleration more or less than half of $g$?

**TR:** (a) Kasa kayar mı, (b) ivme, (c) $4.00\ \mathrm m$ sonra hız ve süre.

<details><summary>Answer / Yanıt</summary>

**(a)** The down-slope pull is $mg\sin\theta$; static friction can hold at most $\mu_s mg\cos\theta$. It stays only if $\tan\theta \le \mu_s$; $\tan 30.0^\circ = 0.577 > 0.350$, so it slides. The mass cancels.

**(b)** Down the slope positive: $mg\sin\theta - \mu_k mg\cos\theta = ma$, so
$$a = g\,(\sin\theta - \mu_k\cos\theta).$$
Limiting cases: $\mu_k \to 0$ gives $g\sin\theta$; $\theta \to 90^\circ$ gives $g$, free fall, because $N \to 0$. Numbers: $a = 9.81\,(0.500 - 0.200 \times 0.866) = 3.21\ \mathrm{m/s^2}$, less than half of $g$ because friction removes about a third of the down-slope component.

**(c)** $v = \sqrt{2ad} = \sqrt{2(3.206)(4.00)} = 5.06\ \mathrm{m/s}$ and $t = v/a = 1.58\ \mathrm s$.

**Interpretation:** the steepest ramp that holds the tote is $\tan^{-1}(0.350) = 19.3^\circ$, so this ramp needs a stop.

</details>


#### P4 — Is the video really 60 frames per second?  ·  L2

A computing student extracts $g$ from a phone video of a ball dropped from rest through $h = 1.80\ \mathrm m$. The metadata says $60.0$ frames per second, but the fall occupies only $18$ frames. (a) Find the fall time in terms of $h$ and $g$; evaluate. (b) How many frames should the fall take at $60$ fps? What rate do $18$ frames imply, and what has probably happened? (c) Find the impact speed from the work–kinetic energy theorem, the distance moved during the last frame at the true rate, and the fractional error in $t$ from one miscounted frame.

**Predict first:** Is $18$ frames at $60$ fps a plausible fall time for $1.80\ \mathrm m$? Estimate before computing.

**TR:** (a) Düşüş süresi, (b) beklenen kare sayısı ve gerçek kare hızı, (c) çarpma hızı, son karedeki yol, bir karelik sayım hatası.

<details><summary>Answer / Yanıt</summary>

**(a)** From $y = \tfrac12 g t^2$, $t = \sqrt{2h/g}$; doubling $h$ multiplies $t$ by $\sqrt2$. $t = \sqrt{2(1.80)/9.81} = 0.606\ \mathrm s$.

**(b)** At $60.0$ fps the fall should take $60.0 \times 0.606 = 36.3$ frames. Eighteen frames imply $18/0.606 = 29.7\ \mathrm{fps}$, the standard $30$ fps: the camera recorded at $30$ fps, or the script read every second frame. A script trusting the header would halve $t$ and, since $g = 2h/t^2$, return $g$ four times too large.

**(c)** Only gravity does work: $mgh = \tfrac12 m v^2$, so $v = \sqrt{2gh}$, mass-independent; $v = \sqrt{2(9.81)(1.80)} = 5.94\ \mathrm{m/s}$. In the last frame at $30$ fps the ball moves about $v\,\Delta t = 5.94/30 = 0.198\ \mathrm m$, so the impact frame is blurred over $20\ \mathrm{cm}$ and the count is uncertain by one frame: $0.0333\ \mathrm s$, or $5.5\ \%$ of $t$ and, since $g \propto 1/t^2$, $11\ \%$ of $g$.

**Interpretation:** check timing data first against a physics estimate of the count, then against the size of the last displacement; longer drops and higher frame rates give a better $g$.

</details>


#### P5 — Inclined conveyor of reactor feedstock  ·  L3

A belt conveyor carries a $40.0\ \mathrm{kg}$ tote of catalyst pellets up an $18.0^\circ$ incline to a reactor charging hatch; the tote sits on the belt without a stop, $\mu_s = 0.450$. The belt starts from rest, reaches $v = 0.800\ \mathrm{m/s}$ with constant acceleration in $1.00\ \mathrm s$, then runs at that speed for $L = 6.00\ \mathrm m$. (a) Find the friction the belt must exert during start-up in terms of $m$, $a$, $g$, $\theta$; does the tote slip? (b) In the steady run, find the friction, the power the belt delivers to the tote and its work over $L$. (c) If start-up is shortened to $0.600\ \mathrm s$, does the tote slip?

**Predict first:** In the steady run, is the friction on the tote up or down the slope? Static or kinetic?

**TR:** (a) Kalkışta gereken sürtünme ve kayma kontrolü, (b) sabit hızda sürtünme, güç ve iş, (c) $0.600\ \mathrm s$'lik kalkışta kayar mı?

<details><summary>Answer / Yanıt</summary>

**(a)** Forces on the tote: weight, normal force, static friction $f$ **up** the slope — the belt drags the tote, so friction drives. Up-slope positive, $f - mg\sin\theta = ma$:
$$f_{\rm needed} = m\,(a + g\sin\theta), \qquad a = \frac{v}{t_{\rm start}} = 0.800\ \mathrm{m/s^2}.$$
Limiting cases: $\theta \to 0$ gives $ma$, a flat belt; $a \to 0$ gives $mg\sin\theta$. Numbers: $f_{\rm needed} = 40.0\,(0.800 + 3.03) = 153\ \mathrm N$; available $\mu_s mg\cos\theta = 0.450 \times 40.0 \times 9.81 \times 0.951 = 168\ \mathrm N$. Since $153 < 168$, no slip.

**(b)** Steady speed, $a = 0$: $f = mg\sin\theta = 121\ \mathrm N$, still static, still up the slope. $P = fv = (121.3)(0.800) = 97.0\ \mathrm W$; over $L$, $W_{\rm belt} = mgL\sin\theta = 728\ \mathrm J$. Check: gravity does $-728\ \mathrm J$, the normal force none, $\Delta K = 0$.

**(c)** $a = 0.800/0.600 = 1.33\ \mathrm{m/s^2}$, so $f_{\rm needed} = 40.0\,(1.33 + 3.03) = 175\ \mathrm N > 168\ \mathrm N$: the tote slips back. Use a gentler ramp-up, a cleat, or a lower angle.

**Interpretation:** the slip condition $a + g\sin\theta \le \mu_s g\cos\theta$ contains no mass, so heavier totes are no safer; at steady speed the belt's $97\ \mathrm W$ all goes against the weight component.

</details>


#### P6 — Mobile robot towing a trolley  ·  L2

A mobile robot ($m_1 = 12.0\ \mathrm{kg}$) tows a parts trolley ($m_2 = 8.00\ \mathrm{kg}$) by a light horizontal cable across a level floor. The drive wheels supply a net forward force $F = 30.0\ \mathrm N$ on the robot; the trolley's rolling resistance is a friction force $\mu m_2 g$, $\mu = 0.0500$. (a) Find the acceleration and the cable tension in symbols; check a limiting case; evaluate. (b) Over the first $d = 2.50\ \mathrm m$ from rest, find the work done on the trolley by the cable and by friction, and its speed from the work–kinetic energy theorem. (c) Confirm the speed by kinematics.

**Predict first:** Is the tension larger or smaller than $F$? Larger or smaller than $F m_2/(m_1 + m_2)$?

**TR:** (a) İvme ve kablo gerilmesi, (b) $2.50\ \mathrm m$ boyunca kablonun ve sürtünmenin işi, hız, (c) kinematik doğrulama.

<details><summary>Answer / Yanıt</summary>

**(a)** Whole system: $F - \mu m_2 g = (m_1 + m_2)\,a$. Trolley alone: $T - \mu m_2 g = m_2 a$. Hence
$$a = \frac{F - \mu m_2 g}{m_1 + m_2}, \qquad T = m_2\,\frac{F + \mu m_1 g}{m_1 + m_2}.$$
Limiting cases: $\mu \to 0$ gives $T = F m_2/(m_1 + m_2)$, the trolley's share of the push; $m_2 \to 0$ gives $T \to 0$. Numbers: $\mu m_2 g = 3.92\ \mathrm N$; $a = (30.0 - 3.92)/20.0 = 1.30\ \mathrm{m/s^2}$; $T = 8.00(1.304) + 3.92 = 14.4\ \mathrm N$, above the frictionless share $12.0\ \mathrm N$ and below $F$.

**(b)** $W_T = Td = (14.35)(2.50) = 35.9\ \mathrm J$; $W_f = -\mu m_2 g\,d = -9.81\ \mathrm J$; weight and normal force do no work. $\Delta K = 26.1\ \mathrm J = \tfrac12 m_2 v^2$, so $v = \sqrt{2(26.08)/8.00} = 2.55\ \mathrm{m/s}$.

**(c)** $v^2 = 2ad = 2(1.304)(2.50) = 6.52\ \mathrm{m^2/s^2}$, $v = 2.55\ \mathrm{m/s}$: the theorem is Newton's second law integrated over the displacement.

**Interpretation:** the cable carries under half the drive force; loading the trolley to $m_2 = m_1$ would raise $T$ to about $17\ \mathrm N$.

</details>


#### P7 — Beacon drop from a moving quadcopter with command latency  ·  L3

A survey quadcopter flying horizontally at $u = 6.00\ \mathrm{m/s}$ at height $h = 20.0\ \mathrm m$ must release a beacon onto a ground marker. The beacon leaves with the drone's velocity; the release command reaches the mechanism after a latency $\tau = 250\ \mathrm{ms}$. Ignore air resistance. (a) Find the fall time and the horizontal distance travelled during the fall, in symbols and numbers. (b) At what horizontal distance $D$ before the marker must the *command* be sent? Give $D$ in terms of $u$, $h$, $g$, $\tau$ and check $\tau \to 0$. (c) On the next pass the drone is also climbing at $w = 1.50\ \mathrm{m/s}$. Find the new fall time and $D$; is the command earlier or later?

**Predict first:** At impact, is the beacon ahead of the drone, behind it, or directly below? Does climbing lengthen or shorten the flight?

**TR:** (a) Düşüş süresi ve yatay yol, (b) komutun hedeften ne kadar önce verileceği, (c) drone yükselirken aynı hesap.

<details><summary>Answer / Yanıt</summary>

**(a)** Origin at release, forward $x$, up $y$; $v_x = u$, $v_y = 0$. Vertical: $-h = -\tfrac12 g t^2$, so $t_1 = \sqrt{2h/g}$; horizontal: $x_1 = u\sqrt{2h/g}$. Limiting case: $u \to 0$ gives $x_1 \to 0$ with $t_1$ unchanged. Numbers: $t_1 = \sqrt{40.0/9.81} = 2.02\ \mathrm s$, $x_1 = 6.00 \times 2.019 = 12.1\ \mathrm m$. The beacon keeps the drone's $v_x$, so at impact the drone is directly overhead.

**(b)** Between command and release the drone covers $u\tau$, with $\tau = 250\ \mathrm{ms} = 0.250\ \mathrm s$:
$$D = u\left(\tau + \sqrt{\frac{2h}{g}}\right).$$
$\tau \to 0$ returns $x_1$. Numbers: $D = 6.00\,(0.250 + 2.019) = 13.6\ \mathrm m$; the latency alone is worth $1.50\ \mathrm m$.

**(c)** Now $v_y = +w$. Up positive: $-h = wt - \tfrac12 g t^2$, whose positive root is
$$t_2 = \frac{w + \sqrt{w^2 + 2gh}}{g};$$
$w \to 0$ returns $\sqrt{2h/g}$. Numbers: $\sqrt{2.25 + 392.4} = 19.9\ \mathrm{m/s}$, $t_2 = (1.50 + 19.87)/9.81 = 2.18\ \mathrm s$, $D_2 = 6.00\,(0.250 + 2.178) = 14.6\ \mathrm m$. The beacon first rises $w^2/2g = 0.115\ \mathrm m$, so the flight is longer and the command must go **earlier**, about $0.95\ \mathrm m$ further out.

**Interpretation:** the horizontal motion is uniform throughout, so the timing lives in the vertical equation; the controller needs the vertical velocity, the height and its own latency in SI units.

</details>


#### P8 — Pick-and-place gripper holding a part by friction  ·  L3

A gantry robot holds a $2.50\ \mathrm{kg}$ part between two side pads that squeeze it with a normal force $N_p$ each; $\mu_s = 0.600$. Friction alone carries the weight and drives every acceleration. The gantry moves the part horizontally from rest at $a = 3.00\ \mathrm{m/s^2}$ for $t = 0.400\ \mathrm s$; later a vertical stage lifts it with upward acceleration $a_y = 2.00\ \mathrm{m/s^2}$. (a) For the horizontal move, draw the friction force with its two components, find its magnitude in terms of $m$, $g$, $a$, and the minimum $N_p$ per pad; check $a \to 0$. (b) For the lift, find the friction needed and the minimum $N_p$; which motion sets the grip design? (c) For the horizontal move, find the speed and distance at $0.400\ \mathrm s$, the work done by the gripper, the average power and the instantaneous power at the end.

**Predict first:** Which needs more friction, the move or the lift? Is the instantaneous power at the end greater or smaller than the average?

**TR:** (a) Yatay ivmede sürtünme vektörü ve en küçük sıkma, (b) kaldırırken aynı hesap, (c) hız, yol, iş, ortalama ve anlık güç.

<details><summary>Answer / Yanıt</summary>

**(a)** Friction has a vertical component $mg$ (it carries the weight) and a horizontal component $ma$ (the only horizontal force):
$$f = m\sqrt{g^2 + a^2}, \qquad N_p \ge \frac{m\sqrt{g^2 + a^2}}{2\mu_s},$$
since two pads supply at most $2\mu_s N_p$. Limiting case: $a \to 0$ gives $f = mg$ and $N_p = mg/2\mu_s$, the squeeze that just holds the part at rest. Numbers: $mg = 24.5\ \mathrm N$, $ma = 7.50\ \mathrm N$, $f = 25.6\ \mathrm N$, $N_p \ge 25.6/1.20 = 21.4\ \mathrm N$ per pad; the components add at right angles, so the horizontal demand adds only about $1\ \mathrm N$.

**(b)** Up positive: $f - mg = ma_y$, so $f = m\,(g + a_y)$; $a_y \to 0$ gives $mg$, $a_y \to -g$ gives $0$. Numbers: $f = 2.50\,(11.81) = 29.5\ \mathrm N$, $N_p \ge 24.6\ \mathrm N$. The lift needs more ($29.5 > 25.6\ \mathrm N$) because $a_y$ adds to $g$ along the same line, so the lift sets the grip.

**(c)** $v = at = 1.20\ \mathrm{m/s}$; $d = \tfrac12 at^2 = 0.240\ \mathrm m$. Horizontal force $F = ma = 7.50\ \mathrm N$; the vertical friction does no work in a horizontal move. $W = Fd = 1.80\ \mathrm J = \tfrac12(2.50)(1.20)^2$. Average power $W/t = 4.50\ \mathrm W$; instantaneous $Fv = 9.00\ \mathrm W$, twice the average, because under constant force from rest the power grows linearly with time.

**Interpretation:** the grip is sized by the lift and the motor by the peak $9.00\ \mathrm W$, not the average; a power figure quoted without saying which is meant is a unit error in disguise.

</details>


### Common errors seen in past midterms / Sık yapılan hatalar

Each item cost marks on more than one script in past sittings. Use them as a checklist on the sample questions and the eight problems above.

1. **The sign of $g$.** $y = v_0 t + \tfrac12 g t^2$ with $g = 9.81$ and up positive. $g$ is the positive number $9.81\ \mathrm{m/s^2}$; choose the axis once, at the top of the page, and then $a_y = -g$ if up is positive.
2. **cos and sin swapped on an incline.** Along the slope the weight component is $mg\sin\theta$, into the slope $mg\cos\theta$. Check with $\theta \to 0$: a flat floor has no along-slope pull, which $\sin 0 = 0$ gives and $\cos 0 = 1$ does not.
3. **Normal force equals weight on an incline.** $N = mg\cos\theta$, reduced by an upward pull (Sample 1) or increased by a downward push. $N$ is whatever the perpendicular balance says, never a given.
4. **Forgetting that the work of friction is negative.** Kinetic friction opposes the displacement, so $W_f = -f_k d$ whether the body speeds up or slows down; the sign comes from $\vec f_k$ against $\Delta\vec x$, not from the speed.
5. **Mixing km/h and m/s.** $72\ \mathrm{km/h}$ put into $v^2 = v_0^2 + 2a\Delta x$ without dividing by $3.6$ gives a braking distance thirteen times too long. Convert every given to SI before the first equation.
6. **Using $\bar v = (v_0 + v)/2$ when the acceleration is not constant.** That formula belongs to constant acceleration only; for a variable force use $W = \int F\,dx$, or the area under the graph, with the work–kinetic energy theorem.
7. **Third-law pair on the same body.** The weight of a crate and the normal force on it both act on the crate, so they are not a pair; the partner of $\vec N$ is the crate's push on the floor. A free-body diagram shows one body's forces only.
8. **Rounding intermediate values.** Rounding $\cos 25^\circ$ to $0.9$ or $\sqrt{2h/g}$ to $0.5\ \mathrm s$ moves the third significant figure and sometimes the second. Carry full precision; round once, at the boxed answer.

**TR:** 1 $g$'nin işareti eksende yaşar; 2 eğim boyunca $\sin\theta$, dik $\cos\theta$; 3 normal kuvvet dik dengenin sonucudur; 4 sürtünme işi hep eksi; 5 km/h'yi önce çevir; 6 $(v_0 + v)/2$ yalnızca sabit ivmede; 7 üçüncü yasa çifti aynı cisimde olmaz; 8 yalnız son kutuda yuvarla.

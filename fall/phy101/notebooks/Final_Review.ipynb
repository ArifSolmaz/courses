## 1. The method, one last time / Yöntem, son bir kez

> **draw → choose a law → rearrange to the symbolic answer → check a limiting case → substitute with units → interpret**

In an exam, steps 1 and 2 take thirty seconds and decide everything. Steps 3–6 are execution.

**What this is worth in marks.** A correct symbolic answer with an arithmetic slip loses almost nothing. A
correct number with no visible derivation cannot be given partial credit, because there is nothing to
credit. And a solution that starts by substituting numbers into a half-remembered formula is the single
most common way to lose a whole question — you cannot check a wrong formula once the symbols are gone.

**One habit above all others:** before writing anything, draw the situation and label what you know with
units. Two-thirds of the errors in this course's exams are not physics errors at all — they are a sign,
an angle measured from the wrong axis, or a length used where a height belonged, and every one of them is
visible in a drawing and invisible in an equation.

**TR:** Sınavda ilk iki adım otuz saniye alır ve her şeyi belirler. Doğru sembolik çözüm artı aritmetik
sürçme neredeyse hiçbir şey kaybettirmez; türetmesi görünmeyen doğru sayıya kısmi puan verilemez, çünkü
puanlanacak bir şey yoktur. **Her şeyden önce gelen alışkanlık:** yazmaya başlamadan önce durumu çiz ve
bildiklerini birimleriyle etiketle. Bu dersin sınavlarındaki hataların üçte ikisi fizik hatası değildir —
bir işaret, yanlış eksenden ölçülmüş bir açı, yüksekliğin yerine kullanılmış bir uzunluk; hepsi bir
şekilde görünür, denklemde görünmez.

## 2. Choosing a principle / İlkeyi seçmek

Almost every problem in this course can be attacked by more than one route, but one route is usually far
shorter. Read the question for what it **gives** and what it **asks**, then use this table.

| The question gives you… and asks for… | Use | Why |
| --- | --- | --- |
| forces, and asks for acceleration (or the reverse) | $\sum\vec F = m\vec a$ | the direct definition |
| forces and a **distance**, asks for a **speed** | work–energy, $W_{\rm net}=\Delta K$ | time never appears |
| forces and a **time**, asks for a **speed** | impulse–momentum, $\vec J = \Delta\vec p$ | distance never appears |
| start and end **heights**, asks for a speed, no friction | energy conservation | the path does not matter |
| a **collision** or an explosion | momentum conservation | internal forces cancel |
| a collision **and** asks whether energy was lost | momentum **and** compare $K$ | two separate statements |
| torques, asks for angular acceleration | $\sum\tau = I\alpha$ | rotational Newton II |
| rolling down a slope, asks for a speed | energy with **both** $\tfrac12mv^2$ and $\tfrac12I\omega^2$ | rolling stores energy twice |
| "at rest" or "in equilibrium", asks for an unknown force | $\sum\vec F=0$ **and** $\sum\tau=0$ | both, always |
| $I$ changes with no external torque | $L = I\omega$ conserved | Week 11 |
| a restoring force proportional to displacement | $T = 2\pi\sqrt{m/k}$ or $2\pi\sqrt{L/g}$ | it must be SHM |

### The three decisions that cost the most marks

**1. Time or distance?** This is the fastest diagnostic in the whole course. If the problem mentions a
distance and no time, reach for energy. If it mentions a time and no distance, reach for impulse. Choosing
the other one forces you to find the missing quantity first, turning a one-line problem into three.

**2. Is energy conserved here?** Only if there is no friction, no collision, and nothing deforms. If any
of those are present, mechanical energy is **not** conserved — but momentum may still be. Momentum survives
collisions; kinetic energy generally does not. That asymmetry is examined almost every year.

**3. Which axis, which pivot?** On an incline, use axes along and perpendicular to the slope. For a torque
problem, choose the pivot **at the point whose force you do not want to know** — that force then has zero
lever arm and vanishes from the equation. Choosing the pivot well can turn two simultaneous equations into
one.

**TR:** **Mesafe mi süre mi?** Problemde mesafe var ve süre yoksa enerjiye git; süre var ve mesafe yoksa
itmeye git. Yanlış seçim, tek satırlık problemi üç satıra çıkarır. **Enerji korunuyor mu?** Yalnızca
sürtünme, çarpışma ve şekil değişimi yoksa. Momentum çarpışmalardan sağ çıkar, kinetik enerji genellikle
çıkmaz. **Hangi eksen, hangi mesnet?** Eğik düzlemde eksenleri eğime paralel ve dik seç; tork probleminde
mesnedi, **bilmek istemediğin kuvvetin** bulunduğu noktaya koy — o kuvvetin moment kolu sıfır olur ve
denklemden düşer.

## 3. Self-diagnosis: twelve questions / Öz değerlendirme: on iki soru

Answer all twelve **before** opening anything. They are conceptual on purpose: they take about fifteen
minutes and they tell you which weeks to revise. Each answer names the week to return to.

1. A block slides to the right at constant speed on a rough table, pushed by a horizontal force. How do
   the sizes of the push and the friction compare?
2. A ball is thrown straight up. At the highest point, what are its velocity and its acceleration?
3. Two balls leave a table edge at the same instant, one dropped and one fired horizontally. Which lands
   first?
4. A book rests on a table. Is the table's upward push the third-law partner of the book's weight?
5. You carry a heavy suitcase horizontally at constant speed for 50 m. How much work do you do on it?
6. A car doubles its speed. By what factor does its stopping distance change, at the same braking force?
7. A light cart and a heavy cart are pushed with the same force for the same **time**. Which ends with more
   momentum? With more kinetic energy?
8. Two equal masses collide head-on at equal speeds and stick together. Where does the kinetic energy go?
9. A solid disc and a ring of the same mass and radius roll down the same slope from rest. Which reaches
   the bottom first?
10. A skater pulls her arms in and spins faster. What pushed her?
11. A pendulum clock is taken up a high mountain. Does it run fast, slow, or the same?
12. Two identical springs, one with a 1 kg mass and one with a 4 kg mass. How do their periods compare?

<details><summary>Answers, with the week to revise / Yanıtlar ve dönülecek hafta</summary>

1. **Equal in size, opposite in direction.** Constant speed means zero acceleration means zero net force.
   "Moving" does not require a net force — that is Newton's first law. → **Week 03–04**
2. **Velocity zero, acceleration still $9.81\ \mathrm{m/s^2}$ downward.** Gravity does not pause at the
   top. If $a$ were zero there, the ball would stay. → **Week 02**
3. **Together.** The vertical motions are identical and independent of the horizontal. → **Week 03**
4. **No.** Both act on the book, so they cannot be a third-law pair. The weight's partner is the book
   pulling the Earth; the normal force's partner is the book pushing the table. → **Week 03–04**
5. **Zero.** Your force is vertical, the displacement horizontal, $\cos 90^\circ = 0$. You get tired
   because of muscle physiology, not because of physics work. → **Week 05**
6. **Four times.** $d = mv^2/2F \propto v^2$. → **Week 05**
7. **Same momentum** ($\Delta p = F\Delta t$, identical for both); **the light cart has more kinetic
   energy** (same force acts through a greater distance, so more work is done on it). This one question
   separates impulse from work better than any other. → **Weeks 05 and 09**
8. **They stop dead, and all of it becomes heat and deformation.** Momentum was zero before and after, so
   nothing is violated — momentum conservation does not require motion. → **Week 09**
9. **The disc.** It has the smaller $I$ ($\tfrac12MR^2$ against $MR^2$), so less of the energy goes into
   rotation and more into translation. Mass and radius are irrelevant — only the *shape* matters. → **Week 11**
10. **Nothing pushed her.** No external torque acts, so $L = I\omega$ is conserved; pulling her arms in
    lowers $I$, so $\omega$ must rise. She does do work, which is where the extra kinetic energy comes
    from. → **Week 11**
11. **Slow.** Higher up, $g$ is smaller, and $T = 2\pi\sqrt{L/g}$ grows as $g$ falls. → **Week 13**
12. **The 4 kg period is twice as long.** $T \propto \sqrt m$, and $\sqrt4 = 2$. → **Week 13**

**Scoring, honestly:** fewer than 8 right means revise the named weeks *conceptually* before touching any
more calculations. Practising arithmetic on top of a wrong concept just makes the wrong answer faster.

**TR:** 8'den az doğru varsa, daha fazla hesap yapmadan önce adı geçen haftaları **kavramsal olarak**
tekrarla. Yanlış bir kavramın üstüne aritmetik çalışmak, yalnızca yanlış cevabı hızlandırır.

</details>

## 4. The formula map / Formül haritası

Grouped by the principle they belong to, not alphabetically — because in an exam you retrieve by principle.

**Kinematics, constant acceleration (Weeks 02–03)**

$$v = v_0 + at, \qquad x = x_0 + v_0t + \tfrac12at^2, \qquad v^2 = v_0^2 + 2a\Delta x$$

Projectiles: the same equations twice, with $a_x = 0$, $a_y = -g$, sharing one $t$.

**Forces (Weeks 03–04)**

$$\sum\vec F = m\vec a, \qquad f_s \le \mu_s N, \qquad f_k = \mu_k N, \qquad
\text{incline: } a = g\sin\theta,\ N = mg\cos\theta$$

Note $f_s \le \mu_s N$ is an **inequality** — static friction takes whatever value is needed, up to a
maximum. Using $f_s = \mu_s N$ when the object is not on the point of slipping is a standard error.

**Energy (Weeks 05, 08)**

$$W = Fd\cos\theta, \qquad W = \int F\,dx \text{ (area under } F\text{–}x), \qquad
K = \tfrac12mv^2, \qquad P = \frac{W}{\Delta t} = Fv$$
$$W_{\rm net} = \Delta K, \qquad U_g = mgh, \qquad U_s = \tfrac12kx^2, \qquad
K_i + U_i + W_{\rm other} = K_f + U_f$$

**Momentum (Week 09)**

$$\vec p = m\vec v, \qquad \vec J = \vec F\Delta t = \Delta\vec p, \qquad
\sum\vec p_i = \sum\vec p_f \text{ (no external impulse)}$$
$$\text{perfectly inelastic: } v' = \frac{m_1v_1 + m_2v_2}{m_1+m_2}$$

**Rotation (Weeks 10–11)**

$$\theta,\ \omega = \frac{d\theta}{dt},\ \alpha = \frac{d\omega}{dt}; \qquad
\omega = \omega_0 + \alpha t, \qquad \theta = \omega_0 t + \tfrac12\alpha t^2, \qquad
\omega^2 = \omega_0^2 + 2\alpha\theta$$
$$v = R\omega, \quad a_t = R\alpha, \quad a_c = \omega^2R; \qquad
\sum\tau = I\alpha = \frac{dL}{dt}, \qquad L = I\omega$$
$$K_{\rm rot} = \tfrac12I\omega^2, \qquad \text{rolling: } K = \tfrac12mv^2 + \tfrac12I\omega^2$$
$$I_{\rm disc} = \tfrac12MR^2, \quad I_{\rm ring} = MR^2, \quad I_{\rm rod,cm} = \tfrac1{12}ML^2,
\quad I_{\rm sphere} = \tfrac25MR^2, \qquad I = I_{\rm cm} + Md^2$$

**Equilibrium and centre of mass (Week 12)**

$$\sum\vec F = 0 \ \textbf{and}\ \sum\tau = 0, \qquad \tau = rF\sin\phi = F r_\perp, \qquad
x_{\rm cm} = \frac{\sum m_ix_i}{\sum m_i}$$

**Periodic motion (Week 13)**

$$x = A\cos(\omega t+\phi), \qquad \omega = \sqrt{\frac{k}{m}}, \qquad T = \frac{2\pi}{\omega}
= 2\pi\sqrt{\frac{m}{k}}, \qquad T_{\rm pend} = 2\pi\sqrt{\frac{L}{g}}$$
$$E = \tfrac12kA^2, \qquad v_{\max} = A\omega, \qquad a_{\max} = A\omega^2$$

**Constant:** $g = 9.81\ \mathrm{m/s^2}$. **Report three significant figures.**

**TR:** Formüller ilkelerine göre gruplanmıştır, alfabetik değil — çünkü sınavda ilkeden hatırlarsın.
$f_s \le \mu_s N$ bir **eşitsizliktir**; cisim kayma sınırında değilken $f_s = \mu_s N$ yazmak standart bir
hatadır.

## 5. Six mixed review routes / Altı karma tekrar rotası

Each combines two or more weeks, like the harder exam questions. Attempt, then open.

#### Route A — forces into energy (Weeks 04 + 05)

A $12.0\ \mathrm{kg}$ crate is pulled $6.00\ \mathrm m$ across a horizontal floor by a $75.0\ \mathrm N$
force at $30.0^\circ$ above the horizontal. The coefficient of kinetic friction is $0.250$. Find the
crate's final speed, starting from rest.

<details><summary>Answer — after your own attempt</summary>

**Forces first, because friction needs the normal force**, and the normal force is not $mg$ here — the
pull has an upward component:

$$N = mg - F\sin\theta = (12.0)(9.81) - (75.0)(0.500) = 117.7 - 37.5 = 80.2\ \mathrm N$$

**Missing that is the trap in this question**, and it changes the answer by a third.

$$f_k = \mu_k N = (0.250)(80.2) = 20.1\ \mathrm N$$

**Now energy.** Three forces do work; $N$ and the vertical pull component do none (perpendicular to
motion):

$$W_{\rm pull} = Fd\cos\theta = (75.0)(6.00)(0.866) = 390\ \mathrm J, \qquad
W_f = -f_kd = -(20.1)(6.00) = -120\ \mathrm J$$
$$W_{\rm net} = 390 - 120 = 270\ \mathrm J = \tfrac12mv^2
\Rightarrow v = \sqrt{\frac{2(270)}{12.0}} = 6.70\ \mathrm{m/s}$$

**Sense check:** if you had wrongly used $N = mg$, then $f_k = 29.4\ \mathrm N$, $W_f = -177\ \mathrm J$
and $v = 5.96\ \mathrm{m/s}$ — plausible-looking and wrong. The check that catches it: a force pulling
*upward* must reduce the normal force, so it must reduce friction.

**TR:** Önce kuvvetler, çünkü sürtünme normal kuvvete ihtiyaç duyar — ve burada $N \ne mg$: çekmenin
yukarı bileşeni normal kuvveti azaltır. Bu soruların tuzağı budur ve cevabı üçte bir oranında değiştirir.

</details>

#### Route B — momentum then energy (Weeks 09 + 08)

A $0.0500\ \mathrm{kg}$ bullet travelling at $400\ \mathrm{m/s}$ embeds itself in a $4.95\ \mathrm{kg}$
block hanging from a long string. Find (a) the speed of block-plus-bullet just after impact and (b) the
height it swings to.

<details><summary>Answer — after your own attempt</summary>

**Two stages, and they need different principles. Mixing them is the classic error.**

**(a) The collision — momentum, not energy.** Energy is *not* conserved here; the bullet deforms the
block.

$$m_bv_b = (m_b+M)v' \Rightarrow v' = \frac{(0.0500)(400)}{5.00} = \frac{20.0}{5.00} = 4.00\ \mathrm{m/s}$$

**(b) The swing — energy, not momentum.** Momentum is *not* conserved here; the string tension is an
external force. After impact nothing dissipates, so:

$$\tfrac12(5.00)(4.00)^2 = (5.00)(9.81)h \Rightarrow h = \frac{v'^2}{2g} = \frac{16.0}{19.62} = 0.815\ \mathrm m$$

**How much energy did the collision destroy?** Before: $\tfrac12(0.0500)(400)^2 = 4000\ \mathrm J$.
After: $\tfrac12(5.00)(4.00)^2 = 40.0\ \mathrm J$. **99% of it is gone** — into heat and deformation.
That is why you must not use energy conservation for stage (a): it would give a wildly wrong answer.

**TR:** İki aşama, iki farklı ilke; karıştırmak klasik hatadır. Çarpışmada **momentum** korunur (enerji
korunmaz, mermi bloğu deforme eder); salınımda **enerji** korunur (momentum korunmaz, ip gerilmesi dış
kuvvettir). Çarpışma enerjinin **%99'unu** yok eder.

</details>

#### Route C — rolling energy (Weeks 08 + 11)

A solid sphere ($I = \tfrac25MR^2$) rolls without slipping from rest down a slope of vertical height
$2.00\ \mathrm m$. Find its speed at the bottom, symbolically and numerically. Compare with a block
sliding down frictionlessly.

<details><summary>Answer — after your own attempt</summary>

**Symbolic first.** Rolling stores energy in two places, and the no-slip condition $v = R\omega$ links them:

$$Mgh = \tfrac12Mv^2 + \tfrac12I\omega^2 = \tfrac12Mv^2 + \tfrac12\left(\tfrac25MR^2\right)\frac{v^2}{R^2}
= \tfrac12Mv^2 + \tfrac15Mv^2 = \tfrac{7}{10}Mv^2$$

$$\boxed{\,v = \sqrt{\frac{10gh}{7}}\,}$$

Both $M$ and $R$ cancel — so a marble and a cannonball arrive together, and the radius is irrelevant.

**Numerically:** $v = \sqrt{10(9.81)(2.00)/7} = \sqrt{28.0} = 5.29\ \mathrm{m/s}$.

**Compare:** the sliding block gets all the energy as translation, $v = \sqrt{2gh} = 6.26\ \mathrm{m/s}$.
The sphere is slower because $\tfrac27$ of its energy is tied up in spin. **Limiting check:** if $I \to 0$
the formula returns $\sqrt{2gh}$ ✅. And the ranking by shape — sphere ($\tfrac25$) beats disc
($\tfrac12$) beats ring ($1$) — depends only on how far the mass sits from the axis.

**TR:** Yuvarlanma enerjiyi iki yerde depolar ve $v = R\omega$ ikisini bağlar; $M$ ve $R$ sadeleşir, yani
bilye ile gülle birlikte varır. Küre kayan bloktan yavaştır çünkü enerjisinin $\tfrac27$'si dönmededir.

</details>

#### Route D — equilibrium with a chosen pivot (Week 12)

A uniform $8.00\ \mathrm m$ ladder of mass $25.0\ \mathrm{kg}$ leans against a frictionless wall at
$60.0^\circ$ to the horizontal. Find the wall's reaction force and the friction force at the floor.

<details><summary>Answer — after your own attempt</summary>

**Choose the pivot at the foot of the ladder**, so the two unknown floor forces have zero lever arm and
drop out of the torque equation. That single choice is the whole solution.

Horizontal distance from foot to the wall contact: $8.00\cos 60.0^\circ = 4.00\ \mathrm m$; vertical
height of the contact: $8.00\sin 60.0^\circ = 6.93\ \mathrm m$. The weight acts at the midpoint, a
horizontal distance $2.00\ \mathrm m$ from the foot.

$$\sum\tau_{\rm foot} = 0: \quad N_{\rm wall}(6.93) = mg(2.00)
\Rightarrow N_{\rm wall} = \frac{(25.0)(9.81)(2.00)}{6.93} = 70.8\ \mathrm N$$

$$\sum F_x = 0: \quad f = N_{\rm wall} = 70.8\ \mathrm N, \qquad
\sum F_y = 0: \quad N_{\rm floor} = mg = 245\ \mathrm N$$

**Limiting check:** as the ladder approaches vertical, $\cos\theta \to 0$, the weight's lever arm
vanishes and $N_{\rm wall} \to 0$ ✅ — a vertical ladder needs no wall. As it flattens, $N_{\rm wall}$
grows without limit, which is why ladders slip at shallow angles. The minimum coefficient of friction
needed here is $f/N_{\rm floor} = 70.8/245 = 0.289$.

**TR:** Mesnedi **ayağa** koy: iki bilinmeyen zemin kuvvetinin moment kolu sıfır olur ve denklemden
düşer; çözüm bu tek seçimdir. Merdiven düşeye yaklaşınca duvar kuvveti sıfıra gider, yatıklaştıkça
sınırsız büyür — merdivenlerin küçük açılarda kaymasının sebebi budur.

</details>

#### Route E — SHM from a static measurement (Week 13)

Hanging a $0.400\ \mathrm{kg}$ mass on a spring stretches it by $8.00\ \mathrm{cm}$. The mass is then
pulled down a further $3.00\ \mathrm{cm}$ and released. Find $k$, the period, the maximum speed and the
maximum acceleration.

<details><summary>Answer — after your own attempt</summary>

**The static stretch gives $k$** — at equilibrium the spring force balances the weight:

$$k = \frac{mg}{x_0} = \frac{(0.400)(9.81)}{0.0800} = 49.1\ \mathrm{N/m}$$

$$\omega = \sqrt{\frac{k}{m}} = \sqrt{\frac{49.1}{0.400}} = 11.1\ \mathrm{rad/s}, \qquad
T = \frac{2\pi}{\omega} = 0.567\ \mathrm s$$

The amplitude is the *extra* displacement from equilibrium, $A = 0.0300\ \mathrm m$ — not the total
stretch of $0.110\ \mathrm m$. **That distinction is the trap here.**

$$v_{\max} = A\omega = (0.0300)(11.1) = 0.332\ \mathrm{m/s}, \qquad
a_{\max} = A\omega^2 = (0.0300)(122.6) = 3.68\ \mathrm{m/s^2}$$

**Neat shortcut worth knowing:** combining $k = mg/x_0$ with $T = 2\pi\sqrt{m/k}$ gives
$T = 2\pi\sqrt{x_0/g}$ — identical in form to the pendulum, with the static stretch playing the role of
the length. So a spring stretched 8 cm by its own load has the same period as an 8 cm pendulum.

**TR:** Statik uzama $k$'yi verir. Genlik, dengeden **ek** yer değiştirmedir ($0.0300\ \mathrm m$), toplam
uzama değil — sorunun tuzağı budur. Güzel bir kısayol: $T = 2\pi\sqrt{x_0/g}$, yani yükü altında 8 cm uzayan
bir yay, 8 cm'lik bir sarkaçla aynı periyoda sahiptir.

</details>

#### Route F — repair a tempting wrong solution (Weeks 02 + 05)

A student writes: *"A 2.0 kg ball is dropped from 5.0 m. Its speed at the ground is
$v = 2gh = 2(9.81)(5.0) = 98\ \mathrm{m/s}$."* Find every error and give the correct answer.

<details><summary>Answer — after your own attempt</summary>

**Three errors, and the third is the one that matters:**

1. The relation is $v^2 = 2gh$, not $v = 2gh$. The square root was dropped.
2. Therefore $v = \sqrt{2(9.81)(5.0)} = \sqrt{98.1} = 9.90\ \mathrm{m/s}$.
3. **The units were never checked.** $2gh$ has units $\mathrm{m^2/s^2}$, which is not a speed. Writing
   $v = 98\ \mathrm{m/s}$ requires ignoring the units of your own expression.

**And a sense check the student skipped:** $98\ \mathrm{m/s}$ is 350 km/h from a five-metre drop. A
two-second sanity check — free fall for about a second reaches roughly $10\ \mathrm{m/s}$ — rejects it
immediately.

**The lesson.** Two independent checks each catch this error in seconds: carrying units through the
algebra, and asking whether the number is physically plausible. Both are free, and neither requires
knowing the right answer in advance. This is what "interpret and check" means in the course method, and
it is why it is a step rather than an afterthought.

**TR:** Üç hata: karekök düşmüş, sonuç yanlış, ve **birimler hiç kontrol edilmemiş** — $2gh$'nin birimi
$\mathrm{m^2/s^2}$'dir, bu bir sürat değildir. Ayrıca beş metreden 350 km/h saçmadır ve iki saniyelik bir
makullük kontrolü bunu hemen reddeder. İki bağımsız kontrol de bedavadır ve doğru cevabı önceden bilmeyi
gerektirmez.

</details>

## 6. Exam technique and the error log / Sınav tekniği ve hata kaydı

### In the examination

1. **Read every question before starting.** Do the ones you can see the route for first; marks are the
   same whichever order you collect them in.
2. **Draw, always.** Even for a one-line question. Most of this course's lost marks are signs and angles.
3. **Symbols before numbers.** It is faster, it is checkable, and it is where the partial credit lives.
4. **Carry units through.** A units mismatch is a free error-detector, as Route F showed.
5. **Check plausibility.** A car at $300\ \mathrm{m/s}$, a mass of $-4\ \mathrm{kg}$, a coefficient of
   friction of 7 — say so if you get one and you cannot find the slip. Naming an implausible result earns
   more than presenting it silently.
6. **Three significant figures, with units.** A number without a unit is not an answer.
7. **If you are stuck, write the principle you would use and why.** An examiner can credit a correct
   choice of law. A blank page has nothing to credit.

### The error log — the highest-return revision there is

For every mistake you make between now and the exam, write **one line**:

| Problem | What I did | What was actually wrong | The check that would have caught it |
| --- | --- | --- | --- |
| W04 P6 | used $N = mg$ on an incline | $N = mg\cos\theta$ | $\theta \to 0$ must give $N \to mg$ |

After a dozen entries the fourth column stops being varied: nearly everyone finds two or three errors
they make repeatedly, and those are worth more attention than any new topic. **Revising your own error
log beats re-reading any notebook, because it is the only revision material written specifically for
you.**

**TR:** Bu hafta ile sınav arasındaki her hatan için **tek satır** yaz. On iki kayıttan sonra dördüncü
sütun tekrar etmeye başlar: hemen herkes iki-üç hatayı sürekli yapar ve bunlar yeni bir konudan daha çok
ilgi hak eder. **Kendi hata kaydını tekrar etmek, herhangi bir notu yeniden okumaktan iyidir** — çünkü
sana özel yazılmış tek tekrar materyalidir.

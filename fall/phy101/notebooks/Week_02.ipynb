**Choose by the missing quantity.** If time is not needed, use $v^2=v_0^2+2a\Delta x$. If final velocity is not needed, use $\Delta x=v_0t+\frac12at^2$. These equations require constant acceleration during the interval.

**Algebra slowly:** isolate elapsed time by applying the same operation to both sides:

$$\begin{aligned}
v&=v_0+at,\\
v-v_0&=at &&\text{subtract }v_0,\\
t&=\frac{v-v_0}{a} &&\text{divide by }a\ne0.
\end{aligned}$$

For a car slowing from $12\,\mathrm{m/s}$ to rest with $a=-3\,\mathrm{m/s^2}$,

$$t=\frac{0-12\,\mathrm{m/s}}{-3\,\mathrm{m/s^2}}=4\,\mathrm s.$$

Both numerator and denominator are negative, so the elapsed time is positive. The unit quotient is seconds.

**Signs come from axes.** With upward positive, free-fall acceleration is $a_y=-9.81\,\mathrm{m/s^2}$ both while rising and while falling. At the highest point, velocity is momentarily zero but acceleration is still downward. Speed is the nonnegative magnitude of velocity.

**Square roots:** $t^2=9\,\mathrm{s^2}$ has mathematical roots $\pm3\,\mathrm s$. A time measured after release uses $+3\,\mathrm s$. A quadratic height equation can instead have two valid positive roots, for the upward and downward crossings.

**Review stop:** explain why $v=-5\,\mathrm{m/s}$ and $a=-2\,\mathrm{m/s^2}$ describe increasing speed in the negative direction. After one second, $v=-5+(-2)(1)=-7\,\mathrm{m/s}$: speed rose from $5$ to $7\,\mathrm{m/s}$. Acceleration with a minus sign does not always mean slowing down.

```python
#@title Run once — prepare the demonstration controls
"""Shared demonstration interface, embedded in every PHY101 notebook.

Only standard ipywidgets, IPython and matplotlib are used, so the notebooks stay
self-contained in Colab and in a local Jupyter. The design goals are:

* A slider change must always produce a visible reaction. The status line under
  the controls says "Updating…" immediately and reports the draw time afterwards.
* The graph is replaced through the same Output-widget route that
  ``ipywidgets.interact`` uses (``clear_output(wait=True)`` followed by a fresh
  display), which is the most widely tested path in Colab and Jupyter. No
  output-capturing context is used: ipykernel 7 dispatches widget messages
  concurrently and IPython's capture object breaks that dispatch.
* Live updates while dragging are switched on when a graph draws quickly and
  switched off (update on release) when it draws slowly, so the kernel never
  falls behind a fast slider.
"""
import functools
import sys
import time
import traceback

import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import HTML, clear_output, display

# Force the inline backend. Otherwise a local kernel may choose a desktop
# backend and block at plt.show(), which looks like a frozen notebook.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

_physics_panels = []
_physics_callback_errors = []

# Draw-time thresholds (seconds) for switching live dragging on and off.
PHYSICS_LIVE_ON = 0.12
PHYSICS_LIVE_OFF = 0.25


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; background: #fff; }
.phy101-controls { padding: 0 0 2px; box-sizing: border-box; }
.phy101-status { font-size: 12px; color: #4a5a6a; padding: 0 2px 6px; min-height: 18px; }
.phy101-status.busy { color: #b45309; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-animation-panel { max-width: 100%; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1; background: white; color: #172433; padding: 4px; }
@media (max-width: 650px) {
  .phy101-controls, .phy101-plot-output { width: 100% !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Lay the controls out as a wrapping toolbar with full-length labels."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "310px"
        control.layout.flex = "0 1 310px"
        control.layout.max_width = "100%"
        control.layout.min_width = "0"
        control.layout.margin = "2px 6px 2px 0"
        if isinstance(control, widgets.Button):
            control.layout.width = "auto"
            control.layout.flex = "0 0 auto"
    # Repeat the style inside the widget tree: Colab isolates output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.Box([style] + flat, layout=widgets.Layout(
        display="flex", flex_flow="row wrap", align_items="center",
        width="100%", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        width="100%", min_width="0", max_width="100%",
        height="auto", overflow="visible", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Button-driven demos: a compact control toolbar directly above the result."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="column",
                              align_items="stretch", width="100%"))
    panel.add_class("phy101-panel")
    return panel


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


class PhysicsPanel:
    """Controls, a status line and one Output widget that shows the latest result."""

    def __init__(self, function, controls):
        self.f = function
        self.controls = controls
        self.out = _physics_output(widgets.Output())
        self.status = widgets.HTML(value="")
        self.status.add_class("phy101-status")
        self.seconds = None
        self.live = True
        self.updates = 0
        self.last_outputs = 0
        self.figures = 0
        self.error = None
        visible = []
        for control in controls.values():
            if isinstance(control, widgets.fixed):
                continue
            visible.append(control)
            if hasattr(control, "continuous_update"):
                control.continuous_update = True
            control.observe(self._changed, names="value")
        self.widget = widgets.VBox([_physics_controls(visible), self.status, self.out],
                                   layout=widgets.Layout(width="100%"))
        self.widget.add_class("phy101-panel")
        self.children = self.widget.children
        _physics_panels.append(self)
        self.render()

    # Compatibility with the earlier validation code.
    @property
    def layout(self):
        return self.widget.layout

    def _changed(self, change):
        self.render()

    def _set_live(self, live):
        if live == self.live:
            return
        self.live = live
        for control in self.controls.values():
            if hasattr(control, "continuous_update"):
                control.continuous_update = live

    def render(self):
        self.status.value = "⏳ Updating… / Güncelleniyor…"
        self.status.add_class("busy")
        started = time.perf_counter()
        kwargs = {name: control.value for name, control in self.controls.items()}
        self.error = None
        self.figures = 0
        # Count everything the demonstration shows (figures, HTML, animations, text)
        # by wrapping the display publisher and stdout for this draw only.
        shell = get_ipython()
        publisher = getattr(shell, "display_pub", None) if shell is not None else None
        if publisher is not None:
            original_publish = publisher.publish

            def counting_publish(*args, **kwargs):
                self.figures += 1
                return original_publish(*args, **kwargs)
            publisher.publish = counting_publish
        stdout = sys.stdout
        original_write = stdout.write

        def counting_write(text):
            if text.strip():
                self.figures += 1
            return original_write(text)
        stdout.write = counting_write
        try:
            # The previous result stays visible until the new one arrives.
            with self.out:
                clear_output(wait=True)
                try:
                    result = self.f(**kwargs)
                    from ipywidgets.widgets.interaction import show_inline_matplotlib_plots
                    show_inline_matplotlib_plots()
                    if result is not None:
                        display(result)
                except Exception:
                    self.error = traceback.format_exc()
                    _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), self.error))
                    print(self.error)
        finally:
            if publisher is not None and publisher.__dict__.get("publish") is counting_publish:
                del publisher.publish
            if stdout.__dict__.get("write") is counting_write:
                del stdout.write
        self.last_outputs = self.figures
        self.seconds = time.perf_counter() - started
        self.updates += 1
        if self.seconds > PHYSICS_LIVE_OFF:
            self._set_live(False)
        elif self.seconds < PHYSICS_LIVE_ON:
            self._set_live(True)
        self.status.remove_class("busy")
        mode = ("updates while you drag / sürüklerken güncellenir" if self.live
                else "updates when you release the slider / kaydırıcıyı bırakınca güncellenir")
        self.status.value = (f"✓ Drawn in {self.seconds:.2f} s · {mode}" if self.error is None
                             else "⚠ The demonstration reported an error; see the message below.")


def physics_interactive(function, **controls):
    """Build a panel like ipywidgets.interactive, returning the panel object."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        return function(*args, **kwargs)

    return PhysicsPanel(checked, controls)


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel.widget
    function.panel = panel
    display(panel.widget)
    return function
```

#### Important Distinctions

- **Displacement vs Distance:** Walking $3\,\mathrm m$ east and $3\,\mathrm m$ west gives $\Delta x=0$ and $d=6\,\mathrm m$. No motion gives zero distance, so distance is not *always positive*.
- **Speed vs Velocity:** Speed is $|v|\ge0$; velocity includes direction.
- **Velocity vs Acceleration:** Velocity describes change of position; acceleration describes change of velocity. A negative acceleration does not by itself mean slowing down.

**Small example — one trip, two averages.** A robot moves from $x=1\,\mathrm m$ to $x=5\,\mathrm m$, then back to $x=3\,\mathrm m$, in $6\,\mathrm s$ total.

$$
\begin{aligned}
\Delta x&=3-1=2\,\mathrm m, & d&=(5-1)+(5-3)=6\,\mathrm m,\\
\bar v&=\frac{2\,\mathrm m}{6\,\mathrm s}=0.333\,\mathrm{m/s}, & \text{average speed}&=\frac{6\,\mathrm m}{6\,\mathrm s}=1.00\,\mathrm{m/s}.
\end{aligned}
$$

### The Kinematic Equations (Constant Acceleration)

Use these equations only during an interval with **constant acceleration**. Let $t$ be time since that interval began, $x_0,v_0$ its initial values, and $x,v$ its final values.

$$\boxed{v=v_0+at}$$

$$\boxed{x=x_0+v_0t+\tfrac12at^2}$$

$$\boxed{v^2=v_0^2+2a(x-x_0)}$$

$$\boxed{x-x_0=\tfrac12(v_0+v)t}$$

There are only four, and the fourth follows from the first three. **Each one is missing a different quantity**, which is what makes the table below the fastest way to choose: find the quantity the problem does not give you and does not ask for, and use the equation that leaves it out.

#### Sign Convention

With right/up positive, $v>0$ means right/up and $v<0$ means left/down. When $v$ and $a$ have the same sign, speed increases; opposite signs mean speed decreases until a possible reversal. At $v=0$, inspect what acceleration does immediately afterward.


### Slopes and areas describe change and accumulation

**Meaning and geometry.** The slope of an $x$–$t$ graph is signed velocity: how much position changes per second. The slope of a $v$–$t$ graph is acceleration: how much velocity changes per second. Draw a small right triangle on each graph and label both axis units. The slope units then follow by division, $\mathrm{m/s}$ and $\mathrm{m/s^2}$.

The signed area under a velocity graph accumulates displacement. A constant velocity makes a rectangle; a steadily changing velocity makes a trapezoid. This is why, **only for constant acceleration**,

$$\Delta x=\frac{v_i+v_f}{2}\Delta t=v_i\Delta t+\frac12a(\Delta t)^2.$$

The rectangle represents motion at the initial velocity; the triangle represents the additional displacement caused by acceleration. The factors are geometric, not calculator instructions to memorise. Distance instead accumulates $|v|$.

### The calculus bridge: velocity is a slope / Hız bir eğimdir

You have four kinematic equations (three independent ones) and they only work when the acceleration is **constant**. Behind them
sits one idea that always works, and you can read it straight off the graphs you have already drawn.

**Velocity is the slope of the position graph.**

$$v = \frac{\Delta x}{\Delta t} \quad \text{over an interval}, \qquad
v = \frac{dx}{dt} \quad \text{at an instant.}$$

The second expression is the first one with the interval shrunk until it is a single point. That is all
$d/dt$ means: *the slope of the graph right here*, rather than the average slope across a stretch. Read
the symbol as the sentence "the slope of $x$ against $t$" and it stops being intimidating.

**Acceleration is the slope of the velocity graph**, by exactly the same reasoning:

$$a = \frac{dv}{dt} = \text{the slope of } v \text{ against } t.$$

And going the other way, **displacement is the area under the velocity graph**:

$$\Delta x = \int v\,dt = \text{the signed area under } v \text{ against } t.$$

This is why the $v$–$t$ area demonstration in this notebook matters more than it looks: that area *is*
the displacement, and area below the axis counts as motion backwards.

| You want | Read this off the graph | In symbols |
| --- | --- | --- |
| Velocity at an instant | slope of the $x$–$t$ curve | $dx/dt$ |
| Acceleration at an instant | slope of the $v$–$t$ curve | $dv/dt$ |
| Displacement over an interval | area under the $v$–$t$ curve | $\int v\,dt$ |
| Change in velocity | area under the $a$–$t$ curve | $\int a\,dt$ |

**Where the familiar equations come from.** If $a$ is constant, the $a$–$t$ graph is a flat line, so the
area under it is just $a t$ — giving $v = v_0 + at$. The $v$–$t$ graph is then a straight line, and the
area under a straight line is a rectangle plus a triangle: $v_0 t + \tfrac12 a t^2$, which is exactly
$\Delta x$. **The equations you are memorising are two areas.** That is worth knowing, because when the
acceleration is *not* constant the equations break but the areas and slopes do not.

altındaki (işaretli) alandır. $d/dt$ yalnızca "tam burada eğim" demektir — aralığı bir noktaya kadar

### Animated 1D Motion

Watch a ball move under constant acceleration. The animation shows both the ball's position and the trail it leaves behind.

```python
#@title Run the animation — observe the physical motion
# ---- Animated 1D Constant-Acceleration Motion ----

# Parameters
x0_anim = 0.0      # initial position (m)
v0_anim = 8.0      # initial velocity (m/s)
a_anim = -2.0      # acceleration (m/s^2) -- braking
t_max_anim = 6.0   # total time (s)
dt_anim = 0.05

t_arr = np.arange(0, t_max_anim, dt_anim)
x_arr = x0_anim + v0_anim * t_arr + 0.5 * a_anim * t_arr**2
v_arr = v0_anim + a_anim * t_arr

fig, (ax_track, ax_xt, ax_vt) = plt.subplots(3, 1, figsize=(10, 8),
                                              gridspec_kw={'height_ratios': [1, 2, 2]})
fig.suptitle('1D Motion: Constant Acceleration', fontsize=14)
fig.tight_layout(pad=2.5)

# Track (top)
ax_track.set_xlim(min(x_arr) - 2, max(x_arr) + 2)
ax_track.set_ylim(-0.5, 0.5)
ax_track.set_yticks([])
ax_track.set_xlabel('Position x (m)')
ax_track.axhline(0, color='gray', lw=1)
ball, = ax_track.plot([], [], 'ro', markersize=18)
trail, = ax_track.plot([], [], 'r-', alpha=0.3, lw=2)

# x-t graph (middle)
ax_xt.set_xlim(0, t_max_anim)
ax_xt.set_ylim(min(x_arr) - 2, max(x_arr) + 2)
ax_xt.set_xlabel('Time (s)')
ax_xt.set_ylabel('Position x (m)')
ax_xt.grid(True, alpha=0.3)
ax_xt.plot(t_arr, x_arr, 'b--', alpha=0.3, label='Full path')
line_xt, = ax_xt.plot([], [], 'b-', lw=2, label='x(t)')
dot_xt, = ax_xt.plot([], [], 'bo', markersize=8)
ax_xt.legend(loc='upper left')

# v-t graph (bottom)
ax_vt.set_xlim(0, t_max_anim)
ax_vt.set_ylim(min(v_arr) - 1, max(v_arr) + 1)
ax_vt.set_xlabel('Time (s)')
ax_vt.set_ylabel('Velocity v (m/s)')
ax_vt.grid(True, alpha=0.3)
ax_vt.axhline(0, color='k', lw=0.5)
ax_vt.plot(t_arr, v_arr, 'r--', alpha=0.3, label='Full v(t)')
line_vt, = ax_vt.plot([], [], 'r-', lw=2, label='v(t)')
dot_vt, = ax_vt.plot([], [], 'ro', markersize=8)
ax_vt.legend(loc='upper right')

time_text = ax_track.text(0.02, 0.85, '', transform=ax_track.transAxes, fontsize=11)

def init_1d():
    ball.set_data([], [])
    trail.set_data([], [])
    line_xt.set_data([], [])
    dot_xt.set_data([], [])
    line_vt.set_data([], [])
    dot_vt.set_data([], [])
    time_text.set_text('')
    return ball, trail, line_xt, dot_xt, line_vt, dot_vt, time_text

def animate_1d(i):
    idx = min(i * 3, len(t_arr) - 1)  # step through faster
    ball.set_data([x_arr[idx]], [0])
    trail.set_data(x_arr[:idx+1], np.zeros(idx+1))
    line_xt.set_data(t_arr[:idx+1], x_arr[:idx+1])
    dot_xt.set_data([t_arr[idx]], [x_arr[idx]])
    line_vt.set_data(t_arr[:idx+1], v_arr[:idx+1])
    dot_vt.set_data([t_arr[idx]], [v_arr[idx]])
    time_text.set_text(f't = {t_arr[idx]:.2f} s | x = {x_arr[idx]:.1f} m | v = {v_arr[idx]:.1f} m/s')
    return ball, trail, line_xt, dot_xt, line_vt, dot_vt, time_text

anim_1d = FuncAnimation(fig, animate_1d, init_func=init_1d,
                        frames=physics_frames(len(t_arr)//3), interval=physics_interval(len(t_arr)//3, 50), blit=True)
plt.close(fig)
physics_animation_html(anim_1d)
```

```python
#@title Run the demonstration — predict, adjust, observe
# ---- Compare x(t), v(t), and a(t) together ----

def plot_kinematics(x0=0.0, v0=5.0, a=-1.0, t_end=10.0):
    t = np.linspace(0, t_end, 500)
    x = x0 + v0 * t + 0.5 * a * t**2
    v = v0 + a * t
    a_const = np.full_like(t, a)

    fig, axes = plt.subplots(1, 3, figsize=(11.6, 3.6), layout='constrained')
    fig.suptitle('Constant acceleration: three views of the same motion', fontsize=12)

    axes[0].plot(t, x, color='#2155a3', lw=2)
    axes[0].set(xlabel='Time t (s)', ylabel='Position x (m)', title='Position')
    if a != 0:
        t_turn = -v0 / a
        if 0 < t_turn < t_end:
            x_turn = x0 + v0 * t_turn + 0.5 * a * t_turn**2
            axes[0].plot(t_turn, x_turn, 'o', color='#bb2525', markersize=6)
            axes[0].set_title(f'Position: v = 0 at {t_turn:.1f} s', fontsize=11)

    axes[1].plot(t, v, color='#bb2525', lw=2)
    axes[1].fill_between(t, v, alpha=0.15, color='#bb2525')
    axes[1].set(xlabel='Time t (s)', ylabel='Velocity v (m/s)',
                title='Velocity: signed area = displacement')

    axes[2].plot(t, a_const, color='#237c49', lw=2)
    axes[2].set(xlabel='Time t (s)', ylabel=r'Acceleration a (m/s$^2$)',
                title='Acceleration', ylim=(a - 2, a + 2))
    for ax in axes:
        ax.grid(True, alpha=0.25)
        ax.axhline(0, color='#555555', lw=0.6)
        ax.tick_params(labelsize=9)
        ax.xaxis.label.set_size(10)
        ax.yaxis.label.set_size(10)
        ax.title.set_fontsize(10)

    # Keep the current values with all three graphs, in the same output.
    fig.supxlabel(
        rf'$x_0={x0:.1f}\,\mathrm{{m}},\quad v_0={v0:.1f}\,\mathrm{{m/s}},\quad '
        rf'a={a:.1f}\,\mathrm{{m/s^2}}$' + '\n' +
        rf'At $t={t_end:.1f}\,\mathrm{{s}}$: $x={x[-1]:.2f}\,\mathrm{{m}},\quad '
        rf'v={v[-1]:.2f}\,\mathrm{{m/s}},\quad \Delta x={x[-1]-x0:.2f}\,\mathrm{{m}}$',
        fontsize=10)
    plt.show()

physics_interact(plot_kinematics,
         x0=FloatSlider(min=-20, max=20, step=1, value=0, description='x0 (m):'),
         v0=FloatSlider(min=-10, max=10, step=0.5, value=5, description='v0 (m/s):'),
         a=FloatSlider(min=-5, max=5, step=0.25, value=-1, description='a (m/s^2):'),
         t_end=FloatSlider(min=2, max=20, step=1, value=10, description='t_end (s):'));
```

### Free Fall Simulator

Choose upward positive. In the ideal model, gravity supplies the constant acceleration $a_y=-g$, where $g=9.81\,\mathrm{m/s^2}$ is a **positive magnitude**.

$$y=y_0+v_0t-\frac12gt^2,\qquad v_y=v_0-gt.$$

For an upward launch ($v_0>0$), $v_y=0$ at $t_{\rm up}=\frac{v_0}{g}$. The **rise above launch** is $\Delta h=\frac{v_0^2}{2g}$; the absolute maximum height is $y_{\max}=y_0+\Delta h$. Upward and downward flight times are equal only when comparing the same starting and ending heights.

**Small check:** With $v_0=9.81\,\mathrm{m/s}$, after $1\,\mathrm s$ the ball has $v_y=9.81-9.81(1)=0$, while $a_y=-9.81\,\mathrm{m/s^2}$ still. Its rise is $\frac12(9.81)(1^2)=4.91\,\mathrm m$.

**Türkçe:** Tepe noktası bir duraklama anıdır, yerçekiminin kapandığı bir an değildir. Yüksekliği zeminden mi, atış noktasından mı ölçtüğünü mutlaka belirt.


```python
#@title Run the animation — observe the physical motion
# ---- Animated Free Fall ----

v0_ff = 15.0   # initial upward velocity (m/s)
g_val = 9.81   # gravitational acceleration (m/s^2)

# Time of flight (thrown up, lands back at y=0)
t_flight = 2 * v0_ff / g_val
dt_ff = 0.02
t_ff = np.arange(0, t_flight + dt_ff, dt_ff)
y_ff = v0_ff * t_ff - 0.5 * g_val * t_ff**2
v_ff = v0_ff - g_val * t_ff

# Max height
t_top = v0_ff / g_val
h_max = v0_ff**2 / (2 * g_val)

fig_ff, (ax_ball, ax_yt, ax_vt2) = plt.subplots(1, 3, figsize=(14, 7),
                                                  gridspec_kw={'width_ratios': [1, 2, 2]})
fig_ff.suptitle(f'Free Fall: v0 = {v0_ff} m/s upward', fontsize=14)

# Ball column
ax_ball.set_xlim(-1, 1)
ax_ball.set_ylim(-2, h_max + 3)
ax_ball.set_xticks([])
ax_ball.set_ylabel('Height y (m)')
ax_ball.axhline(0, color='brown', lw=3)  # ground
ball_ff, = ax_ball.plot([], [], 'ro', markersize=20)
trail_ff, = ax_ball.plot([], [], 'r-', alpha=0.2, lw=1)
ax_ball.axhline(h_max, color='gray', ls='--', alpha=0.5)
ax_ball.text(0.5, h_max + 0.3, f'h_max = {h_max:.1f} m', ha='center', fontsize=9)

# y-t graph
ax_yt.set_xlim(0, t_flight + 0.5)
ax_yt.set_ylim(-2, h_max + 3)
ax_yt.set_xlabel('Time (s)')
ax_yt.set_ylabel('Height y (m)')
ax_yt.grid(True, alpha=0.3)
ax_yt.axhline(0, color='k', lw=0.5)
ax_yt.plot(t_ff, y_ff, 'b--', alpha=0.2)
line_yt, = ax_yt.plot([], [], 'b-', lw=2)
dot_yt, = ax_yt.plot([], [], 'bo', markersize=8)

# v-t graph
ax_vt2.set_xlim(0, t_flight + 0.5)
ax_vt2.set_ylim(min(v_ff) - 2, max(v_ff) + 2)
ax_vt2.set_xlabel('Time (s)')
ax_vt2.set_ylabel('Velocity v (m/s)')
ax_vt2.grid(True, alpha=0.3)
ax_vt2.axhline(0, color='k', lw=0.5)
ax_vt2.plot(t_ff, v_ff, 'r--', alpha=0.2)
line_vt2, = ax_vt2.plot([], [], 'r-', lw=2)
dot_vt2, = ax_vt2.plot([], [], 'ro', markersize=8)

info_ff = ax_ball.text(0, -1.5, '', ha='center', fontsize=9)

def init_ff():
    ball_ff.set_data([], [])
    trail_ff.set_data([], [])
    line_yt.set_data([], [])
    dot_yt.set_data([], [])
    line_vt2.set_data([], [])
    dot_vt2.set_data([], [])
    info_ff.set_text('')
    return ball_ff, trail_ff, line_yt, dot_yt, line_vt2, dot_vt2, info_ff

def animate_ff(i):
    idx = min(i * 2, len(t_ff) - 1)
    ball_ff.set_data([0], [y_ff[idx]])
    trail_ff.set_data(np.zeros(idx+1), y_ff[:idx+1])
    line_yt.set_data(t_ff[:idx+1], y_ff[:idx+1])
    dot_yt.set_data([t_ff[idx]], [y_ff[idx]])
    line_vt2.set_data(t_ff[:idx+1], v_ff[:idx+1])
    dot_vt2.set_data([t_ff[idx]], [v_ff[idx]])
    info_ff.set_text(f't={t_ff[idx]:.1f}s\nv={v_ff[idx]:.1f}m/s')
    return ball_ff, trail_ff, line_yt, dot_yt, line_vt2, dot_vt2, info_ff

anim_ff = FuncAnimation(fig_ff, animate_ff, init_func=init_ff,
                        frames=physics_frames(len(t_ff)//2), interval=physics_interval(len(t_ff)//2, 40), blit=True)
plt.close(fig_ff)
physics_animation_html(anim_ff)
```

```python
#@title Run the demonstration — predict, adjust, observe
# ---- Interactive Drop from Height ----

def free_fall_drop(height=20.0, v0_up=0.0):
    g = 9.81
    # y = height + v0_up * t - 0.5 * g * t^2 = 0
    # Solve: 0.5*g*t^2 - v0_up*t - height = 0
    discriminant = v0_up**2 + 2 * g * height
    if discriminant < 0:
        print("Invalid parameters.")
        return
    t_hit = (v0_up + np.sqrt(discriminant)) / g
    v_hit = v0_up - g * t_hit  # negative = downward
    
    t = np.linspace(0, t_hit, 300)
    y = height + v0_up * t - 0.5 * g * t**2
    v = v0_up - g * t
    
    # Max height (if thrown upward)
    if v0_up > 0:
        t_top = v0_up / g
        y_max = height + v0_up**2 / (2 * g)
    else:
        t_top = 0
        y_max = height
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
    
    # y-t graph
    ax1.plot(t, y, 'b-', lw=2)
    ax1.axhline(0, color='brown', lw=2, label='Ground')
    ax1.set_xlabel('Time (s)')
    ax1.set_ylabel('Height y (m)')
    ax1.set_title('Height vs Time')
    ax1.grid(True, alpha=0.3)
    ax1.plot(t_hit, 0, 'rv', markersize=15, label=f'Impact at t={t_hit:.2f}s')
    if v0_up > 0:
        ax1.plot(t_top, y_max, 'g^', markersize=12, label=f'Peak: {y_max:.1f}m')
    ax1.legend()
    
    # v-t graph
    ax2.plot(t, v, 'r-', lw=2)
    ax2.axhline(0, color='k', lw=0.5)
    ax2.set_xlabel('Time (s)')
    ax2.set_ylabel('Velocity (m/s)')
    ax2.set_title('Velocity vs Time')
    ax2.grid(True, alpha=0.3)
    ax2.plot(t_hit, v_hit, 'rv', markersize=12, label=f'v_impact = {v_hit:.2f} m/s')
    ax2.legend()
    
    # Summary
    speed_hit = abs(v_hit)
    speed_kmh = speed_hit * 3.6
    fig.text(0.5, 0.035,
             f'Time to ground: {t_hit:.2f} s | Impact speed: {speed_hit:.1f} m/s ({speed_kmh:.0f} km/h)',
             ha='center', fontsize=12, bbox=dict(boxstyle='round', facecolor='lightyellow'))
    
    plt.tight_layout(rect=[0, 0.10, 1, 1])
    plt.show()

physics_interact(free_fall_drop,
         height=FloatSlider(min=1, max=100, step=1, value=20, description='Height (m):'),
         v0_up=FloatSlider(min=-10, max=20, step=1, value=0, description='v0 up (m/s):'));
```

### Braking Distance Calculator

Separate the reaction interval (constant speed in this model) from braking (constant negative acceleration). Let $b=|a|>0$ be the braking deceleration magnitude.

$$d_{\rm reaction}=v_0t_r,\qquad d_{\rm brake}=\frac{v_0^2}{2b},\qquad d_{\rm stop}=d_{\rm reaction}+d_{\rm brake}.$$

<p><strong>Illustrative road condition:</strong> Dry asphalt</p>
<ul>
<li><p><strong>Deceleration magnitude used for a classroom model:</strong> $7$–$10\,\mathrm{m/s^2}$</p></li>
</ul>
<p><strong>Illustrative road condition:</strong> Wet asphalt</p>
<ul>
<li><p><strong>Deceleration magnitude used for a classroom model:</strong> $4$–$7\,\mathrm{m/s^2}$</p></li>
</ul>
<p><strong>Illustrative road condition:</strong> Icy road</p>
<ul>
<li><p><strong>Deceleration magnitude used for a classroom model:</strong> $1$–$3\,\mathrm{m/s^2}$</p></li>
</ul>

These are illustrative ranges, not guarantees for a particular vehicle or road. Use the values supplied in each question.

**Worked comparison:** For $v_0=10\,\mathrm{m/s}$, $t_r=1\,\mathrm s$, and $b=5\,\mathrm{m/s^2}$, reaction distance is $10\,\mathrm m$ and braking distance is $100/10=10\,\mathrm m$: total $20\,\mathrm m$. Doubling speed gives $20+400/10=60\,\mathrm m$, not $40\,\mathrm m$.

**Türkçe:** Tepki yolu hızla doğru orantılı, fren yolu hızın karesiyle orantılıdır. Hızı ikiye katlayınca toplam durma yolu için tek bir “iki kat” kuralı kullanamayız.

```python
#@title Run the demonstration — predict, adjust, observe
# ---- Interactive Braking Distance Calculator ----

def braking_distance(speed_kmh=80, reaction_time=1.5, decel=7.0):
    v0 = speed_kmh / 3.6  # convert to m/s
    
    # Reaction phase
    d_reaction = v0 * reaction_time
    
    # Braking phase: v^2 = v0^2 - 2*a*d  =>  d = v0^2 / (2*a)
    d_brake = v0**2 / (2 * decel)
    d_total = d_reaction + d_brake
    
    # Time to stop after braking starts
    t_brake = v0 / decel
    
    # Build time arrays
    t1 = np.linspace(0, reaction_time, 100)
    t2 = np.linspace(reaction_time, reaction_time + t_brake, 100)
    
    # Positions
    x1 = v0 * t1
    x2 = d_reaction + v0 * (t2 - reaction_time) - 0.5 * decel * (t2 - reaction_time)**2
    
    # Velocities
    v1 = np.full_like(t1, v0)
    v2 = v0 - decel * (t2 - reaction_time)
    
    fig, axes = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
    fig.suptitle(f'Braking from {speed_kmh} km/h ({v0:.1f} m/s)', fontsize=14)
    
    # Position plot
    axes[0].plot(t1, x1, 'orange', lw=2.5, label=f'Reaction: {d_reaction:.1f} m')
    axes[0].plot(t2, x2, 'red', lw=2.5, label=f'Braking: {d_brake:.1f} m')
    axes[0].axhline(d_total, color='gray', ls='--', alpha=0.5)
    axes[0].set_ylabel('Distance (m)')
    axes[0].grid(True, alpha=0.3)
    axes[0].legend(fontsize=11)
    # Totals are reported in the reserved footer, not over the curves.
    
    # Velocity plot
    axes[1].plot(t1, v1 * 3.6, 'orange', lw=2.5, label='Reaction (constant v)')
    axes[1].plot(t2, v2 * 3.6, 'red', lw=2.5, label='Braking (decelerating)')
    axes[1].fill_between(t2, v2 * 3.6, alpha=0.15, color='red')
    axes[1].axhline(0, color='k', lw=0.5)
    axes[1].set_ylabel('Speed (km/h)')
    axes[1].set_xlabel('Time (s)')
    axes[1].grid(True, alpha=0.3)
    axes[1].legend(fontsize=11)
    
    # Visual bar at bottom
    fig.text(0.5, 0.035,
             f'Reaction dist: {d_reaction:.1f} m + Braking dist: {d_brake:.1f} m = Total: {d_total:.1f} m  |  '
             f'Stop time: {reaction_time + t_brake:.1f} s',
             ha='center', fontsize=11, bbox=dict(boxstyle='round', facecolor='lightyellow'))
    
    plt.tight_layout(rect=[0, 0.10, 1, 0.96])
    plt.show()

physics_interact(braking_distance,
         speed_kmh=FloatSlider(min=20, max=200, step=10, value=80, description='Speed (km/h):'),
         reaction_time=FloatSlider(min=0.2, max=3.0, step=0.1, value=1.5, description='Reaction (s):'),
         decel=FloatSlider(min=1, max=10, step=0.5, value=7, description='Decel (m/s^2):'));
```

### Comparing Motions: v-t Area Visualization

Signed area under the velocity graph gives displacement, $\Delta x=\int v\,dt$. Area below zero contributes a negative displacement; distance adds the magnitudes of forward and backward pieces.

**Predict:** At $v_0=10\,\mathrm{m/s}$ and $a=-2\,\mathrm{m/s^2}$, turning occurs at $5\,\mathrm s$. By $8\,\mathrm s$, the areas are $+\frac12(5)(10)=25\,\mathrm m$ and $-\frac12(3)(6)=-9\,\mathrm m$. Displacement is $16\,\mathrm m$ and distance is $34\,\mathrm m$.

**Türkçe:** Grafik geri dönüşü de kapsar. Konumun en büyük veya en küçük olduğu noktada hız sıfırdır; eğrinin hangi yöne kıvrıldığı bunun tepe mi çukur mu olduğunu belirler.

```python
#@title Run the demonstration — predict, adjust, observe
# ---- v-t Area = Displacement Demo ----

def vt_area_demo(v0=10.0, a=-2.0, t_end=8.0):
    t = np.linspace(0, t_end, 500)
    v = v0 + a * t
    x = v0 * t + 0.5 * a * t**2
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
    
    # v-t with shaded area
    ax1.plot(t, v, 'r-', lw=2)
    ax1.axhline(0, color='k', lw=0.5)
    ax1.set_ylabel('Velocity (m/s)')
    ax1.set_title('Velocity-Time Graph (shaded area = displacement)', fontsize=13)
    ax1.grid(True, alpha=0.3)
    
    # Shade positive area green, negative area red
    ax1.fill_between(t, v, where=(v >= 0), alpha=0.3, color='green', label='Forward displacement')
    ax1.fill_between(t, v, where=(v < 0), alpha=0.3, color='red', label='Backward displacement')
    ax1.legend(fontsize=10)
    
    # x-t showing displacement
    ax2.plot(t, x, 'b-', lw=2)
    ax2.axhline(0, color='k', lw=0.5)
    ax2.set_ylabel('Position (m)')
    ax2.set_xlabel('Time (s)')
    ax2.set_title('Position-Time Graph', fontsize=13)
    ax2.grid(True, alpha=0.3)
    
    turn_note = "No turning point in this interval."
    # Mark the turning point; its numerical label belongs in the footer.
    if a != 0:
        t_turn = -v0 / a
        if 0 < t_turn < t_end:
            x_max = v0 * t_turn + 0.5 * a * t_turn**2
            ax2.plot(t_turn, x_max, 'ro', markersize=10)
            turn_note = f'Red point: v = 0 at t = {t_turn:.2f} s, x = {x_max:.2f} m.'
    
    final_x = x[-1]
    total_displacement = final_x
    fig.text(0.5, 0.025,
             f'Net displacement at t={t_end:.0f}s: {total_displacement:.1f} m\n{turn_note}',
             ha='center', fontsize=12, fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='lightyellow'))
    
    plt.tight_layout(rect=[0, 0.13, 1, 1])
    plt.show()

physics_interact(vt_area_demo,
         v0=FloatSlider(min=-10, max=15, step=0.5, value=10, description='v0 (m/s):'),
         a=FloatSlider(min=-5, max=5, step=0.25, value=-2, description='a (m/s^2):'),
         t_end=FloatSlider(min=2, max=15, step=0.5, value=8, description='t_end (s):'));
```

### Worked Example 1: Constant Acceleration

**Problem:** A car starts from rest and accelerates uniformly at $3\,\mathrm{m/s^2}$.
- (a) What is its speed after $5\,\mathrm s$?
- (b) How far has it traveled in $5\,\mathrm s$?
- (c) How far has it traveled when it reaches $30\,\mathrm{m/s}$?

#### Hand calculation

Choose forward positive. The initial velocity is $v_0=0$ and acceleration is $a=+3\,\mathrm{m/s^2}$. Let $\Delta x$ denote displacement from the starting point.

**(a) Substitute into the velocity equation:**

$$\begin{aligned}
v&=v_0+at,\\
 &=0+(3\,\mathrm{m/s^2})(5\,\mathrm s)=15\,\mathrm{m/s}.
\end{aligned}$$

**(b) Square time before multiplying:**

$$\begin{aligned}
\Delta x&=v_0t+\frac12at^2,\\
 &=0+\frac12(3\,\mathrm{m/s^2})(5\,\mathrm s)^2,\\
 &=\frac12(3)(25)\,\mathrm m=37.5\,\mathrm m.
\end{aligned}$$

The unit cancellation is $(\mathrm{m/s^2})(\mathrm{s^2})=\mathrm m$.

**(c) Choose the equation without time.** This is a new final event, when $v=30\,\mathrm{m/s}$:

$$\begin{aligned}
v^2&=v_0^2+2a\Delta x,\\
(30\,\mathrm{m/s})^2&=0+2(3\,\mathrm{m/s^2})\Delta x,\\
\Delta x&=\frac{900\,\mathrm{m^2/s^2}}{6\,\mathrm{m/s^2}}=150\,\mathrm m.
\end{aligned}$$

The last step divides both sides by $6\,\mathrm{m/s^2}$. This event occurs at $t=v/a=10\,\mathrm s$.

**Check:** Doubling elapsed time from rest at constant acceleration quadruples distance: $150=4(37.5)$.

**Türkçe:** (c) şıkkında süre artık beş saniye değildir. Yeni son hıza ait denklemi kur; sonra iki tarafı aynı ivme terimine böl. Yolun birimi metre kalmalıdır.

```python
#@title Optional numerical check — the algebra is explained above
# Worked Example 1
v0 = 0       # starts from rest
a = 3.0      # m/s^2
t = 5.0      # s

# (a) v = v0 + a*t
v = v0 + a * t
print(f"(a) Speed after {t} s: v = {v0} + {a}*{t} = {v:.1f} m/s")

# (b) x = v0*t + 0.5*a*t^2
x = v0 * t + 0.5 * a * t**2
print(f"(b) Distance in {t} s: x = {x:.1f} m")

# (c) v^2 = v0^2 + 2*a*x  =>  x = (v^2 - v0^2) / (2*a)
v_target = 30.0
x_30 = (v_target**2 - v0**2) / (2 * a)
print(f"(c) Distance to reach {v_target} m/s: x = ({v_target}^2 - {v0}^2) / (2*{a}) = {x_30:.1f} m")
```

### Worked Example 2: Free Fall

**Problem:** A ball is thrown straight up from the ground with $v_0=20\,\mathrm{m/s}$.
- (a) How high does it go?
- (b) How long until it returns to the ground?
- (c) What is its velocity when it hits the ground?

#### Hand calculation with one sign convention

Take upward positive, $y_0=0$, $v_0=+20\,\mathrm{m/s}$, and $a_y=-g=-9.81\,\mathrm{m/s^2}$. Neglect air resistance. Let $H$ be the rise above the launch point.

**(a) At the peak, set velocity to zero:**

$$\begin{aligned}
0&=v_0^2-2gH,\\
2gH&=v_0^2 &&\text{add }2gH\text{ to both sides},\\
H&=\frac{v_0^2}{2g} &&\text{divide both sides by }2g,\\
 &=\frac{(20\,\mathrm{m/s})^2}{2(9.81\,\mathrm{m/s^2})}=20.4\,\mathrm m.
\end{aligned}$$

**(b) Return means the height is zero again.** Factor the time equation:

$$\begin{aligned}
0&=v_0t-\frac12gt^2=t\left(v_0-\frac12gt\right),\\
t&=0\quad\text{or}\quad v_0-\frac12gt=0,\\
t_{\rm return}&=\frac{2v_0}{g}=\frac{2(20\,\mathrm{m/s})}{9.81\,\mathrm{m/s^2}}=4.08\,\mathrm s.
\end{aligned}$$

The zero root is the launch event; the positive root is the return event.

**(c) Use the signed velocity equation:**

$$v_y=v_0-gt_{\rm return}=20-9.81(4.08)\approx-20.0\,\mathrm{m/s}.$$

The negative sign means downward. Impact **speed** is the magnitude, $20.0\,\mathrm{m/s}$.

**Check:** $t_{\rm peak}=\frac{v_0}{g}=2.04\,\mathrm s$ is half the full flight because launch and landing heights match. Gravity remains downward throughout.

**Türkçe:** Tepe noktasında $v_y=0$ koyuyoruz, $a_y=0$ değil. Çarpanlara ayırma iki farklı olayı görünür yapar: atış ve yere dönüş. Son hızdaki eksi, süratin değil yönün bilgisidir.


```python
#@title Optional numerical check — the algebra is explained above
# Worked Example 2
v0 = 20.0   # m/s upward
g = 9.81    # m/s^2

# (a) At max height, v = 0
# v^2 = v0^2 - 2*g*h  =>  0 = v0^2 - 2*g*h  =>  h = v0^2 / (2*g)
h_max = v0**2 / (2 * g)
print(f"(a) Maximum height: h = {v0}^2 / (2*{g}) = {h_max:.2f} m")

# (b) y = v0*t - 0.5*g*t^2 = 0  =>  t(v0 - 0.5*g*t) = 0
# t = 0 (launch) or t = 2*v0/g
t_return = 2 * v0 / g
print(f"(b) Time to return: t = 2*{v0}/{g} = {t_return:.2f} s")

# (c) v = v0 - g*t
v_ground = v0 - g * t_return
print(f"(c) Velocity at ground: v = {v0} - {g}*{t_return:.2f} = {v_ground:.2f} m/s")
print(f"    (negative means downward, speed = {abs(v_ground):.2f} m/s = same as launch speed!)")
```

### Worked Example 3: Braking Problem

**Problem:** A driver going $100\,\mathrm{km/h}$ sees a child $80\,\mathrm m$ ahead. Reaction time is $1\,\mathrm s$, and braking deceleration has magnitude $8\,\mathrm{m/s^2}$. Does the car stop in time?

#### Hand calculation in two phases

Choose forward positive. First convert the initial speed; the conversion ratios cancel kilometres and hours:

$$v_0=100\,\frac{\mathrm{km}}{\mathrm h}\frac{1000\,\mathrm m}{1\,\mathrm{km}}\frac{1\,\mathrm h}{3600\,\mathrm s}=27.8\,\mathrm{m/s}.$$

**Reaction phase:** Let $d_r$ be the distance before braking. Acceleration is zero during $t_r=1.0\,\mathrm s$:

$$d_r=v_0t_r=(27.8\,\mathrm{m/s})(1.0\,\mathrm s)=27.8\,\mathrm m.$$

**Braking phase:** Let $d_b$ be braking distance. Here $a=-8\,\mathrm{m/s^2}$ and final $v=0$:

$$\begin{aligned}
0&=v_0^2+2ad_b,\\
0&=772\,\mathrm{m^2/s^2}-(16\,\mathrm{m/s^2})d_b,\\
(16\,\mathrm{m/s^2})d_b&=772\,\mathrm{m^2/s^2},\\
d_b&=\frac{772\,\mathrm{m^2/s^2}}{16\,\mathrm{m/s^2}}=48.2\,\mathrm m.
\end{aligned}$$

Add the braking term to both sides, then divide by its positive coefficient. Finally add the two distances:

$$\begin{aligned}
d_{\rm stop}&=d_r+d_b=27.8+48.2=76.0\,\mathrm m,\\
\text{remaining distance}&=80-76.0=4.00\,\mathrm m.
\end{aligned}$$

**Result:** The car stops about $4.00\,\mathrm m$ before the child within the stated model. Keep unrounded values in the calculation.

**Check:** Count reaction distance as well as braking distance. The small positive margin depends on the supplied assumptions.

**Türkçe:** Tepki sırasında araç ilerlemeye devam eder. Önce iki yol parçasını ayrı hesapla, sonra topla. Negatif ivmeyi denkleme işaretiyle yaz; pozitif durma yolunu bulmak için terimleri iki tarafa tutarlı taşı.


```python
#@title Optional numerical check — the algebra is explained above
# Worked Example 3
speed_kmh = 100
v0 = speed_kmh / 3.6  # m/s
t_react = 1.0         # s
a_brake = 8.0         # m/s^2 (magnitude)
d_child = 80.0        # m

print(f"Initial speed: {speed_kmh} km/h = {v0:.2f} m/s")

# Reaction distance
d_react = v0 * t_react
print(f"\nReaction distance: d = {v0:.2f} * {t_react} = {d_react:.2f} m")

# Braking distance
d_brake = v0**2 / (2 * a_brake)
print(f"Braking distance: d = {v0:.2f}^2 / (2*{a_brake}) = {d_brake:.2f} m")

# Total
d_total = d_react + d_brake
print(f"\nTotal stopping distance: {d_react:.2f} + {d_brake:.2f} = {d_total:.2f} m")
print(f"Distance to child: {d_child} m")

if d_total < d_child:
    print(f"\n>>> The car stops {d_child - d_total:.1f} m BEFORE the child. Safe!")
else:
    print(f"\n>>> The car needs {d_total - d_child:.1f} m MORE than available. NOT safe!")
    # Speed at impact
    d_brake_available = d_child - d_react
    if d_brake_available > 0:
        v_impact_sq = v0**2 - 2 * a_brake * d_brake_available
        if v_impact_sq > 0:
            v_impact = np.sqrt(v_impact_sq)
            print(f"    Speed at impact: {v_impact:.1f} m/s ({v_impact*3.6:.0f} km/h)")
```

#### Worked example: Set an obstacle-detection distance

**Problem.** A warehouse vehicle travels at $1.60\,\mathrm{m/s}$. Its controller waits $0.180\,\mathrm s$ after detection before braking at $1.20\,\mathrm{m/s^2}$. It must stop $0.250\,\mathrm m$ before an obstacle. Is a detection distance of $1.50\,\mathrm m$ enough?

**Picture and principle.** Draw three lengths: latency travel, braking travel, and clearance. Constant speed during latency gives $v\tau$; braking gives $v^2/(2b)$, where $b>0$ is the deceleration magnitude. Required detection distance is $D=v\tau+v^2/(2b)+c$.

**Limiting check.** With zero latency the first term vanishes, but braking distance remains. Doubling speed quadruples the braking term.

**Numbers and units.** $$D=(1.60)(0.180)\,\mathrm m+\frac{(1.60\,\mathrm{m/s})^2}{2(1.20\,\mathrm{m/s^2})}+0.250\,\mathrm m=1.60\,\mathrm m.$$

**Interpret.** The setting is insufficient: the shortfall is $0.105\,\mathrm m$. Braking performance alone does not set the stopping envelope.

**Türkçe destek.** Algılama gecikmesinde araç yol almaya devam eder. Güvenli mesafe, gecikme yolu ile frenleme yolu ve bırakılacak boşluğun toplamıdır.


#### Worked example: Synchronise two inspection carts

**Problem.** Cart A passes a launch mark at constant $0.600\,\mathrm{m/s}$. Cart B starts there from rest $0.400\,\mathrm s$ later, with constant acceleration. They must meet at a station $1.80\,\mathrm m$ from the mark. Find B’s acceleration and decide whether meeting also means matching speeds.

**Picture and principle.** Draw two position–time curves crossing at the station. A arrives at $T=L/u$; B has only $T-\tau$. Therefore $a_B=2L/(L/u-\tau)^2$ and $v_B=a_B(T-\tau)$.

**Limiting check.** As the delay approaches A’s arrival time, the required acceleration diverges; a positive travel time is essential.

**Numbers and units.** $$T=3.00\,\mathrm s,\quad T-\tau=2.60\,\mathrm s,\quad a_B=\frac{2(1.80\,\mathrm m)}{(2.60\,\mathrm s)^2}=0.533\,\mathrm{m/s^2}.$$ B arrives at $1.38\,\mathrm{m/s}$, faster than A.

**Interpret.** The positions match but the velocities do not. A transfer operation would need an additional speed-matching phase.

**Türkçe destek.** İki aracın aynı konuma aynı anda gelmesi, hızlarının eşit olduğu anlamına gelmez. Gecikme, ikinci aracın hareket süresinden çıkarılır.


#### Worked example: Decode two beam crossings

**Problem.** A small marker is thrown vertically from a fixture. It crosses a light beam $0.350\,\mathrm s$ and $0.950\,\mathrm s$ after release. Neglect drag. Infer both its launch speed and the beam height above the release point without measuring that height directly.

**Picture and principle.** Sketch a parabolic height–time curve intersecting one horizontal line twice. The roots of $\tfrac12gt^2-v_0t+h=0$ obey $t_1+t_2=2v_0/g$ and $t_1t_2=2h/g$. Hence $v_0=g(t_1+t_2)/2$ and $h=gt_1t_2/2$.

**Limiting check.** If the crossings merge, the beam is at the highest point. The apex time is always their midpoint.

**Numbers and units.** $$v_0=\frac{(9.81)(0.350+0.950)}{2}\,\mathrm{m/s}=6.38\,\mathrm{m/s},\qquad h=\frac{(9.81)(0.350)(0.950)}2\,\mathrm m=1.63\,\mathrm m.$$

**Interpret.** Two timestamps determine two unknowns. The earlier crossing is upward and the later one downward.

**Türkçe destek.** Aynı yüksekliğe çıkışta ve inişte uğranır. İki zamanın toplamı ilk hızı, çarpımı ışık perdesinin yüksekliğini verir.

### Looking ahead: Week 03 is the heaviest week / Önümüzdeki hafta en yoğun hafta

Week 03 carries **two** topics in one three-hour session: motion in two dimensions **and** the
introduction to Newton's laws. That is set by the departmental schedule and it cannot be spread out,
because the common midterm assumes both have been covered. So the honest thing to do is warn you and
give you a way to arrive ready.

**The good news: two-dimensional motion is not a new subject.** A projectile is just *this* week's
one-dimensional motion, running twice, on one shared clock:

| | Horizontal | Vertical |
| --- | --- | --- |
| Acceleration | $a_x = 0$ | $a_y = -g$ |
| Velocity | constant: $v_x = v_0\cos\theta$ | changes: $v_y = v_0\sin\theta - gt$ |
| Position | $x = v_x t$ | $y = y_0 + v_0\sin\theta\,t - \tfrac12 g t^2$ |

Look at the vertical column: it is exactly the free-fall equations you used today. The horizontal column
is the constant-velocity case. **The only new idea is that the two columns share the same $t$** — and $t$
is what connects them, which is why almost every projectile problem is solved by finding the time first.

**A ten-minute investment that pays for itself.** Before the next session:

1. Take any free-fall problem from today and re-solve it, writing $a_y = -g$ and $v_{0y} = v_0\sin\theta$
   with $\theta = 90^\circ$. Confirm you get the same answer. That is the 2D machinery, used on a 1D problem.
2. Read the first table in Week 03's "Theory: Projectile Motion" — just the table, not the whole section.
3. Write down, in one sentence, what you think a force *is*. You will compare it with Newton's answer.

If you do those three things, Week 03 has one new topic in it instead of two.

## 5. Exit check / Çıkış kontrolü

Explain why a thrown ball has zero velocity but nonzero acceleration at its highest point.

Write one sentence naming the physical principle and one line of algebra you can now explain. **TR:** Sonuca nasıl ulaştığını bir cümleyle anlat; emin olmadığın ilk adımı işaretle.

<details>
<summary>Model responses / Örnek yanıtlar — open after trying</summary>

**Entry:** $36\,\mathrm{km/h}\times1000/3600=10\,\mathrm{m/s}$. A leftward velocity is negative if right is positive; a sign needs a stated axis.

**Exit:** At the highest point of a vertical throw, the instantaneous velocity is zero while gravity still gives $a_y=-9.81$ m/s². The velocity changes through zero; the ball does not stay there.

**Visual prediction:** With $v_0=5$ m/s and $a=-1$ m/s², set $v=5-t=0$, so the turning time is 5 s. Before that turn, $\Delta x=5(5)-5^2/2=12.5$ m and distance is also 12.5 m. If motion continues to 10 s, displacement is $5(10)-10^2/2=0$ but distance is $12.5+12.5=25$ m.

</details>

## 1. Prediction, Part A — moment of inertia / Tahmin

### Symbolically first

A mass $m$ hangs from a string wound round a spindle of radius $R$ on a disc of moment of inertia $I$.
Release from rest. Write Newton's second law for the mass and the rotational equation for the disc, and
use the no-slip condition $a = R\alpha$:

$$mg - T = ma, \qquad TR = I\alpha = I\frac{a}{R}$$

Eliminate the tension $T$:

$$\boxed{\,a = \frac{mg}{m + I/R^2} \quad \Longrightarrow \quad I = mR^2\left(\frac{g}{a} - 1\right)\,}$$

### Limiting checks — all three are marks

1. $I \to 0$ (a massless disc) gives $a \to g$: the mass falls freely. ✅ Sensible.
2. $I \to \infty$ gives $a \to 0$: an immovable disc holds the mass up. ✅ Sensible.
3. In $I = mR^2(g/a - 1)$, what happens if you measure $a$ slightly **too large**? Show that $I$ comes out
   too **small**, and note that as $a \to g$ the bracket $\to 0$ — so for a light disc this method becomes
   very sensitive to small errors in $a$. **Choose a hanging mass small enough that $a$ is comfortably less
   than $g$.** That is experimental design following from a limiting case.

### Compare with the geometric prediction

For a uniform solid disc of mass $M$ and radius $r$ about its centre, $I_{\text{disc}} = \tfrac12 M r^2$.
Weigh and measure the disc, compute this, and compare with your dynamical $I$. **Two independent routes to
the same quantity** — that is what makes this a real measurement rather than an exercise.

**TR:** Önce sembolik: $I = mR^2(g/a - 1)$. Üç sınır kontrolünü de yap. Üçüncüsü deney tasarımını belirler:
$a$, $g$'ye yaklaştıkça parantez sıfıra gider ve yöntem küçük hatalara aşırı duyarlı olur — bu yüzden
asılı kütleyi, $a$ rahatça $g$'den küçük kalacak kadar küçük seç. Sonucu ayrıca $I = \tfrac12 M r^2$ ile
karşılaştır: aynı büyüklüğe **iki bağımsız yol.**

## 3. Primer, Part B — everything you need about the pendulum / Sarkaç için gereken her şey

**This section is self-contained.** Week 13 will do it properly; here is the working version.

### Where the period comes from

A mass $m$ on a light string of length $L$, displaced by a small angle $\theta$ from vertical. The
restoring force along the arc is the tangential component of gravity:

$$F = -mg\sin\theta$$

The minus sign says it always pushes back towards the bottom. **This is the whole idea of oscillation: a
force that always points back to equilibrium.**

For small angles, with $\theta$ **in radians**, $\sin\theta \approx \theta$. Since the arc displacement is
$s = L\theta$, we get $\theta = s/L$ and

$$F \approx -mg\frac{s}{L} = -\left(\frac{mg}{L}\right)s.$$

That is a restoring force **proportional to displacement** — exactly the form $F = -ks$ with
$k = mg/L$. Newton's second law $F = ma$ then gives a motion whose period is
$T = 2\pi\sqrt{m/k}$, and substituting $k = mg/L$:

$$\boxed{\,T = 2\pi\sqrt{\frac{L}{g}}\,}$$

**The mass cancelled.** $m$ appears in the restoring force and again in the inertia, and the two uses
divide out. A heavy pendulum and a light one of the same length keep the same time.

### How to test it

Square both sides and rearrange into a straight line:

$$T^2 = \frac{4\pi^2}{g}L \quad \Longrightarrow \quad \text{plot } T^2 \text{ against } L,
\quad \text{slope } m_s = \frac{4\pi^2}{g}, \quad \boxed{\,g = \frac{4\pi^2}{m_s}\,}$$

Intercept should be zero.

### Limiting checks

1. Longer string → longer period. Check the formula agrees with a playground swing.
2. Larger $g$ → shorter period. A pendulum clock would run **fast** on Jupiter.
3. $T$ does not depend on $m$ or on the amplitude (for small angles). **Both are testable today**, and you
   should test them.

### The small-angle limit, honestly

$\sin\theta \approx \theta$ is good to 0.5% at $10°$ and about 2% at $20°$. The next correction makes the
true period slightly **longer** than the formula:

$$T \approx 2\pi\sqrt{\frac{L}{g}}\left(1 + \frac{\theta_0^2}{16}\right)$$

So **keep amplitudes under about $10°$** and your $g$ will not be biased low by the approximation. If you
deliberately use a large amplitude in one run, you should see $T$ come out slightly long — a prediction
worth testing if you have time.

**TR:** Bu bölüm kendi kendine yeterlidir. Küçük açıda geri çağırıcı kuvvet yer değiştirmeyle doğru
orantılıdır ve periyot $T = 2\pi\sqrt{L/g}$ olur; kütle sadeleşir. Test için $T^2$'yi $L$'ye karşı çiz,
$g = 4\pi^2/\text{eğim}$. Genlikleri $10°$ altında tut: yaklaşıklık, gerçek periyodu biraz **uzun** yapar.


## 4. Procedure, Part B / Yöntem — B bölümü

1. Set up a pendulum whose length $L$ you can change. **$L$ is measured from the pivot to the centre of
   the bob**, not to its top — with a 3 cm bob on a 40 cm string that distinction is a 4% error in $L$ and
   hence 4% in $g$.
2. Choose **six lengths** spread as widely as the room allows, e.g. 0.20 to 1.20 m. Wide spread pins the
   slope down.
3. At each length, displace by **less than 10°** and release cleanly — no push, no sideways component, so
   it swings in a plane rather than circling.
4. **Time 20 complete oscillations**, not one. Then $T = t_{20}/20$ and your timing uncertainty is
   divided by 20 as well: a 0.2 s reaction time becomes 0.01 s on $T$. Repeat three times per length.
5. Also do two checks at **one fixed length**:
   - swap the bob for a clearly heavier one — does $T$ change?
   - use a deliberately large amplitude ($\sim 30°$) — does $T$ change, and in which direction?

**Why timing 20 swings is the single most important instruction here.** It is the cheapest uncertainty
reduction in the entire lab programme: the same stopwatch and the same reaction time, divided by 20, for
no extra equipment. Compare with the free-fall experiment, where you could not do this because the fall
happens once. **When a process repeats, time many cycles and divide.**

**TR:** $L$, mesnetten topuzun **merkezine** kadar ölçülür. Altı farklı uzunluk, genlik $10°$ altında ve
her uzunlukta **20 tam salınım** ölç, sonra 20'ye böl: aynı kronometre ve aynı tepki süresiyle belirsizlik
20 kat küçülür. Bu, bütün laboratuvar programının en ucuz belirsizlik azaltma yöntemidir. Bir olay
tekrarlanıyorsa çok çevrim ölç ve böl.

## 6. Discussion / Tartışma

<details><summary>Your two hanging masses give different I. What is the model missing? / İki kütle farklı I veriyorsa model neyi atlıyor?</summary>

Bearing friction. The derivation assumed the only torque on the disc is $TR$, but there is also a
friction torque $\tau_f$ opposing rotation:

$$TR - \tau_f = I\alpha$$

Ignore $\tau_f$ and you attribute its effect to $I$, so the $I$ you extract is too **large** — and the
error is worse for the smaller hanging mass, because $\tau_f$ is then a bigger fraction of the driving
torque. That is exactly the pattern to look for: **the lighter mass should give the larger apparent $I$.**

Check whether your two values follow that pattern. If they do, you have not merely found a discrepancy —
you have identified its mechanism and predicted its direction, which is a complete piece of experimental
reasoning. (Extracting $\tau_f$ properly needs three or more masses and a graph; mention it as the next
step.)

**TR:** Yatak sürtünmesi. Gerçek denklem $TR - \tau_f = I\alpha$; $\tau_f$'yi yok sayarsan etkisini $I$'ye
yüklersin ve $I$ **büyük** çıkar. Hata, küçük kütlede daha belirgindir: **hafif kütle daha büyük görünür
$I$ vermelidir.** İki değerin bu deseni izliyorsa yalnızca bir uyuşmazlık bulmadın, mekanizmasını
belirleyip yönünü öngördün.

</details>

<details><summary>Dynamical I is bigger than ½Mr². Is the disc not uniform? / Dinamik I, ½Mr²'den büyükse disk düzgün değil mi?</summary>

Possibly, but check the mundane causes first, in this order:

1. **The spindle, hub, axle and any clamps also rotate** and carry their own inertia. $\tfrac12 Mr^2$ is
   the bare disc only, so the measured $I$ *should* exceed it. This alone often explains the gap.
2. Friction, as above, inflates the measured $I$.
3. Then, and only then, consider a non-uniform disc — e.g. a thicker rim, which raises $I$ above
   $\tfrac12 Mr^2$ because mass further out counts quadratically.

**The measured value exceeding the geometric one is the expected outcome**, not an error. What would be
alarming is measuring *less* than $\tfrac12 Mr^2$, which would point to a mistake in $R_{\text{eff}}$ or a
slipping string.

**TR:** Önce sıradan sebepler: mil, göbek, eksen ve kelepçeler de döner ve kendi eylemsizliklerini taşır;
$\tfrac12 Mr^2$ yalnızca çıplak disktir, yani ölçülen $I$ bundan **büyük olmalıdır**. Endişe verici olan,
$\tfrac12 Mr^2$'den **küçük** ölçmektir; bu $R_{\text{eff}}$ hatasına veya kayan ipe işaret eder.

</details>

<details><summary>Did the heavier bob change the period? Should it have? / Ağır topuz periyodu değiştirdi mi?</summary>

No, and no. In $T = 2\pi\sqrt{L/g}$ the mass has cancelled: it sets both the restoring force ($mg\sin\theta$)
and the inertia ($m$), and the ratio is what determines the period.

If your heavier bob *did* change $T$ measurably, look for a length change — a heavier bob may stretch the
string, or its centre may sit at a different distance from the pivot. That is an $L$ effect masquerading
as an $m$ effect, and catching it is the point of the check.

**TR:** Hayır ve hayır; kütle sadeleşir. Ölçülebilir bir değişiklik olduysa uzunluk değişimini ara: ağır
topuz ipi gerebilir veya merkezi mesnetten farklı uzaklıkta olabilir. Bu, $m$ etkisi gibi görünen bir $L$
etkisidir ve kontrolün amacı tam olarak onu yakalamaktır.

</details>

<details><summary>At 30° amplitude the period came out longer. Is that an error? / 30°'de periyot uzun çıktıysa hata mı?</summary>

No — that is the small-angle approximation failing, exactly as §3 predicted, and you have just measured
the failure of a model rather than the failure of your technique. The correction
$T \approx T_0(1 + \theta_0^2/16)$ predicts about $+1\%$ at $30°$ ($\theta_0 = 0.52$ rad).

Compare your measured increase with that number. Agreement means you have detected a second-order
physical effect with a stopwatch and a piece of string — a genuinely good result, and worth writing up as
the highlight of your report.

**TR:** Hayır; §3'ün öngördüğü gibi küçük açı yaklaşımının bozulmasıdır ve tekniğinin değil, bir
**modelin** sınırını ölçtün. Düzeltme $30°$ için yaklaşık $+\%1$ öngörür. Ölçtüğün artışı bu sayıyla
karşılaştır: uyuşuyorsa kronometre ve iple ikinci mertebeden bir fiziksel etkiyi saptamışsın.

</details>

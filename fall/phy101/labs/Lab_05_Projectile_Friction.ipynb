# Lab 05 — Projectile motion and friction

## Eğik atış ve sürtünme

**Week 05 laboratory session · 19 – 23 October 2026**
**Official schedule entry / İzlencedeki adı:** *Eğik Atış - Sürtünme*

It measures the physics taught in **Week 03**, so the lecture comes first and the experiment confirms it.

**Aim.** Predict where a projectile will land before firing it, then measure the coefficient of static friction from the angle at which a block starts to slide.

**TR.** Bir cismin nereye düşeceğini atmadan önce tahmin et; sonra bir bloğun kaymaya başladığı açıdan statik sürtünme katsayısını ölç.

**Scope note.** The experiment is scheduled by the department and is the same for every section. The
*analysis technique* in this brief (uncertainty, linearising, counting sigmas) is a laboratory
requirement and is **not** part of the common midterm or final; the *physics* being measured is.

**Before you arrive:** read [Lab 00 — measurement and uncertainty toolkit](Lab_00_Uncertainty_Toolkit.ipynb) and complete the prediction below. You will be asked for your
predicted number at the bench, before any apparatus is switched on.

**TR:** Gelmeden önce el kitabını oku ve aşağıdaki tahmini yap. Tezgâhta, düzenek çalışmadan önce
tahmin ettiğin sayı sorulacak.


## 1. Prediction — before the bench / Tahmin

This experiment has two parts. Do both predictions in your logbook first.

### Part A — where will it land?

A projectile leaves a launcher at speed $v_0$, at angle $\theta$ above the horizontal, from height $y_0$
above the floor. **Symbolically first**, with the no-drag, constant-$g$ model:

$$x(t) = v_0\cos\theta\,t, \qquad y(t) = y_0 + v_0\sin\theta\,t - \tfrac12 g t^2.$$

Set $y = 0$ and solve the quadratic for the flight time, then substitute into $x$:

$$t_f = \frac{v_0\sin\theta + \sqrt{v_0^2\sin^2\theta + 2gy_0}}{g}, \qquad
\boxed{\,R = v_0\cos\theta \cdot t_f\,}$$

**Limiting checks — do all three, they are the marks:**

1. $y_0 = 0$ should give $R = v_0^2\sin(2\theta)/g$. Verify it, and confirm the maximum is at $45°$.
2. $\theta = 0$ should give $R = v_0\sqrt{2y_0/g}$ — the horizontal-launch result. Check it.
3. As $g$ increases, $R$ must decrease. Confirm from the formula.

**Then propagate:** with $y_0 = 0$ for simplicity, $R \propto v_0^2$, so $\delta R/R = 2\,\delta v_0/v_0$.
The launch speed enters **squared**. Measuring $v_0$ well is the entire experiment; the angle is easy and
the height is easy.

### Part B — when will it slide?

A block rests on a plane tilted at angle $\alpha$. At the angle $\alpha_c$ where it *just* begins to
slide, static friction is at its maximum. Resolve along and normal to the surface:

$$mg\sin\alpha_c = \mu_s N, \qquad N = mg\cos\alpha_c \quad \Longrightarrow \quad \boxed{\,\mu_s = \tan\alpha_c\,}$$

**Notice what cancelled.** The mass dropped out entirely. **Predict, before measuring:** does adding a
second block on top change $\alpha_c$? Write down your answer and your reason — this is the single most
instructive prediction in the whole lab programme, and most students get it wrong the first time.

**TR:** İki bölüm var, ikisinin de tahminini önce deftere yaz. A'da üç sınır kontrolünü de yap; $v_0$
kareli girdiği için bütün deney onu iyi ölçmekten ibaret. B'de kütlenin sadeleştiğine dikkat et: üste
ikinci bir blok koymak $\alpha_c$'yi değiştirir mi? Cevabını ve sebebini **ölçmeden** yaz.


## 2. Apparatus / Düzenek

**Part A:** spring or pneumatic launcher with an adjustable angle, protractor, metre rule or tape, carbon
paper or a sand tray to mark the landing point, small dense ball.

**Part B:** adjustable inclined plane (or a plank and a stand), protractor or an angle app, blocks of
different mass with the **same** contact material, and at least two different surface pairs
(e.g. wood-on-wood and rubber-on-wood).

**Measuring $v_0$.** If the launcher is not calibrated, get $v_0$ from a **horizontal** launch first:
fire at $\theta = 0$ from a measured height $y_0$, measure the range $R_0$, then

$$v_0 = R_0\sqrt{\frac{g}{2y_0}}.$$

Do this five times and average. Note the circularity you are avoiding: calibrate at one angle, then
**predict** at a different angle. Predicting the same shot you calibrated with tests nothing.

**TR:** Fırlatıcı kalibre değilse $v_0$'ı önce yatay atıştan bul, beş kez ölç ve ortala. Bir açıda kalibre
edip **başka** bir açıda tahmin yapmaya dikkat et: kalibre ettiğin atışı tahmin etmek hiçbir şeyi test
etmez.


## 3. Procedure / Yöntem

### Part A

1. Calibrate $v_0$ from five horizontal launches as above. Compute $\bar{v}_0 \pm \delta v_0$.
2. Set the launcher to $\theta = 30°$. **Before firing, compute the predicted $R$ with its uncertainty and
   write it in your logbook.** Mark the predicted landing point on the floor.
3. Fire five times and record each landing distance.
4. Repeat for $\theta = 45°$ and $\theta = 60°$.
5. For each angle compare predicted and measured range by counting sigmas.

**Safety:** never sight along the barrel, and announce each shot. Keep the landing zone clear.

### Part B

1. Place the block on the plane at a low angle. Raise the angle **slowly** and continuously until the
   block just begins to slide. Record $\alpha_c$.
2. Repeat **eight times** for the same block and surface. This measurement is much noisier than it looks,
   which is why the repeat count is high.
3. Repeat with a second block of clearly different mass, same materials. Compare $\alpha_c$ with your §1
   prediction.
4. Repeat with a different surface pair.

**Why $\alpha_c$ scatters.** Static friction has a *maximum*, not a fixed value, and the exact angle at
which slipping starts depends on surface dust, the precise patch of contact, and how smoothly you tilted.
Tilt too fast and you overshoot; tilt in jerks and vibration triggers an early slip. Aim for slow and
steady, and accept a spread of a degree or two as honest.

**TR:** Açıyı **yavaş** ve sürekli artır. Bu ölçüm göründüğünden çok gürültülüdür — statik sürtünmenin
sabit değeri değil, bir *üst sınırı* vardır — bu yüzden sekiz tekrar. Hızlı eğersen aşarsın, sarsarsan
titreşim erken kaymayı tetikler. Bir-iki derecelik saçılma dürüst bir sonuçtur.

$n_\sigma = |R_{\text{pred}} - \bar{R}| / \sqrt{(\delta R)^2 + (\delta\bar{R})^2}$.

### Part B — friction

| Surface pair | Block mass (g) | $\alpha_c$ trials 1–8 (°) | $\bar{\alpha}_c \pm \delta$ (°) | $\mu_s = \tan\bar{\alpha}_c$ | $\delta\mu_s$ |

**Propagating through the tangent.** Since $\mu_s = \tan\alpha_c$,

$$\delta\mu_s = \frac{\delta\alpha_c}{\cos^2\alpha_c} \quad \text{with } \delta\alpha_c \textbf{ in radians.}$$

Forgetting to convert degrees to radians here is the most common arithmetic slip in this experiment, and
it inflates $\delta\mu_s$ by a factor of 57.

**TR:** Tanjantta yayılma yukarıdaki gibidir ve $\delta\alpha_c$ **radyan** olmalıdır. Dereceyi radyana
çevirmeyi unutmak bu deneydeki en sık hatadır ve belirsizliği 57 kat şişirir.


## 5. Analysis / Analiz

### Part A

1. For each angle, state predicted vs measured with $n_\sigma$ and a one-word verdict.
2. Plot measured $R$ against $\theta$ and overlay your predicted curve. Is the measured maximum at 45°,
   above it, or below it?
3. If your measured ranges are systematically **shorter** than predicted, say so and name the cause.

### Part B

1. Do the two blocks of different mass give the same $\alpha_c$ within uncertainty? Count sigmas.
2. Report $\mu_s$ for each surface pair with its uncertainty.
3. **Test the model, not just the number:** the mass-independence of $\alpha_c$ *is* the prediction
   $\mu_s = \tan\alpha_c$ being tested. If the two masses agree, you have confirmed that friction scales
   with the normal force. If they disagree by more than $3\sigma$, something is wrong with the surfaces
   (different materials? one block rocking on a rounded edge?) rather than with the physics.

---

### The mass question, resolved

Doubling the mass doubles the driving component $mg\sin\alpha$ **and** doubles the normal force
$mg\cos\alpha$, hence doubles the maximum friction $\mu_s N$. Both sides of the comparison grow by the
same factor, so the tipping angle is unchanged. $\alpha_c$ depends only on the **materials**, not on how
much you put on top.

Most people predict that a heavier block slides at a smaller angle, because heavy things feel harder to
hold up. The algebra says otherwise, and your data will back the algebra. **Getting this prediction
wrong and then seeing it corrected by your own measurement is worth more than getting it right by
luck** — say so in your report if that is what happened.

**TR:** Kütleyi iki katına çıkarmak hem $mg\sin\alpha$'yı hem $mg\cos\alpha$'yı iki katına çıkarır, yani
en büyük sürtünme de iki katına çıkar; karşılaştırmanın iki tarafı aynı oranda büyüdüğü için devrilme
açısı değişmez. $\alpha_c$ yalnızca **malzemelere** bağlıdır. Çoğu kişi tersini tahmin eder; bu tahmini
yanlış yapıp kendi ölçümünle düzeltilmesi, şansa doğru yapmaktan değerlidir.


## 6. Discussion / Tartışma

<details><summary>Your measured ranges are all ~8% short. What is the most likely cause? / Menziller hep %8 kısa çıkıyorsa?</summary>

Almost certainly **air resistance**, which removes horizontal momentum throughout the flight, plus a
slightly overestimated $v_0$ if the launcher weakens with use. The signature that distinguishes them: air
drag makes the shortfall **grow** with range (worst at 45°, where the ball flies longest), while a bad
$v_0$ calibration produces a roughly constant *percentage* shortfall at every angle.

Check which pattern your data shows and say which explanation it supports. That is a genuine piece of
experimental reasoning, and it is available to you from three angles' worth of data.

**TR:** Büyük olasılıkla hava direnci, artı yorulan bir fırlatıcıdan gelen fazla tahmin edilmiş $v_0$.
Ayırt edici iz: hava direncinde eksiklik menzille **büyür** (en çok 45°'de), kötü kalibrasyonda ise her
açıda yaklaşık sabit bir **yüzde** eksiklik olur. Verinin hangi deseni gösterdiğini söyle.

</details>

<details><summary>Why measure v₀ from a horizontal launch instead of using the manufacturer's value? / Neden v₀'ı yatay atıştan ölçüyoruz?</summary>

Because the manufacturer's value is for a new, clean launcher, and yours has been compressed a few
thousand times by previous classes. Using a nominal value imports someone else's systematic error into
your result and you would have no way to detect it.

The deeper habit: **measure the input to your prediction with the same apparatus you will test the
prediction on.** Otherwise a disagreement tells you nothing, because you cannot say whether the model or
the calibration failed.

**TR:** Çünkü katalog değeri yeni ve temiz bir fırlatıcı içindir; senin fırlatıcın binlerce kez sıkışmış.
Genel alışkanlık: tahminin girdisini, tahmini test edeceğin **aynı** düzenekle ölç. Aksi hâlde uyuşmazlık
hiçbir şey söylemez; model mi kalibrasyon mu bozuk, ayırt edemezsin.

</details>

<details><summary>Is 45° really the best angle here? / Burada gerçekten en iyi açı 45° mi?</summary>

Only if the launch and landing heights are equal. Your launcher sits **above** the floor, $y_0 > 0$, and
the optimum angle then drops below 45° — the projectile gets extra flight time for free from the initial
height, so it pays to convert more of $v_0$ into horizontal speed.

You can see this in your own formula: put $y_0 > 0$ into $R(\theta)$ and evaluate at 40° and 45°. With a
launcher a metre up and modest $v_0$, the optimum is typically around 40–43°. Your three-angle data set is
too coarse to resolve that, and saying so is the correct conclusion — **knowing what your data cannot
settle is part of the answer.**

**TR:** Yalnızca atış ve iniş yükseklikleri eşitse. Fırlatıcı yerden yüksekte olduğu için en iyi açı
45°'nin altına iner. Üç açılık verin bunu ayırt edemeyecek kadar kabadır; bunu söylemek doğru sonuçtur —
verinin neyi karara bağlayamadığını bilmek de cevabın parçasıdır.

</details>

<details><summary>Does μₛ depend on the contact area? / μₛ temas alanına bağlı mı?</summary>

In this model, no — and you can test it if a block can be laid on a large face and then on a small face
at the same mass. The tipping angle should be the same within uncertainty.

The reason: a larger area spreads the same total normal force over more contact, so the force per unit
area falls in proportion and the total friction is unchanged. The model $f_s^{\max} = \mu_s N$ has no area
in it at all. (Tyres and polished metals deviate from this, which is why it is a model and not a law.)

**TR:** Bu modelde hayır. Aynı kütleyle geniş ve dar yüzeyi dene; devrilme açısı belirsizlik içinde aynı
çıkmalı. Daha büyük alan aynı normal kuvveti daha geniş temasa yayar, birim alandaki kuvvet oranlı azalır.
$f_s^{\max} = \mu_s N$ modelinde alan hiç yoktur. (Lastikler ve parlatılmış metaller sapar; bu yüzden
yasa değil, modeldir.)

</details>

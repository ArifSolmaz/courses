## 1. What a measurement is / Ölçüm nedir

Every measurement is reported as

$$x = x_{\text{best}} \pm \delta x \quad [\text{unit}]$$

- $x_{\text{best}}$ — your best estimate, usually the mean of your readings.
- $\delta x$ — the **uncertainty**: the size of the range in which the true value plausibly sits.
- The unit is part of the answer. A number without a unit is not physics.

Two kinds of error behave differently, and confusing them is the most common lab mistake:

| | What it is | What it does to your result | How to reduce it |
| --- | --- | --- | --- |
| **Random** | Scatter from reaction time, judging a scale, small draughts | Makes repeats disagree; averages out | Measure repeatedly and average |
| **Systematic** | Zero offset, a ruler that is short, a stopwatch started early every time | Shifts *every* repeat the same way; does **not** average out | Find and correct the cause |

**A systematic error cannot be fixed by taking more data.** If your stopwatch always starts late, a
thousand trials give a very precise wrong answer. This is why you always check the zero of an
instrument before using it.

**TR:** Rastgele hata tekrarlarla küçülür; sistematik hata küçülmez, sadece daha kesin bir *yanlış*
sonuç verir. Bu yüzden ölçüme başlamadan önce cihazın sıfırını kontrol et.

---

### The word you must not use loosely

"Error" in a lab does not mean "mistake". Dropping the mass on your foot is a mistake; the ±0.5 mm you
cannot avoid when reading a ruler is an uncertainty. Mistakes are repeated, not reported.

**TR:** Laboratuvarda "hata" kelimesi "yanlış yapmak" demek değildir; ölçmenin sınırı demektir.

## 2. Reading uncertainty from an instrument / Cihazdan okuma belirsizliği

For a single reading on an analogue scale, take **half the smallest division**; for a digital display,
take **one unit of the last digit**.

| Instrument | Smallest division | Reading uncertainty $\delta x$ |
| --- | --- | --- |
| Metre rule | 1 mm | $\pm 0.5\,\mathrm{mm}$ |
| Vernier caliper | 0.05 mm (or 0.02 mm) | $\pm 0.025\,\mathrm{mm}$ (or $\pm 0.01\,\mathrm{mm}$) |
| Micrometer | 0.01 mm | $\pm 0.005\,\mathrm{mm}$ |
| Digital balance | 0.1 g | $\pm 0.1\,\mathrm{g}$ |
| Hand stopwatch | 0.01 s | $\pm 0.2\,\mathrm{s}$ — **reaction time dominates**, not the display |

That last row matters more than all the others. A stopwatch shows hundredths of a second, but a human
starting and stopping it is good to roughly two tenths. **Quote the uncertainty you actually have, not
the one the display advertises.** This single habit separates a physics report from a list of digits.

**TR:** Analog ölçekte en küçük bölmenin yarısı, dijital göstergede son hanenin bir birimi. Elle tutulan
kronometrede belirleyici olan gösterge değil, senin tepki sürendir: yaklaşık ±0.2 s. Göstergenin
gösterdiği hassasiyeti değil, gerçekten sahip olduğun hassasiyeti yaz.

---

### A worked reading

A vernier caliper with 0.05 mm divisions reads a rod diameter of 12.40 mm.

$$d = 12.40 \pm 0.03\ \mathrm{mm} \qquad \frac{\delta d}{d} = \frac{0.03}{12.40} = 0.24\%$$

Report the absolute uncertainty to **one significant figure** (0.03, not 0.025), and round the best
estimate to the same decimal place. `12.4025 ± 0.025` is wrong twice over.

## 3. Repeated measurements / Tekrarlı ölçümler

Take $N$ readings $x_1 \dots x_N$ of the same quantity.

**Best estimate — the mean:**
$$\bar{x} = \frac{1}{N}\sum_{i=1}^{N} x_i$$

**Spread of a single reading — the standard deviation:**
$$s = \sqrt{\frac{1}{N-1}\sum_{i=1}^{N}\left(x_i-\bar{x}\right)^2}$$

**Uncertainty in the mean — the standard error:**
$$\delta \bar{x} = \frac{s}{\sqrt{N}}$$

Report $\bar{x} \pm s/\sqrt{N}$. The $\sqrt{N}$ is the whole reason we repeat a measurement: four trials
halve the uncertainty, nine trials cut it to a third.

**The catch:** $\sqrt{N}$ only helps against *random* error, and its help shrinks fast. Going from
$N=5$ to $N=10$ buys you a factor $1.4$; going from 10 to 100 buys another factor 3 and an hour of your
life. Five to ten good trials with the systematic errors hunted down beats fifty careless ones.

**TR:** Ortalamayı ve standart hatayı yaz. $\sqrt{N}$ yalnızca rastgele hataya karşı işe yarar ve
kazancı hızla azalır: sistematik hatayı avlanmış on iyi ölçüm, dikkatsiz elli ölçümden iyidir.

---

### Worked example — five fall times

Times for a ball dropped from a fixed height (s): 0.62, 0.58, 0.65, 0.60, 0.61.

$$\bar{t} = \frac{0.62+0.58+0.65+0.60+0.61}{5} = \frac{3.06}{5} = 0.612\ \mathrm{s}$$

Deviations: $+0.008, -0.032, +0.038, -0.012, -0.002$. Squares sum to $0.002680\ \mathrm{s^2}$, so

$$s = \sqrt{\frac{0.002680}{4}} = 0.0259\ \mathrm{s}, \qquad
\delta\bar{t} = \frac{0.0259}{\sqrt{5}} = 0.0116\ \mathrm{s}.$$

Report $t = 0.612 \pm 0.012\ \mathrm{s}$. Note that the scatter ($\pm 0.026$ s for one reading) is close
to the expected reaction-time uncertainty — a good sign that nothing else is going wrong.

## 4. Propagating uncertainty / Belirsizliğin yayılması

You measure $a$ and $b$; you want $y$ built from them. Use the rule that matches the algebra.

| If | then | In words |
| --- | --- | --- |
| $y = a + b$ or $y = a - b$ | $\delta y = \sqrt{(\delta a)^2 + (\delta b)^2}$ | add **absolute** uncertainties in quadrature |
| $y = ab$ or $y = a/b$ | $\dfrac{\delta y}{y} = \sqrt{\left(\dfrac{\delta a}{a}\right)^2 + \left(\dfrac{\delta b}{b}\right)^2}$ | add **relative** uncertainties in quadrature |
| $y = a^n$ | $\dfrac{\delta y}{y} = \lvert n\rvert\,\dfrac{\delta a}{a}$ | multiply the relative uncertainty by the power |
| $y = ca$ ($c$ exact) | $\delta y = \lvert c\rvert\,\delta a$ | an exact constant does not add uncertainty |

**The one thing to notice.** In the power rule the exponent multiplies your uncertainty. In the free-fall
experiment $g = 2h/t^2$, so

$$\frac{\delta g}{g} = \sqrt{\left(\frac{\delta h}{h}\right)^2 + \left(2\frac{\delta t}{t}\right)^2}.$$

A 1% error in time costs you **2%** in $g$, while a 1% error in height costs only 1%. So spend your
effort on timing, not on measuring the height to the nearest millimetre. **Uncertainty analysis is not
paperwork — it tells you where to put your effort before you start.**

**TR:** Kuvvet kuralında üs, belirsizliği çarpar. $g = 2h/t^2$ için zamandaki %1 hata $g$'de %2 olur;
yükseklikteki %1 ise sadece %1. Bu yüzden emeğini zaman ölçümüne harca. Belirsizlik analizi evrak işi
değildir; deneye başlamadan önce nereye çalışacağını söyler.

---

### Worked propagation — g from a drop

$h = 1.200 \pm 0.005\ \mathrm{m}$, $t = 0.612 \pm 0.012\ \mathrm{s}$, and $g = 2h/t^2$.

**Symbolic first:** $g = 2ht^{-2}$, so $\dfrac{\delta g}{g} = \sqrt{\left(\dfrac{\delta h}{h}\right)^2 + \left(2\dfrac{\delta t}{t}\right)^2}$.

**Limiting check:** if timing were perfect ($\delta t \to 0$), $\delta g/g \to \delta h/h$ — the whole
uncertainty would come from the height. That is the right behaviour.

**Now numbers:**
$$g = \frac{2(1.200)}{(0.612)^2} = \frac{2.400}{0.3745} = 6.41\ \mathrm{m/s^2}$$
$$\frac{\delta h}{h} = \frac{0.005}{1.200} = 0.42\%, \qquad 2\frac{\delta t}{t} = 2\frac{0.012}{0.612} = 3.78\%$$
$$\frac{\delta g}{g} = \sqrt{(0.0042)^2 + (0.0378)^2} = 3.80\% \Rightarrow \delta g = 0.24\ \mathrm{m/s^2}$$

**Result:** $g = 6.41 \pm 0.24\ \mathrm{m/s^2}$.

That is nowhere near $9.81$ — and the uncertainty is what tells you so. The gap is $14\sigma$, far too
big to be bad luck, so something **systematic** is wrong. In this experiment it is almost always the
same thing: the timer starts when you *decide* to release, not when the ball leaves your fingers. The
error analysis did not just decorate the answer; it diagnosed the apparatus.

**TR:** Sonuç 9.81'den çok uzak ve bunu anlamamızı sağlayan şey belirsizliğin kendisidir. Fark $14\sigma$
kadar; bu kadarı şans olamaz, yani sistematik bir sorun var. Hata analizi sonucu süslemedi, düzeneği
teşhis etti.

## 5. Significant figures / Anlamlı rakamlar

Three rules cover everything you will meet:

1. **Multiplying or dividing:** the answer carries as many significant figures as the *least* precise input.
   $(2.0)(3.14159) = 6.3$, not $6.28318$.
2. **Adding or subtracting:** the answer is rounded to the *coarsest decimal place*.
   $12.11 + 0.3 = 12.4$, not $12.41$.
3. **Round once, at the end.** Carry extra digits through the algebra; round only the reported answer.

**With an uncertainty, the uncertainty decides the rounding.** Quote $\delta x$ to one significant
figure, then round $x$ to that decimal place:

- $g = 9.7382 \pm 0.2145 \rightarrow g = 9.7 \pm 0.2\ \mathrm{m/s^2}$ ✅
- $g = 9.7382 \pm 0.2145 \rightarrow g = 9.7382 \pm 0.2145\ \mathrm{m/s^2}$ ❌ — claims a precision you do not have

**A calculator has no idea how good your ruler is.** Writing down all its digits is not honesty, it is
the opposite: it advertises a precision you never measured.

**TR:** Belirsizliği bir anlamlı rakama yuvarla, sonucu da aynı basamağa. Hesap makinesinin cetvelin ne
kadar iyi olduğu hakkında bir fikri yok; bütün haneleri yazmak dürüstlük değil, tam tersidir.

## 6. Straightening a curve and reading a slope / Grafiği doğrultmak

Never test a curved law by eye. **Rearrange it into $y = mx$ form**, plot that, and check whether the
points fall on a line through the origin.

| Physical law | Plot this on $y$ | against this on $x$ | Slope $m$ means |
| --- | --- | --- | --- |
| $h = \tfrac12 g t^2$ | $h$ | $t^2$ | $m = g/2$, so $g = 2m$ |
| $T = 2\pi\sqrt{L/g}$ | $T^2$ | $L$ | $m = 4\pi^2/g$, so $g = 4\pi^2/m$ |
| $v^2 = 2as$ | $v^2$ | $s$ | $m = 2a$ |
| $f_s^{\max} = \mu_s N$ | $f_s^{\max}$ | $N$ | $m = \mu_s$ |

Why this is worth the trouble: a straight line makes the *law* testable, not just the number. Curvature,
or a line that misses the origin, is a message. **An intercept that should be zero but isn't is a
systematic error you can read straight off the graph** — in the free-fall experiment, a positive time
intercept is exactly the delay between "release" and "the timer noticing".

**TR:** Eğri bir yasayı gözle kontrol etme; $y = mx$ biçimine çevir. Sıfır olması gereken bir kesim
noktası sıfır değilse, sistematik hatayı doğrudan grafikten okuyorsun demektir.

---

### Reading the slope with its uncertainty

By hand: draw the best line, then the steepest and shallowest lines that still pass through the error
bars. Then
$$m = m_{\text{best}}, \qquad \delta m \approx \frac{m_{\max}-m_{\min}}{2}.$$

That is enough for a PHY101 report. The cell below does the same thing by least squares if you prefer —
paste your own numbers in. **Python is optional in this course; the graph on paper is not.**


```python
#@title Least-squares slope with its uncertainty (optional) {display-mode: "form"}
import numpy as np

# ---- paste your own measurements here (x, y and the uncertainty on each y) ----
x     = np.array([0.20, 0.40, 0.60, 0.80, 1.00])    # e.g. t^2 in s^2, or L in m
y     = np.array([0.98, 1.99, 2.92, 3.95, 4.90])    # e.g. h in m, or T^2 in s^2
y_err = np.array([0.05, 0.05, 0.06, 0.07, 0.08])    # uncertainty on each y value
# -------------------------------------------------------------------------------

w = 1.0 / y_err**2
fit, cov = np.polyfit(x, y, 1, w=np.sqrt(w), cov=True)
m, c = fit
dm, dc = np.sqrt(np.diag(cov))

residual = y - (m * x + c)
chi2_red = float(np.sum(w * residual**2) / (len(x) - 2))

print(f"slope     m = {m:.3f} +/- {dm:.3f}")
print(f"intercept c = {c:.3f} +/- {dc:.3f}")
print(f"reduced chi-squared = {chi2_red:.2f}   (near 1 is healthy)")
print()
print("Is the intercept consistent with zero?",
      "yes" if abs(c) <= 2 * dc else f"NO - {abs(c)/dc:.1f} sigma away, look for a systematic error")
print()
print("Then convert the slope to the physics -- use ONLY the line matching your experiment:")
print(f"  if you plotted h vs t^2 : g = 2m      = {2*m:.2f} +/- {2*dm:.2f} m/s^2")
print(f"  if you plotted T^2 vs L : g = 4pi^2/m = {4*np.pi**2/m:.2f} +/- {4*np.pi**2*dm/m**2:.2f} m/s^2")
```

## 7. Do the numbers agree? / Sonuçlar uyuşuyor mu

You have a measurement $x \pm \delta x$ and something to compare it with — an accepted value, or a
partner's result. **Do not use percentage difference to decide whether they agree.** Count sigmas:

$$n_\sigma = \frac{\lvert x_{\text{measured}} - x_{\text{expected}}\rvert}{\delta x}$$

| $n_\sigma$ | Verdict | What to write |
| --- | --- | --- |
| $< 1$ | Agrees | "Consistent with the accepted value." |
| $1 - 2$ | Agrees | "Consistent within uncertainty." |
| $2 - 3$ | Marginal | "Marginal; a systematic effect may be present." |
| $> 3$ | **Disagrees** | "Inconsistent — there is an unaccounted systematic error." Then name a candidate. |

A 5% discrepancy with a 10% uncertainty is a **success**. A 5% discrepancy with a 0.5% uncertainty is a
**failure that needs explaining**. The percentage on its own tells you neither.

**Comparing two measurements** of the same thing: $n_\sigma = \lvert x_1 - x_2 \rvert / \sqrt{(\delta x_1)^2 + (\delta x_2)^2}$.

**A result that disagrees is not a failed experiment.** A report that measures $g = 6.4 \pm 0.2$, shows
it is $14\sigma$ from $9.81$, identifies the release delay as the cause and supports that with a nonzero
time intercept earns full marks. A report that writes $g = 9.81$ because that is the expected answer
earns none. **In this course you are graded on the quality of your reasoning about the data, never on
how close you landed to the textbook number.**

**TR:** Uyuşmayı yüzde farkla değil, sigma sayarak karar ver. Uyuşmayan bir sonuç başarısız bir deney
değildir: farkı gösteren, sebebini adlandıran ve bunu veriyle destekleyen rapor tam puan alır; beklenen
sayıyı yazan rapor puan almaz. Bu derste veriyle kurduğun akıl yürütmeden not alırsın, kitaptaki sayıya
ne kadar yaklaştığından değil.

## 8. The report checklist / Rapor kontrol listesi

Every PHY101 lab report, one or two pages, in this order. Tick each line before you hand it in.

| ✓ | Section | What earns the mark |
| --- | --- | --- |
| ☐ | **Aim** | One sentence: which quantity, by which method. |
| ☐ | **Prediction** | The symbolic result **before** any data, plus one limiting-case check. |
| ☐ | **Apparatus** | What you used, with each instrument's reading uncertainty. |
| ☐ | **Data** | A table with units in the headers and every raw reading — never only the average. |
| ☐ | **Analysis** | The linearising plot, the slope with its uncertainty, propagation shown step by step. |
| ☐ | **Result** | $x \pm \delta x$ with a unit, correctly rounded, $\delta x$ to one significant figure. |
| ☐ | **Comparison** | $n_\sigma$ against the accepted value, and the verdict in words. |
| ☐ | **Systematics** | One named candidate error, its likely **sign**, and the evidence for it. |

**The sign matters.** "There was air resistance" is a sentence. "Air resistance lengthens the fall, so
our $g$ should read **low**, and it does" is physics.

**TR:** Her rapor bu sırayla ve iki sayfayı geçmeden. Sistematik hatanın yönünü söyle: "hava direnci
vardı" bir cümledir; "hava direnci düşüşü uzatır, bu yüzden $g$ olması gerekenden **küçük** çıkar ve
öyle çıktı" fiziktir.

---

### The four sentences that must be in every report

1. We predicted $\dots$ (symbolically), which behaves correctly in the limit $\dots$
2. We measured $\dots \pm \dots$ (unit).
3. This is $n_\sigma = \dots$ from the accepted value, so it **does / does not** agree.
4. The dominant systematic is $\dots$, which pushes the result **up / down**, and the evidence is $\dots$

**TR:** Her raporda bulunması gereken dört cümle yukarıdadır. Bunları yazabiliyorsan deneyi anlamışsın.

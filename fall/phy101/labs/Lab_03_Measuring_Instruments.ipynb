# Lab 03 — Measuring instruments

## Ölçme cihazları

**Aim.** Use a vernier caliper and a micrometer properly, and determine the density of an unknown metal cylinder with a fully propagated uncertainty.

**TR:** Gelmeden önce el kitabını oku ve aşağıdaki tahmini yap. Tezgâhta, düzenek çalışmadan önce

## 1. Prediction — before the bench / Tahmin

**Symbolically first.** A solid cylinder has diameter $d$, length $L$ and mass $m$. Show that

$$\rho = \frac{m}{V} = \frac{4m}{\pi d^2 L}$$

and that the uncertainty propagates as

$$\frac{\delta\rho}{\rho} = \sqrt{\left(\frac{\delta m}{m}\right)^2 + \left(2\frac{\delta d}{d}\right)^2 + \left(\frac{\delta L}{L}\right)^2}.$$

**Limiting check.** If the mass and length were known perfectly, $\delta\rho/\rho \to 2\,\delta d/d$.
Explain in one sentence why the diameter is the measurement to be most careful about.

**Now decide your strategy before you touch anything.** Which instrument for $d$, which for $L$, and
why? The factor of 2 on $\delta d/d$ should decide it for you. Write your intended plan in the logbook —
you are predicting where the uncertainty will come from, and then you will check whether you were right.

**TR:** Önce sembolik olarak yaz, sonra sınır durumunu kontrol et: kütle ve uzunluk kusursuz bilinse
belirsizlik $2\,\delta d/d$'ye iner. Çapın neden en dikkatli ölçülmesi gereken büyüklük olduğunu bir
cümleyle açıkla, sonra hangi cihazı nerede kullanacağına **ölçmeye başlamadan** karar ver.


## 2. Apparatus / Düzenek

- Metal cylinder (material not disclosed)
- Vernier caliper — read its smallest division and note it
- Micrometer screw gauge — check its zero before use
- Digital balance
- Metre rule for comparison

**Your bench may differ.** If your caliper reads 0.02 mm rather than 0.05 mm, or the balance steps in
0.01 g, use *your* instrument's value — never a number copied from this brief or from another group.

**TR:** Tezgâhın farklı olabilir. Kumpasın 0.05 mm yerine 0.02 mm okuyorsa **kendi** cihazının değerini
kullan; bu metinden veya başka bir gruptan kopyalanmış sayı kullanma.


## 3. Procedure / Yöntem

1. **Zero both instruments.** Record the caliper and micrometer zero errors, with sign. Subtract them
   from every subsequent reading.
2. **Diameter:** measure $d$ with the micrometer at **five different places** along the cylinder, rotating
   it 90° between readings. Record all five.
3. **Length:** measure $L$ five times with the caliper.
4. **Mass:** weigh the cylinder three times, removing and replacing it between weighings.
5. Repeat step 2 with the *caliper* as well, so you can compare the two instruments at the end.

**Why five places and a 90° rotation.** If the cylinder is slightly tapered or oval, a single reading is
biased and you would never know. Five spread-out readings turn that bias into visible scatter — which
your standard deviation then reports honestly. **This is how you convert a hidden systematic error into
an admitted random one**, and it is the most useful trick in experimental physics.

**TR:** Beş farklı yerde ve 90° döndürerek ölçmenin sebebi: silindir hafifçe konik veya oval ise tek
okuma yanlıdır ve bunu asla fark etmezsin. Yayılmış beş okuma bu yanlılığı görünür saçılmaya çevirir;
standart sapma da onu dürüstçe bildirir. Gizli bir sistematik hatayı, kabul edilmiş bir rastgele hataya
çevirmiş olursun — deneysel fiziğin en kullanışlı hilesi budur.

## 5. Analysis / Analiz

1. Convert everything to SI: metres and kilograms. **Do this first**, before any arithmetic —
   mixing mm with g is the most common way this experiment goes wrong.
2. Compute $V = \tfrac{\pi}{4} d^2 L$ and then $\rho = m/V$.
3. Propagate with the expression you derived in §1. Show every term as a percentage so you can see
   which one dominates.
4. Round: $\delta\rho$ to one significant figure, $\rho$ to the matching decimal place.
5. Compare your micrometer-based $d$ with your caliper-based $d$: count sigmas. Do they agree?

**Then identify the metal** from your result and its uncertainty:

| Metal | $\rho$ (kg/m³) |
| --- | --- |
| Aluminium | 2700 |
| Zinc | 7140 |
| Iron / steel | 7870 |
| Brass | 8500 |
| Copper | 8960 |
| Lead | 11340 |

**The honest version of this step.** State which metals your interval $\rho \pm \delta\rho$ *cannot*
exclude, not just the nearest one. If you get $8600 \pm 500$ kg/m³ you have not proved it is brass —
you have shown it is brass or copper and ruled out aluminium, zinc and lead. Saying exactly that is the
correct answer, and it is better physics than a confident guess.

**TR:** Bu adımın dürüst hâli: yalnızca en yakın metali söylemek değil, $\rho \pm \delta\rho$ aralığının
**dışlayamadığı** metalleri söylemektir. $8600 \pm 500$ bulduysan pirinç olduğunu kanıtlamadın; pirinç
veya bakır olduğunu gösterdin, alüminyum, çinko ve kurşunu dışladın. Doğru cevap tam olarak budur.


## 6. Discussion / Tartışma

<details><summary>Which term dominated your uncertainty, and did your §1 prediction get it right? / Belirsizliğe hangi terim hâkim oldu?</summary>

Almost always the diameter, because it enters squared: a 0.5% error in $d$ becomes 1% in $\rho$, while a
0.5% error in $m$ or $L$ stays 0.5%. If the diameter did **not** dominate for you, that is interesting —
it usually means the cylinder is tapered or the balance is drifting, and one of those belongs in your
systematics paragraph.

**TR:** Neredeyse her zaman çap, çünkü kareli giriyor. Çap baskın **çıkmadıysa** bu ilginçtir: genellikle
silindirin konik olduğunu veya terazinin kaydığını gösterir.

</details>

<details><summary>You over-tighten the micrometer. Which way does ρ move? / Mikrometreyi fazla sıkarsan ρ hangi yöne kayar?</summary>

Over-tightening compresses the cylinder slightly, so $d$ reads **too small**. Then $V \propto d^2$ is too
small and $\rho = m/V$ comes out **too large**. Naming the direction like this — not just "there was an
error" — is what a systematics paragraph is for.

**TR:** Fazla sıkmak çapı **küçük** okutur, hacim küçülür, yoğunluk **büyük** çıkar. Sistematik hata
paragrafı tam olarak bunun için vardır: yönünü söylemek.

</details>

<details><summary>Why measure d five times but m only three? / Neden çapı beş, kütleyi üç kez?</summary>

Because the diameter has real physical variation along the cylinder *and* enters squared, so it deserves
the most attention. The mass is a single property of a single object measured by a stable instrument; its
repeats mostly test whether the balance is behaving. Effort follows the propagation formula — that is
what §1 was for.

**TR:** Çapın silindir boyunca gerçek bir değişimi var ve kareli giriyor. Kütle tek bir nesnenin tek bir
özelliği. Emek, yayılma formülünün gösterdiği yere gider.

</details>

<details><summary>Your interval excludes every metal in the table. What now? / Aralığın tablodaki her metali dışlıyorsa?</summary>

Then either the cylinder is an alloy or hollow, or you have a systematic error — and you should say
which you think it is and why. Check: is it hollow (compare mass with the volume you measured), did you
subtract the zero errors, and did you convert mm to m consistently? A factor of 1000 in the density is a
unit-conversion fingerprint, not a new material.

**TR:** Ya silindir alaşım veya içi boş, ya da sistematik hatan var; hangisi olduğunu ve neden öyle
düşündüğünü söyle. Yoğunlukta 1000 kat fark yeni bir malzeme değil, birim çevirme parmak izidir.

</details>

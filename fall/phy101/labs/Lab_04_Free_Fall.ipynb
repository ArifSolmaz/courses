# Lab 04 — Motion in one dimension — free fall

## Bir boyutta hareket — serbest düşme

**Week 04 laboratory session · 12 – 16 October 2026**
**Official schedule entry / İzlencedeki adı:** *Bir Boyutta Hareket - Serbest Düşme*

It measures the physics taught in **Week 02**, so the lecture comes first and the experiment confirms it.

**Aim.** Measure the acceleration due to gravity by timing a fall from several heights, and detect the timing systematic from the graph's intercept.

**TR.** Birkaç farklı yükseklikten düşme süresini ölçerek yerçekimi ivmesini belirle ve zamanlama kaynaklı sistematik hatayı grafiğin kesim noktasından yakala.

**Before you arrive:** read [Lab 00 — measurement and uncertainty toolkit](Lab_00_Uncertainty_Toolkit.ipynb) and complete the prediction below. You will be asked for your
predicted number at the bench, before any apparatus is switched on.

**TR:** Gelmeden önce el kitabını oku ve aşağıdaki tahmini yap. Tezgâhta, düzenek çalışmadan önce
tahmin ettiğin sayı sorulacak.


## 1. Prediction — before the bench / Tahmin

**Symbolically first.** An object released from rest falls a height $h$ in time $t$ with no air
resistance. From $h = \tfrac12 g t^2$:

$$\boxed{\,g = \frac{2h}{t^2}\,}$$

**But do not test it in that form.** A single $h$ and $t$ gives one number with no way to see what is
wrong with it. Instead rearrange into a straight line:

$$h = \frac{g}{2}\,t^2 \quad \Longrightarrow \quad \text{plot } h \text{ against } t^2, \text{ slope } m = \frac{g}{2}, \text{ so } g = 2m.$$

**Predict the intercept.** If the model is right and your timing is unbiased, the line passes through the
origin: at $h = 0$, $t = 0$. **Write down now what a positive intercept on the $t^2$ axis would mean.**
You will almost certainly see one, and it is the most instructive number in this experiment.

**Limiting check.** In $g = 2h/t^2$, what happens to $g$ as $t \to 0$ at fixed $h$? Explain why that
tells you a very short fall is a bad experiment, and choose your smallest height accordingly.

**Where to spend your effort.** From the toolkit's power rule,
$$\frac{\delta g}{g} = \sqrt{\left(\frac{\delta h}{h}\right)^2 + \left(2\frac{\delta t}{t}\right)^2}.$$
Time enters with a factor 2 **and** your stopwatch uncertainty is far worse than your metre rule's, so
timing dominates completely. Use the reaction time you measured in Lab 02, Exercise C — not 0.01 s.

**TR:** Önce sembolik: $g = 2h/t^2$. Ama bu biçimde test etme; $h$'yi $t^2$'ye karşı çizip eğimden
$g = 2m$ bul. Kesim noktasının ne anlama geldiğini **şimdi** yaz. Zaman iki kat ağırlıkla giriyor ve
kronometre belirsizliğin cetvelden çok daha kötü; bu yüzden Lab 02'de bulduğun tepki süresini kullan.


## 2. Apparatus / Düzenek

- Steel ball (dense and small, so air resistance stays negligible)
- Metre rule or tape fixed vertically to a stand
- Stopwatch — **or** an electronic gate/photogate timer if your bench has one
- Soft catch tray

**If you have a photogate,** use it: it removes the reaction-time systematic almost entirely and your
uncertainty drops by more than an order of magnitude. Record which timing method you used — your whole
error budget depends on it, and the two methods give visibly different graphs.

**TR:** Foto-kapı varsa kullan: tepki süresi sistematiğini neredeyse tamamen kaldırır. Hangi yöntemi
kullandığını kaydet; bütün hata bütçen buna bağlı ve iki yöntem gözle görülür biçimde farklı grafik verir.


## 3. Procedure / Yöntem

1. Fix the rule vertically. Define $h$ as the distance from the **bottom of the ball at release** to the
   catch surface, and keep that definition for every trial.
2. Choose **five heights** spread as widely as the bench allows, e.g. 0.40, 0.70, 1.00, 1.30, 1.60 m.
   Spread matters more than number: a wide range pins the slope down far better than five clustered points.
3. At each height, release from rest — **drop, do not throw**. Any downward push adds an initial velocity
   and breaks the model.
4. Time the fall **five times** at each height. Record all 25 readings.
5. Measure each $h$ once carefully with its reading uncertainty.

**The release is the hard part.** Hold the ball between two fingers and open them; do not flick. If you
are using a stopwatch, one person should both release and time — a second person adds their own reaction
delay on top of yours.

**TR:** Bırak, atma: aşağı doğru en küçük itiş bile modeli bozar. Kronometre kullanıyorsan aynı kişi hem
bıraksın hem tutsun; ikinci kişi kendi tepki gecikmesini senin üstüne ekler.


## 4. Data / Veri

| $h$ (m) | $\delta h$ (m) | $t_1$ | $t_2$ | $t_3$ | $t_4$ | $t_5$ (s) | $\bar{t}$ (s) | $\delta\bar{t}$ (s) | $\bar{t}^2$ (s²) | $\delta(t^2)$ (s²) |
| --- | --- | --- | --- | --- | --- | --- | --- | --- | --- | --- |
|  |  |  |  |  |  |  |  |  |  |  |
|  |  |  |  |  |  |  |  |  |  |  |
|  |  |  |  |  |  |  |  |  |  |  |
|  |  |  |  |  |  |  |  |  |  |  |
|  |  |  |  |  |  |  |  |  |  |  |

Two columns people get wrong:

- $\delta\bar{t} = s/\sqrt{5}$ — the standard error of **that height's** five trials, not of all 25.
- $\delta(t^2) = 2\,\bar{t}\,\delta\bar{t}$ — from the power rule, since
  $\delta(t^2)/t^2 = 2\,\delta t/t$. This is the error bar you draw on the horizontal axis.

**TR:** İki sütun sık yanlış yapılır: $\delta\bar{t}$ o yüksekliğin beş denemesinin standart hatasıdır,
25'inin değil; ve $\delta(t^2) = 2\bar{t}\,\delta\bar{t}$'dir.


## 5. Analysis / Analiz

1. Plot $h$ (vertical) against $\bar{t}^2$ (horizontal), **on paper**, with horizontal error bars.
2. Draw the best straight line, then the steepest and shallowest lines that still pass through the error
   bars. Read $m_{\text{best}}$, $m_{\max}$, $m_{\min}$.
3. $g = 2m$ and $\delta g = 2\,\delta m$ with $\delta m = (m_{\max}-m_{\min})/2$.
4. Read the **intercept** and its uncertainty. Is it consistent with zero?
5. Compare with $g = 9.81\ \mathrm{m/s^2}$ by counting sigmas.

You may check your hand fit with the least-squares cell in [Lab 00 — measurement and uncertainty toolkit](Lab_00_Uncertainty_Toolkit.ipynb) §6. The hand-drawn graph is the
requirement; the code is the optional check — never the other way round.

---

### Reading the intercept — the point of the whole experiment

Suppose your line crosses the $t^2$ axis at $t^2 = +0.02\ \mathrm{s^2}$, i.e. the fit says a fall of zero
height still takes about $0.14$ s. Nothing physical takes time to fall nowhere. So that intercept is not
physics, it is **your apparatus**: the stopwatch runs for roughly 0.14 s during which the ball has not
yet been released — the gap between deciding to let go and actually letting go.

Now notice what this does to $g$:

- The intercept shifts every point the **same** way, so it does **not** change the slope much — which is
  exactly why the graph method is better than computing $g = 2h/t^2$ from a single trial. A single trial
  folds the delay into $t$ and gives you a **low** $g$ every time.
- So: **the linearised graph is partly immune to a systematic error that would ruin a single
  measurement, and it also measures that error for you.** That is a real experimental technique, not a
  lab-report ritual.

**Predict the sign before you look:** a delay makes each measured $t$ too **long**, so a single-shot
$g = 2h/t^2$ comes out too **small**. Check whether your own single-trial estimate from the largest
height sits below your slope-based $g$. It usually does.

**TR:** Doğrunun $t^2$ eksenini pozitif bir noktada kesmesi fizik değil, düzeneğindir: bırakmaya karar
vermenle gerçekten bırakman arasındaki gecikme. Bu gecikme her noktayı aynı yöne kaydırdığı için eğimi
fazla değiştirmez — tek bir ölçümü mahvedecek sistematik hataya karşı doğrultulmuş grafiğin kısmen bağışık
olmasının sebebi budur, üstelik o hatayı senin için ölçer de. Gecikme her $t$'yi **uzun** okutur, yani tek
denemelik $g$ **küçük** çıkar.


## 6. Discussion / Tartışma

<details><summary>Does air resistance make your g too big or too small? / Hava direnci g'yi büyük mü küçük mü gösterir?</summary>

Air resistance opposes the fall, so the ball takes **longer** than the vacuum model predicts. Longer $t$
with the same $h$ gives a **smaller** $g$. So air resistance pushes your result **down**.

Two follow-ups worth a sentence each in the report: the effect grows with speed, so it hurts your
**tallest** drops most and should make the graph bend very slightly *downward* at large $t^2$; and it is
much smaller than your timing error for a steel ball over a metre, so you should **not** expect to see
it. Claiming you saw air resistance in this data, with 0.2 s timing, is not defensible.

**TR:** Hava direnci düşüşü uzatır, süre uzar, $g$ **küçük** çıkar. Ama çelik bilye için bu etki senin
zamanlama hatandan çok küçüktür, yani görmen beklenmez. 0.2 s'lik zamanlamayla "hava direncini gördük"
demek savunulabilir değil.

</details>

<details><summary>Why plot h against t² rather than t against h? / Neden h'yi t²'ye karşı çiziyoruz?</summary>

Because $h = \tfrac12 g t^2$ is linear in $t^2$ with zero intercept, so plotting $h$ against $t^2$ gives a
straight line whose slope *is* the physics and whose intercept *is* your systematic error. Plot $t$
against $h$ and you get a curve, from which you can read neither reliably by eye.

The general habit: **rearrange the law until the unknown you want is a slope.** Slopes use all your data
at once and are far more robust than any single point.

**TR:** Çünkü $h$, $t^2$'de doğrusaldır: eğim fiziği, kesim noktası sistematik hatanı verir. Genel
alışkanlık: yasayı, aradığın bilinmeyen bir *eğim* olana kadar düzenle.

</details>

<details><summary>Five heights × five trials, or one height × twenty-five trials? / Beş yükseklik mi, tek yükseklikte yirmi beş deneme mi?</summary>

Five heights, decisively. Twenty-five trials at one height give a beautifully small *random* uncertainty
on one point — and no way whatsoever to detect the release delay, because with a single point you cannot
see an intercept. You would end up with a precise, confidently wrong $g$.

This is the lesson from the toolkit in its sharpest form: **more data does not beat better design.**

**TR:** Kesinlikle beş yükseklik. Tek yükseklikte yirmi beş deneme, bir noktada çok küçük rastgele
belirsizlik verir ama bırakma gecikmesini görmenin hiçbir yolunu bırakmaz: tek noktayla kesim noktası
göremezsin. Sonuç hassas ama kendinden emin biçimde yanlış bir $g$ olur. Daha çok veri, daha iyi
tasarımı yenmez.

</details>

<details><summary>Your g comes out 9.2 ± 0.6 m/s². Is the experiment a success? / 9.2 ± 0.6 çıktıysa deney başarılı mı?</summary>

Yes. $n_\sigma = |9.2-9.81|/0.6 \approx 1.0$, so it is consistent with the accepted value within
uncertainty. Write it that way. Do **not** apologise for a "6% error" — with honest 0.2 s timing, ±0.6
is the uncertainty you earned, and the result agrees.

Now compare with a group who reported $9.81 \pm 0.02$ from the same stopwatches. Their uncertainty is not
credible, and that is the more serious error.

**TR:** Evet. $n_\sigma \approx 1$, yani belirsizlik içinde uyuşuyor; böyle yaz. "%6 hata" diye özür
dileme. Aynı kronometreyle $9.81 \pm 0.02$ bildiren bir grubun belirsizliği inandırıcı değildir ve daha
ciddi bir hatadır.

</details>

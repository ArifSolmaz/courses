# Lab 02 — Laboratory introduction

## Laboratuvar tanıtımı

**Week 02 laboratory session · 28 September – 2 October 2026**
**Official schedule entry / İzlencedeki adı:** *Laboratuvar Tanıtımı*

It introduces the laboratory itself; no lecture topic is being tested.

**Aim.** Learn what the laboratory expects of you: how to work safely, how to keep a logbook, and how to turn a reading into a reportable result.

**TR.** Laboratuvarın senden ne beklediğini öğren: güvenli çalışma, kayıt defteri tutma ve bir okumayı raporlanabilir bir sonuca çevirme.

**Scope note.** The experiment is scheduled by the department and is the same for every section. The
*analysis technique* in this brief (uncertainty, linearising, counting sigmas) is a laboratory
requirement and is **not** part of the common midterm or final; the *physics* being measured is.

**Before you arrive:** read [Lab 00 — measurement and uncertainty toolkit](Lab_00_Uncertainty_Toolkit.ipynb) and complete the prediction below. You will be asked for your
predicted number at the bench, before any apparatus is switched on.

**TR:** Gelmeden önce el kitabını oku ve aşağıdaki tahmini yap. Tezgâhta, düzenek çalışmadan önce
tahmin ettiğin sayı sorulacak.


## 1. Prediction / Tahmin

No physics is measured today, so there is nothing to predict. Instead answer this in your logbook before
the session:

> You measure the width of your bench with a metre rule and get 74.8 cm. Write the result in the form
> $x \pm \delta x$ with a unit. How did you choose $\delta x$?

**TR:** Bugün fizik ölçmüyoruz. Bunun yerine tezgâhın genişliğini metre ile ölçtüğünü varsay ve sonucu
$x \pm \delta x$ biçiminde yaz; $\delta x$'i nasıl seçtiğini açıkla.


### Physical meaning: connect the reading to the object

Sketch the object, the endpoints defining its measured length and the instrument's zero. Before recording a number, state what the instrument actually measures. A stopwatch reading between two hand actions can include human reaction delays as well as the physical event.

**Predict.** Two instruments disagree on the same length. Is the one showing more decimal places necessarily right?

<details><summary>Explain what the digits can establish</summary>

No. Resolution, zero offset, alignment and contact pressure all matter. Compare the readings with their uncertainties and check calibration. For timing, write which action starts and stops the clock; do not name “reaction time” without identifying how it changes the measured interval.

</details>

Apply this to each bench exercise: picture → operational definition → prediction → reading → interpretation. Keep the original observation even when your prediction fails.

**TR:** Ölçtüğün şeyin fiziksel tanımını önce yaz; fazla basamak tek başına doğruluk değildir.

## 2. Safety / Güvenlik

Read, then sign the laboratory register.

- Bags and coats go under the bench, never in the walkway. Trip hazards cause most lab injuries.
- Masses, trolleys and rotating parts: keep hands and hair clear, and **never stand where a falling mass can land on your foot**.
- Nothing is switched on until the demonstrator has checked the assembly.
- Report a broken instrument immediately. A quietly damaged caliper produces a whole semester of
  systematically wrong data for everyone who uses it afterwards.
- No food or drink at the bench.

**TR:** Çantalar tezgâh altına; düşen kütlelerin ayağına gelebileceği yerde durma; düzenek kontrol
edilmeden hiçbir şey çalıştırılmaz; bozuk cihazı hemen bildir — sessizce zarar görmüş bir kumpas, ondan
sonra kullanan herkes için sistematik olarak yanlış veri üretir.


## 3. The logbook / Kayıt defteri

One rule: **write down what you actually did and what the instrument actually showed, at the moment it
showed it.** Not afterwards, not tidied up.

| Do | Do not |
| --- | --- |
| Record every raw reading | Record only the average |
| Cross out a wrong entry with one line, keep it legible | Erase, overwrite, or use correction fluid |
| Write the unit in the column heading | Leave units to be guessed later |
| Note the instrument and its reading uncertainty | Assume you will remember which caliper you used |
| Note anything odd (a draught, a wobble, a late start) | Silently drop a trial you did not like |

**Why crossing out beats erasing.** A discarded reading is data about your experiment. If three of your
ten trials were late starts, the pattern in those three is the evidence for your systematic error. Erase
them and you have destroyed your own diagnosis.

**TR:** Tek kural: gerçekte ne yaptığını ve cihazın gerçekte ne gösterdiğini, o anda yaz. Yanlış kaydı
tek çizgiyle çiz, silme: attığın ölçüm de deneyin hakkında veridir. Üç denemende geç başlama olduysa,
sistematik hatanın kanıtı tam olarak o üç denemedir.


## 4. Bench exercises / Tezgâh alıştırmaları

Work through all four. They take about an hour and they are the whole point of today.

### Exercise A — the same length, three instruments

Measure the **diameter of a metal rod** with a metre rule, then a vernier caliper, then a micrometer.

| Instrument | Reading | $\delta x$ | Relative $\delta x / x$ |
| --- | --- | --- | --- |
| Metre rule |  mm |  mm |  % |
| Vernier caliper |  mm |  mm |  % |
| Micrometer |  mm |  mm |  % |

Do the three results agree within their uncertainties? Count sigmas between the rule and the micrometer.

### Exercise B — find a zero error

Close the caliper jaws fully. Does it read exactly zero? Record the offset. Then measure the rod again
and correct for it. **This is a systematic error you just removed by looking for it** — the only way
systematic errors ever get removed.

### Exercise C — reaction time

Your partner drops a ruler between your fingers; you catch it. Convert the fall distance $d$ to a time
using $d = \tfrac12 g t^2$, i.e. $t = \sqrt{2d/g}$. Repeat five times and report $\bar{t} \pm s/\sqrt{N}$.

**Keep this number.** It is *your* timing uncertainty, and you will use it in the Week 04 free-fall
experiment instead of the stopwatch's advertised 0.01 s.

### Exercise D — one honest result

Measure the mass of the rod on the digital balance and compute its **density** $\rho = m/V$ with
$V = \pi (d/2)^2 L$. Carry the uncertainty through using the product and power rules. Report
$\rho \pm \delta\rho$ to three significant figures, then look up which metal it is.

**TR:** Dört alıştırmayı da yap. C'de bulduğun tepki süresini sakla: 4. hafta serbest düşme deneyinde
kronometrenin ilan ettiği 0.01 s yerine **kendi** zaman belirsizliğin olarak onu kullanacaksın.


## 5. Discussion / Tartışma

<details><summary>Why is the micrometer's answer not automatically the best one? / Mikrometrenin sonucu neden otomatik olarak en iyisi değildir?</summary>

Because precision and accuracy are different things. The micrometer has the smallest reading
uncertainty, so it is the most **precise**. But if it has a zero error, or you over-tighten it and
compress the rod, it is confidently wrong — precise and inaccurate. The metre rule cannot be very
precise but it is hard to bias.

This is Exercise B's real lesson: a better instrument raises the stakes on systematic error. A crude
instrument's mistakes are visible in the scatter; a fine instrument's mistakes hide inside a tight,
trustworthy-looking number.

**TR:** Hassasiyet ve doğruluk farklı şeylerdir. Mikrometre en hassastır, ama sıfır hatası varsa veya
çubuğu sıkarak bastırıyorsan kendinden emin biçimde yanlıştır. Kaba bir cihazın hatası saçılmada
görünür; hassas bir cihazın hatası dar ve güvenilir görünen bir sayının içine saklanır.

</details>

<details><summary>Should you throw away a reading that looks wrong? / Yanlış görünen bir ölçümü atmalı mısın?</summary>

Only if you can name a reason from the logbook — "the trolley clipped the rail on trial 4" — and you
say so in the report. Discarding a point because it spoils the graph is falsifying data.

If a point is far out and you have **no** reason for it, keep it and say that you kept it. An honest
outlier with a large uncertainty is worth more than a tidy graph you cannot defend.

**TR:** Yalnızca defterinde bir sebep varsa ve bunu raporda yazıyorsan. Grafiği bozduğu için bir noktayı
atmak veri tahrifidir. Sebebini bilmiyorsan noktayı tut ve tuttuğunu söyle.

</details>

<details><summary>The balance reads 0.1 g steps. Is δm = 0.1 g the whole story? / Terazi 0.1 g adımlıysa belirsizlik bu kadar mıdır?</summary>

No. That is the *display resolution*. Also check: does it read zero with nothing on it, does the reading
drift while you watch, and is the bench vibrating? Any of those can exceed 0.1 g. Quote the largest
effect you can actually detect, not the smallest one the display can show.

**TR:** Hayır, o yalnızca gösterge çözünürlüğüdür. Boşken sıfır mı gösteriyor, okuma kayıyor mu, tezgâh
titriyor mu? Tespit edebildiğin en büyük etkiyi yaz, göstergenin gösterebildiği en küçüğü değil.

</details>


## Report / Rapor

Hand in the one-to-two page report set out in [Lab 00 — measurement and uncertainty toolkit](Lab_00_Uncertainty_Toolkit.ipynb) §8. The four sentences required there —
prediction, result with uncertainty, $n_\sigma$ verdict, named systematic with its sign — are what the
mark is for.

**You are not graded on how close you land to the accepted value.** You are graded on whether your
reasoning about your own data is sound. A result that disagrees, correctly diagnosed, scores full marks.

**TR:** Not, kabul edilen değere ne kadar yaklaştığından değil, kendi verinle kurduğun akıl yürütmeden
verilir. Doğru teşhis edilmiş, uyuşmayan bir sonuç tam puan alır.

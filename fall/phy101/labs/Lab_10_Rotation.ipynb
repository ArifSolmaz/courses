## 1. Prediction — before the bench / Tahmin

### Symbolically first

A string is wound round a spindle of radius $R$ on a disc that turns freely; a mass $m$ hangs from the
string and is released from rest. The mass falls a height $h$ in time $t$.

**Linear side:** the mass has constant acceleration $a$, so
$$h = \tfrac12 a t^2 \quad \Longrightarrow \quad a = \frac{2h}{t^2}.$$

**The link between linear and angular** — the string does not slip, so
$$\boxed{\,v = R\omega, \qquad a = R\alpha \quad \Longrightarrow \quad \alpha = \frac{a}{R} = \frac{2h}{Rt^2}\,}$$

**Angular kinematics, if $\alpha$ is constant:**
$$\omega = \alpha t, \qquad \theta = \tfrac12\alpha t^2, \qquad \omega^2 = 2\alpha\theta.$$

### Limiting checks

1. A larger spindle radius $R$ with the same $a$ gives a **smaller** $\alpha$. Explain that in words — it
   is the whole reason gearboxes exist.
2. As $R \to 0$, $\alpha \to \infty$ for fixed $a$. Why is a very small spindle a bad experimental choice?
3. $\theta$ should grow as $t^2$ while $\omega$ grows as $t$. Which of the two is easier to measure
   accurately, and why?

### How you will test "constant α"

Do **not** measure one time and divide. Measure $h$ at several values, and plot

$$h \text{ against } t^2 \quad \Longrightarrow \quad \text{slope } = \frac{a}{2}, \text{ intercept } = 0.$$

A straight line through the origin *is* the evidence that $\alpha$ is constant. Curvature would mean the
acceleration changes during the fall — which is what friction in the bearing actually does.

**Predict the sign now:** bearing friction opposes the rotation. Will your measured $\alpha$ come out
**larger** or **smaller** than $a_{\text{ideal}}/R$? Write it down.

**TR:** Önce sembolik: $a = 2h/t^2$ ve kaymama koşulundan $\alpha = a/R$. Sabit $\alpha$'yı tek ölçümle
bölerek test etme; $h$'yi $t^2$'ye karşı çiz. Orijinden geçen doğru, $\alpha$'nın sabit olduğunun
kanıtıdır. Yatak sürtünmesinin $\alpha$'yı **büyük** mü **küçük** mü göstereceğini şimdi yaz.


## 2. Apparatus / Düzenek

- Rotating disc or wheel on a low-friction axle, with a spindle of known radius
- String, hanging masses, metre rule
- Stopwatch, or a photogate counting spokes/marks, or a rotary encoder if available
- Vernier caliper to measure the spindle radius — **measure the diameter and halve it**, and include the
  string thickness if it is a significant fraction of $R$

**A note on $R$.** It enters $\alpha = a/R$ directly, so a 3% error in $R$ is a 3% error in every angular
result. The string wraps *around* the spindle, so the effective radius is
$R_{\text{eff}} = R_{\text{spindle}} + r_{\text{string}}$. With a 10 mm spindle and 1 mm string that is a
**10%** correction — far larger than anything else in this experiment. Measure both.

**TR:** $R$ doğrudan girer, üstelik ip milin **etrafına** sarıldığı için etkin yarıçap
$R_{\text{mil}} + r_{\text{ip}}$'dir. 10 mm mil ve 1 mm ip için bu **%10**'luk bir düzeltmedir; deneydeki
her şeyden büyüktür. İkisini de ölç.


## 3. Procedure / Yöntem

1. Measure the spindle diameter five times with the caliper; measure the string diameter. Compute
   $R_{\text{eff}} \pm \delta R$.
2. Wind the string so the mass hangs at your largest chosen height. Mark the release point.
3. Release from rest and time the fall to the floor. **Five trials** at each of **five heights**
   (e.g. 0.30, 0.50, 0.70, 0.90, 1.10 m).
4. Keep the hanging mass **the same** throughout. You are testing kinematics here, not dynamics — the
   torque–inertia relation is the Week 11 experiment.
5. If you have a photogate on the disc, also record $\omega$ directly at two or three moments during one
   run, so you can check $\omega = \alpha t$ independently.

**Let go, do not push, and keep the string vertical.** A string that rubs on the edge of the spindle
housing adds a variable friction torque, and it will show up as scatter you cannot explain.

**TR:** Bırak, itme; ip düşey kalsın. Mil yuvasının kenarına sürten bir ip değişken sürtünme torku ekler
ve açıklayamayacağın bir saçılma olarak görünür.

## 5. Analysis / Analiz

1. Plot $h$ against $\bar{t}^2$ on paper with error bars. Best line plus steepest/shallowest.
2. Is it straight? Is the intercept consistent with zero? **Answer both explicitly** — the first tests
   constant acceleration, the second tests your timing.
3. $a = 2m_{\text{slope}}$, $\delta a = 2\delta m_{\text{slope}}$.
4. $\alpha = a/R_{\text{eff}}$, with
   $\dfrac{\delta\alpha}{\alpha} = \sqrt{\left(\dfrac{\delta a}{a}\right)^2 + \left(\dfrac{\delta R}{R}\right)^2}$.
5. Test $v = R\omega$ as set out in §4.
6. Convert $\omega$ at the floor into **rev/min** as a sanity check: a disc doing 200 rpm is plausible, one
   doing 20 000 rpm means you dropped a factor somewhere. Also express $\theta$ in revolutions and check
   it against the number of string turns that actually unwound — **a count you can verify with your
   eyes.**

---

### Why this experiment is about units as much as rotation

Almost every failure here is a radian problem, not a physics problem:

- $\alpha = a/R$ gives rad/s² **only if** $R$ is in metres. Millimetres give you an answer 1000× too big.
- $\omega$ in rad/s converts as $1\ \mathrm{rev/s} = 2\pi\ \mathrm{rad/s}$ and
  $1\ \mathrm{rpm} = 2\pi/60\ \mathrm{rad/s}$.
- The radian is dimensionless, so it **disappears silently** from your unit check. $v = R\omega$ looks
  like m = m·(1/s) dimensionally, and dimensional analysis will therefore not catch a degree/radian
  mix-up. You have to catch it yourself.

That last point is why the revolution count in step 6 matters: it is the one check in this experiment
that a wrong angular unit cannot survive. **When a quantity's unit is invisible to dimensional analysis,
find a physical thing to count instead.**

**TR:** Buradaki hataların neredeyse tamamı radyan hatasıdır, fizik hatası değil. Radyan boyutsuz olduğu
için birim kontrolünden **sessizce** kaybolur; $v = R\omega$ boyutsal olarak doğru görünür ve derece/radyan
karışıklığını yakalamaz. 6. adımdaki tur sayımı bu yüzden önemlidir: yanlış açısal birimin atlatamayacağı
tek kontroldür. Bir birim boyut analizine görünmüyorsa, sayabileceğin fiziksel bir şey bul.


## 6. Discussion / Tartışma

<details><summary>Your α is smaller than a/R predicts from an ideal frictionless disc. Why? / α beklenenden küçükse?</summary>

Bearing friction applies a torque **opposing** the rotation, so the net torque is less than the string
tension alone would give, and the disc accelerates less. Your measured $\alpha$ is therefore the **true**
one for your apparatus; it is the frictionless *prediction* that is too large.

Say it that way round in the report. The measurement is not wrong — the idealised model omitted a torque
that is really there.

**TR:** Yatak sürtünmesi dönmeye **karşı** bir tork uygular, net tork azalır, disk daha az ivmelenir.
Ölçtüğün $\alpha$ düzeneğin **gerçek** değeridir; büyük olan, sürtünmesiz **tahmindir**. Ölçüm yanlış
değil; idealleştirilmiş model gerçekten var olan bir torku atlamış.

</details>

<details><summary>Does the hanging mass affect α? / Asılı kütle α'yı etkiler mi?</summary>

Yes — a bigger mass gives a bigger tension, a bigger torque, and a bigger $\alpha$. But that is a
**dynamics** question ($\sum\tau = I\alpha$), and it is the Week 11 experiment.

This week you keep $m$ fixed on purpose, so that the only thing being tested is the *kinematic*
structure: is $\alpha$ constant, and does $v = R\omega$ hold? Separating kinematics from dynamics like
this is exactly why the course teaches Week 10 before Week 11.

**TR:** Evet, ama bu bir **dinamik** sorusudur ($\sum\tau = I\alpha$) ve 11. hafta deneyidir. Bu hafta
$m$'yi bilerek sabit tutuyorsun; test edilen tek şey kinematik yapıdır. Dersin 10. haftayı 11.'den önce
işlemesinin sebebi tam olarak bu ayrımdır.

</details>

<details><summary>Why include the string thickness in R? / İp kalınlığı R'ye neden dahil?</summary>

Because the string's *centre line* is what travels at speed $v$, and that line sits half a string
thickness out from the spindle surface. With a thin spindle this is a large fractional correction, and it
is systematic: leaving it out makes $R$ too small, hence $\alpha = a/R$ too **large**, every single time.

Worth noticing: as the string unwinds, the effective radius changes slightly too, if it wraps in layers.
Wind it in a single layer to avoid that.

**TR:** Çünkü $v$ hızıyla giden şey ipin **orta çizgisidir** ve bu çizgi mil yüzeyinden yarım ip kalınlığı
dışarıdadır. Dahil etmemek $R$'yi küçük, dolayısıyla $\alpha$'yı her seferinde **büyük** gösterir. İp
katlar hâlinde sarılırsa etkin yarıçap da değişir; tek kat sar.

</details>

<details><summary>Your h vs t² graph bends slightly downward at large h. What does that mean? / Grafik büyük h'de hafifçe aşağı kıvrılıyorsa?</summary>

Downward curvature means the acceleration **falls** as the fall proceeds — so $\alpha$ is not quite
constant. The usual cause is a friction torque that grows with angular speed (air drag on the disc,
viscous bearing drag), which bites hardest late in the run when the disc is spinning fastest.

Before claiming that, check the mundane alternative: if the string wound in layers, $R_{\text{eff}}$ shrank
as it unwound, which also bends the graph. Distinguishing a real physical effect from an apparatus
artefact is the more valuable half of this question.

**TR:** Aşağı kıvrım, ivmenin düşüşle birlikte **azaldığını** gösterir; yani $\alpha$ tam olarak sabit
değil. Olağan sebep, açısal hızla büyüyen bir sürtünme torkudur. Ama önce sıradan alternatifi kontrol et:
ip katlar hâlinde sarıldıysa $R_{\text{eff}}$ değişti ve bu da grafiği kıvırır. Gerçek fiziksel etkiyi
düzenek kusurundan ayırmak, bu sorunun daha değerli yarısıdır.

</details>

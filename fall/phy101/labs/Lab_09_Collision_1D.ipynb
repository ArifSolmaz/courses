# Lab 09 — Collision in one dimension

## Bir boyutta çarpışma

**Week 09 laboratory session · 16 – 20 November 2026**
**Official schedule entry / İzlencedeki adı:** *Bir Boyutta Çarpışma*

It measures the physics taught in **Week 09**, so the lecture comes first and the experiment confirms it.

**Aim.** Test whether momentum is conserved in a one-dimensional collision, and classify the collision from the fraction of kinetic energy that survives it.

**TR.** Bir boyutlu çarpışmada momentumun korunup korunmadığını sına ve kinetik enerjinin ne kadarının hayatta kaldığına bakarak çarpışmayı sınıflandır.

**Scope note.** The experiment is scheduled by the department and is the same for every section. The
*analysis technique* in this brief (uncertainty, linearising, counting sigmas) is a laboratory
requirement and is **not** part of the common midterm or final; the *physics* being measured is.

**Before you arrive:** read [Lab 00 — measurement and uncertainty toolkit](Lab_00_Uncertainty_Toolkit.ipynb) and complete the prediction below. You will be asked for your
predicted number at the bench, before any apparatus is switched on.

**TR:** Gelmeden önce el kitabını oku ve aşağıdaki tahmini yap. Tezgâhta, düzenek çalışmadan önce
tahmin ettiğin sayı sorulacak.


## 1. Prediction — before the bench / Tahmin

### Symbolically first

Two trolleys on a track, masses $m_1$ and $m_2$. Trolley 1 arrives at $v_1$, trolley 2 is at rest.

**Perfectly inelastic** (they stick):
$$m_1 v_1 = (m_1+m_2)v' \quad \Longrightarrow \quad \boxed{\,v' = \frac{m_1}{m_1+m_2}v_1\,}$$

**Perfectly elastic** (momentum *and* kinetic energy conserved):
$$\boxed{\,v_1' = \frac{m_1-m_2}{m_1+m_2}v_1, \qquad v_2' = \frac{2m_1}{m_1+m_2}v_1\,}$$

### Limiting checks — all four

1. Equal masses, elastic: $v_1' = 0$, $v_2' = v_1$. Trolley 1 stops dead. Confirm from the formula.
2. $m_2 \gg m_1$, elastic: $v_1' \to -v_1$. It bounces back at nearly its original speed.
3. $m_1 \gg m_2$, elastic: $v_1' \to v_1$, $v_2' \to 2v_1$. The light one leaves at **twice** the speed —
   surprising, worth checking.
4. Inelastic with $m_1 = m_2$: $v' = v_1/2$. Half the speed, and therefore **half** the kinetic energy
   lost. Verify that second part; it is the prediction you will test.

### The energy fraction

For the perfectly inelastic case, show that

$$\frac{K'}{K} = \frac{m_1}{m_1+m_2}$$

so with equal masses exactly **50%** of the kinetic energy is lost. Write down where it went.

**Predict now, in the logbook:** momentum is conserved in *both* collision types, but kinetic energy only
in the elastic one. Which quantity do you expect your apparatus to conserve **better**, and why?

**TR:** Önce sembolik, sonra dört sınır kontrolü. Tam esnek olmayan çarpışmada eşit kütleler için kinetik
enerjinin tam **%50**'si kaybolur; nereye gittiğini yaz. Deftere şunu da tahmin et: momentum her iki
çarpışmada da korunur, kinetik enerji yalnızca esnek olanda. Düzeneğinin hangisini **daha iyi**
koruyacağını ve nedenini yaz.


## 2. Apparatus / Düzenek

- Air track or low-friction track with two trolleys
- Two photogates, or a motion sensor, or a ruler and a stopwatch if that is all the bench has
- Magnetic or spring bumpers for the near-elastic case; Velcro or a pin-and-cork for the inelastic case
- Balance for the masses; additional slotted masses to make $m_1 \ne m_2$

**Level the track first.** Put a trolley on it and let go: if it drifts consistently one way, the track is
tilted and gravity is quietly adding momentum to every run. This takes two minutes and it removes the
largest systematic error in the experiment.

**TR:** Önce rayı dengele. Arabayı bırak: sürekli bir yöne kayıyorsa ray eğik ve yerçekimi her denemeye
sessizce momentum ekliyor. İki dakika sürer ve deneydeki en büyük sistematik hatayı kaldırır.


## 3. Procedure / Yöntem

Run **three configurations**, five trials each.

| Config | Masses | Bumpers | What it tests |
| --- | --- | --- | --- |
| A | $m_1 = m_2$ | sticky | $v' = v_1/2$ and 50% energy loss |
| B | $m_1 = m_2$ | elastic | $v_1' \approx 0$, $v_2' \approx v_1$ |
| C | $m_1 \approx 2m_2$ | elastic | $v_1' > 0$, $v_2' > v_1$ |

For each trial:

1. Weigh both trolleys **with everything attached** — bumpers, flags, added masses.
2. Trolley 2 starts at rest. Give trolley 1 a smooth push and let it travel freely before the gate.
3. Record the velocity **into** and **out of** the collision for both trolleys.
4. Catch the trolleys before they hit the end stops and run back through the gates.

**The velocities must be measured close to the collision.** On a real track friction bleeds speed
continuously, so a velocity measured half a metre away is not the velocity at impact — and it will look
like momentum was lost when it was only lost to friction on the way.

**TR:** Hızları çarpışmaya **yakın** ölç. Gerçek rayda sürtünme hızı sürekli azaltır; yarım metre ötede
ölçülen hız çarpışma anındaki hız değildir ve momentum kaybolmuş gibi görünür — oysa yolda sürtünmeye
gitmiştir.


## 4. Data / Veri

One table per configuration.

**Configuration ......   $m_1 = \dots \pm \dots$ kg   $m_2 = \dots \pm \dots$ kg**

| Trial | $v_1$ (m/s) | $v_1'$ (m/s) | $v_2'$ (m/s) | $p_{\text{before}}$ | $p_{\text{after}}$ | $K_{\text{before}}$ | $K_{\text{after}}$ |
| --- | --- | --- | --- | --- | --- | --- | --- |
| 1 |  |  |  |  |  |  |  |
| 2 |  |  |  |  |  |  |  |
| 3 |  |  |  |  |  |  |  |
| 4 |  |  |  |  |  |  |  |
| 5 |  |  |  |  |  |  |  |
| **Mean** |  |  |  |  |  |  |  |

**Signs are physics, not bookkeeping.** Choose the positive direction once, write it at the top of the
table, and keep it. A trolley that rebounds has a **negative** velocity, and if you record its speed as
positive your momentum will appear not to be conserved — the classic wrong answer in this experiment.

$$p = m_1 v_1 + m_2 v_2, \qquad K = \tfrac12 m_1 v_1^2 + \tfrac12 m_2 v_2^2$$

**Uncertainties:** $\delta p$ from the product rule on each term, added in quadrature. For $K$, note
$K \propto v^2$, so relative uncertainties **double** — your energy comparison will always be weaker than
your momentum comparison, and that is a property of the algebra, not of your technique.

**TR:** İşaret bir muhasebe ayrıntısı değil, fiziktir. Pozitif yönü bir kez seç, tablonun başına yaz ve
koru. Geri sekem arabanın hızı **negatiftir**; onu pozitif yazarsan momentum korunmamış görünür — bu
deneydeki klasik yanlış. Enerjide bağıl belirsizlikler **iki katına** çıkar, yani enerji karşılaştırması
her zaman momentum karşılaştırmasından zayıf olur; bu tekniğin değil, cebirin özelliğidir.


## 5. Analysis / Analiz

For each configuration:

1. Compute $p_{\text{before}}$ and $p_{\text{after}}$ with uncertainties. Count sigmas between them.
2. Compute the fraction retained, $K_{\text{after}}/K_{\text{before}}$, with its uncertainty.
3. Compare the measured $v'$ (or $v_1', v_2'$) with your §1 prediction. Count sigmas.
4. Fill in the verdict table:

| Config | $p$ conserved within uncertainty? | $K'/K$ measured | $K'/K$ predicted | Classification |
| --- | --- | --- | --- | --- |
| A |  |  | $m_1/(m_1+m_2)$ | perfectly inelastic |
| B |  |  | 1 | elastic |
| C |  |  | 1 | elastic |

---

### What the two conservation laws are telling you

Momentum should come out conserved in **all three** configurations, within a few per cent. Kinetic energy
should be conserved only in B and C, and should drop to about half in A.

That asymmetry is the whole content of Week 09. Momentum conservation follows from there being no net
**external** horizontal force, so it does not care what happens inside the collision — sticking,
bouncing, crumpling, all irrelevant. Kinetic energy conservation is an extra assumption about the
*internal* mechanics, and it fails whenever the collision has somewhere to put energy: deformation,
sound, heat in the Velcro.

So: **momentum is a bookkeeping law that always balances; energy conservation of the mechanical kind is
a model that can fail.** If your data show momentum conserved and energy lost, you have not found an
error — you have found the point of the experiment.

**A caution about config B.** Real magnetic bumpers are very good but not perfect, so expect
$K'/K \approx 0.9$–$0.98$, not exactly 1. Reporting $K'/K = 1.00 \pm 0.01$ from a real air track is a
sign that something was rounded into agreement.

**TR:** Momentum üç yapılandırmada da korunmalı; kinetik enerji yalnızca B ve C'de. Bu asimetri 9.
haftanın bütün içeriğidir: momentum, dış kuvvet olmadığı için çarpışmanın içinde ne olduğuna aldırmaz;
kinetik enerjinin korunumu ise iç mekaniğe dair **ek** bir varsayımdır ve enerjinin gidecek bir yeri
olduğunda bozulur. Momentum korunup enerji kaybolduysa hata bulmadın, deneyin amacını buldun.


## 6. Discussion / Tartışma

<details><summary>Momentum came out 4% low in every run. Where did it go? / Momentum her denemede %4 düşük çıktıysa?</summary>

It did not "go" anywhere — the system you measured was not closed. Candidates, in order of likelihood:

1. **Friction between the gates and the collision point.** Every centimetre of travel removes a little
   momentum to the track and the air. This biases the *after* velocities low, consistently.
2. **A tilted track.** If it slopes against the motion, gravity removes momentum on every run. Level it
   and repeat one configuration to check.
3. **Velocities measured too far from the impact** — same cause as 1, made worse.

Note the diagnostic: all three predict a **consistent one-directional** shortfall, which is exactly what
"4% low in every run" looks like. A random error would scatter both ways.

**TR:** Hiçbir yere gitmedi; ölçtüğün sistem kapalı değildi. Sırayla: kapılarla çarpışma arasındaki
sürtünme, eğik ray, çarpışmadan uzakta ölçülen hızlar. Üçü de **tek yönlü tutarlı** bir eksiklik öngörür;
"her denemede %4 düşük" tam olarak buna benzer. Rastgele hata iki yöne saçılırdı.

</details>

<details><summary>Equal masses, elastic: trolley 1 stops completely. Why is that not magic? / Eşit kütlede esnek çarpışmada 1. araba neden tamamen duruyor?</summary>

Because it is the only outcome that satisfies **both** conservation laws at once. Momentum alone allows
many splits of the final velocities; energy conservation picks out one of them. With $m_1 = m_2$ that
solution is the complete transfer: $v_1' = 0$, $v_2' = v_1$.

Worth appreciating: you have two equations and two unknowns, so the outcome is *forced*. The trolley is
not choosing to stop — nothing else is consistent.

**TR:** Çünkü iki korunum yasasını **birlikte** sağlayan tek sonuç budur. Momentum tek başına birçok
bölüşüme izin verir; enerji korunumu bunlardan birini seçer. İki denklem, iki bilinmeyen: sonuç zorunludur.

</details>

<details><summary>Your K'/K in config A is 0.58, not 0.50. Is that a failure? / A'da 0.58 çıktıysa başarısızlık mı?</summary>

Check it against your uncertainty first. Because $K \propto v^2$, a 4% velocity uncertainty gives about
8% on each energy and roughly 11% on the ratio — so $0.58 \pm 0.06$ is about $1.3\sigma$ from 0.50 and
**agrees**. Count sigmas before you apologise.

If it is genuinely far out, the usual cause is that the trolleys did not stick cleanly and separated
slightly, so some kinetic energy survived that the perfectly-inelastic model assumes is gone.

**TR:** Önce belirsizliğe bak. $K \propto v^2$ olduğundan hızdaki %4, oranda yaklaşık %11 yapar:
$0.58 \pm 0.06$, 0.50'den yaklaşık $1.3\sigma$ uzaktır ve **uyuşur**. Özür dilemeden önce sigma say.

</details>

<details><summary>Why must the masses include the bumpers and flags? / Kütlelere tamponlar ve bayraklar neden dahil?</summary>

Because momentum is mass times velocity for whatever is actually moving. A 15 g flag on a 200 g trolley is
7.5% of the mass; leave it out and you have built a 7.5% error into $p$ before taking a single reading — and
it will look exactly like non-conservation.

This is why step 1 of the procedure says to weigh the trolleys fully assembled. **Weigh what moves.**

**TR:** Çünkü momentum, gerçekten hareket eden şeyin kütlesi çarpı hızıdır. 200 g'lık arabada 15 g'lık
bayrak kütlenin %7.5'idir; dahil etmezsen tek okuma yapmadan $p$'ye %7.5 hata koymuş olursun ve bu tam
olarak korunumsuzluk gibi görünür. Hareket edeni tart.

</details>

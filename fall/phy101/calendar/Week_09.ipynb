## Before the equations: signs and two stages / İşaretler ve iki aşama

The required work is a diagram, a few equations, and a calculator. Python is optional verification. Draw **before** and **after** pictures; choose right/east as positive before substituting any velocities.

Momentum has a sign: $p=mv$. Kinetic energy has no directional sign: $K=\tfrac12mv^2$. A $2\,\mathrm{kg}$ object at $-3\,\mathrm{m/s}$ has $p=-6\,\mathrm{kg\,m/s}$ and $K=+9\,\mathrm J$.

**Algebra for objects that stick:**
$$m_1v_1+m_2v_2=(m_1+m_2)v_f,$$
$$v_f=\frac{m_1v_1+m_2v_2}{m_1+m_2}.$$
Divide the **whole numerator** by the **total mass**. For $2\,\mathrm{kg}$ at $+3\,\mathrm{m/s}$ and $1\,\mathrm{kg}$ at $-3\,\mathrm{m/s}$, $v_f=[6-3]/3=+1\,\mathrm{m/s}$. The plus sign gives the direction.

**Do not mix stages.** In a bullet–block collision, momentum is conserved during the short impact if external impulse is negligible. Kinetic energy decreases. During the later upward swing, mechanical energy is conserved if drag is negligible; gravity then changes momentum. Use a new equation for each stage.

In 2D, write two momentum equations, one for $x$ and one for $y$. Only after finding $v_x,v_y$ calculate $v=\sqrt{v_x^2+v_y^2}$ and the angle. Units: $1\,\mathrm{N\,s}=1\,\mathrm{kg\,m/s}$; these are impulse/momentum units, not joules.

**Pause and explain:** Could a collision conserve momentum but lose kinetic energy? Yes: two objects can stick while energy becomes deformation, sound, and heat.

### Analogy: The Freight Train vs. The Sports Car

Think of momentum as "how hard it is to stop something." A slow freight train (huge mass, small velocity) can have the same momentum as a fast sports car (small mass, huge velocity). Equal momentum means equal stopping impulse. Equal stopping time would require equal average force; equal stopping distance instead depends on kinetic energy, which can differ.

### Newton's Second Law in Momentum Form

$$\vec{F}_{\text{net}} = \frac{d\vec{p}}{dt}$$

When mass is constant: $\vec{F}_{\text{net}} = m\vec{a}$ (the familiar form).

## 2. Impulse-Momentum Theorem

**Impulse** is the change in momentum produced by a force acting over a time interval:

$$\vec{J} = \int_{t_1}^{t_2} \vec{F}\, dt = \Delta\vec{p} = m\vec{v}_f - m\vec{v}_i$$

For the average net force over the interval, $\vec J=\vec F_{\mathrm{avg}}\Delta t$. For a constant net force this is simply $\vec J=\vec F\Delta t$. Time is a scalar; this multiplication is not a dot product.

### Engineering Insight

This is why car airbags and crumple zones work! By increasing $\Delta t$, the **average force** on the occupant decreases for the same momentum change. The signed area under the $F$–$t$ curve (impulse) stays the same. A longer duration reduces the average force; the peak force also depends on the pulse shape. For a triangular pulse, $J=\frac12F_{\mathrm{peak}}\Delta t$, so $F_{\mathrm{peak}}=2J/\Delta t$.

## 3. Conservation of Momentum

For an **isolated system** (no external forces), total momentum is conserved:

$$\vec{p}_{\text{total, before}} = \vec{p}_{\text{total, after}}$$

For two objects in 1D:

$$m_1 v_{1i} + m_2 v_{2i} = m_1 v_{1f} + m_2 v_{2f}$$

### Elastic Collision Formulas (1D)

$$v_{1f} = \frac{m_1 - m_2}{m_1 + m_2} v_{1i} + \frac{2m_2}{m_1 + m_2} v_{2i}$$

$$v_{2f} = \frac{2m_1}{m_1 + m_2} v_{1i} + \frac{m_2 - m_1}{m_1 + m_2} v_{2i}$$

### Perfectly Inelastic Collision Formula (1D)

$$v_f = \frac{m_1 v_{1i} + m_2 v_{2i}}{m_1 + m_2}$$

## Worked example: the two carts stick together

A 2.0 kg cart moves right at 3.0 m/s and hits a stationary 1.0 kg cart. They stick. Choose right positive and neglect external impulse during the short collision.
$$m_1v_{1i}+m_2v_{2i}=(m_1+m_2)v_f,$$
$$2(3)+1(0)=3v_f\Rightarrow v_f=2.0\,\mathrm{m/s}\text{ right}.$$
Initial kinetic energy is $K_i=\tfrac12(2)(3^2)=9$ J; final energy is $K_f=\tfrac12(3)(2^2)=6$ J. Momentum stays 6 kg m/s, while 3 J transfers to internal energy. Momentum conservation alone does not make a collision elastic.

**TR:** Yapışma anında önce momentumu kullan. Mekanik enerjinin tamamının korunacağını varsayma.

## 4. Worked Examples

### Example 1: Head-on Elastic Collision

A $2.0\,\mathrm{kg}$ ball moving at $5.0\,\mathrm{m/s}$ collides head-on elastically with a $3.0\,\mathrm{kg}$ ball initially at rest. Find the final velocities.

**Paper solution.** Choose the initial direction as positive. Momentum conservation gives $2(5)+3(0)=2v_1+3v_2$, or $10=2v_1+3v_2$. For a one-dimensional elastic collision, separation speed equals approach speed: $v_2-v_1=5$, so $v_2=v_1+5$.

Substitute into momentum: $10=2v_1+3(v_1+5)=5v_1+15$. Subtract 15 from both sides: $-5=5v_1$. Divide by 5: **$v_1=-1\,\mathrm{m/s}$**, then **$v_2=4\,\mathrm{m/s}$**. The first ball rebounds.

Check momentum: $2(-1)+3(4)=10\,\mathrm{kg\,m/s}$. Check energy: initial $\tfrac12(2)(5^2)=25\,\mathrm J$; final $\tfrac12(2)(1^2)+\tfrac12(3)(4^2)=1+24=25\,\mathrm J$.

**Türkçe:** İki bilinmeyen hız için iki bağımsız denklem kullandık. Eksi işareti hata değildir; birinci top geri dönüyor.

### Gravity transfers momentum during a fall

**Problem.** A stone of mass $2.50\,\mathrm{kg}$ is released from rest $3.75\,\mathrm{m}$ above the ground. Ignore air resistance. Find the impulse delivered by gravity before the stone reaches the ground. Use $g=9.81\,\mathrm{m/s^2}$. The collision with the ground is outside this time interval.

**Reasoning and algebra.** Draw a downward arrow for the weight, and choose downward as the positive direction. Impulse measures the change in momentum, not the distance travelled:

$$
J_y=\Delta p_y=m(v_f-v_i).
$$

We know $v_i=0$, but the final speed is missing. Constant-acceleration kinematics supplies it without requiring the fall time:

$$
\begin{aligned}
v_f^2&=v_i^2+2gh,\\
v_f&=\sqrt{2gh}
=\sqrt{2(9.81\,\mathrm{m/s^2})(3.75\,\mathrm{m})}
=8.5776\,\mathrm{m/s},\\
J_y&=(2.50\,\mathrm{kg})(8.5776\,\mathrm{m/s})
=21.444\,\mathrm{kg\,m/s}.
\end{aligned}
$$

The positive square root is appropriate because downward was chosen as positive and the stone is falling downward. Thus the impulse has magnitude **$21.4\,\mathrm{N\,s}$, directed downward**. The units agree because $1\,\mathrm{N\,s}=1\,\mathrm{kg\,m/s}$.

**Check.** The fall lasts $t=v_f/g\approx0.874\,\mathrm{s}$. Multiplying the constant weight, $mg=24.525\,\mathrm{N}$, by this time gives the same impulse. The force and impulse point in the same direction here because gravity does not reverse during the fall.

**Türkçe açıklama.** İtme, momentum değişimidir; kuvvet ile alınan yolu çarpmak ise işi verir. Önce düşme hızını bulduk, sonra kütleyle çarptık. Aşağı yönü pozitif seçtiğimiz için sonuç pozitiftir. Taşın yere çarpması sırasındaki büyük temas kuvvetini bu hesaba katmıyoruz.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 8, §8.2 Problems, question 2 (`08_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### Find the second velocity before calculating energy loss

**Problem.** A $2.00\,\mathrm{kg}$ object travels east at $20.0\,\mathrm{m/s}$; a $3.00\,\mathrm{kg}$ object travels west at $10.0\,\mathrm{m/s}$. They collide. Afterwards, the first object travels west at $5.00\,\mathrm{m/s}$. Find the kinetic energy lost, assuming negligible external impulse during contact.

**Reasoning and algebra.** Choose east as positive. The signed velocities are $v_{1i}=+20.0$, $v_{2i}=-10.0$, and $v_{1f}=-5.00$, all in $\mathrm{m/s}$. First conserve the total momentum of both objects:

$$
\begin{aligned}
m_1v_{1i}+m_2v_{2i}&=m_1v_{1f}+m_2v_{2f},\\
v_{2f}&=\frac{m_1v_{1i}+m_2v_{2i}-m_1v_{1f}}{m_2},\\
&=\frac{(2)(20)+(3)(-10)-(2)(-5)}{3}
=\frac{20}{3}\,\mathrm{m/s}.
\end{aligned}
$$

The second object therefore moves east after contact. Now calculate energy using squared speeds; kinetic energy is never made negative by a westward velocity:

$$
\begin{aligned}
K_i&=\tfrac12(2)(20)^2+\tfrac12(3)(10)^2=550\,\mathrm{J},\\
K_f&=\tfrac12(2)(5)^2+\tfrac12(3)(20/3)^2
=91.667\,\mathrm{J},\\
E_{\mathrm{lost}}&=K_i-K_f=\boxed{458.3\,\mathrm{J}}.
\end{aligned}
$$

**Interpretation and check.** Initial and final momentum are both $+10.0\,\mathrm{kg\,m/s}$. Kinetic energy decreases while momentum stays constant: deformation, heating and sound can receive energy during the impact. Keep $20/3$ in the intermediate calculation so rounding the velocity does not unnecessarily change the energy result.

**Türkçe açıklama.** Batıya giden hızları momentum denklemine eksi işaretle yazdık. Enerji hesabında hızın karesi bulunduğu için aynı eksi işaret enerjiye taşınmaz. Önce bilinmeyen son hızı çözmeden toplam son enerjiyi hesaplayamayız. Momentumun korunması, kinetik enerjinin de korunacağı anlamına gelmez.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 8, §8.2 Problems, question 14 (`08_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### A collision followed by an upward swing

**Problem — adapted variant.** A $5.00\,\mathrm{kg}$ ball hangs at rest from a long, light wire. A $1.50\,\mathrm{kg}$ stone strikes it horizontally at $12.0\,\mathrm{m/s}$, then rebounds at $6.00\,\mathrm{m/s}$. Find the ball's maximum rise above its starting position. Use $g=9.81\,\mathrm{m/s^2}$. The rebound speed is adapted from the bank so that an ordinary passive collision is energetically possible.

**Reasoning and algebra.** Separate the event into two stages. During the short collision, take the stone's incoming direction as positive and neglect external horizontal impulse. Let $V$ be the ball's speed immediately after impact:

$$
\begin{aligned}
m_sv_{si}+M(0)&=m_sv_{sf}+MV,\\
MV&=m_s(v_{si}-v_{sf}),\\
V&=\frac{(1.50\,\mathrm{kg})[12.0-(-6.00)]\,\mathrm{m/s}}
{5.00\,\mathrm{kg}}
=5.40\,\mathrm{m/s}.
\end{aligned}
$$

The stone's final velocity is negative because it rebounds. The ball gains $27.0\,\mathrm{kg\,m/s}$ of horizontal momentum, exactly what the stone loses.

During the subsequent swing, conserve the ball–Earth mechanical energy. The ideal wire's tension is perpendicular to the ball's motion, so it does no work. At the highest point the ball momentarily stops:

$$
\begin{aligned}
\tfrac12 MV^2&=Mgh,\\
h&=\frac{V^2}{2g}
=\frac{(5.40\,\mathrm{m/s})^2}{2(9.81\,\mathrm{m/s^2})}
=\boxed{1.49\,\mathrm{m}}.
\end{aligned}
$$

**Energy check.** Check the collision separately, including both objects:

$$
\begin{aligned}
K_i&=\tfrac12(1.50)(12.0)^2=108\,\mathrm{J},\\
K_f&=\tfrac12(1.50)(6.00)^2+\tfrac12(5.00)(5.40)^2
=99.9\,\mathrm{J},\\
K_i-K_f&=8.10\,\mathrm{J}.
\end{aligned}
$$

This decrease is consistent with deformation, heating or sound in a passive collision. The ball's mass cancels from the swing equation, but it mattered when finding $V$.

**Türkçe açıklama.** Çarpışmada yatay momentumu, sonraki yükselmede mekanik enerjiyi kullanıyoruz. Geri dönen taşın hızı $-6.00$ olduğundan parantez $12.0+6.00$ olur. Çarpışma sırasında toplam kinetik enerji $8.10\,\mathrm{J}$ azalır; bu enerji kaybolmaz, başka biçimlere dönüşür.

**Source:** Adapted from Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 8, §8.2 Problems, question 20 (`08_testbank.doc`). Rebound speed changed from $8.50$ to $6.00\,\mathrm{m/s}$ for a passive collision; all other givens retained. Wording and worked explanation expanded.

### Two perpendicular momenta produce one final velocity

**Problem.** A $900\,\mathrm{kg}$ car travels east at $15.0\,\mathrm{m/s}$. A $750\,\mathrm{kg}$ car travels north at $20.0\,\mathrm{m/s}$. They collide and remain together. Ignoring external impulse during the collision, find the speed and direction of the joined cars immediately afterwards.

**Reasoning and algebra.** Draw east as the positive $x$-axis and north as the positive $y$-axis. The combined mass is $M=900+750=1650\,\mathrm{kg}$. Momentum conservation applies separately to each direction:

$$
\begin{aligned}
Mv_x&=(900\,\mathrm{kg})(15.0\,\mathrm{m/s}),\\
v_x&=\frac{13500}{1650}\,\mathrm{m/s}=8.1818\,\mathrm{m/s},\\
Mv_y&=(750\,\mathrm{kg})(20.0\,\mathrm{m/s}),\\
v_y&=\frac{15000}{1650}\,\mathrm{m/s}=9.0909\,\mathrm{m/s}.
\end{aligned}
$$

These are perpendicular components of one velocity, so their magnitudes cannot simply be added. Use the right triangle made by $v_x$ and $v_y$:

$$
\begin{aligned}
v&=\sqrt{v_x^2+v_y^2}=\boxed{12.2\,\mathrm{m/s}},\\
\theta&=\tan^{-1}\!\left(\frac{v_y}{v_x}\right)
=\tan^{-1}\!\left(\frac{15000}{13500}\right)
=\boxed{48.0^\circ\text{ north of east}}.
\end{aligned}
$$

**Interpretation and check.** Both components are positive, so the answer belongs in the northeast quadrant. The northward momentum is slightly larger than the eastward momentum; accordingly, the direction is slightly above $45^\circ$. Multiplying each final velocity component by $1650\,\mathrm{kg}$ recovers its initial momentum component. Sticking together permits kinetic-energy loss while preserving total momentum during contact.

**Türkçe açıklama.** Doğu ve kuzey hızlarını önce kendi eksenlerinde işliyoruz. Birleşen araçlar için her iki denklemde de toplam kütle kullanılır. $48.0^\circ$, doğudan kuzeye doğru ölçülür. Kuzeyden başlasaydık aynı yönü $42.0^\circ$ doğu olarak ifade ederdik; referans ekseni mutlaka belirtmeliyiz.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 8, §8.2 Problems, question 23 (`08_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### Why the heavier skater moves more slowly

**Problem.** Two skaters start at rest and push away from each other. One has mass $45.0\,\mathrm{kg}$ and leaves at $0.375\,\mathrm{m/s}$. The other has mass $60.0\,\mathrm{kg}$. Find the second skater's velocity, neglecting external horizontal impulse while they push.

**Reasoning and algebra.** Treat both skaters as one system. Their mutual push is internal to that system. The initial total momentum is zero because both skaters are initially stationary. Choose the lighter skater's final direction as positive:

$$
\begin{aligned}
p_i&=p_f,\\
0&=m_1v_1+m_2v_2,\\
m_2v_2&=-m_1v_1,\\
v_2&=-\frac{m_1}{m_2}v_1
=-\frac{45.0\,\mathrm{kg}}{60.0\,\mathrm{kg}}
(0.375\,\mathrm{m/s})\\
&=-0.28125\,\mathrm{m/s}.
\end{aligned}
$$

The negative sign answers the direction question: the second skater travels opposite to the first. Its **speed is $0.281\,\mathrm{m/s}$**, rounded to three significant figures.

**Interpretation and check.** The two final momenta are $+16.875$ and $-16.875\,\mathrm{kg\,m/s}$, whose sum is zero. Equal and opposite impulses do not imply equal speeds: momentum includes mass. Here the heavier skater's speed is $45/60=0.75$ times the lighter skater's speed, which agrees with the calculated result.

The skaters can both gain kinetic energy despite initially having none. Muscular energy supplies that increase. This does not conflict with momentum conservation, because energy transfer and the balance of external impulse answer different questions.

**Türkçe açıklama.** Eşit ve zıt olan büyüklük hızlar değil, iki kişinin momentum değişimleridir. Büyük kütle aynı momentum büyüklüğü için daha küçük hız kazanır. Eksi işaret olumsuz bir sürat anlamına gelmez; seçtiğimiz pozitif yönün tersini gösterir. Sonucun yönünü ve sürat büyüklüğünü ayrı söylemek bu ayrımı görünür kılar.

**Source:** Young/Freedman, *University Physics with Modern Physics*, 14th Global Edition, Chapter 8, §8.2 Problems, question 25 (`08_testbank.doc`). Original givens retained; wording and worked explanation expanded.

### L1 -- Module 07 P1: Momentum Calculation

A $0.060\,\mathrm{kg}$ tennis ball is served at $55.0\,\mathrm{m/s}$. (a) What is its momentum? (b) A $0.43\,\mathrm{kg}$ soccer ball is kicked at $25.0\,\mathrm{m/s}$. What is its momentum? (c) Which ball is harder to stop?

**Türkçe problem açıklaması:** $0.060\,\mathrm{kg}$ tenis topu $55.0\,\mathrm{m/s}$, $0.43\,\mathrm{kg}$ futbol topu $25.0\,\mathrm{m/s}\,\mathrm s$üratle gidiyor. (a,b) Momentumlarını bulun; (c) hangisini durdurmak daha zordur? “Zor” ifadesini aynı durma süresinde gereken ortalama kuvvet olarak açıklayın. Aynı durma mesafesi verilseydi enerji karşılaştırması gerekirdi.

<details>
<summary>Answer and assumptions</summary>

Use momentum magnitudes $p=mv$:
$$\begin{aligned}\text{(a)}\quad p_\mathrm{tennis}&=(0.060)(55.0)=3.30\,\mathrm{kg\,m/s},\\
\text{(b)}\quad p_\mathrm{soccer}&=(0.43)(25.0)=10.75\,\mathrm{kg\,m/s}.
\end{aligned}$$
(c) To stop, the impulse magnitude equals initial momentum. Therefore
$$\frac{|J_\mathrm{soccer}|}{|J_\mathrm{tennis}|}=\frac{10.75}{3.30}=3.26.$$
For the same stopping time, $|F_\mathrm{avg}|=|J|/\Delta t$, so the soccer ball requires $3.26$ times the average force. Equal stopping distance instead compares kinetic energies; specify the meaning of “harder to stop.”

```python
#@title Run once — prepare the physics demonstrations
# Setup: Run this cell first to import all required libraries
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
import ipywidgets as widgets
from ipywidgets import interact, interactive, FloatSlider, IntSlider, HBox, VBox, Output
import warnings
warnings.filterwarnings('ignore')

# Enable inline plots
%matplotlib inline
print("All libraries loaded successfully!")
```

```python
#@title Run once — prepare compact demonstration controls
"""Embedded in the lecture notebooks by refresh_notebook_interface.py.

Only standard ipywidgets and IPython are used; notebooks remain self-contained.
"""
import functools
import traceback
import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import display, HTML
from IPython.utils.capture import capture_output
from ipywidgets import interactive as _PhysicsInteractive
from ipywidgets.widgets.interaction import show_inline_matplotlib_plots

# Set this explicitly: otherwise a local kernel can choose a desktop backend
# and block at plt.show(), leaving the notebook apparently frozen.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

try:
    from google.colab import output as _colab_output
except ImportError:
    pass
else:
    _colab_output.enable_custom_widget_manager()

_physics_panels = []
_physics_callback_errors = []


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; }
.phy101-controls { padding: 0 0 4px; box-sizing: border-box; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-animation-panel { max-width: 100%; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1;
  background: white; color: #172433; padding: 4px; }
@media (max-width: 650px) {
  .phy101-panel { flex-direction: column !important; }
  .phy101-controls, .phy101-plot-output { width: 100% !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Fit full labels and postpone costly redraws until a slider is released."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "continuous_update"):
            control.continuous_update = False
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "310px"
        control.layout.flex = "0 1 310px"
        control.layout.max_width = "100%"
        control.layout.min_width = "0"
        control.layout.margin = "2px 6px 2px 0"
        if isinstance(control, widgets.Button):
            control.layout.width = "auto"
            control.layout.flex = "0 0 auto"
    # Include the style in each output: Colab may isolate output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.Box([style] + flat, layout=widgets.Layout(
        display="flex", flex_flow="row wrap", align_items="center",
        width="100%", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        width="100%", min_width="0", max_width="100%",
        height="auto", overflow="visible", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Keep a compact control toolbar directly above the complete result."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="column",
                              align_items="stretch", width="100%"))
    panel.add_class("phy101-panel")
    return panel


class _PhysicsStableInteractive(_PhysicsInteractive):
    """Replace one persistent output model only after a redraw has completed.

    Sending clear_output followed by an inline figure depends on callback message
    capture in the frontend. Explicit MIME records keep repeated slider changes
    visible in Jupyter/Colab, without clearing the surrounding controls.
    """
    def update(self, *args):
        self.kwargs = {}
        if self.manual:
            self.manual_button.disabled = True
        try:
            for control in self.kwargs_widgets:
                self.kwargs[control._kwarg] = control.get_interact_value()
            with capture_output(stdout=True, stderr=True, display=True) as captured:
                self.result = self.f(**self.kwargs)
                show_inline_matplotlib_plots()
                if self.auto_display and self.result is not None:
                    display(self.result)
            records = []
            if captured.stdout:
                records.append({"output_type": "stream", "name": "stdout", "text": captured.stdout})
            if captured.stderr:
                records.append({"output_type": "stream", "name": "stderr", "text": captured.stderr})
            records.extend({"output_type": "display_data", "data": output.data,
                            "metadata": output.metadata or {}} for output in captured.outputs)
            # Keep the old result visible during calculation; publish the complete
            # replacement as one widget-state update, including its PNG/HTML data.
            self.out.outputs = tuple(records)
        except Exception:
            _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), traceback.format_exc()))
            self.out.outputs = ({"output_type": "stream", "name": "stderr",
                                 "text": traceback.format_exc()},)
        finally:
            if self.manual:
                self.manual_button.disabled = False


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


def physics_interactive(function, **controls):
    """An interactive widget with the same arguments as ipywidgets.interactive."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        try:
            return function(*args, **kwargs)
        except Exception as error:
            _physics_callback_errors.append((function.__name__, repr(error)))
            raise

    panel = _PhysicsStableInteractive(checked, {"auto_display": True}, **controls)
    inputs = [child for child in panel.children if child is not panel.out]
    panel.children = (_physics_controls(inputs), _physics_output(panel.out))
    panel.layout = widgets.Layout(display="flex", flex_flow="column",
                                  align_items="stretch", width="100%")
    panel.add_class("phy101-panel")
    _physics_panels.append(panel)
    return panel


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel
    display(panel)
    return function
```

### Visual check 1: Interactive Demo 2: Animated 1D Collision Simulator

```python
#@title Run the demonstration — predict, adjust, observe
def collision_1d_animation(m1=2.0,m2=1.0,v1i=3.0,v2i=-1.0,collision_type='Elastic'):
    """Evaluate motion at the exact contact time, independently for every animation frame."""
    if collision_type=='Elastic':
        v1f=((m1-m2)*v1i+2*m2*v2i)/(m1+m2)
        v2f=(2*m1*v1i+(m2-m1)*v2i)/(m1+m2)
    else:v1f=v2f=(m1*v1i+m2*v2i)/(m1+m2)
    size1,size2=.3+.15*m1,.3+.15*m2
    x1_start,x2_start=-4.0,4.0
    gap=x2_start-x1_start-(size1+size2)/2
    t_collision=gap/(v1i-v2i) if v1i>v2i else float('inf')
    total_time=min(3*t_collision,8.0);n_frames=150
    times=np.linspace(0,total_time,n_frames)
    def state(t):
        if t<t_collision:
            return x1_start+v1i*t,x2_start+v2i*t,v1i,v2i,False
        x1_contact=x1_start+v1i*t_collision
        x2_contact=x2_start+v2i*t_collision
        return x1_contact+v1f*(t-t_collision),x2_contact+v2f*(t-t_collision),v1f,v2f,True
    sampled=[state(t) for t in times]
    positions=[x for row in sampled for x in row[:2]]
    padding=2+max(abs(v1i),abs(v2i),abs(v1f),abs(v2f))*.3
    fig,(ax,info)=plt.subplots(1,2,figsize=(14,4.5),gridspec_kw={'width_ratios':[1.7,1]},layout='constrained')
    ax.set(xlim=(min(positions)-padding,max(positions)+padding),ylim=(-.5,1.8),xlabel='Position (m)',title=collision_type+' collision')
    ax.set_yticks([]);ax.axhline(-.05,color='gray',lw=2)
    colors=['#2196F3','#FF5722']
    boxes=[patches.Rectangle((x-size/2,0),size,size,facecolor=color,edgecolor='black',lw=2)
           for x,size,color in [(x1_start,size1,colors[0]),(x2_start,size2,colors[1])]]
    for box in boxes:ax.add_patch(box)
    arrows=[ax.annotate('',xy=(x,sz/2),xytext=(x,sz/2),arrowprops=dict(arrowstyle='->',color=color,lw=2))
            for x,sz,color in [(x1_start,size1,colors[0]),(x2_start,size2,colors[1])]]
    info.axis('off');info.set_title('Signed velocities and conservation')
    mass1_text=info.text(.02,.88,'',transform=info.transAxes,color=colors[0],fontsize=11)
    mass2_text=info.text(.02,.77,'',transform=info.transAxes,color=colors[1],fontsize=11)
    detail=info.text(.02,.62,'',transform=info.transAxes,va='top',fontsize=10.5,linespacing=1.4)
    p_before=m1*v1i+m2*v2i;K_before=.5*m1*v1i**2+.5*m2*v2i**2
    artists=tuple(boxes+arrows+[mass1_text,mass2_text,detail])
    def update(frame):
        t=times[frame];x1,x2,v1,v2,hit=state(t)
        for box,x,size in zip(boxes,[x1,x2],[size1,size2]):box.set_x(x-size/2)
        for arrow,x,v,size in zip(arrows,[x1,x2],[v1,v2],[size1,size2]):
            arrow.xy=(x+.3*v,size/2);arrow.set_position((x,size/2))
        mass1_text.set_text(f'Blue: m1 = {m1:.1f} kg; v1 = {v1:+.2f} m/s')
        mass2_text.set_text(f'Orange: m2 = {m2:.1f} kg; v2 = {v2:+.2f} m/s')
        K=.5*m1*v1**2+.5*m2*v2**2;p=m1*v1+m2*v2
        stage='After contact' if hit else ('Before contact' if t_collision<=total_time else 'No collision in this time window')
        detail.set_text(f't = {t:.2f} s — {stage}\n'
            f'p initial = {p_before:+.3f} kg m/s\np now = {p:+.3f} kg m/s\n'
            f'K initial = {K_before:.3f} J\nK now = {K:.3f} J\n'
            f'K transferred to internal energy = {K_before-K:.3f} J\n\n'
            'External horizontal impulse is neglected.\nMomentum is conserved through the collision.')
        return artists
    anim=FuncAnimation(fig,update,init_func=lambda:update(0),frames=physics_frames(n_frames),
                       interval=physics_interval(n_frames,40),blit=False)
    plt.close(fig)
    return physics_animation_html(anim)

display(HTML("<h4>Adjust parameters and run the cell to see the animation:</h4>"))

physics_interact(collision_1d_animation,
         m1=FloatSlider(value=2.0, min=0.5, max=5.0, step=0.5, description='Mass 1 (kg)'),
         m2=FloatSlider(value=1.0, min=0.5, max=5.0, step=0.5, description='Mass 2 (kg)'),
         v1i=FloatSlider(value=3.0, min=-5.0, max=5.0, step=0.5, description='v1 initial'),
         v2i=FloatSlider(value=-1.0, min=-5.0, max=5.0, step=0.5, description='v2 initial'),
         collision_type=widgets.Dropdown(options=['Elastic', 'Perfectly Inelastic'], value='Elastic', description='Type'));
```

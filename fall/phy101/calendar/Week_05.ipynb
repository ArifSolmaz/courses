## Read and work through the examples / Konu ve çözümlü örnekler

### 2.1 Work by a constant force

When a constant force $\vec{F}$ acts on an object that undergoes displacement $\vec{d}$:

$$W = \vec{F} \cdot \vec{d} = Fd\cos\theta$$

where $\theta$ is the angle between the force and displacement vectors.

### Analogy -- Pushing a shopping cart

If you push a cart horizontally with $50\,\mathrm N$ and it moves $10\,\mathrm m$, you do $W = 50 \times 10 = 500\,\mathrm J$. If you push at $30^\circ$ below horizontal, only the horizontal component counts: $W = 50 \times 10 \times \cos 30^\circ \approx 433\,\mathrm J$.

### 2.2 Work by a variable force

When the force varies with position, work is the signed area under the force component along the path, plotted against position. Area below the axis contributes negative work:

$$W = \int_{x_1}^{x_2} F(x)\,dx$$

You can start with geometry: a constant $6\,\mathrm N$ force over $2\,\mathrm m$ gives rectangle area $(6)(2)=12\,\mathrm J$. If the force increases linearly from zero to $6\,\mathrm N$ over the same distance, triangle area gives $W=\frac12(2)(6)=6\,\mathrm J$. **TR:** İntegral, değişen kuvvetin küçük aralıklardaki işlerini toplar; dikdörtgen ve üçgen alanı bunun basit örnekleridir.

## Kinetic energy: solve for speed one operation at a time

Kinetic energy describes motion: $K=\tfrac12mv^2$, measured in joules ($1\,\mathrm J=1\,\mathrm{kg\,m^2/s^2}$). Net work changes it:
$$W_{\rm net}=K_f-K_i=\tfrac12mv_f^2-\tfrac12mv_i^2.$$
The word **net** matters: add the work of all forces first. Positive work increases kinetic energy, negative work decreases it, and a force perpendicular to displacement does zero work.

A 4.0 kg cart starts at 2.0 m/s and receives 24 J of net work. Its initial energy is $K_i=\tfrac12(4)(2^2)=8$ J. Thus $K_f=8+24=32$ J. Solve $32=\tfrac12(4)v_f^2$:
$$64=4v_f^2\Rightarrow16=v_f^2\Rightarrow v_f=4.0\,\mathrm{m/s}.$$
We multiplied both sides by 2, divided by 4 kg, and took the positive square root for speed. Energy does not give the direction; use the diagram.

**TR:** Bu hafta iş ve kinetik enerji kullanıyoruz. Potansiyel enerji ve korunma hesabını vizeden sonraki 8. haftada kuracağız. Sonucun sürat olması için karekök adımını unutma.

## Worked example: opposite forces do opposite work

A 3.0 kg box already slides to the right at 2.0 m/s. A 12 N horizontal pull and a 3 N friction force act over 4.0 m. Find its final speed.

The pull does $W_{\rm pull}=12(4)\cos0^\circ=48$ J. Friction does $W_f=3(4)\cos180^\circ=-12$ J. Weight and normal are perpendicular to motion and do zero work.
$$W_{\rm net}=48-12=36\,\mathrm J,\qquad K_i=\tfrac12(3)(2^2)=6\,\mathrm J.$$
$$K_f=6+36=42\,\mathrm J=\tfrac12(3)v_f^2
\Rightarrow v_f^2=\frac{2(42)}3=28\Rightarrow v_f=5.292\,\mathrm{m/s}.$$
**Check by Newton:** $a=(12-3)/3=3$ m/s² and $v_f^2=2^2+2(3)(4)=28$ m²/s². Both routes predict the same speed.

**TR:** Sürtünmenin yaptığı iş negatiftir. Ancak toplam iş hâlâ pozitif olduğu için cisim hızlanır.

## 7. Worked Examples

### Example 1 -- Work by gravity on a sliding block

A $5\,\mathrm{kg}$ block slides $3\,\mathrm m$ down a frictionless ramp inclined at $30^\circ$. Find the work done by gravity and the final speed (starting from rest).

**Paper solution / Kâğıt üzerinde çözüm.** Draw the ramp; the distance along it is $3\,\mathrm m$ but the vertical drop is $h=3\sin30^\circ=1.5\,\mathrm m$. Gravity does positive work because the object moves downward. The normal force is perpendicular to motion and does no work.

$$W_g=mgh=(5)(9.81)(1.5)=73.575\ \mathrm J.$$
Starting from rest, $W_{\rm net}=\tfrac12mv^2$ gives $73.575=2.5v^2$. Divide by 2.5 to get $v^2=29.43\,\mathrm{m^2/s^2}$, then take the square root: **$v=5.42\,\mathrm{m/s}$ down the ramp**. The mass cancels if we first write $mgh=\tfrac12mv^2$, giving $v=\sqrt{2gh}$. The same vertical drop gives the same speed on any frictionless path under this particle model.

**Türkçe:** Eğik düzlem uzunluğu ile düşey yüksekliği karıştırma. Enerji denklemi sürati verir; yönü şekilden belirleriz.

## Power: how quickly is work done?

Average power is $P=W/\Delta t$. At constant velocity with force parallel to motion, $P=Fv$. Power is measured in watts: $1\,\mathrm W=1\,\mathrm{J/s}$.

For the module's lifting example, a motor raises a 200 kg crate at constant speed 0.50 m/s. There is no acceleration, so its upward force is $F=mg=200(9.81)=1962$ N.
$$P=Fv=(1962\,\mathrm N)(0.50\,\mathrm{m/s})=981\,\mathrm W.$$
In 4.0 s it lifts 2.0 m and does $W=1962(2)=3924$ J; dividing by 4.0 s again gives 981 W. This is mechanical output power; electrical input would require motor efficiency.

**TR:** İş enerji aktarımıdır; güç bunun ne kadar hızlı yapıldığını gösterir. Sabit hızda kuvvet sıfır değildir, net kuvvet sıfırdır.

### Nonzero applied work at constant speed

**Problem.** A student slides an $80.0\,\mathrm{kg}$ desk $4.00\,\mathrm m$ across a level floor at constant speed. The kinetic friction coefficient is $0.400$. Taking the applied push to be horizontal, find the student's work. Use $g=9.80\,\mathrm{m/s^2}$.

**Distinguish one force's work from net work.** Constant speed means the desk's kinetic energy does not change, so the net work is zero. It does not mean every force does zero work. The student supplies positive work while friction does equal negative work. Weight and normal force are perpendicular to the horizontal displacement and do no work.

The horizontal push leaves the normal force unchanged:

$$N=mg=(80.0)(9.80)=784\,\mathrm N.$$

Because the desk is sliding, use kinetic friction. Because its speed is constant, the push balances that friction:

$$F_{\rm push}-f_k=0,\qquad F_{\rm push}=f_k=\mu_kN=(0.400)(784)=313.6\,\mathrm N.$$

**Calculate the student's work.** The angle between the applied force and displacement is zero:

$$\begin{aligned}
W_{\rm student}&=F_{\rm push}d\cos0^\circ,\\
&=\mu_kmgd\\
&=(0.400)(80.0\,\mathrm{kg})(9.80\,\mathrm{m/s^2})(4.00\,\mathrm m)\\
&=1254.4\,\mathrm J=\boxed{1.25\,\mathrm{kJ}}.
\end{aligned}$$

The frictional work is $W_f=-1254.4\,\mathrm J$. Consequently $W_{\rm net}=W_{\rm student}+W_f=0$, agreeing with $\Delta K=0$. The bank's $1.26\,\mathrm{kJ}$ answer corresponds to using $g=9.81\,\mathrm{m/s^2}$, which gives $1255.68\,\mathrm J$. With our stated $g=9.80\,\mathrm{m/s^2}$, the value is $1254.4\,\mathrm J$, rounded to $1.25\,\mathrm{kJ}$; the difference comes from the chosen gravity value.

**Sense check.** A larger desk mass, larger friction coefficient, or longer travel distance would increase the required work. The units reduce to $\mathrm{N\,m}=\mathrm J$, rather than newtons alone.

**Türkçe destek:** Sabit hız, net işin sıfır olduğunu söyler. Öğrencinin yaptığı iş ile sürtünmenin yaptığı işi ayrı ayrı yazınca çelişki ortadan kalkar. Kuvveti bulurken sabit hızdan $a=0$ sonucuna, işi bulurken ise kuvvet ile yer değiştirme arasındaki açıya başvuruyoruz.

### Work uses the angle between two directions

**Problem.** An airplane travels $120\,\mathrm{km}$ horizontally in a direction $30.0^\circ$ north of east. The net air force is $2.40\,\mathrm{kN}$, directed $10.0^\circ$ south of west. Find the work done by this air force over the displacement.

**Put both directions on one angular scale.** Measure angles counterclockwise from east. The displacement direction is $30.0^\circ$. West is $180^\circ$, so a direction $10.0^\circ$ south of west is $190.0^\circ$. The smaller angle between the two vectors is therefore

$$\phi=190.0^\circ-30.0^\circ=160.0^\circ.$$

The force nearly opposes the displacement. We should expect negative work before doing any arithmetic. The work of a constant force is the dot product

$$W=\mathbf F\cdot\Delta\mathbf r=Fd\cos\phi.$$

**Convert both prefixes.** A kilonewton contains $10^3$ newtons, and a kilometre contains $10^3$ metres:

$$F=2.40\times10^3\,\mathrm N,\qquad d=1.20\times10^5\,\mathrm m.$$

Now substitute:

$$\begin{aligned}
W&=(2.40\times10^3)(1.20\times10^5)\cos160.0^\circ\ \mathrm{N\,m}\\
&=(2.88\times10^8)(-0.93969)\,\mathrm J\\
&=\boxed{-2.71\times10^8\,\mathrm J}.
\end{aligned}$$

**Interpret the sign.** The negative value describes energy removed from the airplane's translational motion by this particular force. It is not automatically the net work on the airplane: engine thrust can also do work. Constant altitude tells us the displacement is horizontal, not that the speed is necessarily constant.

**Sense check.** The magnitude must not exceed $Fd=2.88\times10^8\,\mathrm J$ because $|\cos\phi|\le1$. Our answer satisfies that bound and has the expected negative sign.

**Türkçe destek:** İş formülündeki açı, kuvvetin veya yolun tek başına doğuyla yaptığı açı değildir; iki vektör arasındaki açıdır. Her ikisini aynı başlangıç ekseninden ölçmek $160^\circ$ sonucunu açık hale getirir. İki ayrı kilo dönüşümü, birlikte $10^6$ çarpanı oluşturur.

*Source: 06_testbank.doc, Section 6.2 Problems, Question 6. Original numerical givens retained; restated as an open worked example.*

### Lifting work includes the final kinetic energy

**Problem.** A worker lifts a $20.0\,\mathrm{kg}$ bucket from rest at ground level to the top of a $20.0\,\mathrm m$ building. At the top its speed is $4.0\,\mathrm{m/s}$. Find the minimum work done by the worker, neglecting dissipative losses and using $g=9.80\,\mathrm{m/s^2}$.

**Use the work–kinetic energy theorem directly.** Take the bucket as the system. The worker's lifting force does positive work, while gravity does negative work over the upward displacement. For the minimum required work, neglect air resistance and other energy losses:

$$W_{\rm worker}+W_g=K_f-K_i.$$

The bucket starts from rest, so $K_i=0$. Gravity points opposite its upward displacement, giving

$$W_g=mgh\cos180^\circ=-mgh.$$

**Isolate the worker's contribution.** Insert the negative gravitational work, then add $mgh$ to both sides:

$$\begin{aligned}
W_{\rm worker}-mgh&=\frac12mv_f^2,\\
W_{\rm worker}&=mgh+\frac12mv_f^2.
\end{aligned}$$

Evaluate the two parts separately so their meanings remain visible:

$$\begin{aligned}
mgh&=(20.0)(9.80)(20.0)\,\mathrm J=3920\,\mathrm J,\\
\frac12mv_f^2&=\frac12(20.0)(4.0)^2\,\mathrm J=160\,\mathrm J,\\
W_{\rm worker}&=3920+160=\boxed{4080\,\mathrm J=4.08\,\mathrm{kJ}}.
\end{aligned}$$

**Interpret and check.** The worker must both counter gravity's negative work and leave the bucket with kinetic energy. A result of $3920\,\mathrm J$ would correspond to reaching the top with zero speed. Our answer is larger by exactly the final kinetic energy. This method needs the starting and ending speeds, not the detailed speed at every intermediate height.

**Türkçe destek:** Yerçekiminin işi yukarı çıkışta negatiftir. Denklemdeki $-mgh$ terimini karşıya almak, iki tarafa da $mgh$ eklemek demektir. Son hız sıfır verilmediğinden $\tfrac12mv_f^2$ terimini atamayız. Burada yalnızca kuvvetlerin işi ve kinetik enerji değişimini kullanıyoruz.

*Source: 06_testbank.doc, Section 6.2 Problems, Question 27. Original numerical givens retained; restated as an open worked example.*

### Find friction from the distance needed to stop

**Problem.** A $5.00\,\mathrm{kg}$ box initially slides at $3.00\,\mathrm{m/s}$ on a level floor and stops after $4.00\,\mathrm m$. If kinetic friction is the only horizontal force, determine its coefficient. Use $g=9.80\,\mathrm{m/s^2}$.

**Account for the work.** The normal force and weight are perpendicular to the displacement, so neither does work. Friction opposes motion throughout the slide and removes the box's initial kinetic energy. Because the floor is level and there is no extra vertical force, $N=mg$ and $f_k=\mu_kmg$.

The work–kinetic energy theorem is therefore

$$W_f=K_f-K_i,\qquad -\mu_kmgd=0-\frac12mv_i^2.$$

**Simplify without losing the signs.** Both sides are negative. Multiply the complete equation by $-1$; then divide by the nonzero mass:

$$\begin{aligned}
\mu_kmgd&=\frac12mv_i^2,\\
\mu_kgd&=\frac12v_i^2,\\
\mu_k&=\frac{v_i^2}{2gd}.
\end{aligned}$$

Now substitute the measured quantities:

$$\mu_k=\frac{(3.00\,\mathrm{m/s})^2}{2(9.80\,\mathrm{m/s^2})(4.00\,\mathrm m)}
=\frac{9.00}{78.4}=\boxed{0.115}.$$

**Interpret the result.** A friction coefficient is dimensionless. The given mass disappears from this idealized stopping-distance relation, but the actual friction force still depends on mass. Here $f_k=\mu_kmg=5.625\,\mathrm N$ before rounding.

**Sense check.** The initial kinetic energy is $\tfrac12(5.00)(3.00)^2=22.5\,\mathrm J$. The frictional work is $-(5.625)(4.00)=-22.5\,\mathrm J$, exactly what is needed to stop. If the same initial speed stopped over a larger distance, the inferred coefficient would be smaller.

**Türkçe destek:** Negatif iş ile azalan kinetik enerji birbirine uygundur; eksi işaretleri hatalı değildir. İki tarafı birden $-1$ ile çarpınca ikisi de pozitif olur. Kütle yalnızca bütün terimlerde ortak çarpan olduğu için sadeleşir. Sonuca newton veya joule birimi yazma; $\mu_k$ birimsizdir.

*Source: 06_testbank.doc, Section 6.2 Problems, Question 32. Original numerical givens retained; restated as an open worked example.*

### Average engine power from a kinetic-energy increase

**Problem.** A $1500\,\mathrm{kg}$ car accelerates from rest to $25\,\mathrm{m/s}$ in $7.0\,\mathrm s$. Neglect friction and air resistance. Find the average useful engine power in watts and horsepower, using $1\,\mathrm{hp}=746\,\mathrm W$.

**Choose an energy calculation.** Power describes how rapidly work is done. Under the stated loss-free model, the engine's useful work becomes the car's translational kinetic energy. We therefore need the change in kinetic energy and the elapsed time, rather than the detailed acceleration history:

$$\overline P=\frac{W}{\Delta t}=\frac{\Delta K}{\Delta t}.$$

Since the initial speed is zero,

$$\begin{aligned}
\Delta K&=\frac12m(v_f^2-v_i^2)\\
&=\frac12(1500\,\mathrm{kg})\left[(25\,\mathrm{m/s})^2-0\right]\\
&=468{,}750\,\mathrm J.
\end{aligned}$$

**Divide by the full time interval.**

$$\overline P=\frac{468{,}750\,\mathrm J}{7.0\,\mathrm s}
=66{,}964\,\mathrm W\approx\boxed{67\,\mathrm{kW}}.$$

To convert watts into horsepower, multiply by a conversion factor equal to one:

$$\overline P=(66{,}964\,\mathrm W)
\left(\frac{1\,\mathrm{hp}}{746\,\mathrm W}\right)
=89.8\,\mathrm{hp}\approx\boxed{90\,\mathrm{hp}}.$$

**Interpret and check.** This is an interval average, not necessarily the instantaneous power at the end. For constant acceleration, the force would be constant while instantaneous power $P=Fv$ rises with speed. Using the final speed in $Fv$ would then give twice the interval average. No assumption of constant acceleration was needed for the energy-based result above.

**Türkçe destek:** Önce hızın karesini alarak kinetik enerji değişimini hesapla, sonra zamana böl. Watt, joule bölü saniyedir. Beygir gücüne çevirirken $746$ ile bölmemizin nedeni, her bir beygir gücünün $746$ watt içermesidir. Ortalama güç ile yolculuğun son anındaki güç aynı kavram değildir; verilen süre ortalamayı belirler.

### L2 -- Module 06 P8: Work-Energy with Variable Force

A force $F(x)=6.0x-2.0$ (in newtons, with $x$ in metres) acts on a $2.0\,\mathrm{kg}$ object. (a) How much work does this force do as the object moves from $x=1.0\,\mathrm m$ to $x=4.0\,\mathrm m$? (b) If the object starts with a speed of $3.0\,\mathrm{m/s}$ at $x=1.0\,\mathrm m$, what is its speed at $x=4.0\,\mathrm m$?

**Türkçe problem açıklaması:** $2.0\,\mathrm{kg}$ cisme $F(x)=6.0x-2.0$ kuvveti uygulanıyor; $x$ metre olduğunda kuvvet newtondur. (a) $x=1.0\,\mathrm m$ ile $4.0\,\mathrm m$ arasındaki işi, (b) ilk sürat $3.0\,\mathrm{m/s}$ ise son sürati bulun. Önce kuvvet–konum grafiğinin işaretli alanını hesaplayın, sonra bunu kinetik enerji değişimine bağlayın.

<details>
<summary>Answer</summary>

$$\begin{aligned}\text{(a)}\quad W&=\int_1^4(6x-2)\,dx=\left[3x^2-2x\right]_1^4\\&=(48-8)-(3-2)=40-1=39.0\,\mathrm J.\end{aligned}$$

The coefficients carry the units implied by the statement: $6.0\,\mathrm{N/m}$ and $2.0\,\mathrm N$.

$$\begin{aligned}\text{(b)}\quad\frac12mv_f^2&=\frac12mv_i^2+W,\\v_f^2&=v_i^2+\frac{2W}{m}=9+\frac{2(39)}{2}=48\,\mathrm{m^2/s^2},\\v_f&=\sqrt{48}=6.93\,\mathrm{m/s}.\end{aligned}$$

</details>

```python
#@title Run once — prepare the physics demonstrations
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display, clear_output
import ipywidgets as widgets
from ipywidgets import interact, interactive, FloatSlider, IntSlider, HBox, VBox
import warnings
warnings.filterwarnings('ignore')

plt.rcParams.update({
    'figure.figsize': (8, 6),
    'font.size': 12,
    'axes.grid': True,
    'grid.alpha': 0.3
})

print('All imports successful.')
```

```python
#@title Run once — prepare compact demonstration controls
"""Embedded in the lecture notebooks by refresh_notebook_interface.py.

Only standard ipywidgets and IPython are used; notebooks remain self-contained.
"""
import functools
import traceback
import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import display, HTML
from IPython.utils.capture import capture_output
from ipywidgets import interactive as _PhysicsInteractive
from ipywidgets.widgets.interaction import show_inline_matplotlib_plots

# Set this explicitly: otherwise a local kernel can choose a desktop backend
# and block at plt.show(), leaving the notebook apparently frozen.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

try:
    from google.colab import output as _colab_output
except ImportError:
    pass
else:
    _colab_output.enable_custom_widget_manager()

_physics_panels = []
_physics_callback_errors = []


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; }
.phy101-controls { padding: 0 0 4px; box-sizing: border-box; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-animation-panel { max-width: 100%; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1;
  background: white; color: #172433; padding: 4px; }
@media (max-width: 650px) {
  .phy101-panel { flex-direction: column !important; }
  .phy101-controls, .phy101-plot-output { width: 100% !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Fit full labels and postpone costly redraws until a slider is released."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "continuous_update"):
            control.continuous_update = False
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "310px"
        control.layout.flex = "0 1 310px"
        control.layout.max_width = "100%"
        control.layout.min_width = "0"
        control.layout.margin = "2px 6px 2px 0"
        if isinstance(control, widgets.Button):
            control.layout.width = "auto"
            control.layout.flex = "0 0 auto"
    # Include the style in each output: Colab may isolate output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.Box([style] + flat, layout=widgets.Layout(
        display="flex", flex_flow="row wrap", align_items="center",
        width="100%", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        width="100%", min_width="0", max_width="100%",
        height="auto", overflow="visible", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Keep a compact control toolbar directly above the complete result."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="column",
                              align_items="stretch", width="100%"))
    panel.add_class("phy101-panel")
    return panel


class _PhysicsStableInteractive(_PhysicsInteractive):
    """Replace one persistent output model only after a redraw has completed.

    Sending clear_output followed by an inline figure depends on callback message
    capture in the frontend. Explicit MIME records keep repeated slider changes
    visible in Jupyter/Colab, without clearing the surrounding controls.
    """
    def update(self, *args):
        self.kwargs = {}
        if self.manual:
            self.manual_button.disabled = True
        try:
            for control in self.kwargs_widgets:
                self.kwargs[control._kwarg] = control.get_interact_value()
            with capture_output(stdout=True, stderr=True, display=True) as captured:
                self.result = self.f(**self.kwargs)
                show_inline_matplotlib_plots()
                if self.auto_display and self.result is not None:
                    display(self.result)
            records = []
            if captured.stdout:
                records.append({"output_type": "stream", "name": "stdout", "text": captured.stdout})
            if captured.stderr:
                records.append({"output_type": "stream", "name": "stderr", "text": captured.stderr})
            records.extend({"output_type": "display_data", "data": output.data,
                            "metadata": output.metadata or {}} for output in captured.outputs)
            # Keep the old result visible during calculation; publish the complete
            # replacement as one widget-state update, including its PNG/HTML data.
            self.out.outputs = tuple(records)
        except Exception:
            _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), traceback.format_exc()))
            self.out.outputs = ({"output_type": "stream", "name": "stderr",
                                 "text": traceback.format_exc()},)
        finally:
            if self.manual:
                self.manual_button.disabled = False


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


def physics_interactive(function, **controls):
    """An interactive widget with the same arguments as ipywidgets.interactive."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        try:
            return function(*args, **kwargs)
        except Exception as error:
            _physics_callback_errors.append((function.__name__, repr(error)))
            raise

    panel = _PhysicsStableInteractive(checked, {"auto_display": True}, **controls)
    inputs = [child for child in panel.children if child is not panel.out]
    panel.children = (_physics_controls(inputs), _physics_output(panel.out))
    panel.layout = widgets.Layout(display="flex", flex_flow="column",
                                  align_items="stretch", width="100%")
    panel.add_class("phy101-panel")
    _physics_panels.append(panel)
    return panel


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel
    display(panel)
    return function
```

### Visual check 1: 6. Interactive Demo 4 -- Work Done by a Variable Force (Area Under F-x Curve)

Use the controls below. Predict → adjust one value → explain the result. / Tahmin et → tek değeri değiştir → sonucu açıkla.

```python
#@title Run the demonstration — predict, adjust, observe
try:
    _trapz = np.trapezoid
except AttributeError:
    _trapz = np.trapz

def work_area_under_curve(x_end=5.0, force_type=1):
    """Interactive F-x diagram showing work as area under the curve."""
    x = np.linspace(0, 8, 500)

    if force_type == 1:
        # Linear spring force: F = kx
        k = 10.0
        F = k * x
        label = f'Spring force F = kx  (k = {k} N/m)'
        W_exact = 0.5 * k * x_end**2
        formula = f'W = (1/2)kx^2 = (1/2)({k})({x_end:.1f})^2 = {W_exact:.1f} J'
    elif force_type == 2:
        # Constant force
        F0 = 15.0
        F = F0 * np.ones_like(x)
        label = f'Constant force F = {F0} N'
        W_exact = F0 * x_end
        formula = f'W = F*d = {F0}*{x_end:.1f} = {W_exact:.1f} J'
    else:
        # Varying force: F = 20*sin(x) + 5
        F = 20 * np.sin(0.5 * x) + 5
        label = 'F(x) = 20 sin(0.5x) + 5 N'
        mask = x <= x_end
        W_exact = _trapz(F[mask], x[mask])
        formula = f'W = integral F dx = {W_exact:.1f} J (numerical)'

    fig, ax = plt.subplots(figsize=(10, 6))

    ax.plot(x, F, 'b-', lw=2, label=label)

    # Shade area under curve up to x_end
    mask = x <= x_end
    ax.fill_between(x[mask], 0, F[mask], alpha=0.3, color='dodgerblue')

    # Boundary line
    ax.axvline(x=x_end, color='red', ls='--', lw=2, label=f'x = {x_end:.1f} m')

    ax.set_xlim(0, 8)
    ax.set_ylim(0, max(F)*1.2)
    ax.set_xlabel('Position x (m)', fontsize=12)
    ax.set_ylabel('Force F (N)', fontsize=12)
    ax.set_title('Work = Area Under the F-x Curve', fontsize=14)
    ax.legend(fontsize=11)

    # Annotate work
    ax.text(x_end/2, max(F)*0.8,
            f'W = {W_exact:.1f} J',
            fontsize=16, fontweight='bold', color='darkblue',
            ha='center',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='lightyellow', alpha=0.9))

    ax.text(0.5, max(F)*0.55, formula, fontsize=11, color='navy')

    plt.tight_layout()
    plt.show()

physics_interact(work_area_under_curve,
         x_end=FloatSlider(min=0.5, max=7.5, step=0.25, value=5.0, description='x_end (m)'),
         force_type=widgets.fixed(2));
```

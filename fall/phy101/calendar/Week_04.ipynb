## Read and work through the examples / Konu ve çözümlü örnekler

## Before calculating: a free-body diagram becomes algebra

**Core route:** select one object → draw only forces acting on it → choose axes → write $\sum F_x=ma_x$ and $\sum F_y=ma_y$ → solve by hand. The simulations are optional checks; you are learning mechanics, not Python syntax.

**Forces are interactions.** Weight $mg$ comes from Earth, the normal force $N$ from a surface, tension $T$ from a string, and friction from contact. 'Net force' is their sum, not an extra arrow. A third-law partner acts on the other object and does not belong in the same free-body diagram.

**Algebra bridge:** for a lifted mass, take up positive. Isolate tension $T$:

$$\begin{aligned}
T-mg&=ma,\\
T&=ma+mg &&\text{add }mg\text{ to both sides},\\
T&=m(a+g) &&\text{factor the common mass }m.
\end{aligned}$$

Both terms in the parentheses have acceleration units. Thus $\mathrm{kg}\times\mathrm{m/s^2}=\mathrm N$. If $a=0$, the expression reduces to $T=mg$.

**Inclines:** choose one axis along the slope and one perpendicular. Weight components are $mg\sin\theta$ down the slope and $mg\cos\theta$ into it. If no other perpendicular force acts and there is no perpendicular acceleration, $N=mg\cos\theta$. On a horizontal surface with an angled push, $N$ need not equal $mg$.

**Friction decision:** while at rest, $0\le\lvert f_s\rvert\le\mu_sN$; static friction adjusts to what is needed. Check that limit before assuming motion. During sliding, $\lvert f_k\rvert=\mu_kN$ opposite relative sliding. Knowing only $\mu_k$ is insufficient to decide whether a resting box starts moving.

**Review stop:** a block moving right at constant speed has zero net horizontal force. It can still have two nonzero opposing horizontal forces. Draw them and show their cancellation.

**Türkçe:** Önce hangi cismi incelediğini belirle. Kuvvetler toplamını yazmadan $F=ma$ içine tek bir kuvveti koyma. Sürtünmenin yönünü ve cismin kayıp kaymadığını ayrıca kontrol et.

**Small force balance:** A $2\,\mathrm{kg}$ box is pushed downward by an additional vertical $5\,\mathrm N$ while resting on a floor. Upward positive gives

$$N-mg-5\,\mathrm N=0\quad\Rightarrow\quad N=(2)(9.81)+5=24.62\,\mathrm N.$$

Adding $mg+5\,\mathrm N$ to both sides isolates $N$. It exceeds the weight because the surface supports the extra push too.

**Türkçe:** Normal kuvvetini otomatik olarak $mg$ yazma. Düşey ivme sıfır olsa bile ek aşağı kuvvet, zeminin daha büyük tepki kuvveti vermesini gerektirir.

### 2.3 Friction Model

Static friction adjusts to the needed value until its limit. At equality the contact is at impending slip; a larger demand cannot be supported by static friction. During sliding use the kinetic magnitude and the sliding direction.

**Think–Pair–Explain:** A horizontal contact has $N=20\,\mathrm N$, $\mu_s=0.40$, and $\mu_k=0.30$. A resting box is pulled with $5\,\mathrm N$. What is friction? What changes for a $10\,\mathrm N$ pull?

**Model response:** The static limit is $(0.40)(20)=8\,\mathrm N$. At a $5\,\mathrm N$ pull, static friction is $5\,\mathrm N$, not $8\,\mathrm N$. A $10\,\mathrm N$ pull exceeds the limit; once sliding begins, kinetic friction is $(0.30)(20)=6\,\mathrm N$, giving net force $10-6=4\,\mathrm N$.

**Türkçe:** $\mu_sN$ her durumda sürtünmenin kendisi değildir, sağlayabileceği en büyük değerdir. Kayma başladıktan sonra farklı bir model ve $\mu_k$ kullanılır.

### 2.4 Strategy for Solving Force Problems

1. Select an object and identify its contacts.
2. Draw only the real forces acting on it; net force is their sum, not an extra arrow.
3. Choose axes; tilt them with an incline when useful.
4. Write $\sum F_x=ma_x$ and $\sum F_y=ma_y$ with signs.
5. Rearrange symbolically, substitute units, then test direction and limiting cases.

## 8. Worked Examples

### Example 1: Block on a Flat Surface

A $10\,\mathrm{kg}$ block on a horizontal surface ($\mu_k=0.25$) is pushed with $40\,\mathrm N$. Find the acceleration.

### Hand calculation and the missing assumption

Assume the block is already sliding in the direction of the horizontal push. Only $\mu_k$ is supplied, so we cannot determine whether a resting block starts moving.

Choose forward and upward positive. Vertical acceleration is zero. First find normal force $N$, then kinetic-friction magnitude $f_k$:

$$\begin{aligned}
N-mg&=0,\\
N&=mg=(10\,\mathrm{kg})(9.81\,\mathrm{m/s^2})=98.1\,\mathrm N,\\
f_k&=\mu_kN=(0.25)(98.1\,\mathrm N)=24.525\,\mathrm N.
\end{aligned}$$

Friction points backward. Write its minus sign in the horizontal sum, then divide both sides by mass:

$$\begin{aligned}
F-f_k&=ma,\\
40\,\mathrm N-24.525\,\mathrm N&=(10\,\mathrm{kg})a,\\
a&=\frac{15.475\,\mathrm N}{10\,\mathrm{kg}}=1.5475\,\mathrm{m/s^2}.
\end{aligned}$$

**Result:** Acceleration is $1.55\,\mathrm{m/s^2}$ forward under the sliding assumption.

**Check:** It is smaller than the frictionless value $F/m=4\,\mathrm{m/s^2}$. The net force, not the applied force alone, equals $ma$.

**Türkçe:** Önce düşey dengeyle normal kuvveti bul; sürtünme bu kuvvete bağlıdır. Ardından yatay kuvvetleri işaretleriyle topla. Kinetik katsayı kayma hesabına yeter, fakat başlangıçta durgun kutunun harekete geçip geçmeyeceğini belirlemez.

### Example 2: Block on an Incline

A $5\,\mathrm{kg}$ block is placed on a $30^\circ$ incline with $\mu_s=0.4$ and $\mu_k=0.3$. Does it slide? If yes, what is its acceleration?

### Hand calculation: first decide whether it moves

Choose downhill positive. Let $W_{\parallel}$ be the downhill weight component. There is no acceleration perpendicular to the plane:

$$\begin{aligned}
N&=mg\cos30^\circ=(5)(9.81)\cos30^\circ=42.4785\,\mathrm N,\\
W_{\parallel}&=mg\sin30^\circ=(5)(9.81)\sin30^\circ=24.525\,\mathrm N,\\
f_{s,\max}&=\mu_sN=(0.4)(42.4785)=16.9914\,\mathrm N.
\end{aligned}$$

The needed uphill friction exceeds the static limit: $24.525\,\mathrm N>16.9914\,\mathrm N$. Therefore **the block slides downhill**.

After sliding starts, replace static friction by kinetic friction. Keep its uphill direction as a minus sign:

$$\begin{aligned}
f_k&=\mu_kN=(0.3)(42.4785)=12.7436\,\mathrm N,\\
W_{\parallel}-f_k&=ma,\\
a&=\frac{24.525\,\mathrm N-12.7436\,\mathrm N}{5\,\mathrm{kg}}=2.3563\,\mathrm{m/s^2}.
\end{aligned}$$

The same result follows symbolically by factoring and dividing by nonzero mass:

$$m\bigl(g\sin\theta-\mu_kg\cos\theta\bigr)=ma
\quad\Rightarrow\quad a=g(\sin\theta-\mu_k\cos\theta).$$

**Check:** The acceleration lies between zero and the frictionless value $g\sin30^\circ=4.905\,\mathrm{m/s^2}$. The critical static angle satisfies $\tan\theta_c=\mu_s$, and $30^\circ>\tan^{-1}(0.4)$.

**Türkçe:** Kayma kararını kuvvetlerin büyüklüklerini karşılaştırarak ver. Kararda $\mu_s$, hareket başladıktan sonraki ivmede $\mu_k$ kullanılır. Son bölümde her terimde aynı kütle bulunduğu için denklemin iki tarafını kütleye bölüyoruz.

### Example 3: Atwood Machine

Two masses $m_1=7\,\mathrm{kg}$ and $m_2=3\,\mathrm{kg}$ are connected by a string over a frictionless pulley. Find acceleration and tension.

### Hand calculation: add two force equations

Assume a light, inextensible string and an ideal pulley. Take downward positive for $m_1$ and upward positive for $m_2$. Their acceleration magnitudes are the same, denoted $a$; the string tension is $T$ on each side.

$$\begin{aligned}
m_1g-T&=m_1a &&\text{heavier mass},\\
T-m_2g&=m_2a &&\text{lighter mass}.
\end{aligned}$$

Add the equations. The opposite tension terms cancel:

$$\begin{aligned}
(m_1-m_2)g&=(m_1+m_2)a,\\
a&=\frac{m_1-m_2}{m_1+m_2}g,\\
 &=\frac{7-3}{7+3}(9.81\,\mathrm{m/s^2})=3.924\,\mathrm{m/s^2}.
\end{aligned}$$

To find tension, use the lighter mass and add its weight to both sides:

$$\begin{aligned}
T-m_2g&=m_2a,\\
T&=m_2(g+a),\\
 &=(3\,\mathrm{kg})(9.81+3.924)\,\mathrm{m/s^2}=41.202\,\mathrm N.
\end{aligned}$$

**Check:** The heavier mass gives the same answer, $T=m_1(g-a)=(7)(9.81-3.924)=41.202\,\mathrm N$. Acceleration is less than $g$; tension lies between the two weights.

**Türkçe:** İki kütle için pozitif yönleri farklı seçiyoruz, çünkü biri aşağı inerken diğeri yukarı çıkar. Bu seçimle aynı pozitif $a$ büyüklüğü kullanılabilir. Denklemleri toplarken gerilme yok olur; gerilmeyi daha sonra tek kütlenin denkleminden geri buluruz.

### A descending elevator can make the scale read more

**Problem.** A $60.0\,\mathrm{kg}$ person stands on a scale inside an elevator. The elevator is moving downward but slowing at $2.00\,\mathrm{m/s^2}$. Find the scale reading in newtons, using $g=9.80\,\mathrm{m/s^2}$.

**Separate velocity from acceleration.** Moving downward tells us the velocity direction. Slowing downward motion requires an upward acceleration. Choose upward as positive, so the correct acceleration is $a_y=+2.00\,\mathrm{m/s^2}$. The scale measures the upward normal contact force $N$ on the person's feet; it does not directly measure the gravitational force.

For the person, the forces are $N$ upward and $mg$ downward. Newton's second law gives

$$N-mg=ma_y.$$

**Rearrange the force equation.** Add $mg$ to both sides and factor out the common mass:

$$\begin{aligned}
N&=ma_y+mg,\\
N&=m(a_y+g)\\
&=(60.0\,\mathrm{kg})(2.00+9.80)\,\mathrm{m/s^2}\\
&=\boxed{708\,\mathrm N}.
\end{aligned}$$

The person's ordinary weight is only

$$mg=(60.0)(9.80)\,\mathrm N=588\,\mathrm N.$$

**Interpret the difference.** The additional $708-588=120\,\mathrm N$ provides the upward net force that reduces the downward speed. The person can feel heavier while travelling downward; acceleration determines the change in apparent weight. If the scale displays a mass-like reading, its calibrated display would be $N/g\approx72.2\,\mathrm{kg}$, although the person's actual mass remains $60.0\,\mathrm{kg}$.

**Sense check.** A normal force below the weight would give downward acceleration, making the elevator speed up downward. That would contradict the stated slowing motion. Our larger normal force has the required sign.

**Türkçe destek:** “Aşağı gidiyor” ile “aşağı ivmeleniyor” aynı değildir. Aşağı yönlü hız azalırken ivme yukarı yönlüdür. Baskülün gösterdiği kuvvet $N$'dir; kişinin kütlesi değişmez. İşaret seçimini önce kelimelerle belirlemek, formülde yanlışlıkla $g-a$ kullanmayı önler.

### Up and back on a frictionless incline

**Problem.** A brief push gives a block an initial speed of $12.0\,\mathrm{m/s}$ up a frictionless incline at $20.0^\circ$ above horizontal. Once the push ends, find its distance up the slope before stopping and the total time until it returns to its starting point. Use $g=9.80\,\mathrm{m/s^2}$.

**Choose the slope as the axis.** Positive displacement points uphill. The normal force cancels the perpendicular component of weight, while gravity's parallel component points downhill. After the brief push, there is no continuing applied force. Thus

$$ma=-mg\sin\theta\quad\Rightarrow\quad a=-g\sin\theta=-3.3518\,\mathrm{m/s^2}.$$

The mass cancels because it multiplies both sides. This acceleration remains downhill throughout the motion, including at the instant the block stops.

**Find the turning-point distance.** At the top, $v=0$. Rearrange the equation without time:

$$\begin{aligned}
0&=v_0^2+2as,\\
s&=-\frac{v_0^2}{2a}
=\frac{v_0^2}{2g\sin\theta}\\
&=\frac{(12.0)^2}{2(9.80)\sin20.0^\circ}\,\mathrm m
=\boxed{21.5\,\mathrm m}.
\end{aligned}$$

**Find the complete travel time.** From $v=v_0+at$, the uphill time is $t_{\rm up}=v_0/(g\sin\theta)=3.58\,\mathrm s$. Returning to the original position requires the nonzero solution of

$$0=v_0t-\frac12g\sin\theta\,t^2
=t\left(v_0-\frac12g\sin\theta\,t\right).$$

The factor $t=0$ is the initial instant. The return occurs at

$$t_{\rm return}=\frac{2v_0}{g\sin\theta}=\boxed{7.16\,\mathrm s}.$$

**Sense check.** The return time is twice the ascent time, as expected for this frictionless constant-acceleration motion. The final velocity is $-12.0\,\mathrm{m/s}$: equal speed, opposite direction.

**Türkçe destek:** En üst noktada hız sıfırdır ama ivme sıfır değildir; yerçekimi etkisini sürdürür. Dönüş zamanı denkleminde iki kök bulunur. $t=0$ başlangıcı, diğer kök ise bloğun gerçekten geri döndüğü anı gösterir.

### Static friction adjusts; kinetic friction has a different value

**Problem.** A $50.0\,\mathrm{kg}$ box rests on a horizontal floor. Its friction coefficients are $\mu_s=0.300$ and $\mu_k=0.200$. Determine the friction force when the box is subjected separately to horizontal pushes of $140\,\mathrm N$ and $175\,\mathrm N$. Start each case from rest and use $g=9.80\,\mathrm{m/s^2}$.

**First determine whether motion starts.** The push is horizontal, so it does not change vertical force balance:

$$N=mg=(50.0)(9.80)\,\mathrm N=490\,\mathrm N.$$

Static friction can take any required magnitude up to a maximum:

$$0\le f_s\le\mu_sN,\qquad f_{s,\max}=(0.300)(490)=147\,\mathrm N.$$

The expression $\mu_sN$ is a limit, not the friction force in every stationary situation. Compare the push with this limit before choosing the friction model.

**Case (a): the push is $140\,\mathrm N$.** Static friction can match this because $140<147$. The horizontal force equation for a stationary box is

$$F-f_s=0\quad\Rightarrow\quad f_s=F=\boxed{140\,\mathrm N}.$$

It acts opposite the push. There is no acceleration and no sliding.

**Case (b): the push is $175\,\mathrm N$.** Static friction cannot supply $175\,\mathrm N$, so the box starts sliding. Once it slides, use the kinetic coefficient:

$$f_k=\mu_kN=(0.200)(490)=\boxed{98.0\,\mathrm N}.$$

The resulting horizontal net force is $175-98.0=77.0\,\mathrm N$, giving $a=77.0/50.0=1.54\,\mathrm{m/s^2}$ while this push continues.

**Sense check.** At a zero push, static friction would be zero, even though $\mu_sN$ is still $147\,\mathrm N$. This limiting example shows why automatically substituting $f_s=\mu_sN$ is incorrect.

**Türkçe destek:** Statik sürtünme, cismi durgun tutmak için gereken değere ayarlanır; $147\,\mathrm N$ onun ulaşabileceği en büyük değerdir. Eşik aşılınca hesapta $\mu_s$ yerine $\mu_k$ kullanılır. İki durum bağımsızdır; ikinci durumu birincinin devamı olarak yorumlamıyoruz.

### Friction accelerates an object onto a moving belt

**Problem.** A $10\,\mathrm{kg}$ computer is placed on a conveyor belt moving at $3.1\,\mathrm{m/s}$. Initially its horizontal speed relative to the floor is zero. The coefficients are $\mu_s=0.50$ and $\mu_k=0.30$. Find its horizontal travel distance before it moves together with the belt. Model the motion after landing; use $g=9.80\,\mathrm{m/s^2}$.

**Decide the friction direction.** At first the belt slides forward underneath the computer. Friction opposes this relative sliding, so the belt's frictional force on the computer points forward. Friction can therefore increase an object's speed; it does not always point opposite that object's velocity relative to the room.

During sliding the normal force is $N=mg$. Newton's second law gives

$$\begin{aligned}
f_k&=\mu_kmg=ma,\\
a&=\mu_kg=(0.30)(9.80)=2.94\,\mathrm{m/s^2}.
\end{aligned}$$

We cancel the same nonzero mass from both sides. The computer accelerates until its floor-frame speed reaches the belt speed. At that point the surfaces no longer slide relative to each other.

**Calculate the distance.** With $v_0=0$ and $v=3.1\,\mathrm{m/s}$,

$$\begin{aligned}
v^2-v_0^2&=2as,\\
s&=\frac{v^2}{2a}
=\frac{(3.1\,\mathrm{m/s})^2}{2(2.94\,\mathrm{m/s^2})}\\
&=\boxed{1.63\,\mathrm m\approx1.6\,\mathrm m}.
\end{aligned}$$

**Check with time.** The matching time is $t=v/a=1.05\,\mathrm s$. The computer's average speed during constant acceleration is $(0+3.1)/2=1.55\,\mathrm{m/s}$; multiplying by the unrounded time reproduces $1.63\,\mathrm m$. After matching speeds on an ideal horizontal belt, no horizontal force is needed to maintain that constant speed.

**Türkçe destek:** Sürtünmenin yönünü zemine göre hıza bakarak seçme; temas eden yüzeylerin birbirine göre kaymasına bak. Başlangıçta kayma olduğu için $\mu_k$ gerekir. Verilen $\mu_s$, bu kayma evresinin hesabına girmez. Son hızın kayış hızına eşitlenmesi, hangi anda hesabı durduracağımızı belirler.

### Resolve gravity and friction on a downhill slope

**Problem.** A $6.0\,\mathrm{kg}$ box slides down a slope inclined at $39^\circ$ above horizontal. The kinetic friction coefficient is $0.19$. Find its acceleration down the slope, using $g=9.80\,\mathrm{m/s^2}$.

**Rotate the coordinate axes to match the surface.** Let positive $x$ point downhill and positive $y$ point perpendicular to the surface, away from it. Weight has a downhill component $mg\sin\theta$ and a perpendicular component $mg\cos\theta$ into the surface. Friction acts uphill because the box is already sliding downhill.

There is no acceleration through the surface. Therefore

$$N-mg\cos\theta=0\quad\Rightarrow\quad N=mg\cos\theta.$$

This is why the normal force on an incline is generally smaller than $mg$. The kinetic friction magnitude is

$$f_k=\mu_kN=\mu_kmg\cos\theta.$$

**Write and rearrange the downhill equation.** Keeping directions explicit,

$$\begin{aligned}
mg\sin\theta-f_k&=ma,\\
mg\sin\theta-\mu_kmg\cos\theta&=ma,\\
a&=g\left(\sin\theta-\mu_k\cos\theta\right).
\end{aligned}$$

Substitute the angle in degrees, taking care that both trigonometric terms use the same angle:

$$a=(9.80)\left[\sin39^\circ-(0.19)\cos39^\circ\right]
=\boxed{4.72\,\mathrm{m/s^2}\ \text{downhill}}.$$

**Interpret and check.** Without friction the acceleration would be $g\sin39^\circ\approx6.17\,\mathrm{m/s^2}$. Friction reduces it, as our smaller answer shows. The result is still positive, so the already sliding box speeds up downhill. The mass cancels from this ideal model, although the individual normal and friction forces both depend on mass.

**Türkçe destek:** Normal kuvveti önce dik doğrultudaki dengeden bul. Sonra sürtünmeyi $\mu_kN$ ile hesapla; doğrudan $\mu_kmg$ kullanmak eğimin etkisini kaybettirir. Eksi işareti sürtünmenin seçtiğimiz pozitif yöne, yani yokuş aşağıya, karşı olmasından gelir. Ortak $m$ çarpanı ancak bütün terimlerde bulunduğu için sadeleşir.

### L1 -- Module 04 P2: Weight and Normal Force

A $12.0\,\mathrm{kg}$ box sits on a horizontal floor. A person pushes downward on the box with a force of $40.0\,\mathrm N$ at an angle of $30^\circ$ below the horizontal. Find (a) the normal force on the box, and (b) the acceleration if the coefficient of kinetic friction is $\mu_k=0.20$.

```python
#@title Run once — prepare the physics demonstrations
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import matplotlib.transforms as transforms
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display, clear_output
import ipywidgets as widgets

plt.rcParams.update({
    'figure.figsize': (10, 6),
    'font.size': 13,
    'axes.grid': True,
    'grid.alpha': 0.3
})

g = 9.81  # m/s^2
print("All imports ready.")
```

```python
#@title Run once — prepare compact demonstration controls
"""Embedded in the lecture notebooks by refresh_notebook_interface.py.

Only standard ipywidgets and IPython are used; notebooks remain self-contained.
"""
import functools
import traceback
import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import display, HTML
from IPython.utils.capture import capture_output
from ipywidgets import interactive as _PhysicsInteractive
from ipywidgets.widgets.interaction import show_inline_matplotlib_plots

# Set this explicitly: otherwise a local kernel can choose a desktop backend
# and block at plt.show(), leaving the notebook apparently frozen.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

try:
    from google.colab import output as _colab_output
except ImportError:
    pass
else:
    _colab_output.enable_custom_widget_manager()

_physics_panels = []
_physics_callback_errors = []


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; }
.phy101-controls { padding: 0 0 4px; box-sizing: border-box; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-animation-panel { max-width: 100%; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1;
  background: white; color: #172433; padding: 4px; }
@media (max-width: 650px) {
  .phy101-panel { flex-direction: column !important; }
  .phy101-controls, .phy101-plot-output { width: 100% !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Fit full labels and postpone costly redraws until a slider is released."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "continuous_update"):
            control.continuous_update = False
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "310px"
        control.layout.flex = "0 1 310px"
        control.layout.max_width = "100%"
        control.layout.min_width = "0"
        control.layout.margin = "2px 6px 2px 0"
        if isinstance(control, widgets.Button):
            control.layout.width = "auto"
            control.layout.flex = "0 0 auto"
    # Include the style in each output: Colab may isolate output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.Box([style] + flat, layout=widgets.Layout(
        display="flex", flex_flow="row wrap", align_items="center",
        width="100%", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        width="100%", min_width="0", max_width="100%",
        height="auto", overflow="visible", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Keep a compact control toolbar directly above the complete result."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="column",
                              align_items="stretch", width="100%"))
    panel.add_class("phy101-panel")
    return panel


class _PhysicsStableInteractive(_PhysicsInteractive):
    """Replace one persistent output model only after a redraw has completed.

    Sending clear_output followed by an inline figure depends on callback message
    capture in the frontend. Explicit MIME records keep repeated slider changes
    visible in Jupyter/Colab, without clearing the surrounding controls.
    """
    def update(self, *args):
        self.kwargs = {}
        if self.manual:
            self.manual_button.disabled = True
        try:
            for control in self.kwargs_widgets:
                self.kwargs[control._kwarg] = control.get_interact_value()
            with capture_output(stdout=True, stderr=True, display=True) as captured:
                self.result = self.f(**self.kwargs)
                show_inline_matplotlib_plots()
                if self.auto_display and self.result is not None:
                    display(self.result)
            records = []
            if captured.stdout:
                records.append({"output_type": "stream", "name": "stdout", "text": captured.stdout})
            if captured.stderr:
                records.append({"output_type": "stream", "name": "stderr", "text": captured.stderr})
            records.extend({"output_type": "display_data", "data": output.data,
                            "metadata": output.metadata or {}} for output in captured.outputs)
            # Keep the old result visible during calculation; publish the complete
            # replacement as one widget-state update, including its PNG/HTML data.
            self.out.outputs = tuple(records)
        except Exception:
            _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), traceback.format_exc()))
            self.out.outputs = ({"output_type": "stream", "name": "stderr",
                                 "text": traceback.format_exc()},)
        finally:
            if self.manual:
                self.manual_button.disabled = False


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


def physics_interactive(function, **controls):
    """An interactive widget with the same arguments as ipywidgets.interactive."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        try:
            return function(*args, **kwargs)
        except Exception as error:
            _physics_callback_errors.append((function.__name__, repr(error)))
            raise

    panel = _PhysicsStableInteractive(checked, {"auto_display": True}, **controls)
    inputs = [child for child in panel.children if child is not panel.out]
    panel.children = (_physics_controls(inputs), _physics_output(panel.out))
    panel.layout = widgets.Layout(display="flex", flex_flow="column",
                                  align_items="stretch", width="100%")
    panel.add_class("phy101-panel")
    _physics_panels.append(panel)
    return panel


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel
    display(panel)
    return function
```

### Visual check 1: 6. Interactive Demo 4 — Friction Force Visualizer

Use the controls below. Predict → adjust one value → explain the result. / Tahmin et → tek değeri değiştir → sonucu açıkla.

```python
#@title Run the demonstration — predict, adjust, observe
def friction_visualizer(mass=5.0, mu_s=0.5, mu_k=0.35):
    """Show friction force vs applied force."""
    N = mass * g
    f_s_max = mu_s * N
    f_k = mu_k * N

    F_applied = np.linspace(0, max(f_s_max * 1.8, f_k * 1.8, 1.0), 500)
    f_friction = np.where(F_applied <= f_s_max, F_applied, f_k)
    a_net = np.where(F_applied <= f_s_max, 0, (F_applied - f_k) / mass)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.6, 4.2))
    fig.subplots_adjust(left=0.07, right=0.98, bottom=0.29, top=0.83, wspace=0.30)
    fig.suptitle(rf'$m={mass:.1f}\,\mathrm{{kg}},\quad N=mg={N:.2f}\,\mathrm{{N}},\quad '
                 rf'\mu_s={mu_s:.2f},\quad \mu_k={mu_k:.2f}$', fontsize=11)

    # Left: friction vs applied
    ax1.plot(F_applied, f_friction, 'b-', linewidth=2.5, label='Friction force')
    ax1.plot(F_applied, F_applied, 'k--', linewidth=1, alpha=0.4, label='f = F (reference)')
    ax1.axhline(f_s_max, color='red', linestyle=':', alpha=0.7, label=f'f_s(max) = {f_s_max:.1f} N')
    ax1.axhline(f_k, color='orange', linestyle=':', alpha=0.7, label=f'f_k = {f_k:.1f} N')
    ax1.axvline(f_s_max, color='gray', linestyle='--', alpha=0.3)

    # Shade regions
    ax1.axvspan(0, f_s_max, alpha=0.08, color='green', label='Static region')
    ax1.axvspan(f_s_max, F_applied[-1], alpha=0.08, color='red', label='Kinetic region')

    ax1.set_xlabel('Applied Force (N)', fontsize=12)
    ax1.set_ylabel('Friction Force (N)', fontsize=12)
    ax1.set_title('Friction vs Applied Force', fontsize=14)
    ax1.set_xlim(0, F_applied[-1])
    ax1.set_ylim(0, max(f_s_max * 1.5, f_k * 1.5, 1.0))

    # Right: acceleration vs applied
    ax2.plot(F_applied, a_net, 'r-', linewidth=2.5)
    ax2.axvline(f_s_max, color='gray', linestyle='--', alpha=0.3)
    ax2.axvspan(0, f_s_max, alpha=0.08, color='green')
    ax2.axvspan(f_s_max, F_applied[-1], alpha=0.08, color='red')
    ax2.set_xlabel('Applied Force (N)', fontsize=12)
    ax2.set_ylabel('Acceleration (m/s²)', fontsize=12)
    ax2.set_title('Acceleration vs Applied Force\nStatic: a = 0 | Kinetic: sliding', fontsize=11)

    # A figure-level legend has its own space, below both x-axis labels.
    handles, labels = ax1.get_legend_handles_labels()
    fig.legend(handles, labels, loc='lower center', bbox_to_anchor=(0.5, 0.02),
               ncol=3, fontsize=9, frameon=False)
    for ax in (ax1, ax2):
        ax.tick_params(labelsize=9)
        ax.xaxis.label.set_size(10)
        ax.yaxis.label.set_size(10)
        ax.title.set_fontsize(11)
    plt.show()


@physics_interact(
    mass=widgets.FloatSlider(value=5, min=1, max=20, step=0.5, description='Mass (kg):'),
    mu_s=widgets.FloatSlider(value=0.5, min=0.1, max=1.0, step=0.05,
                             description='μ_s:', readout_format='.2f'),
    mu_k=widgets.FloatSlider(value=0.35, min=0.05, max=0.9, step=0.05,
                             description='μ_k:', readout_format='.2f')
)
def _friction_viz(mass, mu_s, mu_k):
    if mu_k > mu_s:
        print("⚠️  μ_k should be ≤ μ_s. Adjusting μ_k = μ_s.")
        mu_k = mu_s
    friction_visualizer(mass, mu_s, mu_k)
```

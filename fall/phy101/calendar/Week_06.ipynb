## Review A — motion plus force, with a complete worked route

A 2.0 kg cart begins at rest on a horizontal track. A 10 N force pulls right while a 2 N friction force acts left for 3.0 s. Assume the cart is already in the sliding regime once motion begins and these forces stay constant.

1. Draw the cart: right pull, left friction, upward normal, downward weight. Choose right positive.
2. Net horizontal force is $10-2=8$ N. From $F_{\rm net}=ma$, divide by 2.0 kg: $a=4.0$ m/s².
3. $v=0+at=4.0(3.0)=12.0$ m/s and $d=0+at^2/2=4.0(9.0)/2=18.0$ m.
4. Check the same motion with work: $W_{\rm net}=F_{\rm net}d=8(18)=144$ J. The change in kinetic energy is $\Delta K=\tfrac12(2)(12^2)-0=144$ J.

**Explain:** Why is $10/2$ the wrong acceleration? Why is friction work negative even while speed increases? **Answer:** acceleration uses the net 8 N, and the negative friction work is smaller than the positive pulling work. **TR:** Aynı durumu iki yöntemle kontrol etmek, yalnızca son sayıyı kontrol etmekten daha değerlidir.

## Review B — one clock for two directions

A ball leaves a 1.25 m high table horizontally at 3.0 m/s. Ignore drag. Set upward positive and floor height to zero.
$$0=1.25-\tfrac12gt^2\Rightarrow t^2=\frac{2(1.25)}{9.81}
\Rightarrow t=0.50482\,\mathrm s.$$
Horizontal range is $x=v_xt=3.0(0.50482)=1.5145$ m. Just before impact $v_y=-gt=-4.9523$ m/s. Speed is
$$v=\sqrt{v_x^2+v_y^2}=\sqrt{3.0^2+4.9523^2}=5.7901\,\mathrm{m/s}.$$
The vertical force is $mg$ down throughout the flight; horizontal net force is zero. At release the vertical speed is zero but acceleration is already downward.

**Try then check:** If horizontal launch speed doubles, time stays 0.50482 s and range doubles to 3.0289 m. **TR:** Yatay hız iki katına çıkınca düşey denklem değişmez; ortak süre değişmeden yatay yol artar.

## Review C — repair a tempting wrong solution

A 1000 kg vehicle slows from 72 km/h to 36 km/h. A learner writes “energy halves because speed halves.” Correct the reasoning.

Convert the speeds first: $72/3.6=20$ m/s and $36/3.6=10$ m/s.
$$K_i=\tfrac12(1000)(20^2)=200000\,\mathrm J,
\quad K_f=\tfrac12(1000)(10^2)=50000\,\mathrm J.$$
$$W_{\rm net}=K_f-K_i=-150000\,\mathrm J.$$
Energy becomes one quarter, because speed is squared. If the net retarding force is constant at 3000 N, take forward positive and write $W=-Fd$:
$$-150000=-3000d\Rightarrow d=50.0\,\mathrm m.$$
**Check:** $a=-3000/1000=-3.0$ m/s², and $10^2=20^2+2(-3)d$ also gives 50.0 m. **TR:** Eksi işaretini kaybetme; fren kuvveti hareket yönüne terstir.

### L1 -- Module 01 P2: Dimensional Analysis Check

Check $F=C\rho v^2A$, where $C$ is dimensionless, $\rho$ is density in $\mathrm{kg/m^3}$, $v$ is speed in $\mathrm{m/s}$, and $A$ is cross-sectional area in $\mathrm{m^2}$. Does the right side have force dimensions?

**Türkçe — nasıl başlamalı?** Hızın hem sayısının hem biriminin karesi alınır. Aynı tabanın üslerini toplayarak metre ve saniye üslerini sadeleştir.

<details><summary>Answer and steps / Yanıt ve işlemler</summary>

$$[C\rho v^2A]=(ML^{-3})(L^2\mathsf T^{-2})(L^2)=ML^{-3+2+2}\mathsf T^{-2}=ML\mathsf T^{-2}.$$
In units this is $\mathrm{kg\,m\,s^{-2}}=\mathrm N$. Dimensional consistency does not establish the value of $C$.

</details>

### L1 -- Module 02 P4: Kinematic Equation Selection

A train moving at $30.0\,\mathrm{m/s}$ applies brakes and decelerates uniformly at $1.50\,\mathrm{m/s^2}$. How far does it travel before stopping?

<summary>Answer / Kısa kontrol</summary>

### L1 -- Module 03 P2: Horizontal Launch

A marble rolls off the edge of a table $1.20\,\mathrm m$ above the floor with a horizontal speed of $2.50\,\mathrm{m/s}$. (a) How long does it take to hit the floor? (b) How far from the base of the table does it land?

### L1 -- Module 04 P1: Newton's Second Law

A net force of $24.0\,\mathrm N$ acts on a $6.0\,\mathrm{kg}$ object initially at rest. (a) What is the acceleration? (b) What is the speed after $5.0\,\mathrm s$? (c) How far has the object moved in $5.0\,\mathrm s$?

**Türkçe — nasıl düşünmeli:** Verilen kuvvet zaten net kuvvettir. Önce ivmeyi bul, ardından sabit ivmeli hareket denklemleriyle aynı süre içindeki hız ve yolu hesapla.

<details>
<summary>Answer / Kısa kontrol</summary>

(a) $a=\frac{F_{\rm net}}{m}=24.0/6.0=4.0\,\mathrm{m/s^2}$. (b) $v=at=(4.0)(5.0)=20.0\,\mathrm{m/s}$. (c) $\Delta x=\frac12at^2=\frac12(4.0)(25)=50.0\,\mathrm m$.

</details>

### L1 -- Module 06 P1: Work by a Constant Force

A worker pushes a $40.0\,\mathrm{kg}$ crate $8.0\,\mathrm m$ across a floor by applying a $200\,\mathrm N$ force at $25^\circ$ below the horizontal. How much work does the worker do on the crate?

**Türkçe problem açıklaması:** $40.0\,\mathrm{kg}$ kütleli kasa yatayda $8.0\,\mathrm m$ ilerlerken işçi yatayın $25^\circ$ altında $200\,\mathrm N$ kuvvet uyguluyor. İşçinin yaptığı işi bulun. Yer değiştirme yatay olduğundan yalnızca kuvvetin yatay bileşeni iş yapar; burada net işi değil, bu kuvvetin işini soruyoruz.

<details>
<summary>Answer</summary>

$$W=Fd\cos\theta=(200\,\mathrm N)(8.0\,\mathrm m)\cos25^\circ=1450.09\,\mathrm J\approx1.45\times10^3\,\mathrm J.$$

Use the unrounded cosine for the calculation. The force's vertical component does no work on the horizontal displacement.

</details>

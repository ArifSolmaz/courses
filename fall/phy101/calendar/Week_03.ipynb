## Read and work through the examples / Konu ve çözümlü örnekler

## Before calculating: two motions sharing one clock

**Physics first:** understand the diagram and solve the equations on paper. Use a trajectory plot to test your prediction; programming the plot is optional.

For ideal projectile motion, ignore air resistance, choose $+x$ horizontal and $+y$ upward, and write:
$$\begin{aligned}
v_{0x}&=v_0\cos\theta, & v_{0y}&=v_0\sin\theta,\\
x&=x_0+v_{0x}t, & y&=y_0+v_{0y}t-\frac12gt^2.
\end{aligned}$$
Horizontal acceleration is zero; vertical acceleration is $-g$. The two directions have different equations but **the same elapsed time**.

**A small algebra bridge:** for a horizontal launch from height $h$, $0=h-\frac12gt^2$. Add $\frac12gt^2$ to both sides, multiply by 2, divide by $g$, then take the positive square root:
$$\frac12gt^2=h\ \Rightarrow\ gt^2=2h\ \Rightarrow\ t^2=\frac{2h}{g}\ \Rightarrow\ t=\sqrt{\frac{2h}{g}}.$$
The units under the root are $\mathrm m/(\mathrm{m/s^2})=\mathrm{s^2}$, so the result is in seconds. Only after finding $t$ use $x=v_{0x}t$.

**When the shortcut works:** $T=\frac{2v_0\sin\theta}{g}$ and $R=\frac{v_0^2\sin(2\theta)}{g}$ assume landing and launch at the **same height**. For a cliff or hillside, use the position equations and the actual landing condition. At the peak, $v_y=0$; usually $v_x\ne0$.

**Two roots:** in $At^2+Bt+C=0$, keep the signs inside $t=\dfrac{-B\pm\sqrt{B^2-4AC}}{2A}$. Test each root against the time interval and the physical event; do not automatically discard the smaller root.

**Review stop:** two objects released horizontally at different speeds from the same height land simultaneously if both have zero initial vertical velocity. Explain this from the vertical equation before viewing a plot.

## 2. Theory: Projectile Motion

### 2.1 What is a Projectile?

A **projectile** moves under gravity after release; the launch push has ended. Our first model neglects air resistance and assumes constant $g=9.81\,\mathrm{m/s^2}$. Take $+x$ horizontal, $+y$ upward, and measure $\theta$ above $+x$. The initial point $(x_0,y_0)$ need not be the ground.

Positions have units $\mathrm m$, velocities $\mathrm{m/s}$, accelerations $\mathrm{m/s^2}$, and time $\mathrm s$. Do not calculate different horizontal and vertical flight times for one landing event.

**Small example — build both coordinates.** Release a ball from $(0,2\,\mathrm m)$ with $v_0=10\,\mathrm{m/s}$ at $30^\circ$. After $0.50\,\mathrm s$:

$$
\begin{aligned}
v_{0x}&=10\cos30^\circ=8.660\,\mathrm{m/s}, &v_{0y}&=10\sin30^\circ=5.000\,\mathrm{m/s},\\
x&=0+(8.660)(0.50)=4.330\,\mathrm m,\\
y&=2+(5.000)(0.50)-\tfrac12(9.81)(0.50)^2\\
 &=2+2.500-1.22625=3.27375\,\mathrm m,\\
v_y&=5.000-9.81(0.50)=0.095\,\mathrm{m/s}.
\end{aligned}
$$

Square time before multiplying the gravity term. The ball is almost at its apex but is still rising slightly. At the apex $v_y=0$, while $v_x=8.660\,\mathrm{m/s}$ and $a_y=-g$ remain: neither the full velocity nor acceleration is zero.

For a cliff or an inclined landing surface, solve the actual vertical landing equation for $t$ first, then use $x=x_0+v_{0x}t$.

### 2.2 Key Insight — Independence of Motion

A dropped ball and a horizontally launched ball released together from the **same height** have the same $y_0$, $v_{0y}=0$, and $a_y=-g$. Their vertical equations are identical, so they land together on level ground in this ideal model. Horizontal speed changes where they land, not when.

**Think–Pair–Explain:** From $h=4.905\,\mathrm m$, one ball is dropped and one is launched horizontally at $3\,\mathrm{m/s}$. Predict their flight times and horizontal distances. **Model reasoning:** $0=h-\frac12gt^2$ gives $t=\sqrt{\frac{2h}{g}}=1.00\,\mathrm s$ for both. Distances are zero and $(3)(1)=3.00\,\mathrm m$.

**Türkçe:** Bağımsızlık “ayrı zamanlar” demek değildir. Düşey başlangıç koşulları aynıysa uçuş süreleri aynıdır; yatay fark sadece menzilde görünür.

### 2.3 Trajectory Equation (Eliminating Time)

For $v_{0x}\ne0$, subtract $x_0$ and divide by $v_{0x}$:

$$
\begin{aligned}
t&=\frac{x-x_0}{v_{0x}},\\
y-y_0&=v_{0y}\frac{x-x_0}{v_{0x}}-\frac g2\left(\frac{x-x_0}{v_{0x}}\right)^2\\
&=(x-x_0)\tan\theta-\frac{g(x-x_0)^2}{2v_0^2\cos^2\theta}.
\end{aligned}
$$

Setting $x_0=y_0=0$ gives the usual parabola equation. A vertical launch has $v_{0x}=0$; division by it is invalid, so keep the time equations instead.

**Türkçe:** Zamanı yok ederken yerine yazdığın kesrin tamamının karesini al. Başlangıç noktası sıfır değilse $x$ yerine $x-x_0$ ve $y$ yerine $y-y_0$ gerekir.

## 8. Worked Examples

### Example 1: Football Kick

A football is kicked with $v_0=20\,\mathrm{m/s}$ at $\theta=37^\circ$. Find:
- (a) Time of flight
- (b) Maximum height
- (c) Range

### Hand calculation

Assume landing at launch height and no air resistance. Choose horizontal forward and vertical up positive. First split the initial velocity:

$$\begin{aligned}
v_{0x}&=v_0\cos\theta=(20)\cos37^\circ=15.9727\,\mathrm{m/s},\\
v_{0y}&=v_0\sin\theta=(20)\sin37^\circ=12.0363\,\mathrm{m/s}.
\end{aligned}$$

**(a) Find the shared flight time $T$.** Landing has zero vertical displacement:

$$\begin{aligned}
0&=v_{0y}t-\frac12gt^2=t\left(v_{0y}-\frac12gt\right),\\
T&=\frac{2v_{0y}}{g}=\frac{2(12.0363\,\mathrm{m/s})}{9.81\,\mathrm{m/s^2}}=2.45388\,\mathrm s.
\end{aligned}$$

Factoring gives the launch root $t=0$ and the later landing root $T$; use the latter.

**(b) Find rise $H$ by setting $v_y=0$ at the peak:**

$$\begin{aligned}
0&=v_{0y}^2-2gH,\\
H&=\frac{v_{0y}^2}{2g}=\frac{(12.0363\,\mathrm{m/s})^2}{19.62\,\mathrm{m/s^2}}=7.3839\,\mathrm m.
\end{aligned}$$

Add $2gH$ to both sides, then divide by $2g$.

**(c) Use that same flight time in the horizontal equation:**

$$R=v_{0x}T=(15.9727\,\mathrm{m/s})(2.45388\,\mathrm s)=39.1952\,\mathrm m.$$

**Check:** Equal launch/landing heights give equal up/down times. At the peak the ball still moves horizontally at $15.9727\,\mathrm{m/s}$.

**Türkçe:** Önce iki hız bileşenini ayır. Düşey denklem süreyi verir; menzilde aynı süreyi kullanırız. Tepe için yalnızca düşey hız sıfırlanır. Ara sayıları hesap makinesinde tam hassasiyetle sakla.

### Example 2: Cliff Launch

A ball is thrown horizontally from a cliff $45\,\mathrm m$ high with $v_0=15\,\mathrm{m/s}$. How far from the base does it land?

### Hand calculation

Set ground level to $y=0$ and choose upward positive. The release conditions are $y_0=45\,\mathrm m$, $v_{0y}=0$, and $v_{0x}=15\,\mathrm{m/s}$. Set $x_0=0$ at the cliff.

**The vertical equation gives time:**

$$\begin{aligned}
0&=y_0-\frac12gt^2,\\
\frac12gt^2&=y_0 &&\text{add the gravity term to both sides},\\
t^2&=\frac{2y_0}{g} &&\text{multiply by }2/g,\\
t&=\sqrt{\frac{2(45\,\mathrm m)}{9.81\,\mathrm{m/s^2}}}=3.02891\,\mathrm s.
\end{aligned}$$

Choose the positive square root because we want time after release. Now use the horizontal equation:

$$x=v_{0x}t=(15\,\mathrm{m/s})(3.02891\,\mathrm s)=45.4337\,\mathrm m.$$

**Result:** The ball lands about $45.4\,\mathrm m$ from the cliff base.

**Check:** Horizontal speed does not appear in the fall-time equation. Doubling it doubles the range while leaving fall time unchanged.

**Türkçe:** İlk düşey hız sıfır olduğu için düşey denklem sadeleşir. Sürenin karesinden süreye geçerken pozitif karekökü al; sonra aynı süreyi yatay harekete taşı.

### 2.1 The Three Laws

Use an inertial reference frame and a constant mass. A free-body diagram contains forces acting **on the selected object**, not forces it exerts on others.

**Small example:** A $2\,\mathrm{kg}$ cart has a $7\,\mathrm N$ force right and a $3\,\mathrm N$ force left. With right positive:

$$\sum F_x=7-3=4\,\mathrm N,\qquad a_x=\frac{4\,\mathrm N}{2\,\mathrm{kg}}=2\,\mathrm{m/s^2}.$$

If both forces become $3\,\mathrm N$, the net force and acceleration are zero. A cart already moving right then continues right at constant velocity; zero net force does not mean zero velocity.

**Türkçe:** $ma$'ya yalnızca büyük kuvveti değil, işaretli toplamı eşitliyoruz. Etki–tepki kuvvetleri farklı cisimlere etki eder; aynı cismin diyagramında birbirlerini götüren bir çift değildir.

## A first force diagram: which interactions act on one object?

For today's Newton introduction, practise weight, support and an applied force. Draw only forces **on the chosen object**. A third-law partner acts on the other object.

A 2.0 kg cart on a horizontal frictionless track is pulled right by 6.0 N. Choose right and up positive. Vertical forces balance: $N-mg=0$, so $N=2.0(9.81)=19.62$ N. Horizontally,
$$\sum F_x=ma_x\Rightarrow6.0=2.0a_x\Rightarrow a_x=3.0\,\mathrm{m/s^2}.$$
From rest, after 2.0 s its velocity is $v_x=0+(3.0)(2.0)=6.0$ m/s. The force on the cart causes its acceleration. The opposite force **on the hand or rope** does not cancel it on the cart's diagram.

**Check:** With the pull removed, the ideal cart keeps its velocity; zero force means zero acceleration, not necessarily zero velocity. **TR:** Bu hafta önce tek cismin kuvvetlerini ayırıyoruz. Sürtünmenin türleri ve eğik düzlem uygulamaları gelecek haftanın konusudur.

### Worked example: A horizontal launch develops a downward velocity

**Problem.** A puck leaves the horizontal edge of a table at $28.0\,\mathrm{m/s}$. The tabletop is $2.00\,\mathrm{m}$ above the floor. Find the angle below horizontal of the puck's velocity just before impact. Ignore air resistance and use $g=9.80\,\mathrm{m/s^2}$.

**Picture and principle.** Draw the horizontal velocity to the right and vertical velocity downward at impact. These perpendicular arrows form a right triangle whose diagonal is the total velocity. A horizontal launch means $v_{y0}=0$; it does not mean the vertical velocity remains zero.

With rightward and upward chosen positive,

$$v_x=v_{x0}=28.0\,\mathrm{m/s},\qquad a_y=-g,\qquad\Delta y=-h.$$

We can obtain the impact vertical component without finding time. Start with

$$v_y^2=v_{y0}^2+2a_y\Delta y=0+2(-g)(-h)=2gh.$$

Taking a square root gives a magnitude. The downward direction determines the sign:

$$v_y=-\sqrt{2gh}=-\sqrt{2(9.80\,\mathrm{m/s^2})(2.00\,\mathrm{m})}=-6.261\,\mathrm{m/s}.$$

For the positive angle $\alpha$ measured below horizontal, use the velocity triangle:

$$\tan\alpha=\frac{|v_y|}{v_x}\quad\Rightarrow\quad\alpha=\tan^{-1}\left(\frac{6.261}{28.0}\right)=\boxed{12.6^\circ\text{ below horizontal}}.$$

**Meaning and check.** Horizontal velocity is much greater than downward velocity, so the impact direction should be shallow, not nearly vertical. An independent time calculation gives $t=\sqrt{2h/g}=0.639\,\mathrm{s}$, and then $v_y=-gt=-6.261\,\mathrm{m/s}$ again. The angle concerns the tangent to the trajectory at impact; it is not the angle of the straight line joining launch and landing points.

**Türkçe destek.** Karekök yalnızca büyüklüğü verir. Düşey hızın eksi işaretini cismin aşağı hareket etmesinden belirleriz. Açı yataydan ölçüldüğü için tanjantın payında düşey, paydasında yatay hız vardır. Yatay hareketin sürmesi, düşey hareketin başlamasını engellemez; ikisi aynı süre boyunca birlikte gerçekleşir.

### Worked example: An elevated launch needs the full vertical equation

**Problem.** A ball is launched at $20.0\,\mathrm{m/s}$ and $60.0^\circ$ above horizontal from a point $2.00\,\mathrm{m}$ above level ground. Determine the time until it reaches the ground. Ignore air resistance and use $g=9.80\,\mathrm{m/s^2}$.

**Picture and principle.** Draw the ground as $y=0$ and place the launch point at $y_0=2.00\,\mathrm{m}$. The ball first rises and then falls below its original height. Only the initial vertical component determines the flight time:

$$v_{y0}=v_0\sin\theta=(20.0\,\mathrm{m/s})\sin60.0^\circ=17.3205\ldots\,\mathrm{m/s}.$$

The vertical position is $y=y_0+v_{y0}t-\tfrac12gt^2$. At landing, set $y=0$ and rearrange into quadratic form:

$$\frac12gt^2-v_{y0}t-y_0=0.$$

Applying the quadratic formula before inserting values gives

$$t=\frac{v_{y0}\pm\sqrt{v_{y0}^2+2gy_0}}{g}.$$

The positive root describes the future landing event:

$$t=\frac{17.3205+\sqrt{17.3205^2+2(9.80)(2.00)}}{9.80}\,\mathrm{s}=\boxed{3.65\,\mathrm{s}}.$$

The other root is approximately $-0.112\,\mathrm{s}$. It belongs to a backward mathematical extension of the parabola and does not describe motion after release. Keeping extra digits gives $3.6467\,\mathrm{s}$ for the stated gravitational acceleration.

**Meaning and check.** Returning merely to launch height would take $2v_{y0}/g\approx3.53\,\mathrm{s}$. The ground is lower, so the actual flight must last longer; $3.65\,\mathrm{s}$ passes this check. Using $g=9.81\,\mathrm{m/s^2}$ instead gives about $3.64\,\mathrm{s}$, explaining the nearby rounded value in the source bank.

**Türkçe destek.** $2v_{y0}/g$ bağıntısı, başlangıç ve bitiş yükseklikleri eşitse geçerlidir. Burada top iki metre daha aşağıya indiği için bu kısa yol kullanılamaz. Denklemde $y_0$'ı koru; kök bulduktan sonra olayın $t\ge0$ koşulunu sağladığını kontrol et.

### Worked example: Choose the ascending root before finding horizontal distance

**Problem.** A rock is thrown from $2.00\,\mathrm{m}$ above the ground at $30.0\,\mathrm{m/s}$ and $40.0^\circ$ above horizontal. It reaches a window $18.0\,\mathrm{m}$ above the ground while still rising. Find the horizontal launch distance from the window. Ignore air resistance; use $g=9.80\,\mathrm{m/s^2}$.

**Picture and principle.** Place the origin at the launch point, with $+x$ toward the window and $+y$ upward. The required vertical displacement is $\Delta y=18.0-2.00=16.0\,\mathrm{m}$. Resolve the initial velocity:

$$v_{x0}=30.0\cos40.0^\circ=22.9813\,\mathrm{m/s},$$

$$v_{y0}=30.0\sin40.0^\circ=19.2836\,\mathrm{m/s}.$$

First solve the vertical motion for time; then use that same time horizontally:

$$\Delta y=v_{y0}t-\frac12gt^2\quad\Rightarrow\quad\frac12gt^2-v_{y0}t+\Delta y=0,$$

$$t=\frac{v_{y0}\pm\sqrt{v_{y0}^2-2g\Delta y}}{g}.$$

Numerical substitution produces two positive roots:

$$t=\frac{19.2836\pm\sqrt{19.2836^2-2(9.80)(16.0)}}{9.80}\,\mathrm{s},$$

$$t_1=1.18887\,\mathrm{s},\qquad t_2=2.74657\,\mathrm{s}.$$

“Still rising” selects the earlier root. With zero horizontal acceleration,

$$x=v_{x0}t_1=(22.9813\,\mathrm{m/s})(1.18887\,\mathrm{s})=\boxed{27.3\,\mathrm{m}}.$$

**Meaning and check.** At the selected time, $v_y=v_{y0}-gt_1\approx+7.63\,\mathrm{m/s}$, so the rock is indeed rising. The later root gives a downward vertical velocity and belongs to the descending crossing of the same height. Both roots solve the height equation, but only one satisfies the stated condition.

The highest point occurs at $v_{y0}/g\approx1.97\,\mathrm{s}$, between the two roots. This gives another way to distinguish the upward and downward crossings. 

**Türkçe destek.** Pencerenin yerden yüksekliği doğrudan yer değiştirme değildir: başlangıç yüksekliğini çıkarmalıyız. İki pozitif kökten birini sırf daha küçük olduğu için seçmiyoruz; “yükselirken” koşulunu $v_y>0$ ile kontrol ediyoruz. Yatay mesafeyi bulurken seçtiğimiz aynı zamanı kullanmak zorundayız.

### Worked example: Reconstruct a launch from its range and flight time

**Problem.** A projectile travels $76.2\,\mathrm{m}$ horizontally and returns to its launch height after $4.08\,\mathrm{s}$. Determine its initial speed. Air resistance is negligible. Use $g=9.80\,\mathrm{m/s^2}$ and choose the horizontal travel direction and upward direction as positive.

**Picture and principle.** Draw a parabolic trajectory whose two endpoints are at the same height. Instead of starting with launch speed and angle, reconstruct the two initial velocity components separately. The horizontal displacement determines $v_{x0}$; the zero net vertical displacement determines $v_{y0}$.

With no horizontal acceleration,

$$R=v_{x0}T\quad\Rightarrow\quad v_{x0}=\frac{R}{T}=\frac{76.2\,\mathrm{m}}{4.08\,\mathrm{s}}=18.6765\ldots\,\mathrm{m/s}.$$

The vertical displacement over the complete flight is zero:

$$0=v_{y0}T-\frac12gT^2.$$

Since $T\ne0$, divide both sides by $T$, then move the gravitational term to the other side:

$$v_{y0}=\frac12gT=\frac12(9.80\,\mathrm{m/s^2})(4.08\,\mathrm{s})=19.992\,\mathrm{m/s}.$$

These components are perpendicular, so their magnitudes are not added directly. Use Pythagoras:

$$v_0=\sqrt{v_{x0}^2+v_{y0}^2}=\sqrt{(18.6765)^2+(19.992)^2}\,\mathrm{m/s}=\boxed{27.4\,\mathrm{m/s}}.$$

**Meaning and check.** The initial velocity has slightly more upward than horizontal motion. Its direction would be $\tan^{-1}(19.992/18.6765)\approx46.9^\circ$ above horizontal. The time to maximum height, $v_{y0}/g=2.04\,\mathrm{s}$, is exactly half the total flight time, as expected for equal launch and landing heights. Reconstructed horizontal motion gives $v_{x0}T=76.2\,\mathrm{m}$.

### From stopping distance to braking force

**Problem.** A car of mass $1100\,\mathrm{kg}$ travels at $27.0\,\mathrm{m/s}$ and stops after $578\,\mathrm m$. Find the magnitude of the effective average braking force. Represent the braking by a constant horizontal force over this distance.

**Start from the physics.** Choose the car's original direction as positive. Its displacement is positive, but its acceleration is negative because its speed decreases. On a level road the vertical forces balance. The horizontal net force therefore determines the braking acceleration. We know distance rather than stopping time, so use the kinematic equation that does not contain time:

$$v_f^2=v_i^2+2a\Delta x.$$

**Rearrange before substituting.** Subtract $v_i^2$ from both sides, then divide both sides by $2\Delta x$:

$$\begin{aligned}
v_f^2-v_i^2&=2a\Delta x,\\
a&=\frac{v_f^2-v_i^2}{2\Delta x}
=\frac{0-(27.0\,\mathrm{m/s})^2}{2(578\,\mathrm m)}\\
&=-0.6306\,\mathrm{m/s^2}.
\end{aligned}$$

Now apply Newton's second law to the car, using its mass rather than its weight:

$$F_x=ma=(1100\,\mathrm{kg})(-0.6306\,\mathrm{m/s^2})=-694\,\mathrm N.$$

**Interpret the result.** The requested magnitude is $\boxed{694\,\mathrm N}$, approximately $690\,\mathrm N$ at two significant figures. The signed force points opposite the motion. A real braking force may vary; this constant-force model gives the equivalent force over the stated stopping distance, not a claim about its detailed time history.

**Sense check.** The unit $\mathrm{kg\,m/s^2}$ is a newton. Substitution gives $27.0^2+2(-0.6306)(578)\approx0$, consistent with stopping.

**Türkçe destek:** Eksi ivme, aracın geri gittiği anlamına gelmez; seçtiğimiz pozitif yöndeki hızın azaldığını gösterir. Önce hızın karesini al, sonra çıkarma yap. Kuvvetin büyüklüğü sorulduğu için son yanıtta $|-694|=694\,\mathrm N$ kullanılır.

*Source: 04_testbank.doc, Section 4.2 Problems, Question 11. Original numerical givens retained; restated as an open worked example.*

### A velocity reversal: subtract signed velocities

**Problem.** During a test, a $1.50\,\mathrm g$ bullet passes through a $28.0\,\mathrm{kg}$ block in $11.4\,\mathrm{ms}$. The block's velocity changes uniformly from $1.75\,\mathrm{m/s}$ rightward to $1.20\,\mathrm{m/s}$ leftward. Find the magnitude of the horizontal force exerted by the bullet on the block.

**Identify the system.** We are analysing the block, so Newton's second law uses the block's mass, $28.0\,\mathrm{kg}$. The bullet's mass is supplied by the bank but is not needed for this particular question. Take rightward as positive. The initial and final velocities must therefore have opposite signs:

$$v_i=+1.75\,\mathrm{m/s},\qquad v_f=-1.20\,\mathrm{m/s}.$$

Convert the short time interval before dividing:

$$\Delta t=11.4\,\mathrm{ms}=11.4\times10^{-3}\,\mathrm s=0.0114\,\mathrm s.$$

**Calculate acceleration explicitly.** A change means final minus initial, not the difference between two positive speed readings:

$$\begin{aligned}
\Delta v&=v_f-v_i=(-1.20)-(+1.75)=-2.95\,\mathrm{m/s},\\
a_x&=\frac{\Delta v}{\Delta t}
=\frac{-2.95\,\mathrm{m/s}}{0.0114\,\mathrm s}
=-258.77\,\mathrm{m/s^2}.
\end{aligned}$$

The horizontal interaction force is the horizontal net force on the block during the stated idealized test:

$$F_x=ma_x=(28.0)(-258.77)\,\mathrm N=-7.25\times10^3\,\mathrm N.$$

**Interpret and check.** Its magnitude is $\boxed{7.25\times10^3\,\mathrm N}$, directed leftward. This large value is reasonable because the reversal occurs in only about one hundredth of a second. The same change spread over twice the time would require half the force. Dividing the answer by the block mass recovers the acceleration.

**Türkçe destek:** $-1.20-1.75=-2.95$ işlemi önemlidir. Cisim önce sağa giderken durur, sonra sola hızlanır; bu yüzden hız değişiminin büyüklüğü iki hız büyüklüğünün toplamıdır. Ayrıca milisaniyeyi saniyeye çevirmeden hesap yapmak kuvveti bin kat küçük verir.

### Rocket thrust must overcome weight and accelerate

**Problem.** A rocket of mass $10{,}000\,\mathrm{kg}$ accelerates vertically upward at $2.00\,\mathrm{m/s^2}$ near Earth's surface. Neglect air resistance. Find the upward engine thrust, using $g=9.80\,\mathrm{m/s^2}$. Treat the given mass as constant during this calculation.

**Draw the force picture in words.** For the rocket as one system, thrust $T$ acts upward and weight $mg$ acts downward. Choose upward as the positive direction. These forces oppose one another, so the net force is their difference. The equation is not $T=ma$ because thrust is only one of the forces acting on the rocket:

$$\sum F_y=T-mg=ma.$$

**Isolate the unknown.** Add $mg$ to both sides. This cancels the negative weight term on the left:

$$\begin{aligned}
T-mg+mg&=ma+mg,\\
T&=ma+mg=m(a+g).
\end{aligned}$$

Now substitute compatible SI units:

$$\begin{aligned}
mg&=(10{,}000)(9.80)\,\mathrm N=98{,}000\,\mathrm N,\\
ma&=(10{,}000)(2.00)\,\mathrm N=20{,}000\,\mathrm N,\\
T&=98{,}000\,\mathrm N+20{,}000\,\mathrm N\\
&=\boxed{1.18\times10^5\,\mathrm N\ \text{upward}}.
\end{aligned}$$

**Understand the two contributions.** The first $98{,}000\,\mathrm N$ balances gravity. The additional $20{,}000\,\mathrm N$ produces the specified upward acceleration. The rocket does not need its thrust to equal its weight unless its vertical acceleration is zero.

**Sense check.** Subtracting the weight from the thrust gives $20{,}000\,\mathrm N$; dividing by $10{,}000\,\mathrm{kg}$ returns $2.00\,\mathrm{m/s^2}$. If the desired acceleration approached zero, the formula would become $T=mg$, as expected for hovering in this simple model.

**Türkçe destek:** $ma$ bir kuvvetin adı değil, net kuvvetin değeridir. $-mg$ terimini diğer tarafa geçirirken işaretinin artıya dönüşmesi, aslında denklemin iki tarafına da $mg$ eklememizden kaynaklanır. İtki, hem ağırlığı dengelemeli hem de yukarı yönlü ivme sağlamalıdır.

### Infer a force from a measured travel time

**Problem.** A $50\,\mathrm{kg}$ box starts from rest on a frictionless horizontal surface. A constant horizontal force makes it travel $28\,\mathrm m$ in $6.9\,\mathrm s$. Find the force magnitude.

**Connect motion to force in two stages.** The measured distance and time describe the motion; they do not directly give a force. First find the acceleration, then use Newton's second law. Because the force and mass are constant, the horizontal acceleration is constant. The box starts from rest, so $v_0=0$. Choose the force direction as positive and set the starting position to zero.

The displacement equation simplifies to

$$\Delta x=v_0t+\frac12at^2=\frac12at^2.$$

**Do the algebra before using numbers.** Multiplying both sides by $2$ removes the fraction; dividing both sides by $t^2$ isolates $a$:

$$\begin{aligned}
2\Delta x&=at^2,\\
a&=\frac{2\Delta x}{t^2}
=\frac{2(28\,\mathrm m)}{(6.9\,\mathrm s)^2}\\
&=\frac{56}{47.61}\,\mathrm{m/s^2}
=1.1762\,\mathrm{m/s^2}.
\end{aligned}$$

On the frictionless surface, weight and normal force cancel vertically, leaving the applied force as the horizontal net force:

$$F=ma=(50\,\mathrm{kg})(1.1762\,\mathrm{m/s^2})
=\boxed{58.8\,\mathrm N\approx59\,\mathrm N}.$$

**Sense check.** The average speed is $28/6.9=4.06\,\mathrm{m/s}$. With constant acceleration from rest, the final speed should be twice this value: $8.12\,\mathrm{m/s}$. Calculating $at$ gives the same result. This is an independent check on the factor of two in the displacement equation.

**Türkçe destek:** Zamanın karesi $(6.9)^2=47.61$ alınmalıdır; $2t$ ile $t^2$ aynı şey değildir. Ortalama hız $\Delta x/t$, başlangıçtan sona hız değişimi değildir. Önce ivmeyi bulup sonra kütleyle çarpmak, kinematik ile dinamiği açık biçimde birbirine bağlar.

*Source: 04_testbank.doc, Section 4.2 Problems, Question 16. Original numerical givens retained; restated as an open worked example.*

### An angled pull changes both acceleration and support force

**Problem.** A worker pulls a $40.0\,\mathrm{kg}$ tool box across a smooth horizontal floor with a $100\,\mathrm N$ force directed $37.0^\circ$ above horizontal. Describe its free-body diagram, calculate its horizontal acceleration, and find the normal force from the floor. Use $g=9.80\,\mathrm{m/s^2}$.

**Start with three forces.** The box has weight $mg$ downward, normal force $N$ upward, and the applied pull $F$ tilted upward and forward. Smooth means friction is neglected. Choose horizontal $+x$ forward and vertical $+y$ upward. The angle is measured from horizontal, so the adjacent component uses cosine:

$$F_x=F\cos\theta,\qquad F_y=F\sin\theta.$$

**Horizontal equation.** Only the horizontal part of the pull accelerates the box along the floor:

$$\begin{aligned}
F\cos\theta&=ma_x,\\
a_x&=\frac{F\cos\theta}{m}
=\frac{(100\,\mathrm N)\cos37.0^\circ}{40.0\,\mathrm{kg}}\\
&=\boxed{2.00\,\mathrm{m/s^2}}.
\end{aligned}$$

**Vertical equation.** The box stays on the floor rather than accelerating upward. Therefore $a_y=0$, even though the pull has a vertical component:

$$\begin{aligned}
N+F\sin\theta-mg&=0,\\
N&=mg-F\sin\theta\\
&=(40.0)(9.80)-(100)\sin37.0^\circ\ \mathrm N\\
&=392-60.18=\boxed{332\,\mathrm N}.
\end{aligned}$$

**Interpret and check.** The worker partially supports the box, so the floor supplies less than the full weight. Our result is positive and smaller than $392\,\mathrm N$, consistent with continued contact. In the limiting case of a horizontal pull, $\theta=0$, the formulas give $a_x=F/m$ and $N=mg$.

**Türkçe destek:** Sinüs ve kosinüs seçimini ezberlemek yerine açının hangi eksenden ölçüldüğünü kontrol et. Burada yatay bileşen açıya komşudur. Dikey ivmenin sıfır olması dikey kuvvetlerin tek tek sıfır olduğu anlamına gelmez; toplamları sıfırdır. Bu nedenle normal kuvveti doğrudan $mg$ almak yanlış olur.

```python
#@title Run once — prepare compact demonstration controls
"""Embedded in the lecture notebooks by refresh_notebook_interface.py.

Only standard ipywidgets and IPython are used; notebooks remain self-contained.
"""
import functools
import traceback
import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import display, HTML
from IPython.utils.capture import capture_output
from ipywidgets import interactive as _PhysicsInteractive
from ipywidgets.widgets.interaction import show_inline_matplotlib_plots

# Set this explicitly: otherwise a local kernel can choose a desktop backend
# and block at plt.show(), leaving the notebook apparently frozen.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

try:
    from google.colab import output as _colab_output
except ImportError:
    pass
else:
    _colab_output.enable_custom_widget_manager()

_physics_panels = []
_physics_callback_errors = []


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; }
.phy101-controls { padding: 0 0 4px; box-sizing: border-box; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-animation-panel { max-width: 100%; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1;
  background: white; color: #172433; padding: 4px; }
@media (max-width: 650px) {
  .phy101-panel { flex-direction: column !important; }
  .phy101-controls, .phy101-plot-output { width: 100% !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Fit full labels and postpone costly redraws until a slider is released."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "continuous_update"):
            control.continuous_update = False
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "310px"
        control.layout.flex = "0 1 310px"
        control.layout.max_width = "100%"
        control.layout.min_width = "0"
        control.layout.margin = "2px 6px 2px 0"
        if isinstance(control, widgets.Button):
            control.layout.width = "auto"
            control.layout.flex = "0 0 auto"
    # Include the style in each output: Colab may isolate output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.Box([style] + flat, layout=widgets.Layout(
        display="flex", flex_flow="row wrap", align_items="center",
        width="100%", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        width="100%", min_width="0", max_width="100%",
        height="auto", overflow="visible", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Keep a compact control toolbar directly above the complete result."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="column",
                              align_items="stretch", width="100%"))
    panel.add_class("phy101-panel")
    return panel


class _PhysicsStableInteractive(_PhysicsInteractive):
    """Replace one persistent output model only after a redraw has completed.

    Sending clear_output followed by an inline figure depends on callback message
    capture in the frontend. Explicit MIME records keep repeated slider changes
    visible in Jupyter/Colab, without clearing the surrounding controls.
    """
    def update(self, *args):
        self.kwargs = {}
        if self.manual:
            self.manual_button.disabled = True
        try:
            for control in self.kwargs_widgets:
                self.kwargs[control._kwarg] = control.get_interact_value()
            with capture_output(stdout=True, stderr=True, display=True) as captured:
                self.result = self.f(**self.kwargs)
                show_inline_matplotlib_plots()
                if self.auto_display and self.result is not None:
                    display(self.result)
            records = []
            if captured.stdout:
                records.append({"output_type": "stream", "name": "stdout", "text": captured.stdout})
            if captured.stderr:
                records.append({"output_type": "stream", "name": "stderr", "text": captured.stderr})
            records.extend({"output_type": "display_data", "data": output.data,
                            "metadata": output.metadata or {}} for output in captured.outputs)
            # Keep the old result visible during calculation; publish the complete
            # replacement as one widget-state update, including its PNG/HTML data.
            self.out.outputs = tuple(records)
        except Exception:
            _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), traceback.format_exc()))
            self.out.outputs = ({"output_type": "stream", "name": "stderr",
                                 "text": traceback.format_exc()},)
        finally:
            if self.manual:
                self.manual_button.disabled = False


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


def physics_interactive(function, **controls):
    """An interactive widget with the same arguments as ipywidgets.interactive."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        try:
            return function(*args, **kwargs)
        except Exception as error:
            _physics_callback_errors.append((function.__name__, repr(error)))
            raise

    panel = _PhysicsStableInteractive(checked, {"auto_display": True}, **controls)
    inputs = [child for child in panel.children if child is not panel.out]
    panel.children = (_physics_controls(inputs), _physics_output(panel.out))
    panel.layout = widgets.Layout(display="flex", flex_flow="column",
                                  align_items="stretch", width="100%")
    panel.add_class("phy101-panel")
    _physics_panels.append(panel)
    return panel


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel
    display(panel)
    return function
```

### Visual check 1: 4. Interactive Demo 2 — Range vs Launch Angle Optimizer

Use the controls below. Predict → adjust one value → explain the result. / Tahmin et → tek değeri değiştir → sonucu açıkla.

```python
#@title Run the demonstration — predict, adjust, observe
def range_vs_angle(v0=30):
    """Interactive range-vs-angle plot."""
    angles = np.linspace(0, 90, 500)
    ranges = v0**2 * np.sin(2 * np.radians(angles)) / g

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.plot(angles, ranges, 'b-', linewidth=2)
    ax.set_xlabel('Launch Angle (°)')
    ax.set_ylabel('Range (m)')
    ax.set_title(f'Range vs Launch Angle  (v₀ = {v0} m/s)')

    # Highlight 45°
    R_max = v0**2 / g
    ax.axvline(45, color='r', linestyle='--', alpha=0.5)
    ax.plot(45, R_max, 'r*', markersize=18, label=f'Max range = {R_max:.1f} m at 45°')

    # Show complementary angles give same range
    for a in [30, 60]:
        R_a = v0**2 * np.sin(2 * np.radians(a)) / g
        ax.plot(a, R_a, 'go', markersize=10, label=f'{a}°: range {R_a:.1f} m')

    ax.legend(fontsize=10, loc='upper center', bbox_to_anchor=(0.5, -0.18), ncol=2)
    plt.tight_layout(rect=[0, 0.10, 1, 1])
    plt.show()


@physics_interact(v0=widgets.IntSlider(value=30, min=5, max=80, step=5,
                                       description='v₀ (m/s):'))
def _range_plot(v0):
    range_vs_angle(v0)
```

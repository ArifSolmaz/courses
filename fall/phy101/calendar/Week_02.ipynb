**Choose by the missing quantity.** If time is not needed, use $v^2=v_0^2+2a\Delta x$. If final velocity is not needed, use $\Delta x=v_0t+\frac12at^2$. These equations require constant acceleration during the interval.

**Algebra slowly:** isolate elapsed time by applying the same operation to both sides:

$$\begin{aligned}
v&=v_0+at,\\
v-v_0&=at &&\text{subtract }v_0,\\
t&=\frac{v-v_0}{a} &&\text{divide by }a\ne0.
\end{aligned}$$

For a car slowing from $12\,\mathrm{m/s}$ to rest with $a=-3\,\mathrm{m/s^2}$,

$$t=\frac{0-12\,\mathrm{m/s}}{-3\,\mathrm{m/s^2}}=4\,\mathrm s.$$

Both numerator and denominator are negative, so the elapsed time is positive. The unit quotient is seconds.

**Signs come from axes.** With upward positive, free-fall acceleration is $a_y=-9.81\,\mathrm{m/s^2}$ both while rising and while falling. At the highest point, velocity is momentarily zero but acceleration is still downward. Speed is the nonnegative magnitude of velocity.

**Square roots:** $t^2=9\,\mathrm{s^2}$ has mathematical roots $\pm3\,\mathrm s$. A time measured after release uses $+3\,\mathrm s$. A quadratic height equation can instead have two valid positive roots, for the upward and downward crossings.

**Review stop:** explain why $v=-5\,\mathrm{m/s}$ and $a=-2\,\mathrm{m/s^2}$ describe increasing speed in the negative direction. After one second, $v=-5+(-2)(1)=-7\,\mathrm{m/s}$: speed rose from $5$ to $7\,\mathrm{m/s}$. Acceleration with a minus sign does not always mean slowing down.

### Important Distinctions

- **Displacement vs Distance:** Walking $3\,\mathrm m$ east and $3\,\mathrm m$ west gives $\Delta x=0$ and $d=6\,\mathrm m$. No motion gives zero distance, so distance is not *always positive*.
- **Speed vs Velocity:** Speed is $|v|\ge0$; velocity includes direction.
- **Velocity vs Acceleration:** Velocity describes change of position; acceleration describes change of velocity. A negative acceleration does not by itself mean slowing down.

**Small example — one trip, two averages.** A robot moves from $x=1\,\mathrm m$ to $x=5\,\mathrm m$, then back to $x=3\,\mathrm m$, in $6\,\mathrm s$ total.

$$
\begin{aligned}
\Delta x&=3-1=2\,\mathrm m, & d&=(5-1)+(5-3)=6\,\mathrm m,\\
\bar v&=\frac{2\,\mathrm m}{6\,\mathrm s}=0.333\,\mathrm{m/s}, & \text{average speed}&=\frac{6\,\mathrm m}{6\,\mathrm s}=1.00\,\mathrm{m/s}.
\end{aligned}
$$

**Türkçe:** Yer değiştirmede yalnızca başlangıç ve son konumu çıkarıyoruz. Toplam yolda geri dönüş bölümünü de pozitif uzunluk olarak ekliyoruz. Aynı yolculuğun ortalama hız vektörü ile ortalama sürati bu nedenle farklı olabilir.

**Worked selection:** A cart slows from $6\,\mathrm{m/s}$ to rest with $a=-2\,\mathrm{m/s^2}$. Find its displacement without finding time first. Equation 3 avoids time:

$$
\begin{aligned}
0&=(6\,\mathrm{m/s})^2+2(-2\,\mathrm{m/s^2})\Delta x,\\
(4\,\mathrm{m/s^2})\Delta x&=36\,\mathrm{m^2/s^2} &&\text{add }(4\,\mathrm{m/s^2})\Delta x\text{ to both sides},\\
\Delta x&=\frac{36\,\mathrm{m^2/s^2}}{4\,\mathrm{m/s^2}}=9\,\mathrm m &&\text{divide both sides by }4\,\mathrm{m/s^2}.
\end{aligned}
$$

**Türkçe:** Denklem seçerken bilinmeyen sayısını azaltıyoruz. İvme negatif olsa da araba durana kadar ileri gittiği için yer değiştirme pozitiftir. Birimleri de böl: $(\mathrm{m^2/s^2})/(\mathrm{m/s^2})=\mathrm m$.

### Sign Convention

With right/up positive, $v>0$ means right/up and $v<0$ means left/down. When $v$ and $a$ have the same sign, speed increases; opposite signs mean speed decreases until a possible reversal. At $v=0$, inspect what acceleration does immediately afterward.

## Worked Example 1: Constant Acceleration

**Problem:** A car starts from rest and accelerates uniformly at $3\,\mathrm{m/s^2}$.
- (a) What is its speed after $5\,\mathrm s$?
- (b) How far has it traveled in $5\,\mathrm s$?
- (c) How far has it traveled when it reaches $30\,\mathrm{m/s}$?

### Hand calculation

Choose forward positive. The initial velocity is $v_0=0$ and acceleration is $a=+3\,\mathrm{m/s^2}$. Let $\Delta x$ denote displacement from the starting point.

**(a) Substitute into the velocity equation:**

$$\begin{aligned}
v&=v_0+at,\\
 &=0+(3\,\mathrm{m/s^2})(5\,\mathrm s)=15\,\mathrm{m/s}.
\end{aligned}$$

**(b) Square time before multiplying:**

$$\begin{aligned}
\Delta x&=v_0t+\frac12at^2,\\
 &=0+\frac12(3\,\mathrm{m/s^2})(5\,\mathrm s)^2,\\
 &=\frac12(3)(25)\,\mathrm m=37.5\,\mathrm m.
\end{aligned}$$

The unit cancellation is $(\mathrm{m/s^2})(\mathrm{s^2})=\mathrm m$.

**(c) Choose the equation without time.** This is a new final event, when $v=30\,\mathrm{m/s}$:

$$\begin{aligned}
v^2&=v_0^2+2a\Delta x,\\
(30\,\mathrm{m/s})^2&=0+2(3\,\mathrm{m/s^2})\Delta x,\\
\Delta x&=\frac{900\,\mathrm{m^2/s^2}}{6\,\mathrm{m/s^2}}=150\,\mathrm m.
\end{aligned}$$

The last step divides both sides by $6\,\mathrm{m/s^2}$. This event occurs at $t=v/a=10\,\mathrm s$.

**Check:** Doubling elapsed time from rest at constant acceleration quadruples distance: $150=4(37.5)$.

**Türkçe:** (c) şıkkında süre artık beş saniye değildir. Yeni son hıza ait denklemi kur; sonra iki tarafı aynı ivme terimine böl. Yolun birimi metre kalmalıdır.

## Worked Example 2: Free Fall

**Problem:** A ball is thrown straight up from the ground with $v_0=20\,\mathrm{m/s}$.
- (a) How high does it go?
- (b) How long until it returns to the ground?
- (c) What is its velocity when it hits the ground?

### Hand calculation with one sign convention

Take upward positive, $y_0=0$, $v_0=+20\,\mathrm{m/s}$, and $a_y=-g=-9.81\,\mathrm{m/s^2}$. Neglect air resistance. Let $H$ be the rise above the launch point.

**(a) At the peak, set velocity to zero:**

$$\begin{aligned}
0&=v_0^2-2gH,\\
2gH&=v_0^2 &&\text{add }2gH\text{ to both sides},\\
H&=\frac{v_0^2}{2g} &&\text{divide both sides by }2g,\\
 &=\frac{(20\,\mathrm{m/s})^2}{2(9.81\,\mathrm{m/s^2})}=20.387\,\mathrm m.
\end{aligned}$$

**(b) Return means the height is zero again.** Factor the time equation:

$$\begin{aligned}
0&=v_0t-\frac12gt^2=t\left(v_0-\frac12gt\right),\\
t&=0\quad\text{or}\quad v_0-\frac12gt=0,\\
t_{\rm return}&=\frac{2v_0}{g}=\frac{2(20\,\mathrm{m/s})}{9.81\,\mathrm{m/s^2}}=4.0775\,\mathrm s.
\end{aligned}$$

The zero root is the launch event; the positive root is the return event.

**(c) Use the signed velocity equation:**

$$v_y=v_0-gt_{\rm return}=20-9.81(4.0775)\approx-20.0\,\mathrm{m/s}.$$

The negative sign means downward. Impact **speed** is the magnitude, $20.0\,\mathrm{m/s}$.

**Check:** $t_{\rm peak}=\frac{v_0}{g}=2.0387\,\mathrm s$ is half the full flight because launch and landing heights match. Gravity remains downward throughout.

**Türkçe:** Tepe noktasında $v_y=0$ koyuyoruz, $a_y=0$ değil. Çarpanlara ayırma iki farklı olayı görünür yapar: atış ve yere dönüş. Son hızdaki eksi, süratin değil yönün bilgisidir.

## Worked Example 3: Braking Problem

**Problem:** A driver going $100\,\mathrm{km/h}$ sees a child $80\,\mathrm m$ ahead. Reaction time is $1\,\mathrm s$, and braking deceleration has magnitude $8\,\mathrm{m/s^2}$. Does the car stop in time?

### Hand calculation in two phases

Choose forward positive. First convert the initial speed; the conversion ratios cancel kilometres and hours:

$$v_0=100\,\frac{\mathrm{km}}{\mathrm h}\frac{1000\,\mathrm m}{1\,\mathrm{km}}\frac{1\,\mathrm h}{3600\,\mathrm s}=27.7778\,\mathrm{m/s}.$$

**Reaction phase:** Let $d_r$ be the distance before braking. Acceleration is zero during $t_r=1.0\,\mathrm s$:

$$d_r=v_0t_r=(27.7778\,\mathrm{m/s})(1.0\,\mathrm s)=27.7778\,\mathrm m.$$

**Braking phase:** Let $d_b$ be braking distance. Here $a=-8\,\mathrm{m/s^2}$ and final $v=0$:

$$\begin{aligned}
0&=v_0^2+2ad_b,\\
0&=771.605\,\mathrm{m^2/s^2}-(16\,\mathrm{m/s^2})d_b,\\
(16\,\mathrm{m/s^2})d_b&=771.605\,\mathrm{m^2/s^2},\\
d_b&=\frac{771.605\,\mathrm{m^2/s^2}}{16\,\mathrm{m/s^2}}=48.2253\,\mathrm m.
\end{aligned}$$

Add the braking term to both sides, then divide by its positive coefficient. Finally add the two distances:

$$\begin{aligned}
d_{\rm stop}&=d_r+d_b=27.7778+48.2253=76.0031\,\mathrm m,\\
\text{remaining distance}&=80-76.0031=3.9969\,\mathrm m.
\end{aligned}$$

**Result:** The car stops about $4.00\,\mathrm m$ before the child within the stated model. Keep unrounded values in the calculation.

**Check:** Count reaction distance as well as braking distance. The small positive margin depends on the supplied assumptions.

**Türkçe:** Tepki sırasında araç ilerlemeye devam eder. Önce iki yol parçasını ayrı hesapla, sonra topla. Negatif ivmeyi denkleme işaretiyle yaz; pozitif durma yolunu bulmak için terimleri iki tarafa tutarlı taşı.

### Worked example: Average velocity and average speed describe different things

**Problem.** A walker travels south at $2.00\,\mathrm{m/s}$ for $60.0\,\mathrm{min}$, then turns and walks $3000\,\mathrm{m}$ north in $25.0\,\mathrm{min}$. Find her average velocity and average speed over the complete trip. Include the direction of the average velocity.

**Picture and principle.** Draw one north–south axis with north positive. The first displacement is negative; the return displacement is positive. Average velocity uses signed displacement, while average speed uses total path length:

$$\bar v=\frac{\Delta x_1+\Delta x_2}{t_1+t_2},\qquad\overline{\text{speed}}=\frac{d_1+d_2}{t_1+t_2}.$$

Convert both durations before combining them with speeds in metres per second:

$$t_1=60.0(60)=3600\,\mathrm{s},\quad t_2=25.0(60)=1500\,\mathrm{s},\quad T=5100\,\mathrm{s}.$$

For the first leg, rearranging $v=\Delta x/t$ gives $\Delta x=vt$. Hence

$$\Delta x_1=(-2.00\,\mathrm{m/s})(3600\,\mathrm{s})=-7200\,\mathrm{m},$$

$$\Delta x_2=+3000\,\mathrm{m},\qquad\Delta x=-4200\,\mathrm{m}.$$

The complete-trip results are

$$\bar v=\frac{-4200\,\mathrm{m}}{5100\,\mathrm{s}}=-0.823529\ldots\,\mathrm{m/s},$$

$$\overline{\text{speed}}=\frac{7200+3000}{5100}\,\mathrm{m/s}=\boxed{2.00\,\mathrm{m/s}}.$$

The average velocity is therefore $\boxed{0.824\,\mathrm{m/s}\text{ south}}$. The negative sign simply records the chosen north-positive axis.

**Meaning and check.** She finishes south of her starting point even though she spends part of the trip moving north. Her return-leg speed is $3000/1500=2.00\,\mathrm{m/s}$, so a total average speed of $2.00\,\mathrm{m/s}$ is consistent. Also, average speed must be at least the magnitude of average velocity: $2.00>0.824$.

**Türkçe destek.** Ortalama hızın payında yer değiştirme vardır; dönüşte yapılan yol önceki yer değiştirmenin bir bölümünü götürür. Ortalama süratte ise iki yol uzunluğu da pozitiftir. “Güneye doğru” bilgisini ya negatif işaretle ya yön sözcüğüyle göster; ikisini fiziksel olarak çelişen biçimde kullanma.

### Worked example: A dragster: final speed is twice the average speed

**Problem.** A dragster starts from rest and covers a quarter mile in $6.70\,\mathrm{s}$ with constant acceleration. Find its velocity at the finish line, reporting its speed in miles per hour. Take the forward track direction as positive and use $1\,\mathrm{mi}=1609.344\,\mathrm{m}$.

**Picture and principle.** On a velocity–time graph, the motion is a straight line from zero to $v_f$. The area under this line is a triangle and equals displacement:

$$d=\frac12tv_f.$$

The same relation follows from $d=\tfrac12at^2$ and $v_f=at$. Rearrange before substituting: multiply by two and divide by the elapsed time,

$$v_f=\frac{2d}{t}.$$

Convert the quarter-mile distance to SI units:

$$d=\frac14(1609.344\,\mathrm{m})=402.336\,\mathrm{m}.$$

Then calculate the finishing velocity:

$$v_f=\frac{2(402.336\,\mathrm{m})}{6.70\,\mathrm{s}}=120.100\ldots\,\mathrm{m/s}.$$

To express the speed in miles per hour, cancel metres and seconds using conversion fractions:

$$v_f=120.100\ldots\frac{\mathrm{m}}{\mathrm{s}}\left(\frac{1\,\mathrm{mi}}{1609.344\,\mathrm{m}}\right)\left(\frac{3600\,\mathrm{s}}{1\,\mathrm{h}}\right)=\boxed{269\,\mathrm{mi/h}}.$$

Its velocity is forward along the track. The positive constant acceleration inferred from the same data is $a=v_f/t\approx17.9\,\mathrm{m/s^2}$.

**Meaning and check.** The calculation $d/t$ alone gives the average speed, about $134\,\mathrm{mi/h}$, not the finishing speed. Starting from rest and increasing velocity linearly makes the average exactly half the final value. Reinsert the inferred acceleration into $d=\tfrac12at^2$ to recover $402.336\,\mathrm{m}$.

**Türkçe destek.** Yol bölü zaman her durumda son hızı vermez. Bu harekette hız sıfırdan düzenli arttığı için ortalama hız $(0+v_f)/2$ olur. Denklemdeki $1/2$ katsayısı bu nedenle vardır. Önce $v_f$'yi yalnız bırakmak, sayıları yanlış sırayla çarpma riskini azaltır.

**Source:** `02_testbank.doc`, §2.2 Problems, Question 18. Original givens preserved; worked explanation added.

### Worked example: Stopping distance includes the reaction interval

**Problem.** A car approaches a stop sign at $40.0\,\mathrm{m/s}$. The sign is $200\,\mathrm{m}$ away when the driver notices it. The driver takes $0.200\,\mathrm{s}$ to apply the brakes. Find the magnitude of the constant acceleration needed during braking to stop exactly at the sign. Assume the speed remains constant during the reaction interval.

**Picture and principle.** Split the road into two adjacent lengths: reaction distance followed by braking distance. During the first interval there is no braking acceleration. During the second, acceleration points opposite to the forward velocity. Choose forward as positive.

The reaction distance follows from constant-speed motion:

$$d_r=v_0t_r=(40.0\,\mathrm{m/s})(0.200\,\mathrm{s})=8.00\,\mathrm{m}.$$

Subtract it from the total available distance:

$$d_b=D-d_r=200-8.00=192\,\mathrm{m}.$$

Braking time is unknown, so use the constant-acceleration relation that eliminates time:

$$v_f^2=v_0^2+2ad_b.$$

Subtract $v_0^2$ and divide by $2d_b$ to isolate the signed acceleration:

$$a=\frac{v_f^2-v_0^2}{2d_b}=\frac{0^2-(40.0\,\mathrm{m/s})^2}{2(192\,\mathrm{m})}=-4.1667\ldots\,\mathrm{m/s^2}.$$

The requested magnitude is $\boxed{|a|=4.17\,\mathrm{m/s^2}}$. The minus sign in $a$ indicates that the acceleration opposes the chosen positive direction.

**Meaning and check.** From $v_f=v_0+at$, the braking time is $t_b=-v_0/a=9.60\,\mathrm{s}$. Its average speed is $20.0\,\mathrm{m/s}$, so braking covers $20.0(9.60)=192\,\mathrm{m}$. Adding the reaction distance recovers $200\,\mathrm{m}$. Ignoring reaction would incorrectly give a smaller required deceleration, $4.00\,\mathrm{m/s^2}$.

**Türkçe destek.** Tepki süresinde araç durmaz; fren uygulanana kadar aynı hızla yol alır. Önce bu yolu toplam mesafeden çıkarmalıyız. İvmenin büyüklüğü pozitif istenir, fakat hareket denkleminde yönü göstermek için $a$ negatiftir. Eksi işaretini hesap bitmeden silme.

**Source:** `02_testbank.doc`, §2.2 Problems, Question 24. Original givens preserved; worked explanation added.

### Worked example: A delayed pursuit uses two different elapsed times

**Problem.** A car moving at a constant $30.0\,\mathrm{m/s}$ passes a stationary police car. The police driver waits $1.00\,\mathrm{s}$, then accelerates uniformly from rest. Determine the acceleration needed to catch the moving car at a point $300\,\mathrm{m}$ from the police car's starting position.

**Picture and principle.** Put $x=0$ where the cars first pass and $x=300\,\mathrm{m}$ at the meeting point. Start a common clock at the passing event. Both vehicles must occupy the same position when they meet, but they have not been moving for the same length of time.

The first car moves uniformly, so its travel time to the meeting point satisfies

$$D=vt_{\mathrm{meet}}\quad\Rightarrow\quad t_{\mathrm{meet}}=\frac{D}{v}=\frac{300\,\mathrm{m}}{30.0\,\mathrm{m/s}}=10.0\,\mathrm{s}.$$

The police car accelerates only after the delay $\tau$:

$$t_p=t_{\mathrm{meet}}-\tau=10.0-1.00=9.00\,\mathrm{s}.$$

Starting from rest gives $D=\tfrac12at_p^2$. Multiply by two and divide by the square of the police travel time:

$$a=\frac{2D}{t_p^2}=\frac{2D}{(D/v-\tau)^2}=\frac{2(300\,\mathrm{m})}{(9.00\,\mathrm{s})^2}=\boxed{7.41\,\mathrm{m/s^2}}.$$

**Meaning and check.** The police car reaches the meeting point with speed $at_p\approx66.7\,\mathrm{m/s}$. Its average speed over acceleration is half that value, $33.3\,\mathrm{m/s}$; multiplying by $9.00\,\mathrm{s}$ gives $300\,\mathrm{m}$. Meanwhile the other car travels $30.0(10.0)=300\,\mathrm{m}$. Equal meeting positions do not require equal instantaneous velocities.

### Worked example: One height can correspond to two times

**Problem.** A ball is hit straight upward at $30.0\,\mathrm{m/s}$. Find its time to maximum height, maximum rise, and both times when it is $25.0\,\mathrm{m}$ above the launch point. Ignore air resistance and use $g=9.80\,\mathrm{m/s^2}$.

**Picture and principle.** Put $y=0$ at launch and choose upward positive. Gravity gives constant acceleration $a_y=-g$ throughout the flight, including the highest point. Thus

$$v_y=v_0-gt,\qquad y=v_0t-\frac12gt^2.$$

At the top only the vertical velocity is zero. Solving $0=v_0-gt_{\mathrm{top}}$ gives

$$t_{\mathrm{top}}=\frac{v_0}{g}=\frac{30.0}{9.80}\,\mathrm{s}=3.06\,\mathrm{s}.$$

Using $0=v_0^2-2gh$ and isolating height,

$$h=\frac{v_0^2}{2g}=\frac{(30.0\,\mathrm{m/s})^2}{2(9.80\,\mathrm{m/s^2})}=\boxed{45.9\,\mathrm{m}}.$$

For $y=25.0\,\mathrm{m}$, move all terms to one side:

$$\frac12gt^2-v_0t+y=0.$$

The quadratic formula, with coefficients $g/2$, $-v_0$, and $y$, becomes

$$t=\frac{v_0\pm\sqrt{v_0^2-2gy}}{g}=\frac{30.0\pm\sqrt{30.0^2-2(9.80)(25.0)}}{9.80}\,\mathrm{s}.$$

Therefore $\boxed{t_1=0.995\,\mathrm{s}}$ on the way up and $\boxed{t_2=5.13\,\mathrm{s}}$ on the way down.

**Meaning and check.** The target height lies below the maximum, so two crossings are physically possible. The earlier root occurs before $3.06\,\mathrm{s}$ and the later root after it. Their midpoint equals $t_{\mathrm{top}}$, reflecting the up–down symmetry. Both roots satisfy the same height equation; neither should be discarded.

The positive discriminant also confirms two distinct crossings. A target higher than the maximum would give no real crossing time. 

**Türkçe destek.** İkinci dereceden denklemdeki iki kök burada iki farklı fiziksel olaydır. Top önce yükselirken, sonra inerken aynı yükseklikten geçer. Tepe noktasında hız sıfır olsa bile ivme sıfır değildir: yerçekimi aşağıya doğru etkisini sürdürür. Kökleri yorumlamak için tepeye çıkış süresiyle karşılaştır.

**Source:** `02_testbank.doc`, §2.2 Problems, Question 31. Original givens preserved; worked explanation added.

```python
#@title Run once — prepare compact demonstration controls
"""Embedded in the lecture notebooks by refresh_notebook_interface.py.

Only standard ipywidgets and IPython are used; notebooks remain self-contained.
"""
import functools
import traceback
import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import display, HTML
from IPython.utils.capture import capture_output
from ipywidgets import interactive as _PhysicsInteractive
from ipywidgets.widgets.interaction import show_inline_matplotlib_plots

# Set this explicitly: otherwise a local kernel can choose a desktop backend
# and block at plt.show(), leaving the notebook apparently frozen.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

try:
    from google.colab import output as _colab_output
except ImportError:
    pass
else:
    _colab_output.enable_custom_widget_manager()

_physics_panels = []
_physics_callback_errors = []


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; }
.phy101-controls { padding: 0 0 4px; box-sizing: border-box; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-animation-panel { max-width: 100%; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1;
  background: white; color: #172433; padding: 4px; }
@media (max-width: 650px) {
  .phy101-panel { flex-direction: column !important; }
  .phy101-controls, .phy101-plot-output { width: 100% !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Fit full labels and postpone costly redraws until a slider is released."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "continuous_update"):
            control.continuous_update = False
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "310px"
        control.layout.flex = "0 1 310px"
        control.layout.max_width = "100%"
        control.layout.min_width = "0"
        control.layout.margin = "2px 6px 2px 0"
        if isinstance(control, widgets.Button):
            control.layout.width = "auto"
            control.layout.flex = "0 0 auto"
    # Include the style in each output: Colab may isolate output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.Box([style] + flat, layout=widgets.Layout(
        display="flex", flex_flow="row wrap", align_items="center",
        width="100%", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        width="100%", min_width="0", max_width="100%",
        height="auto", overflow="visible", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Keep a compact control toolbar directly above the complete result."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="column",
                              align_items="stretch", width="100%"))
    panel.add_class("phy101-panel")
    return panel


class _PhysicsStableInteractive(_PhysicsInteractive):
    """Replace one persistent output model only after a redraw has completed.

    Sending clear_output followed by an inline figure depends on callback message
    capture in the frontend. Explicit MIME records keep repeated slider changes
    visible in Jupyter/Colab, without clearing the surrounding controls.
    """
    def update(self, *args):
        self.kwargs = {}
        if self.manual:
            self.manual_button.disabled = True
        try:
            for control in self.kwargs_widgets:
                self.kwargs[control._kwarg] = control.get_interact_value()
            with capture_output(stdout=True, stderr=True, display=True) as captured:
                self.result = self.f(**self.kwargs)
                show_inline_matplotlib_plots()
                if self.auto_display and self.result is not None:
                    display(self.result)
            records = []
            if captured.stdout:
                records.append({"output_type": "stream", "name": "stdout", "text": captured.stdout})
            if captured.stderr:
                records.append({"output_type": "stream", "name": "stderr", "text": captured.stderr})
            records.extend({"output_type": "display_data", "data": output.data,
                            "metadata": output.metadata or {}} for output in captured.outputs)
            # Keep the old result visible during calculation; publish the complete
            # replacement as one widget-state update, including its PNG/HTML data.
            self.out.outputs = tuple(records)
        except Exception:
            _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), traceback.format_exc()))
            self.out.outputs = ({"output_type": "stream", "name": "stderr",
                                 "text": traceback.format_exc()},)
        finally:
            if self.manual:
                self.manual_button.disabled = False


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


def physics_interactive(function, **controls):
    """An interactive widget with the same arguments as ipywidgets.interactive."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        try:
            return function(*args, **kwargs)
        except Exception as error:
            _physics_callback_errors.append((function.__name__, repr(error)))
            raise

    panel = _PhysicsStableInteractive(checked, {"auto_display": True}, **controls)
    inputs = [child for child in panel.children if child is not panel.out]
    panel.children = (_physics_controls(inputs), _physics_output(panel.out))
    panel.layout = widgets.Layout(display="flex", flex_flow="column",
                                  align_items="stretch", width="100%")
    panel.add_class("phy101-panel")
    _physics_panels.append(panel)
    return panel


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel
    display(panel)
    return function
```

### Visual check 1: 2.2 Interactive x-t, v-t, a-t Graph Explorer

Use the controls below. Predict → adjust one value → explain the result. / Tahmin et → tek değeri değiştir → sonucu açıkla.

```python
#@title Run the demonstration — predict, adjust, observe
# ---- Compare x(t), v(t), and a(t) together ----

def plot_kinematics(x0=0.0, v0=5.0, a=-1.0, t_end=10.0):
    t = np.linspace(0, t_end, 500)
    x = x0 + v0 * t + 0.5 * a * t**2
    v = v0 + a * t
    a_const = np.full_like(t, a)

    fig, axes = plt.subplots(1, 3, figsize=(11.6, 3.6), layout='constrained')
    fig.suptitle('Constant acceleration: three views of the same motion', fontsize=12)

    axes[0].plot(t, x, color='#2155a3', lw=2)
    axes[0].set(xlabel='Time t (s)', ylabel='Position x (m)', title='Position')
    if a != 0:
        t_turn = -v0 / a
        if 0 < t_turn < t_end:
            x_turn = x0 + v0 * t_turn + 0.5 * a * t_turn**2
            axes[0].plot(t_turn, x_turn, 'o', color='#bb2525', markersize=6)
            axes[0].set_title(f'Position: v = 0 at {t_turn:.1f} s', fontsize=11)

    axes[1].plot(t, v, color='#bb2525', lw=2)
    axes[1].fill_between(t, v, alpha=0.15, color='#bb2525')
    axes[1].set(xlabel='Time t (s)', ylabel='Velocity v (m/s)',
                title='Velocity: signed area = displacement')

    axes[2].plot(t, a_const, color='#237c49', lw=2)
    axes[2].set(xlabel='Time t (s)', ylabel=r'Acceleration a (m/s$^2$)',
                title='Acceleration', ylim=(a - 2, a + 2))
    for ax in axes:
        ax.grid(True, alpha=0.25)
        ax.axhline(0, color='#555555', lw=0.6)
        ax.tick_params(labelsize=9)
        ax.xaxis.label.set_size(10)
        ax.yaxis.label.set_size(10)
        ax.title.set_fontsize(10)

    # Keep the current values with all three graphs, in the same output.
    fig.supxlabel(
        rf'$x_0={x0:.1f}\,\mathrm{{m}},\quad v_0={v0:.1f}\,\mathrm{{m/s}},\quad '
        rf'a={a:.1f}\,\mathrm{{m/s^2}}$' + '\n' +
        rf'At $t={t_end:.1f}\,\mathrm{{s}}$: $x={x[-1]:.2f}\,\mathrm{{m}},\quad '
        rf'v={v[-1]:.2f}\,\mathrm{{m/s}},\quad \Delta x={x[-1]-x0:.2f}\,\mathrm{{m}}$',
        fontsize=10)
    plt.show()

physics_interact(plot_kinematics,
         x0=FloatSlider(min=-20, max=20, step=1, value=0, description='x0 (m):'),
         v0=FloatSlider(min=-10, max=10, step=0.5, value=5, description='v0 (m/s):'),
         a=FloatSlider(min=-5, max=5, step=0.25, value=-1, description='a (m/s^2):'),
         t_end=FloatSlider(min=2, max=20, step=1, value=10, description='t_end (s):'));
```

## Before the equations: a force can turn an object / Kuvvet ve döndürme etkisi

Work on paper first; the interactive pictures are optional ways to test a prediction. For a body at rest we need **both** $\sum F_x=0,\ \sum F_y=0$ and $\sum\tau=0$. A zero net force alone does not prevent rotation.

Draw every force at its application point. Replace a uniform rod's distributed weight by $mg$ at its midpoint. Choose a pivot that removes an unknown force: a force whose line passes through the pivot has zero lever arm.

**The geometry bridge:** $\tau=F d_\perp=rF\sin\theta$. The lever arm $d_\perp$ is the shortest distance from the pivot to the **line of the force**, not always the length of the object. A vertical weight on a rod at angle $\phi$ above horizontal has lever arm $r\cos\phi$. A horizontal force at its end has lever arm $r\sin\phi$.

**The algebra bridge:** If $F(4\,\mathrm m)=(200\,\mathrm N)(2\,\mathrm m)$, divide both sides by $4\,\mathrm m$ to get $F=100\,\mathrm N$. The metres cancel. A negative reaction force means the real direction is opposite to the arrow you guessed; a pin can push or pull, but a simple support cannot pull downward.

For center of mass, first calculate each $m_ix_i$ and $m_iy_i$, add each column, then divide by total mass: $x_{\rm cm}=\frac{\sum_i m_ix_i}{\sum_i m_i}$. A weighted average lies within the range of the positive-mass positions.

**Pause:** A vertical forearm hangs directly below the elbow. Gravity's line passes through the elbow, so its gravitational torque about that joint is zero.

**Türkçe:** Denge için hem kuvvetlerin hem torkların toplamı sıfır olmalı. Torkta kuvvetin etki çizgisine olan **dik uzaklığı** kullan. Önce şekli kontrol et; doğru cebir, yanlış geometriden doğru sonuç çıkaramaz.

## 1. Center of Mass

### Definition

The **center of mass** (COM) is the mass-weighted average position of all particles in a system:

$$x_{\text{cm}} = \frac{\sum_i m_i x_i}{\sum_i m_i} = \frac{m_1 x_1 + m_2 x_2 + \cdots}{m_1 + m_2 + \cdots}$$

$$y_{\text{cm}} = \frac{\sum_i m_i y_i}{\sum_i m_i}$$

### Why It Matters

| Property | Description |
|----------|-------------|
| **Motion** | The COM moves as if all mass is concentrated there and all external forces act on it |
| **Momentum** | $\vec{p}_{\text{total}} = M_{\text{total}} \cdot \vec{v}_{\text{cm}}$ |
| **Stability** | For a body resting on a horizontal support under gravity, the vertical projection of the COM must stay inside its support region to avoid tipping |
| **Balance** | In a uniform gravitational field, a support through the COM gives zero gravitational torque about that support |

### Analogy: The Team Captain

Imagine a group of people on a raft. Even if individuals move around, the raft itself moves based on where the "average person" (the center of mass) is. The heavier people pull the average closer to them, just like heavier masses pull the COM closer.

**Two-coordinate calculation / İki koordinatı ayrı hesapla.** For $1\,\mathrm{kg}$ at $(0,0)$ and $3\,\mathrm{kg}$ at $(4,2)\,\mathrm m$:

$$\begin{aligned}M&=1+3=4\,\mathrm{kg},\\x_\mathrm{cm}&=\frac{1(0)+3(4)}{4}=3.0\,\mathrm m,\\y_\mathrm{cm}&=\frac{1(0)+3(2)}{4}=1.5\,\mathrm m.\end{aligned}$$

The center lies three quarters of the way toward the larger mass. **TR:** Aynı kütle paydasını kullanırken $x$ ve $y$ paylarını birbirine karıştırmayın. Sıfır toplam kütle için bu ortalama tanımlı değildir.

## 4. Worked Examples

### Example 1: Center of Mass of a System

Three masses are placed on the $x$-axis: $2\,\mathrm{kg}$ at $x=0$, $5\,\mathrm{kg}$ at $x=3\,\mathrm m$, and $3\,\mathrm{kg}$ at $x=7\,\mathrm m$. Find the center of mass.

**Paper solution.** A center of mass is a mass-weighted average. Make a small table:

| $m$ (kg) | $x$ (m) | $mx$ (kg m) |
|---|---|---|
| 2 | 0 | 0 |
| 5 | 3 | 15 |
| 3 | 7 | 21 |
| **Total 10** | | **Total 36** |

$$x_{\rm cm}=\frac{0+15+21}{2+5+3}=\frac{36\ \mathrm{kg\,m}}{10\ \mathrm{kg}}=\boxed{3.6\ \mathrm m}.$$
The result lies between 0 and $7\,\mathrm m$, as required for positive masses. It lies right of $3\,\mathrm m$ because the mass at $7\,\mathrm m$ has a larger rightward moment than the mass at $0\,\mathrm m$ has leftward moment about $3\,\mathrm m$.

**Türkçe:** Konumların normal ortalamasını değil, kütleyle ağırlıklandırılmış ortalamasını alıyoruz. Birimlerde kg sadeleşip metre kalır.

## 3. Static Equilibrium

### Two Conditions

An object initially at rest remains in **static equilibrium** when both conditions hold. If it is already moving, zero net force and torque do not by themselves make it stop:

1. **Translational equilibrium** (no net force): $\sum \vec{F} = 0$
2. **Rotational equilibrium** (no net torque about any point): $\sum \tau = 0$

### Problem-Solving Strategy

1. Draw a **free-body diagram** showing all forces
2. Choose a convenient **pivot point** (often where an unknown force acts, to eliminate it)
3. Write $\sum F_x = 0$, $\sum F_y = 0$, and $\sum \tau = 0$
4. Solve the system of equations

### Key Insight

You can choose ANY point as the pivot for the torque equation. Choosing wisely (where unknown forces act) simplifies your math enormously!

**Why two equations are needed.** Two forces $+10\,\mathrm N$ upward at the right end and $-10\,\mathrm N$ downward at the left end of a $2.0\,\mathrm m$ bar give zero net force. About the center, each contributes $+10\,\mathrm{N\,m}$, so net torque is $+20\,\mathrm{N\,m}$. The bar is not in static equilibrium.

**TR:** Eşit ve zıt iki kuvvet farklı doğrularda etki ederse bir kuvvet çifti oluşturabilir. Kuvvet toplamı sıfır olsa da dönme etkileri aynı yönde toplanır.

### Example 2: Beam with Two Supports

A uniform $4\,\mathrm m$ beam weighing $200\,\mathrm N$ is supported at both ends (A at $x=0$, B at $x=4\,\mathrm m$). A $500\,\mathrm N$ weight is placed at $x=1\,\mathrm m$ from end A. Find the reaction forces at A and B.

**Paper solution.** Draw upward reactions $R_A,R_B$. The uniform beam's own $200\,\mathrm N$ weight acts at its center, $x=2\,\mathrm m$; the $500\,\mathrm N$ load acts at $x=1\,\mathrm m$. Choose A as the torque pivot, counterclockwise positive:
$$R_B(4)-200(2)-500(1)=0.$$
Add the two clockwise moments: $4R_B=400+500=900\,\mathrm{N\,m}$. Divide by $4\,\mathrm m$: **$R_B=225\,\mathrm N$**.

Next use vertical force balance: $R_A+R_B-200-500=0$, so $R_A=700-225=475\,\mathrm N$. Check: $475+225=700\,\mathrm N$ equals the total load. Taking moments about B also gives $475(4)=200(2)+500(3)=1900\,\mathrm{N\,m}$. A carries more load because the crate is nearer A.

**Türkçe:** Önce bir desteği dönme noktası seçerek onun bilinmeyen kuvvetini tork denkleminden çıkar. Sonra kuvvet dengesiyle diğer desteği bul.

### L1 (Basic) — Module 08 P3

A uniform beam of length $6.0\,\mathrm m$ and mass $30\,\mathrm{kg}$ is supported at its left end by a pivot. What upward force must be applied at the right end to keep the beam horizontal? Take $g = 9.81\,\mathrm{m/s^2}$.

**Türkçe problem açıklaması:** $6.0\,\mathrm m$ uzunluklu, $30\,\mathrm{kg}$ düzgün kiriş sol uçtan mafsallı. Sağ uca hangi yukarı kuvvet uygulanırsa yatay dengede kalır? $g=9.81\,\mathrm{m/s^2}$. Kiriş ağırlığını orta noktaya yerleştirin ve sol uç etrafındaki torkları dengeleyin.

<details>
<summary>Answer</summary>

Take torques about the left pivot. The uniform beam's weight acts $L/2=3.0\,\mathrm m$ from that pivot:
$$\begin{aligned}FL-mg\frac L2&=0,\\
F&=\frac{mg(L/2)}{L}=\frac{mg}{2}
=\frac{(30)(9.81)}{2}=147.15\,\mathrm N\quad\text{upward}.
\end{aligned}$$
The pivot supplies the remaining upward force; the force at the free end need not equal the whole weight.

</details>

### L2 (Intermediate) — Module 08 P6

A ladder of length $4.0\,\mathrm m$ and mass $12\,\mathrm{kg}$ leans against a smooth (frictionless) wall at an angle of $65^\circ$ with the floor. A $75\,\mathrm{kg}$ person stands $3.0\,\mathrm m$ up the ladder from the bottom. Find the normal force from the wall and the friction force at the floor. Take $g = 9.81\,\mathrm{m/s^2}$.

**Türkçe problem açıklaması:** $4.0\,\mathrm m$, $12\,\mathrm{kg}$ merdiven sürtünmesiz duvara, zeminle $65^\circ$ yaparak dayanıyor. $75\,\mathrm{kg}$ kişi merdiven boyunca alttan $3.0\,\mathrm m$ yukarıda. <details>
<summary>Answer</summary>

$$\begin{aligned}N_\mathrm{wall}(4.0\sin65^\circ)&=[12(9.81)(2.0)+75(9.81)(3.0)]\cos65^\circ,\\
N_\mathrm{wall}&=\frac{1032.32\,\mathrm{N\,m}}{3.62524\,\mathrm m}=284.76\,\mathrm N.
\end{aligned}$$
Horizontal equilibrium gives $f=N_\mathrm{wall}=284.76\,\mathrm N$, with floor friction directed toward the wall. Vertically, $N_\mathrm{floor}=(12+75)g=853.47\,\mathrm N$, so static equilibrium requires
$$\mu_s\ge\frac{f}{N_\mathrm{floor}}=0.334.$$
**[CORRECTED]** The previous $290.7\,\mathrm N$ used approximately $g=10\,\mathrm{m/s^2}$ with rounded trigonometry, rather than the stated $g=9.81\,\mathrm{m/s^2}$.

```python
#@title Run once — keep controls beside figures
"""Embedded in the lecture notebooks by refresh_notebook_interface.py.

Only standard ipywidgets and IPython are used; notebooks remain self-contained.
"""
import functools
import traceback
import ipywidgets as widgets
from IPython import get_ipython
from IPython.display import display, HTML
from IPython.utils.capture import capture_output
from ipywidgets import interactive as _PhysicsInteractive
from ipywidgets.widgets.interaction import show_inline_matplotlib_plots

# Set this explicitly: otherwise a local kernel can choose a desktop backend
# and block at plt.show(), leaving the notebook apparently frozen.
if get_ipython() is not None:
    get_ipython().run_line_magic("matplotlib", "inline")

try:
    from google.colab import output as _colab_output
except ImportError:
    pass
else:
    _colab_output.enable_custom_widget_manager()

_physics_panels = []
_physics_callback_errors = []


def physics_frames(frame_count, maximum=60):
    """Sample display frames, retaining both endpoints and all simulation data."""
    count = int(frame_count)
    shown = min(count, maximum)
    if shown <= 1:
        return list(range(shown))
    return [round(index * (count - 1) / (shown - 1)) for index in range(shown)]


def physics_interval(frame_count, interval_ms):
    """Preserve first-to-last playback duration when display frames are sampled."""
    shown = len(physics_frames(frame_count))
    return interval_ms if shown <= 1 else interval_ms * (int(frame_count) - 1) / (shown - 1)


PHYSICS_STYLE = """<style>
.phy101-panel { border: 1px solid #a9b9c9; border-radius: 8px; padding: 8px; }
.phy101-controls { padding: 4px 8px 4px 0; box-sizing: border-box; overflow-x: hidden; }
.phy101-plot-output img { max-width: 100%; height: auto; object-fit: contain; }
.phy101-plot-output .output_area { overflow: visible; }
.phy101-plot-output table { font-size: 13px; width: 100%; }
.phy101-plot-output .animation { max-width: 100%; }
.phy101-plot-output .animation img { max-height: 390px; width: auto; object-fit: contain; }
.phy101-animation-panel { max-height: 470px; overflow: auto; }
.phy101-animation-panel .animation { display: flex; flex-direction: column; }
.phy101-animation-panel .animation img { order: 2; max-width: 100%; height: auto; }
.phy101-animation-panel .anim-controls { order: 1; position: sticky; top: 0;
  z-index: 2; background: white; color: #172433; padding: 4px; }
.phy101-plot-output .phy101-animation-panel { max-height: 410px; }
@media (max-width: 650px) {
  .phy101-panel { flex-direction: column !important; }
  .phy101-controls { width: 100% !important; flex-basis: auto !important;
                     max-height: 170px; overflow-y: auto; }
  .phy101-plot-output { width: 100% !important; flex-basis: auto !important;
                       max-height: 350px !important; }
}
</style>"""
display(HTML(PHYSICS_STYLE))


def physics_animation_html(animation):
    """A self-contained animation pane with playback controls kept in view."""
    return HTML(PHYSICS_STYLE + '<div class="phy101-animation-panel">' +
                animation.to_jshtml(default_mode="once") + "</div>")


def _physics_controls(items):
    """Fit full labels and postpone costly redraws until a slider is released."""
    flat = []
    for item in items:
        if isinstance(item, (widgets.HBox, widgets.VBox)):
            flat.extend(item.children)
        else:
            flat.append(item)
    for control in flat:
        if hasattr(control, "continuous_update"):
            control.continuous_update = False
        if hasattr(control, "style") and "description_width" in control.style.traits():
            control.style.description_width = "initial"
        control.layout.width = "calc(100% - 8px)"
        control.layout.min_width = "0"
    # Include the style in each output: Colab may isolate output frames.
    style = widgets.HTML(value=PHYSICS_STYLE, layout=widgets.Layout(display="none"))
    box = widgets.VBox([style] + flat, layout=widgets.Layout(
        flex="0 0 245px", width="245px", min_width="0", max_width="100%"))
    box.add_class("phy101-controls")
    return box


def _physics_output(output):
    output.layout = widgets.Layout(
        flex="1 1 320px", min_width="0", max_width="100%",
        height="480px", overflow="auto", margin="0")
    output.add_class("phy101-plot-output")
    return output


def physics_panel(controls, output):
    """Keep custom button controls beside a separately scrolling result pane."""
    panel = widgets.Box([_physics_controls(controls), _physics_output(output)],
        layout=widgets.Layout(display="flex", flex_flow="row wrap",
                              align_items="flex-start", width="100%"))
    panel.add_class("phy101-panel")
    return panel


class _PhysicsStableInteractive(_PhysicsInteractive):
    """Replace one persistent output model only after a redraw has completed.

    Sending clear_output followed by an inline figure depends on callback message
    capture in the frontend. Explicit MIME records keep repeated slider changes
    visible in Jupyter/Colab, without clearing the surrounding controls.
    """
    def update(self, *args):
        self.kwargs = {}
        if self.manual:
            self.manual_button.disabled = True
        try:
            for control in self.kwargs_widgets:
                self.kwargs[control._kwarg] = control.get_interact_value()
            with capture_output(stdout=True, stderr=True, display=True) as captured:
                self.result = self.f(**self.kwargs)
                show_inline_matplotlib_plots()
                if self.auto_display and self.result is not None:
                    display(self.result)
            records = []
            if captured.stdout:
                records.append({"output_type": "stream", "name": "stdout", "text": captured.stdout})
            if captured.stderr:
                records.append({"output_type": "stream", "name": "stderr", "text": captured.stderr})
            records.extend({"output_type": "display_data", "data": output.data,
                            "metadata": output.metadata or {}} for output in captured.outputs)
            # Keep the old result visible during calculation; publish the complete
            # replacement as one widget-state update, including its PNG/HTML data.
            self.out.outputs = tuple(records)
        except Exception:
            _physics_callback_errors.append((getattr(self.f, "__name__", "callback"), traceback.format_exc()))
            self.out.outputs = ({"output_type": "stream", "name": "stderr",
                                 "text": traceback.format_exc()},)
        finally:
            if self.manual:
                self.manual_button.disabled = False


def physics_show_figure(figure):
    """Display one inline figure and close its pyplot registration afterwards."""
    import matplotlib.pyplot as plt
    display(figure)
    plt.close(figure)


def physics_vector_axes(axes, points):
    """Equal x/y scales and limits covering all arrow endpoints, including sums."""
    import numpy as np
    coordinates = np.asarray(points, dtype=float).reshape(-1, 2)
    span = max(1.0, float(np.max(np.abs(coordinates)))) * 1.22
    axes.set(xlim=(-span, span), ylim=(-span, span), xlabel="x component", ylabel="y component")
    axes.set_aspect("equal", adjustable="box")
    axes.axhline(0, color="#718096", linewidth=0.7)
    axes.axvline(0, color="#718096", linewidth=0.7)
    axes.grid(alpha=0.2)


def physics_interactive(function, **controls):
    """An interactive widget with the same arguments as ipywidgets.interactive."""
    @functools.wraps(function)
    def checked(*args, **kwargs):
        try:
            return function(*args, **kwargs)
        except Exception as error:
            _physics_callback_errors.append((function.__name__, repr(error)))
            raise

    panel = _PhysicsStableInteractive(checked, {"auto_display": True}, **controls)
    inputs = [child for child in panel.children if child is not panel.out]
    panel.children = (_physics_controls(inputs), _physics_output(panel.out))
    panel.layout = widgets.Layout(display="flex", flex_flow="row wrap",
                                  align_items="flex-start", width="100%")
    panel.add_class("phy101-panel")
    _physics_panels.append(panel)
    return panel


def physics_interact(function=None, **controls):
    """Support both @physics_interact(...) and physics_interact(function, ...)."""
    if function is None:
        return lambda function: physics_interact(function, **controls)
    panel = physics_interactive(function, **controls)
    function.widget = panel
    display(panel)
    return function
```

### Visual check 1: Interactive Demo 1: Center of Mass Calculator & Visualizer

Use the controls below. Predict → adjust one value → explain the result. / Tahmin et → tek değeri değiştir → sonucu açıkla.

```python
#@title Run the demonstration — predict, adjust, observe
def com_visualizer(m1=3.0, x1=1.0, y1=1.0,
                   m2=2.0, x2=5.0, y2=1.0,
                   m3=4.0, x3=3.0, y3=5.0,
                   m4=1.0, x4=6.0, y4=4.0):
    """Keep original mass identities and put coincident-point labels in a separate table."""
    masses, xs, ys = [m1,m2,m3,m4], [x1,x2,x3,x4], [y1,y2,y3,y4]
    active = [(i,m,x,y) for i,(m,x,y) in enumerate(zip(masses,xs,ys)) if m > 0]
    if not active:
        print('Center of mass is undefined when total mass is zero. / Toplam kütle sıfırken kütle merkezi tanımsızdır.')
        return
    M = sum(m for _,m,x,y in active)
    x_cm = sum(m*x for _,m,x,y in active)/M
    y_cm = sum(m*y for _,m,x,y in active)/M
    fig,(ax,ax_info)=plt.subplots(1,2,figsize=(13,6),gridspec_kw={'width_ratios':[1.2,1]},layout='constrained')
    ax.set(xlim=(-1,8),ylim=(-1,8),xlabel='x (m)',ylabel='y (m)',title='Center of Mass — positions')
    ax.set_aspect('equal'); ax.grid(True,alpha=0.25)
    colors=['#2196F3','#FF5722','#4CAF50','#9C27B0']
    for i,m,x,y in active:
        ax.scatter(x,y,s=100+150*m,color=colors[i],edgecolors='black',alpha=0.6,zorder=5,
                   label=f'm{i+1}')
        ax.plot([x,x_cm],[y,y_cm],'--',color=colors[i],alpha=0.4)
    ax.scatter(x_cm,y_cm,s=260,color='red',marker='X',edgecolors='black',zorder=10)
    ax_info.axis('off')
    ax_info.set_title('Masses and calculation',fontsize=13)
    rows=[[f'm{i+1}',f'{m:.1f}',f'{x:.1f}',f'{y:.1f}',f'{m*x:.2f}',f'{m*y:.2f}'] for i,m,x,y in active]
    table=ax_info.table(cellText=rows,colLabels=['ID','m (kg)','x (m)','y (m)','mx','my'],
                        cellLoc='center',bbox=[0,0.49,1,0.42])
    table.auto_set_font_size(False); table.set_fontsize(10)
    for row,(i,m,x,y) in enumerate(active,1):
        table[(row,0)].get_text().set_color(colors[i])
    summary=(f'Total mass = {M:.2f} kg\n'
             f'x_cm = {sum(m*x for _,m,x,y in active):.2f} / {M:.2f} = {x_cm:.3f} m\n'
             f'y_cm = {sum(m*y for _,m,x,y in active):.2f} / {M:.2f} = {y_cm:.3f} m\n\n'
             'Red X = center of mass.\nmx and my columns have units kg m.\n'
             'Coincident points remain separate rows.\nA zero mass is omitted, not renumbered.')
    ax_info.text(0.02,0.42,summary,transform=ax_info.transAxes,va='top',fontsize=10.5,linespacing=1.4)
    plt.show()
    print(f'Total mass: M = {M:.1f} kg; center of mass: ({x_cm:.3f}, {y_cm:.3f}) m')

physics_interact(com_visualizer,
         m1=FloatSlider(value=3.0, min=0, max=10, step=0.5, description='m1 (kg)'),
         x1=FloatSlider(value=1.0, min=0, max=7, step=0.5, description='x1 (m)'),
         y1=FloatSlider(value=1.0, min=0, max=7, step=0.5, description='y1 (m)'),
         m2=FloatSlider(value=2.0, min=0, max=10, step=0.5, description='m2 (kg)'),
         x2=FloatSlider(value=5.0, min=0, max=7, step=0.5, description='x2 (m)'),
         y2=FloatSlider(value=1.0, min=0, max=7, step=0.5, description='y2 (m)'),
         m3=FloatSlider(value=4.0, min=0, max=10, step=0.5, description='m3 (kg)'),
         x3=FloatSlider(value=3.0, min=0, max=7, step=0.5, description='x3 (m)'),
         y3=FloatSlider(value=5.0, min=0, max=7, step=0.5, description='y3 (m)'),
         m4=FloatSlider(value=1.0, min=0, max=10, step=0.5, description='m4 (kg)'),
         x4=FloatSlider(value=6.0, min=0, max=7, step=0.5, description='x4 (m)'),
         y4=FloatSlider(value=4.0, min=0, max=7, step=0.5, description='y4 (m)'));
```

### Visual check 2: Interactive Demo 4: Seesaw/Lever Equilibrium

Use the controls below. Predict → adjust one value → explain the result. / Tahmin et → tek değeri değiştir → sonucu açıkla.

```python
#@title Run the demonstration — predict, adjust, observe
def seesaw_interactive(m_left=60.0,d_left=2.0,m_right=40.0,d_right=3.0):
    """Separate torque data from geometry and show the heavy side tending downward."""
    g=9.81;half_len=4.0
    torque_left=m_left*g*d_left;torque_right=m_right*g*d_right
    net_torque=torque_left-torque_right
    required=m_left*d_left/m_right
    # Positive CCW means the left endpoint moves downward.
    tilt=-np.clip(net_torque*.005,-12,12)
    ct,st=np.cos(np.radians(tilt)),np.sin(np.radians(tilt))
    fig,(ax,info)=plt.subplots(1,2,figsize=(14,6),gridspec_kw={'width_ratios':[1.4,1]},layout='constrained')
    ax.set(xlim=(-5,5),ylim=(-2,3),title='8 m seesaw — turning tendency')
    ax.set_aspect('equal');ax.axis('off')
    ax.plot([-5,5],[-1,-1],'k-',lw=2)
    ax.add_patch(plt.Polygon([(-.4,-1),(.4,-1),(0,0)],facecolor='#607D8B',edgecolor='black'))
    ax.plot([-half_len*ct,half_len*ct],[half_len*st,-half_len*st],color='#795548',lw=10)
    for mass,distance,sign,color in [(m_left,d_left,-1,'#1565C0'),(m_right,d_right,1,'#D84315')]:
        x=sign*distance*ct;y=-sign*distance*st
        ax.add_patch(plt.Circle((x,y+.95),.15,facecolor=color,edgecolor='black'))
        ax.plot([x,x],[y+.05,y+.8],color=color,lw=3)
        ax.plot([x-.3,x+.3],[y+.5,y+.5],color=color,lw=2)
        ax.text(x,y+1.3,f'{mass:.0f} kg',ha='center',fontsize=11,color=color)
    balanced=abs(net_torque)<1e-8
    status='Balanced' if balanced else ('Left side tends downward (CCW)' if net_torque>0 else 'Right side tends downward (CW)')
    info.axis('off');info.set_title('Horizontal-position torque account')
    rows=[f'Left: {m_left:.0f} × 9.81 × {d_left:.2f}',f'      {torque_left:.2f} N m (CCW)',
          f'Right: {m_right:.0f} × 9.81 × {d_right:.2f}',f'       {torque_right:.2f} N m (CW)',
          '',f'Net = {net_torque:+.2f} N m',status,'',
          f'Right distance for balance = {required:.3f} m']
    if required>half_len:
        rows += ['Outside the 4 m half-board: no feasible', 'balance by moving only the right person.']
    elif not .5<=required<=3.5:
        rows += ['On the board, but outside this slider range.', 'Adjust another mass or position as well.']
    else: rows += ['This position is available on the slider.']
    rows += ['', 'The drawn tilt indicates direction only;', 'it is not a predicted equilibrium angle.']
    info.text(.02,.94,'\n'.join(rows),transform=info.transAxes,va='top',fontsize=10.5,linespacing=1.35)
    plt.show()

physics_interact(seesaw_interactive,
         m_left=FloatSlider(value=60, min=10, max=120, step=5, description='Left (kg)'),
         d_left=FloatSlider(value=2.0, min=0.5, max=3.5, step=0.25, description='d left (m)'),
         m_right=FloatSlider(value=40, min=10, max=120, step=5, description='Right (kg)'),
         d_right=FloatSlider(value=3.0, min=0.5, max=3.5, step=0.25, description='d right (m)'));
```

## Part 1: Why Organize Code?

So far we have been writing all our code in a **single notebook** or a **single file**. This works fine for small programs, but as projects grow, things get messy.

Think about a mechatronics project:

| Component | Code Responsibility |
|-----------|--------------------|
| Sensor | Read data from sensors |
| Motor | Control motor speed |
| Display | Show information on screen |
| Logger | Save data to files |

If all this code is in one file, it becomes very hard to:
- **Find** the code you need
- **Fix bugs** without breaking other parts
- **Reuse** code in another project
- **Work in a team** without conflicts

The solution? **Split your code into separate files** (modules) and **organize them into folders** (packages).

**Figure 11.1** — Single file vs. organized project:

```
Bad:                          Good:
everything.py (500 lines)     src/
                                sensor.py
                                motor.py
                                display.py
                                logger.py
```

## Part 2: Modules — One File = One Module

A **module** is simply a `.py` file that contains Python code (variables, functions, classes).

Every `.py` file you create is automatically a module!

**Figure 11.2** — A module is just a Python file:

```
math_tools.py    <-- This is a module named "math_tools"
```

Let's create our first module in Colab. We will use the `%%writefile` magic command to create a `.py` file.

````python
%%writefile math_tools.py
# math_tools.py - A simple module with math helper functions

def add(a, b):
    """Add two numbers and return the result."""
    return a + b

def multiply(a, b):
    """Multiply two numbers and return the result."""
    return a * b

PI = 3.14159

def circle_area(radius):
    """Calculate the area of a circle."""
    return PI * radius * radius
````

Now we have a file called `math_tools.py` in our Colab working directory. This file is a **module**.

Key points about modules:
- The **file name** (without `.py`) is the **module name**
- A module can contain **functions**, **classes**, and **variables**
- You can **import** a module into other code to use it

## Part 3: Import Statements

To use code from a module, we **import** it. The simplest way is:

```python
import module_name
```

After importing, you access things inside the module using **dot notation**:

```python
module_name.function_name()
module_name.variable_name
```

**Figure 11.1** — Module imports

````python
# Import our module
import math_tools

# Use functions from the module
result = math_tools.add(5, 3)
print(f"5 + 3 = {result}")

area = math_tools.circle_area(10)
print(f"Area of circle with radius 10 = {area}")

# Access a variable from the module
print(f"PI = {math_tools.PI}")
````

### Import with an Alias

If a module name is long, you can give it a shorter **alias**:

```python
import module_name as short_name
```

You have already seen this with popular libraries:

````python
# Common aliases you have seen before
import math_tools as mt

# Now use the short name
print(mt.add(10, 20))
print(mt.PI)
````

### Python Built-in Modules

Python comes with many built-in modules. You have probably used some:

| Module | What It Does |
|--------|-------------|
| `math` | Math functions (sqrt, sin, cos) |
| `random` | Random number generation |
| `os` | Operating system interaction |
| `datetime` | Date and time handling |
| `csv` | Reading/writing CSV files |

````python
# Examples of built-in modules
import math
print(f"Square root of 16 = {math.sqrt(16)}")

import random
print(f"Random number = {random.randint(1, 100)}")

import datetime
print(f"Today = {datetime.date.today()}")
````

## Part 4: From...Import Syntax

Sometimes you only need **one or two things** from a module. Instead of importing the whole module, you can import specific items:

```python
from module_name import function_name
from module_name import function1, function2
from module_name import variable_name
```

When you use `from...import`, you **do not need the module name prefix** anymore.

**Figure 11.2** — Module imports

````python
# Import specific items
from math_tools import add, PI

# Use them directly (no prefix needed)
result = add(100, 200)
print(f"100 + 200 = {result}")
print(f"PI = {PI}")
````

### Import Everything (Not Recommended)

You can import everything from a module with `*`:

```python
from module_name import *
```

This is **not recommended** because:
- You cannot tell which names came from which module
- Names from different modules can **conflict** (overwrite each other)

**Figure 11.3** — Comparison of import styles:

| Style | Example | When to Use |
|-------|---------|------------|
| `import X` | `import math_tools` | When you use many things from a module |
| `import X as Y` | `import math_tools as mt` | When the name is long |
| `from X import Y` | `from math_tools import add` | When you need only a few items |
| `from X import *` | `from math_tools import *` | Avoid this! |

````python
# Example showing why 'import *' can be dangerous

# Suppose we have our own variable called PI
PI = 3.14
print(f"Our PI = {PI}")

# If we do 'from math_tools import *', it overwrites our PI!
from math_tools import *
print(f"After import *, PI = {PI}")  # PI is now from math_tools!
````

## Part 5: Creating a Package (Folder with `__init__.py`)

A **package** is a **folder** that contains Python modules. To make a folder a package, it needs a special file called `__init__.py`.

**Figure 11.4** — Package structure:

```
my_package/              <-- This is a package (folder)
    __init__.py          <-- Makes it a package (can be empty)
    module_a.py          <-- A module inside the package
    module_b.py          <-- Another module
```

### What is `__init__.py`?

- It is a special file that tells Python "this folder is a package"
- It can be **empty** or contain initialization code
- Without this file, Python may not recognize the folder as a package

Let's create a package in Colab:

````python
import os

# Create a folder for our package
os.makedirs("my_package", exist_ok=True)

print("Folder 'my_package' created!")
````

**Figure 11.3** — Creating a module file

````python
%%writefile my_package/__init__.py
# This file makes 'my_package' a Python package.
# It can be empty, but we add a print to see when it runs.
print("my_package has been imported!")
````

**Figure 11.4** — Creating a module file

````python
%%writefile my_package/greetings.py
# greetings.py - A module inside our package

def hello(name):
    """Return a greeting message."""
    return f"Hello, {name}! Welcome to OOP."

def goodbye(name):
    """Return a farewell message."""
    return f"Goodbye, {name}! See you next week."
````

**Figure 11.5** — Creating a module file

````python
%%writefile my_package/calculator.py
# calculator.py - Another module inside our package

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def divide(a, b):
    if b == 0:
        return "Error: division by zero"
    return a / b
````

**Figure 11.6** — Module imports

````python
# Now let's import from our package
from my_package import greetings

print(greetings.hello("Student"))
print(greetings.goodbye("Student"))
````

**Figure 11.7** — Module imports

````python
# Import a specific function from a module inside the package
from my_package.calculator import add, divide

print(f"10 + 5 = {add(10, 5)}")
print(f"10 / 3 = {divide(10, 3):.2f}")
print(f"10 / 0 = {divide(10, 0)}")
````

## Part 6: The `src/` Structure in Colab

In real projects, it is common to put all your code in a `src/` folder. This keeps things organized.

**Figure 11.5** — A typical `src/` project structure:

```
project/
    src/
        __init__.py
        sensor.py
        motor.py
        utils.py
    main.py
    README.md
```

In Colab, we sometimes need to tell Python **where to find our modules**. We do this with `sys.path`.

### What is `sys.path`?

`sys.path` is a **list of folders** where Python looks for modules when you use `import`. If your module is not in one of these folders, Python will not find it.

````python
import sys

# Show where Python looks for modules
print("Python searches for modules in these folders:")
for path in sys.path:
    print(f"  - {path}")
````

````python
# Create a src/ folder structure
import os

os.makedirs("src", exist_ok=True)

print("Created src/ folder")
````

**Figure 11.8** — Creating a module file

````python
%%writefile src/__init__.py
# Makes src/ a package
````

**Figure 11.9** — Creating a module file

````python
%%writefile src/converter.py
# converter.py - Unit conversion utilities for engineering

def celsius_to_fahrenheit(c):
    """Convert Celsius to Fahrenheit."""
    return c * 9 / 5 + 32

def meters_to_feet(m):
    """Convert meters to feet."""
    return m * 3.28084

def kg_to_pounds(kg):
    """Convert kilograms to pounds."""
    return kg * 2.20462
````

**Figure 11.10** — Module imports

````python
import sys

# Add the current directory to sys.path so Python can find src/
if '.' not in sys.path:
    sys.path.insert(0, '.')

# Now we can import from src
from src.converter import celsius_to_fahrenheit, meters_to_feet

print(f"100 C = {celsius_to_fahrenheit(100)} F")
print(f"5 meters = {meters_to_feet(5):.2f} feet")
````

## Part 7: Putting It Together — A Toolkit Package

Let's build a complete **engineering toolkit** package step by step. This will be a package called `toolkit` with several modules.

**Figure 11.6** — Our toolkit structure:

```
toolkit/
    __init__.py
    math_utils.py       # Math helper functions
    string_utils.py     # String helper functions
    physics.py           # Basic physics formulas
```

````python
import os

# Create the toolkit folder
os.makedirs("toolkit", exist_ok=True)
print("Created toolkit/ folder")
````

**Figure 11.11** — Creating a module file

````python
%%writefile toolkit/__init__.py
# toolkit package
# This file makes the toolkit/ folder a Python package.

VERSION = "1.0"
AUTHOR = "OOP Student"
````

**Figure 11.12** — Creating a module file

````python
%%writefile toolkit/math_utils.py
# math_utils.py - Math helper functions

def average(numbers):
    """Calculate the average of a list of numbers."""
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

def clamp(value, min_val, max_val):
    """Keep a value within a min/max range."""
    if value < min_val:
        return min_val
    if value > max_val:
        return max_val
    return value

def is_even(n):
    """Check if a number is even."""
    return n % 2 == 0
````

**Figure 11.13** — Creating a module file

````python
%%writefile toolkit/string_utils.py
# string_utils.py - String helper functions

def shout(text):
    """Convert text to uppercase with exclamation mark."""
    return text.upper() + "!"

def whisper(text):
    """Convert text to lowercase with ellipsis."""
    return text.lower() + "..."

def repeat(text, times=3):
    """Repeat text a given number of times."""
    return (text + " ") * times
````

**Figure 11.14** — Creating a module file

````python
%%writefile toolkit/physics.py
# physics.py - Basic physics formulas

GRAVITY = 9.81  # m/s^2

def force(mass, acceleration):
    """Calculate force: F = m * a"""
    return mass * acceleration

def weight(mass):
    """Calculate weight on Earth: W = m * g"""
    return mass * GRAVITY

def kinetic_energy(mass, velocity):
    """Calculate kinetic energy: KE = 0.5 * m * v^2"""
    return 0.5 * mass * velocity ** 2
````

**Figure 11.15** — Module imports

````python
# Let's use our complete toolkit!
import toolkit
from toolkit import math_utils, string_utils, physics

# Check the package info
print(f"Toolkit version: {toolkit.VERSION}")
print(f"Author: {toolkit.AUTHOR}")
print()

# Math utils
scores = [85, 92, 78, 95, 88]
print(f"Average score: {math_utils.average(scores)}")
print(f"Clamped 150 to [0,100]: {math_utils.clamp(150, 0, 100)}")
print()

# String utils
print(string_utils.shout("hello world"))
print(string_utils.whisper("BE QUIET"))
print()

# Physics
print(f"Force (10kg, 5m/s^2): {physics.force(10, 5)} N")
print(f"Weight of 70kg person: {physics.weight(70):.1f} N")
print(f"KE (2kg, 3m/s): {physics.kinetic_energy(2, 3)} J")
````

## Part 2: The `__init__` Method --- Automatic Setup

The `__init__` method (short for "initialize") is a special function that Python calls **automatically** every time you create a new object.

**Figure 2.2** --- Your first `__init__`

````python
class Motor:
    def __init__(self, name, speed):
        self.name = name        # Store name inside the object
        self.speed = speed      # Store speed inside the object
        self.is_running = False  # Default value --- always starts as False

# Now we create a Motor by passing arguments directly
m1 = Motor("Left Motor", 100)
m2 = Motor("Right Motor", 200)

print(m1.name, "speed:", m1.speed, "running:", m1.is_running)
print(m2.name, "speed:", m2.speed, "running:", m2.is_running)
````

Let's break this down:

```python
def __init__(self, name, speed):
```

- `def` --- we are defining a function (method) inside the class
- `__init__` --- the special name Python looks for (double underscores on each side)
- `self` --- refers to **the object being created** (more on this in Part 5)
- `name, speed` --- parameters we want the user to provide

```python
self.name = name
```

- `self.name` --- creates an attribute **on the object**
- `= name` --- assigns the value passed by the user

When we write `Motor("Left Motor", 100)`, Python does two things:
1. Creates a new empty Motor object
2. Calls `__init__(new_object, "Left Motor", 100)` automatically

**Figure 2.3** --- `__init__` with default values

````python
class Sensor:
    def __init__(self, name, unit, reading=0.0):
        self.name = name
        self.unit = unit
        self.reading = reading  # Default is 0.0 if not provided

# Providing all arguments
s1 = Sensor("Thermometer", "Celsius", 23.5)
print(f"{s1.name}: {s1.reading} {s1.unit}")

# Using the default value for reading
s2 = Sensor("Barometer", "hPa")
print(f"{s2.name}: {s2.reading} {s2.unit}")
````

## Part 3: Why `__init__` Matters

**Figure 2.4** --- Without `__init__`, things can go wrong

````python
# BAD: Without __init__, you might forget attributes
class BadMotor:
    pass

m = BadMotor()
m.name = "Motor X"
# Oops! Forgot to set m.speed

try:
    print(f"{m.name} speed: {m.speed}")  # AttributeError!
except AttributeError as e:
    print(f"Error: {e}")
````

**Figure 2.5** — Creating objects from a class

````python
# GOOD: With __init__, every object is set up correctly
class GoodMotor:
    def __init__(self, name, speed):
        self.name = name
        self.speed = speed

# Cannot forget --- Python requires the arguments!
try:
    m = GoodMotor()  # Missing arguments!
except TypeError as e:
    print(f"Error: {e}")

# This works perfectly
m = GoodMotor("Motor X", 100)
print(f"{m.name} speed: {m.speed}")
````

| Without `__init__` | With `__init__` |
|---|---|
| Attributes may be missing | All attributes guaranteed |
| No error if you forget | Python tells you what is missing |
| Objects may be inconsistent | Every object has the same structure |
| Multiple lines to set up | One line to create |

## Part 4: Adding Methods --- Giving Objects Behavior

So far, our objects can only **store data**. But real objects **do things**:
- A motor can **start** and **stop**
- A sensor can **take a reading**
- An LED can **turn on** and **turn off**

A **method** is a function defined inside a class. It describes what an object can **do**.

**Figure 2.5** --- Adding methods to a class

````python
class Motor:
    def __init__(self, name, max_speed):
        self.name = name
        self.max_speed = max_speed
        self.speed = 0
        self.is_running = False

    def start(self):                  # A method --- no extra parameters
        self.is_running = True
        print(f"{self.name} started.")

    def stop(self):                   # Another method
        self.is_running = False
        self.speed = 0
        print(f"{self.name} stopped.")

    def set_speed(self, new_speed):   # A method with a parameter
        if new_speed > self.max_speed:
            print(f"Warning: {new_speed} exceeds max speed {self.max_speed}!")
            self.speed = self.max_speed
        else:
            self.speed = new_speed
        print(f"{self.name} speed set to {self.speed} RPM.")
````

**Figure 2.6** — Working with object attributes

````python
# Using the methods
m = Motor("Drive Motor", 500)

m.start()            # Call a method with dot notation
m.set_speed(200)     # Pass an argument to a method
m.set_speed(600)     # Try to exceed max speed
m.stop()
````

Notice the pattern:
- Methods are defined **inside** the class, indented under `class Motor:`
- Every method takes `self` as its **first** parameter
- We call methods with `object.method()` --- same dot notation as attributes
- Methods can **read** and **modify** the object's attributes through `self`

**Figure 2.6** --- Methods that return values

````python
class Sensor:
    def __init__(self, name, unit):
        self.name = name
        self.unit = unit
        self.readings = []      # Empty list to store readings

    def add_reading(self, value):
        self.readings.append(value)        # Add to list

    def get_average(self):                 # Returns a value
        if len(self.readings) == 0:
            return 0.0
        return sum(self.readings) / len(self.readings)

    def get_count(self):                   # Returns a value
        return len(self.readings)

# Use the sensor
temp = Sensor("Thermometer", "Celsius")
temp.add_reading(22.0)
temp.add_reading(23.5)
temp.add_reading(21.8)

avg = temp.get_average()   # Store the returned value
print(f"Average of {temp.get_count()} readings: {avg:.1f} {temp.unit}")
````

## Part 5: The `self` Parameter

`self` is Python's way of letting a method know **which object** it is working on.

**Figure 2.7** --- Understanding `self`

````python
class LED:
    def __init__(self, color):
        self.color = color
        self.is_on = False

    def turn_on(self):
        self.is_on = True
        print(f"{self.color} LED is ON")

    def turn_off(self):
        self.is_on = False
        print(f"{self.color} LED is OFF")

# Create two LEDs
red_led = LED("Red")
green_led = LED("Green")

# When we call red_led.turn_on(), Python translates it to:
# LED.turn_on(red_led)
# So 'self' becomes 'red_led'

red_led.turn_on()     # self = red_led  --> self.color = "Red"
green_led.turn_on()   # self = green_led --> self.color = "Green"
````

Key points about `self`:

| Rule | Explanation |
|---|---|
| Always the **first** parameter | `def turn_on(self):` |
| You **never** pass it manually | `red_led.turn_on()` not `red_led.turn_on(red_led)` |
| Python fills it in for you | `self` becomes the object before the dot |
| Use it to access attributes | `self.color`, `self.is_on` |
| Use it to call other methods | `self.turn_off()` from inside another method |

**Figure 2.8** --- Common error: forgetting `self`

````python
# ERROR: Forgetting self as parameter
class BrokenLED:
    def __init__(self, color):
        self.color = color

    def turn_on():          # Missing self!
        print("LED is ON")

led = BrokenLED("Blue")
try:
    led.turn_on()  # TypeError!
except TypeError as e:
    print(f"Error: {e}")
    print("Fix: Add 'self' as the first parameter of turn_on.")
````

## Part 6: `__repr__` --- Making Objects Readable

What happens when you print an object right now?

**Figure 2.9** --- The default print output is not helpful

````python
class Motor:
    def __init__(self, name, speed):
        self.name = name
        self.speed = speed

m = Motor("Motor A", 100)
print(m)  # Not very helpful! Just shows memory address
````

We can fix this by adding a `__repr__` method. Python calls `__repr__` whenever it needs a **string representation** of the object.

**Figure 2.10** --- Adding `__repr__`

````python
class Motor:
    def __init__(self, name, speed):
        self.name = name
        self.speed = speed

    def __repr__(self):  # Must return a string
        return f"Motor(name='{self.name}', speed={self.speed})"

m = Motor("Motor A", 100)
print(m)   # Now it's readable!

# Also works in lists
motors = [Motor("M1", 100), Motor("M2", 200), Motor("M3", 50)]
print(motors)
````

The `__repr__` method:
- Must take only `self` as a parameter
- Must **return** a string (not print it!)
- Should ideally show how to **recreate** the object
- Is used by `print()`, in lists, and in the interactive console

---
## Part 7: `__str__` vs `__repr__`

Python has **two** methods for converting objects to strings:

| Method | Purpose | When it is called |
|---|---|---|
| `__repr__` | Developer-friendly representation | `repr(obj)`, in lists, in console |
| `__str__` | User-friendly display | `print(obj)`, `str(obj)` |

If only one is defined, `__repr__` is used everywhere.  
If both are defined, `print()` uses `__str__` and the console uses `__repr__`.

**Figure 2.11** --- `__str__` vs `__repr__` in action

````python
class Sensor:
    def __init__(self, name, unit, reading):
        self.name = name
        self.unit = unit
        self.reading = reading

    def __repr__(self):
        # Technical: shows how to recreate the object
        return f"Sensor('{self.name}', '{self.unit}', {self.reading})"

    def __str__(self):
        # Friendly: for end users
        return f"{self.name}: {self.reading} {self.unit}"

s = Sensor("Thermometer", "Celsius", 23.5)

print(str(s))     # Uses __str__ --- friendly
print(repr(s))    # Uses __repr__ --- technical
print(s)          # print() uses __str__ if available
print([s])        # Lists always use __repr__
````

**Tip for beginners:** If you only define one, define `__repr__`. It covers more cases.

## Part 8: Putting It All Together

Let's build a complete, user-friendly class step by step.

**Figure 2.12** --- A complete `LED` class

````python
class LED:
    def __init__(self, color, brightness=100):
        self.color = color             # e.g., "red", "green"
        self.brightness = brightness   # 0 to 100
        self.is_on = False             # Starts OFF

    def turn_on(self):
        self.is_on = True

    def turn_off(self):
        self.is_on = False

    def set_brightness(self, level):
        if level < 0:
            self.brightness = 0
        elif level > 100:
            self.brightness = 100
        else:
            self.brightness = level

    def __repr__(self):
        return f"LED('{self.color}', brightness={self.brightness})"

    def __str__(self):
        status = "ON" if self.is_on else "OFF"
        return f"{self.color} LED [{status}] brightness={self.brightness}%"

# Demo
led = LED("red")
print(led)              # OFF by default

led.turn_on()
print(led)              # Now ON

led.set_brightness(50)
print(led)              # Dimmed

print(repr(led))        # Technical representation
````

**Figure 2.13** --- A complete `Servo` class

````python
class Servo:
    def __init__(self, servo_id, min_angle=0, max_angle=180):
        self.servo_id = servo_id
        self.min_angle = min_angle
        self.max_angle = max_angle
        self.angle = min_angle       # Start at minimum

    def move_to(self, angle):
        if angle < self.min_angle:
            self.angle = self.min_angle
        elif angle > self.max_angle:
            self.angle = self.max_angle
        else:
            self.angle = angle

    def center(self):
        self.angle = (self.min_angle + self.max_angle) // 2

    def __repr__(self):
        return f"Servo({self.servo_id}, angle={self.angle})"

    def __str__(self):
        return f"Servo #{self.servo_id} at {self.angle} degrees"

# Demo
s = Servo(1)
print(s)

s.move_to(90)
print(s)

s.move_to(999)   # Clamped to max
print(s)

s.center()
print(s)
````

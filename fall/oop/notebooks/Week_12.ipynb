## Part 2: Building the Sensor Class

The `Sensor` class simulates a real sensor. In mechatronics, sensors measure things like temperature, pressure, distance, and speed. Real sensors always have some **noise** (small random errors).

Our `Sensor` will:
- Have a **name** (e.g., "Temperature Sensor")
- Know its **range** (min and max values)
- **Generate** random readings within that range
- **Store** all readings in a list

**Figure 12.4** — Sensor class diagram:

```
+---------------------------+
|         Sensor            |
+---------------------------+
| - name: str               |
| - min_val: float          |
| - max_val: float          |
| - readings: list          |
+---------------------------+
| + generate(n) -> list     |
| + get_readings() -> list  |
| + get_last() -> float     |
| + __str__() -> str        |
+---------------------------+
```

Here is the starter code. Study it carefully:

````python
%%writefile mech_toolkit/sensor.py
# sensor.py - Simulated Sensor class

import random

class Sensor:
    """A simulated sensor that generates random readings."""

    def __init__(self, name, min_val=0.0, max_val=100.0):
        """
        Create a new sensor.

        Args:
            name: Name of the sensor (e.g., "Temperature")
            min_val: Minimum possible reading
            max_val: Maximum possible reading
        """
        self.name = name
        self.min_val = min_val
        self.max_val = max_val
        self.readings = []   # Store all readings here

    def generate(self, n=10):
        """
        Generate n random readings and add them to the readings list.

        Args:
            n: Number of readings to generate (default 10)

        Returns:
            List of newly generated readings
        """
        new_readings = []
        for i in range(n):
            # Generate a random value within the sensor range
            value = random.uniform(self.min_val, self.max_val)
            # Round to 2 decimal places
            value = round(value, 2)
            new_readings.append(value)

        # Add new readings to the full list
        self.readings.extend(new_readings)
        return new_readings

    def get_readings(self):
        """Return all collected readings."""
        return self.readings.copy()

    def get_last(self):
        """Return the last reading, or None if no readings exist."""
        if len(self.readings) == 0:
            return None
        return self.readings[-1]

    def __str__(self):
        """Return a string description of the sensor."""
        count = len(self.readings)
        return f"Sensor('{self.name}', range=[{self.min_val}, {self.max_val}], readings={count})"
````

## Part 3: Building the Filter Class

Real sensor data is **noisy**. To get useful information, we need to **filter** (smooth) the data. Two common simple filters are:

1. **Moving Average** — replace each value with the average of nearby values
2. **Median Filter** — replace each value with the median of nearby values

**Figure 12.5** — How a moving average works (window size = 3):

```
Raw data:      [10, 15, 12, 18, 14, 20, 16]
                 |       |       |
Window 1:      [10, 15, 12]  -> average = 12.33
Window 2:          [15, 12, 18]  -> average = 15.0
Window 3:              [12, 18, 14]  -> average = 14.67
...
```

The `window_size` controls how much smoothing is applied. A larger window means more smoothing.

**Figure 12.6** — Filter class diagram:

```
+----------------------------------+
|            Filter                |
+----------------------------------+
| - window_size: int               |
+----------------------------------+
| + moving_average(data) -> list   |
| + median_filter(data) -> list    |
| + __str__() -> str               |
+----------------------------------+
```

````python
%%writefile mech_toolkit/filter.py
# filter.py - Data filtering class

class Filter:
    """A simple data filter for smoothing sensor readings."""

    def __init__(self, window_size=3):
        """
        Create a new filter.

        Args:
            window_size: Number of values to average together (default 3)
        """
        if window_size < 1:
            window_size = 1
        self.window_size = window_size

    def moving_average(self, data):
        """
        Apply a moving average filter to the data.

        Args:
            data: List of numbers to filter

        Returns:
            List of filtered (smoothed) values
        """
        if len(data) == 0:
            return []

        result = []
        for i in range(len(data)):
            # Get the window of values around position i
            start = max(0, i - self.window_size // 2)
            end = min(len(data), start + self.window_size)
            # Adjust start if we are near the end
            start = max(0, end - self.window_size)

            window = data[start:end]
            avg = sum(window) / len(window)
            result.append(round(avg, 2))

        return result

    def median_filter(self, data):
        """
        Apply a median filter to the data.

        Args:
            data: List of numbers to filter

        Returns:
            List of filtered values
        """
        if len(data) == 0:
            return []

        result = []
        for i in range(len(data)):
            start = max(0, i - self.window_size // 2)
            end = min(len(data), start + self.window_size)
            start = max(0, end - self.window_size)

            window = sorted(data[start:end])
            mid = len(window) // 2
            median_val = window[mid]
            result.append(round(median_val, 2))

        return result

    def __str__(self):
        return f"Filter(window_size={self.window_size})"
````

**Figure 12.3** — Module imports

````python
# Test the Filter class
from mech_toolkit.filter import Filter

# Create some noisy test data
noisy_data = [20.0, 22.5, 19.0, 25.0, 21.0, 28.0, 20.5, 23.0, 27.0, 22.0]
print(f"Raw data:      {noisy_data}")

# Apply moving average with window size 3
f = Filter(window_size=3)
smoothed = f.moving_average(noisy_data)
print(f"Moving avg(3): {smoothed}")

# Apply median filter with window size 3
median = f.median_filter(noisy_data)
print(f"Median(3):     {median}")

# Try a larger window
f5 = Filter(window_size=5)
smoothed5 = f5.moving_average(noisy_data)
print(f"Moving avg(5): {smoothed5}")
````

## Part 4: Building the Logger Class

The final component is the `Logger`. In real mechatronics projects, you always need to **save data** for analysis. CSV (Comma-Separated Values) is one of the simplest and most common formats.

**Figure 12.7** — What a CSV file looks like:

```
index,raw,filtered
0,20.0,20.5
1,22.5,20.83
2,19.0,22.17
...
```

**Figure 12.8** — Logger class diagram:

```
+---------------------------------------+
|              Logger                   |
+---------------------------------------+
| - filename: str                       |
+---------------------------------------+
| + save(data, headers) -> None         |
| + save_two_columns(raw, filt) -> None |
| + __str__() -> str                    |
+---------------------------------------+
```

````python
%%writefile mech_toolkit/logger.py
# logger.py - CSV data logger class

import csv

class Logger:
    """A simple CSV logger for saving sensor data."""

    def __init__(self, filename="output.csv"):
        """
        Create a new logger.

        Args:
            filename: Name of the CSV file to write (default: output.csv)
        """
        self.filename = filename

    def save(self, data, headers=None):
        """
        Save a list of rows to a CSV file.

        Args:
            data: List of lists (each inner list is a row)
            headers: Optional list of column headers
        """
        with open(self.filename, "w", newline="") as f:
            writer = csv.writer(f)
            if headers:
                writer.writerow(headers)
            for row in data:
                writer.writerow(row)

        print(f"Data saved to '{self.filename}' ({len(data)} rows)")

    def save_two_columns(self, raw, filtered):
        """
        Save raw and filtered data side by side.

        Args:
            raw: List of raw sensor readings
            filtered: List of filtered readings
        """
        headers = ["index", "raw", "filtered"]
        rows = []
        for i in range(len(raw)):
            raw_val = raw[i]
            filt_val = filtered[i] if i < len(filtered) else ""
            rows.append([i, raw_val, filt_val])

        self.save(rows, headers=headers)

    def __str__(self):
        return f"Logger(filename='{self.filename}')"
````

## Part 5: Wiring Everything Together

Now we have all three components. Let's connect them into a complete pipeline!

**Figure 12.9** — Complete data pipeline:

```
1. Create a Sensor      --> generate readings
2. Create a Filter       --> smooth the readings
3. Create a Logger       --> save to CSV
4. Print a summary       --> display results
```

````python
# Complete pipeline demonstration
from mech_toolkit.sensor import Sensor
from mech_toolkit.filter import Filter
from mech_toolkit.logger import Logger

# Step 1: Create and use the sensor
print("=" * 50)
print("MECHATRONICS TOOLKIT - Data Pipeline")
print("=" * 50)

sensor = Sensor("Temperature", min_val=18.0, max_val=30.0)
raw_data = sensor.generate(20)
print(f"\n1. Sensor: {sensor}")
print(f"   Raw data (first 5): {raw_data[:5]}")

# Step 2: Filter the data
f = Filter(window_size=3)
avg_data = f.moving_average(raw_data)
med_data = f.median_filter(raw_data)
print(f"\n2. Filter: {f}")
print(f"   Avg data (first 5): {avg_data[:5]}")
print(f"   Med data (first 5): {med_data[:5]}")

# Step 3: Log the data
logger = Logger("sensor_report.csv")
logger.save_two_columns(raw_data, avg_data)
print(f"\n3. Logger: {logger}")

# Step 4: Summary
print(f"\n{'=' * 50}")
print(f"Pipeline complete!")
print(f"  Readings collected: {len(raw_data)}")
print(f"  Raw average: {sum(raw_data)/len(raw_data):.2f}")
print(f"  Filtered average: {sum(avg_data)/len(avg_data):.2f}")
print(f"  Data saved to: {logger.filename}")
````

## Part 5b: Visualizing the Pipeline Output

Numbers in a table are useful, but a **plot** makes the effect of filtering immediately clear. Let's use `matplotlib` (from CP2) to visualize our pipeline's raw vs filtered data.

**Figure 12.10** — Plotting raw vs filtered sensor data:

````python
# Figure 12.10 - Visualizing raw vs filtered sensor data

import matplotlib.pyplot as plt
from mech_toolkit.sensor import Sensor
from mech_toolkit.filter import Filter

# Generate noisy sensor data
sensor = Sensor("Temperature", min_val=18.0, max_val=30.0)
raw = sensor.generate(30)

# Apply two different filters
f3 = Filter(window_size=3)
f7 = Filter(window_size=7)
avg3 = f3.moving_average(raw)
avg7 = f7.moving_average(raw)
med3 = f3.median_filter(raw)

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: moving average comparison
axes[0].plot(raw, 'o-', color='gray', alpha=0.4, label='Raw', markersize=4)
axes[0].plot(avg3, 's-', color='steelblue', label='Moving Avg (w=3)', linewidth=2)
axes[0].plot(avg7, 'd-', color='tomato', label='Moving Avg (w=7)', linewidth=2)
axes[0].set_xlabel('Reading #')
axes[0].set_ylabel('Temperature (°C)')
axes[0].set_title('Moving Average: Window Size Comparison')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Right: moving average vs median
axes[1].plot(raw, 'o-', color='gray', alpha=0.4, label='Raw', markersize=4)
axes[1].plot(avg3, 's-', color='steelblue', label='Moving Avg (w=3)', linewidth=2)
axes[1].plot(med3, '^-', color='seagreen', label='Median (w=3)', linewidth=2)
axes[1].set_xlabel('Reading #')
axes[1].set_ylabel('Temperature (°C)')
axes[1].set_title('Moving Average vs Median Filter')
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Tip: The median filter is better at removing sudden spikes (outliers).")
print("The moving average is smoother overall but keeps some spike influence.")
````

## Part 6: Basic Self-Tests

In real software projects, we write **tests** to make sure our code works correctly. Let's write some simple tests for each component.

A basic test checks if the output matches what we expect. If the test **passes**, we print a success message. If it **fails**, we print what went wrong.

**Figure 12.5** — check() function

````python
# Simple test helper function
def check(condition, test_name):
    """Print PASS or FAIL for a test."""
    if condition:
        print(f"  PASS: {test_name}")
    else:
        print(f"  FAIL: {test_name}")
````

**Figure 12.6** — Module imports

````python
# Tests for the Sensor class
from mech_toolkit.sensor import Sensor

print("Testing Sensor class:")

# Test 1: Sensor starts with no readings
s = Sensor("Test", 0, 100)
check(len(s.readings) == 0, "Sensor starts with empty readings")

# Test 2: Generate creates the right number of readings
s.generate(5)
check(len(s.readings) == 5, "generate(5) creates 5 readings")

# Test 3: All readings are within range
all_in_range = all(0 <= r <= 100 for r in s.readings)
check(all_in_range, "All readings within [0, 100]")

# Test 4: get_last returns the last reading
check(s.get_last() == s.readings[-1], "get_last() returns correct value")

# Test 5: get_last on empty sensor returns None
empty_sensor = Sensor("Empty", 0, 10)
check(empty_sensor.get_last() is None, "get_last() on empty sensor returns None")
````

**Figure 12.7** — Module imports

````python
# Tests for the Filter class
from mech_toolkit.filter import Filter

print("Testing Filter class:")

# Test 1: Moving average of identical values should be the same value
f = Filter(window_size=3)
same = f.moving_average([5.0, 5.0, 5.0, 5.0, 5.0])
check(all(v == 5.0 for v in same), "Moving avg of identical values stays the same")

# Test 2: Filter returns same length as input
data = [1, 2, 3, 4, 5]
result = f.moving_average(data)
check(len(result) == len(data), "Output length equals input length")

# Test 3: Empty input returns empty output
check(f.moving_average([]) == [], "Empty input returns empty output")

# Test 4: Median filter with known data
med_result = f.median_filter([1, 100, 3])
# With window 3, the median of [1, 100, 3] sorted is [1, 3, 100], median = 3
check(med_result[1] == 3, "Median filter removes outlier")
````

**Figure 12.8** — Module imports

````python
# Tests for the Logger class
from mech_toolkit.logger import Logger
import os

print("Testing Logger class:")

# Test 1: Logger creates a file
log = Logger("test_check.csv")
log.save([[1, 2], [3, 4]], headers=["a", "b"])
check(os.path.exists("test_check.csv"), "CSV file was created")

# Test 2: File has correct content
with open("test_check.csv", "r") as f:
    lines = f.readlines()
check(len(lines) == 3, "File has header + 2 data rows")
check("a,b" in lines[0], "Header row is correct")

# Test 3: save_two_columns works
log2 = Logger("test_two_col.csv")
log2.save_two_columns([1.0, 2.0, 3.0], [1.5, 2.5, 3.5])
with open("test_two_col.csv", "r") as f:
    lines = f.readlines()
check(len(lines) == 4, "Two-column file has header + 3 rows")

# Clean up test files
os.remove("test_check.csv")
os.remove("test_two_col.csv")
print("Test files cleaned up.")
````

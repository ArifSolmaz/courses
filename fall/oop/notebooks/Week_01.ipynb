## Part 1: What is OOP and Why?

Look around you. Everything is an **object**:

| Real-World Object | Data (State) | Behavior |
|---|---|---|
| A motor | speed, voltage, is_running | start, stop, change_speed |
| A sensor | reading, unit, is_calibrated | read, calibrate, reset |
| A robot | position, battery_level | move, turn, charge |
| An LED | color, brightness, is_on | turn_on, turn_off, dim |

In programming, we often need to represent things like these.  
OOP lets us **bundle data and behavior together** into a single unit.

### Without OOP (Procedural Approach)

Imagine tracking two motors using plain variables:

**Figure 1.1** — Working with object attributes

````python
# Procedural approach: separate variables for each motor
motor1_name = "Left Motor"
motor1_speed = 0
motor1_is_running = False

motor2_name = "Right Motor"
motor2_speed = 0
motor2_is_running = False

# What if we have 10 motors? 100 motors?
# This gets messy very quickly!
print(motor1_name, "speed:", motor1_speed)
````

The procedural approach has problems:
- Variables are **scattered** --- nothing groups them together
- Easy to make **mistakes** (e.g., using `motor1_speed` when you meant `motor2_speed`)
- Hard to **scale** to many objects

OOP solves this by letting us create a **blueprint** and stamp out objects from it.

---
## Part 2: Classes and Objects --- The Blueprint Analogy

Think of a **class** as a **blueprint** (or a cookie cutter).  
Think of an **object** as the **actual thing** built from that blueprint (or the cookie).

| Term | Analogy | Example |
|---|---|---|
| **Class** | Blueprint / Template | The design for a DC motor |
| **Object** | Actual thing built from the blueprint | Motor #1 on your robot |
| **Attribute** | A piece of data stored in the object | `speed = 100` |
| **Instance** | Another word for object | Motor #1 is an *instance* of the Motor class |

One class can produce **many** objects, just like one blueprint can produce many houses.

---
## Part 3: Your First Class

Let's create the simplest possible class in Python.

**Figure 1.1** --- A minimal class definition

````python
# Define a class called Motor
class Motor:
    pass  # 'pass' means "this class is empty for now"

# That's it! We just created a class.
print(Motor)       # Shows the class itself
print(type(Motor)) # Motor is of type 'type'
````

Key rules:
- Use the keyword `class` followed by the class name
- Class names use **CamelCase** (capital first letter, no underscores): `Motor`, `TemperatureSensor`, `RobotArm`
- The body is indented (just like `if`, `for`, `def`)
- `pass` is a placeholder that does nothing --- we will replace it soon

---
## Part 4: Creating Objects (Instances)

To create an object from a class, we **call** the class like a function.

**Figure 1.2** --- Creating objects from a class

````python
# Create two Motor objects
motor1 = Motor()  # First instance
motor2 = Motor()  # Second instance

print(motor1)  # Each object has a unique identity (memory address)
print(motor2)  # Different address = different object
````

**Figure 1.3** — Defining a minimal class

````python
# Are they the same object?
print(motor1 == motor2)     # False --- they are different objects
print(type(motor1))         # <class '__main__.Motor'>
print(isinstance(motor1, Motor))  # True --- motor1 IS a Motor
````

Each time we write `Motor()`, Python creates a **new, independent** object in memory.

---
## Part 5: Adding More Attributes

An empty class is not very useful. Let's give our objects some **data**.

We can attach attributes to an object using **dot notation**: `object.attribute = value`

**Figure 1.3** --- Adding attributes to objects

````python
# Create a Motor object
motor1 = Motor()

# Add attributes using dot notation
motor1.name = "Left Motor"    # A string attribute
motor1.speed = 0              # An integer attribute
motor1.voltage = 12.0         # A float attribute
motor1.is_running = False     # A boolean attribute

# Read attributes using dot notation
print("Name:", motor1.name)
print("Speed:", motor1.speed)
print("Voltage:", motor1.voltage)
print("Running:", motor1.is_running)
````

**Figure 1.4** — Creating objects from a class

````python
# Create a second motor with different data
motor2 = Motor()
motor2.name = "Right Motor"
motor2.speed = 150
motor2.voltage = 12.0
motor2.is_running = True

# Each object has its OWN copy of the attributes
print(motor1.name, "speed:", motor1.speed)
print(motor2.name, "speed:", motor2.speed)
````

Notice: `motor1` and `motor2` have the **same attribute names** but **different values**.  
Each object is independent --- changing one does not affect the other.

**Figure 1.4** --- Modifying attributes

````python
# We can change attributes after creation
motor1.speed = 200
motor1.is_running = True

print(motor1.name, "is now running at speed", motor1.speed)

# motor2 is not affected
print(motor2.name, "speed is still", motor2.speed)
````

## Part 6: State --- The Data Inside Objects

The **state** of an object is the collection of all its attribute values at a given moment.

| | motor1 | motor2 |
|---|---|---|
| name | "Left Motor" | "Right Motor" |
| speed | 200 | 150 |
| voltage | 12.0 | 12.0 |
| is_running | True | True |

As your program runs, the state **changes** --- motors speed up, sensors take new readings, batteries drain.

**Figure 1.5** --- Watching state change over time

````python
# Let's simulate a sensor taking readings
class Sensor:
    pass

temp_sensor = Sensor()
temp_sensor.name = "Temperature Sensor"
temp_sensor.unit = "Celsius"
temp_sensor.reading = 22.5

print(f"Reading: {temp_sensor.reading} {temp_sensor.unit}")

# The sensor takes a new reading --- state changes!
temp_sensor.reading = 23.1
print(f"Reading: {temp_sensor.reading} {temp_sensor.unit}")

# Another reading
temp_sensor.reading = 24.8
print(f"Reading: {temp_sensor.reading} {temp_sensor.unit}")
````

**Figure 1.6** --- Using objects in calculations

````python
# Objects can be used in expressions just like regular variables
class LED:
    pass

led1 = LED()
led1.color = "red"
led1.brightness = 80   # percent
led1.is_on = True

led2 = LED()
led2.color = "green"
led2.brightness = 50
led2.is_on = True

# Calculate average brightness
avg_brightness = (led1.brightness + led2.brightness) / 2
print(f"Average LED brightness: {avg_brightness}%")
````

**Figure 1.7** --- Storing objects in a list

````python
# Objects can be stored in lists, just like numbers or strings
leds = [led1, led2]

# Loop through the list
for led in leds:
    print(f"{led.color} LED, brightness: {led.brightness}%")
````

## Part 7: Common Errors with Classes

Here are mistakes beginners often make. Let's see each one and learn how to fix it.

**Figure 1.8** --- Error: accessing an attribute that does not exist

````python
# ERROR: Accessing an attribute that was never set
class Robot:
    pass

robot1 = Robot()
robot1.name = "WallE"

# This will cause an AttributeError
try:
    print(robot1.battery_level)  # We never set this!
except AttributeError as e:
    print(f"Error: {e}")
    print("Fix: Make sure to set the attribute before reading it.")
````

**Figure 1.9** --- Error: forgetting the parentheses

````python
# ERROR: Forgetting parentheses when creating an object
wrong = Motor     # This does NOT create an object!
right = Motor()   # This DOES create an object

print("wrong:", wrong)   # This is the class itself, not an object
print("right:", right)   # This is an actual Motor object
print()
print("Always remember the parentheses: Motor()")
````

**Figure 1.10** --- Error: objects are independent

````python
# Common misconception: setting an attribute on one object does NOT set it on others
m1 = Motor()
m2 = Motor()

m1.speed = 100

# m2 does NOT have speed set
try:
    print(m2.speed)
except AttributeError as e:
    print(f"Error: {e}")
    print("Each object is independent. You must set attributes on each one.")
````
